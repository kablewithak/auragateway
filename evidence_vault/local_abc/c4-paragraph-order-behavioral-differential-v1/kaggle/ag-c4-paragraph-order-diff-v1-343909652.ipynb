```python
"""Generated C4 paragraph-order differential transaction-bound wrapper V1."""

from __future__ import annotations

import base64
import hashlib
import json
import subprocess
import sys
from datetime import UTC, datetime
from pathlib import Path

_AUTHORIZATION_B64 = (
    "eyJhdXRob3JpemF0aW9uIjp7ImF1dGhvcml6YXRpb25fY2hhbGxlbmdlX3NoYTI1NiI6IjIz"
    "MTRkNTk1MWYxYWY2ZWJmN2NlNTA1ODhhYjYxNTIwMmNkMGQxNzljMThkNDAxM2U4MWZhMDE3"
    "ZWM3ZGUzNDciLCJhdXRob3JpemF0aW9uX2Rlc2lnbl9tZXJnZV9jb21taXQiOiIyZTBlMDli"
    "NjUwMjMzOTdmMzBkNzNhNDA2ZjNmNjBlNzA5MGM4NWIxIiwiYXV0aG9yaXphdGlvbl9kZXNp"
    "Z25fcmVjb3JkX3NoYTI1NiI6IjgzMDVlYmIxNTNmOTYyMDE1YzI4ZGU5OGJkNmZjZjZmZWVi"
    "MjAyNDgyMTYzYzZjY2UzZjBjYWYwOGNjM2QxNDMiLCJhdXRob3JpemF0aW9uX2lkIjoiYzRw"
    "by1hNWVhZTkyMDBmY2M0NGIwY2EwNjQ0M2RlZWFmNTg5MSIsImF1dGhvcml6YXRpb25fcmV1"
    "c2FibGUiOmZhbHNlLCJhdXRob3JpemF0aW9uX3dpbmRvd19taW51dGVzIjoxODAsImJ1ZGdl"
    "dCI6eyJtYXhpbXVtX2JlbmNobWFya190cmFqZWN0b3J5X3JlcXVlc3RzIjowLCJtYXhpbXVt"
    "X2V4dGVybmFsX25ldHdvcmtfcmVxdWVzdHMiOjAsIm1heGltdW1fZXh0ZXJuYWxfc3BlbmQi"
    "OjAsIm1heGltdW1faGlkZGVuX3JldHJpZXMiOjAsIm1heGltdW1fa2FnZ2xlX3Nlc3Npb25z"
    "IjoxLCJtYXhpbXVtX21vZGVsX2xvYWRzIjo2LCJtYXhpbXVtX21vZGVsX3JlcXVlc3RzIjo2"
    "LCJtYXhpbXVtX291dHB1dF90b2tlbnNfcGVyX3JlcXVlc3QiOjMyLCJtYXhpbXVtX3JlcGxh"
    "Y2VtZW50X29ic2VydmF0aW9ucyI6MCwibWF4aW11bV9ydW50aW1lX2ltcG9ydF9jbG9zdXJl"
    "X3Byb2JlcyI6MSwibWF4aW11bV9ydW50aW1lX2luc3RhbGxfYXR0ZW1wdHMiOjEsIm1heGlt"
    "dW1fc2F2ZV9hbmRfcnVuX2FsbF9hY3Rpb25zIjoxLCJtYXhpbXVtX3dvcmtlcl9zdGFydHMi"
    "OjYsIm1heGltdW1fd29ya2VyX3RlYXJkb3ducyI6Nn0sImJ1ZGdldF9jb250cmFjdF9zaGEy"
    "NTYiOiI2MzljNDgwZjkxYzUyNjJkMzQzMjVlNzlhYzcxY2NiNzgxMjk2MGI2NmM5ODgzODE4"
    "YWVjMjc3NTA4N2Y2YjgxIiwiZGVjaXNpb24iOiJBVVRIT1JJWkVEIiwiZHVyYWJsZV9wbGF0"
    "Zm9ybV9vYnNlcnZhdGlvbl9yZXF1aXJlZCI6dHJ1ZSwiZXZlcnlfdGVybWluYWxfYXR0ZW1w"
    "dF9jb25zdW1lc19hdXRob3JpemF0aW9uIjp0cnVlLCJleHBlcmltZW50Ijp7ImFzc2lzdGFu"
    "dF9hY2tub3dsZWRnZW1lbnQiOiJTeW50aGV0aWMgZGV0ZXJtaW5pc3RpYyBjb250ZXh0IGFj"
    "a25vd2xlZGdlZC4iLCJjYW5vbmljYWxfY29ycHVzX2dsb2JhbF9pbnZhbGlkYXRpb25fcGVy"
    "bWl0dGVkIjpmYWxzZSwiY2Fub25pY2FsX2ZpbmFsX29iamVjdCI6IntcInByb2JlXCI6XCJl"
    "eGFjdC1ydW50aW1lLXA1LXA2XCIsXCJ2YWx1ZVwiOjF9IiwiY2Fub25pY2FsX29iamVjdF9z"
    "aGEyNTYiOiI0NDhmYWQzZDNhYzVjMmYxMWY0YzA5YjBkZjFlN2U2MjM3Y2UyYTA5MTg1Zjk5"
    "NTAzOTQ2MzExODc1ZjVlMTEzIiwiY2hhcmFjdGVyX2NvdW50X3ByZXNlcnZlZCI6dHJ1ZSwi"
    "Y29tbW9uX3N1ZmZpeF90b2tlbl9jb3VudCI6MTIyLCJjb250cm9sX2FuY2hvcl9oaXN0b3Jp"
    "Y2FsX3BhcnNlZF9pZGVudGl0eV9yZXF1aXJlZCI6dHJ1ZSwiY29udHJvbF9hbmNob3JfbXVz"
    "dF9yZXByb2R1Y2VfemVyb19vZl90aHJlZV9leGFjdCI6dHJ1ZSwiY29udHJvbF9hbmNob3Jf"
    "dmFsaWRfanNvbl90aHJlZV9vZl90aHJlZV9yZXF1aXJlZCI6dHJ1ZSwiY29udHJvbF9jb25k"
    "aXRpb25faWQiOiJDT05UUk9MX09SSUdJTkFMX0M0IiwiY29udHJvbF9wYXJhZ3JhcGhfb3Jk"
    "ZXIiOlsxLDIsMyw0LDUsNiw3LDgsOSwxMF0sImNvbnRyb2xfcGF5bG9hZF9zaGEyNTYiOiJh"
    "ODg4YzE3ZWQ4ZTgyMzYwZmRkNDZkMGJiNjgzM2RiOWRiMmRjM2ZiYmZiMTRkODYxZjcwNjNi"
    "MjcxMDYzNzg4IiwiY29udHJvbF90b2tlbl9zaGEyNTYiOiJmMDA5YjE0OWIwYjhjY2YwOGE0"
    "MjMzNDZlMTczNmJlODE5MjcwOTU5MDdiNTIyMWUwNzBlNTlmZmM2ZDg3ZjRjIiwiZGVjaXNp"
    "b25fc3RhdGVzIjpbIkNPTlRST0xfQU5DSE9SX05PTlJFUFJPRFVDVElPTl9JTlZBTElEQVRF"
    "U19JTkZFUkVOQ0UiLCJPUkRFUl9JTlRFUlZFTlRJT05fUkVTVE9SRVNfQkVIQVZJT1IiLCJP"
    "UkRFUl9JTlRFUlZFTlRJT05fRE9FU19OT1RfQ0hBTkdFX09CU0VSVkVEX1BIRU5PVFlQRSIs"
    "Ik9SREVSX0lOVEVSVkVOVElPTl9DSEFOR0VTX0ZBSUxVUkVfUEhFTk9UWVBFIiwiT1JERVJf"
    "SU5URVJWRU5USU9OX0VGRkVDVF9BTUJJR1VPVVMiLCJESUFHTk9TVElDX0lOVkFMSUQiXSwi"
    "ZXhwZXJpbWVudF9pZCI6IkM0X1BBUkFHUkFQSF9PUkRFUl9CRUhBVklPUkFMX0RJRkZFUkVO"
    "VElBTF9WMSIsImZpbmFsX3VzZXJfYm91bmRhcnlfcGVyX2NvbmRpdGlvbiI6ODgwLCJmaXJz"
    "dF9wYXJhZ3JhcGhfcHJlc2VydmVkIjp0cnVlLCJmcmVzaF93b3JrZXJfcHJvY2Vzc19wZXJf"
    "b2JzZXJ2YXRpb24iOnRydWUsImhpc3RvcmljYWxfY29udHJvbF9wYXJzZWRfb2JqZWN0X3No"
    "YTI1NiI6ImZiOGNiZmRlMGZmZWZmNDhjNDc3M2NlZTk1YzU3NmY4MjFiMjJmODRiMDBkYzEw"
    "NTk0MTA4NTY1MDIyNTZhYmEiLCJpbnZhbGlkX2pzb25fcmV0YWluZWRfYXNfb2JzZXJ2YXRp"
    "b24iOnRydWUsImxhc3RfcGFyYWdyYXBoX3ByZXNlcnZlZCI6dHJ1ZSwibWF4aW11bV9vdXRw"
    "dXRfdG9rZW5zIjozMiwibWVzc2FnZV9ib3VuZGFyeV9wcm9maWxlX3Blcl9jb25kaXRpb24i"
    "OlszLDI4LDg2OSw4ODBdLCJtZXNzYWdlX3JvbGVzIjpbInN5c3RlbSIsInVzZXIiLCJhc3Np"
    "c3RhbnQiLCJ1c2VyIl0sIm1peGVkX3Jlc3VsdF9wZXJtaXRzX3BhcmFncmFwaF9vcmRlcl9j"
    "bGFpbSI6ZmFsc2UsIm5vcnRoX3N0YXJfYWJjX2VmZmVjdF9jbGFpbV9hdXRob3JpemVkIjpm"
    "YWxzZSwib2JzZXJ2YXRpb25zX3Blcl9jb25kaXRpb24iOjMsIm91dHB1dF9tb2RlIjoiVU5D"
    "T05TVFJBSU5FRCIsInA1X3A2X3JlcXVhbGlmaWNhdGlvbl9hdXRob3JpemVkIjpmYWxzZSwi"
    "cGFyYWdyYXBoX2NvbnRlbnRfbXVsdGlzZXRfcHJlc2VydmVkIjp0cnVlLCJwYXJhZ3JhcGhf"
    "Y291bnRfcGVyX2NvbmRpdGlvbiI6MTAsInBhcmFncmFwaF9vcmRlcl9yb290X2NhdXNlX2Ns"
    "YWltX3Blcm1pdHRlZCI6ZmFsc2UsInBvc3RfaG9jX3R3b19vZl90aHJlZV9pbnRlcnByZXRh"
    "dGlvbl9wZXJtaXR0ZWQiOmZhbHNlLCJwcmVfcmVxdWVzdF9pZGVudGl0eV9wZXJzaXN0ZWRf"
    "YmVmb3JlX21vZGVsX3JlcXVlc3RfYnVkZ2V0Ijp0cnVlLCJwcmVfcmVxdWVzdF90b2tlbl9p"
    "ZGVudGl0eV9qb3VybmFsX3JlcXVpcmVkIjp0cnVlLCJwcmlvcl9yZXF1ZXN0X2NhY2hlX2Nh"
    "cnJ5b3Zlcl9wZXJtaXR0ZWQiOmZhbHNlLCJwcm9tcHRfdG9rZW5fY291bnRfcGVyX2NvbmRp"
    "dGlvbiI6ODk5LCJyZXBldGl0aW9uX3BlbmFsdHkiOjEuMSwicmVxdWVzdF9vcmRlciI6WyJD"
    "T05UUk9MX09SSUdJTkFMX0M0IiwiVFJFQVRNRU5UX1JFVkVSU0VEX01JRERMRV9FSUdIVCIs"
    "IlRSRUFUTUVOVF9SRVZFUlNFRF9NSURETEVfRUlHSFQiLCJDT05UUk9MX09SSUdJTkFMX0M0"
    "IiwiQ09OVFJPTF9PUklHSU5BTF9DNCIsIlRSRUFUTUVOVF9SRVZFUlNFRF9NSURETEVfRUlH"
    "SFQiXSwicmVzcG9uc2VfZm9ybWF0X3ByZXNlbnQiOmZhbHNlLCJydW50aW1lX3JlbWVkaWF0"
    "aW9uX2F1dGhvcml6ZWQiOmZhbHNlLCJzZWVkIjo3LCJzdHJlYW0iOmZhbHNlLCJ0ZWFyZG93"
    "bl9yZXF1aXJlZF9iZXR3ZWVuX29ic2VydmF0aW9ucyI6dHJ1ZSwidGVtcGVyYXR1cmUiOjAs"
    "InRocmVzaG9sZF9zZWFyY2hfYXV0aG9yaXplZCI6ZmFsc2UsInRva2VuX2lkX211bHRpc2V0"
    "X2lkZW50aWNhbCI6dHJ1ZSwidG9wX3AiOjEsInRyZWF0bWVudF9jb25kaXRpb25faWQiOiJU"
    "UkVBVE1FTlRfUkVWRVJTRURfTUlERExFX0VJR0hUIiwidHJlYXRtZW50X3BhcmFncmFwaF9v"
    "cmRlciI6WzEsOSw4LDcsNiw1LDQsMywyLDEwXSwidHJlYXRtZW50X3BheWxvYWRfc2hhMjU2"
    "IjoiNDdjNTE5YzI0ZWZkNDBlM2JhYjRiZmEyZWFlYzFjZjNkNjJjOTFhNjQ4ODcwZTYzMTcy"
    "MTYyNTU2N2YyMGI1ZSIsInRyZWF0bWVudF90aHJlZV9vZl90aHJlZV9leGFjdF9tZWFuc19y"
    "ZXN0b3JhdGlvbiI6dHJ1ZSwidHJlYXRtZW50X3Rva2VuX3NoYTI1NiI6IjE0ZDZhNjg1NmZm"
    "YjVjNGNhYTRhNGVkMjI5ZmEwYzk0YWMwNmI4NmZiZWY0NzNiZTAwMWRkNmQ4ZTM2OThjY2Ui"
    "LCJ2YXJpYWJsZV91bmRlcl90ZXN0IjoiR0xPQkFMX1BBUkFHUkFQSF9PUkRFUl9XSVRIX1RP"
    "S0VOX0lOVkVOVE9SWV9BTkRfTE9DQUxfVEFJTF9MT0NLRUQiLCJ6ZXJvX2NhY2hlZF9wcmVm"
    "aXhfYmFzZWxpbmVfcmVxdWlyZWQiOnRydWV9LCJleHBlcmltZW50X2NvbnRyYWN0X3NoYTI1"
    "NiI6ImRkNjdiZTMwZmE5N2MyODUyNDQ4ZGFiZGViYjE0NjhiMzMzZGY1NWZmZTAyOWFkMjE4"
    "MDBmNmQwOTlmZGE2MjMiLCJleHBpcmVzX2F0IjoiMjAyNi0wOC0yMVQxMzoxNDoxMS4xNjI3"
    "NDFaIiwiZ2VuZXJhdG9yX2NvbnRyYWN0X3NoYTI1NiI6IjdhNjI0NDVlNDhhNTFmMzJjNDlh"
    "YTc2ZDM3ODEyN2E4MzM5YTkxZTIyZTBjMjJlNGRlMDE1MGY3NTcxYjlmMjMiLCJpbXBsZW1l"
    "bnRhdGlvbl9tZXJnZV9jb21taXQiOiI3ZTAzNzU5NmRlMWE3NDAzODU4M2E4NWVkODFkNDZl"
    "YzEyZGViYmFjIiwiaW1wbGVtZW50YXRpb25fcmVjb3JkX3NoYTI1NiI6ImM1NjNiZjAxMmM3"
    "ZWM1ODcwODliN2IyOGFmNTA3NDIwN2EzODljNWZiNzM4MWI5YzEyMTMyOTlkM2I0ODkzODYi"
    "LCJpbXBsZW1lbnRhdGlvbl9yZXZpZXdfc2hhMjU2IjoiMzU1YTZiN2Y3ODcxZTY0OGQ4YmZh"
    "ZjRjNzg0MWU5ZTYzNDZmOWI1OWViYTY1YWM5OGMwMGI1NWQ5NDBkMjU5NSIsImlzc3VlZF9h"
    "dCI6IjIwMjYtMDgtMjFUMTA6MTQ6MTEuMTYyNzQxWiIsImlzc3Vlcl9tZXJnZV9jb21taXQi"
    "OiJhYWE4Njg4OGRmMmI4YTljM2QwMTY4MGUxYTFlNGI4ZWJiYTFlY2M3IiwiaXNzdWVyX3Nv"
    "dXJjZV9zaGEyNTYiOiJlY2FjY2M2NzQ0NzhjZjViOGQzNzliNDU2YjQ2MmI2NDMxZTA3MWYz"
    "NTY3Y2ZiNWU5OGMyNDVjYTFmZDNmMDMwIiwibGlmZWN5Y2xlIjoiSVNTVUVEIiwibm9ydGhf"
    "c3Rhcl9hYmNfZWZmZWN0X2NsYWltX2F1dGhvcml6ZWQiOmZhbHNlLCJvcGVyYXRvcl9jb25m"
    "aXJtYXRpb25fbWV0aG9kIjoiUkVUWVBFX0RZTkFNSUNfU0hBMjU2X0NIQUxMRU5HRSIsIm9w"
    "ZXJhdG9yX2NvbmZpcm1hdGlvbl9yZWNvcmRlZCI6dHJ1ZSwib3BlcmF0b3JfY29uZmlybWVk"
    "X2F0IjoiMjAyNi0wOC0yMVQxMDoxNDoxMS4xNjI3NDFaIiwicDVfcDZfcmVxdWFsaWZpY2F0"
    "aW9uX2F1dGhvcml6ZWQiOmZhbHNlLCJwYXJhZ3JhcGhfb3JkZXJfcm9vdF9jYXVzZV9jbGFp"
    "bV9hdXRob3JpemVkIjpmYWxzZSwicGxhdGZvcm1fY29udHJhY3Rfc2hhMjU2IjoiNzU2MDQw"
    "OWU2NmY5Zjc1ZTk1MjhjZWRmODZmMjNmNjAzZjc1YjgxYWMzZjAxNTJiNDkyOGQyYjQzNWI1"
    "ZWE2NiIsInBsYXRmb3JtX29ic2VydmF0aW9uX2NvbnRyb2xfaWQiOiJQRVJTSVNUX0RVUkFC"
    "TEVfT0JTRVJWQVRJT05fQkVGT1JFX1NBVkVfQU5EX1JVTl9BTEwiLCJyZXF1aXJlZF9wbGF0"
    "Zm9ybSI6eyJhY2NlbGVyYXRvciI6IlQ0X1gyIiwiYWxsb2NhdGVkX2dwdV9jb3VudCI6Miwi"
    "Y3JlZGVudGlhbHNfcGVybWl0dGVkIjpmYWxzZSwiY3VzdG9tZXJfZGF0YV9wZXJtaXR0ZWQi"
    "OmZhbHNlLCJleHRlcm5hbF9uZXR3b3JrX2FjY2Vzc19wZXJtaXR0ZWQiOmZhbHNlLCJmcmVz"
    "aF9wb3N0X2FydGlmYWN0X29ic2VydmF0aW9uX3JlcXVpcmVkIjp0cnVlLCJpbnRlcm5ldF9l"
    "bmFibGVkIjpmYWxzZSwibWFjaGluZV9vYnNlcnZhYmxlX3J1bnRpbWVfdG9wb2xvZ3lfY2hl"
    "Y2tfcmVxdWlyZWQiOnRydWUsIm9ic2VydmF0aW9uX21vdW50ZWRfYXNfcnVudGltZV9pbnB1"
    "dCI6ZmFsc2UsIm9ic2VydmF0aW9uX3ByZWNlZGVzX3NhdmVfYW5kX3J1bl9hbGwiOnRydWUs"
    "InByZWlzc3VhbmNlX3BsYXRmb3JtX29ic2VydmF0aW9uX3JlcXVpcmVkIjpmYWxzZX0sInJ1"
    "bnRpbWUiOnsiYXR0ZW50aW9uX2JhY2tlbmQiOiJUUklUT05fQVRUTiIsImNhY2hlX2Jsb2Nr"
    "X3NpemUiOjE2LCJjdWRhX3ZhcmlhbnQiOiJjdTEyOSIsImdwdV90b3BvbG9neSI6IlQ0X1gy"
    "IiwibWF4X21vZGVsX2xlbiI6NDA5NiwibW9kZWxfZGlyZWN0b3J5X3NoYTI1NiI6Ijg0OTY5"
    "ZjZiZTJlZDhjNjY4NWUwNDAxMGYyN2I0M2ZkOTE3YzVkYzQzODczMDBjOTIyNDEwNGI1ZDNi"
    "MzFjOTQiLCJtb2RlbF9yZXBvc2l0b3J5IjoiUXdlbi9Rd2VuMi41LTAuNUItSW5zdHJ1Y3Qi"
    "LCJtb2RlbF9yZXZpc2lvbiI6IjdhZTU1NzYwNGFkZjY3YmU1MDQxN2Y1OWMyYzJmMTY3ZGVm"
    "OWE3NzUiLCJwcmVmaXhfY2FjaGluZ19lbmFibGVkIjp0cnVlLCJweXRob24iOiIzLjEyIiwi"
    "cmVxdWlyZWRfbmF0aXZlX21vZHVsZSI6InZsbG0uX0Nfc3RhYmxlX2xpYnRvcmNoIiwidG9r"
    "ZW5pemVyX3JldmlzaW9uIjoiN2FlNTU3NjA0YWRmNjdiZTUwNDE3ZjU5YzJjMmYxNjdkZWY5"
    "YTc3NSIsInRvcmNoIjoiMi4xMS4wK2N1MTI5IiwidG9yY2hfY3VkYV92ZXJzaW9uIjoiMTIu"
    "OSIsInRyYW5zZm9ybWVycyI6IjUuMTQuMSIsInRyaXRvbiI6IjMuNi4wIiwidmxsbV9kaXN0"
    "cmlidXRpb24iOiIwLjI1LjErY3UxMjkiLCJ2bGxtX3B1YmxpY19zZW1hbnRpY192ZXJzaW9u"
    "IjoiMC4yNS4xIiwid29ya2VyX2dwdV9pbmRleCI6MH0sInJ1bnRpbWVfYW50aV9yZXBsYXlf"
    "ZXN0YWJsaXNoZWQiOmZhbHNlLCJydW50aW1lX2NvbnRyYWN0X3NoYTI1NiI6IjQ3ZWM2NGE2"
    "Y2QwMjU2NDU0NDJiZjJiMjMzM2RhNTNjYzBmMzQ1NjdmMDg1ZjhjZTk5MzZhNWJjZDM1MjBi"
    "NmEiLCJydW50aW1lX2V4ZWN1dGlvbl9hdXRob3JpemVkIjp0cnVlLCJydW50aW1lX3BheWxv"
    "YWRfc2hhMjU2IjoiMWQwNTVkZmFiOWY4M2EyNzA2ZjUzMzViNDUyOWRmOThkNDVlNDVkZTUy"
    "MTBhOGM2YzIxYzJiOTFlNmE3MmRmMCIsInJ1bnRpbWVfcmVtZWRpYXRpb25fYXV0aG9yaXpl"
    "ZCI6ZmFsc2UsInNjaGVtYV92ZXJzaW9uIjoiMS4wLjAiLCJzY29wZSI6IkM0X1BBUkFHUkFQ"
    "SF9PUkRFUl9CRUhBVklPUkFMX0RJRkZFUkVOVElBTF9WMSIsInNpbmdsZV91c2UiOnRydWUs"
    "InRocmVzaG9sZF9zZWFyY2hfYXV0aG9yaXplZCI6ZmFsc2UsInVuY2hhbmdlZF9yZXBsYXlf"
    "YXV0aG9yaXplZCI6ZmFsc2V9LCJzY2hlbWFfdmVyc2lvbiI6IjEuMC4wIiwidHJhbnNhY3Rp"
    "b25faWQiOiIxOWZjMDNhNmI0Y2E3NGEwMjVmOGQ5YjhjZDIxYmU3ZTFjYjE0YTRlNzc2OTk1"
    "NzYwYzY3OWE1ODEzNDBmMTIyIn0="
)
_RUNTIME_PAYLOAD_B64 = (
    "ZnJvbSBfX2Z1dHVyZV9fIGltcG9ydCBhbm5vdGF0aW9ucwoKaW1wb3J0IGhhc2hsaWIKaW1w"
    "b3J0IGpzb24KaW1wb3J0IG9zCmltcG9ydCByZQppbXBvcnQgc2h1dGlsCmltcG9ydCBzb2Nr"
    "ZXQKaW1wb3J0IHN0YXQKaW1wb3J0IHN1YnByb2Nlc3MKaW1wb3J0IHN5cwppbXBvcnQgdGhy"
    "ZWFkaW5nCmltcG9ydCB0aW1lCmltcG9ydCB1cmxsaWIuZXJyb3IKaW1wb3J0IHVybGxpYi5w"
    "YXJzZQppbXBvcnQgdXJsbGliLnJlcXVlc3QKaW1wb3J0IHppcGZpbGUKZnJvbSBkYXRhY2xh"
    "c3NlcyBpbXBvcnQgYXNkaWN0LCBkYXRhY2xhc3MKZnJvbSBkYXRldGltZSBpbXBvcnQgVVRD"
    "LCBkYXRldGltZQpmcm9tIGVudW0gaW1wb3J0IFN0ckVudW0KZnJvbSBwYXRobGliIGltcG9y"
    "dCBQYXRoCmZyb20gdHlwaW5nIGltcG9ydCBGaW5hbAoKTk9URUJPT0tfTkFNRTogRmluYWwg"
    "PSAiYWctYzQtcGFyYWdyYXBoLW9yZGVyLWRpZmYtdjEiClNPVVJDRV9NQUlOX0NPTU1JVDog"
    "RmluYWwgPSAiNTQ0NWNhZGRmOGY1MzMxYmJmMGY5ZjVjYmIwNmZkNzY4MjMwMDY3ZiIKSU1Q"
    "TEVNRU5UQVRJT05fUkVWSUVXX1NIQTI1NjogRmluYWwgPSAoCiAgICAiNGRlNTJhNGI4NTJk"
    "NzZkYTVjODcxM2Y1OTBmNTU5OWEzZjBhZGIxODVhOTgxNjM1YjJmNGMxZGQ5MWYzNDdlNSIK"
    "KQpERVNJR05fUkVDT1JEX1NIQTI1NjogRmluYWwgPSAiYTQyNjgyYzFmYWEwNWIyZTFkMjIw"
    "YTBjMjM5YzIxNzE4MzQxMjcwZjcxNzQyMGFhZjExYmUyZTI0NDFlOTcyOCIKVjVfQUNDRVBU"
    "QU5DRV9TSEEyNTY6IEZpbmFsID0gImI4NjMxNGJkOGM5YTcxNzY2ODg0YWM3MTQzYjdmZmYz"
    "MTk4ZTk4NmRkOTljNjA2NTgxNGI0NWM4ZDEwOTVlYjEiCk1PREVMX1NOQVBTSE9UX1NIQTI1"
    "NjogRmluYWwgPSAiODQ5NjlmNmJlMmVkOGM2Njg1ZTA0MDEwZjI3YjQzZmQ5MTdjNWRjNDM4"
    "NzMwMGM5MjI0MTA0YjVkM2IzMWM5NCIKRVhQRUNURURfQkFDS0VORF9MT0dfTUFSS0VSOiBG"
    "aW5hbCA9ICJVc2luZyBBdHRlbnRpb25CYWNrZW5kRW51bS5UUklUT05fQVRUTiBiYWNrZW5k"
    "LiIKQkFDS0VORF9NQVJLRVJfTE9HX1BSRUZJWF9QQVRURVJOOiBGaW5hbCA9IHJlLmNvbXBp"
    "bGUoCiAgICByIl4oPzouKlxiSU5GT1xiLiopP1VzaW5nIEF0dGVudGlvbkJhY2tlbmRFbnVt"
    "XC5UUklUT05fQVRUTiBiYWNrZW5kXC4kIgopCkJBQ0tFTkRfTUFSS0VSX0VYQ0xVREVEX1RP"
    "S0VOUzogRmluYWwgPSAoCiAgICAiLS1hdHRlbnRpb24tYmFja2VuZCIsCiAgICAiZXhwZWN0"
    "ZWRfYmFja2VuZF9sb2dfbWFya2VyIiwKICAgICJyZXR1cm4gIiwKICAgICJhc3NlcnQgIiwK"
    "ICAgICJwcmludCgiLAopCkdQVV9NRU1PUllfUkVUVVJOX1RPTEVSQU5DRV9NSUI6IEZpbmFs"
    "ID0gMTI4Ck1BWF9URUFSRE9XTl9QT0xMUzogRmluYWwgPSAyMApURUFSRE9XTl9QT0xMX1NF"
    "Q09ORFM6IEZpbmFsID0gMC41CgpJTlBVVF9ST09UID0gUGF0aCgiL2thZ2dsZS9pbnB1dCIp"
    "LnJlc29sdmUoKQpXT1JLX1JPT1QgPSBQYXRoKCIva2FnZ2xlL3dvcmtpbmciKS5yZXNvbHZl"
    "KCkKT1VUUFVUX1JPT1Q6IEZpbmFsID0gV09SS19ST09UIC8gImM0X3BhcmFncmFwaF9vcmRl"
    "cl9iZWhhdmlvcmFsX2RpZmZlcmVudGlhbF92MSIKU0NSQVRDSF9ST09UOiBGaW5hbCA9IFdP"
    "UktfUk9PVCAvICJjNF9wYXJhZ3JhcGhfb3JkZXJfYmVoYXZpb3JhbF9kaWZmZXJlbnRpYWxf"
    "djFfc2NyYXRjaCIKVEFSR0VUX1JPT1QgPSBTQ1JBVENIX1JPT1QgLyAidGFyZ2V0X3J1bnRp"
    "bWUiClRBUkdFVF9TSVRFID0gVEFSR0VUX1JPT1QgLyAibGliIiAvICJweXRob24zLjEyIiAv"
    "ICJzaXRlLXBhY2thZ2VzIgpUQVJHRVRfUFlUSE9OID0gVEFSR0VUX1JPT1QgLyAiYmluIiAv"
    "ICJweXRob24iCkxPR19ST09UID0gT1VUUFVUX1JPT1QgLyAid29ya2VyX2xvZ3MiCkVWSURF"
    "TkNFX1pJUDogRmluYWwgPSBXT1JLX1JPT1QgLyAiYWctYzQtcGFyYWdyYXBoLW9yZGVyLWRp"
    "ZmYtZXZpZGVuY2UtdjEuemlwIgpQUkVfUkVRVUVTVF9UT0tFTl9JREVOVElUWV9KT1VSTkFM"
    "ID0gT1VUUFVUX1JPT1QgLyAicHJlX3JlcXVlc3RfdG9rZW5faWRlbnRpdHlfam91cm5hbF92"
    "MS5qc29uIgoKUlVOVElNRV9PVVRQVVRfRElSRUNUT1JZID0gImF1cmFnYXRld2F5X3ByZWZs"
    "aWdodF92M19leGFjdF9ydW50aW1lX3doZWVsaG91c2VfdjEiCk1PREVMX1JFVklTSU9OID0g"
    "IjdhZTU1NzYwNGFkZjY3YmU1MDQxN2Y1OWMyYzJmMTY3ZGVmOWE3NzUiCk1PREVMX1JFUE9T"
    "SVRPUlkgPSAiUXdlbi9Rd2VuMi41LTAuNUItSW5zdHJ1Y3QiClNFUlZFRF9NT0RFTF9OQU1F"
    "ID0gImxvY2FsLXF3ZW4yLjUtMC41Yi1pbnN0cnVjdCIKRVhQRUNURURfVkxMTV9ESVNUUklC"
    "VVRJT05fVkVSU0lPTiA9ICIwLjI1LjErY3UxMjkiCkVYUEVDVEVEX1ZMTE1fTU9EVUxFX1ZF"
    "UlNJT04gPSAiMC4yNS4xIgpFWFBFQ1RFRF9UT1JDSF9WRVJTSU9OID0gIjIuMTEuMCtjdTEy"
    "OSIKRVhQRUNURURfVE9SQ0hfQ1VEQSA9ICIxMi45IgpFWFBFQ1RFRF9UUklUT05fVkVSU0lP"
    "TiA9ICIzLjYuMCIKRVhQRUNURURfVFJBTlNGT1JNRVJTX1ZFUlNJT04gPSAiNS4xNC4xIgpS"
    "RVFVSVJFRF9OQVRJVkVfTU9EVUxFID0gInZsbG0uX0Nfc3RhYmxlX2xpYnRvcmNoIgpFWFBF"
    "Q1RFRF9HUFVfTkFNRSA9ICJUZXNsYSBUNCIKRVhQRUNURURfQ09NUFVURV9DQVBBQklMSVRZ"
    "ID0gIjcuNSIKRVhQRUNURURfQkFDS0VORCA9ICJUUklUT05fQVRUTiIKRVhQRUNURURfQkFD"
    "S0VORF9DTEFTUyA9ICJ2bGxtLnYxLmF0dGVudGlvbi5iYWNrZW5kcy50cml0b25fYXR0bi5U"
    "cml0b25BdHRlbnRpb25CYWNrZW5kIgpSRUFMX0RSSVZFUl9ESVJFQ1RPUlkgPSAiL3Vzci9s"
    "b2NhbC9udmlkaWEvbGliNjQiCkVYUEVDVEVEX0NISUxEX1BZVEhPTiA9IFRBUkdFVF9QWVRI"
    "T04KSU1QT1JUX0NMT1NVUkVfTU9EVUxFUyA9ICgKICAgICJ2bGxtIiwKICAgICJ0b3JjaCIs"
    "CiAgICAidHJpdG9uIiwKICAgICJ0cmFuc2Zvcm1lcnMiLAogICAgInZsbG0ubW9kZWxfZXhl"
    "Y3V0b3IubW9kZWxzLnJlZ2lzdHJ5IiwKKQoKTUFYX0hFQUxUSF9QT0xMUyA9IDkwCkhFQUxU"
    "SF9QT0xMX1NFQ09ORFMgPSAyLjAKSFRUUF9USU1FT1VUX1NFQ09ORFMgPSAxMjAuMApNQVhf"
    "U1RSRUFNX0JZVEVTID0gMTMxMDcyCk1BWF9JTlNUQUxMX0VYQ0VSUFRfQ0hBUkFDVEVSUyA9"
    "IDE2MDAwCk1BWF9FVklERU5DRV9aSVBfQllURVMgPSAyICogMTAyNCoqMgpNQVhfTU9ERUxf"
    "QllURVMgPSAxNiAqIDEwMjQqKjMKCkVYUEVDVEVEX0NPTlRST0xfSEFTSEVTID0gewogICAg"
    "InJlc29sdXRpb25fbG9jay5qc29uIjogKCIxMjk0Mzk0YWM0NzYzMzZiMTAzYjAzNmQ4NjU0"
    "YTQ5ZTRhZTc4YzI1YzkxMmNhNTcyOWNkOTRmOTgyMzg0ZjNjIiksCiAgICAicmVxdWlyZW1l"
    "bnRzLmxvY2sudHh0IjogKCJjZjVkNzczZWY1YzI2ZjJlNDJhN2FmZDc2ZjBlNDY2YzIxODQ3"
    "MTY5OTg2ZjE0ZmU1YTdhYzlhZDAyZjBhM2MzIiksCiAgICAibWF0ZXJpYWxpemF0aW9uLmxv"
    "Y2sudHh0IjogKAogICAgICAgICI3NzQ0NjE1MDg3OTRkODA0MjQ0YjJmMGRiZmYwNWU1MmZk"
    "Y2NjOGVmYmUxOWFmOGNmYjhkMGZhZWRjYjI1MzM5IgogICAgKSwKICAgICJydW50aW1lX21h"
    "bmlmZXN0Lmpzb24iOiAoImNiOWM2MjMyMWVhMTY1MWRlYWMyNjAxMjZkYjc1YzM5NTI1ZTRi"
    "YTcxMWVlMzcwOGZlNWY3YTViNTBmZmQ2ZWQiKSwKICAgICJzaGEyNTZfbWFuaWZlc3QuanNv"
    "biI6ICgiMDBkYmRhNGZkNzM0Y2Y5NGI2ZjVkZmRlMjYxOWY4M2VkNmE0ZGI3NzYxYTRjM2M1"
    "YWNlNmIwZjFlYmU2M2IwOCIpLAogICAgIm1hdGVyaWFsaXphdGlvbl9yZWNlaXB0Lmpzb24i"
    "OiAoCiAgICAgICAgIjU1YmM4ZDA3OGFmOTk2MGQ1ZjZhNjBiZjdkOTYzODgyMGJlOWZkZGEw"
    "ZWU3Njc1NGE5NDYyZDQ2ZWIwNTNmZTAiCiAgICApLAp9CgpPVVRQVVRfTkFNRVMgPSAoCiAg"
    "ICAicnVudGltZV9zb3VyY2VfaWRlbnRpdHlfcmVwb3J0X3YxLmpzb24iLAogICAgInJ1bnRp"
    "bWVfaW5zdGFsbF9yZXBvcnRfdjEuanNvbiIsCiAgICAicnVudGltZV9lbnZpcm9ubWVudF9y"
    "ZXBvcnRfdjEuanNvbiIsCiAgICAicnVudGltZV9pbXBvcnRfY2xvc3VyZV9yZXBvcnRfdjEu"
    "anNvbiIsCiAgICAiYzFfbW9kZWxfY29uc3RydWN0aW9uX3JlcG9ydF92MS5qc29uIiwKICAg"
    "ICJjMl93b3JrZXJfc3RhcnR1cF9yZXBvcnRfdjEuanNvbiIsCiAgICAiYzNfc2luZ2xlX3Jl"
    "cXVlc3RfcmVwb3J0X3YxLmpzb24iLAogICAgImM0X291dHB1dF9jb250cmFjdF9yZXBvcnRf"
    "djEuanNvbiIsCiAgICAicHJlX3JlcXVlc3RfdG9rZW5faWRlbnRpdHlfam91cm5hbF92MS5q"
    "c29uIiwKICAgICJwNV9jYWNoZV9iZWhhdmlvcl9yZXBvcnRfdjEuanNvbiIsCiAgICAicDVf"
    "cG9zdF9yZXN0YXJ0X25hdGl2ZV9vcmlnaW5fcmVwb3J0X3YxLmpzb24iLAogICAgInA2X3N0"
    "YWdlX2NoZWNrcG9pbnRfcmVwb3J0X3YxLmpzb24iLAogICAgInA2X25hdGl2ZV9vcmlnaW5f"
    "cmVwb3J0X3YxLmpzb24iLAogICAgInA2X3dvcmtlcl9zdGF0ZV9pc29sYXRpb25fcmVwb3J0"
    "X3YxLmpzb24iLAogICAgIndvcmtlcl90ZWFyZG93bl9yZXBvcnRfdjEuanNvbiIsCiAgICAi"
    "c2NyYXRjaF9jbGVhbnVwX3JlcG9ydF92MS5qc29uIiwKICAgICJwNV9wNl9leGFjdF9ydW50"
    "aW1lX3JlcXVhbGlmaWNhdGlvbl9zdW1tYXJ5X3YxLmpzb24iLAogICAgImZhaWx1cmVfcmVw"
    "b3J0X3YxLmpzb24iLAogICAgImJ1bmRsZV9tYW5pZmVzdF92MS5qc29uIiwKICAgICJodW1h"
    "bl9yZXBvcnRfdjEubWQiLAopCgpDUkVERU5USUFMX0VOVl9OQU1FUyA9ICgKICAgICJBTlRI"
    "Uk9QSUNfQVBJX0tFWSIsCiAgICAiQVdTX0FDQ0VTU19LRVlfSUQiLAogICAgIkFXU19TRUNS"
    "RVRfQUNDRVNTX0tFWSIsCiAgICAiQVpVUkVfT1BFTkFJX0FQSV9LRVkiLAogICAgIkdPT0dM"
    "RV9BUElfS0VZIiwKICAgICJIRl9UT0tFTiIsCiAgICAiSFVHR0lOR19GQUNFX0hVQl9UT0tF"
    "TiIsCiAgICAiT1BFTkFJX0FQSV9LRVkiLAogICAgIk9QRU5ST1VURVJfQVBJX0tFWSIsCikK"
    "ClNZU1RFTV9QUk9NUFQgPSAoCiAgICAiUmV0dXJuIG9ubHkgdGhlIGV4YWN0IEpTT04gb2Jq"
    "ZWN0IHN1cHBsaWVkIGluIHRoZSBmaW5hbCB1c2VyIG1lc3NhZ2UsICIKICAgICJ3aXRoIG5v"
    "IG1hcmtkb3duIG9yIGFkZGl0aW9uYWwgdGV4dC4iCikKU1lOVEhFVElDX0NBQ0hFX0NPTlRF"
    "WFRfQSA9ICgKICAgICJBdXJhR2F0ZXdheSBkZXRlcm1pbmlzdGljIHByZWZpeC1jYWNoZSBy"
    "ZWxpYWJpbGl0eSBwcm9iZSB2YXJpYW50IEEuICIKICAgICJUaGlzIHRleHQgaXMgc3ludGhl"
    "dGljLCBjb250YWlucyBubyBjdXN0b21lciBkYXRhLCBhbmQgbXVzdCByZW1haW4gIgogICAg"
    "ImJ5dGUtaWRlbnRpY2FsIHdoZW5ldmVyIHByZWZpeCB2YXJpYW50IEEgaXMgcmVxdWVzdGVk"
    "LiAiCikgKiAyNCArICgKICAgICJSZXR1cm4gb25seSB0aGUgZXhhY3QgSlNPTiBvYmplY3Qg"
    "c3VwcGxpZWQgaW4gdGhlIGZpbmFsIHVzZXIgbWVzc2FnZSwgIgogICAgIndpdGggbm8gbWFy"
    "a2Rvd24gb3IgYWRkaXRpb25hbCB0ZXh0LiIKKQpTWU5USEVUSUNfQ0FDSEVfQ09OVEVYVF9C"
    "ID0gKAogICAgIkF1cmFHYXRld2F5IGRlbGliZXJhdGVseSBjaGFuZ2VkIHByZWZpeC1jYWNo"
    "ZSBjb250cm9sIHZhcmlhbnQgQi4gIgogICAgIlRoaXMgdGV4dCBpcyBzeW50aGV0aWMsIGNv"
    "bnRhaW5zIG5vIGN1c3RvbWVyIGRhdGEsIGFuZCBpbnRlbnRpb25hbGx5ICIKICAgICJkaXZl"
    "cmdlcyBmcm9tIHZhcmlhbnQgQSBiZWZvcmUgdGhlIHJlcGVhdGVkIGNvbnRyb2wgYm9keS4g"
    "IgopICogMjQgKyAoCiAgICAiUmV0dXJuIG9ubHkgdGhlIGV4YWN0IEpTT04gb2JqZWN0IHN1"
    "cHBsaWVkIGluIHRoZSBmaW5hbCB1c2VyIG1lc3NhZ2UsICIKICAgICJ3aXRoIG5vIG1hcmtk"
    "b3duIG9yIGFkZGl0aW9uYWwgdGV4dC4iCikKU1lOVEhFVElDX0FTU0lTVEFOVF9BQ0sgPSAi"
    "U3ludGhldGljIGRldGVybWluaXN0aWMgY29udGV4dCBhY2tub3dsZWRnZWQuIgpFWFBFQ1RF"
    "RF9PQkpFQ1QgPSB7InByb2JlIjogImV4YWN0LXJ1bnRpbWUtcDUtcDYiLCAidmFsdWUiOiAx"
    "fQpFWFBFQ1RFRF9PQkpFQ1RfQ0FOT05JQ0FMID0ganNvbi5kdW1wcygKICAgIEVYUEVDVEVE"
    "X09CSkVDVCwKICAgIGVuc3VyZV9hc2NpaT1UcnVlLAogICAgc2VwYXJhdG9ycz0oIiwiLCAi"
    "OiIpLAogICAgc29ydF9rZXlzPVRydWUsCikKQ0FDSEVfQkxPQ0tfU0laRSA9IDE2CgpBQ1RJ"
    "T05fQlVER0VUX0xJTUlUUzogRmluYWwgPSB7CiAgICAicnVudGltZV9pbnN0YWxsX2F0dGVt"
    "cHRzIjogMSwKICAgICJydW50aW1lX2ltcG9ydF9jbG9zdXJlX3Byb2JlcyI6IDEsCiAgICAi"
    "bW9kZWxfbG9hZHMiOiA2LAogICAgIndvcmtlcl9zdGFydHMiOiA2LAogICAgIm1vZGVsX3Jl"
    "cXVlc3RzIjogNiwKfQoKRkFJTFVSRV9DT0RFUyA9IHsKICAgICJQM19QNl9SVU5USU1FX1NP"
    "VVJDRV9JREVOVElUWV9NSVNNQVRDSCIsCiAgICAiUDNfUDZfUlVOVElNRV9QUk9DRVNTX1RS"
    "RUVfSU1QT1JUX0NMT1NVUkVfRkFJTEVEIiwKICAgICJQM19QNl9OQVRJVkVfT1JJR0lOX0NM"
    "T1NVUkVfRkFJTEVEIiwKICAgICJQM19QNl9QTEFURk9STV9JREVOVElUWV9NSVNNQVRDSCIs"
    "CiAgICAiUDNfUDZfV0hFRUxIT1VTRV9JTlZBTElEIiwKICAgICJQM19QNl9SVU5USU1FX0lO"
    "U1RBTExfRkFJTEVEIiwKICAgICJQM19QNl9SVU5USU1FX0lOU1RBTExfTk9OWkVST19FWElU"
    "IiwKICAgICJQM19QNl9SVU5USU1FX0lOU1RBTExfVElNRU9VVCIsCiAgICAiUDNfUDZfUlVO"
    "VElNRV9JTlNUQUxMX0xBVU5DSF9GQUlMRUQiLAogICAgIlAzX1A2X01PREVMX0lERU5USVRZ"
    "X01JU01BVENIIiwKICAgICJQM19QNl9FWFBMSUNJVF9CQUNLRU5EX05PVF9SRUFMSVpFRCIs"
    "CiAgICAiUDNfUDZfV09SS0VSX1NUQVJUVVBfRkFJTEVEIiwKICAgICJQM19QNl9NT0RFTF9J"
    "TlZFTlRPUllfTUlTTUFUQ0giLAogICAgIlAzX1A2X1JFUVVFU1RfRkFJTEVEIiwKICAgICJQ"
    "M19QNl9NRVRSSUNfU0VNQU5USUNfVU5BVkFJTEFCTEUiLAogICAgIlAzX1A2X0NBQ0hFX1JF"
    "VVNFX05PVF9PQlNFUlZFRCIsCiAgICAiUDNfUDZfUkVTRVRfTk9UX1BST1ZFTiIsCiAgICAi"
    "UDZfV09SS0VSXzJfU1RBUlRVUF9GQUlMRUQiLAogICAgIlA2X1BST0NFU1NfSVNPTEFUSU9O"
    "X0ZBSUxFRCIsCiAgICAiUDZfR1BVX0lTT0xBVElPTl9GQUlMRUQiLAogICAgIlA2X1BPUlRf"
    "SVNPTEFUSU9OX0ZBSUxFRCIsCiAgICAiUDZfV09SS0VSXzFfUk9VVEVfVFJBTlNQT1JUX0ZB"
    "SUxFRCIsCiAgICAiUDZfV09SS0VSXzFfUkVTUE9OU0VfRU5WRUxPUEVfSU5WQUxJRCIsCiAg"
    "ICAiUDZfV09SS0VSXzFfTUVUUklDX0FUVFJJQlVUSU9OX0ZBSUxFRCIsCiAgICAiUDZfV09S"
    "S0VSXzJfUk9VVEVfVFJBTlNQT1JUX0ZBSUxFRCIsCiAgICAiUDZfV09SS0VSXzJfUkVTUE9O"
    "U0VfRU5WRUxPUEVfSU5WQUxJRCIsCiAgICAiUDZfV09SS0VSXzJfTUVUUklDX0FUVFJJQlVU"
    "SU9OX0ZBSUxFRCIsCiAgICAiUDZfUkVRVUVTVF9DT1VOVEVSX1JFQ09OQ0lMSUFUSU9OX0ZB"
    "SUxFRCIsCiAgICAiUDZfQ0hFQ0tQT0lOVF9TRVJJQUxJWkFUSU9OX0ZBSUxFRCIsCiAgICAi"
    "UDNfUDZfQUNUSU9OX0JVREdFVF9FWENFRURFRCIsCiAgICAiUDNfUDZfUFJJVkFDWV9CT1VO"
    "REFSWV9WSU9MQVRJT04iLAogICAgIlAzX1A2X1dPUktFUl9URUFSRE9XTl9GQUlMRUQiLAog"
    "ICAgIlAzX1A2X1NDUkFUQ0hfQ0xFQU5VUF9GQUlMRUQiLAp9CgpGQUlMVVJFX1RBWE9OT01Z"
    "ID0gewogICAgIklOUFVUX0lERU5USVRZX0ZBSUxVUkUiLAogICAgIk1PREVMX0FSVElGQUNU"
    "X0ZBSUxVUkUiLAogICAgIlRPS0VOSVpFUl9BUlRJRkFDVF9GQUlMVVJFIiwKICAgICJNT0RF"
    "TF9DT05TVFJVQ1RJT05fRkFJTFVSRSIsCiAgICAiV09SS0VSX1NUQVJUVVBfRkFJTFVSRSIs"
    "CiAgICAiV09SS0VSX0lERU5USVRZX0ZBSUxVUkUiLAogICAgIkRFVklDRV9SRUFMSVpBVElP"
    "Tl9GQUlMVVJFIiwKICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJFIiwKICAgICJPVVRQ"
    "VVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAiUDVfQ0FDSEVfRU5BQkxFTUVOVF9GQUlMVVJF"
    "IiwKICAgICJQNV9TVEFSVElOR19TVEFURV9GQUlMVVJFIiwKICAgICJQNV9DQUNIRV9PQlNF"
    "UlZBVElPTl9GQUlMVVJFIiwKICAgICJQNV9CRUhBVklPUl9GQUlMVVJFIiwKICAgICJQNl9S"
    "T1VURV9SRUFMSVpBVElPTl9GQUlMVVJFIiwKICAgICJQNl9XT1JLRVJfR0VORVJBVElPTl9G"
    "QUlMVVJFIiwKICAgICJQNl9TVEFURV9JU09MQVRJT05fRkFJTFVSRSIsCiAgICAiUDZfQkVI"
    "QVZJT1JfRkFJTFVSRSIsCiAgICAiTUVUUklDX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgIk1F"
    "VFJJQ19BVFRSSUJVVElPTl9BTUJJR1VPVVMiLAogICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJ"
    "T05fRkFJTFVSRSIsCiAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAiSEFSTkVTU19TRU1B"
    "TlRJQ19GQUlMVVJFIiwKICAgICJFVklERU5DRV9QUk9KRUNUSU9OX0ZBSUxVUkUiLAogICAg"
    "IkFVVEhPUklUWV9GQUlMVVJFIiwKICAgICJOT05fREVURVJNSU5JU1RJQ19GQUlMVVJFIiwK"
    "ICAgICJESUFHTk9TVElDX0lOVkFMSUQiLAp9CgoKZGVmIGNsYXNzaWZ5X2ZhaWx1cmVfZGV0"
    "YWlsKGRldGFpbF9jb2RlOiBzdHIpIC0+IHN0cjoKICAgIGlmIGRldGFpbF9jb2RlIGluIEZB"
    "SUxVUkVfVEFYT05PTVk6CiAgICAgICAgcmV0dXJuIGRldGFpbF9jb2RlCiAgICBleHBsaWNp"
    "dCA9IHsKICAgICAgICAiTU9ERUxfQ09OU1RSVUNUSU9OX0ZBSUxVUkUiOiAiTU9ERUxfQ09O"
    "U1RSVUNUSU9OX0ZBSUxVUkUiLAogICAgICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJF"
    "IjogIlJFUVVFU1RfRVhFQ1VUSU9OX0ZBSUxVUkUiLAogICAgfQogICAgaWYgZGV0YWlsX2Nv"
    "ZGUgaW4gZXhwbGljaXQ6CiAgICAgICAgcmV0dXJuIGV4cGxpY2l0W2RldGFpbF9jb2RlXQog"
    "ICAgaWYgZGV0YWlsX2NvZGUgaW4gewogICAgICAgICJQM19QNl9SVU5USU1FX1NPVVJDRV9J"
    "REVOVElUWV9NSVNNQVRDSCIsCiAgICAgICAgIlAzX1A2X1dIRUVMSE9VU0VfSU5WQUxJRCIs"
    "CiAgICAgICAgIlAzX1A2X1BSSVZBQ1lfQk9VTkRBUllfVklPTEFUSU9OIiwKICAgIH06CiAg"
    "ICAgICAgcmV0dXJuICJBVVRIT1JJVFlfRkFJTFVSRSIKICAgIGlmIGRldGFpbF9jb2RlIGlu"
    "IHsKICAgICAgICAiUDNfUDZfTU9ERUxfSURFTlRJVFlfTUlTTUFUQ0giLAogICAgfToKICAg"
    "ICAgICByZXR1cm4gIk1PREVMX0FSVElGQUNUX0ZBSUxVUkUiCiAgICBpZiBkZXRhaWxfY29k"
    "ZSBpbiB7CiAgICAgICAgIlAzX1A2X1JVTlRJTUVfSU5TVEFMTF9GQUlMRUQiLAogICAgICAg"
    "ICJQM19QNl9SVU5USU1FX0lOU1RBTExfTk9OWkVST19FWElUIiwKICAgICAgICAiUDNfUDZf"
    "UlVOVElNRV9JTlNUQUxMX1RJTUVPVVQiLAogICAgICAgICJQM19QNl9SVU5USU1FX0lOU1RB"
    "TExfTEFVTkNIX0ZBSUxFRCIsCiAgICAgICAgIlAzX1A2X1JVTlRJTUVfUFJPQ0VTU19UUkVF"
    "X0lNUE9SVF9DTE9TVVJFX0ZBSUxFRCIsCiAgICAgICAgIlAzX1A2X05BVElWRV9PUklHSU5f"
    "Q0xPU1VSRV9GQUlMRUQiLAogICAgICAgICJQM19QNl9QTEFURk9STV9JREVOVElUWV9NSVNN"
    "QVRDSCIsCiAgICAgICAgIlAzX1A2X01PREVMX0lOVkVOVE9SWV9NSVNNQVRDSCIsCiAgICAg"
    "ICAgIlAzX1A2X0VYUExJQ0lUX0JBQ0tFTkRfTk9UX1JFQUxJWkVEIiwKICAgIH06CiAgICAg"
    "ICAgcmV0dXJuICJNT0RFTF9DT05TVFJVQ1RJT05fRkFJTFVSRSIKICAgIGlmIGRldGFpbF9j"
    "b2RlIGluIHsKICAgICAgICAiUDNfUDZfV09SS0VSX1NUQVJUVVBfRkFJTEVEIiwKICAgICAg"
    "ICAiUDZfV09SS0VSXzJfU1RBUlRVUF9GQUlMRUQiLAogICAgfToKICAgICAgICByZXR1cm4g"
    "IldPUktFUl9TVEFSVFVQX0ZBSUxVUkUiCiAgICBpZiBkZXRhaWxfY29kZSBpbiB7CiAgICAg"
    "ICAgIlAzX1A2X1JFUVVFU1RfRkFJTEVEIiwKICAgICAgICAiUDZfV09SS0VSXzFfUk9VVEVf"
    "VFJBTlNQT1JUX0ZBSUxFRCIsCiAgICAgICAgIlA2X1dPUktFUl8yX1JPVVRFX1RSQU5TUE9S"
    "VF9GQUlMRUQiLAogICAgfToKICAgICAgICByZXR1cm4gIlJFUVVFU1RfRVhFQ1VUSU9OX0ZB"
    "SUxVUkUiCiAgICBpZiBkZXRhaWxfY29kZSBpbiB7CiAgICAgICAgIlA2X1dPUktFUl8xX1JF"
    "U1BPTlNFX0VOVkVMT1BFX0lOVkFMSUQiLAogICAgICAgICJQNl9XT1JLRVJfMl9SRVNQT05T"
    "RV9FTlZFTE9QRV9JTlZBTElEIiwKICAgIH06CiAgICAgICAgcmV0dXJuICJPVVRQVVRfQ09O"
    "VFJBQ1RfRkFJTFVSRSIKICAgIGlmIGRldGFpbF9jb2RlID09ICJQM19QNl9NRVRSSUNfU0VN"
    "QU5USUNfVU5BVkFJTEFCTEUiOgogICAgICAgIHJldHVybiAiTUVUUklDX1NFTUFOVElDX0ZB"
    "SUxVUkUiCiAgICBpZiBkZXRhaWxfY29kZSA9PSAiUDNfUDZfQ0FDSEVfUkVVU0VfTk9UX09C"
    "U0VSVkVEIjoKICAgICAgICByZXR1cm4gIlA1X0JFSEFWSU9SX0ZBSUxVUkUiCiAgICBpZiBk"
    "ZXRhaWxfY29kZSA9PSAiUDNfUDZfUkVTRVRfTk9UX1BST1ZFTiI6CiAgICAgICAgcmV0dXJu"
    "ICJQNV9TVEFSVElOR19TVEFURV9GQUlMVVJFIgogICAgaWYgZGV0YWlsX2NvZGUgPT0gIlA2"
    "X1BST0NFU1NfSVNPTEFUSU9OX0ZBSUxFRCI6CiAgICAgICAgcmV0dXJuICJXT1JLRVJfSURF"
    "TlRJVFlfRkFJTFVSRSIKICAgIGlmIGRldGFpbF9jb2RlID09ICJQNl9HUFVfSVNPTEFUSU9O"
    "X0ZBSUxFRCI6CiAgICAgICAgcmV0dXJuICJERVZJQ0VfUkVBTElaQVRJT05fRkFJTFVSRSIK"
    "ICAgIGlmIGRldGFpbF9jb2RlID09ICJQNl9QT1JUX0lTT0xBVElPTl9GQUlMRUQiOgogICAg"
    "ICAgIHJldHVybiAiUDZfUk9VVEVfUkVBTElaQVRJT05fRkFJTFVSRSIKICAgIGlmIGRldGFp"
    "bF9jb2RlIGluIHsKICAgICAgICAiUDZfV09SS0VSXzFfTUVUUklDX0FUVFJJQlVUSU9OX0ZB"
    "SUxFRCIsCiAgICAgICAgIlA2X1dPUktFUl8yX01FVFJJQ19BVFRSSUJVVElPTl9GQUlMRUQi"
    "LAogICAgfToKICAgICAgICByZXR1cm4gIk1FVFJJQ19BVFRSSUJVVElPTl9BTUJJR1VPVVMi"
    "CiAgICBpZiBkZXRhaWxfY29kZSA9PSAiUDZfUkVRVUVTVF9DT1VOVEVSX1JFQ09OQ0lMSUFU"
    "SU9OX0ZBSUxFRCI6CiAgICAgICAgcmV0dXJuICJSRVFVRVNUX1JFQ09OQ0lMSUFUSU9OX0ZB"
    "SUxVUkUiCiAgICBpZiBkZXRhaWxfY29kZSA9PSAiUDZfQ0hFQ0tQT0lOVF9TRVJJQUxJWkFU"
    "SU9OX0ZBSUxFRCI6CiAgICAgICAgcmV0dXJuICJFVklERU5DRV9QUk9KRUNUSU9OX0ZBSUxV"
    "UkUiCiAgICBpZiBkZXRhaWxfY29kZSA9PSAiUDNfUDZfQUNUSU9OX0JVREdFVF9FWENFRURF"
    "RCI6CiAgICAgICAgcmV0dXJuICJBVVRIT1JJVFlfRkFJTFVSRSIKICAgIGlmIGRldGFpbF9j"
    "b2RlIGluIHsKICAgICAgICAiUDNfUDZfV09SS0VSX1RFQVJET1dOX0ZBSUxFRCIsCiAgICAg"
    "ICAgIlAzX1A2X1NDUkFUQ0hfQ0xFQU5VUF9GQUlMRUQiLAogICAgfToKICAgICAgICByZXR1"
    "cm4gIlRFQVJET1dOX0ZBSUxVUkUiCiAgICByZXR1cm4gIkRJQUdOT1NUSUNfSU5WQUxJRCIK"
    "CgpjbGFzcyBCZWhhdmlvclN0YXRlKFN0ckVudW0pOgogICAgUEFTUyA9ICJQQVNTIgogICAg"
    "RkFJTCA9ICJGQUlMIgogICAgQU1CSUdVT1VTID0gIkFNQklHVU9VUyIKCgpjbGFzcyBPYnNl"
    "cnZhdGlvblN0YXRlKFN0ckVudW0pOgogICAgTk9UX0VYUE9TRUQgPSAiTk9UX0VYUE9TRUQi"
    "CiAgICBOT1RfT0JTRVJWRUQgPSAiTk9UX09CU0VSVkVEIgogICAgWkVSTyA9ICJaRVJPIgog"
    "ICAgUE9TSVRJVkUgPSAiUE9TSVRJVkUiCiAgICBJTlZBTElEID0gIklOVkFMSUQiCiAgICBB"
    "TUJJR1VPVVMgPSAiQU1CSUdVT1VTIgoKCkBkYXRhY2xhc3MoZnJvemVuPVRydWUpCmNsYXNz"
    "IFJhd01ldHJpY1NhbXBsZToKICAgIG5hbWU6IHN0cgogICAgbGFiZWxzOiB0dXBsZVt0dXBs"
    "ZVtzdHIsIHN0cl0sIC4uLl0KICAgIHZhbHVlOiBmbG9hdAoKCkBkYXRhY2xhc3MoZnJvemVu"
    "PVRydWUpCmNsYXNzIE1ldHJpY1NuYXBzaG90T2JzZXJ2YXRpb246CiAgICBwcmVmaXhfY2Fj"
    "aGVfcXVlcmllczogZmxvYXQKICAgIHByZWZpeF9jYWNoZV9oaXRzOiBmbG9hdAogICAgbG9j"
    "YWxfY29tcHV0ZTogZmxvYXQKICAgIGxvY2FsX2NhY2hlX2hpdDogZmxvYXQKICAgIGV4dGVy"
    "bmFsX2t2X3RyYW5zZmVyOiBmbG9hdAogICAgY2FjaGVkX3Byb21wdF90b2tlbnM6IGZsb2F0"
    "CiAgICBuZXdseV9jb21wdXRlZF9wcmVmaWxsX3Rva2VuczogZmxvYXQKCiAgICBAcHJvcGVy"
    "dHkKICAgIGRlZiB0b3RhbF9wcm9tcHRfdG9rZW5zKHNlbGYpIC0+IGZsb2F0OgogICAgICAg"
    "IHJldHVybiBzZWxmLmxvY2FsX2NvbXB1dGUgKyBzZWxmLmxvY2FsX2NhY2hlX2hpdCArIHNl"
    "bGYuZXh0ZXJuYWxfa3ZfdHJhbnNmZXIKCgpAZGF0YWNsYXNzKGZyb3plbj1UcnVlKQpjbGFz"
    "cyBNZXRyaWNEZWx0YU9ic2VydmF0aW9uOgogICAgcHJlZml4X2NhY2hlX3F1ZXJpZXM6IGZs"
    "b2F0CiAgICBwcmVmaXhfY2FjaGVfaGl0czogZmxvYXQKICAgIGxvY2FsX2NvbXB1dGU6IGZs"
    "b2F0CiAgICBsb2NhbF9jYWNoZV9oaXQ6IGZsb2F0CiAgICBleHRlcm5hbF9rdl90cmFuc2Zl"
    "cjogZmxvYXQKICAgIGNhY2hlZF9wcm9tcHRfdG9rZW5zOiBmbG9hdAogICAgbmV3bHlfY29t"
    "cHV0ZWRfcHJlZmlsbF90b2tlbnM6IGZsb2F0CgogICAgQHByb3BlcnR5CiAgICBkZWYgdG90"
    "YWxfcHJvbXB0X3Rva2VucyhzZWxmKSAtPiBmbG9hdDoKICAgICAgICByZXR1cm4gc2VsZi5s"
    "b2NhbF9jb21wdXRlICsgc2VsZi5sb2NhbF9jYWNoZV9oaXQgKyBzZWxmLmV4dGVybmFsX2t2"
    "X3RyYW5zZmVyCgoKQGRhdGFjbGFzcyhmcm96ZW49VHJ1ZSkKY2xhc3MgVG9rZW5JZGVudGl0"
    "eU9ic2VydmF0aW9uOgogICAgcmVxdWVzdF9yb2xlOiBzdHIKICAgIHByZWZpeF92YXJpYW50"
    "OiBzdHIKICAgIHRva2VuX2NvdW50OiBpbnQKICAgIHRva2VuX3NoYTI1Njogc3RyCiAgICB0"
    "b2tlbl9pZHM6IHR1cGxlW2ludCwgLi4uXQoKCkBkYXRhY2xhc3MoZnJvemVuPVRydWUpCmNs"
    "YXNzIFJvdXRlT2JzZXJ2YXRpb246CiAgICByZXF1ZXN0X2lkOiBzdHIKICAgIHJlcXVlc3Rf"
    "cm9sZTogc3RyCiAgICBpbnRlbmRlZF93b3JrZXI6IHN0cgogICAgcmVhbGl6ZWRfd29ya2Vy"
    "OiBzdHIKICAgIHdvcmtlcl9nZW5lcmF0aW9uOiBpbnQKICAgIGVuZHBvaW50X3BvcnQ6IGlu"
    "dAogICAgbWV0cmljX2VuZHBvaW50X2lkZW50aXR5OiBzdHIKICAgIHJvdXRlX3JlYXNvbjog"
    "c3RyCiAgICBmYWxsYmFja19yZWFzb246IHN0ciB8IE5vbmUKICAgIG91dHB1dF9zaGEyNTY6"
    "IHN0cgoKCkBkYXRhY2xhc3MoZnJvemVuPVRydWUpCmNsYXNzIEJlaGF2aW9yRGVjaXNpb246"
    "CiAgICBjYXBhYmlsaXR5OiBzdHIKICAgIHN0YXRlOiBCZWhhdmlvclN0YXRlCiAgICBmYWls"
    "dXJlX2NsYXNzOiBzdHIgfCBOb25lCiAgICByZWFzb25zOiB0dXBsZVtzdHIsIC4uLl0KCgpA"
    "ZGF0YWNsYXNzKGZyb3plbj1UcnVlKQpjbGFzcyBFdmlkZW5jZVByb2plY3Rpb246CiAgICBk"
    "ZWNpc2lvbjogQmVoYXZpb3JEZWNpc2lvbgogICAgcHVibGljX3N1bW1hcnk6IGRpY3Rbc3Ry"
    "LCBvYmplY3RdCgoKZGVmIG9ic2VydmF0aW9uX3N0YXRlKHZhbHVlOiBmbG9hdCB8IE5vbmUp"
    "IC0+IE9ic2VydmF0aW9uU3RhdGU6CiAgICBpZiB2YWx1ZSBpcyBOb25lOgogICAgICAgIHJl"
    "dHVybiBPYnNlcnZhdGlvblN0YXRlLk5PVF9PQlNFUlZFRAogICAgaWYgdmFsdWUgPCAwOgog"
    "ICAgICAgIHJldHVybiBPYnNlcnZhdGlvblN0YXRlLklOVkFMSUQKICAgIGlmIHZhbHVlID09"
    "IDA6CiAgICAgICAgcmV0dXJuIE9ic2VydmF0aW9uU3RhdGUuWkVSTwogICAgcmV0dXJuIE9i"
    "c2VydmF0aW9uU3RhdGUuUE9TSVRJVkUKCgpGQUlMVVJFX0NPREVTIHw9IEZBSUxVUkVfVEFY"
    "T05PTVkKCgpjbGFzcyBEaWFnbm9zdGljRmFpbHVyZShSdW50aW1lRXJyb3IpOgogICAgZGVm"
    "IF9faW5pdF9fKHNlbGYsIGVycm9yX2NvZGU6IHN0ciwgc2FmZV9tZXNzYWdlOiBzdHIpIC0+"
    "IE5vbmU6CiAgICAgICAgaWYgZXJyb3JfY29kZSBub3QgaW4gRkFJTFVSRV9DT0RFUzoKICAg"
    "ICAgICAgICAgcmFpc2UgVmFsdWVFcnJvcigidW5zdXBwb3J0ZWQgZXhhY3QtcnVudGltZSBQ"
    "NS9QNiBmYWlsdXJlIGNvZGUiKQogICAgICAgIHN1cGVyKCkuX19pbml0X18oc2FmZV9tZXNz"
    "YWdlKQogICAgICAgIHNlbGYuZXJyb3JfY29kZSA9IGVycm9yX2NvZGUKICAgICAgICBzZWxm"
    "LnNhZmVfbWVzc2FnZSA9IHNhZmVfbWVzc2FnZQoKCmNsYXNzIERpYWdub3N0aWNBbWJpZ3Vp"
    "dHkoUnVudGltZUVycm9yKToKICAgIGRlZiBfX2luaXRfXyhzZWxmLCBmYWlsdXJlX2NsYXNz"
    "OiBzdHIsIHNhZmVfbWVzc2FnZTogc3RyKSAtPiBOb25lOgogICAgICAgIGlmIGZhaWx1cmVf"
    "Y2xhc3Mgbm90IGluIEZBSUxVUkVfVEFYT05PTVk6CiAgICAgICAgICAgIHJhaXNlIFZhbHVl"
    "RXJyb3IoInVuc3VwcG9ydGVkIGFtYmlndWl0eSBmYWlsdXJlIGNsYXNzIikKICAgICAgICBz"
    "dXBlcigpLl9faW5pdF9fKHNhZmVfbWVzc2FnZSkKICAgICAgICBzZWxmLmZhaWx1cmVfY2xh"
    "c3MgPSBmYWlsdXJlX2NsYXNzCiAgICAgICAgc2VsZi5zYWZlX21lc3NhZ2UgPSBzYWZlX21l"
    "c3NhZ2UKCgpkZWYgcmVxdWlyZV90cmFuc2FjdGlvbl9ib3VuZF9jb250ZXh0KCkgLT4gZGlj"
    "dFtzdHIsIG9iamVjdF06CiAgICB0cmFuc2FjdGlvbl9pZCA9IGdsb2JhbHMoKS5nZXQoIkFV"
    "UkFHQVRFV0FZX1RSQU5TQUNUSU9OX0lEIikKICAgIGlmIG5vdCBpc2luc3RhbmNlKHRyYW5z"
    "YWN0aW9uX2lkLCBzdHIpIG9yIHJlLmZ1bGxtYXRjaChyIlswLTlhLWZdezY0fSIsIHRyYW5z"
    "YWN0aW9uX2lkKSBpcyBOb25lOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAog"
    "ICAgICAgICAgICAiQVVUSE9SSVRZX0ZBSUxVUkUiLAogICAgICAgICAgICAidHJhbnNhY3Rp"
    "b24tYm91bmQgd3JhcHBlciBhZG1pc3Npb24gY29udGV4dCBpcyBtaXNzaW5nIG9yIGludmFs"
    "aWQiLAogICAgICAgICkKICAgIHJldHVybiB7CiAgICAgICAgInRyYW5zYWN0aW9uX2lkIjog"
    "dHJhbnNhY3Rpb25faWQsCiAgICAgICAgImF1dGhvcml6YXRpb25fdHJhbnNwb3J0IjogIkVN"
    "QkVEREVEX1dSQVBQRVJfQURNSVNTSU9OIiwKICAgICAgICAiYXV0aG9yaXphdGlvbl9zcGVj"
    "aWZpY19rYWdnbGVfaW5wdXRzIjogMCwKICAgICAgICAiYXV0aG9yaXphdGlvbl9wcm9kdWNl"
    "cl9ub3RlYm9va3MiOiAwLAogICAgICAgICJydW50aW1lX2V4ZWN1dGlvbl9hdXRob3JpemVk"
    "IjogVHJ1ZSwKICAgIH0KCgpkZWYgY29uc3VtZV9hY3Rpb25zKAogICAgY291bnRlcnM6IGRp"
    "Y3Rbc3RyLCBpbnRdLAogICAgKmFjdGlvbl9uYW1lczogc3RyLAopIC0+IE5vbmU6CiAgICBm"
    "b3IgbmFtZSBpbiBhY3Rpb25fbmFtZXM6CiAgICAgICAgbGltaXQgPSBBQ1RJT05fQlVER0VU"
    "X0xJTUlUUy5nZXQobmFtZSkKICAgICAgICBpZiBsaW1pdCBpcyBOb25lOgogICAgICAgICAg"
    "ICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgICAgICJQM19QNl9BQ1RJ"
    "T05fQlVER0VUX0VYQ0VFREVEIiwKICAgICAgICAgICAgICAgIGYidW5rbm93biBib3VuZGVk"
    "IGFjdGlvbjoge25hbWV9IiwKICAgICAgICAgICAgKQogICAgICAgIGlmIGNvdW50ZXJzW25h"
    "bWVdID49IGxpbWl0OgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgICAgICJQM19QNl9BQ1RJT05fQlVER0VUX0VYQ0VFREVEIiwKICAgICAgICAg"
    "ICAgICAgIGYiYWN0aW9uIGJ1ZGdldCBleGhhdXN0ZWQ6IHtuYW1lfSIsCiAgICAgICAgICAg"
    "ICkKICAgIGZvciBuYW1lIGluIGFjdGlvbl9uYW1lczoKICAgICAgICBjb3VudGVyc1tuYW1l"
    "XSArPSAxCgoKZGVmIGNhbm9uaWNhbF9qc29uKHBheWxvYWQ6IG9iamVjdCkgLT4gc3RyOgog"
    "ICAgcmV0dXJuIGpzb24uZHVtcHMoCiAgICAgICAgcGF5bG9hZCwKICAgICAgICBlbnN1cmVf"
    "YXNjaWk9VHJ1ZSwKICAgICAgICBzZXBhcmF0b3JzPSgiLCIsICI6IiksCiAgICAgICAgc29y"
    "dF9rZXlzPVRydWUsCiAgICApCgoKZGVmIHNoYTI1Nl9ieXRlcyhwYXlsb2FkOiBieXRlcykg"
    "LT4gc3RyOgogICAgcmV0dXJuIGhhc2hsaWIuc2hhMjU2KHBheWxvYWQpLmhleGRpZ2VzdCgp"
    "CgoKZGVmIHNoYTI1Nl90ZXh0KHBheWxvYWQ6IHN0cikgLT4gc3RyOgogICAgcmV0dXJuIHNo"
    "YTI1Nl9ieXRlcyhwYXlsb2FkLmVuY29kZSgidXRmLTgiKSkKCgpkZWYgZmlsZV9zaGEyNTYo"
    "cGF0aDogUGF0aCkgLT4gc3RyOgogICAgZGlnZXN0ID0gaGFzaGxpYi5zaGEyNTYoKQogICAg"
    "d2l0aCBwYXRoLm9wZW4oInJiIikgYXMgaGFuZGxlOgogICAgICAgIGZvciBjaHVuayBpbiBp"
    "dGVyKGxhbWJkYTogaGFuZGxlLnJlYWQoMTAyNCAqIDEwMjQpLCBiIiIpOgogICAgICAgICAg"
    "ICBkaWdlc3QudXBkYXRlKGNodW5rKQogICAgcmV0dXJuIGRpZ2VzdC5oZXhkaWdlc3QoKQoK"
    "CmRlZiB3cml0ZV9ieXRlc19hdG9taWMocGF0aDogUGF0aCwgcGF5bG9hZDogYnl0ZXMpIC0+"
    "IE5vbmU6CiAgICBwYXRoLnBhcmVudC5ta2RpcihwYXJlbnRzPVRydWUsIGV4aXN0X29rPVRy"
    "dWUpCiAgICB0ZW1wb3JhcnkgPSBwYXRoLndpdGhfbmFtZShmIntwYXRoLm5hbWV9Lntvcy5n"
    "ZXRwaWQoKX0udG1wIikKICAgIGlmIHRlbXBvcmFyeS5leGlzdHMoKToKICAgICAgICByYWlz"
    "ZSBSdW50aW1lRXJyb3IoInRlbXBvcmFyeSBldmlkZW5jZSBwYXRoIGFscmVhZHkgZXhpc3Rz"
    "IikKICAgIHRyeToKICAgICAgICB0ZW1wb3Jhcnkud3JpdGVfYnl0ZXMocGF5bG9hZCkKICAg"
    "ICAgICB0ZW1wb3JhcnkucmVwbGFjZShwYXRoKQogICAgZmluYWxseToKICAgICAgICBpZiB0"
    "ZW1wb3JhcnkuZXhpc3RzKCk6CiAgICAgICAgICAgIHRlbXBvcmFyeS51bmxpbmsoKQoKCmRl"
    "ZiB3cml0ZV9qc29uKHBhdGg6IFBhdGgsIHBheWxvYWQ6IG9iamVjdCkgLT4gTm9uZToKICAg"
    "IHdyaXRlX2J5dGVzX2F0b21pYyhwYXRoLCBjYW5vbmljYWxfanNvbihwYXlsb2FkKS5lbmNv"
    "ZGUoInV0Zi04IikpCgoKZGVmIHdyaXRlX3RleHQocGF0aDogUGF0aCwgcGF5bG9hZDogc3Ry"
    "KSAtPiBOb25lOgogICAgd3JpdGVfYnl0ZXNfYXRvbWljKHBhdGgsIHBheWxvYWQucmVwbGFj"
    "ZSgiXHJcbiIsICJcbiIpLmVuY29kZSgidXRmLTgiKSkKCgpkZWYgcnVudGltZV9zb3VyY2Vf"
    "aWRlbnRpdHkoKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHZhbHVlID0gZ2xvYmFscygp"
    "LmdldCgiRVhFQ1VURURfUlVOVElNRV9TQ1JJUFRfU0hBMjU2IikKICAgIGlmIG5vdCBpc2lu"
    "c3RhbmNlKHZhbHVlLCBzdHIpIG9yIHJlLmZ1bGxtYXRjaChyIlswLTlhLWZdezY0fSIsIHZh"
    "bHVlKSBpcyBOb25lOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiUDNfUDZfUlVOVElNRV9TT1VSQ0VfSURFTlRJVFlfTUlTTUFUQ0giLAogICAgICAg"
    "ICAgICAicnVudGltZSBzb3VyY2UgaWRlbnRpdHkgd3JhcHBlciBpcyBtaXNzaW5nIG9yIGlu"
    "dmFsaWQiLAogICAgICAgICkKICAgIHJldHVybiB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9u"
    "IjogIjEuMC4wIiwKICAgICAgICAicmVwb3J0X2lkIjogImF1cmFnYXRld2F5LXA1LXA2LWV4"
    "YWN0LXJ1bnRpbWUtcmVxdWFsaWZpY2F0aW9uLXJ1bnRpbWUtc291cmNlLWlkZW50aXR5LXYx"
    "IiwKICAgICAgICAic3RhdHVzIjogIlBBU1NFRCIsCiAgICAgICAgImRlY2lzaW9uIjogIkVY"
    "RUNVVEVEX1JVTlRJTUVfU0NSSVBUX0lERU5USVRZX1ZFUklGSUVEIiwKICAgICAgICAibm90"
    "ZWJvb2tfbmFtZSI6IE5PVEVCT09LX05BTUUsCiAgICAgICAgInNvdXJjZV9tYWluX2NvbW1p"
    "dCI6IFNPVVJDRV9NQUlOX0NPTU1JVCwKICAgICAgICAiZXhlY3V0ZWRfcnVudGltZV9zY3Jp"
    "cHRfc2hhMjU2IjogdmFsdWUsCiAgICAgICAgIndyYXBwZXJfaGFzaF92ZXJpZmljYXRpb25f"
    "cGFzc2VkIjogVHJ1ZSwKICAgIH0KCgpkZWYgd3JpdGVfcnVudGltZV9zb3VyY2VfaWRlbnRp"
    "dHlfcmVwb3J0KCkgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICByZXBvcnQgPSBydW50aW1l"
    "X3NvdXJjZV9pZGVudGl0eSgpCiAgICB3cml0ZV9qc29uKAogICAgICAgIE9VVFBVVF9ST09U"
    "IC8gInJ1bnRpbWVfc291cmNlX2lkZW50aXR5X3JlcG9ydF92MS5qc29uIiwKICAgICAgICBy"
    "ZXBvcnQsCiAgICApCiAgICByZXR1cm4gcmVwb3J0CgoKZGVmIGJvdW5kZWRfbG9vcGJhY2so"
    "dXJsOiBzdHIpIC0+IHN0cjoKICAgIHBhcnNlZCA9IHVybGxpYi5wYXJzZS51cmxzcGxpdCh1"
    "cmwpCiAgICBpZiBwYXJzZWQuc2NoZW1lICE9ICJodHRwIjoKICAgICAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoInJ1bnRpbWUgSFRUUCBtdXN0IHVzZSBwbGFpbiBsb29wYmFjayBIVFRQIikK"
    "ICAgIGlmIHBhcnNlZC5ob3N0bmFtZSBub3QgaW4geyIxMjcuMC4wLjEiLCAibG9jYWxob3N0"
    "In06CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJydW50aW1lIEhUVFAgaXMgcmVzdHJp"
    "Y3RlZCB0byBsb29wYmFjayIpCiAgICBpZiBwYXJzZWQudXNlcm5hbWUgaXMgbm90IE5vbmUg"
    "b3IgcGFyc2VkLnBhc3N3b3JkIGlzIG5vdCBOb25lOgogICAgICAgIHJhaXNlIFJ1bnRpbWVF"
    "cnJvcigibG9vcGJhY2sgVVJMcyBjYW5ub3QgY29udGFpbiBjcmVkZW50aWFscyIpCiAgICBp"
    "ZiBwYXJzZWQucXVlcnkgb3IgcGFyc2VkLmZyYWdtZW50OgogICAgICAgIHJhaXNlIFJ1bnRp"
    "bWVFcnJvcigibG9vcGJhY2sgVVJMcyBjYW5ub3QgY29udGFpbiBxdWVyeSBvciBmcmFnbWVu"
    "dCIpCiAgICByZXR1cm4gdXJsCgoKZGVmIGdldF90ZXh0KHVybDogc3RyLCB0aW1lb3V0OiBm"
    "bG9hdCA9IDEwLjApIC0+IHN0cjoKICAgIHJlcXVlc3QgPSB1cmxsaWIucmVxdWVzdC5SZXF1"
    "ZXN0KGJvdW5kZWRfbG9vcGJhY2sodXJsKSwgbWV0aG9kPSJHRVQiKQogICAgd2l0aCB1cmxs"
    "aWIucmVxdWVzdC51cmxvcGVuKHJlcXVlc3QsIHRpbWVvdXQ9dGltZW91dCkgYXMgcmVzcG9u"
    "c2U6CiAgICAgICAgcmV0dXJuIHJlc3BvbnNlLnJlYWQoKS5kZWNvZGUoInV0Zi04IikgICMg"
    "dHlwZTogaWdub3JlW25vLWFueS1yZXR1cm5dCgoKZGVmIGdldF9qc29uKHVybDogc3RyLCB0"
    "aW1lb3V0OiBmbG9hdCA9IDEwLjApIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgcGF5bG9h"
    "ZCA9IGpzb24ubG9hZHMoZ2V0X3RleHQodXJsLCB0aW1lb3V0PXRpbWVvdXQpKQogICAgaWYg"
    "bm90IGlzaW5zdGFuY2UocGF5bG9hZCwgZGljdCk6CiAgICAgICAgcmFpc2UgUnVudGltZUVy"
    "cm9yKCJsb29wYmFjayByZXNwb25zZSByb290IG11c3QgYmUgb25lIEpTT04gb2JqZWN0IikK"
    "ICAgIHJldHVybiBwYXlsb2FkCgoKZGVmIHBvc3RfanNvbigKICAgIHVybDogc3RyLAogICAg"
    "cGF5bG9hZDogZGljdFtzdHIsIG9iamVjdF0sCiAgICB0aW1lb3V0OiBmbG9hdCA9IEhUVFBf"
    "VElNRU9VVF9TRUNPTkRTLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgZW5jb2RlZCA9"
    "IGNhbm9uaWNhbF9qc29uKHBheWxvYWQpLmVuY29kZSgidXRmLTgiKQogICAgcmVxdWVzdCA9"
    "IHVybGxpYi5yZXF1ZXN0LlJlcXVlc3QoCiAgICAgICAgYm91bmRlZF9sb29wYmFjayh1cmwp"
    "LAogICAgICAgIGRhdGE9ZW5jb2RlZCwKICAgICAgICBoZWFkZXJzPXsiQ29udGVudC1UeXBl"
    "IjogImFwcGxpY2F0aW9uL2pzb24ifSwKICAgICAgICBtZXRob2Q9IlBPU1QiLAogICAgKQog"
    "ICAgd2l0aCB1cmxsaWIucmVxdWVzdC51cmxvcGVuKHJlcXVlc3QsIHRpbWVvdXQ9dGltZW91"
    "dCkgYXMgcmVzcG9uc2U6CiAgICAgICAgb2JzZXJ2ZWQgPSBqc29uLmxvYWRzKHJlc3BvbnNl"
    "LnJlYWQoKS5kZWNvZGUoInV0Zi04IikpCiAgICBpZiBub3QgaXNpbnN0YW5jZShvYnNlcnZl"
    "ZCwgZGljdCk6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJsb29wYmFjayByZXNwb25z"
    "ZSByb290IG11c3QgYmUgb25lIEpTT04gb2JqZWN0IikKICAgIHJldHVybiBvYnNlcnZlZAoK"
    "CmRlZiBwb3J0X29wZW4ocG9ydDogaW50KSAtPiBib29sOgogICAgdHJ5OgogICAgICAgIHdp"
    "dGggc29ja2V0LmNyZWF0ZV9jb25uZWN0aW9uKCgiMTI3LjAuMC4xIiwgcG9ydCksIHRpbWVv"
    "dXQ9MS4wKToKICAgICAgICAgICAgcmV0dXJuIFRydWUKICAgIGV4Y2VwdCBPU0Vycm9yOgog"
    "ICAgICAgIHJldHVybiBGYWxzZQoKCmRlZiByZXF1aXJlX3ByaXZhdGVfZW52aXJvbm1lbnQo"
    "KSAtPiBOb25lOgogICAgcHJlc2VudCA9IHR1cGxlKG5hbWUgZm9yIG5hbWUgaW4gQ1JFREVO"
    "VElBTF9FTlZfTkFNRVMgaWYgb3MuZW52aXJvbi5nZXQobmFtZSkpCiAgICBpZiBwcmVzZW50"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUDNfUDZf"
    "UFJJVkFDWV9CT1VOREFSWV9WSU9MQVRJT04iLAogICAgICAgICAgICAiY3JlZGVudGlhbC1i"
    "ZWFyaW5nIGVudmlyb25tZW50IHZhcmlhYmxlcyBhcmUgcHJvaGliaXRlZCIsCiAgICAgICAg"
    "KQogICAgaWYgb3MuZW52aXJvbi5nZXQoIkFVUkFHQVRFV0FZX0NVU1RPTUVSX0RBVEFfUFJF"
    "U0VOVCIpID09ICIxIjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIlAzX1A2X1BSSVZBQ1lfQk9VTkRBUllfVklPTEFUSU9OIiwKICAgICAgICAgICAg"
    "ImN1c3RvbWVyIGRhdGEgaXMgcHJvaGliaXRlZCIsCiAgICAgICAgKQoKCmRlZiBzYWZlX2Rp"
    "cmVjdG9yeV9zaGEyNTYocm9vdDogUGF0aCwgbWF4aW11bV9ieXRlczogaW50KSAtPiBzdHI6"
    "CiAgICBpZiBub3Qgcm9vdC5pc19kaXIoKSBvciByb290LmlzX3N5bWxpbmsoKToKICAgICAg"
    "ICByYWlzZSBSdW50aW1lRXJyb3IoImV4cGVjdGVkIG9uZSByZWFsIGRpcmVjdG9yeSIpCiAg"
    "ICBlbnRyaWVzOiBsaXN0W2RpY3Rbc3RyLCBvYmplY3RdXSA9IFtdCiAgICB0b3RhbCA9IDAK"
    "ICAgIGZvciBwYXRoIGluIHNvcnRlZChyb290LnJnbG9iKCIqIiksIGtleT1sYW1iZGEgaXRl"
    "bTogaXRlbS5hc19wb3NpeCgpKToKICAgICAgICBpZiBwYXRoLmlzX3N5bWxpbmsoKToKICAg"
    "ICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJkaXJlY3RvcnkgY29udGFpbnMgYSBzeW1i"
    "b2xpYyBsaW5rIikKICAgICAgICBpZiBwYXRoLmlzX2RpcigpOgogICAgICAgICAgICBjb250"
    "aW51ZQogICAgICAgIG1vZGUgPSBwYXRoLnN0YXQoKS5zdF9tb2RlCiAgICAgICAgaWYgbm90"
    "IHN0YXQuU19JU1JFRyhtb2RlKToKICAgICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJk"
    "aXJlY3RvcnkgY29udGFpbnMgYSBub24tcmVndWxhciBtZW1iZXIiKQogICAgICAgIHRvdGFs"
    "ICs9IHBhdGguc3RhdCgpLnN0X3NpemUKICAgICAgICBpZiB0b3RhbCA+IG1heGltdW1fYnl0"
    "ZXM6CiAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigiZGlyZWN0b3J5IGV4Y2VlZHMg"
    "dGhlIGJ5dGUgYnVkZ2V0IikKICAgICAgICBlbnRyaWVzLmFwcGVuZCgKICAgICAgICAgICAg"
    "ewogICAgICAgICAgICAgICAgInBhdGgiOiBwYXRoLnJlbGF0aXZlX3RvKHJvb3QpLmFzX3Bv"
    "c2l4KCksCiAgICAgICAgICAgICAgICAic2hhMjU2IjogZmlsZV9zaGEyNTYocGF0aCksCiAg"
    "ICAgICAgICAgICAgICAic2l6ZV9ieXRlcyI6IHBhdGguc3RhdCgpLnN0X3NpemUsCiAgICAg"
    "ICAgICAgIH0KICAgICAgICApCiAgICBpZiBub3QgZW50cmllczoKICAgICAgICByYWlzZSBS"
    "dW50aW1lRXJyb3IoImRpcmVjdG9yeSBpcyBlbXB0eSIpCiAgICBlbnZlbG9wZSA9IHsic2No"
    "ZW1hX3ZlcnNpb24iOiAiMS4wLjAiLCAiZmlsZXMiOiBlbnRyaWVzfQogICAgcmV0dXJuIHNo"
    "YTI1Nl90ZXh0KGNhbm9uaWNhbF9qc29uKGVudmVsb3BlKSkKCgpkZWYgZGlzY292ZXJfb25l"
    "X2RpcmVjdG9yeShuYW1lOiBzdHIpIC0+IFBhdGg6CiAgICBtYXRjaGVzID0gdHVwbGUoCiAg"
    "ICAgICAgcGF0aC5yZXNvbHZlKCkgZm9yIHBhdGggaW4gSU5QVVRfUk9PVC5yZ2xvYihuYW1l"
    "KSBpZiBwYXRoLmlzX2RpcigpIGFuZCBub3QgcGF0aC5pc19zeW1saW5rKCkKICAgICkKICAg"
    "IHVuaXF1ZSA9IHR1cGxlKGRpY3QuZnJvbWtleXMobWF0Y2hlcykpCiAgICBpZiBsZW4odW5p"
    "cXVlKSAhPSAxOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcihmImV4cGVjdGVkIG9uZSB7"
    "bmFtZX0gZGlyZWN0b3J5OyBvYnNlcnZlZCB7bGVuKHVuaXF1ZSl9IikKICAgIHJldHVybiB1"
    "bmlxdWVbMF0KCgpkZWYgZGlzY292ZXJfbW9kZWxfc25hcHNob3QoKSAtPiBQYXRoOgogICAg"
    "bWF0Y2hlcyA9IHR1cGxlKAogICAgICAgIHBhdGgucGFyZW50LnJlc29sdmUoKQogICAgICAg"
    "IGZvciBwYXRoIGluIElOUFVUX1JPT1Qucmdsb2IoZiJzbmFwc2hvdHMve01PREVMX1JFVklT"
    "SU9OfS9jb25maWcuanNvbiIpCiAgICAgICAgaWYgcGF0aC5pc19maWxlKCkgYW5kIG5vdCBw"
    "YXRoLmlzX3N5bWxpbmsoKQogICAgKQogICAgdW5pcXVlID0gdHVwbGUoZGljdC5mcm9ta2V5"
    "cyhtYXRjaGVzKSkKICAgIGlmIGxlbih1bmlxdWUpICE9IDE6CiAgICAgICAgcmFpc2UgUnVu"
    "dGltZUVycm9yKGYiZXhwZWN0ZWQgb25lIGV4YWN0IGV4cGFuZGVkIG1vZGVsIHNuYXBzaG90"
    "OyBvYnNlcnZlZCB7bGVuKHVuaXF1ZSl9IikKICAgIHNuYXBzaG90ID0gdW5pcXVlWzBdCiAg"
    "ICBpZiBzYWZlX2RpcmVjdG9yeV9zaGEyNTYoc25hcHNob3QsIE1BWF9NT0RFTF9CWVRFUykg"
    "IT0gTU9ERUxfU05BUFNIT1RfU0hBMjU2OgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigi"
    "bW9kZWwgc25hcHNob3QgaWRlbnRpdHkgZHJpZnRlZCIpCiAgICByZXR1cm4gc25hcHNob3QK"
    "CgpkZWYgcHJlcGFyZV9tb2RlbF9ob21lKHNuYXBzaG90OiBQYXRoKSAtPiB0dXBsZVtQYXRo"
    "LCBQYXRoXToKICAgIG1vZGVsX2hvbWUgPSBTQ1JBVENIX1JPT1QgLyAibW9kZWxfaG9tZSIK"
    "ICAgIGRlc3RpbmF0aW9uID0gKAogICAgICAgIG1vZGVsX2hvbWUgLyAiaHViIiAvICJtb2Rl"
    "bHMtLVF3ZW4tLVF3ZW4yLjUtMC41Qi1JbnN0cnVjdCIgLyAic25hcHNob3RzIiAvIE1PREVM"
    "X1JFVklTSU9OCiAgICApCiAgICBpZiBtb2RlbF9ob21lLmV4aXN0cygpOgogICAgICAgIHJh"
    "aXNlIFJ1bnRpbWVFcnJvcigid3JpdGFibGUgbW9kZWwgaG9tZSBhbHJlYWR5IGV4aXN0cyIp"
    "CiAgICB0b3RhbCA9IDAKICAgIGZpbGVzOiBsaXN0W3R1cGxlW1BhdGgsIFBhdGhdXSA9IFtd"
    "CiAgICBmb3IgcGF0aCBpbiBzb3J0ZWQoc25hcHNob3Qucmdsb2IoIioiKSwga2V5PWxhbWJk"
    "YSBpdGVtOiBpdGVtLmFzX3Bvc2l4KCkpOgogICAgICAgIGlmIHBhdGguaXNfc3ltbGluaygp"
    "OgogICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoIm1vZGVsIHNuYXBzaG90IGNvbnRh"
    "aW5zIGEgc3ltYm9saWMgbGluayIpCiAgICAgICAgaWYgcGF0aC5pc19kaXIoKToKICAgICAg"
    "ICAgICAgY29udGludWUKICAgICAgICBpZiBub3Qgc3RhdC5TX0lTUkVHKHBhdGguc3RhdCgp"
    "LnN0X21vZGUpOgogICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoIm1vZGVsIHNuYXBz"
    "aG90IGNvbnRhaW5zIGEgbm9uLXJlZ3VsYXIgbWVtYmVyIikKICAgICAgICB0b3RhbCArPSBw"
    "YXRoLnN0YXQoKS5zdF9zaXplCiAgICAgICAgaWYgdG90YWwgPiBNQVhfTU9ERUxfQllURVM6"
    "CiAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigibW9kZWwgc25hcHNob3QgZXhjZWVk"
    "cyB0aGUgY29weSBidWRnZXQiKQogICAgICAgIGZpbGVzLmFwcGVuZCgocGF0aCwgZGVzdGlu"
    "YXRpb24gLyBwYXRoLnJlbGF0aXZlX3RvKHNuYXBzaG90KSkpCiAgICBpZiBub3QgZmlsZXM6"
    "CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJtb2RlbCBzbmFwc2hvdCBpcyBlbXB0eSIp"
    "CiAgICBmb3Igc291cmNlLCB0YXJnZXQgaW4gZmlsZXM6CiAgICAgICAgdGFyZ2V0LnBhcmVu"
    "dC5ta2RpcihwYXJlbnRzPVRydWUsIGV4aXN0X29rPVRydWUpCiAgICAgICAgc2h1dGlsLmNv"
    "cHkyKHNvdXJjZSwgdGFyZ2V0LCBmb2xsb3dfc3ltbGlua3M9RmFsc2UpCiAgICBpZiBzYWZl"
    "X2RpcmVjdG9yeV9zaGEyNTYoZGVzdGluYXRpb24sIE1BWF9NT0RFTF9CWVRFUykgIT0gTU9E"
    "RUxfU05BUFNIT1RfU0hBMjU2OgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigid3JpdGFi"
    "bGUgbW9kZWwgc25hcHNob3QgaWRlbnRpdHkgZHJpZnRlZCIpCiAgICByZXR1cm4gbW9kZWxf"
    "aG9tZSwgZGVzdGluYXRpb24KCgpkZWYgdmFsaWRhdGVfd2hlZWxob3VzZShyb290OiBQYXRo"
    "KSAtPiBOb25lOgogICAgZm9yIG5hbWUsIGV4cGVjdGVkIGluIEVYUEVDVEVEX0NPTlRST0xf"
    "SEFTSEVTLml0ZW1zKCk6CiAgICAgICAgcGF0aCA9IHJvb3QgLyBuYW1lCiAgICAgICAgaWYg"
    "bm90IHBhdGguaXNfZmlsZSgpIG9yIHBhdGguaXNfc3ltbGluaygpOgogICAgICAgICAgICBy"
    "YWlzZSBSdW50aW1lRXJyb3IoZiJ3aGVlbGhvdXNlIGNvbnRyb2wgZmlsZSBtaXNzaW5nOiB7"
    "bmFtZX0iKQogICAgICAgIGlmIGZpbGVfc2hhMjU2KHBhdGgpICE9IGV4cGVjdGVkOgogICAg"
    "ICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoZiJ3aGVlbGhvdXNlIGNvbnRyb2wgaWRlbnRp"
    "dHkgZHJpZnRlZDoge25hbWV9IikKICAgIG1hbmlmZXN0ID0ganNvbi5sb2Fkcygocm9vdCAv"
    "ICJzaGEyNTZfbWFuaWZlc3QuanNvbiIpLnJlYWRfdGV4dChlbmNvZGluZz0idXRmLTgiKSkK"
    "ICAgIGlmIG5vdCBpc2luc3RhbmNlKG1hbmlmZXN0LCBkaWN0KToKICAgICAgICByYWlzZSBS"
    "dW50aW1lRXJyb3IoIndoZWVsaG91c2UgY2hlY2tzdW0gbWFuaWZlc3QgaXMgaW52YWxpZCIp"
    "CiAgICBlbnRyaWVzID0gbWFuaWZlc3QuZ2V0KCJlbnRyaWVzIikKICAgIGlmIG5vdCBpc2lu"
    "c3RhbmNlKGVudHJpZXMsIGxpc3QpIG9yIGxlbihlbnRyaWVzKSAhPSAyMDA6CiAgICAgICAg"
    "cmFpc2UgUnVudGltZUVycm9yKCJ3aGVlbGhvdXNlIG1hbmlmZXN0IGVudHJ5IGNvdW50IGRy"
    "aWZ0ZWQiKQogICAgdmVyaWZpZWQgPSAwCiAgICB3aGVlbHMgPSAwCiAgICBmb3IgcmF3IGlu"
    "IGVudHJpZXM6CiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2UocmF3LCBkaWN0KToKICAgICAg"
    "ICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJ3aGVlbGhvdXNlIG1hbmlmZXN0IGVudHJ5IGlz"
    "IGludmFsaWQiKQogICAgICAgIG5hbWUgPSByYXcuZ2V0KCJwYXRoIikgICMgdHlwZTogaWdu"
    "b3JlW2Fzc2lnbm1lbnRdCiAgICAgICAgZGlnZXN0ID0gcmF3LmdldCgic2hhMjU2IikKICAg"
    "ICAgICBzaXplID0gcmF3LmdldCgic2l6ZV9ieXRlcyIpCiAgICAgICAgaWYgbm90IGlzaW5z"
    "dGFuY2UobmFtZSwgc3RyKToKICAgICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJ3aGVl"
    "bGhvdXNlIG1hbmlmZXN0IHBhdGggaXMgaW52YWxpZCIpCiAgICAgICAgaWYgbm90IGlzaW5z"
    "dGFuY2UoZGlnZXN0LCBzdHIpIG9yIG5vdCBpc2luc3RhbmNlKHNpemUsIGludCk6CiAgICAg"
    "ICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigid2hlZWxob3VzZSBtYW5pZmVzdCBpZGVudGl0"
    "eSBpcyBpbnZhbGlkIikKICAgICAgICBwYXRoID0gcm9vdCAvIG5hbWUKICAgICAgICBpZiBu"
    "b3QgcGF0aC5pc19maWxlKCkgb3IgcGF0aC5pc19zeW1saW5rKCk6CiAgICAgICAgICAgIHJh"
    "aXNlIFJ1bnRpbWVFcnJvcihmIndoZWVsaG91c2UgbWVtYmVyIG1pc3Npbmc6IHtuYW1lfSIp"
    "CiAgICAgICAgaWYgZmlsZV9zaGEyNTYocGF0aCkgIT0gZGlnZXN0IG9yIHBhdGguc3RhdCgp"
    "LnN0X3NpemUgIT0gc2l6ZToKICAgICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKGYid2hl"
    "ZWxob3VzZSBtZW1iZXIgZHJpZnRlZDoge25hbWV9IikKICAgICAgICB2ZXJpZmllZCArPSAx"
    "CiAgICAgICAgaWYgcGF0aC5zdWZmaXggPT0gIi53aGwiOgogICAgICAgICAgICB3aGVlbHMg"
    "Kz0gMQogICAgaWYgdmVyaWZpZWQgIT0gMjAwIG9yIHdoZWVscyAhPSAxOTY6CiAgICAgICAg"
    "cmFpc2UgUnVudGltZUVycm9yKCJ3aGVlbGhvdXNlIHZlcmlmaWNhdGlvbiBjb3VudHMgZHJp"
    "ZnRlZCIpCgoKY2xhc3MgQm91bmRlZENhcHR1cmU6CiAgICBkZWYgX19pbml0X18oc2VsZiwg"
    "cGF0aDogUGF0aCkgLT4gTm9uZToKICAgICAgICBzZWxmLnBhdGggPSBwYXRoCiAgICAgICAg"
    "c2VsZi5idWZmZXIgPSBieXRlYXJyYXkoKQogICAgICAgIHNlbGYub2JzZXJ2ZWQgPSAwCiAg"
    "ICAgICAgc2VsZi5sb2NrID0gdGhyZWFkaW5nLkxvY2soKQogICAgICAgIHNlbGYudGhyZWFk"
    "OiB0aHJlYWRpbmcuVGhyZWFkIHwgTm9uZSA9IE5vbmUKICAgICAgICBzZWxmLmZpbmFsaXpl"
    "ZCA9IEZhbHNlCgogICAgZGVmIHN0YXJ0KHNlbGYsIHNvdXJjZTogb2JqZWN0KSAtPiBOb25l"
    "OgogICAgICAgIGlmIHNvdXJjZSBpcyBOb25lOgogICAgICAgICAgICByZXR1cm4KICAgICAg"
    "ICBpZiBzZWxmLnRocmVhZCBpcyBub3QgTm9uZToKICAgICAgICAgICAgcmFpc2UgUnVudGlt"
    "ZUVycm9yKCJjYXB0dXJlIGhhcyBhbHJlYWR5IGJlZW4gc3RhcnRlZCIpCiAgICAgICAgc2Vs"
    "Zi5wYXRoLnBhcmVudC5ta2RpcihwYXJlbnRzPVRydWUsIGV4aXN0X29rPVRydWUpCiAgICAg"
    "ICAgc2VsZi50aHJlYWQgPSB0aHJlYWRpbmcuVGhyZWFkKAogICAgICAgICAgICB0YXJnZXQ9"
    "c2VsZi5jb25zdW1lLAogICAgICAgICAgICBhcmdzPShzb3VyY2UsKSwKICAgICAgICAgICAg"
    "ZGFlbW9uPVRydWUsCiAgICAgICAgKQogICAgICAgIHNlbGYudGhyZWFkLnN0YXJ0KCkKCiAg"
    "ICBkZWYgY29uc3VtZShzZWxmLCBzb3VyY2U6IG9iamVjdCkgLT4gTm9uZToKICAgICAgICB0"
    "cnk6CiAgICAgICAgICAgIHdoaWxlIFRydWU6CiAgICAgICAgICAgICAgICBjaHVuayA9IHNv"
    "dXJjZS5yZWFkKDgxOTIpICAjIHR5cGU6IGlnbm9yZVthdHRyLWRlZmluZWRdCiAgICAgICAg"
    "ICAgICAgICBpZiBub3QgY2h1bms6CiAgICAgICAgICAgICAgICAgICAgYnJlYWsKICAgICAg"
    "ICAgICAgICAgIHdpdGggc2VsZi5sb2NrOgogICAgICAgICAgICAgICAgICAgIHNlbGYub2Jz"
    "ZXJ2ZWQgKz0gbGVuKGNodW5rKQogICAgICAgICAgICAgICAgICAgIHNlbGYuYnVmZmVyLmV4"
    "dGVuZChjaHVuaykKICAgICAgICAgICAgICAgICAgICBvdmVyZmxvdyA9IGxlbihzZWxmLmJ1"
    "ZmZlcikgLSBNQVhfU1RSRUFNX0JZVEVTCiAgICAgICAgICAgICAgICAgICAgaWYgb3ZlcmZs"
    "b3cgPiAwOgogICAgICAgICAgICAgICAgICAgICAgICBkZWwgc2VsZi5idWZmZXJbOm92ZXJm"
    "bG93XQogICAgICAgICAgICAgICAgICAgIHNlbGYucGF0aC53cml0ZV9ieXRlcyhieXRlcyhz"
    "ZWxmLmJ1ZmZlcikpCiAgICAgICAgZmluYWxseToKICAgICAgICAgICAgc291cmNlLmNsb3Nl"
    "KCkgICMgdHlwZTogaWdub3JlW2F0dHItZGVmaW5lZF0KCiAgICBkZWYgZmluYWxpemUoc2Vs"
    "ZiwgdGltZW91dF9zZWNvbmRzOiBmbG9hdCA9IDUuMCkgLT4gTm9uZToKICAgICAgICBpZiBz"
    "ZWxmLnRocmVhZCBpcyBub3QgTm9uZToKICAgICAgICAgICAgc2VsZi50aHJlYWQuam9pbih0"
    "aW1lb3V0PXRpbWVvdXRfc2Vjb25kcykKICAgICAgICAgICAgaWYgc2VsZi50aHJlYWQuaXNf"
    "YWxpdmUoKToKICAgICAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigiY2FwdHVyZSB0"
    "aHJlYWQgZGlkIG5vdCBmaW5hbGl6ZSIpCiAgICAgICAgc2VsZi5maW5hbGl6ZWQgPSBUcnVl"
    "CgogICAgZGVmIHNuYXBzaG90KHNlbGYpIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgICAg"
    "IHdpdGggc2VsZi5sb2NrOgogICAgICAgICAgICBwYXlsb2FkID0gYnl0ZXMoc2VsZi5idWZm"
    "ZXIpCiAgICAgICAgICAgIHRocmVhZF9hbGl2ZSA9IEZhbHNlIGlmIHNlbGYudGhyZWFkIGlz"
    "IE5vbmUgZWxzZSBzZWxmLnRocmVhZC5pc19hbGl2ZSgpCiAgICAgICAgICAgIHJldHVybiB7"
    "CiAgICAgICAgICAgICAgICAib2JzZXJ2ZWRfYnl0ZXMiOiBzZWxmLm9ic2VydmVkLAogICAg"
    "ICAgICAgICAgICAgInJldGFpbmVkX2J5dGVzIjogbGVuKHBheWxvYWQpLAogICAgICAgICAg"
    "ICAgICAgInRhaWxfc2hhMjU2Ijogc2hhMjU2X2J5dGVzKHBheWxvYWQpLAogICAgICAgICAg"
    "ICAgICAgImNhcHR1cmVfZmluYWxpemVkIjogc2VsZi5maW5hbGl6ZWQsCiAgICAgICAgICAg"
    "ICAgICAiY2FwdHVyZV90aHJlYWRfYWxpdmUiOiB0aHJlYWRfYWxpdmUsCiAgICAgICAgICAg"
    "IH0KCiAgICBkZWYgdGV4dChzZWxmKSAtPiBzdHI6CiAgICAgICAgd2l0aCBzZWxmLmxvY2s6"
    "CiAgICAgICAgICAgIHJldHVybiBieXRlcyhzZWxmLmJ1ZmZlcikuZGVjb2RlKCJ1dGYtOCIs"
    "IGVycm9ycz0icmVwbGFjZSIpCgoKZGVmIHNhbml0aXplX2V4Y2VycHQodmFsdWU6IHN0cikg"
    "LT4gc3RyOgogICAgYm91bmRlZCA9IHZhbHVlWy1NQVhfSU5TVEFMTF9FWENFUlBUX0NIQVJB"
    "Q1RFUlM6XQogICAgcmVwbGFjZW1lbnRzID0gewogICAgICAgIHN0cihJTlBVVF9ST09UKTog"
    "IjxpbnB1dD4iLAogICAgICAgIHN0cihXT1JLX1JPT1QpOiAiPHdvcmtpbmc+IiwKICAgICAg"
    "ICBvcy5lbnZpcm9uLmdldCgiSE9NRSIsICIiKTogIjxob21lPiIsCiAgICB9CiAgICBmb3Ig"
    "c291cmNlLCByZXBsYWNlbWVudCBpbiByZXBsYWNlbWVudHMuaXRlbXMoKToKICAgICAgICBp"
    "ZiBzb3VyY2U6CiAgICAgICAgICAgIGJvdW5kZWQgPSBib3VuZGVkLnJlcGxhY2Uoc291cmNl"
    "LCByZXBsYWNlbWVudCkKICAgIHJldHVybiBib3VuZGVkCgoKZGVmIGRpc2tfc25hcHNob3Qo"
    "cGF0aDogUGF0aCkgLT4gZGljdFtzdHIsIGludF06CiAgICB1c2FnZSA9IHNodXRpbC5kaXNr"
    "X3VzYWdlKHBhdGgpCiAgICByZXR1cm4gewogICAgICAgICJ0b3RhbF9ieXRlcyI6IHVzYWdl"
    "LnRvdGFsLAogICAgICAgICJ1c2VkX2J5dGVzIjogdXNhZ2UudXNlZCwKICAgICAgICAiZnJl"
    "ZV9ieXRlcyI6IHVzYWdlLmZyZWUsCiAgICB9CgoKZGVmIGRpcmVjdG9yeV9zbmFwc2hvdChw"
    "YXRoOiBQYXRoKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIG5vdCBwYXRoLmV4aXN0"
    "cygpOgogICAgICAgIHJldHVybiB7CiAgICAgICAgICAgICJleGlzdHMiOiBGYWxzZSwKICAg"
    "ICAgICAgICAgImZpbGVfY291bnQiOiAwLAogICAgICAgICAgICAic3ltbGlua19jb3VudCI6"
    "IDAsCiAgICAgICAgICAgICJzaXplX2J5dGVzIjogMCwKICAgICAgICB9CiAgICBpZiBub3Qg"
    "cGF0aC5pc19kaXIoKSBvciBwYXRoLmlzX3N5bWxpbmsoKToKICAgICAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoInRhcmdldCBydW50aW1lIHBhdGggaXMgbm90IG9uZSByZWFsIGRpcmVjdG9y"
    "eSIpCiAgICBmaWxlX2NvdW50ID0gMAogICAgc3ltbGlua19jb3VudCA9IDAKICAgIHNpemVf"
    "Ynl0ZXMgPSAwCiAgICBmb3IgbWVtYmVyIGluIHBhdGgucmdsb2IoIioiKToKICAgICAgICBp"
    "ZiBtZW1iZXIuaXNfc3ltbGluaygpOgogICAgICAgICAgICBzeW1saW5rX2NvdW50ICs9IDEK"
    "ICAgICAgICAgICAgY29udGludWUKICAgICAgICBpZiBtZW1iZXIuaXNfZGlyKCk6CiAgICAg"
    "ICAgICAgIGNvbnRpbnVlCiAgICAgICAgaWYgbm90IHN0YXQuU19JU1JFRyhtZW1iZXIuc3Rh"
    "dCgpLnN0X21vZGUpOgogICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoInRhcmdldCBy"
    "dW50aW1lIGNvbnRhaW5zIGEgbm9uLXJlZ3VsYXIgbWVtYmVyIikKICAgICAgICBmaWxlX2Nv"
    "dW50ICs9IDEKICAgICAgICBzaXplX2J5dGVzICs9IG1lbWJlci5zdGF0KCkuc3Rfc2l6ZQog"
    "ICAgcmV0dXJuIHsKICAgICAgICAiZXhpc3RzIjogVHJ1ZSwKICAgICAgICAiZmlsZV9jb3Vu"
    "dCI6IGZpbGVfY291bnQsCiAgICAgICAgInN5bWxpbmtfY291bnQiOiBzeW1saW5rX2NvdW50"
    "LAogICAgICAgICJzaXplX2J5dGVzIjogc2l6ZV9ieXRlcywKICAgIH0KCgpkZWYgaW5zdGFs"
    "bF9mYWlsdXJlX3NpZ25hbHMoc3Rkb3V0X3RhaWw6IHN0ciwgc3RkZXJyX3RhaWw6IHN0cikg"
    "LT4gdHVwbGVbc3RyLCAuLi5dOgogICAgdGV4dCA9IChzdGRvdXRfdGFpbCArICJcbiIgKyBz"
    "dGRlcnJfdGFpbCkubG93ZXIoKQogICAgc2lnbmFsczogbGlzdFtzdHJdID0gW10KICAgIHBh"
    "dHRlcm5zID0gKAogICAgICAgICgibm8gc3BhY2UgbGVmdCBvbiBkZXZpY2UiLCAiRElTS19F"
    "WEhBVVNUSU9OX1NJR05BTCIpLAogICAgICAgICgidGhlc2UgcGFja2FnZXMgZG8gbm90IG1h"
    "dGNoIHRoZSBoYXNoZXMiLCAiSEFTSF9NSVNNQVRDSF9TSUdOQUwiKSwKICAgICAgICAoImhh"
    "c2hlcyBhcmUgcmVxdWlyZWQiLCAiSEFTSF9SRVFVSVJFTUVOVF9TSUdOQUwiKSwKICAgICAg"
    "ICAoIm5vIG1hdGNoaW5nIGRpc3RyaWJ1dGlvbiBmb3VuZCIsICJESVNUUklCVVRJT05fVU5B"
    "VkFJTEFCTEVfU0lHTkFMIiksCiAgICAgICAgKCJjb3VsZCBub3QgZmluZCBhIHZlcnNpb24g"
    "dGhhdCBzYXRpc2ZpZXMiLCAiRElTVFJJQlVUSU9OX1VOQVZBSUxBQkxFX1NJR05BTCIpLAog"
    "ICAgICAgICgibm90IGEgc3VwcG9ydGVkIHdoZWVsIG9uIHRoaXMgcGxhdGZvcm0iLCAiVU5T"
    "VVBQT1JURURfV0hFRUxfU0lHTkFMIiksCiAgICAgICAgKCJyZXNvbHV0aW9uaW1wb3NzaWJs"
    "ZSIsICJERVBFTkRFTkNZX1JFU09MVVRJT05fU0lHTkFMIiksCiAgICAgICAgKCJjb25mbGlj"
    "dGluZyBkZXBlbmRlbmNpZXMiLCAiREVQRU5ERU5DWV9SRVNPTFVUSU9OX1NJR05BTCIpLAog"
    "ICAgKQogICAgZm9yIHBocmFzZSwgc2lnbmFsIGluIHBhdHRlcm5zOgogICAgICAgIGlmIHBo"
    "cmFzZSBpbiB0ZXh0IGFuZCBzaWduYWwgbm90IGluIHNpZ25hbHM6CiAgICAgICAgICAgIHNp"
    "Z25hbHMuYXBwZW5kKHNpZ25hbCkKICAgIHJldHVybiB0dXBsZShzaWduYWxzKQoKCmRlZiBy"
    "dW5fYm91bmRlZF9wcm9jZXNzKAogICAgcm9sZTogc3RyLAogICAgYXJndjogbGlzdFtzdHJd"
    "LAogICAgKiwKICAgIHRpbWVvdXRfc2Vjb25kczogZmxvYXQsCiAgICBlbnZpcm9ubWVudDog"
    "ZGljdFtzdHIsIHN0cl0sCiAgICBjYXB0dXJlX3Jvb3Q6IFBhdGgsCikgLT4gZGljdFtzdHIs"
    "IG9iamVjdF06CiAgICBzdGFydGVkX21vbm90b25pYyA9IHRpbWUubW9ub3RvbmljKCkKICAg"
    "IHN0YXJ0ZWRfYXQgPSBkYXRldGltZS5ub3coVVRDKS5pc29mb3JtYXQodGltZXNwZWM9InNl"
    "Y29uZHMiKQogICAgc3Rkb3V0X2NhcHR1cmUgPSBCb3VuZGVkQ2FwdHVyZShjYXB0dXJlX3Jv"
    "b3QgLyBmIntyb2xlfS5zdGRvdXQubG9nIikKICAgIHN0ZGVycl9jYXB0dXJlID0gQm91bmRl"
    "ZENhcHR1cmUoY2FwdHVyZV9yb290IC8gZiJ7cm9sZX0uc3RkZXJyLmxvZyIpCiAgICBwcm9j"
    "ZXNzOiBzdWJwcm9jZXNzLlBvcGVuW2J5dGVzXSB8IE5vbmUgPSBOb25lCiAgICBsYXVuY2hf"
    "ZXJyb3JfdHlwZTogc3RyIHwgTm9uZSA9IE5vbmUKICAgIGxhdW5jaF9lcnJvcl9tZXNzYWdl"
    "OiBzdHIgfCBOb25lID0gTm9uZQogICAgdGltZWRfb3V0ID0gRmFsc2UKICAgIHRyeToKICAg"
    "ICAgICBwcm9jZXNzID0gc3VicHJvY2Vzcy5Qb3BlbigKICAgICAgICAgICAgYXJndiwKICAg"
    "ICAgICAgICAgZW52PWVudmlyb25tZW50LAogICAgICAgICAgICBzdGRpbj1zdWJwcm9jZXNz"
    "LkRFVk5VTEwsCiAgICAgICAgICAgIHN0ZG91dD1zdWJwcm9jZXNzLlBJUEUsCiAgICAgICAg"
    "ICAgIHN0ZGVycj1zdWJwcm9jZXNzLlBJUEUsCiAgICAgICAgICAgIHRleHQ9RmFsc2UsCiAg"
    "ICAgICAgICAgIGJ1ZnNpemU9MCwKICAgICAgICApCiAgICAgICAgc3Rkb3V0X2NhcHR1cmUu"
    "c3RhcnQocHJvY2Vzcy5zdGRvdXQpCiAgICAgICAgc3RkZXJyX2NhcHR1cmUuc3RhcnQocHJv"
    "Y2Vzcy5zdGRlcnIpCiAgICAgICAgdHJ5OgogICAgICAgICAgICBwcm9jZXNzLndhaXQodGlt"
    "ZW91dD10aW1lb3V0X3NlY29uZHMpCiAgICAgICAgZXhjZXB0IHN1YnByb2Nlc3MuVGltZW91"
    "dEV4cGlyZWQ6CiAgICAgICAgICAgIHRpbWVkX291dCA9IFRydWUKICAgICAgICAgICAgcHJv"
    "Y2Vzcy50ZXJtaW5hdGUoKQogICAgICAgICAgICB0cnk6CiAgICAgICAgICAgICAgICBwcm9j"
    "ZXNzLndhaXQodGltZW91dD0xMCkKICAgICAgICAgICAgZXhjZXB0IHN1YnByb2Nlc3MuVGlt"
    "ZW91dEV4cGlyZWQ6CiAgICAgICAgICAgICAgICBwcm9jZXNzLmtpbGwoKQogICAgICAgICAg"
    "ICAgICAgcHJvY2Vzcy53YWl0KHRpbWVvdXQ9MTApCiAgICBleGNlcHQgT1NFcnJvciBhcyBl"
    "cnJvcjoKICAgICAgICBsYXVuY2hfZXJyb3JfdHlwZSA9IHR5cGUoZXJyb3IpLl9fbmFtZV9f"
    "CiAgICAgICAgbGF1bmNoX2Vycm9yX21lc3NhZ2UgPSBzYW5pdGl6ZV9leGNlcnB0KHN0cihl"
    "cnJvcikpCiAgICBmaW5hbGx5OgogICAgICAgIGlmIHN0ZG91dF9jYXB0dXJlLnRocmVhZCBp"
    "cyBub3QgTm9uZToKICAgICAgICAgICAgc3Rkb3V0X2NhcHR1cmUudGhyZWFkLmpvaW4odGlt"
    "ZW91dD01KQogICAgICAgIGlmIHN0ZGVycl9jYXB0dXJlLnRocmVhZCBpcyBub3QgTm9uZToK"
    "ICAgICAgICAgICAgc3RkZXJyX2NhcHR1cmUudGhyZWFkLmpvaW4odGltZW91dD01KQoKICAg"
    "IGZpbmlzaGVkX2F0ID0gZGF0ZXRpbWUubm93KFVUQykuaXNvZm9ybWF0KHRpbWVzcGVjPSJz"
    "ZWNvbmRzIikKICAgIHN0ZG91dF90YWlsID0gc2FuaXRpemVfZXhjZXJwdChzdGRvdXRfY2Fw"
    "dHVyZS50ZXh0KCkpCiAgICBzdGRlcnJfdGFpbCA9IHNhbml0aXplX2V4Y2VycHQoc3RkZXJy"
    "X2NhcHR1cmUudGV4dCgpKQogICAgcmV0dXJuY29kZSA9IE5vbmUgaWYgcHJvY2VzcyBpcyBO"
    "b25lIGVsc2UgcHJvY2Vzcy5yZXR1cm5jb2RlCiAgICBpZiBsYXVuY2hfZXJyb3JfdHlwZSBp"
    "cyBub3QgTm9uZToKICAgICAgICBwcm9jZXNzX291dGNvbWUgPSAiTEFVTkNIX0VSUk9SIgog"
    "ICAgZWxpZiB0aW1lZF9vdXQ6CiAgICAgICAgcHJvY2Vzc19vdXRjb21lID0gIlRJTUVPVVQi"
    "CiAgICBlbGlmIHJldHVybmNvZGUgPT0gMDoKICAgICAgICBwcm9jZXNzX291dGNvbWUgPSAi"
    "WkVST19FWElUIgogICAgZWxzZToKICAgICAgICBwcm9jZXNzX291dGNvbWUgPSAiTk9OWkVS"
    "T19FWElUIgogICAgcmV0dXJuIHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAi"
    "LAogICAgICAgICJjb21tYW5kX3JvbGUiOiByb2xlLAogICAgICAgICJzdGF0dXMiOiAiUEFT"
    "U0VEIiBpZiBwcm9jZXNzX291dGNvbWUgPT0gIlpFUk9fRVhJVCIgZWxzZSAiRkFJTEVEIiwK"
    "ICAgICAgICAicHJvY2Vzc19vdXRjb21lIjogcHJvY2Vzc19vdXRjb21lLAogICAgICAgICJh"
    "cmd2IjogW3Nhbml0aXplX2V4Y2VycHQoaXRlbSkgZm9yIGl0ZW0gaW4gYXJndl0sCiAgICAg"
    "ICAgImFyZ3Zfc2hhMjU2Ijogc2hhMjU2X3RleHQoY2Fub25pY2FsX2pzb24oYXJndikpLAog"
    "ICAgICAgICJzdGFydGVkX2F0Ijogc3RhcnRlZF9hdCwKICAgICAgICAiZmluaXNoZWRfYXQi"
    "OiBmaW5pc2hlZF9hdCwKICAgICAgICAiZHVyYXRpb25fbXMiOiBpbnQoKHRpbWUubW9ub3Rv"
    "bmljKCkgLSBzdGFydGVkX21vbm90b25pYykgKiAxMDAwKSwKICAgICAgICAicmV0dXJuY29k"
    "ZSI6IHJldHVybmNvZGUsCiAgICAgICAgInRpbWVkX291dCI6IHRpbWVkX291dCwKICAgICAg"
    "ICAibGF1bmNoX2Vycm9yX3R5cGUiOiBsYXVuY2hfZXJyb3JfdHlwZSwKICAgICAgICAibGF1"
    "bmNoX2Vycm9yX21lc3NhZ2UiOiBsYXVuY2hfZXJyb3JfbWVzc2FnZSwKICAgICAgICAic3Rk"
    "b3V0X29ic2VydmVkX2J5dGVzIjogc3Rkb3V0X2NhcHR1cmUub2JzZXJ2ZWQsCiAgICAgICAg"
    "InN0ZGVycl9vYnNlcnZlZF9ieXRlcyI6IHN0ZGVycl9jYXB0dXJlLm9ic2VydmVkLAogICAg"
    "ICAgICJzdGRvdXRfdGFpbCI6IHN0ZG91dF90YWlsLAogICAgICAgICJzdGRlcnJfdGFpbCI6"
    "IHN0ZGVycl90YWlsLAogICAgICAgICJmYWlsdXJlX3NpZ25hbHMiOiBpbnN0YWxsX2ZhaWx1"
    "cmVfc2lnbmFscyhzdGRvdXRfdGFpbCwgc3RkZXJyX3RhaWwpLAogICAgfQoKCmRlZiBpbnN0"
    "YWxsX3J1bnRpbWUod2hlZWxob3VzZTogUGF0aCwgY291bnRlcnM6IGRpY3Rbc3RyLCBpbnRd"
    "KSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIFRBUkdFVF9ST09ULmV4aXN0cygpOgog"
    "ICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigidGFyZ2V0IHJ1bnRpbWUgYWxyZWFkeSBleGlz"
    "dHMiKQogICAgd2hlZWxzID0gd2hlZWxob3VzZSAvICJ3aGVlbHMiCiAgICBpZiBub3Qgd2hl"
    "ZWxzLmlzX2RpcigpIG9yIHdoZWVscy5pc19zeW1saW5rKCk6CiAgICAgICAgcmFpc2UgUnVu"
    "dGltZUVycm9yKCJ3aGVlbGhvdXNlIHdoZWVscyBkaXJlY3RvcnkgaXMgbWlzc2luZyBvciB1"
    "bnNhZmUiKQoKICAgIGNyZWF0ZV9wcm9jZXNzID0gcnVuX2JvdW5kZWRfcHJvY2VzcygKICAg"
    "ICAgICAidGFyZ2V0X2Vudmlyb25tZW50X2NyZWF0aW9uIiwKICAgICAgICBbCiAgICAgICAg"
    "ICAgIHN5cy5leGVjdXRhYmxlLAogICAgICAgICAgICAiLW0iLAogICAgICAgICAgICAidmVu"
    "diIsCiAgICAgICAgICAgICItLXdpdGhvdXQtcGlwIiwKICAgICAgICAgICAgc3RyKFRBUkdF"
    "VF9ST09UKSwKICAgICAgICBdLAogICAgICAgIHRpbWVvdXRfc2Vjb25kcz0xMjAuMCwKICAg"
    "ICAgICBlbnZpcm9ubWVudD17Kipvcy5lbnZpcm9ufSwKICAgICAgICBjYXB0dXJlX3Jvb3Q9"
    "U0NSQVRDSF9ST09UIC8gInRhcmdldF9lbnZpcm9ubWVudF9sb2dzIiwKICAgICkKICAgIGlm"
    "IGNyZWF0ZV9wcm9jZXNzWyJwcm9jZXNzX291dGNvbWUiXSAhPSAiWkVST19FWElUIjoKICAg"
    "ICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk1PREVMX0NPTlNU"
    "UlVDVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgImlzb2xhdGVkIHRhcmdldCBlbnZpcm9u"
    "bWVudCBjcmVhdGlvbiBmYWlsZWQiLAogICAgICAgICkKICAgIGlmIG5vdCBUQVJHRVRfUFlU"
    "SE9OLmlzX2ZpbGUoKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIk1PREVMX0NPTlNUUlVDVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgImlzb2xh"
    "dGVkIHRhcmdldCBQeXRob24gZXhlY3V0YWJsZSBpcyB1bmF2YWlsYWJsZSIsCiAgICAgICAg"
    "KQoKICAgIGFyZ3YgPSBbCiAgICAgICAgc3lzLmV4ZWN1dGFibGUsCiAgICAgICAgIi1tIiwK"
    "ICAgICAgICAicGlwIiwKICAgICAgICAiLS1pc29sYXRlZCIsCiAgICAgICAgIi0tZGlzYWJs"
    "ZS1waXAtdmVyc2lvbi1jaGVjayIsCiAgICAgICAgIi0tcHl0aG9uIiwKICAgICAgICBzdHIo"
    "VEFSR0VUX1JPT1QpLAogICAgICAgICJpbnN0YWxsIiwKICAgICAgICAiLS1uby1pbmRleCIs"
    "CiAgICAgICAgIi0tbm8tY2FjaGUtZGlyIiwKICAgICAgICAiLS1uby1kZXBzIiwKICAgICAg"
    "ICAiLS1maW5kLWxpbmtzIiwKICAgICAgICBzdHIod2hlZWxzKSwKICAgICAgICAiLS1yZXF1"
    "aXJlLWhhc2hlcyIsCiAgICAgICAgIi1yIiwKICAgICAgICBzdHIod2hlZWxob3VzZSAvICJy"
    "ZXF1aXJlbWVudHMubG9jay50eHQiKSwKICAgIF0KICAgIGVudmlyb25tZW50ID0geyoqb3Mu"
    "ZW52aXJvbn0KICAgIGVudmlyb25tZW50LnBvcCgiUElQX0lOREVYX1VSTCIsIE5vbmUpCiAg"
    "ICBlbnZpcm9ubWVudC5wb3AoIlBJUF9FWFRSQV9JTkRFWF9VUkwiLCBOb25lKQogICAgZW52"
    "aXJvbm1lbnQudXBkYXRlKAogICAgICAgIHsKICAgICAgICAgICAgIlBJUF9OT19JTkRFWCI6"
    "ICIxIiwKICAgICAgICAgICAgIlBJUF9ESVNBQkxFX1BJUF9WRVJTSU9OX0NIRUNLIjogIjEi"
    "LAogICAgICAgICAgICAiUElQX05PX0NBQ0hFX0RJUiI6ICIxIiwKICAgICAgICB9CiAgICAp"
    "CiAgICBiZWZvcmVfZGlzayA9IGRpc2tfc25hcHNob3QoV09SS19ST09UKQogICAgY29uc3Vt"
    "ZV9hY3Rpb25zKGNvdW50ZXJzLCAicnVudGltZV9pbnN0YWxsX2F0dGVtcHRzIikKICAgIHBy"
    "b2Nlc3MgPSBydW5fYm91bmRlZF9wcm9jZXNzKAogICAgICAgICJvZmZsaW5lX3RhcmdldF9y"
    "dW50aW1lX2luc3RhbGwiLAogICAgICAgIGFyZ3YsCiAgICAgICAgdGltZW91dF9zZWNvbmRz"
    "PTE4MDAuMCwKICAgICAgICBlbnZpcm9ubWVudD1lbnZpcm9ubWVudCwKICAgICAgICBjYXB0"
    "dXJlX3Jvb3Q9U0NSQVRDSF9ST09UIC8gImluc3RhbGxfbG9ncyIsCiAgICApCiAgICByZXBv"
    "cnQgPSB7CiAgICAgICAgKipwcm9jZXNzLAogICAgICAgICJyZXBvcnRfaWQiOiAiYXVyYWdh"
    "dGV3YXktcDUtcDYtZXhhY3QtcnVudGltZS1yZXF1YWxpZmljYXRpb24taW5zdGFsbC12MSIs"
    "CiAgICAgICAgImV4ZWN1dG9yIjogIkJBU0VfUFlUSE9OX1BJUF9WRU5WX1RBUkdFVCIsCiAg"
    "ICAgICAgImZpbmRfbGlua3Nfc2NvcGUiOiAid2hlZWxob3VzZS93aGVlbHMiLAogICAgICAg"
    "ICJyZXF1aXJlbWVudHNfbG9ja19zaGEyNTYiOiBmaWxlX3NoYTI1Nih3aGVlbGhvdXNlIC8g"
    "InJlcXVpcmVtZW50cy5sb2NrLnR4dCIpLAogICAgICAgICJ3aGVlbGhvdXNlX21hbmlmZXN0"
    "X3NoYTI1NiI6IGZpbGVfc2hhMjU2KHdoZWVsaG91c2UgLyAic2hhMjU2X21hbmlmZXN0Lmpz"
    "b24iKSwKICAgICAgICAid29ya2luZ19kaXNrX2JlZm9yZSI6IGJlZm9yZV9kaXNrLAogICAg"
    "ICAgICJ3b3JraW5nX2Rpc2tfYWZ0ZXIiOiBkaXNrX3NuYXBzaG90KFdPUktfUk9PVCksCiAg"
    "ICAgICAgInRhcmdldF9ydW50aW1lX2FmdGVyIjogZGlyZWN0b3J5X3NuYXBzaG90KFRBUkdF"
    "VF9ST09UKSwKICAgICAgICAidGFyZ2V0X3B5dGhvbiI6IHN0cihUQVJHRVRfUFlUSE9OKSwK"
    "ICAgICAgICAibmV0d29ya19hY2Nlc3NfcmVxdWVzdGVkIjogRmFsc2UsCiAgICAgICAgImhp"
    "ZGRlbl9yZXRyeV9jb3VudCI6IDAsCiAgICAgICAgInJvb3RfY2F1c2VfcmV2aWV3X3JlcXVp"
    "cmVkIjogcHJvY2Vzc1sic3RhdHVzIl0gIT0gIlBBU1NFRCIsCiAgICB9CiAgICB3cml0ZV9q"
    "c29uKE9VVFBVVF9ST09UIC8gInJ1bnRpbWVfaW5zdGFsbF9yZXBvcnRfdjEuanNvbiIsIHJl"
    "cG9ydCkKICAgIG91dGNvbWUgPSBwcm9jZXNzWyJwcm9jZXNzX291dGNvbWUiXQogICAgaWYg"
    "b3V0Y29tZSA9PSAiTEFVTkNIX0VSUk9SIjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIk1PREVMX0NPTlNUUlVDVElPTl9GQUlMVVJFIiwKICAgICAg"
    "ICAgICAgIm9mZmxpbmUgdGFyZ2V0LXJ1bnRpbWUgaW5zdGFsbGVyIGNvdWxkIG5vdCBiZSBs"
    "YXVuY2hlZCIsCiAgICAgICAgKQogICAgaWYgb3V0Y29tZSA9PSAiVElNRU9VVCI6CiAgICAg"
    "ICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJNT0RFTF9DT05TVFJV"
    "Q1RJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJvZmZsaW5lIHRhcmdldC1ydW50aW1lIGlu"
    "c3RhbGxhdGlvbiB0aW1lZCBvdXQiLAogICAgICAgICkKICAgIGlmIG91dGNvbWUgPT0gIk5P"
    "TlpFUk9fRVhJVCI6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJNT0RFTF9DT05TVFJVQ1RJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJvZmZsaW5l"
    "IHRhcmdldC1ydW50aW1lIGluc3RhbGxhdGlvbiByZXR1cm5lZCBub256ZXJvIiwKICAgICAg"
    "ICApCiAgICByZXR1cm4gcmVwb3J0CgoKZGVmIHRhcmdldF9saWJyYXJ5X2RpcmVjdG9yaWVz"
    "KCkgLT4gdHVwbGVbUGF0aCwgLi4uXToKICAgIG5hbWVzID0gKAogICAgICAgICJudmlkaWEv"
    "Y3VibGFzL2xpYiIsCiAgICAgICAgIm52aWRpYS9jdWRhX2N1cHRpL2xpYiIsCiAgICAgICAg"
    "Im52aWRpYS9jdWRhX252cnRjL2xpYiIsCiAgICAgICAgIm52aWRpYS9jdWRhX3J1bnRpbWUv"
    "bGliIiwKICAgICAgICAibnZpZGlhL2N1ZG5uL2xpYiIsCiAgICAgICAgIm52aWRpYS9jdWZm"
    "dC9saWIiLAogICAgICAgICJudmlkaWEvY3VmaWxlL2xpYiIsCiAgICAgICAgIm52aWRpYS9j"
    "dXJhbmQvbGliIiwKICAgICAgICAibnZpZGlhL2N1c29sdmVyL2xpYiIsCiAgICAgICAgIm52"
    "aWRpYS9jdXNwYXJzZS9saWIiLAogICAgICAgICJudmlkaWEvbmNjbC9saWIiLAogICAgICAg"
    "ICJudmlkaWEvbnZqaXRsaW5rL2xpYiIsCiAgICAgICAgIm52aWRpYS9udnNobWVtL2xpYiIs"
    "CiAgICApCiAgICByZXN1bHQgPSB0dXBsZShwYXRoIGZvciBuYW1lIGluIG5hbWVzIGlmIChw"
    "YXRoIDo9IFRBUkdFVF9TSVRFIC8gbmFtZSkuaXNfZGlyKCkpCiAgICBpZiBub3QgcmVzdWx0"
    "OgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigidGFyZ2V0IE5WSURJQSBsaWJyYXJ5IGRp"
    "cmVjdG9yaWVzIGFyZSB1bmF2YWlsYWJsZSIpCiAgICByZXR1cm4gcmVzdWx0CgoKQk9PVFNU"
    "UkFQID0gciIiIgppbXBvcnQgcnVucHkKaW1wb3J0IHNpdGUKaW1wb3J0IHN5cwppbXBvcnQg"
    "dHlwZXMKZnJvbSBwYXRobGliIGltcG9ydCBQYXRoCgp0YXJnZXRfc2l0ZSA9IFBhdGgoc3lz"
    "LmFyZ3YucG9wKDEpKS5yZXNvbHZlKCkKbW9kdWxlX25hbWUgPSBzeXMuYXJndi5wb3AoMSkK"
    "CmRlZiBzZW50aW5lbChuYW1lKToKICAgIG1vZHVsZSA9IHR5cGVzLk1vZHVsZVR5cGUobmFt"
    "ZSkKICAgIG1vZHVsZS5fX2ZpbGVfXyA9IGYiPGF1cmFnYXRld2F5LXN1cHByZXNzZWQte25h"
    "bWV9PiIKICAgIHJldHVybiBtb2R1bGUKCnN5cy5tb2R1bGVzWyJzaXRlY3VzdG9taXplIl0g"
    "PSBzZW50aW5lbCgic2l0ZWN1c3RvbWl6ZSIpCnN5cy5tb2R1bGVzWyJ1c2VyY3VzdG9taXpl"
    "Il0gPSBzZW50aW5lbCgidXNlcmN1c3RvbWl6ZSIpCnNpdGUubWFpbigpCgpjbGVhbmVkID0g"
    "W10KZm9yIHZhbHVlIGluIHN5cy5wYXRoOgogICAgaWYgbm90IHZhbHVlOgogICAgICAgIGNs"
    "ZWFuZWQuYXBwZW5kKHZhbHVlKQogICAgICAgIGNvbnRpbnVlCiAgICBwYXRoID0gUGF0aCh2"
    "YWx1ZSkucmVzb2x2ZSgpCiAgICBpc190YXJnZXQgPSBwYXRoID09IHRhcmdldF9zaXRlIG9y"
    "IHRhcmdldF9zaXRlIGluIHBhdGgucGFyZW50cwogICAgaXNfcGFja2FnZV9wYXRoID0gYW55"
    "KAogICAgICAgIHBhcnQgaW4geyJzaXRlLXBhY2thZ2VzIiwgImRpc3QtcGFja2FnZXMifQog"
    "ICAgICAgIGZvciBwYXJ0IGluIHBhdGgucGFydHMKICAgICkKICAgIGlmIGlzX3BhY2thZ2Vf"
    "cGF0aCBhbmQgbm90IGlzX3RhcmdldDoKICAgICAgICBjb250aW51ZQogICAgY2xlYW5lZC5h"
    "cHBlbmQodmFsdWUpCgppZiBzdHIodGFyZ2V0X3NpdGUpIG5vdCBpbiBjbGVhbmVkOgogICAg"
    "Y2xlYW5lZC5pbnNlcnQoMCwgc3RyKHRhcmdldF9zaXRlKSkKc3lzLnBhdGhbOl0gPSBjbGVh"
    "bmVkCnN5cy5hcmd2ID0gW21vZHVsZV9uYW1lLCAqc3lzLmFyZ3ZbMTpdXQpydW5weS5ydW5f"
    "bW9kdWxlKG1vZHVsZV9uYW1lLCBydW5fbmFtZT0iX19tYWluX18iKQoiIiIKCgpkZWYgY29u"
    "dHJvbGxlZF9weXRob24obW9kdWxlOiBzdHIsICphcmdzOiBzdHIpIC0+IGxpc3Rbc3RyXToK"
    "ICAgIGlmIG5vdCBUQVJHRVRfUFlUSE9OLmlzX2ZpbGUoKToKICAgICAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoInRhcmdldCBQeXRob24gZXhlY3V0YWJsZSBpcyB1bmF2YWlsYWJsZSIpCiAg"
    "ICByZXR1cm4gWwogICAgICAgIHN0cihUQVJHRVRfUFlUSE9OKSwKICAgICAgICAiLVMiLAog"
    "ICAgICAgICItYyIsCiAgICAgICAgQk9PVFNUUkFQLAogICAgICAgIHN0cihUQVJHRVRfU0lU"
    "RSksCiAgICAgICAgbW9kdWxlLAogICAgICAgICphcmdzLAogICAgXQoKClBST0hJQklURURf"
    "TElCUkFSWV9QQVRIX01BUktFUlMgPSAoCiAgICAiL3Vzci9sb2NhbC9jdWRhL2xpYjY0L3N0"
    "dWJzIiwKICAgICIvdXNyL2xvY2FsL2N1ZGEvY29tcGF0IiwKICAgICIvY3VkYS9saWI2NC9z"
    "dHVicyIsCiAgICAiL2N1ZGEvY29tcGF0IiwKKQoKCmRlZiBfaXNfcHJvaGliaXRlZF9saWJy"
    "YXJ5X3BhdGgodmFsdWU6IHN0cikgLT4gYm9vbDoKICAgIG5vcm1hbGl6ZWQgPSB2YWx1ZS5y"
    "ZXBsYWNlKCJcXCIsICIvIikucnN0cmlwKCIvIikKICAgIHJldHVybiBhbnkobWFya2VyIGlu"
    "IG5vcm1hbGl6ZWQgZm9yIG1hcmtlciBpbiBQUk9ISUJJVEVEX0xJQlJBUllfUEFUSF9NQVJL"
    "RVJTKQoKCmRlZiBwcm9jZXNzX3RyZWVfZW52aXJvbm1lbnQoCiAgICBncHVfaW5kZXg6IGlu"
    "dCwKICAgIG1vZGVsX2hvbWU6IFBhdGgsCikgLT4gZGljdFtzdHIsIHN0cl06CiAgICBpbmhl"
    "cml0ZWQgPSBbCiAgICAgICAgaXRlbQogICAgICAgIGZvciBpdGVtIGluIG9zLmVudmlyb24u"
    "Z2V0KCJMRF9MSUJSQVJZX1BBVEgiLCAiIikuc3BsaXQob3MucGF0aHNlcCkKICAgICAgICBp"
    "ZiBpdGVtIGFuZCBub3QgX2lzX3Byb2hpYml0ZWRfbGlicmFyeV9wYXRoKGl0ZW0pCiAgICBd"
    "CiAgICB0YXJnZXRfbGlicmFyaWVzID0gW3N0cihwYXRoKSBmb3IgcGF0aCBpbiB0YXJnZXRf"
    "bGlicmFyeV9kaXJlY3RvcmllcygpXQogICAgb3JkZXJlZDogbGlzdFtzdHJdID0gW10KICAg"
    "IGZvciBpdGVtIGluIFsqdGFyZ2V0X2xpYnJhcmllcywgKmluaGVyaXRlZCwgUkVBTF9EUklW"
    "RVJfRElSRUNUT1JZXToKICAgICAgICBpZiBpdGVtIG5vdCBpbiBvcmRlcmVkOgogICAgICAg"
    "ICAgICBvcmRlcmVkLmFwcGVuZChpdGVtKQogICAgZW52aXJvbm1lbnQgPSB7Kipvcy5lbnZp"
    "cm9ufQogICAgZW52aXJvbm1lbnQucG9wKCJQWVRIT05QQVRIIiwgTm9uZSkKICAgIGVudmly"
    "b25tZW50LnBvcCgiTERfUFJFTE9BRCIsIE5vbmUpCiAgICBlbnZpcm9ubWVudC51cGRhdGUo"
    "CiAgICAgICAgewogICAgICAgICAgICAiQ1VEQV9WSVNJQkxFX0RFVklDRVMiOiBzdHIoZ3B1"
    "X2luZGV4KSwKICAgICAgICAgICAgIkhGX0hPTUUiOiBzdHIobW9kZWxfaG9tZSksCiAgICAg"
    "ICAgICAgICJIRl9IVUJfT0ZGTElORSI6ICIxIiwKICAgICAgICAgICAgIlRSQU5TRk9STUVS"
    "U19PRkZMSU5FIjogIjEiLAogICAgICAgICAgICAiUElQX05PX0lOREVYIjogIjEiLAogICAg"
    "ICAgICAgICAiUFlUSE9OTk9VU0VSU0lURSI6ICIxIiwKICAgICAgICAgICAgIlBZVEhPTlBB"
    "VEgiOiBzdHIoVEFSR0VUX1NJVEUpLAogICAgICAgICAgICAiTElCUkFSWV9QQVRIIjogUkVB"
    "TF9EUklWRVJfRElSRUNUT1JZLAogICAgICAgICAgICAiTERGTEFHUyI6ICgiLUwvdXNyL2xv"
    "Y2FsL252aWRpYS9saWI2NCAtV2wsLXJwYXRoLC91c3IvbG9jYWwvbnZpZGlhL2xpYjY0Iiks"
    "CiAgICAgICAgICAgICJMRF9MSUJSQVJZX1BBVEgiOiBvcy5wYXRoc2VwLmpvaW4ob3JkZXJl"
    "ZCksCiAgICAgICAgICAgICJWTExNX0FUVEVOVElPTl9CQUNLRU5EIjogRVhQRUNURURfQkFD"
    "S0VORCwKICAgICAgICB9CiAgICApCiAgICBpZiBhbnkoX2lzX3Byb2hpYml0ZWRfbGlicmFy"
    "eV9wYXRoKGl0ZW0pIGZvciBpdGVtIGluIG9yZGVyZWQpOgogICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiTU9ERUxfQ09OU1RSVUNUSU9OX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAicHJvaGliaXRlZCBDVURBIHN0dWIgcGF0aCBzdXJ2aXZlZCBlbnZp"
    "cm9ubWVudCBjb25zdHJ1Y3Rpb24iLAogICAgICAgICkKICAgIHJldHVybiBlbnZpcm9ubWVu"
    "dAoKCmRlZiBydW50aW1lX2Vudmlyb25tZW50X3JlcG9ydCgKICAgIGVudmlyb25tZW50OiBk"
    "aWN0W3N0ciwgc3RyXSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHBhdGhzID0gZW52"
    "aXJvbm1lbnRbIkxEX0xJQlJBUllfUEFUSCJdLnNwbGl0KG9zLnBhdGhzZXApCiAgICB0YXJn"
    "ZXRfbGlicmFyaWVzID0gW3N0cihwYXRoKSBmb3IgcGF0aCBpbiB0YXJnZXRfbGlicmFyeV9k"
    "aXJlY3RvcmllcygpXQogICAgcmV0dXJuIHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAi"
    "MS4wLjAiLAogICAgICAgICJzdGF0dXMiOiAiUEFTU0VEIiwKICAgICAgICAicHl0aG9ucGF0"
    "aF9leGFjdF90YXJnZXRfc2l0ZSI6IChlbnZpcm9ubWVudC5nZXQoIlBZVEhPTlBBVEgiKSA9"
    "PSBzdHIoVEFSR0VUX1NJVEUpKSwKICAgICAgICAibGRfcHJlbG9hZF9hYnNlbnQiOiAiTERf"
    "UFJFTE9BRCIgbm90IGluIGVudmlyb25tZW50LAogICAgICAgICJwcm9oaWJpdGVkX3N0dWJf"
    "cGF0aF9wcmVzZW50IjogYW55KF9pc19wcm9oaWJpdGVkX2xpYnJhcnlfcGF0aChpdGVtKSBm"
    "b3IgaXRlbSBpbiBwYXRocyksCiAgICAgICAgInRhcmdldF9saWJyYXJ5X3ByZWZpeF9jb3Vu"
    "dCI6IGxlbih0YXJnZXRfbGlicmFyaWVzKSwKICAgICAgICAidGFyZ2V0X2xpYnJhcmllc19w"
    "cmVjZWRlX2luaGVyaXRlZCI6IChwYXRoc1s6IGxlbih0YXJnZXRfbGlicmFyaWVzKV0gPT0g"
    "dGFyZ2V0X2xpYnJhcmllcyksCiAgICAgICAgInJlYWxfZHJpdmVyX2RpcmVjdG9yeV9wcmVz"
    "ZW50IjogUkVBTF9EUklWRVJfRElSRUNUT1JZIGluIHBhdGhzLAogICAgICAgICJsaWJyYXJ5"
    "X3BhdGgiOiBlbnZpcm9ubWVudC5nZXQoIkxJQlJBUllfUEFUSCIpLAogICAgICAgICJhdHRl"
    "bnRpb25fYmFja2VuZCI6IGVudmlyb25tZW50LmdldCgiVkxMTV9BVFRFTlRJT05fQkFDS0VO"
    "RCIpLAogICAgICAgICJyYXdfZW52aXJvbm1lbnRfcmV0YWluZWQiOiBGYWxzZSwKICAgIH0K"
    "CgpkZWYgdmFsaWRhdGVfdGFyZ2V0X3J1bnRpbWUoKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToK"
    "ICAgIHByb2dyYW0gPSByIiIiCmltcG9ydCBpbXBvcnRsaWIKaW1wb3J0IGltcG9ydGxpYi5t"
    "ZXRhZGF0YQppbXBvcnQganNvbgppbXBvcnQgdG9yY2gKaW1wb3J0IHRyYW5zZm9ybWVycwpp"
    "bXBvcnQgdHJpdG9uCmltcG9ydCB2bGxtCgpyZXF1aXJlZF9uYXRpdmUgPSBpbXBvcnRsaWIu"
    "aW1wb3J0X21vZHVsZSgidmxsbS5fQ19zdGFibGVfbGlidG9yY2giKQpwcmludChqc29uLmR1"
    "bXBzKHsKICAgICJ0b3JjaCI6IHRvcmNoLl9fdmVyc2lvbl9fLAogICAgImN1ZGEiOiB0b3Jj"
    "aC52ZXJzaW9uLmN1ZGEsCiAgICAidHJhbnNmb3JtZXJzIjogaW1wb3J0bGliLm1ldGFkYXRh"
    "LnZlcnNpb24oInRyYW5zZm9ybWVycyIpLAogICAgInRyaXRvbiI6IGltcG9ydGxpYi5tZXRh"
    "ZGF0YS52ZXJzaW9uKCJ0cml0b24iKSwKICAgICJ2bGxtX2Rpc3RyaWJ1dGlvbiI6IGltcG9y"
    "dGxpYi5tZXRhZGF0YS52ZXJzaW9uKCJ2bGxtIiksCiAgICAidmxsbV9tb2R1bGUiOiB2bGxt"
    "Ll9fdmVyc2lvbl9fLAogICAgInJlcXVpcmVkX25hdGl2ZV9tb2R1bGUiOiByZXF1aXJlZF9u"
    "YXRpdmUuX19uYW1lX18sCiAgICAicmVxdWlyZWRfbmF0aXZlX21vZHVsZV9maWxlIjogZ2V0"
    "YXR0cihyZXF1aXJlZF9uYXRpdmUsICJfX2ZpbGVfXyIsIE5vbmUpLAogICAgImN1ZGFfYXZh"
    "aWxhYmxlIjogdG9yY2guY3VkYS5pc19hdmFpbGFibGUoKSwKICAgICJkZXZpY2VfbmFtZSI6"
    "IHRvcmNoLmN1ZGEuZ2V0X2RldmljZV9uYW1lKDApLAogICAgImNvbXB1dGVfY2FwYWJpbGl0"
    "eSI6IGxpc3QodG9yY2guY3VkYS5nZXRfZGV2aWNlX2NhcGFiaWxpdHkoMCkpLAp9LCBzZXBh"
    "cmF0b3JzPSgiLCIsICI6IiksIHNvcnRfa2V5cz1UcnVlKSkKIiIiCiAgICBhcmd2ID0gWwog"
    "ICAgICAgIHN0cihUQVJHRVRfUFlUSE9OKSwKICAgICAgICAiLVMiLAogICAgICAgICItYyIs"
    "CiAgICAgICAgKAogICAgICAgICAgICAiaW1wb3J0IHNpdGUsc3lzLHR5cGVzOyIKICAgICAg"
    "ICAgICAgZiJ0YXJnZXQ9e3N0cihUQVJHRVRfU0lURSkhcn07IgogICAgICAgICAgICAic3lz"
    "Lm1vZHVsZXNbJ3NpdGVjdXN0b21pemUnXT10eXBlcy5Nb2R1bGVUeXBlKCdzaXRlY3VzdG9t"
    "aXplJyk7IgogICAgICAgICAgICAic3lzLm1vZHVsZXNbJ3VzZXJjdXN0b21pemUnXT10eXBl"
    "cy5Nb2R1bGVUeXBlKCd1c2VyY3VzdG9taXplJyk7IgogICAgICAgICAgICAic2l0ZS5tYWlu"
    "KCk7IgogICAgICAgICAgICAic3lzLnBhdGhbOl09W3RhcmdldF0rW3AgZm9yIHAgaW4gc3lz"
    "LnBhdGggaWYgIgogICAgICAgICAgICAiKCdzaXRlLXBhY2thZ2VzJyBub3QgaW4gcCBhbmQg"
    "J2Rpc3QtcGFja2FnZXMnIG5vdCBpbiBwKSBvciBwPT10YXJnZXRdOyIgKyBwcm9ncmFtCiAg"
    "ICAgICAgKSwKICAgIF0KICAgIHJlc3VsdCA9IHN1YnByb2Nlc3MucnVuKAogICAgICAgIGFy"
    "Z3YsCiAgICAgICAgY2hlY2s9RmFsc2UsCiAgICAgICAgY2FwdHVyZV9vdXRwdXQ9VHJ1ZSwK"
    "ICAgICAgICB0ZXh0PVRydWUsCiAgICAgICAgdGltZW91dD0xODAsCiAgICAgICAgZW52PXBy"
    "b2Nlc3NfdHJlZV9lbnZpcm9ubWVudCgwLCBPVVRQVVRfUk9PVCAvICJtb2RlbF9ob21lIiks"
    "CiAgICApCiAgICBpZiByZXN1bHQucmV0dXJuY29kZSAhPSAwOgogICAgICAgIHJhaXNlIERp"
    "YWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiTU9ERUxfQ09OU1RSVUNUSU9OX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAidGFyZ2V0IHJ1bnRpbWUgaWRlbnRpdHkgdmFsaWRhdGlvbiBm"
    "YWlsZWQiLAogICAgICAgICkKICAgIHBheWxvYWQgPSBqc29uLmxvYWRzKHJlc3VsdC5zdGRv"
    "dXQuc3RyaXAoKS5zcGxpdGxpbmVzKClbLTFdKQogICAgZXhwZWN0ZWQgPSB7CiAgICAgICAg"
    "InRvcmNoIjogRVhQRUNURURfVE9SQ0hfVkVSU0lPTiwKICAgICAgICAiY3VkYSI6IEVYUEVD"
    "VEVEX1RPUkNIX0NVREEsCiAgICAgICAgInRyYW5zZm9ybWVycyI6IEVYUEVDVEVEX1RSQU5T"
    "Rk9STUVSU19WRVJTSU9OLAogICAgICAgICJ0cml0b24iOiBFWFBFQ1RFRF9UUklUT05fVkVS"
    "U0lPTiwKICAgICAgICAidmxsbV9kaXN0cmlidXRpb24iOiBFWFBFQ1RFRF9WTExNX0RJU1RS"
    "SUJVVElPTl9WRVJTSU9OLAogICAgICAgICJ2bGxtX21vZHVsZSI6IEVYUEVDVEVEX1ZMTE1f"
    "TU9EVUxFX1ZFUlNJT04sCiAgICAgICAgInJlcXVpcmVkX25hdGl2ZV9tb2R1bGUiOiBSRVFV"
    "SVJFRF9OQVRJVkVfTU9EVUxFLAogICAgICAgICJjdWRhX2F2YWlsYWJsZSI6IFRydWUsCiAg"
    "ICAgICAgImRldmljZV9uYW1lIjogRVhQRUNURURfR1BVX05BTUUsCiAgICAgICAgImNvbXB1"
    "dGVfY2FwYWJpbGl0eSI6IFs3LCA1XSwKICAgIH0KICAgIGZvciBrZXksIGV4cGVjdGVkX3Zh"
    "bHVlIGluIGV4cGVjdGVkLml0ZW1zKCk6CiAgICAgICAgaWYgcGF5bG9hZC5nZXQoa2V5KSAh"
    "PSBleHBlY3RlZF92YWx1ZToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICAgICAiTU9ERUxfQ09OU1RSVUNUSU9OX0ZBSUxVUkUiLAogICAgICAg"
    "ICAgICAgICAgZiJ0YXJnZXQgcnVudGltZSBpZGVudGl0eSBkcmlmdGVkOiB7a2V5fSIsCiAg"
    "ICAgICAgICAgICkKICAgIG5hdGl2ZV9maWxlID0gcGF5bG9hZC5nZXQoInJlcXVpcmVkX25h"
    "dGl2ZV9tb2R1bGVfZmlsZSIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShuYXRpdmVfZmlsZSwg"
    "c3RyKSBvciBub3QgX3BhdGhfd2l0aGluX3RhcmdldChuYXRpdmVfZmlsZSk6CiAgICAgICAg"
    "cmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJNT0RFTF9DT05TVFJVQ1RJ"
    "T05fRkFJTFVSRSIsCiAgICAgICAgICAgICJyZXF1aXJlZCB2TExNIG5hdGl2ZSBtb2R1bGUg"
    "cmVzb2x2ZWQgb3V0c2lkZSB0YXJnZXQgcnVudGltZSIsCiAgICAgICAgKQogICAgcmV0dXJu"
    "IHBheWxvYWQgICMgdHlwZTogaWdub3JlW25vLWFueS1yZXR1cm5dCgoKZGVmIF9wYXRoX3dp"
    "dGhpbl90YXJnZXQodmFsdWU6IHN0cikgLT4gYm9vbDoKICAgIHBhdGggPSBQYXRoKHZhbHVl"
    "KS5yZXNvbHZlKCkKICAgIHRhcmdldCA9IFRBUkdFVF9TSVRFLnJlc29sdmUoKQogICAgcmV0"
    "dXJuIHBhdGggPT0gdGFyZ2V0IG9yIHRhcmdldCBpbiBwYXRoLnBhcmVudHMKCgpkZWYgdmFs"
    "aWRhdGVfaW1wb3J0X2Nsb3N1cmVfcGF5bG9hZCgKICAgIHBheWxvYWQ6IGRpY3Rbc3RyLCBv"
    "YmplY3RdLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgcGFyZW50X2V4ZWN1dGFibGUg"
    "PSBwYXlsb2FkLmdldCgicGFyZW50X3B5dGhvbl9leGVjdXRhYmxlIikKICAgIHBhcmVudF9w"
    "eXRob25wYXRoID0gcGF5bG9hZC5nZXQoInBhcmVudF9weXRob25wYXRoIikKICAgIGNoaWxk"
    "X3JldHVybmNvZGUgPSBwYXlsb2FkLmdldCgiY2hpbGRfcmV0dXJuY29kZSIpCiAgICBjaGls"
    "ZCA9IHBheWxvYWQuZ2V0KCJjaGlsZCIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShwYXJlbnRf"
    "ZXhlY3V0YWJsZSwgc3RyKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgIlAzX1A2X1JVTlRJTUVfUFJPQ0VTU19UUkVFX0lNUE9SVF9DTE9TVVJFX0ZB"
    "SUxFRCIsCiAgICAgICAgICAgICJpbXBvcnQtY2xvc3VyZSBwYXJlbnQgZXhlY3V0YWJsZSBp"
    "cyBpbnZhbGlkIiwKICAgICAgICApCiAgICBpZiBwYXJlbnRfcHl0aG9ucGF0aCAhPSBzdHIo"
    "VEFSR0VUX1NJVEUpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiUDNfUDZfUlVOVElNRV9QUk9DRVNTX1RSRUVfSU1QT1JUX0NMT1NVUkVfRkFJTEVE"
    "IiwKICAgICAgICAgICAgImltcG9ydC1jbG9zdXJlIHBhcmVudCBQWVRIT05QQVRIIGRyaWZ0"
    "ZWQiLAogICAgICAgICkKICAgIGlmIGNoaWxkX3JldHVybmNvZGUgIT0gMCBvciBub3QgaXNp"
    "bnN0YW5jZShjaGlsZCwgZGljdCk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJQM19QNl9SVU5USU1FX1BST0NFU1NfVFJFRV9JTVBPUlRfQ0xPU1VS"
    "RV9GQUlMRUQiLAogICAgICAgICAgICAibmVzdGVkIGltcG9ydC1jbG9zdXJlIGNoaWxkIGZh"
    "aWxlZCIsCiAgICAgICAgKQogICAgY2hpbGRfZXhlY3V0YWJsZSA9IGNoaWxkLmdldCgicHl0"
    "aG9uX2V4ZWN1dGFibGUiKQogICAgY2hpbGRfcHl0aG9ucGF0aCA9IGNoaWxkLmdldCgicHl0"
    "aG9ucGF0aCIpCiAgICBtb2R1bGVzID0gY2hpbGQuZ2V0KCJtb2R1bGVzIikKICAgIGlmIG5v"
    "dCBpc2luc3RhbmNlKGNoaWxkX2V4ZWN1dGFibGUsIHN0cik6CiAgICAgICAgcmFpc2UgRGlh"
    "Z25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJQM19QNl9SVU5USU1FX1BST0NFU1NfVFJF"
    "RV9JTVBPUlRfQ0xPU1VSRV9GQUlMRUQiLAogICAgICAgICAgICAiaW1wb3J0LWNsb3N1cmUg"
    "Y2hpbGQgZXhlY3V0YWJsZSBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICBpZiBjaGlsZF9w"
    "eXRob25wYXRoICE9IHN0cihUQVJHRVRfU0lURSk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICJQM19QNl9SVU5USU1FX1BST0NFU1NfVFJFRV9JTVBP"
    "UlRfQ0xPU1VSRV9GQUlMRUQiLAogICAgICAgICAgICAiaW1wb3J0LWNsb3N1cmUgY2hpbGQg"
    "UFlUSE9OUEFUSCBkcmlmdGVkIiwKICAgICAgICApCiAgICBpZiBub3QgaXNpbnN0YW5jZSht"
    "b2R1bGVzLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIlAzX1A2X1JVTlRJTUVfUFJPQ0VTU19UUkVFX0lNUE9SVF9DTE9TVVJFX0ZBSUxF"
    "RCIsCiAgICAgICAgICAgICJpbXBvcnQtY2xvc3VyZSBtb2R1bGUgaW52ZW50b3J5IGlzIGlu"
    "dmFsaWQiLAogICAgICAgICkKICAgIGlmIHNldChtb2R1bGVzKSAhPSBzZXQoSU1QT1JUX0NM"
    "T1NVUkVfTU9EVUxFUyk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAg"
    "ICAgICAgICJQM19QNl9SVU5USU1FX1BST0NFU1NfVFJFRV9JTVBPUlRfQ0xPU1VSRV9GQUlM"
    "RUQiLAogICAgICAgICAgICAiaW1wb3J0LWNsb3N1cmUgbW9kdWxlIGludmVudG9yeSBkcmlm"
    "dGVkIiwKICAgICAgICApCiAgICBvcmlnaW5zOiBkaWN0W3N0ciwgc3RyXSA9IHt9CiAgICBm"
    "b3IgbmFtZSBpbiBJTVBPUlRfQ0xPU1VSRV9NT0RVTEVTOgogICAgICAgIGl0ZW0gPSBtb2R1"
    "bGVzLmdldChuYW1lKQogICAgICAgIGlmIG5vdCBpc2luc3RhbmNlKGl0ZW0sIGRpY3QpOgog"
    "ICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgICAgICJQ"
    "M19QNl9SVU5USU1FX1BST0NFU1NfVFJFRV9JTVBPUlRfQ0xPU1VSRV9GQUlMRUQiLAogICAg"
    "ICAgICAgICAgICAgZiJpbXBvcnQtY2xvc3VyZSBtb2R1bGUgcmVjb3JkIGlzIGludmFsaWQ6"
    "IHtuYW1lfSIsCiAgICAgICAgICAgICkKICAgICAgICBvcmlnaW4gPSBpdGVtLmdldCgib3Jp"
    "Z2luIikKICAgICAgICBpZiBub3QgaXNpbnN0YW5jZShvcmlnaW4sIHN0cikgb3Igbm90IF9w"
    "YXRoX3dpdGhpbl90YXJnZXQob3JpZ2luKToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiUDNfUDZfUlVOVElNRV9QUk9DRVNTX1RSRUVf"
    "SU1QT1JUX0NMT1NVUkVfRkFJTEVEIiwKICAgICAgICAgICAgICAgIGYiY3JpdGljYWwgbW9k"
    "dWxlIHJlc29sdmVkIG91dHNpZGUgdGFyZ2V0IHNpdGU6IHtuYW1lfSIsCiAgICAgICAgICAg"
    "ICkKICAgICAgICBvcmlnaW5zW25hbWVdID0gb3JpZ2luCiAgICByZXR1cm4gewogICAgICAg"
    "ICJwYXJlbnRfcHl0aG9uX2V4ZWN1dGFibGUiOiBwYXJlbnRfZXhlY3V0YWJsZSwKICAgICAg"
    "ICAicGFyZW50X3B5dGhvbnBhdGgiOiBwYXJlbnRfcHl0aG9ucGF0aCwKICAgICAgICAiY2hp"
    "bGRfcHl0aG9uX2V4ZWN1dGFibGUiOiBjaGlsZF9leGVjdXRhYmxlLAogICAgICAgICJjaGls"
    "ZF9weXRob25wYXRoIjogY2hpbGRfcHl0aG9ucGF0aCwKICAgICAgICAiY3JpdGljYWxfbW9k"
    "dWxlX29yaWdpbnMiOiBvcmlnaW5zLAogICAgICAgICJhbGxfY3JpdGljYWxfb3JpZ2luc193"
    "aXRoaW5fdGFyZ2V0X3NpdGUiOiBUcnVlLAogICAgfQoKCmRlZiB2YWxpZGF0ZV9wcm9jZXNz"
    "X3RyZWVfaW1wb3J0X2Nsb3N1cmUoCiAgICBjb3VudGVyczogZGljdFtzdHIsIGludF0sCikg"
    "LT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBpZiBub3QgRVhQRUNURURfQ0hJTERfUFlUSE9O"
    "LmlzX2ZpbGUoKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAg"
    "ICAgIlAzX1A2X1JVTlRJTUVfUFJPQ0VTU19UUkVFX0lNUE9SVF9DTE9TVVJFX0ZBSUxFRCIs"
    "CiAgICAgICAgICAgICJleHBlY3RlZCBuZXN0ZWQgUHl0aG9uIGV4ZWN1dGFibGUgaXMgdW5h"
    "dmFpbGFibGUiLAogICAgICAgICkKICAgIGNoaWxkX3Byb2dyYW0gPSByIiIiCmltcG9ydCBp"
    "bXBvcnRsaWIKaW1wb3J0IGpzb24KaW1wb3J0IG9zCmltcG9ydCBzeXMKCm1vZHVsZV9uYW1l"
    "cyA9ICgKICAgICJ2bGxtIiwKICAgICJ0b3JjaCIsCiAgICAidHJpdG9uIiwKICAgICJ0cmFu"
    "c2Zvcm1lcnMiLAogICAgInZsbG0ubW9kZWxfZXhlY3V0b3IubW9kZWxzLnJlZ2lzdHJ5IiwK"
    "KQptb2R1bGVzID0ge30KZm9yIG5hbWUgaW4gbW9kdWxlX25hbWVzOgogICAgbW9kdWxlID0g"
    "aW1wb3J0bGliLmltcG9ydF9tb2R1bGUobmFtZSkKICAgIG9yaWdpbiA9IGdldGF0dHIobW9k"
    "dWxlLCAiX19maWxlX18iLCBOb25lKQogICAgaWYgbm90IGlzaW5zdGFuY2Uob3JpZ2luLCBz"
    "dHIpOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcihmIm1vZHVsZSBvcmlnaW4gaXMgdW5h"
    "dmFpbGFibGU6IHtuYW1lfSIpCiAgICBtb2R1bGVzW25hbWVdID0geyJvcmlnaW4iOiBvcmln"
    "aW59CnByaW50KGpzb24uZHVtcHMoewogICAgInB5dGhvbl9leGVjdXRhYmxlIjogc3lzLmV4"
    "ZWN1dGFibGUsCiAgICAicHl0aG9ucGF0aCI6IG9zLmVudmlyb24uZ2V0KCJQWVRIT05QQVRI"
    "IiksCiAgICAibW9kdWxlcyI6IG1vZHVsZXMsCn0sIHNlcGFyYXRvcnM9KCIsIiwgIjoiKSwg"
    "c29ydF9rZXlzPVRydWUpKQoiIiIKICAgIHBhcmVudF9wcm9ncmFtID0gciIiIgppbXBvcnQg"
    "anNvbgppbXBvcnQgb3MKaW1wb3J0IHN1YnByb2Nlc3MKaW1wb3J0IHN5cwoKY2hpbGRfcHJv"
    "Z3JhbSA9IHN5cy5hcmd2WzFdCmNoaWxkX2V4ZWN1dGFibGUgPSBzeXMuYXJndlsyXQpjb21w"
    "bGV0ZWQgPSBzdWJwcm9jZXNzLnJ1bigKICAgIFtjaGlsZF9leGVjdXRhYmxlLCAiLWMiLCBj"
    "aGlsZF9wcm9ncmFtXSwKICAgIGNoZWNrPUZhbHNlLAogICAgY2FwdHVyZV9vdXRwdXQ9VHJ1"
    "ZSwKICAgIHRleHQ9VHJ1ZSwKICAgIHRpbWVvdXQ9OTAsCikKY2hpbGRfcGF5bG9hZCA9IE5v"
    "bmUKaWYgY29tcGxldGVkLnN0ZG91dC5zdHJpcCgpOgogICAgY2hpbGRfcGF5bG9hZCA9IGpz"
    "b24ubG9hZHMoY29tcGxldGVkLnN0ZG91dC5zdHJpcCgpLnNwbGl0bGluZXMoKVstMV0pCnBy"
    "aW50KGpzb24uZHVtcHMoewogICAgInBhcmVudF9weXRob25fZXhlY3V0YWJsZSI6IHN5cy5l"
    "eGVjdXRhYmxlLAogICAgInBhcmVudF9weXRob25wYXRoIjogb3MuZW52aXJvbi5nZXQoIlBZ"
    "VEhPTlBBVEgiKSwKICAgICJjaGlsZF9yZXR1cm5jb2RlIjogY29tcGxldGVkLnJldHVybmNv"
    "ZGUsCiAgICAiY2hpbGQiOiBjaGlsZF9wYXlsb2FkLAp9LCBzZXBhcmF0b3JzPSgiLCIsICI6"
    "IiksIHNvcnRfa2V5cz1UcnVlKSkKaWYgY29tcGxldGVkLnN0ZGVycjoKICAgIHByaW50KGNv"
    "bXBsZXRlZC5zdGRlcnJbLTE2MDAwOl0sIGZpbGU9c3lzLnN0ZGVycikKcmFpc2UgU3lzdGVt"
    "RXhpdCgwIGlmIGNvbXBsZXRlZC5yZXR1cm5jb2RlID09IDAgZWxzZSAzKQoiIiIKICAgIGVu"
    "dmlyb25tZW50ID0gcHJvY2Vzc190cmVlX2Vudmlyb25tZW50KAogICAgICAgIDAsCiAgICAg"
    "ICAgU0NSQVRDSF9ST09UIC8gImltcG9ydF9jbG9zdXJlX21vZGVsX2hvbWUiLAogICAgKQog"
    "ICAgY2FwdHVyZV9yb290ID0gU0NSQVRDSF9ST09UIC8gImltcG9ydF9jbG9zdXJlX2xvZ3Mi"
    "CiAgICBjb25zdW1lX2FjdGlvbnMoY291bnRlcnMsICJydW50aW1lX2ltcG9ydF9jbG9zdXJl"
    "X3Byb2JlcyIpCiAgICBwcm9jZXNzID0gcnVuX2JvdW5kZWRfcHJvY2VzcygKICAgICAgICAi"
    "cnVudGltZV9wcm9jZXNzX3RyZWVfaW1wb3J0X2Nsb3N1cmUiLAogICAgICAgIFsKICAgICAg"
    "ICAgICAgc3RyKEVYUEVDVEVEX0NISUxEX1BZVEhPTiksCiAgICAgICAgICAgICItYyIsCiAg"
    "ICAgICAgICAgIHBhcmVudF9wcm9ncmFtLAogICAgICAgICAgICBjaGlsZF9wcm9ncmFtLAog"
    "ICAgICAgICAgICBzdHIoRVhQRUNURURfQ0hJTERfUFlUSE9OKSwKICAgICAgICBdLAogICAg"
    "ICAgIHRpbWVvdXRfc2Vjb25kcz0xMjAuMCwKICAgICAgICBlbnZpcm9ubWVudD1lbnZpcm9u"
    "bWVudCwKICAgICAgICBjYXB0dXJlX3Jvb3Q9Y2FwdHVyZV9yb290LAogICAgKQogICAgc3Rk"
    "b3V0X3BhdGggPSBjYXB0dXJlX3Jvb3QgLyAicnVudGltZV9wcm9jZXNzX3RyZWVfaW1wb3J0"
    "X2Nsb3N1cmUuc3Rkb3V0LmxvZyIKICAgIHBheWxvYWQ6IGRpY3Rbc3RyLCBvYmplY3RdIHwg"
    "Tm9uZSA9IE5vbmUKICAgIGlmIHN0ZG91dF9wYXRoLmlzX2ZpbGUoKToKICAgICAgICBsaW5l"
    "cyA9IHR1cGxlKAogICAgICAgICAgICBsaW5lCiAgICAgICAgICAgIGZvciBsaW5lIGluIHN0"
    "ZG91dF9wYXRoLnJlYWRfdGV4dCgKICAgICAgICAgICAgICAgIGVuY29kaW5nPSJ1dGYtOCIs"
    "CiAgICAgICAgICAgICAgICBlcnJvcnM9InJlcGxhY2UiLAogICAgICAgICAgICApLnNwbGl0"
    "bGluZXMoKQogICAgICAgICAgICBpZiBsaW5lLnN0cmlwKCkKICAgICAgICApCiAgICAgICAg"
    "aWYgbGluZXM6CiAgICAgICAgICAgIHJhdyA9IGpzb24ubG9hZHMobGluZXNbLTFdKQogICAg"
    "ICAgICAgICBpZiBpc2luc3RhbmNlKHJhdywgZGljdCk6CiAgICAgICAgICAgICAgICBwYXls"
    "b2FkID0gcmF3CiAgICByZXBvcnQgPSB7CiAgICAgICAgKipwcm9jZXNzLAogICAgICAgICJy"
    "ZXBvcnRfaWQiOiAoImF1cmFnYXRld2F5LXA1LXA2LWV4YWN0LXJ1bnRpbWUtcmVxdWFsaWZp"
    "Y2F0aW9uLXJ1bnRpbWUtaW1wb3J0LWNsb3N1cmUtdjEiKSwKICAgICAgICAidGFyZ2V0X3Np"
    "dGUiOiBzdHIoVEFSR0VUX1NJVEUpLAogICAgICAgICJweXRob25wYXRoX2V4YWN0X3Rhcmdl"
    "dF9zaXRlIjogKGVudmlyb25tZW50LmdldCgiUFlUSE9OUEFUSCIpID09IHN0cihUQVJHRVRf"
    "U0lURSkpLAogICAgICAgICJpbmhlcml0ZWRfcHl0aG9ucGF0aF9yZXBsYWNlZCI6IFRydWUs"
    "CiAgICAgICAgIm5lc3RlZF9pbnRlcnByZXRlcl9kZXB0aCI6IDIsCiAgICAgICAgImNyaXRp"
    "Y2FsX21vZHVsZXMiOiBsaXN0KElNUE9SVF9DTE9TVVJFX01PRFVMRVMpLAogICAgICAgICJt"
    "b2RlbF9sb2Fkc19jb25zdW1lZCI6IDAsCiAgICAgICAgIndvcmtlcl9zdGFydHNfY29uc3Vt"
    "ZWQiOiAwLAogICAgICAgICJuZXR3b3JrX2FjY2Vzc19yZXF1ZXN0ZWQiOiBGYWxzZSwKICAg"
    "ICAgICAiaGlkZGVuX3JldHJ5X2NvdW50IjogMCwKICAgIH0KICAgIGlmIHByb2Nlc3NbInN0"
    "YXR1cyJdICE9ICJQQVNTRUQiIG9yIHBheWxvYWQgaXMgTm9uZToKICAgICAgICB3cml0ZV9q"
    "c29uKAogICAgICAgICAgICBPVVRQVVRfUk9PVCAvICJydW50aW1lX2ltcG9ydF9jbG9zdXJl"
    "X3JlcG9ydF92MS5qc29uIiwKICAgICAgICAgICAgcmVwb3J0LAogICAgICAgICkKICAgICAg"
    "ICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlAzX1A2X1JVTlRJTUVf"
    "UFJPQ0VTU19UUkVFX0lNUE9SVF9DTE9TVVJFX0ZBSUxFRCIsCiAgICAgICAgICAgICJuZXN0"
    "ZWQgdGFyZ2V0LXJ1bnRpbWUgaW1wb3J0LWNsb3N1cmUgcHJvYmUgZmFpbGVkIiwKICAgICAg"
    "ICApCiAgICB2YWxpZGF0ZWQgPSB2YWxpZGF0ZV9pbXBvcnRfY2xvc3VyZV9wYXlsb2FkKHBh"
    "eWxvYWQpCiAgICByZXBvcnQudXBkYXRlKHZhbGlkYXRlZCkKICAgIHJlcG9ydFsic3RhdHVz"
    "Il0gPSAiUEFTU0VEIgogICAgcmVwb3J0WyJkZWNpc2lvbiJdID0gIlBST0NFU1NfVFJFRV9J"
    "TVBPUlRfQ0xPU1VSRV9QQVNTRUQiCiAgICB3cml0ZV9qc29uKAogICAgICAgIE9VVFBVVF9S"
    "T09UIC8gInJ1bnRpbWVfaW1wb3J0X2Nsb3N1cmVfcmVwb3J0X3YxLmpzb24iLAogICAgICAg"
    "IHJlcG9ydCwKICAgICkKICAgIHJldHVybiByZXBvcnQKCgpkZWYgYmFja2VuZF9tYXJrZXJf"
    "ZXZpZGVuY2VfZnJvbV90ZXh0cygKICAgIHN0ZG91dF90ZXh0OiBzdHIsCiAgICBzdGRlcnJf"
    "dGV4dDogc3RyLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZToKICAgIG1hdGNoZXM6"
    "IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dID0gW10KICAgIGZvciBzdHJlYW1fbmFtZSwgcGF5"
    "bG9hZCBpbiAoCiAgICAgICAgKCJzdGRvdXQiLCBzdGRvdXRfdGV4dCksCiAgICAgICAgKCJz"
    "dGRlcnIiLCBzdGRlcnJfdGV4dCksCiAgICApOgogICAgICAgIGZvciBsaW5lX251bWJlciwg"
    "cmF3X2xpbmUgaW4gZW51bWVyYXRlKHBheWxvYWQuc3BsaXRsaW5lcygpLCBzdGFydD0xKToK"
    "ICAgICAgICAgICAgbm9ybWFsaXplZCA9IHJhd19saW5lLnN0cmlwKCkKICAgICAgICAgICAg"
    "aWYgRVhQRUNURURfQkFDS0VORF9MT0dfTUFSS0VSIG5vdCBpbiBub3JtYWxpemVkOgogICAg"
    "ICAgICAgICAgICAgY29udGludWUKICAgICAgICAgICAgbG93ZXJlZCA9IG5vcm1hbGl6ZWQu"
    "bG93ZXIoKQogICAgICAgICAgICBpZiBhbnkodG9rZW4gaW4gbG93ZXJlZCBmb3IgdG9rZW4g"
    "aW4gQkFDS0VORF9NQVJLRVJfRVhDTFVERURfVE9LRU5TKToKICAgICAgICAgICAgICAgIGNv"
    "bnRpbnVlCiAgICAgICAgICAgIGlmIEJBQ0tFTkRfTUFSS0VSX0xPR19QUkVGSVhfUEFUVEVS"
    "Ti5mdWxsbWF0Y2gobm9ybWFsaXplZCkgaXMgTm9uZToKICAgICAgICAgICAgICAgIGNvbnRp"
    "bnVlCiAgICAgICAgICAgIG1hdGNoZXMuYXBwZW5kKAogICAgICAgICAgICAgICAgewogICAg"
    "ICAgICAgICAgICAgICAgICJzdHJlYW0iOiBzdHJlYW1fbmFtZSwKICAgICAgICAgICAgICAg"
    "ICAgICAibGluZV9udW1iZXIiOiBsaW5lX251bWJlciwKICAgICAgICAgICAgICAgICAgICAi"
    "bWFya2VyIjogRVhQRUNURURfQkFDS0VORF9MT0dfTUFSS0VSLAogICAgICAgICAgICAgICAg"
    "ICAgICJub3JtYWxpemVkX2xpbmVfc2hhMjU2Ijogc2hhMjU2X3RleHQobm9ybWFsaXplZCks"
    "CiAgICAgICAgICAgICAgICAgICAgIm5vcm1hbGl6ZWRfbGluZV9sZW5ndGgiOiBsZW4obm9y"
    "bWFsaXplZCksCiAgICAgICAgICAgICAgICAgICAgImxpbmVfbG9jYWxfbWF0Y2giOiBUcnVl"
    "LAogICAgICAgICAgICAgICAgICAgICJjbGlfZWNob19yZWplY3RlZCI6IFRydWUsCiAgICAg"
    "ICAgICAgICAgICB9CiAgICAgICAgICAgICkKICAgIGlmIGxlbihtYXRjaGVzKSA+IDE6CiAg"
    "ICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJiYWNrZW5kIG1hcmtlciBldmlkZW5jZSBpcyBh"
    "bWJpZ3VvdXMiKQogICAgcmV0dXJuIE5vbmUgaWYgbm90IG1hdGNoZXMgZWxzZSBtYXRjaGVz"
    "WzBdCgoKY2xhc3MgV29ya2VyOgogICAgZGVmIF9faW5pdF9fKAogICAgICAgIHNlbGYsCiAg"
    "ICAgICAgd29ya2VyX2lkOiBzdHIsCiAgICAgICAgZ3B1X2luZGV4OiBpbnQsCiAgICAgICAg"
    "cG9ydDogaW50LAogICAgICAgIG1vZGVsX2hvbWU6IFBhdGgsCiAgICAgICAgc25hcHNob3Q6"
    "IFBhdGgsCiAgICAgICAgZ2VuZXJhdGlvbjogaW50ID0gMSwKICAgICkgLT4gTm9uZToKICAg"
    "ICAgICBzZWxmLndvcmtlcl9pZCA9IHdvcmtlcl9pZAogICAgICAgIHNlbGYuZ3B1X2luZGV4"
    "ID0gZ3B1X2luZGV4CiAgICAgICAgc2VsZi5wb3J0ID0gcG9ydAogICAgICAgIHNlbGYubW9k"
    "ZWxfaG9tZSA9IG1vZGVsX2hvbWUKICAgICAgICBzZWxmLnNuYXBzaG90ID0gc25hcHNob3QK"
    "ICAgICAgICBzZWxmLmdlbmVyYXRpb24gPSBnZW5lcmF0aW9uCiAgICAgICAgc2VsZi5wcm9j"
    "ZXNzOiBzdWJwcm9jZXNzLlBvcGVuW2J5dGVzXSB8IE5vbmUgPSBOb25lCiAgICAgICAgc2Vs"
    "Zi5iYWNrZW5kX21hcmtlcl9wb2xsX2NvdW50ID0gMAogICAgICAgIHNlbGYuYmFja2VuZF9l"
    "dmlkZW5jZTogZGljdFtzdHIsIG9iamVjdF0gfCBOb25lID0gTm9uZQogICAgICAgIHNlbGYu"
    "c3RhcnRlZF9hdDogc3RyIHwgTm9uZSA9IE5vbmUKICAgICAgICBzZWxmLnBhcmVudF9waWQ6"
    "IGludCB8IE5vbmUgPSBOb25lCiAgICAgICAgc2VsZi5wcm9jZXNzX3N0YXJ0X3RpY2tzOiBp"
    "bnQgfCBOb25lID0gTm9uZQogICAgICAgIHNlbGYuZ3B1X2lkZW50aXR5OiBkaWN0W3N0ciwg"
    "b2JqZWN0XSB8IE5vbmUgPSBOb25lCiAgICAgICAgc2VsZi5tZW1vcnlfYmVmb3JlX3N0YXJ0"
    "X21pYjogaW50IHwgTm9uZSA9IE5vbmUKICAgICAgICBzZWxmLnRlYXJkb3duX3JlcG9ydDog"
    "ZGljdFtzdHIsIG9iamVjdF0gfCBOb25lID0gTm9uZQogICAgICAgIHNlbGYuc3Rkb3V0ID0g"
    "Qm91bmRlZENhcHR1cmUoTE9HX1JPT1QgLyBmInt3b3JrZXJfaWR9LWd7Z2VuZXJhdGlvbn0u"
    "c3Rkb3V0LmxvZyIpCiAgICAgICAgc2VsZi5zdGRlcnIgPSBCb3VuZGVkQ2FwdHVyZShMT0df"
    "Uk9PVCAvIGYie3dvcmtlcl9pZH0tZ3tnZW5lcmF0aW9ufS5zdGRlcnIubG9nIikKICAgICAg"
    "ICBzZWxmLmFyZ3YgPSBjb250cm9sbGVkX3B5dGhvbigKICAgICAgICAgICAgInZsbG0uZW50"
    "cnlwb2ludHMub3BlbmFpLmFwaV9zZXJ2ZXIiLAogICAgICAgICAgICAiLS1tb2RlbCIsCiAg"
    "ICAgICAgICAgIE1PREVMX1JFUE9TSVRPUlksCiAgICAgICAgICAgICItLXJldmlzaW9uIiwK"
    "ICAgICAgICAgICAgTU9ERUxfUkVWSVNJT04sCiAgICAgICAgICAgICItLXRva2VuaXplciIs"
    "CiAgICAgICAgICAgIE1PREVMX1JFUE9TSVRPUlksCiAgICAgICAgICAgICItLXRva2VuaXpl"
    "ci1yZXZpc2lvbiIsCiAgICAgICAgICAgIE1PREVMX1JFVklTSU9OLAogICAgICAgICAgICAi"
    "LS1zZXJ2ZWQtbW9kZWwtbmFtZSIsCiAgICAgICAgICAgIFNFUlZFRF9NT0RFTF9OQU1FLAog"
    "ICAgICAgICAgICAiLS1ob3N0IiwKICAgICAgICAgICAgIjEyNy4wLjAuMSIsCiAgICAgICAg"
    "ICAgICItLXBvcnQiLAogICAgICAgICAgICBzdHIocG9ydCksCiAgICAgICAgICAgICItLWR0"
    "eXBlIiwKICAgICAgICAgICAgImF1dG8iLAogICAgICAgICAgICAiLS1tYXgtbW9kZWwtbGVu"
    "IiwKICAgICAgICAgICAgIjQwOTYiLAogICAgICAgICAgICAiLS1ncHUtbWVtb3J5LXV0aWxp"
    "emF0aW9uIiwKICAgICAgICAgICAgIjAuODUiLAogICAgICAgICAgICAiLS1tYXgtbnVtLXNl"
    "cXMiLAogICAgICAgICAgICAiOCIsCiAgICAgICAgICAgICItLWVuYWJsZS1wcmVmaXgtY2Fj"
    "aGluZyIsCiAgICAgICAgICAgICItLWJsb2NrLXNpemUiLAogICAgICAgICAgICBzdHIoQ0FD"
    "SEVfQkxPQ0tfU0laRSksCiAgICAgICAgICAgICItLWF0dGVudGlvbi1iYWNrZW5kIiwKICAg"
    "ICAgICAgICAgRVhQRUNURURfQkFDS0VORCwKICAgICAgICAgICAgIi0tbm8tZW5hYmxlLWxv"
    "Zy1yZXF1ZXN0cyIsCiAgICAgICAgKQogICAgICAgIHNlbGYuZW52ID0gcHJvY2Vzc190cmVl"
    "X2Vudmlyb25tZW50KGdwdV9pbmRleCwgbW9kZWxfaG9tZSkKCiAgICBAcHJvcGVydHkKICAg"
    "IGRlZiBpbnN0YW5jZV9pZChzZWxmKSAtPiBzdHI6CiAgICAgICAgcmV0dXJuIGYie3NlbGYu"
    "d29ya2VyX2lkfS1ne3NlbGYuZ2VuZXJhdGlvbn0iCgogICAgZGVmIHN0YXJ0KHNlbGYsIGNv"
    "dW50ZXJzOiBkaWN0W3N0ciwgaW50XSkgLT4gTm9uZToKICAgICAgICBpZiBzZWxmLnByb2Nl"
    "c3MgaXMgbm90IE5vbmU6CiAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigid29ya2Vy"
    "IGhhcyBhbHJlYWR5IGJlZW4gc3RhcnRlZCIpCiAgICAgICAgaWYgcG9ydF9vcGVuKHNlbGYu"
    "cG9ydCk6CiAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcihmIndvcmtlciBwb3J0IGFs"
    "cmVhZHkgb3Blbjoge3NlbGYucG9ydH0iKQogICAgICAgIGludmVudG9yeSA9IGdwdV9pbnZl"
    "bnRvcnkoKQogICAgICAgIGlkZW50aXR5ID0gaW52ZW50b3J5LmdldChzZWxmLmdwdV9pbmRl"
    "eCkKICAgICAgICBpZiBpZGVudGl0eSBpcyBOb25lOgogICAgICAgICAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoIndvcmtlciBHUFUgaWRlbnRpdHkgaXMgdW5hdmFpbGFibGUiKQogICAgICAg"
    "IGNvbnN1bWVfYWN0aW9ucyhjb3VudGVycywgIndvcmtlcl9zdGFydHMiLCAibW9kZWxfbG9h"
    "ZHMiKQogICAgICAgIHNlbGYubWVtb3J5X2JlZm9yZV9zdGFydF9taWIgPSBpbnQoaWRlbnRp"
    "dHlbIm1lbW9yeV91c2VkX21pYiJdKSAgIyB0eXBlOiBpZ25vcmVbY2FsbC1vdmVybG9hZF0K"
    "ICAgICAgICBzZWxmLmdwdV9pZGVudGl0eSA9IGlkZW50aXR5CiAgICAgICAgc2VsZi5wYXJl"
    "bnRfcGlkID0gb3MuZ2V0cGlkKCkKICAgICAgICBzZWxmLnN0YXJ0ZWRfYXQgPSBkYXRldGlt"
    "ZS5ub3coVVRDKS5pc29mb3JtYXQodGltZXNwZWM9Im1pbGxpc2Vjb25kcyIpCiAgICAgICAg"
    "c2VsZi5wcm9jZXNzID0gc3VicHJvY2Vzcy5Qb3BlbigKICAgICAgICAgICAgc2VsZi5hcmd2"
    "LAogICAgICAgICAgICBlbnY9c2VsZi5lbnYsCiAgICAgICAgICAgIHN0ZGluPXN1YnByb2Nl"
    "c3MuREVWTlVMTCwKICAgICAgICAgICAgc3Rkb3V0PXN1YnByb2Nlc3MuUElQRSwKICAgICAg"
    "ICAgICAgc3RkZXJyPXN1YnByb2Nlc3MuUElQRSwKICAgICAgICAgICAgdGV4dD1GYWxzZSwK"
    "ICAgICAgICAgICAgYnVmc2l6ZT0wLAogICAgICAgICkKICAgICAgICBzZWxmLnByb2Nlc3Nf"
    "c3RhcnRfdGlja3MgPSBwcm9jZXNzX3N0YXJ0X3RpY2tzKHNlbGYucHJvY2Vzcy5waWQpCiAg"
    "ICAgICAgc2VsZi5zdGRvdXQuc3RhcnQoc2VsZi5wcm9jZXNzLnN0ZG91dCkKICAgICAgICBz"
    "ZWxmLnN0ZGVyci5zdGFydChzZWxmLnByb2Nlc3Muc3RkZXJyKQoKICAgIGRlZiB3YWl0X3Jl"
    "YWR5KHNlbGYpIC0+IE5vbmU6CiAgICAgICAgaWYgc2VsZi5wcm9jZXNzIGlzIE5vbmU6CiAg"
    "ICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigid29ya2VyIHdhcyBub3Qgc3RhcnRlZCIp"
    "CiAgICAgICAgZm9yIGluZGV4IGluIHJhbmdlKE1BWF9IRUFMVEhfUE9MTFMpOgogICAgICAg"
    "ICAgICByZXR1cm5jb2RlID0gc2VsZi5wcm9jZXNzLnBvbGwoKQogICAgICAgICAgICBpZiBy"
    "ZXR1cm5jb2RlIGlzIG5vdCBOb25lOgogICAgICAgICAgICAgICAgcmFpc2UgUnVudGltZUVy"
    "cm9yKGYie3NlbGYuaW5zdGFuY2VfaWR9IGV4aXRlZCBiZWZvcmUgcmVhZGluZXNzOiB7cmV0"
    "dXJuY29kZX0iKQogICAgICAgICAgICB0cnk6CiAgICAgICAgICAgICAgICByZXF1ZXN0ID0g"
    "dXJsbGliLnJlcXVlc3QuUmVxdWVzdCgKICAgICAgICAgICAgICAgICAgICBmImh0dHA6Ly8x"
    "MjcuMC4wLjE6e3NlbGYucG9ydH0vaGVhbHRoIiwKICAgICAgICAgICAgICAgICAgICBtZXRo"
    "b2Q9IkdFVCIsCiAgICAgICAgICAgICAgICApCiAgICAgICAgICAgICAgICB3aXRoIHVybGxp"
    "Yi5yZXF1ZXN0LnVybG9wZW4ocmVxdWVzdCwgdGltZW91dD0yKSBhcyByZXNwb25zZToKICAg"
    "ICAgICAgICAgICAgICAgICBpZiByZXNwb25zZS5zdGF0dXMgPT0gMjAwOgogICAgICAgICAg"
    "ICAgICAgICAgICAgICByZXR1cm4KICAgICAgICAgICAgZXhjZXB0IChPU0Vycm9yLCB1cmxs"
    "aWIuZXJyb3IuVVJMRXJyb3IpOgogICAgICAgICAgICAgICAgcGFzcwogICAgICAgICAgICBp"
    "ZiBpbmRleCArIDEgPT0gTUFYX0hFQUxUSF9QT0xMUzoKICAgICAgICAgICAgICAgIHJhaXNl"
    "IFJ1bnRpbWVFcnJvcihmIntzZWxmLmluc3RhbmNlX2lkfSBmYWlsZWQgYm91bmRlZCByZWFk"
    "aW5lc3MgcG9sbGluZyIpCiAgICAgICAgICAgIHRpbWUuc2xlZXAoSEVBTFRIX1BPTExfU0VD"
    "T05EUykKCiAgICBkZWYgdmFsaWRhdGVfbW9kZWwoc2VsZikgLT4gTm9uZToKICAgICAgICBw"
    "YXlsb2FkID0gZ2V0X2pzb24oZiJodHRwOi8vMTI3LjAuMC4xOntzZWxmLnBvcnR9L3YxL21v"
    "ZGVscyIpCiAgICAgICAgZGF0YSA9IHBheWxvYWQuZ2V0KCJkYXRhIikKICAgICAgICBpZiBu"
    "b3QgaXNpbnN0YW5jZShkYXRhLCBsaXN0KSBvciBsZW4oZGF0YSkgIT0gMToKICAgICAgICAg"
    "ICAgcmFpc2UgUnVudGltZUVycm9yKCJzZXJ2ZWQtbW9kZWwgaW52ZW50b3J5IGlzIGludmFs"
    "aWQiKQogICAgICAgIGl0ZW0gPSBkYXRhWzBdCiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2Uo"
    "aXRlbSwgZGljdCkgb3IgaXRlbS5nZXQoImlkIikgIT0gU0VSVkVEX01PREVMX05BTUU6CiAg"
    "ICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigic2VydmVkLW1vZGVsIGlkZW50aXR5IGRy"
    "aWZ0ZWQiKQoKICAgIGRlZiBiYWNrZW5kX21hcmtlcl9ldmlkZW5jZShzZWxmKSAtPiBkaWN0"
    "W3N0ciwgb2JqZWN0XSB8IE5vbmU6CiAgICAgICAgcmV0dXJuIGJhY2tlbmRfbWFya2VyX2V2"
    "aWRlbmNlX2Zyb21fdGV4dHMoCiAgICAgICAgICAgIHNlbGYuc3Rkb3V0LnRleHQoKSwKICAg"
    "ICAgICAgICAgc2VsZi5zdGRlcnIudGV4dCgpLAogICAgICAgICkKCiAgICBkZWYgYmFja2Vu"
    "ZF9tYXJrZXIoc2VsZikgLT4gYm9vbDoKICAgICAgICByZXR1cm4gc2VsZi5iYWNrZW5kX21h"
    "cmtlcl9ldmlkZW5jZSgpIGlzIG5vdCBOb25lCgogICAgZGVmIHdhaXRfYmFja2VuZF9tYXJr"
    "ZXIoc2VsZikgLT4gTm9uZToKICAgICAgICBpZiBzZWxmLnByb2Nlc3MgaXMgTm9uZToKICAg"
    "ICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJ3b3JrZXIgd2FzIG5vdCBzdGFydGVkIikK"
    "ICAgICAgICBmb3IgaW5kZXggaW4gcmFuZ2UoMjApOgogICAgICAgICAgICBldmlkZW5jZSA9"
    "IHNlbGYuYmFja2VuZF9tYXJrZXJfZXZpZGVuY2UoKQogICAgICAgICAgICBpZiBldmlkZW5j"
    "ZSBpcyBub3QgTm9uZToKICAgICAgICAgICAgICAgIHNlbGYuYmFja2VuZF9tYXJrZXJfcG9s"
    "bF9jb3VudCA9IGluZGV4ICsgMQogICAgICAgICAgICAgICAgc2VsZi5iYWNrZW5kX2V2aWRl"
    "bmNlID0gZXZpZGVuY2UKICAgICAgICAgICAgICAgIHJldHVybgogICAgICAgICAgICBpZiBz"
    "ZWxmLnByb2Nlc3MucG9sbCgpIGlzIG5vdCBOb25lOgogICAgICAgICAgICAgICAgcmFpc2Ug"
    "UnVudGltZUVycm9yKGYie3NlbGYuaW5zdGFuY2VfaWR9IGV4aXRlZCBiZWZvcmUgYmFja2Vu"
    "ZCByZWFsaXphdGlvbiIpCiAgICAgICAgICAgIHRpbWUuc2xlZXAoMC4yNSkKICAgICAgICBy"
    "YWlzZSBSdW50aW1lRXJyb3IoZiJ7c2VsZi5pbnN0YW5jZV9pZH0gZXhwbGljaXQgYmFja2Vu"
    "ZCBtYXJrZXIgd2FzIG5vdCBvYnNlcnZlZCIpCgogICAgZGVmIG1ldHJpY19zbmFwc2hvdChz"
    "ZWxmKSAtPiBNZXRyaWNTbmFwc2hvdE9ic2VydmF0aW9uOgogICAgICAgIHJhd19tZXRyaWNz"
    "ID0gZ2V0X3RleHQoZiJodHRwOi8vMTI3LjAuMC4xOntzZWxmLnBvcnR9L21ldHJpY3MiKQog"
    "ICAgICAgIHJldHVybiBtZXRyaWNfc25hcHNob3QocmF3X21ldHJpY3MpCgogICAgZGVmIHJl"
    "cG9ydChzZWxmKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgICAgICBpZiBzZWxmLnByb2Nl"
    "c3MgaXMgTm9uZToKICAgICAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJ3b3JrZXIgd2Fz"
    "IG5vdCBzdGFydGVkIikKICAgICAgICBpZiBzZWxmLmdwdV9pZGVudGl0eSBpcyBOb25lOgog"
    "ICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoIndvcmtlciBHUFUgaWRlbnRpdHkgd2Fz"
    "IG5vdCBjYXB0dXJlZCIpCiAgICAgICAgZXZpZGVuY2UgPSBzZWxmLmJhY2tlbmRfZXZpZGVu"
    "Y2Ugb3Igc2VsZi5iYWNrZW5kX21hcmtlcl9ldmlkZW5jZSgpCiAgICAgICAgcmV0dXJuIHsK"
    "ICAgICAgICAgICAgIndvcmtlcl9pZCI6IHNlbGYud29ya2VyX2lkLAogICAgICAgICAgICAi"
    "d29ya2VyX2luc3RhbmNlX2lkIjogc2VsZi5pbnN0YW5jZV9pZCwKICAgICAgICAgICAgImdl"
    "bmVyYXRpb24iOiBzZWxmLmdlbmVyYXRpb24sCiAgICAgICAgICAgICJncHVfaW5kZXgiOiBz"
    "ZWxmLmdwdV9pbmRleCwKICAgICAgICAgICAgImdwdV9pZGVudGl0eSI6IHNlbGYuZ3B1X2lk"
    "ZW50aXR5LAogICAgICAgICAgICAicG9ydCI6IHNlbGYucG9ydCwKICAgICAgICAgICAgInBp"
    "ZCI6IHNlbGYucHJvY2Vzcy5waWQsCiAgICAgICAgICAgICJwYXJlbnRfcGlkIjogc2VsZi5w"
    "YXJlbnRfcGlkLAogICAgICAgICAgICAicHJvY2Vzc19zdGFydF90aWNrcyI6IHNlbGYucHJv"
    "Y2Vzc19zdGFydF90aWNrcywKICAgICAgICAgICAgInN0YXJ0ZWRfYXQiOiBzZWxmLnN0YXJ0"
    "ZWRfYXQsCiAgICAgICAgICAgICJhcmd2X3NoYTI1NiI6IHNoYTI1Nl90ZXh0KGNhbm9uaWNh"
    "bF9qc29uKHNlbGYuYXJndikpLAogICAgICAgICAgICAiZXhwbGljaXRfYXR0ZW50aW9uX2Jh"
    "Y2tlbmQiOiBFWFBFQ1RFRF9CQUNLRU5ELAogICAgICAgICAgICAiYmFja2VuZF9sb2dfbWFy"
    "a2VyX29ic2VydmVkIjogZXZpZGVuY2UgaXMgbm90IE5vbmUsCiAgICAgICAgICAgICJiYWNr"
    "ZW5kX21hcmtlcl9ldmlkZW5jZSI6IGV2aWRlbmNlLAogICAgICAgICAgICAiYmFja2VuZF9t"
    "YXJrZXJfcG9sbF9jb3VudCI6IHNlbGYuYmFja2VuZF9tYXJrZXJfcG9sbF9jb3VudCwKICAg"
    "ICAgICAgICAgInN0ZG91dCI6IHNlbGYuc3Rkb3V0LnNuYXBzaG90KCksCiAgICAgICAgICAg"
    "ICJzdGRlcnIiOiBzZWxmLnN0ZGVyci5zbmFwc2hvdCgpLAogICAgICAgIH0KCiAgICBkZWYg"
    "c3RvcF9hbmRfcmVwb3J0KHNlbGYsIHJlYXNvbjogc3RyKSAtPiBkaWN0W3N0ciwgb2JqZWN0"
    "XToKICAgICAgICBpZiBzZWxmLnRlYXJkb3duX3JlcG9ydCBpcyBub3QgTm9uZToKICAgICAg"
    "ICAgICAgcmV0dXJuIHNlbGYudGVhcmRvd25fcmVwb3J0CiAgICAgICAgaWYgc2VsZi5wcm9j"
    "ZXNzIGlzIE5vbmU6CiAgICAgICAgICAgIHNlbGYuc3Rkb3V0LmZpbmFsaXplKCkKICAgICAg"
    "ICAgICAgc2VsZi5zdGRlcnIuZmluYWxpemUoKQogICAgICAgICAgICBzZWxmLnRlYXJkb3du"
    "X3JlcG9ydCA9IHsKICAgICAgICAgICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiOiBzZWxm"
    "Lmluc3RhbmNlX2lkLAogICAgICAgICAgICAgICAgInN0YXR1cyI6ICJOT1RfU1RBUlRFRCIs"
    "CiAgICAgICAgICAgICAgICAicmVhc29uIjogcmVhc29uLAogICAgICAgICAgICB9CiAgICAg"
    "ICAgICAgIHJldHVybiBzZWxmLnRlYXJkb3duX3JlcG9ydAoKICAgICAgICByb290X3BpZCA9"
    "IHNlbGYucHJvY2Vzcy5waWQKICAgICAgICB0cmVlX2JlZm9yZTogc2V0W2ludF0gPSB7cm9v"
    "dF9waWR9CiAgICAgICAgZ3B1X3BpZHNfYmVmb3JlOiBzZXRbaW50XSA9IHNldCgpCiAgICAg"
    "ICAgdGVhcmRvd25fZXJyb3JfdHlwZTogc3RyIHwgTm9uZSA9IE5vbmUKICAgICAgICB0ZWFy"
    "ZG93bl9lcnJvcl9tZXNzYWdlOiBzdHIgfCBOb25lID0gTm9uZQogICAgICAgIHRyeToKICAg"
    "ICAgICAgICAgdHJlZV9iZWZvcmUgPSBkZXNjZW5kYW50cyhyb290X3BpZCwgcHJvY2Vzc19w"
    "YXJlbnRfbWFwKCkpCiAgICAgICAgICAgIGdwdV9wcm9jZXNzZXNfYmVmb3JlID0gY29tcHV0"
    "ZV9wcm9jZXNzZXMoKQogICAgICAgICAgICBncHVfcGlkc19iZWZvcmUgPSB7cGlkIGZvciBw"
    "aWQgaW4gdHJlZV9iZWZvcmUgaWYgcGlkIGluIGdwdV9wcm9jZXNzZXNfYmVmb3JlfQogICAg"
    "ICAgIGV4Y2VwdCAoT1NFcnJvciwgUnVudGltZUVycm9yLCBWYWx1ZUVycm9yKSBhcyBlcnJv"
    "cjoKICAgICAgICAgICAgdGVhcmRvd25fZXJyb3JfdHlwZSA9IHR5cGUoZXJyb3IpLl9fbmFt"
    "ZV9fCiAgICAgICAgICAgIHRlYXJkb3duX2Vycm9yX21lc3NhZ2UgPSBzYW5pdGl6ZV9leGNl"
    "cnB0KHN0cihlcnJvcikpWzo1MTJdCgogICAgICAgIHRlcm1pbmF0aW9uX21ldGhvZCA9ICJB"
    "TFJFQURZX0VYSVRFRCIKICAgICAgICBpZiBzZWxmLnByb2Nlc3MucG9sbCgpIGlzIE5vbmU6"
    "CiAgICAgICAgICAgIHRlcm1pbmF0aW9uX21ldGhvZCA9ICJTSUdURVJNIgogICAgICAgICAg"
    "ICBzZWxmLnByb2Nlc3MudGVybWluYXRlKCkKICAgICAgICB0cnk6CiAgICAgICAgICAgIHNl"
    "bGYucHJvY2Vzcy53YWl0KHRpbWVvdXQ9MjApCiAgICAgICAgZXhjZXB0IHN1YnByb2Nlc3Mu"
    "VGltZW91dEV4cGlyZWQ6CiAgICAgICAgICAgIHRlcm1pbmF0aW9uX21ldGhvZCA9ICJTSUdL"
    "SUxMIgogICAgICAgICAgICBzZWxmLnByb2Nlc3Mua2lsbCgpCiAgICAgICAgICAgIHNlbGYu"
    "cHJvY2Vzcy53YWl0KHRpbWVvdXQ9MTApCgogICAgICAgIHNlbGYuc3Rkb3V0LmZpbmFsaXpl"
    "KCkKICAgICAgICBzZWxmLnN0ZGVyci5maW5hbGl6ZSgpCgogICAgICAgIHByb2Nlc3NfdHJl"
    "ZV9hYnNlbnQgPSBGYWxzZQogICAgICAgIGdwdV9wcm9jZXNzZXNfYWJzZW50ID0gRmFsc2UK"
    "ICAgICAgICBwb3J0X2Nsb3NlZCA9IEZhbHNlCiAgICAgICAgbWVtb3J5X3JldHVybmVkID0g"
    "RmFsc2UKICAgICAgICBtZW1vcnlfYWZ0ZXJfc3RvcF9taWI6IGludCB8IE5vbmUgPSBOb25l"
    "CiAgICAgICAgcG9sbHMgPSAwCiAgICAgICAgZm9yIGluZGV4IGluIHJhbmdlKE1BWF9URUFS"
    "RE9XTl9QT0xMUyk6CiAgICAgICAgICAgIHBvbGxzID0gaW5kZXggKyAxCiAgICAgICAgICAg"
    "IGN1cnJlbnRfcGFyZW50cyA9IHByb2Nlc3NfcGFyZW50X21hcCgpCiAgICAgICAgICAgIGN1"
    "cnJlbnRfZ3B1X3Byb2Nlc3NlcyA9IGNvbXB1dGVfcHJvY2Vzc2VzKCkKICAgICAgICAgICAg"
    "Y3VycmVudF9waWRzID0gc2V0KGN1cnJlbnRfcGFyZW50cykKICAgICAgICAgICAgcHJvY2Vz"
    "c190cmVlX2Fic2VudCA9IG5vdCBib29sKHRyZWVfYmVmb3JlICYgY3VycmVudF9waWRzKQog"
    "ICAgICAgICAgICBncHVfcHJvY2Vzc2VzX2Fic2VudCA9IG5vdCBib29sKHRyZWVfYmVmb3Jl"
    "ICYgc2V0KGN1cnJlbnRfZ3B1X3Byb2Nlc3NlcykpCiAgICAgICAgICAgIHBvcnRfY2xvc2Vk"
    "ID0gbm90IHBvcnRfb3BlbihzZWxmLnBvcnQpCiAgICAgICAgICAgIGlmIHNlbGYuZ3B1X2lk"
    "ZW50aXR5IGlzIG5vdCBOb25lOgogICAgICAgICAgICAgICAgbWVtb3J5X2FmdGVyX3N0b3Bf"
    "bWliID0gZ3B1X21lbW9yeV91c2VkX21pYihzdHIoc2VsZi5ncHVfaWRlbnRpdHlbInV1aWQi"
    "XSkpCiAgICAgICAgICAgIGlmIHNlbGYubWVtb3J5X2JlZm9yZV9zdGFydF9taWIgaXMgbm90"
    "IE5vbmUgYW5kIG1lbW9yeV9hZnRlcl9zdG9wX21pYiBpcyBub3QgTm9uZToKICAgICAgICAg"
    "ICAgICAgIG1lbW9yeV9yZXR1cm5lZCA9ICgKICAgICAgICAgICAgICAgICAgICBtZW1vcnlf"
    "YWZ0ZXJfc3RvcF9taWIKICAgICAgICAgICAgICAgICAgICA8PSBzZWxmLm1lbW9yeV9iZWZv"
    "cmVfc3RhcnRfbWliICsgR1BVX01FTU9SWV9SRVRVUk5fVE9MRVJBTkNFX01JQgogICAgICAg"
    "ICAgICAgICAgKQogICAgICAgICAgICBpZiBwcm9jZXNzX3RyZWVfYWJzZW50IGFuZCBncHVf"
    "cHJvY2Vzc2VzX2Fic2VudCBhbmQgcG9ydF9jbG9zZWQgYW5kIG1lbW9yeV9yZXR1cm5lZDoK"
    "ICAgICAgICAgICAgICAgIGJyZWFrCiAgICAgICAgICAgIHRpbWUuc2xlZXAoVEVBUkRPV05f"
    "UE9MTF9TRUNPTkRTKQoKICAgICAgICBjYXB0dXJlc19maW5hbGl6ZWQgPSAoCiAgICAgICAg"
    "ICAgIGJvb2woc2VsZi5zdGRvdXQuc25hcHNob3QoKVsiY2FwdHVyZV9maW5hbGl6ZWQiXSkK"
    "ICAgICAgICAgICAgYW5kIGJvb2woc2VsZi5zdGRlcnIuc25hcHNob3QoKVsiY2FwdHVyZV9m"
    "aW5hbGl6ZWQiXSkKICAgICAgICAgICAgYW5kIG5vdCBib29sKHNlbGYuc3Rkb3V0LnNuYXBz"
    "aG90KClbImNhcHR1cmVfdGhyZWFkX2FsaXZlIl0pCiAgICAgICAgICAgIGFuZCBub3QgYm9v"
    "bChzZWxmLnN0ZGVyci5zbmFwc2hvdCgpWyJjYXB0dXJlX3RocmVhZF9hbGl2ZSJdKQogICAg"
    "ICAgICkKICAgICAgICBzdGF0dXMgPSAoCiAgICAgICAgICAgICJQQVNTRUQiCiAgICAgICAg"
    "ICAgIGlmICgKICAgICAgICAgICAgICAgIHByb2Nlc3NfdHJlZV9hYnNlbnQKICAgICAgICAg"
    "ICAgICAgIGFuZCBncHVfcHJvY2Vzc2VzX2Fic2VudAogICAgICAgICAgICAgICAgYW5kIHBv"
    "cnRfY2xvc2VkCiAgICAgICAgICAgICAgICBhbmQgbWVtb3J5X3JldHVybmVkCiAgICAgICAg"
    "ICAgICAgICBhbmQgY2FwdHVyZXNfZmluYWxpemVkCiAgICAgICAgICAgICAgICBhbmQgdGVh"
    "cmRvd25fZXJyb3JfdHlwZSBpcyBOb25lCiAgICAgICAgICAgICkKICAgICAgICAgICAgZWxz"
    "ZSAiRkFJTEVEIgogICAgICAgICkKICAgICAgICBzZWxmLnRlYXJkb3duX3JlcG9ydCA9IHsK"
    "ICAgICAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgInJl"
    "cG9ydF9pZCI6ICJhdXJhZ2F0ZXdheS1wNS1wNi1leGFjdC1ydW50aW1lLXJlcXVhbGlmaWNh"
    "dGlvbi13b3JrZXItdGVhcmRvd24tdjEiLAogICAgICAgICAgICAid29ya2VyX2lkIjogc2Vs"
    "Zi53b3JrZXJfaWQsCiAgICAgICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiOiBzZWxmLmlu"
    "c3RhbmNlX2lkLAogICAgICAgICAgICAiZ2VuZXJhdGlvbiI6IHNlbGYuZ2VuZXJhdGlvbiwK"
    "ICAgICAgICAgICAgInJlYXNvbiI6IHJlYXNvbiwKICAgICAgICAgICAgInN0YXR1cyI6IHN0"
    "YXR1cywKICAgICAgICAgICAgInJvb3RfcGlkIjogcm9vdF9waWQsCiAgICAgICAgICAgICJw"
    "cm9jZXNzX3RyZWVfcGlkc19iZWZvcmUiOiBzb3J0ZWQodHJlZV9iZWZvcmUpLAogICAgICAg"
    "ICAgICAiZ3B1X3Byb2Nlc3NfcGlkc19iZWZvcmUiOiBzb3J0ZWQoZ3B1X3BpZHNfYmVmb3Jl"
    "KSwKICAgICAgICAgICAgInJldHVybmNvZGUiOiBzZWxmLnByb2Nlc3MucmV0dXJuY29kZSwK"
    "ICAgICAgICAgICAgInRlcm1pbmF0aW9uX21ldGhvZCI6IHRlcm1pbmF0aW9uX21ldGhvZCwK"
    "ICAgICAgICAgICAgInByb2Nlc3NfdHJlZV9hYnNlbnRfYWZ0ZXIiOiBwcm9jZXNzX3RyZWVf"
    "YWJzZW50LAogICAgICAgICAgICAiZ3B1X3Byb2Nlc3Nlc19hYnNlbnRfYWZ0ZXIiOiBncHVf"
    "cHJvY2Vzc2VzX2Fic2VudCwKICAgICAgICAgICAgInBvcnRfY2xvc2VkX2FmdGVyIjogcG9y"
    "dF9jbG9zZWQsCiAgICAgICAgICAgICJjYXB0dXJlX3RocmVhZHNfZmluYWxpemVkIjogY2Fw"
    "dHVyZXNfZmluYWxpemVkLAogICAgICAgICAgICAibWVtb3J5X2JlZm9yZV9zdGFydF9taWIi"
    "OiBzZWxmLm1lbW9yeV9iZWZvcmVfc3RhcnRfbWliLAogICAgICAgICAgICAibWVtb3J5X2Fm"
    "dGVyX3N0b3BfbWliIjogbWVtb3J5X2FmdGVyX3N0b3BfbWliLAogICAgICAgICAgICAibWVt"
    "b3J5X3JldHVybl90b2xlcmFuY2VfbWliIjogR1BVX01FTU9SWV9SRVRVUk5fVE9MRVJBTkNF"
    "X01JQiwKICAgICAgICAgICAgIm1lbW9yeV9yZXR1cm5lZF93aXRoaW5fdG9sZXJhbmNlIjog"
    "bWVtb3J5X3JldHVybmVkLAogICAgICAgICAgICAiYm91bmRlZF9wb2xsX2NvdW50IjogcG9s"
    "bHMsCiAgICAgICAgICAgICJ0ZWFyZG93bl9lcnJvcl90eXBlIjogdGVhcmRvd25fZXJyb3Jf"
    "dHlwZSwKICAgICAgICAgICAgInRlYXJkb3duX2Vycm9yX21lc3NhZ2UiOiB0ZWFyZG93bl9l"
    "cnJvcl9tZXNzYWdlLAogICAgICAgIH0KICAgICAgICByZXR1cm4gc2VsZi50ZWFyZG93bl9y"
    "ZXBvcnQKCiAgICBkZWYgZmFpbHVyZV9kaWFnbm9zdGljcyhzZWxmKSAtPiBkaWN0W3N0ciwg"
    "b2JqZWN0XToKICAgICAgICByZXR1cm4gewogICAgICAgICAgICAid29ya2VyX2lkIjogc2Vs"
    "Zi53b3JrZXJfaWQsCiAgICAgICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiOiBzZWxmLmlu"
    "c3RhbmNlX2lkLAogICAgICAgICAgICAiZ2VuZXJhdGlvbiI6IHNlbGYuZ2VuZXJhdGlvbiwK"
    "ICAgICAgICAgICAgImdwdV9pbmRleCI6IHNlbGYuZ3B1X2luZGV4LAogICAgICAgICAgICAi"
    "Z3B1X2lkZW50aXR5Ijogc2VsZi5ncHVfaWRlbnRpdHksCiAgICAgICAgICAgICJwb3J0Ijog"
    "c2VsZi5wb3J0LAogICAgICAgICAgICAicmV0dXJuY29kZSI6IChOb25lIGlmIHNlbGYucHJv"
    "Y2VzcyBpcyBOb25lIGVsc2Ugc2VsZi5wcm9jZXNzLnBvbGwoKSksCiAgICAgICAgICAgICJw"
    "aWQiOiBOb25lIGlmIHNlbGYucHJvY2VzcyBpcyBOb25lIGVsc2Ugc2VsZi5wcm9jZXNzLnBp"
    "ZCwKICAgICAgICAgICAgInBhcmVudF9waWQiOiBzZWxmLnBhcmVudF9waWQsCiAgICAgICAg"
    "ICAgICJwcm9jZXNzX3N0YXJ0X3RpY2tzIjogc2VsZi5wcm9jZXNzX3N0YXJ0X3RpY2tzLAog"
    "ICAgICAgICAgICAic3RhcnRlZF9hdCI6IHNlbGYuc3RhcnRlZF9hdCwKICAgICAgICAgICAg"
    "ImFyZ3Zfc2hhMjU2Ijogc2hhMjU2X3RleHQoY2Fub25pY2FsX2pzb24oc2VsZi5hcmd2KSks"
    "CiAgICAgICAgICAgICJweXRob25wYXRoIjogc2VsZi5lbnYuZ2V0KCJQWVRIT05QQVRIIiks"
    "CiAgICAgICAgICAgICJweXRob25wYXRoX2V4YWN0X3RhcmdldF9zaXRlIjogKHNlbGYuZW52"
    "LmdldCgiUFlUSE9OUEFUSCIpID09IHN0cihUQVJHRVRfU0lURSkpLAogICAgICAgICAgICAi"
    "YmFja2VuZF9tYXJrZXJfZXZpZGVuY2UiOiBzZWxmLmJhY2tlbmRfZXZpZGVuY2UsCiAgICAg"
    "ICAgICAgICJzdGRvdXQiOiBzZWxmLnN0ZG91dC5zbmFwc2hvdCgpLAogICAgICAgICAgICAi"
    "c3RkZXJyIjogc2VsZi5zdGRlcnIuc25hcHNob3QoKSwKICAgICAgICAgICAgInN0ZG91dF90"
    "YWlsIjogc2FuaXRpemVfZXhjZXJwdChzZWxmLnN0ZG91dC50ZXh0KCkpLAogICAgICAgICAg"
    "ICAic3RkZXJyX3RhaWwiOiBzYW5pdGl6ZV9leGNlcnB0KHNlbGYuc3RkZXJyLnRleHQoKSks"
    "CiAgICAgICAgICAgICJ0ZWFyZG93biI6IHNlbGYudGVhcmRvd25fcmVwb3J0LAogICAgICAg"
    "IH0KCgpQUk9NRVRIRVVTX0xJTkUgPSByZS5jb21waWxlKAogICAgciJeKD9QPG5hbWU+W2Et"
    "ekEtWl86XVthLXpBLVowLTlfOl0qKSIKICAgIHIiKD86XHsoP1A8bGFiZWxzPi4qKVx9KT9c"
    "cysiCiAgICByIig/UDx2YWx1ZT5bLStdPyg/OlxkKyg/OlwuXGQqKT98XC5cZCspKD86W2VF"
    "XVstK10/XGQrKT8pJCIKKQpQUk9NRVRIRVVTX0xBQkVMID0gcmUuY29tcGlsZShyJyg/Ol58"
    "LCkoP1A8a2V5PlthLXpBLVpfXVthLXpBLVowLTlfXSopPSIoP1A8dmFsdWU+KD86XFwufFte"
    "Il0pKikiJykKCk1FVFJJQ19TQU1QTEVfTkFNRVMgPSB7CiAgICAicHJlZml4X2NhY2hlX3F1"
    "ZXJpZXMiOiAidmxsbTpwcmVmaXhfY2FjaGVfcXVlcmllc190b3RhbCIsCiAgICAicHJlZml4"
    "X2NhY2hlX2hpdHMiOiAidmxsbTpwcmVmaXhfY2FjaGVfaGl0c190b3RhbCIsCiAgICAibG9j"
    "YWxfY29tcHV0ZSI6ICJ2bGxtOnByb21wdF90b2tlbnNfYnlfc291cmNlX3RvdGFsIiwKICAg"
    "ICJsb2NhbF9jYWNoZV9oaXQiOiAidmxsbTpwcm9tcHRfdG9rZW5zX2J5X3NvdXJjZV90b3Rh"
    "bCIsCiAgICAiZXh0ZXJuYWxfa3ZfdHJhbnNmZXIiOiAidmxsbTpwcm9tcHRfdG9rZW5zX2J5"
    "X3NvdXJjZV90b3RhbCIsCiAgICAiY2FjaGVkX3Byb21wdF90b2tlbnMiOiAidmxsbTpwcm9t"
    "cHRfdG9rZW5zX2NhY2hlZF90b3RhbCIsCiAgICAibmV3bHlfY29tcHV0ZWRfcHJlZmlsbF90"
    "b2tlbnMiOiAoInZsbG06cmVxdWVzdF9wcmVmaWxsX2t2X2NvbXB1dGVkX3Rva2Vuc19zdW0i"
    "KSwKfQoKTUVUUklDX1NPVVJDRV9MQUJFTFMgPSB7CiAgICAibG9jYWxfY29tcHV0ZSI6ICJs"
    "b2NhbF9jb21wdXRlIiwKICAgICJsb2NhbF9jYWNoZV9oaXQiOiAibG9jYWxfY2FjaGVfaGl0"
    "IiwKICAgICJleHRlcm5hbF9rdl90cmFuc2ZlciI6ICJleHRlcm5hbF9rdl90cmFuc2ZlciIs"
    "Cn0KCgpkZWYgX3VuZXNjYXBlX3Byb21ldGhldXNfbGFiZWwodmFsdWU6IHN0cikgLT4gc3Ry"
    "OgogICAgcmV0dXJuIHZhbHVlLnJlcGxhY2UociJcXCIsICJcXCIpLnJlcGxhY2UociJcIiIs"
    "ICciJykucmVwbGFjZShyIlxuIiwgIlxuIikKCgpkZWYgX3BhcnNlX3Byb21ldGhldXNfbGFi"
    "ZWxzKHBheWxvYWQ6IHN0cikgLT4gdHVwbGVbdHVwbGVbc3RyLCBzdHJdLCAuLi5dOgogICAg"
    "aWYgbm90IHBheWxvYWQ6CiAgICAgICAgcmV0dXJuICgpCiAgICBwb3NpdGlvbiA9IDAKICAg"
    "IGl0ZW1zOiBsaXN0W3R1cGxlW3N0ciwgc3RyXV0gPSBbXQogICAgd2hpbGUgcG9zaXRpb24g"
    "PCBsZW4ocGF5bG9hZCk6CiAgICAgICAgbWF0Y2ggPSBQUk9NRVRIRVVTX0xBQkVMLm1hdGNo"
    "KHBheWxvYWQsIHBvc2l0aW9uKQogICAgICAgIGlmIG1hdGNoIGlzIE5vbmU6CiAgICAgICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNBbWJpZ3VpdHkoCiAgICAgICAgICAgICAgICAiTUVUUklD"
    "X1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgIlByb21ldGhldXMgbGFiZWxz"
    "IGNvdWxkIG5vdCBiZSBwYXJzZWQgbG9zc2xlc3NseSIsCiAgICAgICAgICAgICkKICAgICAg"
    "ICBpdGVtcy5hcHBlbmQoCiAgICAgICAgICAgICgKICAgICAgICAgICAgICAgIG1hdGNoLmdy"
    "b3VwKCJrZXkiKSwKICAgICAgICAgICAgICAgIF91bmVzY2FwZV9wcm9tZXRoZXVzX2xhYmVs"
    "KG1hdGNoLmdyb3VwKCJ2YWx1ZSIpKSwKICAgICAgICAgICAgKQogICAgICAgICkKICAgICAg"
    "ICBwb3NpdGlvbiA9IG1hdGNoLmVuZCgpCiAgICByZXR1cm4gdHVwbGUoc29ydGVkKGl0ZW1z"
    "KSkKCgpkZWYgcGFyc2VfbWV0cmljX3NhbXBsZXMocGF5bG9hZDogc3RyKSAtPiB0dXBsZVtS"
    "YXdNZXRyaWNTYW1wbGUsIC4uLl06CiAgICBzYW1wbGVzOiBsaXN0W1Jhd01ldHJpY1NhbXBs"
    "ZV0gPSBbXQogICAgZm9yIHJhd19saW5lIGluIHBheWxvYWQuc3BsaXRsaW5lcygpOgogICAg"
    "ICAgIHN0cmlwcGVkID0gcmF3X2xpbmUuc3RyaXAoKQogICAgICAgIGlmIG5vdCBzdHJpcHBl"
    "ZCBvciBzdHJpcHBlZC5zdGFydHN3aXRoKCIjIik6CiAgICAgICAgICAgIGNvbnRpbnVlCiAg"
    "ICAgICAgbWF0Y2ggPSBQUk9NRVRIRVVTX0xJTkUuZnVsbG1hdGNoKHN0cmlwcGVkKQogICAg"
    "ICAgIGlmIG1hdGNoIGlzIE5vbmU6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgbmFt"
    "ZSA9IG1hdGNoLmdyb3VwKCJuYW1lIikKICAgICAgICBpZiBuYW1lIG5vdCBpbiBzZXQoTUVU"
    "UklDX1NBTVBMRV9OQU1FUy52YWx1ZXMoKSk6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAg"
    "ICAgdHJ5OgogICAgICAgICAgICB2YWx1ZSA9IGZsb2F0KG1hdGNoLmdyb3VwKCJ2YWx1ZSIp"
    "KQogICAgICAgIGV4Y2VwdCBWYWx1ZUVycm9yIGFzIGVycm9yOgogICAgICAgICAgICByYWlz"
    "ZSBEaWFnbm9zdGljQW1iaWd1aXR5KAogICAgICAgICAgICAgICAgIk1FVFJJQ19TRU1BTlRJ"
    "Q19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJyZWxldmFudCBQcm9tZXRoZXVzIG1ldHJp"
    "YyB2YWx1ZSBpcyBpbnZhbGlkIiwKICAgICAgICAgICAgKSBmcm9tIGVycm9yCiAgICAgICAg"
    "aWYgdmFsdWUgPCAwOgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljQW1iaWd1aXR5KAog"
    "ICAgICAgICAgICAgICAgIk1FVFJJQ19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAg"
    "ICAgICJyZWxldmFudCBQcm9tZXRoZXVzIGNvdW50ZXIgb3IgaGlzdG9ncmFtIHN1bSBpcyBu"
    "ZWdhdGl2ZSIsCiAgICAgICAgICAgICkKICAgICAgICBzYW1wbGVzLmFwcGVuZCgKICAgICAg"
    "ICAgICAgUmF3TWV0cmljU2FtcGxlKAogICAgICAgICAgICAgICAgbmFtZT1uYW1lLAogICAg"
    "ICAgICAgICAgICAgbGFiZWxzPV9wYXJzZV9wcm9tZXRoZXVzX2xhYmVscyhtYXRjaC5ncm91"
    "cCgibGFiZWxzIikgb3IgIiIpLAogICAgICAgICAgICAgICAgdmFsdWU9dmFsdWUsCiAgICAg"
    "ICAgICAgICkKICAgICAgICApCiAgICByZXR1cm4gdHVwbGUoc2FtcGxlcykKCgpkZWYgX2xh"
    "YmVsc19kaWN0KHNhbXBsZTogUmF3TWV0cmljU2FtcGxlKSAtPiBkaWN0W3N0ciwgc3RyXToK"
    "ICAgIHJldHVybiBkaWN0KHNhbXBsZS5sYWJlbHMpCgoKZGVmIF9tZXRyaWNfdmFsdWUoCiAg"
    "ICBzYW1wbGVzOiB0dXBsZVtSYXdNZXRyaWNTYW1wbGUsIC4uLl0sCiAgICBzZW1hbnRpY19y"
    "b2xlOiBzdHIsCikgLT4gZmxvYXQ6CiAgICBuYW1lID0gTUVUUklDX1NBTVBMRV9OQU1FU1tz"
    "ZW1hbnRpY19yb2xlXQogICAgc291cmNlID0gTUVUUklDX1NPVVJDRV9MQUJFTFMuZ2V0KHNl"
    "bWFudGljX3JvbGUpCiAgICBtYXRjaGVzOiBsaXN0W1Jhd01ldHJpY1NhbXBsZV0gPSBbXQog"
    "ICAgZm9yIHNhbXBsZSBpbiBzYW1wbGVzOgogICAgICAgIGlmIHNhbXBsZS5uYW1lICE9IG5h"
    "bWU6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgbGFiZWxzID0gX2xhYmVsc19kaWN0"
    "KHNhbXBsZSkKICAgICAgICBpZiBsYWJlbHMuZ2V0KCJtb2RlbF9uYW1lIikgIT0gU0VSVkVE"
    "X01PREVMX05BTUU6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgaWYgbGFiZWxzLmdl"
    "dCgiZW5naW5lIikgIT0gIjAiOgogICAgICAgICAgICBjb250aW51ZQogICAgICAgIGlmIHNv"
    "dXJjZSBpcyBub3QgTm9uZSBhbmQgbGFiZWxzLmdldCgic291cmNlIikgIT0gc291cmNlOgog"
    "ICAgICAgICAgICBjb250aW51ZQogICAgICAgIGlmIHNvdXJjZSBpcyBOb25lIGFuZCAic291"
    "cmNlIiBpbiBsYWJlbHM6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgbWF0Y2hlcy5h"
    "cHBlbmQoc2FtcGxlKQogICAgaWYgbGVuKG1hdGNoZXMpICE9IDE6CiAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0FtYmlndWl0eSgKICAgICAgICAgICAgIk1FVFJJQ19BVFRSSUJVVElPTl9B"
    "TUJJR1VPVVMiLAogICAgICAgICAgICAoZiJleHBlY3RlZCBleGFjdGx5IG9uZSBtZXRyaWMg"
    "c2VyaWVzIGZvciB7c2VtYW50aWNfcm9sZX07IG9ic2VydmVkPXtsZW4obWF0Y2hlcyl9Iiks"
    "CiAgICAgICAgKQogICAgcmV0dXJuIG1hdGNoZXNbMF0udmFsdWUKCgpkZWYgbWV0cmljX3Nu"
    "YXBzaG90KHBheWxvYWQ6IHN0cikgLT4gTWV0cmljU25hcHNob3RPYnNlcnZhdGlvbjoKICAg"
    "IHNhbXBsZXMgPSBwYXJzZV9tZXRyaWNfc2FtcGxlcyhwYXlsb2FkKQogICAgb2JzZXJ2YXRp"
    "b24gPSBNZXRyaWNTbmFwc2hvdE9ic2VydmF0aW9uKAogICAgICAgIHByZWZpeF9jYWNoZV9x"
    "dWVyaWVzPV9tZXRyaWNfdmFsdWUoc2FtcGxlcywgInByZWZpeF9jYWNoZV9xdWVyaWVzIiks"
    "CiAgICAgICAgcHJlZml4X2NhY2hlX2hpdHM9X21ldHJpY192YWx1ZShzYW1wbGVzLCAicHJl"
    "Zml4X2NhY2hlX2hpdHMiKSwKICAgICAgICBsb2NhbF9jb21wdXRlPV9tZXRyaWNfdmFsdWUo"
    "c2FtcGxlcywgImxvY2FsX2NvbXB1dGUiKSwKICAgICAgICBsb2NhbF9jYWNoZV9oaXQ9X21l"
    "dHJpY192YWx1ZShzYW1wbGVzLCAibG9jYWxfY2FjaGVfaGl0IiksCiAgICAgICAgZXh0ZXJu"
    "YWxfa3ZfdHJhbnNmZXI9X21ldHJpY192YWx1ZShzYW1wbGVzLCAiZXh0ZXJuYWxfa3ZfdHJh"
    "bnNmZXIiKSwKICAgICAgICBjYWNoZWRfcHJvbXB0X3Rva2Vucz1fbWV0cmljX3ZhbHVlKHNh"
    "bXBsZXMsICJjYWNoZWRfcHJvbXB0X3Rva2VucyIpLAogICAgICAgIG5ld2x5X2NvbXB1dGVk"
    "X3ByZWZpbGxfdG9rZW5zPV9tZXRyaWNfdmFsdWUoCiAgICAgICAgICAgIHNhbXBsZXMsCiAg"
    "ICAgICAgICAgICJuZXdseV9jb21wdXRlZF9wcmVmaWxsX3Rva2VucyIsCiAgICAgICAgKSwK"
    "ICAgICkKICAgIGlmICgKICAgICAgICBhYnMoCiAgICAgICAgICAgIG9ic2VydmF0aW9uLmNh"
    "Y2hlZF9wcm9tcHRfdG9rZW5zCiAgICAgICAgICAgIC0gKG9ic2VydmF0aW9uLmxvY2FsX2Nh"
    "Y2hlX2hpdCArIG9ic2VydmF0aW9uLmV4dGVybmFsX2t2X3RyYW5zZmVyKQogICAgICAgICkK"
    "ICAgICAgICA+IDFlLTkKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0FtYmlndWl0"
    "eSgKICAgICAgICAgICAgIk1FVFJJQ19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAg"
    "ImNhY2hlZC1wcm9tcHQtdG9rZW4gc291cmNlIGludmFyaWFudCBpcyB2aW9sYXRlZCIsCiAg"
    "ICAgICAgKQogICAgcmV0dXJuIG9ic2VydmF0aW9uCgoKZGVmIG1ldHJpY19kZWx0YSgKICAg"
    "IGJlZm9yZTogTWV0cmljU25hcHNob3RPYnNlcnZhdGlvbiwKICAgIGFmdGVyOiBNZXRyaWNT"
    "bmFwc2hvdE9ic2VydmF0aW9uLAopIC0+IE1ldHJpY0RlbHRhT2JzZXJ2YXRpb246CiAgICB2"
    "YWx1ZXM6IGRpY3Rbc3RyLCBmbG9hdF0gPSB7fQogICAgZm9yIGZpZWxkX25hbWUgaW4gKAog"
    "ICAgICAgICJwcmVmaXhfY2FjaGVfcXVlcmllcyIsCiAgICAgICAgInByZWZpeF9jYWNoZV9o"
    "aXRzIiwKICAgICAgICAibG9jYWxfY29tcHV0ZSIsCiAgICAgICAgImxvY2FsX2NhY2hlX2hp"
    "dCIsCiAgICAgICAgImV4dGVybmFsX2t2X3RyYW5zZmVyIiwKICAgICAgICAiY2FjaGVkX3By"
    "b21wdF90b2tlbnMiLAogICAgICAgICJuZXdseV9jb21wdXRlZF9wcmVmaWxsX3Rva2VucyIs"
    "CiAgICApOgogICAgICAgIGRlbHRhID0gZmxvYXQoZ2V0YXR0cihhZnRlciwgZmllbGRfbmFt"
    "ZSkpIC0gZmxvYXQoZ2V0YXR0cihiZWZvcmUsIGZpZWxkX25hbWUpKQogICAgICAgIGlmIGRl"
    "bHRhIDwgLTFlLTk6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNBbWJpZ3VpdHkoCiAg"
    "ICAgICAgICAgICAgICAiTUVUUklDX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAg"
    "ICAgZiJtZXRyaWMgY291bnRlciByZWdyZXNzZWQ6IHtmaWVsZF9uYW1lfSIsCiAgICAgICAg"
    "ICAgICkKICAgICAgICB2YWx1ZXNbZmllbGRfbmFtZV0gPSAwLjAgaWYgYWJzKGRlbHRhKSA8"
    "PSAxZS05IGVsc2UgZGVsdGEKICAgIG9ic2VydmF0aW9uID0gTWV0cmljRGVsdGFPYnNlcnZh"
    "dGlvbigqKnZhbHVlcykKICAgIGlmICgKICAgICAgICBhYnMoCiAgICAgICAgICAgIG9ic2Vy"
    "dmF0aW9uLmNhY2hlZF9wcm9tcHRfdG9rZW5zCiAgICAgICAgICAgIC0gKG9ic2VydmF0aW9u"
    "LmxvY2FsX2NhY2hlX2hpdCArIG9ic2VydmF0aW9uLmV4dGVybmFsX2t2X3RyYW5zZmVyKQog"
    "ICAgICAgICkKICAgICAgICA+IDFlLTkKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0FtYmlndWl0eSgKICAgICAgICAgICAgIk1FVFJJQ19TRU1BTlRJQ19GQUlMVVJFIiwKICAg"
    "ICAgICAgICAgInJlcXVlc3QgbWV0cmljLWRlbHRhIGNhY2hlZC1zb3VyY2UgaW52YXJpYW50"
    "IGlzIHZpb2xhdGVkIiwKICAgICAgICApCiAgICByZXR1cm4gb2JzZXJ2YXRpb24KCgpkZWYg"
    "X3ByZWZpeF9jb250ZXh0KHByZWZpeF92YXJpYW50OiBzdHIpIC0+IHN0cjoKICAgIGlmIHBy"
    "ZWZpeF92YXJpYW50ID09ICJBIjoKICAgICAgICByZXR1cm4gU1lOVEhFVElDX0NBQ0hFX0NP"
    "TlRFWFRfQQogICAgaWYgcHJlZml4X3ZhcmlhbnQgPT0gIkIiOgogICAgICAgIHJldHVybiBT"
    "WU5USEVUSUNfQ0FDSEVfQ09OVEVYVF9CCiAgICByYWlzZSBSdW50aW1lRXJyb3IoInVua25v"
    "d24gcHJlZml4IHZhcmlhbnQiKQoKCmRlZiByZXF1ZXN0X21lc3NhZ2VzKHByZWZpeF92YXJp"
    "YW50OiBzdHIpIC0+IGxpc3RbZGljdFtzdHIsIHN0cl1dOgogICAgcmV0dXJuIFsKICAgICAg"
    "ICB7InJvbGUiOiAic3lzdGVtIiwgImNvbnRlbnQiOiBTWVNURU1fUFJPTVBUfSwKICAgICAg"
    "ICB7InJvbGUiOiAidXNlciIsICJjb250ZW50IjogX3ByZWZpeF9jb250ZXh0KHByZWZpeF92"
    "YXJpYW50KX0sCiAgICAgICAgeyJyb2xlIjogImFzc2lzdGFudCIsICJjb250ZW50IjogU1lO"
    "VEhFVElDX0FTU0lTVEFOVF9BQ0t9LAogICAgICAgIHsicm9sZSI6ICJ1c2VyIiwgImNvbnRl"
    "bnQiOiBFWFBFQ1RFRF9PQkpFQ1RfQ0FOT05JQ0FMfSwKICAgIF0KCgpkZWYgcmVxdWVzdF9w"
    "YXlsb2FkKHByZWZpeF92YXJpYW50OiBzdHIpIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAg"
    "cmV0dXJuIHsKICAgICAgICAibW9kZWwiOiBTRVJWRURfTU9ERUxfTkFNRSwKICAgICAgICAi"
    "bWVzc2FnZXMiOiByZXF1ZXN0X21lc3NhZ2VzKHByZWZpeF92YXJpYW50KSwKICAgICAgICAi"
    "dGVtcGVyYXR1cmUiOiAwLAogICAgICAgICJ0b3BfcCI6IDEsCiAgICAgICAgInJlcGV0aXRp"
    "b25fcGVuYWx0eSI6IDEuMSwKICAgICAgICAic2VlZCI6IDcsCiAgICAgICAgIm1heF90b2tl"
    "bnMiOiAzMiwKICAgICAgICAic3RyZWFtIjogRmFsc2UsCiAgICB9CgoKZGVmIHRva2VuaXpl"
    "X3BheWxvYWQocHJlZml4X3ZhcmlhbnQ6IHN0cikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAg"
    "ICByZXR1cm4gewogICAgICAgICJtb2RlbCI6IFNFUlZFRF9NT0RFTF9OQU1FLAogICAgICAg"
    "ICJtZXNzYWdlcyI6IHJlcXVlc3RfbWVzc2FnZXMocHJlZml4X3ZhcmlhbnQpLAogICAgICAg"
    "ICJhZGRfZ2VuZXJhdGlvbl9wcm9tcHQiOiBUcnVlLAogICAgICAgICJjb250aW51ZV9maW5h"
    "bF9tZXNzYWdlIjogRmFsc2UsCiAgICAgICAgImFkZF9zcGVjaWFsX3Rva2VucyI6IEZhbHNl"
    "LAogICAgICAgICJyZXR1cm5fdG9rZW5fc3RycyI6IEZhbHNlLAogICAgfQoKCmRlZiB0b2tl"
    "bml6ZV9yZXF1ZXN0KAogICAgd29ya2VyOiBXb3JrZXIsCiAgICByZXF1ZXN0X3JvbGU6IHN0"
    "ciwKICAgIHByZWZpeF92YXJpYW50OiBzdHIsCikgLT4gVG9rZW5JZGVudGl0eU9ic2VydmF0"
    "aW9uOgogICAgcmVzcG9uc2UgPSBwb3N0X2pzb24oCiAgICAgICAgZiJodHRwOi8vMTI3LjAu"
    "MC4xOnt3b3JrZXIucG9ydH0vdG9rZW5pemUiLAogICAgICAgIHRva2VuaXplX3BheWxvYWQo"
    "cHJlZml4X3ZhcmlhbnQpLAogICAgKQogICAgcmF3X3Rva2VucyA9IHJlc3BvbnNlLmdldCgi"
    "dG9rZW5zIikKICAgIGNvdW50ID0gcmVzcG9uc2UuZ2V0KCJjb3VudCIpCiAgICBpZiAoCiAg"
    "ICAgICAgbm90IGlzaW5zdGFuY2UocmF3X3Rva2VucywgbGlzdCkKICAgICAgICBvciBpc2lu"
    "c3RhbmNlKGNvdW50LCBib29sKQogICAgICAgIG9yIG5vdCBpc2luc3RhbmNlKGNvdW50LCBp"
    "bnQpCiAgICAgICAgb3IgY291bnQgIT0gbGVuKHJhd190b2tlbnMpCiAgICApOgogICAgICAg"
    "IHJhaXNlIERpYWdub3N0aWNBbWJpZ3VpdHkoCiAgICAgICAgICAgICJUT0tFTklaRVJfQVJU"
    "SUZBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJzZXJ2ZXIgdG9rZW5pemF0aW9uIHJlc3Bv"
    "bnNlIHNoYXBlIGlzIGludmFsaWQiLAogICAgICAgICkKICAgIHRva2VuczogbGlzdFtpbnRd"
    "ID0gW10KICAgIGZvciByYXdfdG9rZW4gaW4gcmF3X3Rva2VuczoKICAgICAgICBpZiBpc2lu"
    "c3RhbmNlKHJhd190b2tlbiwgYm9vbCkgb3Igbm90IGlzaW5zdGFuY2UocmF3X3Rva2VuLCBp"
    "bnQpOgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljQW1iaWd1aXR5KAogICAgICAgICAg"
    "ICAgICAgIlRPS0VOSVpFUl9BUlRJRkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJz"
    "ZXJ2ZXIgdG9rZW5pemF0aW9uIHJldHVybmVkIGEgbm9uLWludGVnZXIgdG9rZW4gaWQiLAog"
    "ICAgICAgICAgICApCiAgICAgICAgaWYgcmF3X3Rva2VuIDwgMDoKICAgICAgICAgICAgcmFp"
    "c2UgRGlhZ25vc3RpY0FtYmlndWl0eSgKICAgICAgICAgICAgICAgICJUT0tFTklaRVJfQVJU"
    "SUZBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICAgICAic2VydmVyIHRva2VuaXphdGlvbiBy"
    "ZXR1cm5lZCBhIG5lZ2F0aXZlIHRva2VuIGlkIiwKICAgICAgICAgICAgKQogICAgICAgIHRv"
    "a2Vucy5hcHBlbmQocmF3X3Rva2VuKQogICAgdG9rZW5fYnl0ZXMgPSBjYW5vbmljYWxfanNv"
    "bih0b2tlbnMpLmVuY29kZSgidXRmLTgiKQogICAgcmV0dXJuIFRva2VuSWRlbnRpdHlPYnNl"
    "cnZhdGlvbigKICAgICAgICByZXF1ZXN0X3JvbGU9cmVxdWVzdF9yb2xlLAogICAgICAgIHBy"
    "ZWZpeF92YXJpYW50PXByZWZpeF92YXJpYW50LAogICAgICAgIHRva2VuX2NvdW50PWxlbih0"
    "b2tlbnMpLAogICAgICAgIHRva2VuX3NoYTI1Nj1zaGEyNTZfYnl0ZXModG9rZW5fYnl0ZXMp"
    "LAogICAgICAgIHRva2VuX2lkcz10dXBsZSh0b2tlbnMpLAogICAgKQoKCmRlZiBjb21tb25f"
    "cHJlZml4X3Rva2VuX2NvdW50KAogICAgbGVmdDogVG9rZW5JZGVudGl0eU9ic2VydmF0aW9u"
    "LAogICAgcmlnaHQ6IFRva2VuSWRlbnRpdHlPYnNlcnZhdGlvbiwKKSAtPiBpbnQ6CiAgICBj"
    "b3VudCA9IDAKICAgIGZvciBsZWZ0X3Rva2VuLCByaWdodF90b2tlbiBpbiB6aXAobGVmdC50"
    "b2tlbl9pZHMsIHJpZ2h0LnRva2VuX2lkcywgc3RyaWN0PUZhbHNlKToKICAgICAgICBpZiBs"
    "ZWZ0X3Rva2VuICE9IHJpZ2h0X3Rva2VuOgogICAgICAgICAgICBicmVhawogICAgICAgIGNv"
    "dW50ICs9IDEKICAgIHJldHVybiBjb3VudAoKCmRlZiBjYWNoZWFibGVfY29tbW9uX3ByZWZp"
    "eF9ib3VuZCgKICAgIGxlZnQ6IFRva2VuSWRlbnRpdHlPYnNlcnZhdGlvbiwKICAgIHJpZ2h0"
    "OiBUb2tlbklkZW50aXR5T2JzZXJ2YXRpb24sCikgLT4gaW50OgogICAgY29tbW9uID0gY29t"
    "bW9uX3ByZWZpeF90b2tlbl9jb3VudChsZWZ0LCByaWdodCkKICAgIHJldHVybiAoY29tbW9u"
    "IC8vIENBQ0hFX0JMT0NLX1NJWkUpICogQ0FDSEVfQkxPQ0tfU0laRQoKCmRlZiB0b2tlbl9p"
    "ZGVudGl0eV9ldmlkZW5jZSgKICAgIG9ic2VydmF0aW9uOiBUb2tlbklkZW50aXR5T2JzZXJ2"
    "YXRpb24sCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICByZXR1cm4gewogICAgICAgICJy"
    "ZXF1ZXN0X3JvbGUiOiBvYnNlcnZhdGlvbi5yZXF1ZXN0X3JvbGUsCiAgICAgICAgInByZWZp"
    "eF92YXJpYW50Ijogb2JzZXJ2YXRpb24ucHJlZml4X3ZhcmlhbnQsCiAgICAgICAgInRva2Vu"
    "X2NvdW50Ijogb2JzZXJ2YXRpb24udG9rZW5fY291bnQsCiAgICAgICAgInRva2VuX3NoYTI1"
    "NiI6IG9ic2VydmF0aW9uLnRva2VuX3NoYTI1NiwKICAgICAgICAidG9rZW5faWRzIjogbGlz"
    "dChvYnNlcnZhdGlvbi50b2tlbl9pZHMpLAogICAgfQoKCmRlZiBfcmVhZF9wcmVfcmVxdWVz"
    "dF90b2tlbl9pZGVudGl0eV9qb3VybmFsKCkgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBp"
    "ZiBub3QgUFJFX1JFUVVFU1RfVE9LRU5fSURFTlRJVFlfSk9VUk5BTC5pc19maWxlKCk6CiAg"
    "ICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSBq"
    "b3VybmFsIGlzIHVuYXZhaWxhYmxlIikKICAgIHRyeToKICAgICAgICByYXc6IG9iamVjdCA9"
    "IGpzb24ubG9hZHMoUFJFX1JFUVVFU1RfVE9LRU5fSURFTlRJVFlfSk9VUk5BTC5yZWFkX3Rl"
    "eHQoZW5jb2Rpbmc9InV0Zi04IikpCiAgICBleGNlcHQgKGpzb24uSlNPTkRlY29kZUVycm9y"
    "LCBVbmljb2RlRGVjb2RlRXJyb3IpIGFzIGVycm9yOgogICAgICAgIHJhaXNlIFJ1bnRpbWVF"
    "cnJvcigicHJlLXJlcXVlc3QgdG9rZW4taWRlbnRpdHkgam91cm5hbCBpcyBpbnZhbGlkIikg"
    "ZnJvbSBlcnJvcgogICAgaWYgbm90IGlzaW5zdGFuY2UocmF3LCBkaWN0KToKICAgICAgICBy"
    "YWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuLWlkZW50aXR5IGpvdXJuYWwg"
    "cm9vdCBpcyBpbnZhbGlkIikKICAgIHJlc3VsdDogZGljdFtzdHIsIG9iamVjdF0gPSB7fQog"
    "ICAgZm9yIGtleSwgdmFsdWUgaW4gcmF3Lml0ZW1zKCk6CiAgICAgICAgaWYgbm90IGlzaW5z"
    "dGFuY2Uoa2V5LCBzdHIpOgogICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoInByZS1y"
    "ZXF1ZXN0IHRva2VuLWlkZW50aXR5IGpvdXJuYWwga2V5IGlzIGludmFsaWQiKQogICAgICAg"
    "IHJlc3VsdFtrZXldID0gdmFsdWUKICAgIHJldHVybiByZXN1bHQKCgpkZWYgaW5pdGlhbGl6"
    "ZV9wcmVfcmVxdWVzdF90b2tlbl9pZGVudGl0eV9qb3VybmFsKCkgLT4gTm9uZToKICAgIGlm"
    "IFBSRV9SRVFVRVNUX1RPS0VOX0lERU5USVRZX0pPVVJOQUwuZXhpc3RzKCk6CiAgICAgICAg"
    "cmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSBqb3VybmFs"
    "IGFscmVhZHkgZXhpc3RzIikKICAgIHdyaXRlX2pzb24oCiAgICAgICAgUFJFX1JFUVVFU1Rf"
    "VE9LRU5fSURFTlRJVFlfSk9VUk5BTCwKICAgICAgICB7CiAgICAgICAgICAgICJzY2hlbWFf"
    "dmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgICAgICJqb3VybmFsX2lkIjogKAogICAgICAg"
    "ICAgICAgICAgImF1cmFnYXRld2F5LXA0LXA1LWNvbXBvc2l0aW9uLXJlbWVkaWF0aW9uLXBy"
    "ZS1yZXF1ZXN0LXRva2VuLWlkZW50aXR5LXYxIgogICAgICAgICAgICApLAogICAgICAgICAg"
    "ICAiZW50cmllcyI6IFtdLAogICAgICAgICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZh"
    "bHNlLAogICAgICAgICAgICAicmF3X21vZGVsX291dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAog"
    "ICAgICAgIH0sCiAgICApCgoKZGVmIHBlcnNpc3RfcHJlX3JlcXVlc3RfdG9rZW5faWRlbnRp"
    "dHkoCiAgICByZXF1ZXN0X29yZGluYWw6IGludCwKICAgIHRva2VuX2lkZW50aXR5OiBUb2tl"
    "bklkZW50aXR5T2JzZXJ2YXRpb24sCiAgICBwYXlsb2FkX3NoYTI1Njogc3RyLAopIC0+IE5v"
    "bmU6CiAgICBqb3VybmFsID0gX3JlYWRfcHJlX3JlcXVlc3RfdG9rZW5faWRlbnRpdHlfam91"
    "cm5hbCgpCiAgICBlbnRyaWVzID0gam91cm5hbC5nZXQoImVudHJpZXMiKQogICAgaWYgbm90"
    "IGlzaW5zdGFuY2UoZW50cmllcywgbGlzdCk6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9y"
    "KCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSBqb3VybmFsIGVudHJpZXMgYXJlIGludmFs"
    "aWQiKQogICAgaWYgcmVxdWVzdF9vcmRpbmFsICE9IGxlbihlbnRyaWVzKSArIDE6CiAgICAg"
    "ICAgcmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSByZXF1"
    "ZXN0IG9yZGluYWwgZHJpZnRlZCIpCiAgICBpZiByZS5mdWxsbWF0Y2gociJbMC05YS1mXXs2"
    "NH0iLCBwYXlsb2FkX3NoYTI1NikgaXMgTm9uZToKICAgICAgICByYWlzZSBSdW50aW1lRXJy"
    "b3IoInByZS1yZXF1ZXN0IHBheWxvYWQgaWRlbnRpdHkgaXMgaW52YWxpZCIpCiAgICBlbnRy"
    "eSA9IHsKICAgICAgICAicmVxdWVzdF9vcmRpbmFsIjogcmVxdWVzdF9vcmRpbmFsLAogICAg"
    "ICAgICJyZXF1ZXN0X3JvbGUiOiB0b2tlbl9pZGVudGl0eS5yZXF1ZXN0X3JvbGUsCiAgICAg"
    "ICAgInByZWZpeF92YXJpYW50IjogdG9rZW5faWRlbnRpdHkucHJlZml4X3ZhcmlhbnQsCiAg"
    "ICAgICAgInRva2VuX2NvdW50IjogdG9rZW5faWRlbnRpdHkudG9rZW5fY291bnQsCiAgICAg"
    "ICAgInRva2VuX3NoYTI1NiI6IHRva2VuX2lkZW50aXR5LnRva2VuX3NoYTI1NiwKICAgICAg"
    "ICAidG9rZW5faWRzIjogbGlzdCh0b2tlbl9pZGVudGl0eS50b2tlbl9pZHMpLAogICAgICAg"
    "ICJwYXlsb2FkX3NoYTI1NiI6IHBheWxvYWRfc2hhMjU2LAogICAgICAgICJwZXJzaXN0ZWRf"
    "YmVmb3JlX21vZGVsX3JlcXVlc3QiOiBUcnVlLAogICAgfQogICAgdXBkYXRlZCA9IHsqKmpv"
    "dXJuYWwsICJlbnRyaWVzIjogWyplbnRyaWVzLCBlbnRyeV19CiAgICB3cml0ZV9qc29uKFBS"
    "RV9SRVFVRVNUX1RPS0VOX0lERU5USVRZX0pPVVJOQUwsIHVwZGF0ZWQpCgoKZGVmIHZhbGlk"
    "YXRlX3N0cnVjdHVyZWRfcmVzcG9uc2UoCiAgICBjb250ZW50OiBzdHIsCiAgICBleHBlY3Rl"
    "ZF9qc29uOiBzdHIsCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICB0cnk6CiAgICAgICAg"
    "b2JzZXJ2ZWQgPSBqc29uLmxvYWRzKGNvbnRlbnQpCiAgICAgICAgZXhwZWN0ZWQgPSBqc29u"
    "LmxvYWRzKGV4cGVjdGVkX2pzb24pCiAgICBleGNlcHQganNvbi5KU09ORGVjb2RlRXJyb3Ig"
    "YXMgZXJyb3I6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJtb2RlbCByZXNwb25zZSBp"
    "cyBub3QgdmFsaWQgSlNPTiIpIGZyb20gZXJyb3IKICAgIGlmIG5vdCBpc2luc3RhbmNlKG9i"
    "c2VydmVkLCBkaWN0KSBvciBub3QgaXNpbnN0YW5jZShleHBlY3RlZCwgZGljdCk6CiAgICAg"
    "ICAgcmFpc2UgUnVudGltZUVycm9yKCJzdHJ1Y3R1cmVkIHJlc3BvbnNlIHJvb3QgbXVzdCBi"
    "ZSBvbmUgb2JqZWN0IikKICAgIGlmIG9ic2VydmVkICE9IGV4cGVjdGVkOgogICAgICAgIHJh"
    "aXNlIFJ1bnRpbWVFcnJvcigic3RydWN0dXJlZCByZXNwb25zZSBkaWZmZXJzIGZyb20gdGhl"
    "IHJlcXVlc3RlZCBvYmplY3QiKQogICAgcmV0dXJuIG9ic2VydmVkCgoKZGVmIHJ1bl9zdHJ1"
    "Y3R1cmVkX3JlcXVlc3QoCiAgICB3b3JrZXI6IFdvcmtlciwKICAgIHJlcXVlc3Rfcm9sZTog"
    "c3RyLAogICAgcHJlZml4X3ZhcmlhbnQ6IHN0ciwKICAgIGNvdW50ZXJzOiBkaWN0W3N0ciwg"
    "aW50XSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHRva2VuX2lkZW50aXR5ID0gdG9r"
    "ZW5pemVfcmVxdWVzdCgKICAgICAgICB3b3JrZXIsCiAgICAgICAgcmVxdWVzdF9yb2xlPXJl"
    "cXVlc3Rfcm9sZSwKICAgICAgICBwcmVmaXhfdmFyaWFudD1wcmVmaXhfdmFyaWFudCwKICAg"
    "ICkKICAgIHJlcXVlc3Rfb3JkaW5hbCA9IGNvdW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdICsg"
    "MQogICAgcGF5bG9hZCA9IHJlcXVlc3RfcGF5bG9hZChwcmVmaXhfdmFyaWFudCkKICAgIHBl"
    "cnNpc3RfcHJlX3JlcXVlc3RfdG9rZW5faWRlbnRpdHkoCiAgICAgICAgcmVxdWVzdF9vcmRp"
    "bmFsLAogICAgICAgIHRva2VuX2lkZW50aXR5LAogICAgICAgIHNoYTI1Nl90ZXh0KGNhbm9u"
    "aWNhbF9qc29uKHBheWxvYWQpKSwKICAgICkKICAgIGJlZm9yZSA9IHdvcmtlci5tZXRyaWNf"
    "c25hcHNob3QoKQogICAgY29uc3VtZV9hY3Rpb25zKGNvdW50ZXJzLCAibW9kZWxfcmVxdWVz"
    "dHMiKQogICAgcmVxdWVzdF9pZCA9ICgKICAgICAgICBmIntyZXF1ZXN0X3JvbGUubG93ZXIo"
    "KX0tIgogICAgICAgIGYie3dvcmtlci53b3JrZXJfaWR9LWd7d29ya2VyLmdlbmVyYXRpb259"
    "LSIKICAgICAgICBmIntjb3VudGVyc1snbW9kZWxfcmVxdWVzdHMnXX0iCiAgICApCiAgICBy"
    "ZXNwb25zZSA9IHBvc3RfanNvbigKICAgICAgICBmImh0dHA6Ly8xMjcuMC4wLjE6e3dvcmtl"
    "ci5wb3J0fS92MS9jaGF0L2NvbXBsZXRpb25zIiwKICAgICAgICBwYXlsb2FkLAogICAgKQog"
    "ICAgYWZ0ZXIgPSB3b3JrZXIubWV0cmljX3NuYXBzaG90KCkKICAgIGRlbHRhID0gbWV0cmlj"
    "X2RlbHRhKGJlZm9yZSwgYWZ0ZXIpCiAgICB1c2FnZSA9IHJlc3BvbnNlLmdldCgidXNhZ2Ui"
    "KQogICAgY2hvaWNlcyA9IHJlc3BvbnNlLmdldCgiY2hvaWNlcyIpCiAgICBpZiBub3QgaXNp"
    "bnN0YW5jZSh1c2FnZSwgZGljdCk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAg"
    "InJlc3BvbnNlIHVzYWdlIGlzIG1pc3NpbmciLAogICAgICAgICkKICAgIGlmIG5vdCBpc2lu"
    "c3RhbmNlKGNob2ljZXMsIGxpc3QpIG9yIGxlbihjaG9pY2VzKSAhPSAxOgogICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZB"
    "SUxVUkUiLAogICAgICAgICAgICAicmVzcG9uc2UgY2hvaWNlcyBhcmUgaW52YWxpZCIsCiAg"
    "ICAgICAgKQogICAgY2hvaWNlID0gY2hvaWNlc1swXQogICAgaWYgbm90IGlzaW5zdGFuY2Uo"
    "Y2hvaWNlLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInJlc3BvbnNl"
    "IGNob2ljZSBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICBtZXNzYWdlID0gY2hvaWNlLmdl"
    "dCgibWVzc2FnZSIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShtZXNzYWdlLCBkaWN0KToKICAg"
    "ICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05U"
    "UkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInJlc3BvbnNlIG1lc3NhZ2UgaXMgaW52YWxp"
    "ZCIsCiAgICAgICAgKQogICAgY29udGVudCA9IG1lc3NhZ2UuZ2V0KCJjb250ZW50IikKICAg"
    "IGlmIG5vdCBpc2luc3RhbmNlKGNvbnRlbnQsIHN0cikgb3Igbm90IGNvbnRlbnQ6CiAgICAg"
    "ICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJB"
    "Q1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJyZXNwb25zZSBjb250ZW50IGlzIGVtcHR5IiwK"
    "ICAgICAgICApCiAgICBvdXRwdXRfdG9rZW5zID0gdXNhZ2UuZ2V0KCJjb21wbGV0aW9uX3Rv"
    "a2VucyIpCiAgICBwcm9tcHRfdG9rZW5zID0gdXNhZ2UuZ2V0KCJwcm9tcHRfdG9rZW5zIikK"
    "ICAgIGlmICgKICAgICAgICBpc2luc3RhbmNlKG91dHB1dF90b2tlbnMsIGJvb2wpCiAgICAg"
    "ICAgb3Igbm90IGlzaW5zdGFuY2Uob3V0cHV0X3Rva2VucywgaW50KQogICAgICAgIG9yIG5v"
    "dCAxIDw9IG91dHB1dF90b2tlbnMgPD0gMzIKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAg"
    "ICAgICAgICAgICJjb21wbGV0aW9uIHRva2VuIGJ1ZGdldCBkcmlmdGVkIiwKICAgICAgICAp"
    "CiAgICBpZiBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGJvb2wpIG9yIG5vdCBpc2luc3Rh"
    "bmNlKHByb21wdF90b2tlbnMsIGludCkgb3IgcHJvbXB0X3Rva2VucyA8PSAwOgogICAgICAg"
    "IHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNU"
    "X0ZBSUxVUkUiLAogICAgICAgICAgICAicHJvbXB0IHRva2VuIGNvdW50IGlzIGludmFsaWQi"
    "LAogICAgICAgICkKICAgIHN0cnVjdHVyZWQgPSB2YWxpZGF0ZV9zdHJ1Y3R1cmVkX3Jlc3Bv"
    "bnNlKAogICAgICAgIGNvbnRlbnQsCiAgICAgICAgRVhQRUNURURfT0JKRUNUX0NBTk9OSUNB"
    "TCwKICAgICkKICAgIG91dHB1dF9zaGEgPSBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihz"
    "dHJ1Y3R1cmVkKSkKICAgIHJvdXRlID0gUm91dGVPYnNlcnZhdGlvbigKICAgICAgICByZXF1"
    "ZXN0X2lkPXJlcXVlc3RfaWQsCiAgICAgICAgcmVxdWVzdF9yb2xlPXJlcXVlc3Rfcm9sZSwK"
    "ICAgICAgICBpbnRlbmRlZF93b3JrZXI9d29ya2VyLndvcmtlcl9pZCwKICAgICAgICByZWFs"
    "aXplZF93b3JrZXI9d29ya2VyLndvcmtlcl9pZCwKICAgICAgICB3b3JrZXJfZ2VuZXJhdGlv"
    "bj13b3JrZXIuZ2VuZXJhdGlvbiwKICAgICAgICBlbmRwb2ludF9wb3J0PXdvcmtlci5wb3J0"
    "LAogICAgICAgIG1ldHJpY19lbmRwb2ludF9pZGVudGl0eT1zaGEyNTZfdGV4dChmImh0dHA6"
    "Ly8xMjcuMC4wLjE6e3dvcmtlci5wb3J0fS9tZXRyaWNzIiksCiAgICAgICAgcm91dGVfcmVh"
    "c29uPSJESVJFQ1RfTE9PUEJBQ0tfRU5EUE9JTlQiLAogICAgICAgIGZhbGxiYWNrX3JlYXNv"
    "bj1Ob25lLAogICAgICAgIG91dHB1dF9zaGEyNTY9b3V0cHV0X3NoYSwKICAgICkKICAgIHJl"
    "dHVybiB7CiAgICAgICAgInJlcXVlc3RfaWQiOiByZXF1ZXN0X2lkLAogICAgICAgICJyZXF1"
    "ZXN0X3JvbGUiOiByZXF1ZXN0X3JvbGUsCiAgICAgICAgInByZWZpeF92YXJpYW50IjogcHJl"
    "Zml4X3ZhcmlhbnQsCiAgICAgICAgInRva2VuX2lkZW50aXR5IjogdG9rZW5faWRlbnRpdHks"
    "CiAgICAgICAgInJlc3BvbnNlX2NvbnRlbnRfc2hhMjU2Ijogc2hhMjU2X3RleHQoY29udGVu"
    "dCksCiAgICAgICAgInN0cnVjdHVyZWRfb3V0cHV0X3NoYTI1NiI6IG91dHB1dF9zaGEsCiAg"
    "ICAgICAgInN0cnVjdHVyZWRfb3V0cHV0X3ZhbGlkIjogVHJ1ZSwKICAgICAgICAiZmluaXNo"
    "X3JlYXNvbiI6IGNob2ljZS5nZXQoImZpbmlzaF9yZWFzb24iKSwKICAgICAgICAicHJvbXB0"
    "X3Rva2VucyI6IHByb21wdF90b2tlbnMsCiAgICAgICAgImNvbXBsZXRpb25fdG9rZW5zIjog"
    "b3V0cHV0X3Rva2VucywKICAgICAgICAibWV0cmljX2RlbHRhIjogZGVsdGEsCiAgICAgICAg"
    "InJvdXRlX29ic2VydmF0aW9uIjogcm91dGUsCiAgICB9CgoKZGVmIGdwdV9pbnZlbnRvcnko"
    "KSAtPiBkaWN0W2ludCwgZGljdFtzdHIsIG9iamVjdF1dOgogICAgcmVzdWx0ID0gc3VicHJv"
    "Y2Vzcy5ydW4oCiAgICAgICAgWwogICAgICAgICAgICAibnZpZGlhLXNtaSIsCiAgICAgICAg"
    "ICAgICgiLS1xdWVyeS1ncHU9aW5kZXgsdXVpZCxwY2kuYnVzX2lkLG5hbWUsY29tcHV0ZV9j"
    "YXAsbWVtb3J5LnVzZWQiKSwKICAgICAgICAgICAgIi0tZm9ybWF0PWNzdixub2hlYWRlcixu"
    "b3VuaXRzIiwKICAgICAgICBdLAogICAgICAgIGNoZWNrPUZhbHNlLAogICAgICAgIGNhcHR1"
    "cmVfb3V0cHV0PVRydWUsCiAgICAgICAgdGV4dD1UcnVlLAogICAgICAgIHRpbWVvdXQ9MzAs"
    "CiAgICApCiAgICBpZiByZXN1bHQucmV0dXJuY29kZSAhPSAwOgogICAgICAgIHJhaXNlIFJ1"
    "bnRpbWVFcnJvcigiR1BVIGlkZW50aXR5IGNvbW1hbmQgZmFpbGVkIikKICAgIGludmVudG9y"
    "eTogZGljdFtpbnQsIGRpY3Rbc3RyLCBvYmplY3RdXSA9IHt9CiAgICBmb3IgbGluZSBpbiBy"
    "ZXN1bHQuc3Rkb3V0LnNwbGl0bGluZXMoKToKICAgICAgICBwYXJ0cyA9IHR1cGxlKHBhcnQu"
    "c3RyaXAoKSBmb3IgcGFydCBpbiBsaW5lLnNwbGl0KCIsIikpCiAgICAgICAgaWYgbGVuKHBh"
    "cnRzKSAhPSA2OgogICAgICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoIkdQVSBpZGVudGl0"
    "eSBvdXRwdXQgaXMgbWFsZm9ybWVkIikKICAgICAgICBpbmRleCA9IGludChwYXJ0c1swXSkK"
    "ICAgICAgICBuYW1lID0gIlRlc2xhIFQ0IiBpZiBwYXJ0c1szXSA9PSAiTlZJRElBIFQ0IiBl"
    "bHNlIHBhcnRzWzNdCiAgICAgICAgaWYgaW5kZXggbm90IGluIHswLCAxfToKICAgICAgICAg"
    "ICAgcmFpc2UgUnVudGltZUVycm9yKCJ1bmV4cGVjdGVkIEdQVSBpbmRleCIpCiAgICAgICAg"
    "aWYgbmFtZSAhPSBFWFBFQ1RFRF9HUFVfTkFNRSBvciBwYXJ0c1s0XSAhPSBFWFBFQ1RFRF9D"
    "T01QVVRFX0NBUEFCSUxJVFk6CiAgICAgICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigiR1BV"
    "IGlkZW50aXR5IGRyaWZ0ZWQiKQogICAgICAgIGludmVudG9yeVtpbmRleF0gPSB7CiAgICAg"
    "ICAgICAgICJpbmRleCI6IGluZGV4LAogICAgICAgICAgICAidXVpZCI6IHBhcnRzWzFdLAog"
    "ICAgICAgICAgICAicGNpX2J1c19pZCI6IHBhcnRzWzJdLAogICAgICAgICAgICAibmFtZSI6"
    "IG5hbWUsCiAgICAgICAgICAgICJjb21wdXRlX2NhcGFiaWxpdHkiOiBwYXJ0c1s0XSwKICAg"
    "ICAgICAgICAgIm1lbW9yeV91c2VkX21pYiI6IGludChwYXJ0c1s1XSksCiAgICAgICAgfQog"
    "ICAgaWYgc2V0KGludmVudG9yeSkgIT0gezAsIDF9OgogICAgICAgIHJhaXNlIFJ1bnRpbWVF"
    "cnJvcigiZXhwZWN0ZWQgZXhhY3RseSB0d28gR1BVcyIpCiAgICByZXR1cm4gaW52ZW50b3J5"
    "CgoKZGVmIGdwdV91dWlkX21hcCgpIC0+IGRpY3RbaW50LCBzdHJdOgogICAgcmV0dXJuIHtp"
    "bmRleDogc3RyKGlkZW50aXR5WyJ1dWlkIl0pIGZvciBpbmRleCwgaWRlbnRpdHkgaW4gZ3B1"
    "X2ludmVudG9yeSgpLml0ZW1zKCl9CgoKZGVmIGdwdV9tZW1vcnlfdXNlZF9taWIoZ3B1X3V1"
    "aWQ6IHN0cikgLT4gaW50OgogICAgZm9yIGlkZW50aXR5IGluIGdwdV9pbnZlbnRvcnkoKS52"
    "YWx1ZXMoKToKICAgICAgICBpZiBpZGVudGl0eVsidXVpZCJdID09IGdwdV91dWlkOgogICAg"
    "ICAgICAgICByZXR1cm4gaW50KGlkZW50aXR5WyJtZW1vcnlfdXNlZF9taWIiXSkgICMgdHlw"
    "ZTogaWdub3JlW25vLWFueS1yZXR1cm4sIGNhbGwtb3ZlcmxvYWRdCiAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoIkdQVSBVVUlEIGlzIHVuYXZhaWxhYmxlIikKCgpkZWYgcHJvY2Vzc19zdGFy"
    "dF90aWNrcyhwaWQ6IGludCkgLT4gaW50OgogICAgc3RhdF9wYXRoID0gUGF0aChmIi9wcm9j"
    "L3twaWR9L3N0YXQiKQogICAgcGF5bG9hZCA9IHN0YXRfcGF0aC5yZWFkX3RleHQoZW5jb2Rp"
    "bmc9InV0Zi04IikKICAgIGNsb3NpbmcgPSBwYXlsb2FkLnJmaW5kKCIpIikKICAgIGlmIGNs"
    "b3NpbmcgPCAwOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicHJvY2VzcyBzdGF0IHJl"
    "Y29yZCBpcyBtYWxmb3JtZWQiKQogICAgZmllbGRzID0gcGF5bG9hZFtjbG9zaW5nICsgMiA6"
    "XS5zcGxpdCgpCiAgICBpZiBsZW4oZmllbGRzKSA8PSAxOToKICAgICAgICByYWlzZSBSdW50"
    "aW1lRXJyb3IoInByb2Nlc3Mgc3RhdCByZWNvcmQgaXMgaW5jb21wbGV0ZSIpCiAgICByZXR1"
    "cm4gaW50KGZpZWxkc1sxOV0pCgoKZGVmIHByb2Nlc3NfcGFyZW50X21hcCgpIC0+IGRpY3Rb"
    "aW50LCBpbnRdOgogICAgcmVzdWx0ID0gc3VicHJvY2Vzcy5ydW4oCiAgICAgICAgWyJwcyIs"
    "ICItZW8iLCAicGlkPSxwcGlkPSJdLAogICAgICAgIGNoZWNrPUZhbHNlLAogICAgICAgIGNh"
    "cHR1cmVfb3V0cHV0PVRydWUsCiAgICAgICAgdGV4dD1UcnVlLAogICAgICAgIHRpbWVvdXQ9"
    "MTAsCiAgICApCiAgICBpZiByZXN1bHQucmV0dXJuY29kZSAhPSAwOgogICAgICAgIHJhaXNl"
    "IFJ1bnRpbWVFcnJvcigicHJvY2VzcyB0b3BvbG9neSBjb21tYW5kIGZhaWxlZCIpCiAgICBt"
    "YXBwaW5nOiBkaWN0W2ludCwgaW50XSA9IHt9CiAgICBmb3IgbGluZSBpbiByZXN1bHQuc3Rk"
    "b3V0LnNwbGl0bGluZXMoKToKICAgICAgICBmaWVsZHMgPSBsaW5lLnNwbGl0KCkKICAgICAg"
    "ICBpZiBsZW4oZmllbGRzKSA9PSAyOgogICAgICAgICAgICBtYXBwaW5nW2ludChmaWVsZHNb"
    "MF0pXSA9IGludChmaWVsZHNbMV0pCiAgICByZXR1cm4gbWFwcGluZwoKCmRlZiBkZXNjZW5k"
    "YW50cyhyb290X3BpZDogaW50LCBwYXJlbnRzOiBkaWN0W2ludCwgaW50XSkgLT4gc2V0W2lu"
    "dF06CiAgICByZXN1bHQgPSB7cm9vdF9waWR9CiAgICBjaGFuZ2VkID0gVHJ1ZQogICAgd2hp"
    "bGUgY2hhbmdlZDoKICAgICAgICBjaGFuZ2VkID0gRmFsc2UKICAgICAgICBmb3IgcGlkLCBw"
    "YXJlbnQgaW4gcGFyZW50cy5pdGVtcygpOgogICAgICAgICAgICBpZiBwYXJlbnQgaW4gcmVz"
    "dWx0IGFuZCBwaWQgbm90IGluIHJlc3VsdDoKICAgICAgICAgICAgICAgIHJlc3VsdC5hZGQo"
    "cGlkKQogICAgICAgICAgICAgICAgY2hhbmdlZCA9IFRydWUKICAgIHJldHVybiByZXN1bHQK"
    "CgpkZWYgY29tcHV0ZV9wcm9jZXNzZXMoKSAtPiBkaWN0W2ludCwgc3RyXToKICAgIHJlc3Vs"
    "dCA9IHN1YnByb2Nlc3MucnVuKAogICAgICAgIFsKICAgICAgICAgICAgIm52aWRpYS1zbWki"
    "LAogICAgICAgICAgICAiLS1xdWVyeS1jb21wdXRlLWFwcHM9cGlkLGdwdV91dWlkIiwKICAg"
    "ICAgICAgICAgIi0tZm9ybWF0PWNzdixub2hlYWRlcixub3VuaXRzIiwKICAgICAgICBdLAog"
    "ICAgICAgIGNoZWNrPUZhbHNlLAogICAgICAgIGNhcHR1cmVfb3V0cHV0PVRydWUsCiAgICAg"
    "ICAgdGV4dD1UcnVlLAogICAgICAgIHRpbWVvdXQ9MzAsCiAgICApCiAgICBpZiByZXN1bHQu"
    "cmV0dXJuY29kZSAhPSAwOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigiR1BVIHByb2Nl"
    "c3MgY29tbWFuZCBmYWlsZWQiKQogICAgbWFwcGluZzogZGljdFtpbnQsIHN0cl0gPSB7fQog"
    "ICAgZm9yIGxpbmUgaW4gcmVzdWx0LnN0ZG91dC5zcGxpdGxpbmVzKCk6CiAgICAgICAgcGFy"
    "dHMgPSB0dXBsZShwYXJ0LnN0cmlwKCkgZm9yIHBhcnQgaW4gbGluZS5zcGxpdCgiLCIpKQog"
    "ICAgICAgIGlmIGxlbihwYXJ0cykgPT0gMiBhbmQgcGFydHNbMF0uaXNkaWdpdCgpOgogICAg"
    "ICAgICAgICBtYXBwaW5nW2ludChwYXJ0c1swXSldID0gcGFydHNbMV0KICAgIHJldHVybiBt"
    "YXBwaW5nCgoKZGVmIHZhbGlkYXRlX3Byb2Nlc3NfaXNvbGF0aW9uKAogICAgd29ya2VyXzE6"
    "IFdvcmtlciwKICAgIHdvcmtlcl8yOiBXb3JrZXIsCikgLT4gZGljdFtzdHIsIG9iamVjdF06"
    "CiAgICBpZiB3b3JrZXJfMS5wcm9jZXNzIGlzIE5vbmUgb3Igd29ya2VyXzIucHJvY2VzcyBp"
    "cyBOb25lOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAi"
    "UDZfUFJPQ0VTU19JU09MQVRJT05fRkFJTEVEIiwKICAgICAgICAgICAgIndvcmtlcnMgYXJl"
    "IG5vdCBydW5uaW5nIiwKICAgICAgICApCiAgICBwYXJlbnRzID0gcHJvY2Vzc19wYXJlbnRf"
    "bWFwKCkKICAgIHdvcmtlcl8xX3BpZHMgPSBkZXNjZW5kYW50cyh3b3JrZXJfMS5wcm9jZXNz"
    "LnBpZCwgcGFyZW50cykKICAgIHdvcmtlcl8yX3BpZHMgPSBkZXNjZW5kYW50cyh3b3JrZXJf"
    "Mi5wcm9jZXNzLnBpZCwgcGFyZW50cykKICAgIG92ZXJsYXAgPSB3b3JrZXJfMV9waWRzICYg"
    "d29ya2VyXzJfcGlkcwogICAgaWYgb3ZlcmxhcDoKICAgICAgICByYWlzZSBEaWFnbm9zdGlj"
    "RmFpbHVyZSgKICAgICAgICAgICAgIlA2X1BST0NFU1NfSVNPTEFUSU9OX0ZBSUxFRCIsCiAg"
    "ICAgICAgICAgICJ3b3JrZXIgcHJvY2VzcyB0cmVlcyBvdmVybGFwIiwKICAgICAgICApCiAg"
    "ICByZXR1cm4gewogICAgICAgICJ3b3JrZXJfMV9yb290X3BpZCI6IHdvcmtlcl8xLnByb2Nl"
    "c3MucGlkLAogICAgICAgICJ3b3JrZXJfMl9yb290X3BpZCI6IHdvcmtlcl8yLnByb2Nlc3Mu"
    "cGlkLAogICAgICAgICJ3b3JrZXJfMV9wcm9jZXNzX2NvdW50IjogbGVuKHdvcmtlcl8xX3Bp"
    "ZHMpLAogICAgICAgICJ3b3JrZXJfMl9wcm9jZXNzX2NvdW50IjogbGVuKHdvcmtlcl8yX3Bp"
    "ZHMpLAogICAgICAgICJ3b3JrZXJfcHJvY2Vzc190cmVlc19kaXNqb2ludCI6IFRydWUsCiAg"
    "ICB9CgoKZGVmIHZhbGlkYXRlX2dwdV9pc29sYXRpb24oCiAgICB3b3JrZXJfMTogV29ya2Vy"
    "LAogICAgd29ya2VyXzI6IFdvcmtlciwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlm"
    "IHdvcmtlcl8xLnByb2Nlc3MgaXMgTm9uZSBvciB3b3JrZXJfMi5wcm9jZXNzIGlzIE5vbmU6"
    "CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJQNl9HUFVf"
    "SVNPTEFUSU9OX0ZBSUxFRCIsCiAgICAgICAgICAgICJ3b3JrZXJzIGFyZSBub3QgcnVubmlu"
    "ZyIsCiAgICAgICAgKQogICAgdXVpZHMgPSBncHVfdXVpZF9tYXAoKQogICAgcGFyZW50cyA9"
    "IHByb2Nlc3NfcGFyZW50X21hcCgpCiAgICBncHVfcHJvY2Vzc2VzID0gY29tcHV0ZV9wcm9j"
    "ZXNzZXMoKQogICAgd29ya2VyXzFfcGlkcyA9IGRlc2NlbmRhbnRzKHdvcmtlcl8xLnByb2Nl"
    "c3MucGlkLCBwYXJlbnRzKQogICAgd29ya2VyXzJfcGlkcyA9IGRlc2NlbmRhbnRzKHdvcmtl"
    "cl8yLnByb2Nlc3MucGlkLCBwYXJlbnRzKQogICAgd29ya2VyXzFfZ3B1X3BpZHMgPSB7cGlk"
    "IGZvciBwaWQgaW4gd29ya2VyXzFfcGlkcyBpZiBncHVfcHJvY2Vzc2VzLmdldChwaWQpID09"
    "IHV1aWRzWzBdfQogICAgd29ya2VyXzJfZ3B1X3BpZHMgPSB7cGlkIGZvciBwaWQgaW4gd29y"
    "a2VyXzJfcGlkcyBpZiBncHVfcHJvY2Vzc2VzLmdldChwaWQpID09IHV1aWRzWzFdfQogICAg"
    "d3JvbmdfMSA9IHtwaWQgZm9yIHBpZCBpbiB3b3JrZXJfMV9waWRzIGlmIGdwdV9wcm9jZXNz"
    "ZXMuZ2V0KHBpZCkgPT0gdXVpZHNbMV19CiAgICB3cm9uZ18yID0ge3BpZCBmb3IgcGlkIGlu"
    "IHdvcmtlcl8yX3BpZHMgaWYgZ3B1X3Byb2Nlc3Nlcy5nZXQocGlkKSA9PSB1dWlkc1swXX0K"
    "ICAgIGlmIG5vdCB3b3JrZXJfMV9ncHVfcGlkcyBvciBub3Qgd29ya2VyXzJfZ3B1X3BpZHM6"
    "CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJQNl9HUFVf"
    "SVNPTEFUSU9OX0ZBSUxFRCIsCiAgICAgICAgICAgICJHUFUgcHJvY2VzcyBhdHRyaWJ1dGlv"
    "biBpcyBpbmNvbXBsZXRlIiwKICAgICAgICApCiAgICBpZiB3cm9uZ18xIG9yIHdyb25nXzI6"
    "CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJQNl9HUFVf"
    "SVNPTEFUSU9OX0ZBSUxFRCIsCiAgICAgICAgICAgICJ3b3JrZXIgR1BVIHByb2Nlc3MgaXNv"
    "bGF0aW9uIGZhaWxlZCIsCiAgICAgICAgKQogICAgcmV0dXJuIHsKICAgICAgICAid29ya2Vy"
    "XzFfZ3B1X3Byb2Nlc3NfY291bnQiOiBsZW4od29ya2VyXzFfZ3B1X3BpZHMpLAogICAgICAg"
    "ICJ3b3JrZXJfMl9ncHVfcHJvY2Vzc19jb3VudCI6IGxlbih3b3JrZXJfMl9ncHVfcGlkcyks"
    "CiAgICAgICAgIndvcmtlcl8xX2JvdW5kX3RvX2dwdV8wIjogVHJ1ZSwKICAgICAgICAid29y"
    "a2VyXzJfYm91bmRfdG9fZ3B1XzEiOiBUcnVlLAogICAgfQoKCkNSSVRJQ0FMX05BVElWRV9U"
    "T0tFTlMgPSAoCiAgICAibGliY3VkYSIsCiAgICAibGliY3VkYXJ0IiwKICAgICJsaWJjdWJs"
    "YXMiLAogICAgImxpYmN1c3BhcnNlIiwKICAgICJsaWJudmppdGxpbmsiLAogICAgImxpYmN1"
    "ZG5uIiwKICAgICJsaWJuY2NsIiwKICAgICJsaWJudnJ0YyIsCiAgICAiL3RvcmNoLyIsCiAg"
    "ICAiL3RyaXRvbi8iLAogICAgIi92bGxtLyIsCikKVEFSR0VUX1JFUVVJUkVEX05BVElWRV9U"
    "T0tFTlMgPSAoImxpYmN1c3BhcnNlIiwgImxpYm52aml0bGluayIpCgoKZGVmIG5hdGl2ZV9w"
    "YXRoc19mb3JfcHJvY2VzcyhwaWQ6IGludCkgLT4gc2V0W3N0cl06CiAgICBtYXBzX3BhdGgg"
    "PSBQYXRoKGYiL3Byb2Mve3BpZH0vbWFwcyIpCiAgICBpZiBub3QgbWFwc19wYXRoLmlzX2Zp"
    "bGUoKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlAz"
    "X1A2X05BVElWRV9PUklHSU5fQ0xPU1VSRV9GQUlMRUQiLAogICAgICAgICAgICAicHJvY2Vz"
    "cyBuYXRpdmUtbWFwIGV2aWRlbmNlIGlzIHVuYXZhaWxhYmxlIiwKICAgICAgICApCiAgICBw"
    "YXRoczogc2V0W3N0cl0gPSBzZXQoKQogICAgZm9yIGxpbmUgaW4gbWFwc19wYXRoLnJlYWRf"
    "dGV4dChlbmNvZGluZz0idXRmLTgiKS5zcGxpdGxpbmVzKCk6CiAgICAgICAgZmllbGRzID0g"
    "bGluZS5zcGxpdChtYXhzcGxpdD01KQogICAgICAgIGlmIGxlbihmaWVsZHMpICE9IDYgb3Ig"
    "bm90IGZpZWxkc1s1XS5zdGFydHN3aXRoKCIvIik6CiAgICAgICAgICAgIGNvbnRpbnVlCiAg"
    "ICAgICAgb2JzZXJ2ZWQgPSBmaWVsZHNbNV0ucmVwbGFjZSgiIChkZWxldGVkKSIsICIiKQog"
    "ICAgICAgIGxvd2VyZWQgPSBvYnNlcnZlZC5sb3dlcigpCiAgICAgICAgaWYgYW55KHRva2Vu"
    "IGluIGxvd2VyZWQgZm9yIHRva2VuIGluIENSSVRJQ0FMX05BVElWRV9UT0tFTlMpOgogICAg"
    "ICAgICAgICBwYXRocy5hZGQob2JzZXJ2ZWQpCiAgICByZXR1cm4gcGF0aHMKCgpkZWYgY2xh"
    "c3NpZnlfbmF0aXZlX29yaWdpbihwYXRoOiBzdHIpIC0+IHN0cjoKICAgIG5vcm1hbGl6ZWQg"
    "PSBzdHIoUGF0aChwYXRoKS5yZXNvbHZlKCkpLnJlcGxhY2UoIlxcIiwgIi8iKQogICAgdGFy"
    "Z2V0X3ByZWZpeCA9IHN0cihUQVJHRVRfUk9PVC5yZXNvbHZlKCkpLnJlcGxhY2UoIlxcIiwg"
    "Ii8iKS5yc3RyaXAoIi8iKSArICIvIgogICAgcmVhbF9kcml2ZXJfcHJlZml4ID0gKAogICAg"
    "ICAgIHN0cihQYXRoKFJFQUxfRFJJVkVSX0RJUkVDVE9SWSkucmVzb2x2ZSgpKS5yZXBsYWNl"
    "KCJcXCIsICIvIikucnN0cmlwKCIvIikgKyAiLyIKICAgICkKICAgIGxvd2VyZWQgPSBub3Jt"
    "YWxpemVkLmxvd2VyKCkKICAgIGlmIF9pc19wcm9oaWJpdGVkX2xpYnJhcnlfcGF0aChub3Jt"
    "YWxpemVkKToKICAgICAgICByZXR1cm4gIlJFSkVDVEVEX1NUVUIiCiAgICBpZiBub3JtYWxp"
    "emVkLnN0YXJ0c3dpdGgodGFyZ2V0X3ByZWZpeCk6CiAgICAgICAgcmV0dXJuICJUQVJHRVRf"
    "UlVOVElNRSIKICAgIGlmIG5vcm1hbGl6ZWQuc3RhcnRzd2l0aChyZWFsX2RyaXZlcl9wcmVm"
    "aXgpIGFuZCAibGliY3VkYSIgaW4gbG93ZXJlZDoKICAgICAgICByZXR1cm4gIlJFQUxfTlZJ"
    "RElBX0RSSVZFUiIKICAgIHJldHVybiAiSE9TVF9PUl9BTUJJRU5UX0xJQlJBUlkiCgoKZGVm"
    "IHNhbml0aXplZF9uYXRpdmVfcGF0aChwYXRoOiBzdHIpIC0+IHN0cjoKICAgIG5vcm1hbGl6"
    "ZWQgPSBzdHIoUGF0aChwYXRoKS5yZXNvbHZlKCkpCiAgICB0YXJnZXRfcHJlZml4ID0gc3Ry"
    "KFRBUkdFVF9ST09ULnJlc29sdmUoKSkKICAgIGRyaXZlcl9wcmVmaXggPSBzdHIoUGF0aChS"
    "RUFMX0RSSVZFUl9ESVJFQ1RPUlkpLnJlc29sdmUoKSkKICAgIGlmIG5vcm1hbGl6ZWQuc3Rh"
    "cnRzd2l0aCh0YXJnZXRfcHJlZml4KToKICAgICAgICByZXR1cm4gIjx0YXJnZXRfcnVudGlt"
    "ZT4iICsgbm9ybWFsaXplZFtsZW4odGFyZ2V0X3ByZWZpeCkgOl0KICAgIGlmIG5vcm1hbGl6"
    "ZWQuc3RhcnRzd2l0aChkcml2ZXJfcHJlZml4KToKICAgICAgICByZXR1cm4gIjxyZWFsX2Ry"
    "aXZlcj4iICsgbm9ybWFsaXplZFtsZW4oZHJpdmVyX3ByZWZpeCkgOl0KICAgIHJldHVybiBz"
    "YW5pdGl6ZV9leGNlcnB0KG5vcm1hbGl6ZWQpCgoKZGVmIHZhbGlkYXRlX25hdGl2ZV9vcmln"
    "aW5fY2xvc3VyZSgKICAgICp3b3JrZXJzOiBXb3JrZXIsCikgLT4gZGljdFtzdHIsIG9iamVj"
    "dF06CiAgICBpZiBub3Qgd29ya2VyczoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVy"
    "ZSgKICAgICAgICAgICAgIlAzX1A2X05BVElWRV9PUklHSU5fQ0xPU1VSRV9GQUlMRUQiLAog"
    "ICAgICAgICAgICAibm8gd29ya2VycyB3ZXJlIHN1cHBsaWVkIGZvciBuYXRpdmUtb3JpZ2lu"
    "IGluc3BlY3Rpb24iLAogICAgICAgICkKICAgIHBhcmVudHMgPSBwcm9jZXNzX3BhcmVudF9t"
    "YXAoKQogICAgcmVwb3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0gPSBbXQogICAgcmVq"
    "ZWN0ZWQ6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dID0gW10KICAgIHJlcXVpcmVkOiBkaWN0"
    "W3N0ciwgbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV1dID0gewogICAgICAgIHRva2VuOiBbXSBm"
    "b3IgdG9rZW4gaW4gVEFSR0VUX1JFUVVJUkVEX05BVElWRV9UT0tFTlMKICAgIH0KICAgIGZv"
    "ciB3b3JrZXIgaW4gd29ya2VyczoKICAgICAgICBpZiB3b3JrZXIucHJvY2VzcyBpcyBOb25l"
    "OgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgICAg"
    "ICJQM19QNl9OQVRJVkVfT1JJR0lOX0NMT1NVUkVfRkFJTEVEIiwKICAgICAgICAgICAgICAg"
    "ICJ3b3JrZXIgcHJvY2VzcyBkaXNhcHBlYXJlZCBkdXJpbmcgbmF0aXZlLW9yaWdpbiBpbnNw"
    "ZWN0aW9uIiwKICAgICAgICAgICAgKQogICAgICAgIHByb2Nlc3NfaWRzID0gc29ydGVkKGRl"
    "c2NlbmRhbnRzKHdvcmtlci5wcm9jZXNzLnBpZCwgcGFyZW50cykpCiAgICAgICAgZm9yIHBp"
    "ZCBpbiBwcm9jZXNzX2lkczoKICAgICAgICAgICAgZm9yIHBhdGggaW4gc29ydGVkKG5hdGl2"
    "ZV9wYXRoc19mb3JfcHJvY2VzcyhwaWQpKToKICAgICAgICAgICAgICAgIGNsYXNzaWZpY2F0"
    "aW9uID0gY2xhc3NpZnlfbmF0aXZlX29yaWdpbihwYXRoKQogICAgICAgICAgICAgICAgaXRl"
    "bSA9IHsKICAgICAgICAgICAgICAgICAgICAid29ya2VyX2lkIjogd29ya2VyLndvcmtlcl9p"
    "ZCwKICAgICAgICAgICAgICAgICAgICAid29ya2VyX2luc3RhbmNlX2lkIjogd29ya2VyLmlu"
    "c3RhbmNlX2lkLAogICAgICAgICAgICAgICAgICAgICJwaWQiOiBwaWQsCiAgICAgICAgICAg"
    "ICAgICAgICAgInBhdGgiOiBzYW5pdGl6ZWRfbmF0aXZlX3BhdGgocGF0aCksCiAgICAgICAg"
    "ICAgICAgICAgICAgImNsYXNzaWZpY2F0aW9uIjogY2xhc3NpZmljYXRpb24sCiAgICAgICAg"
    "ICAgICAgICB9CiAgICAgICAgICAgICAgICByZXBvcnRzLmFwcGVuZChpdGVtKQogICAgICAg"
    "ICAgICAgICAgaWYgY2xhc3NpZmljYXRpb24gPT0gIlJFSkVDVEVEX1NUVUIiOgogICAgICAg"
    "ICAgICAgICAgICAgIHJlamVjdGVkLmFwcGVuZChpdGVtKQogICAgICAgICAgICAgICAgbG93"
    "ZXJlZCA9IFBhdGgocGF0aCkubmFtZS5sb3dlcigpCiAgICAgICAgICAgICAgICBmb3IgdG9r"
    "ZW4gaW4gVEFSR0VUX1JFUVVJUkVEX05BVElWRV9UT0tFTlM6CiAgICAgICAgICAgICAgICAg"
    "ICAgaWYgdG9rZW4gaW4gbG93ZXJlZDoKICAgICAgICAgICAgICAgICAgICAgICAgcmVxdWly"
    "ZWRbdG9rZW5dLmFwcGVuZChpdGVtKQogICAgaWYgbm90IHJlcG9ydHM6CiAgICAgICAgcmFp"
    "c2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJQM19QNl9OQVRJVkVfT1JJR0lO"
    "X0NMT1NVUkVfRkFJTEVEIiwKICAgICAgICAgICAgIm5vIGNyaXRpY2FsIG5hdGl2ZSBvcmln"
    "aW5zIHdlcmUgb2JzZXJ2ZWQiLAogICAgICAgICkKICAgIGlmIHJlamVjdGVkOgogICAgICAg"
    "IHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUDNfUDZfTkFUSVZFX09S"
    "SUdJTl9DTE9TVVJFX0ZBSUxFRCIsCiAgICAgICAgICAgICJvbmUgb3IgbW9yZSBwcm9oaWJp"
    "dGVkIENVREEgc3R1YiBvcmlnaW5zIHdlcmUgb2JzZXJ2ZWQiLAogICAgICAgICkKICAgIHJl"
    "cXVpcmVkX3N0YXR1czogZGljdFtzdHIsIGRpY3Rbc3RyLCBvYmplY3RdXSA9IHt9CiAgICBm"
    "b3IgdG9rZW4sIG9ic2VydmF0aW9ucyBpbiByZXF1aXJlZC5pdGVtcygpOgogICAgICAgIGFs"
    "bF9mcm9tX3RhcmdldCA9IGJvb2wob2JzZXJ2YXRpb25zKSBhbmQgYWxsKAogICAgICAgICAg"
    "ICBpdGVtWyJjbGFzc2lmaWNhdGlvbiJdID09ICJUQVJHRVRfUlVOVElNRSIgZm9yIGl0ZW0g"
    "aW4gb2JzZXJ2YXRpb25zCiAgICAgICAgKQogICAgICAgIHJlcXVpcmVkX3N0YXR1c1t0b2tl"
    "bl0gPSB7CiAgICAgICAgICAgICJvYnNlcnZlZCI6IGJvb2wob2JzZXJ2YXRpb25zKSwKICAg"
    "ICAgICAgICAgImFsbF9mcm9tX3RhcmdldCI6IGFsbF9mcm9tX3RhcmdldCwKICAgICAgICAg"
    "ICAgIm9ic2VydmF0aW9uX2NvdW50IjogbGVuKG9ic2VydmF0aW9ucyksCiAgICAgICAgfQog"
    "ICAgICAgIGlmIG5vdCBhbGxfZnJvbV90YXJnZXQ6CiAgICAgICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlAzX1A2X05BVElWRV9PUklHSU5fQ0xP"
    "U1VSRV9GQUlMRUQiLAogICAgICAgICAgICAgICAgZiJyZXF1aXJlZCB0YXJnZXQgbmF0aXZl"
    "IG9yaWdpbiBmYWlsZWQ6IHt0b2tlbn0iLAogICAgICAgICAgICApCiAgICByZXR1cm4gewog"
    "ICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgInJlcG9ydF9pZCI6"
    "ICgiYXVyYWdhdGV3YXktcDUtcDYtZXhhY3QtcnVudGltZS1yZXF1YWxpZmljYXRpb24tcnVu"
    "dGltZS1uYXRpdmUtb3JpZ2luLXYxIiksCiAgICAgICAgInN0YXR1cyI6ICJQQVNTRUQiLAog"
    "ICAgICAgICJkZWNpc2lvbiI6ICJSVU5USU1FX05BVElWRV9PUklHSU5fQ0xPU1VSRV9QQVNT"
    "RUQiLAogICAgICAgICJvYnNlcnZhdGlvbnMiOiByZXBvcnRzLAogICAgICAgICJyZXF1aXJl"
    "ZF90YXJnZXRfb3JpZ2lucyI6IHJlcXVpcmVkX3N0YXR1cywKICAgICAgICAicmVqZWN0ZWRf"
    "b3JpZ2luX2NvdW50IjogMCwKICAgICAgICAiY3VkYV9zdHViX29yaWdpbl9vYnNlcnZlZCI6"
    "IEZhbHNlLAogICAgICAgICJhbWJpZW50X25vbl9zdHViX29yaWdpbnNfcGVybWl0dGVkIjog"
    "VHJ1ZSwKICAgIH0KCgpkZWYgbmV3X3A2X2NoZWNrcG9pbnQoY291bnRlcnM6IGRpY3Rbc3Ry"
    "LCBpbnRdKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHJldHVybiB7CiAgICAgICAgInNj"
    "aGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAicHJvYmVfaWQiOiAiUDYiLAogICAg"
    "ICAgICJzdGF0dXMiOiAiSU5fUFJPR1JFU1MiLAogICAgICAgICJjdXJyZW50X3N0YWdlIjog"
    "IlA2X1NUQVJURUQiLAogICAgICAgICJldmVudHMiOiBbXSwKICAgICAgICAid29ya2VyX3Jl"
    "cXVlc3RfY291bnRlcnMiOiB7CiAgICAgICAgICAgICJ3b3JrZXJfMSI6IHsiYXR0ZW1wdGVk"
    "IjogMCwgImNvbXBsZXRlZCI6IDB9LAogICAgICAgICAgICAid29ya2VyXzIiOiB7ImF0dGVt"
    "cHRlZCI6IDAsICJjb21wbGV0ZWQiOiAwfSwKICAgICAgICB9LAogICAgICAgICJnbG9iYWxf"
    "bW9kZWxfcmVxdWVzdHMiOiBjb3VudGVyc1sibW9kZWxfcmVxdWVzdHMiXSwKICAgICAgICAi"
    "c3RhcnRpbmdfbW9kZWxfcmVxdWVzdHMiOiBjb3VudGVyc1sibW9kZWxfcmVxdWVzdHMiXSwK"
    "ICAgICAgICAicmF3X3Byb21wdF9sb2dnZWQiOiBGYWxzZSwKICAgICAgICAicmF3X291dHB1"
    "dF9sb2dnZWQiOiBGYWxzZSwKICAgIH0KCgpkZWYgcGVyc2lzdF9wNl9jaGVja3BvaW50KAog"
    "ICAgY2hlY2twb2ludDogZGljdFtzdHIsIG9iamVjdF0sCiAgICBzdGFnZTogc3RyLAogICAg"
    "Y291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAogICAgZGV0YWlsczogZGljdFtzdHIsIG9iamVj"
    "dF0gfCBOb25lID0gTm9uZSwKKSAtPiBOb25lOgogICAgZXZlbnRzID0gY2hlY2twb2ludC5n"
    "ZXQoImV2ZW50cyIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShldmVudHMsIGxpc3QpOgogICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUDZfQ0hFQ0tQT0lO"
    "VF9TRVJJQUxJWkFUSU9OX0ZBSUxFRCIsCiAgICAgICAgICAgICJQNiBjaGVja3BvaW50IGV2"
    "ZW50IGxlZGdlciBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICBldmVudCA9IHsKICAgICAg"
    "ICAic3RhZ2UiOiBzdGFnZSwKICAgICAgICAib2JzZXJ2ZWRfYXQiOiBkYXRldGltZS5ub3co"
    "VVRDKS5pc29mb3JtYXQodGltZXNwZWM9Im1pbGxpc2Vjb25kcyIpLAogICAgICAgICJnbG9i"
    "YWxfbW9kZWxfcmVxdWVzdHMiOiBjb3VudGVyc1sibW9kZWxfcmVxdWVzdHMiXSwKICAgICAg"
    "ICAiZGV0YWlscyI6IGRldGFpbHMgb3Ige30sCiAgICB9CiAgICBldmVudHMuYXBwZW5kKGV2"
    "ZW50KQogICAgY2hlY2twb2ludFsiY3VycmVudF9zdGFnZSJdID0gc3RhZ2UKICAgIGNoZWNr"
    "cG9pbnRbImdsb2JhbF9tb2RlbF9yZXF1ZXN0cyJdID0gY291bnRlcnNbIm1vZGVsX3JlcXVl"
    "c3RzIl0KICAgIHRyeToKICAgICAgICB3cml0ZV9qc29uKAogICAgICAgICAgICBPVVRQVVRf"
    "Uk9PVCAvICJwNl9zdGFnZV9jaGVja3BvaW50X3JlcG9ydF92MS5qc29uIiwKICAgICAgICAg"
    "ICAgY2hlY2twb2ludCwKICAgICAgICApCiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGVycm9y"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUDZfQ0hF"
    "Q0tQT0lOVF9TRVJJQUxJWkFUSU9OX0ZBSUxFRCIsCiAgICAgICAgICAgICJQNiBjaGVja3Bv"
    "aW50IHNlcmlhbGl6YXRpb24gZmFpbGVkIiwKICAgICAgICApIGZyb20gZXJyb3IKCgpkZWYg"
    "cmVxdWVzdF9jb3VudGVyKAogICAgY2hlY2twb2ludDogZGljdFtzdHIsIG9iamVjdF0sCiAg"
    "ICB3b3JrZXJfaWQ6IHN0ciwKKSAtPiBkaWN0W3N0ciwgaW50XToKICAgIGNvdW50ZXJzID0g"
    "Y2hlY2twb2ludC5nZXQoIndvcmtlcl9yZXF1ZXN0X2NvdW50ZXJzIikKICAgIGlmIG5vdCBp"
    "c2luc3RhbmNlKGNvdW50ZXJzLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIlA2X0NIRUNLUE9JTlRfU0VSSUFMSVpBVElPTl9GQUlMRUQi"
    "LAogICAgICAgICAgICAiUDYgd29ya2VyIHJlcXVlc3QgY291bnRlciBtYXAgaXMgaW52YWxp"
    "ZCIsCiAgICAgICAgKQogICAgaXRlbSA9IGNvdW50ZXJzLmdldCh3b3JrZXJfaWQpCiAgICBp"
    "ZiBub3QgaXNpbnN0YW5jZShpdGVtLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGlj"
    "RmFpbHVyZSgKICAgICAgICAgICAgIlA2X0NIRUNLUE9JTlRfU0VSSUFMSVpBVElPTl9GQUlM"
    "RUQiLAogICAgICAgICAgICAiUDYgd29ya2VyIHJlcXVlc3QgY291bnRlciBpcyBpbnZhbGlk"
    "IiwKICAgICAgICApCiAgICBhdHRlbXB0ZWQgPSBpdGVtLmdldCgiYXR0ZW1wdGVkIikKICAg"
    "IGNvbXBsZXRlZCA9IGl0ZW0uZ2V0KCJjb21wbGV0ZWQiKQogICAgaWYgbm90IGlzaW5zdGFu"
    "Y2UoYXR0ZW1wdGVkLCBpbnQpIG9yIG5vdCBpc2luc3RhbmNlKGNvbXBsZXRlZCwgaW50KToK"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlA2X0NIRUNL"
    "UE9JTlRfU0VSSUFMSVpBVElPTl9GQUlMRUQiLAogICAgICAgICAgICAiUDYgd29ya2VyIHJl"
    "cXVlc3QgY291bnRlciB2YWx1ZXMgYXJlIGludmFsaWQiLAogICAgICAgICkKICAgIHJldHVy"
    "biBpdGVtCgoKZGVmIHZhbGlkYXRlX3Jlc3BvbnNlX2VudmVsb3BlKHJlc3BvbnNlOiBkaWN0"
    "W3N0ciwgb2JqZWN0XSkgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICB1c2FnZSA9IHJlc3Bv"
    "bnNlLmdldCgidXNhZ2UiKQogICAgY2hvaWNlcyA9IHJlc3BvbnNlLmdldCgiY2hvaWNlcyIp"
    "CiAgICBtb2RlbCA9IHJlc3BvbnNlLmdldCgibW9kZWwiKQogICAgaWYgbW9kZWwgIT0gU0VS"
    "VkVEX01PREVMX05BTUU6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJyZXNwb25zZSBt"
    "b2RlbCBpZGVudGl0eSBkcmlmdGVkIikKICAgIGlmIG5vdCBpc2luc3RhbmNlKHVzYWdlLCBk"
    "aWN0KToKICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoInJlc3BvbnNlIHVzYWdlIGlzIG1p"
    "c3NpbmciKQogICAgaWYgbm90IGlzaW5zdGFuY2UoY2hvaWNlcywgbGlzdCkgb3IgbGVuKGNo"
    "b2ljZXMpICE9IDE6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJyZXNwb25zZSBjaG9p"
    "Y2VzIGFyZSBpbnZhbGlkIikKICAgIGNob2ljZSA9IGNob2ljZXNbMF0KICAgIGlmIG5vdCBp"
    "c2luc3RhbmNlKGNob2ljZSwgZGljdCk6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJy"
    "ZXNwb25zZSBjaG9pY2UgaXMgaW52YWxpZCIpCiAgICBtZXNzYWdlID0gY2hvaWNlLmdldCgi"
    "bWVzc2FnZSIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShtZXNzYWdlLCBkaWN0KToKICAgICAg"
    "ICByYWlzZSBSdW50aW1lRXJyb3IoInJlc3BvbnNlIG1lc3NhZ2UgaXMgaW52YWxpZCIpCiAg"
    "ICBjb250ZW50ID0gbWVzc2FnZS5nZXQoImNvbnRlbnQiKQogICAgaWYgbm90IGlzaW5zdGFu"
    "Y2UoY29udGVudCwgc3RyKSBvciBub3QgY29udGVudDoKICAgICAgICByYWlzZSBSdW50aW1l"
    "RXJyb3IoInJlc3BvbnNlIGNvbnRlbnQgaXMgZW1wdHkiKQogICAgb3V0cHV0X3Rva2VucyA9"
    "IHVzYWdlLmdldCgiY29tcGxldGlvbl90b2tlbnMiKQogICAgcHJvbXB0X3Rva2VucyA9IHVz"
    "YWdlLmdldCgicHJvbXB0X3Rva2VucyIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShvdXRwdXRf"
    "dG9rZW5zLCBpbnQpIG9yIG5vdCAxIDw9IG91dHB1dF90b2tlbnMgPD0gMzI6CiAgICAgICAg"
    "cmFpc2UgUnVudGltZUVycm9yKCJjb21wbGV0aW9uIHRva2VuIGJ1ZGdldCBkcmlmdGVkIikK"
    "ICAgIGlmIG5vdCBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGludCkgb3IgcHJvbXB0X3Rv"
    "a2VucyA8PSAwOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicHJvbXB0IHRva2VuIGNv"
    "dW50IGlzIGludmFsaWQiKQogICAgcmV0dXJuIHsKICAgICAgICAicmVzcG9uc2VfY29udGVu"
    "dF9zaGEyNTYiOiBzaGEyNTZfdGV4dChjb250ZW50KSwKICAgICAgICAicmVzcG9uc2VfY29u"
    "dGVudF9sZW5ndGgiOiBsZW4oY29udGVudCksCiAgICAgICAgImZpbmlzaF9yZWFzb24iOiBj"
    "aG9pY2UuZ2V0KCJmaW5pc2hfcmVhc29uIiksCiAgICAgICAgInByb21wdF90b2tlbnMiOiBw"
    "cm9tcHRfdG9rZW5zLAogICAgICAgICJjb21wbGV0aW9uX3Rva2VucyI6IG91dHB1dF90b2tl"
    "bnMsCiAgICAgICAgIm1vZGVsX2lkZW50aXR5X3ZhbGlkIjogVHJ1ZSwKICAgICAgICAicmVz"
    "cG9uc2VfZW52ZWxvcGVfdmFsaWQiOiBUcnVlLAogICAgfQoKCmRlZiBfYWxsX21ldHJpY192"
    "YWx1ZXNfemVybyhkZWx0YTogTWV0cmljRGVsdGFPYnNlcnZhdGlvbikgLT4gYm9vbDoKICAg"
    "IHJldHVybiBhbGwoCiAgICAgICAgYWJzKGZsb2F0KGdldGF0dHIoZGVsdGEsIG5hbWUpKSkg"
    "PD0gMWUtOQogICAgICAgIGZvciBuYW1lIGluICgKICAgICAgICAgICAgInByZWZpeF9jYWNo"
    "ZV9xdWVyaWVzIiwKICAgICAgICAgICAgInByZWZpeF9jYWNoZV9oaXRzIiwKICAgICAgICAg"
    "ICAgImxvY2FsX2NvbXB1dGUiLAogICAgICAgICAgICAibG9jYWxfY2FjaGVfaGl0IiwKICAg"
    "ICAgICAgICAgImV4dGVybmFsX2t2X3RyYW5zZmVyIiwKICAgICAgICAgICAgImNhY2hlZF9w"
    "cm9tcHRfdG9rZW5zIiwKICAgICAgICAgICAgIm5ld2x5X2NvbXB1dGVkX3ByZWZpbGxfdG9r"
    "ZW5zIiwKICAgICAgICApCiAgICApCgoKZGVmIHJ1bl9hdHRyaWJ1dGVkX3JlcXVlc3QoCiAg"
    "ICB0YXJnZXQ6IFdvcmtlciwKICAgIG5vbl90YXJnZXQ6IFdvcmtlciwKICAgIHJlcXVlc3Rf"
    "cm9sZTogc3RyLAogICAgcHJlZml4X3ZhcmlhbnQ6IHN0ciwKICAgIGNvdW50ZXJzOiBkaWN0"
    "W3N0ciwgaW50XSwKICAgIGNoZWNrcG9pbnQ6IGRpY3Rbc3RyLCBvYmplY3RdLAopIC0+IGRp"
    "Y3Rbc3RyLCBvYmplY3RdOgogICAgaWYgdGFyZ2V0Lndvcmtlcl9pZCA9PSBub25fdGFyZ2V0"
    "Lndvcmtlcl9pZDoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAg"
    "ICAgIlA2X1JPVVRFX1JFQUxJWkFUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICAidGFyZ2V0"
    "IGFuZCBub24tdGFyZ2V0IHdvcmtlcnMgbXVzdCBiZSBkaXN0aW5jdCIsCiAgICAgICAgKQog"
    "ICAgdGFyZ2V0X2NvdW50ZXIgPSByZXF1ZXN0X2NvdW50ZXIoY2hlY2twb2ludCwgdGFyZ2V0"
    "Lndvcmtlcl9pZCkKICAgIHRhcmdldF9jb3VudGVyWyJhdHRlbXB0ZWQiXSArPSAxCiAgICBw"
    "ZXJzaXN0X3A2X2NoZWNrcG9pbnQoCiAgICAgICAgY2hlY2twb2ludCwKICAgICAgICBmInty"
    "ZXF1ZXN0X3JvbGV9X0FUVEVNUFRFRCIsCiAgICAgICAgY291bnRlcnMsCiAgICAgICAgewog"
    "ICAgICAgICAgICAicmVxdWVzdF9yb2xlIjogcmVxdWVzdF9yb2xlLAogICAgICAgICAgICAi"
    "aW50ZW5kZWRfd29ya2VyIjogdGFyZ2V0Lndvcmtlcl9pZCwKICAgICAgICAgICAgIndvcmtl"
    "cl9pbnN0YW5jZV9pZCI6IHRhcmdldC5pbnN0YW5jZV9pZCwKICAgICAgICAgICAgInBvcnQi"
    "OiB0YXJnZXQucG9ydCwKICAgICAgICAgICAgInByZWZpeF92YXJpYW50IjogcHJlZml4X3Zh"
    "cmlhbnQsCiAgICAgICAgfSwKICAgICkKICAgIGJlZm9yZV90YXJnZXQgPSB0YXJnZXQubWV0"
    "cmljX3NuYXBzaG90KCkKICAgIGJlZm9yZV9ub25fdGFyZ2V0ID0gbm9uX3RhcmdldC5tZXRy"
    "aWNfc25hcHNob3QoKQogICAgdHJ5OgogICAgICAgIHJlcXVlc3QgPSBydW5fc3RydWN0dXJl"
    "ZF9yZXF1ZXN0KAogICAgICAgICAgICB0YXJnZXQsCiAgICAgICAgICAgIHJlcXVlc3Rfcm9s"
    "ZSwKICAgICAgICAgICAgcHJlZml4X3ZhcmlhbnQsCiAgICAgICAgICAgIGNvdW50ZXJzLAog"
    "ICAgICAgICkKICAgIGV4Y2VwdCBEaWFnbm9zdGljQW1iaWd1aXR5OgogICAgICAgIHJhaXNl"
    "CiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGVycm9yOgogICAgICAgIHJhaXNlIERpYWdub3N0"
    "aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVRVUVTVF9FWEVDVVRJT05fRkFJTFVSRSIsCiAg"
    "ICAgICAgICAgIHNhbml0aXplX2V4Y2VycHQoc3RyKGVycm9yKSlbOjUxMl0gb3IgInJvdXRl"
    "IHJlcXVlc3QgZmFpbGVkIiwKICAgICAgICApIGZyb20gZXJyb3IKICAgIGFmdGVyX3Rhcmdl"
    "dCA9IHRhcmdldC5tZXRyaWNfc25hcHNob3QoKQogICAgYWZ0ZXJfbm9uX3RhcmdldCA9IG5v"
    "bl90YXJnZXQubWV0cmljX3NuYXBzaG90KCkKICAgIHRhcmdldF9kZWx0YSA9IG1ldHJpY19k"
    "ZWx0YShiZWZvcmVfdGFyZ2V0LCBhZnRlcl90YXJnZXQpCiAgICBub25fdGFyZ2V0X2RlbHRh"
    "ID0gbWV0cmljX2RlbHRhKGJlZm9yZV9ub25fdGFyZ2V0LCBhZnRlcl9ub25fdGFyZ2V0KQoK"
    "ICAgIGlmIHRhcmdldF9kZWx0YS50b3RhbF9wcm9tcHRfdG9rZW5zIDw9IDA6CiAgICAgICAg"
    "cmFpc2UgRGlhZ25vc3RpY0FtYmlndWl0eSgKICAgICAgICAgICAgIk1FVFJJQ19BVFRSSUJV"
    "VElPTl9BTUJJR1VPVVMiLAogICAgICAgICAgICBmIntyZXF1ZXN0X3JvbGV9IGRpZCBub3Qg"
    "Y2hhbmdlIHRhcmdldC13b3JrZXIgcHJvbXB0IG1ldHJpY3MiLAogICAgICAgICkKICAgIGlm"
    "IG5vdCBfYWxsX21ldHJpY192YWx1ZXNfemVybyhub25fdGFyZ2V0X2RlbHRhKToKICAgICAg"
    "ICByYWlzZSBEaWFnbm9zdGljQW1iaWd1aXR5KAogICAgICAgICAgICAiTUVUUklDX0FUVFJJ"
    "QlVUSU9OX0FNQklHVU9VUyIsCiAgICAgICAgICAgIGYie3JlcXVlc3Rfcm9sZX0gY2hhbmdl"
    "ZCBub24tdGFyZ2V0LXdvcmtlciByZXF1ZXN0IG1ldHJpY3MiLAogICAgICAgICkKICAgIHRh"
    "cmdldF9jb3VudGVyWyJjb21wbGV0ZWQiXSArPSAxCiAgICByZXF1ZXN0WyJ0YXJnZXRfbWV0"
    "cmljX2RlbHRhIl0gPSBhc2RpY3QodGFyZ2V0X2RlbHRhKQogICAgcmVxdWVzdFsibm9uX3Rh"
    "cmdldF9tZXRyaWNfZGVsdGEiXSA9IGFzZGljdChub25fdGFyZ2V0X2RlbHRhKQogICAgcmVx"
    "dWVzdFsicm91dGVfYWNrbm93bGVkZ2VkIl0gPSBUcnVlCiAgICBwZXJzaXN0X3A2X2NoZWNr"
    "cG9pbnQoCiAgICAgICAgY2hlY2twb2ludCwKICAgICAgICBmIntyZXF1ZXN0X3JvbGV9X0FU"
    "VFJJQlVURUQiLAogICAgICAgIGNvdW50ZXJzLAogICAgICAgIHsKICAgICAgICAgICAgInJl"
    "cXVlc3Rfcm9sZSI6IHJlcXVlc3Rfcm9sZSwKICAgICAgICAgICAgInJlYWxpemVkX3dvcmtl"
    "ciI6IHRhcmdldC53b3JrZXJfaWQsCiAgICAgICAgICAgICJyb3V0ZV9hY2tub3dsZWRnZWQi"
    "OiBUcnVlLAogICAgICAgIH0sCiAgICApCiAgICByZXR1cm4gcmVxdWVzdAoKCmRlZiByb3V0"
    "ZV9pc29sYXRpb24oCiAgICB3b3JrZXJfMTogV29ya2VyLAogICAgd29ya2VyXzI6IFdvcmtl"
    "ciwKICAgIGNvdW50ZXJzOiBkaWN0W3N0ciwgaW50XSwKICAgIGNoZWNrcG9pbnQ6IGRpY3Rb"
    "c3RyLCBvYmplY3RdLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgY3Jvc3Nfd29ya2Vy"
    "ID0gcnVuX2F0dHJpYnV0ZWRfcmVxdWVzdCgKICAgICAgICB3b3JrZXJfMiwKICAgICAgICB3"
    "b3JrZXJfMSwKICAgICAgICAiQ1JPU1NfV09SS0VSX0NPTEQiLAogICAgICAgICJBIiwKICAg"
    "ICAgICBjb3VudGVycywKICAgICAgICBjaGVja3BvaW50LAogICAgKQogICAgcmV0ZW50aW9u"
    "ID0gcnVuX2F0dHJpYnV0ZWRfcmVxdWVzdCgKICAgICAgICB3b3JrZXJfMSwKICAgICAgICB3"
    "b3JrZXJfMiwKICAgICAgICAiV09SS0VSMV9SRVRFTlRJT04iLAogICAgICAgICJBIiwKICAg"
    "ICAgICBjb3VudGVycywKICAgICAgICBjaGVja3BvaW50LAogICAgKQoKICAgIGV4cGVjdGVk"
    "X3RvdGFsID0gc3VtKAogICAgICAgIHJlcXVlc3RfY291bnRlcihjaGVja3BvaW50LCB3b3Jr"
    "ZXJfaWQpWyJhdHRlbXB0ZWQiXQogICAgICAgIGZvciB3b3JrZXJfaWQgaW4gKCJ3b3JrZXJf"
    "MSIsICJ3b3JrZXJfMiIpCiAgICApCiAgICBzdGFydGluZ19tb2RlbF9yZXF1ZXN0cyA9IGNo"
    "ZWNrcG9pbnQuZ2V0KCJzdGFydGluZ19tb2RlbF9yZXF1ZXN0cyIpCiAgICBpZiBub3QgaXNp"
    "bnN0YW5jZShzdGFydGluZ19tb2RlbF9yZXF1ZXN0cywgaW50KToKICAgICAgICByYWlzZSBE"
    "aWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05f"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICJQNiBzdGFydGluZyByZXF1ZXN0IGNvdW50IGlzIGlu"
    "dmFsaWQiLAogICAgICAgICkKICAgIHA2X3JlcXVlc3RfdG90YWwgPSBjb3VudGVyc1sibW9k"
    "ZWxfcmVxdWVzdHMiXSAtIHN0YXJ0aW5nX21vZGVsX3JlcXVlc3RzCiAgICBpZiBleHBlY3Rl"
    "ZF90b3RhbCAhPSAyIG9yIHA2X3JlcXVlc3RfdG90YWwgIT0gZXhwZWN0ZWRfdG90YWw6CiAg"
    "ICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX1JF"
    "Q09OQ0lMSUFUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICAiUDYgcmVxdWVzdCBjb3VudGVy"
    "cyBkbyBub3QgcmVjb25jaWxlIiwKICAgICAgICApCiAgICBmb3Igd29ya2VyX2lkIGluICgi"
    "d29ya2VyXzEiLCAid29ya2VyXzIiKToKICAgICAgICBpdGVtID0gcmVxdWVzdF9jb3VudGVy"
    "KGNoZWNrcG9pbnQsIHdvcmtlcl9pZCkKICAgICAgICBpZiBpdGVtWyJhdHRlbXB0ZWQiXSAh"
    "PSAxIG9yIGl0ZW1bImNvbXBsZXRlZCJdICE9IDE6CiAgICAgICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05f"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICAgICAicGVyLXdvcmtlciBQNiByZXF1ZXN0IGNvdW50"
    "ZXJzIGRvIG5vdCByZWNvbmNpbGUiLAogICAgICAgICAgICApCiAgICBwZXJzaXN0X3A2X2No"
    "ZWNrcG9pbnQoCiAgICAgICAgY2hlY2twb2ludCwKICAgICAgICAiUkVRVUVTVF9DT1VOVEVS"
    "U19SRUNPTkNJTEVEIiwKICAgICAgICBjb3VudGVycywKICAgICAgICB7InA2X3JlcXVlc3Rf"
    "dG90YWwiOiBwNl9yZXF1ZXN0X3RvdGFsfSwKICAgICkKICAgIHJldHVybiB7CiAgICAgICAg"
    "ImNyb3NzX3dvcmtlcl9yZXF1ZXN0IjogY3Jvc3Nfd29ya2VyLAogICAgICAgICJ3b3JrZXJf"
    "MV9yZXRlbnRpb25fcmVxdWVzdCI6IHJldGVudGlvbiwKICAgICAgICAid29ya2VyXzFfcm91"
    "dGVfaXNvbGF0ZWQiOiBUcnVlLAogICAgICAgICJ3b3JrZXJfMl9yb3V0ZV9pc29sYXRlZCI6"
    "IFRydWUsCiAgICAgICAgInJlcXVlc3RfY291bnRlcnNfcmVjb25jaWxlZCI6IFRydWUsCiAg"
    "ICB9CgoKZGVmIF9yZXF1ZXN0X21ldHJpY19kZWx0YShwYXlsb2FkOiBkaWN0W3N0ciwgb2Jq"
    "ZWN0XSkgLT4gTWV0cmljRGVsdGFPYnNlcnZhdGlvbjoKICAgIG9ic2VydmF0aW9uID0gcGF5"
    "bG9hZC5nZXQoIm1ldHJpY19kZWx0YSIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShvYnNlcnZh"
    "dGlvbiwgTWV0cmljRGVsdGFPYnNlcnZhdGlvbik6CiAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAg"
    "ICAgICAgICAicmVxdWVzdCBtZXRyaWMgZGVsdGEgaXMgbm90IGEgdHlwZWQgb2JzZXJ2YXRp"
    "b24iLAogICAgICAgICkKICAgIHJldHVybiBvYnNlcnZhdGlvbgoKCmRlZiBfcmVxdWVzdF90"
    "b2tlbl9pZGVudGl0eShwYXlsb2FkOiBkaWN0W3N0ciwgb2JqZWN0XSkgLT4gVG9rZW5JZGVu"
    "dGl0eU9ic2VydmF0aW9uOgogICAgb2JzZXJ2YXRpb24gPSBwYXlsb2FkLmdldCgidG9rZW5f"
    "aWRlbnRpdHkiKQogICAgaWYgbm90IGlzaW5zdGFuY2Uob2JzZXJ2YXRpb24sIFRva2VuSWRl"
    "bnRpdHlPYnNlcnZhdGlvbik6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAg"
    "ICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAicmVx"
    "dWVzdCB0b2tlbiBpZGVudGl0eSBpcyBub3QgYSB0eXBlZCBvYnNlcnZhdGlvbiIsCiAgICAg"
    "ICAgKQogICAgcmV0dXJuIG9ic2VydmF0aW9uCgoKZGVmIF9yZXF1ZXN0X3JvdXRlKHBheWxv"
    "YWQ6IGRpY3Rbc3RyLCBvYmplY3RdKSAtPiBSb3V0ZU9ic2VydmF0aW9uOgogICAgb2JzZXJ2"
    "YXRpb24gPSBwYXlsb2FkLmdldCgicm91dGVfb2JzZXJ2YXRpb24iKQogICAgaWYgbm90IGlz"
    "aW5zdGFuY2Uob2JzZXJ2YXRpb24sIFJvdXRlT2JzZXJ2YXRpb24pOgogICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlM"
    "VVJFIiwKICAgICAgICAgICAgInJlcXVlc3Qgcm91dGUgaXMgbm90IGEgdHlwZWQgb2JzZXJ2"
    "YXRpb24iLAogICAgICAgICkKICAgIHJldHVybiBvYnNlcnZhdGlvbgoKCmRlZiBfc2FtZV90"
    "b2tlbl9pZGVudGl0eSgKICAgIGxlZnQ6IFRva2VuSWRlbnRpdHlPYnNlcnZhdGlvbiwKICAg"
    "IHJpZ2h0OiBUb2tlbklkZW50aXR5T2JzZXJ2YXRpb24sCikgLT4gYm9vbDoKICAgIHJldHVy"
    "biAoCiAgICAgICAgbGVmdC50b2tlbl9jb3VudCA9PSByaWdodC50b2tlbl9jb3VudAogICAg"
    "ICAgIGFuZCBsZWZ0LnRva2VuX3NoYTI1NiA9PSByaWdodC50b2tlbl9zaGEyNTYKICAgICAg"
    "ICBhbmQgbGVmdC50b2tlbl9pZHMgPT0gcmlnaHQudG9rZW5faWRzCiAgICApCgoKZGVmIGRl"
    "Y2lkZV9wNSgKICAgIGNvbGQ6IGRpY3Rbc3RyLCBvYmplY3RdLAogICAgd2FybTogZGljdFtz"
    "dHIsIG9iamVjdF0sCiAgICBuZWdhdGl2ZV9wcmVmaXg6IGRpY3Rbc3RyLCBvYmplY3RdLAog"
    "ICAgcG9zdF9yZXNldF9jb2xkOiBkaWN0W3N0ciwgb2JqZWN0XSwKICAgIGNyb3NzX3dvcmtl"
    "cl9jb2xkOiBkaWN0W3N0ciwgb2JqZWN0XSwKKSAtPiBCZWhhdmlvckRlY2lzaW9uOgogICAg"
    "Y29sZF9tZXRyaWMgPSBfcmVxdWVzdF9tZXRyaWNfZGVsdGEoY29sZCkKICAgIHdhcm1fbWV0"
    "cmljID0gX3JlcXVlc3RfbWV0cmljX2RlbHRhKHdhcm0pCiAgICBuZWdhdGl2ZV9tZXRyaWMg"
    "PSBfcmVxdWVzdF9tZXRyaWNfZGVsdGEobmVnYXRpdmVfcHJlZml4KQogICAgcmVzZXRfbWV0"
    "cmljID0gX3JlcXVlc3RfbWV0cmljX2RlbHRhKHBvc3RfcmVzZXRfY29sZCkKICAgIGNyb3Nz"
    "X21ldHJpYyA9IF9yZXF1ZXN0X21ldHJpY19kZWx0YShjcm9zc193b3JrZXJfY29sZCkKCiAg"
    "ICBjb2xkX3Rva2VucyA9IF9yZXF1ZXN0X3Rva2VuX2lkZW50aXR5KGNvbGQpCiAgICB3YXJt"
    "X3Rva2VucyA9IF9yZXF1ZXN0X3Rva2VuX2lkZW50aXR5KHdhcm0pCiAgICBuZWdhdGl2ZV90"
    "b2tlbnMgPSBfcmVxdWVzdF90b2tlbl9pZGVudGl0eShuZWdhdGl2ZV9wcmVmaXgpCiAgICBy"
    "ZXNldF90b2tlbnMgPSBfcmVxdWVzdF90b2tlbl9pZGVudGl0eShwb3N0X3Jlc2V0X2NvbGQp"
    "CiAgICBjcm9zc190b2tlbnMgPSBfcmVxdWVzdF90b2tlbl9pZGVudGl0eShjcm9zc193b3Jr"
    "ZXJfY29sZCkKCiAgICByZWFzb25zOiBsaXN0W3N0cl0gPSBbXQogICAgZmFpbGVkID0gRmFs"
    "c2UKCiAgICBmb3IgbGFiZWwsIG1ldHJpYyBpbiAoCiAgICAgICAgKCJCQVNFX0NPTEQiLCBj"
    "b2xkX21ldHJpYyksCiAgICAgICAgKCJCQVNFX1dBUk0iLCB3YXJtX21ldHJpYyksCiAgICAg"
    "ICAgKCJORUdBVElWRV9QUkVGSVgiLCBuZWdhdGl2ZV9tZXRyaWMpLAogICAgICAgICgiUE9T"
    "VF9SRVNFVF9DT0xEIiwgcmVzZXRfbWV0cmljKSwKICAgICAgICAoIkNST1NTX1dPUktFUl9D"
    "T0xEIiwgY3Jvc3NfbWV0cmljKSwKICAgICk6CiAgICAgICAgaWYgbWV0cmljLmV4dGVybmFs"
    "X2t2X3RyYW5zZmVyICE9IDA6CiAgICAgICAgICAgIGZhaWxlZCA9IFRydWUKICAgICAgICAg"
    "ICAgcmVhc29ucy5hcHBlbmQoZiJ7bGFiZWx9IG9ic2VydmVkIGV4dGVybmFsIEtWIHRyYW5z"
    "ZmVyIikKCiAgICBpZiBub3QgKAogICAgICAgIF9zYW1lX3Rva2VuX2lkZW50aXR5KGNvbGRf"
    "dG9rZW5zLCB3YXJtX3Rva2VucykKICAgICAgICBhbmQgX3NhbWVfdG9rZW5faWRlbnRpdHko"
    "Y29sZF90b2tlbnMsIHJlc2V0X3Rva2VucykKICAgICAgICBhbmQgX3NhbWVfdG9rZW5faWRl"
    "bnRpdHkoY29sZF90b2tlbnMsIGNyb3NzX3Rva2VucykKICAgICk6CiAgICAgICAgZmFpbGVk"
    "ID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5kKCJwcmVmaXggQSB0b2tlbiBpZGVudGl0"
    "eSBkcmlmdGVkIGFjcm9zcyBQNSBjb250cm9scyIpCgogICAgaWYgX3NhbWVfdG9rZW5faWRl"
    "bnRpdHkoY29sZF90b2tlbnMsIG5lZ2F0aXZlX3Rva2Vucyk6CiAgICAgICAgZmFpbGVkID0g"
    "VHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5kKCJORUdBVElWRV9QUkVGSVggZGlkIG5vdCBk"
    "aXZlcmdlIGF0IHRva2VuIGxldmVsIikKCiAgICBuZWdhdGl2ZV9ib3VuZCA9IGNhY2hlYWJs"
    "ZV9jb21tb25fcHJlZml4X2JvdW5kKAogICAgICAgIGNvbGRfdG9rZW5zLAogICAgICAgIG5l"
    "Z2F0aXZlX3Rva2VucywKICAgICkKCiAgICBpZiBjb2xkX21ldHJpYy5sb2NhbF9jYWNoZV9o"
    "aXQgIT0gMDoKICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAgcmVhc29ucy5hcHBlbmQo"
    "IkJBU0VfQ09MRCB1bmV4cGVjdGVkbHkgb2JzZXJ2ZWQgbG9jYWwgY2FjaGUtaGl0IHRva2Vu"
    "cyIpCiAgICBpZiBjb2xkX21ldHJpYy5sb2NhbF9jb21wdXRlIDw9IDA6CiAgICAgICAgZmFp"
    "bGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5kKCJCQVNFX0NPTEQgZGlkIG5vdCBv"
    "YnNlcnZlIGxvY2FsIGNvbXB1dGUiKQoKICAgIGlmIHdhcm1fbWV0cmljLmxvY2FsX2NhY2hl"
    "X2hpdCA8PSAwOgogICAgICAgIGZhaWxlZCA9IFRydWUKICAgICAgICByZWFzb25zLmFwcGVu"
    "ZCgiQkFTRV9XQVJNIGRpZCBub3Qgb2JzZXJ2ZSBsb2NhbCBjYWNoZS1oaXQgdG9rZW5zIikK"
    "ICAgIGlmIHdhcm1fbWV0cmljLnByZWZpeF9jYWNoZV9oaXRzIDw9IDA6CiAgICAgICAgZmFp"
    "bGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5kKCJCQVNFX1dBUk0gZGlkIG5vdCBv"
    "YnNlcnZlIHByZWZpeC1jYWNoZSBoaXRzIikKICAgIGlmIHdhcm1fbWV0cmljLm5ld2x5X2Nv"
    "bXB1dGVkX3ByZWZpbGxfdG9rZW5zID49IChjb2xkX21ldHJpYy5uZXdseV9jb21wdXRlZF9w"
    "cmVmaWxsX3Rva2Vucyk6CiAgICAgICAgZmFpbGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMu"
    "YXBwZW5kKCJCQVNFX1dBUk0gZGlkIG5vdCByZWR1Y2UgY29tcHV0ZWQgcHJlZmlsbCB0b2tl"
    "bnMiKQoKICAgIGlmIG5lZ2F0aXZlX21ldHJpYy5sb2NhbF9jYWNoZV9oaXQgPiBmbG9hdChu"
    "ZWdhdGl2ZV9ib3VuZCk6CiAgICAgICAgZmFpbGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMu"
    "YXBwZW5kKCJORUdBVElWRV9QUkVGSVggcmV1c2UgZXhjZWVkZWQgaXRzIGNhY2hlYWJsZSBj"
    "b21tb24tcHJlZml4IGJvdW5kIikKICAgIGlmIG5lZ2F0aXZlX21ldHJpYy5sb2NhbF9jYWNo"
    "ZV9oaXQgPj0gd2FybV9tZXRyaWMubG9jYWxfY2FjaGVfaGl0OgogICAgICAgIGZhaWxlZCA9"
    "IFRydWUKICAgICAgICByZWFzb25zLmFwcGVuZCgiTkVHQVRJVkVfUFJFRklYIHJldXNlIHdh"
    "cyBub3QgbG93ZXIgdGhhbiBCQVNFX1dBUk0gcmV1c2UiKQoKICAgIGlmIHJlc2V0X21ldHJp"
    "Yy5sb2NhbF9jYWNoZV9oaXQgIT0gMDoKICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAg"
    "cmVhc29ucy5hcHBlbmQoIlBPU1RfUkVTRVRfQ09MRCBpbmhlcml0ZWQgcHJvaGliaXRlZCBs"
    "b2NhbCBjYWNoZSBzdGF0ZSIpCiAgICBpZiByZXNldF9tZXRyaWMubG9jYWxfY29tcHV0ZSA8"
    "PSAwOgogICAgICAgIGZhaWxlZCA9IFRydWUKICAgICAgICByZWFzb25zLmFwcGVuZCgiUE9T"
    "VF9SRVNFVF9DT0xEIGRpZCBub3QgcmVjb21wdXRlIHByb21wdCB0b2tlbnMiKQoKICAgIGlm"
    "IGNyb3NzX21ldHJpYy5sb2NhbF9jYWNoZV9oaXQgIT0gMDoKICAgICAgICBmYWlsZWQgPSBU"
    "cnVlCiAgICAgICAgcmVhc29ucy5hcHBlbmQoIkNST1NTX1dPUktFUl9DT0xEIGluaGVyaXRl"
    "ZCBwcm9oaWJpdGVkIGxvY2FsIGNhY2hlIHN0YXRlIikKICAgIGlmIGNyb3NzX21ldHJpYy5s"
    "b2NhbF9jb21wdXRlIDw9IDA6CiAgICAgICAgZmFpbGVkID0gVHJ1ZQogICAgICAgIHJlYXNv"
    "bnMuYXBwZW5kKCJDUk9TU19XT1JLRVJfQ09MRCBkaWQgbm90IGNvbXB1dGUgcHJvbXB0IHRv"
    "a2VucyIpCgogICAgaWYgZmFpbGVkOgogICAgICAgIHJldHVybiBCZWhhdmlvckRlY2lzaW9u"
    "KAogICAgICAgICAgICBjYXBhYmlsaXR5PSJQNV9QUkVGSVhfQ0FDSEVfQkVIQVZJT1IiLAog"
    "ICAgICAgICAgICBzdGF0ZT1CZWhhdmlvclN0YXRlLkZBSUwsCiAgICAgICAgICAgIGZhaWx1"
    "cmVfY2xhc3M9IlA1X0JFSEFWSU9SX0ZBSUxVUkUiLAogICAgICAgICAgICByZWFzb25zPXR1"
    "cGxlKHJlYXNvbnMpLAogICAgICAgICkKCiAgICByZXR1cm4gQmVoYXZpb3JEZWNpc2lvbigK"
    "ICAgICAgICBjYXBhYmlsaXR5PSJQNV9QUkVGSVhfQ0FDSEVfQkVIQVZJT1IiLAogICAgICAg"
    "IHN0YXRlPUJlaGF2aW9yU3RhdGUuUEFTUywKICAgICAgICBmYWlsdXJlX2NsYXNzPU5vbmUs"
    "CiAgICAgICAgcmVhc29ucz0oCiAgICAgICAgICAgICJjYWNoZS1zcGVjaWZpYyB0b2tlbiBl"
    "dmlkZW5jZSBzYXRpc2ZpZWQgYWxsIGZyb3plbiBQNSBjb250cm9scyIsCiAgICAgICAgICAg"
    "IGYibmVnYXRpdmVfcHJlZml4X2NhY2hlYWJsZV9jb21tb25fcHJlZml4X2JvdW5kPXtuZWdh"
    "dGl2ZV9ib3VuZH0iLAogICAgICAgICksCiAgICApCgoKZGVmIGRlY2lkZV9wNigKICAgIGNy"
    "b3NzX3dvcmtlcl9jb2xkOiBkaWN0W3N0ciwgb2JqZWN0XSwKICAgIHdvcmtlcl8xX3JldGVu"
    "dGlvbjogZGljdFtzdHIsIG9iamVjdF0sCiAgICBwcm9jZXNzX2lzb2xhdGlvbjogZGljdFtz"
    "dHIsIG9iamVjdF0sCiAgICBncHVfaXNvbGF0aW9uOiBkaWN0W3N0ciwgb2JqZWN0XSwKKSAt"
    "PiBCZWhhdmlvckRlY2lzaW9uOgogICAgY3Jvc3Nfcm91dGUgPSBfcmVxdWVzdF9yb3V0ZShj"
    "cm9zc193b3JrZXJfY29sZCkKICAgIHJldGVudGlvbl9yb3V0ZSA9IF9yZXF1ZXN0X3JvdXRl"
    "KHdvcmtlcl8xX3JldGVudGlvbikKICAgIGNyb3NzX21ldHJpYyA9IF9yZXF1ZXN0X21ldHJp"
    "Y19kZWx0YShjcm9zc193b3JrZXJfY29sZCkKICAgIHJldGVudGlvbl9tZXRyaWMgPSBfcmVx"
    "dWVzdF9tZXRyaWNfZGVsdGEod29ya2VyXzFfcmV0ZW50aW9uKQoKICAgIHJlYXNvbnM6IGxp"
    "c3Rbc3RyXSA9IFtdCiAgICBmYWlsZWQgPSBGYWxzZQoKICAgIGlmIHByb2Nlc3NfaXNvbGF0"
    "aW9uLmdldCgid29ya2VyX3Byb2Nlc3NfdHJlZXNfZGlzam9pbnQiKSBpcyBub3QgVHJ1ZToK"
    "ICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAgcmVhc29ucy5hcHBlbmQoIndvcmtlciBw"
    "cm9jZXNzIHRyZWVzIGFyZSBub3QgZGlzam9pbnQiKQogICAgaWYgZ3B1X2lzb2xhdGlvbi5n"
    "ZXQoIndvcmtlcl8xX2JvdW5kX3RvX2dwdV8wIikgaXMgbm90IFRydWU6CiAgICAgICAgZmFp"
    "bGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5kKCJ3b3JrZXIgMSBHUFUgcmVhbGl6"
    "YXRpb24gaXMgaW52YWxpZCIpCiAgICBpZiBncHVfaXNvbGF0aW9uLmdldCgid29ya2VyXzJf"
    "Ym91bmRfdG9fZ3B1XzEiKSBpcyBub3QgVHJ1ZToKICAgICAgICBmYWlsZWQgPSBUcnVlCiAg"
    "ICAgICAgcmVhc29ucy5hcHBlbmQoIndvcmtlciAyIEdQVSByZWFsaXphdGlvbiBpcyBpbnZh"
    "bGlkIikKCiAgICBpZiAoCiAgICAgICAgY3Jvc3Nfcm91dGUuaW50ZW5kZWRfd29ya2VyICE9"
    "ICJ3b3JrZXJfMiIKICAgICAgICBvciBjcm9zc19yb3V0ZS5yZWFsaXplZF93b3JrZXIgIT0g"
    "Indvcmtlcl8yIgogICAgICAgIG9yIGNyb3NzX3JvdXRlLndvcmtlcl9nZW5lcmF0aW9uICE9"
    "IDEKICAgICk6CiAgICAgICAgZmFpbGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5k"
    "KCJDUk9TU19XT1JLRVJfQ09MRCByb3V0ZSByZWFsaXphdGlvbiBkcmlmdGVkIikKICAgIGlm"
    "ICgKICAgICAgICByZXRlbnRpb25fcm91dGUuaW50ZW5kZWRfd29ya2VyICE9ICJ3b3JrZXJf"
    "MSIKICAgICAgICBvciByZXRlbnRpb25fcm91dGUucmVhbGl6ZWRfd29ya2VyICE9ICJ3b3Jr"
    "ZXJfMSIKICAgICAgICBvciByZXRlbnRpb25fcm91dGUud29ya2VyX2dlbmVyYXRpb24gIT0g"
    "MgogICAgKToKICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAgcmVhc29ucy5hcHBlbmQo"
    "IldPUktFUjFfUkVURU5USU9OIHJvdXRlIHJlYWxpemF0aW9uIGRyaWZ0ZWQiKQoKICAgIGZv"
    "ciByb3V0ZSBpbiAoY3Jvc3Nfcm91dGUsIHJldGVudGlvbl9yb3V0ZSk6CiAgICAgICAgaWYg"
    "cm91dGUucm91dGVfcmVhc29uICE9ICJESVJFQ1RfTE9PUEJBQ0tfRU5EUE9JTlQiOgogICAg"
    "ICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAgICAgIHJlYXNvbnMuYXBwZW5kKGYie3Jv"
    "dXRlLnJlcXVlc3Rfcm9sZX0gdXNlZCBhbiB1bmV4cGVjdGVkIHJvdXRlIHJlYXNvbiIpCiAg"
    "ICAgICAgaWYgcm91dGUuZmFsbGJhY2tfcmVhc29uIGlzIG5vdCBOb25lOgogICAgICAgICAg"
    "ICBmYWlsZWQgPSBUcnVlCiAgICAgICAgICAgIHJlYXNvbnMuYXBwZW5kKGYie3JvdXRlLnJl"
    "cXVlc3Rfcm9sZX0gcmVwb3J0ZWQgZmFsbGJhY2siKQoKICAgIGlmIGNyb3NzX21ldHJpYy5s"
    "b2NhbF9jYWNoZV9oaXQgIT0gMDoKICAgICAgICBmYWlsZWQgPSBUcnVlCiAgICAgICAgcmVh"
    "c29ucy5hcHBlbmQoIndvcmtlciAyIGluaGVyaXRlZCBwcm9oaWJpdGVkIHdvcmtlci0xIGxv"
    "Y2FsIGNhY2hlIHN0YXRlIikKICAgIGlmIHJldGVudGlvbl9tZXRyaWMubG9jYWxfY2FjaGVf"
    "aGl0IDw9IDA6CiAgICAgICAgZmFpbGVkID0gVHJ1ZQogICAgICAgIHJlYXNvbnMuYXBwZW5k"
    "KCJ3b3JrZXIgMSBkaWQgbm90IHJldGFpbiBhdHRyaWJ1dGFibGUgbG9jYWwgY2FjaGUgc3Rh"
    "dGUiKQoKICAgIGlmIGZhaWxlZDoKICAgICAgICByZXR1cm4gQmVoYXZpb3JEZWNpc2lvbigK"
    "ICAgICAgICAgICAgY2FwYWJpbGl0eT0iUDZfV09SS0VSX1NUQVRFX0lTT0xBVElPTiIsCiAg"
    "ICAgICAgICAgIHN0YXRlPUJlaGF2aW9yU3RhdGUuRkFJTCwKICAgICAgICAgICAgZmFpbHVy"
    "ZV9jbGFzcz0iUDZfQkVIQVZJT1JfRkFJTFVSRSIsCiAgICAgICAgICAgIHJlYXNvbnM9dHVw"
    "bGUocmVhc29ucyksCiAgICAgICAgKQoKICAgIHJldHVybiBCZWhhdmlvckRlY2lzaW9uKAog"
    "ICAgICAgIGNhcGFiaWxpdHk9IlA2X1dPUktFUl9TVEFURV9JU09MQVRJT04iLAogICAgICAg"
    "IHN0YXRlPUJlaGF2aW9yU3RhdGUuUEFTUywKICAgICAgICBmYWlsdXJlX2NsYXNzPU5vbmUs"
    "CiAgICAgICAgcmVhc29ucz0oIndvcmtlciBpZGVudGl0eSwgcm91dGUgcmVhbGl6YXRpb24s"
    "IHN0YXRlIGlzb2xhdGlvbiwgYW5kIHJldGVudGlvbiBwYXNzZWQiLCksCiAgICApCgoKZGVm"
    "IGV2aWRlbmNlX3Byb2plY3Rpb24oCiAgICBkZWNpc2lvbjogQmVoYXZpb3JEZWNpc2lvbiwK"
    "ICAgIG9ic2VydmF0aW9uczogZGljdFtzdHIsIG9iamVjdF0sCikgLT4gRXZpZGVuY2VQcm9q"
    "ZWN0aW9uOgogICAgcmV0dXJuIEV2aWRlbmNlUHJvamVjdGlvbigKICAgICAgICBkZWNpc2lv"
    "bj1kZWNpc2lvbiwKICAgICAgICBwdWJsaWNfc3VtbWFyeT17CiAgICAgICAgICAgICJjYXBh"
    "YmlsaXR5IjogZGVjaXNpb24uY2FwYWJpbGl0eSwKICAgICAgICAgICAgInN0YXRlIjogZGVj"
    "aXNpb24uc3RhdGUudmFsdWUsCiAgICAgICAgICAgICJmYWlsdXJlX2NsYXNzIjogZGVjaXNp"
    "b24uZmFpbHVyZV9jbGFzcywKICAgICAgICAgICAgInJlYXNvbnMiOiBsaXN0KGRlY2lzaW9u"
    "LnJlYXNvbnMpLAogICAgICAgICAgICAib2JzZXJ2YXRpb25zIjogb2JzZXJ2YXRpb25zLAog"
    "ICAgICAgIH0sCiAgICApCgoKZGVmIGRlY2lzaW9uX2ludmFyaWFudF91bmRlcl9wcm9qZWN0"
    "aW9uX3BvbGljeSgKICAgIGRlY2lzaW9uOiBCZWhhdmlvckRlY2lzaW9uLAopIC0+IGJvb2w6"
    "CiAgICBmaXJzdCA9IGV2aWRlbmNlX3Byb2plY3Rpb24oCiAgICAgICAgZGVjaXNpb24sCiAg"
    "ICAgICAgeyJkaXNwbGF5X3BhdGgiOiAiPHdvcmtpbmc+IiwgImV4Y2VycHRfbGltaXQiOiAy"
    "NTZ9LAogICAgKQogICAgc2Vjb25kID0gZXZpZGVuY2VfcHJvamVjdGlvbigKICAgICAgICBk"
    "ZWNpc2lvbiwKICAgICAgICB7ImRpc3BsYXlfcGF0aCI6ICI8cmVkYWN0ZWQ+IiwgImV4Y2Vy"
    "cHRfbGltaXQiOiA0MDk2fSwKICAgICkKICAgIHJldHVybiBmaXJzdC5kZWNpc2lvbiA9PSBz"
    "ZWNvbmQuZGVjaXNpb24gPT0gZGVjaXNpb24KCgpkZWYgcmVxdWVzdF9wdWJsaWNfZXZpZGVu"
    "Y2UocGF5bG9hZDogZGljdFtzdHIsIG9iamVjdF0pIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgog"
    "ICAgbWV0cmljID0gX3JlcXVlc3RfbWV0cmljX2RlbHRhKHBheWxvYWQpCiAgICB0b2tlbiA9"
    "IF9yZXF1ZXN0X3Rva2VuX2lkZW50aXR5KHBheWxvYWQpCiAgICByb3V0ZSA9IF9yZXF1ZXN0"
    "X3JvdXRlKHBheWxvYWQpCiAgICByZXR1cm4gewogICAgICAgICJyZXF1ZXN0X2lkIjogcGF5"
    "bG9hZFsicmVxdWVzdF9pZCJdLAogICAgICAgICJyZXF1ZXN0X3JvbGUiOiBwYXlsb2FkWyJy"
    "ZXF1ZXN0X3JvbGUiXSwKICAgICAgICAicHJlZml4X3ZhcmlhbnQiOiBwYXlsb2FkWyJwcmVm"
    "aXhfdmFyaWFudCJdLAogICAgICAgICJ0b2tlbl9pZGVudGl0eSI6IHRva2VuX2lkZW50aXR5"
    "X2V2aWRlbmNlKHRva2VuKSwKICAgICAgICAibWV0cmljX2RlbHRhIjogYXNkaWN0KG1ldHJp"
    "YyksCiAgICAgICAgInJvdXRlX29ic2VydmF0aW9uIjogYXNkaWN0KHJvdXRlKSwKICAgICAg"
    "ICAic3RydWN0dXJlZF9vdXRwdXRfc2hhMjU2IjogcGF5bG9hZFsic3RydWN0dXJlZF9vdXRw"
    "dXRfc2hhMjU2Il0sCiAgICAgICAgInN0cnVjdHVyZWRfb3V0cHV0X3ZhbGlkIjogcGF5bG9h"
    "ZFsic3RydWN0dXJlZF9vdXRwdXRfdmFsaWQiXSwKICAgICAgICAicHJvbXB0X3Rva2VucyI6"
    "IHBheWxvYWRbInByb21wdF90b2tlbnMiXSwKICAgICAgICAiY29tcGxldGlvbl90b2tlbnMi"
    "OiBwYXlsb2FkWyJjb21wbGV0aW9uX3Rva2VucyJdLAogICAgfQoKCmRlZiBidW5kbGVfb3V0"
    "cHV0cygpIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgcmVxdWlyZWRfYmVmb3JlX21hbmlm"
    "ZXN0ID0gc2V0KE9VVFBVVF9OQU1FUykgLSB7ImJ1bmRsZV9tYW5pZmVzdF92MS5qc29uIn0K"
    "ICAgIG9ic2VydmVkX2JlZm9yZV9tYW5pZmVzdCA9IHtwYXRoLm5hbWUgZm9yIHBhdGggaW4g"
    "T1VUUFVUX1JPT1QuaXRlcmRpcigpIGlmIHBhdGguaXNfZmlsZSgpfQogICAgbWlzc2luZyA9"
    "IHJlcXVpcmVkX2JlZm9yZV9tYW5pZmVzdCAtIG9ic2VydmVkX2JlZm9yZV9tYW5pZmVzdAog"
    "ICAgdW5leHBlY3RlZCA9IG9ic2VydmVkX2JlZm9yZV9tYW5pZmVzdCAtIHJlcXVpcmVkX2Jl"
    "Zm9yZV9tYW5pZmVzdAogICAgaWYgbWlzc2luZyBvciB1bmV4cGVjdGVkOgogICAgICAgIHJh"
    "aXNlIFJ1bnRpbWVFcnJvcigKICAgICAgICAgICAgInJ1bnRpbWUgZXZpZGVuY2Ugb3V0cHV0"
    "IGNvbnRyYWN0IGRyaWZ0ZWQ6ICIKICAgICAgICAgICAgKyBjYW5vbmljYWxfanNvbigKICAg"
    "ICAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICAgICAibWlzc2luZyI6IHNvcnRlZCht"
    "aXNzaW5nKSwKICAgICAgICAgICAgICAgICAgICAidW5leHBlY3RlZCI6IHNvcnRlZCh1bmV4"
    "cGVjdGVkKSwKICAgICAgICAgICAgICAgIH0KICAgICAgICAgICAgKQogICAgICAgICkKICAg"
    "IGVudHJpZXMgPSBbXQogICAgZm9yIG5hbWUgaW4gT1VUUFVUX05BTUVTOgogICAgICAgIHBh"
    "dGggPSBPVVRQVVRfUk9PVCAvIG5hbWUKICAgICAgICBpZiBub3QgcGF0aC5pc19maWxlKCk6"
    "CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgZW50cmllcy5hcHBlbmQoCiAgICAgICAg"
    "ICAgIHsKICAgICAgICAgICAgICAgICJwYXRoIjogbmFtZSwKICAgICAgICAgICAgICAgICJz"
    "aGEyNTYiOiBmaWxlX3NoYTI1NihwYXRoKSwKICAgICAgICAgICAgICAgICJzaXplX2J5dGVz"
    "IjogcGF0aC5zdGF0KCkuc3Rfc2l6ZSwKICAgICAgICAgICAgfQogICAgICAgICkKICAgIG1h"
    "bmlmZXN0X3BheWxvYWQgPSB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwK"
    "ICAgICAgICAiZGlhZ25vc3RpY19pZCI6ICJhdXJhZ2F0ZXdheS1wNS1wNi1leGFjdC1ydW50"
    "aW1lLXJlcXVhbGlmaWNhdGlvbi1ydW50aW1lLXF1YWxpZmljYXRpb24tdjEiLAogICAgICAg"
    "ICJzb3VyY2VfbWFpbl9jb21taXQiOiBTT1VSQ0VfTUFJTl9DT01NSVQsCiAgICAgICAgIm1l"
    "bWJlcnMiOiBbaXRlbSBmb3IgaXRlbSBpbiBlbnRyaWVzIGlmIGl0ZW1bInBhdGgiXSAhPSAi"
    "YnVuZGxlX21hbmlmZXN0X3YxLmpzb24iXSwKICAgICAgICAic2NyYXRjaF9kaXJlY3Rvcmll"
    "c19pbmNsdWRlZCI6IEZhbHNlLAogICAgICAgICJ3b3JrZXJfbG9nX2RpcmVjdG9yeV9pbmNs"
    "dWRlZCI6IEZhbHNlLAogICAgfQogICAgd3JpdGVfanNvbihPVVRQVVRfUk9PVCAvICJidW5k"
    "bGVfbWFuaWZlc3RfdjEuanNvbiIsIG1hbmlmZXN0X3BheWxvYWQpCiAgICB3aXRoIHppcGZp"
    "bGUuWmlwRmlsZSgKICAgICAgICBFVklERU5DRV9aSVAsCiAgICAgICAgInciLAogICAgICAg"
    "IGNvbXByZXNzaW9uPXppcGZpbGUuWklQX0RFRkxBVEVELAogICAgICAgIGNvbXByZXNzbGV2"
    "ZWw9OSwKICAgICkgYXMgYXJjaGl2ZToKICAgICAgICBmb3IgbmFtZSBpbiBPVVRQVVRfTkFN"
    "RVM6CiAgICAgICAgICAgIHBhdGggPSBPVVRQVVRfUk9PVCAvIG5hbWUKICAgICAgICAgICAg"
    "aWYgcGF0aC5pc19maWxlKCk6CiAgICAgICAgICAgICAgICBhcmNoaXZlLndyaXRlKHBhdGgs"
    "IGFyY25hbWU9bmFtZSkKICAgIGlmIEVWSURFTkNFX1pJUC5zdGF0KCkuc3Rfc2l6ZSA+IE1B"
    "WF9FVklERU5DRV9aSVBfQllURVM6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJydW50"
    "aW1lIGV2aWRlbmNlIFpJUCBleGNlZWRzIHRoZSBieXRlIGJ1ZGdldCIpCiAgICByZXR1cm4g"
    "ewogICAgICAgICJldmlkZW5jZV96aXAiOiBzdHIoRVZJREVOQ0VfWklQKSwKICAgICAgICAi"
    "ZXZpZGVuY2VfemlwX3NoYTI1NiI6IGZpbGVfc2hhMjU2KEVWSURFTkNFX1pJUCksCiAgICAg"
    "ICAgImV2aWRlbmNlX3ppcF9zaXplX2J5dGVzIjogRVZJREVOQ0VfWklQLnN0YXQoKS5zdF9z"
    "aXplLAogICAgfQoKCmRlZiB3cml0ZV9jYXBhYmlsaXR5X3Rlcm1pbmFsX3JlcG9ydHMoCiAg"
    "ICBjb21wbGV0ZWQ6IGxpc3Rbc3RyXSwKICAgIGZhaWxlZF9jYXBhYmlsaXR5OiBzdHIgfCBO"
    "b25lLAogICAgZmFpbHVyZV9jbGFzczogc3RyIHwgTm9uZSwKICAgIGNvdW50ZXJzOiBkaWN0"
    "W3N0ciwgaW50XSwKKSAtPiBOb25lOgogICAgcmVwb3J0cyA9ICgKICAgICAgICAoIkMxIiwg"
    "ImMxX21vZGVsX2NvbnN0cnVjdGlvbl9yZXBvcnRfdjEuanNvbiIpLAogICAgICAgICgiQzIi"
    "LCAiYzJfd29ya2VyX3N0YXJ0dXBfcmVwb3J0X3YxLmpzb24iKSwKICAgICAgICAoIkMzIiwg"
    "ImMzX3NpbmdsZV9yZXF1ZXN0X3JlcG9ydF92MS5qc29uIiksCiAgICAgICAgKCJDNCIsICJj"
    "NF9vdXRwdXRfY29udHJhY3RfcmVwb3J0X3YxLmpzb24iKSwKICAgICAgICAoIlA1IiwgInA1"
    "X2NhY2hlX2JlaGF2aW9yX3JlcG9ydF92MS5qc29uIiksCiAgICAgICAgKCJQNiIsICJwNl93"
    "b3JrZXJfc3RhdGVfaXNvbGF0aW9uX3JlcG9ydF92MS5qc29uIiksCiAgICApCiAgICBmb3Ig"
    "Y2FwYWJpbGl0eV9pZCwgbmFtZSBpbiByZXBvcnRzOgogICAgICAgIHBhdGggPSBPVVRQVVRf"
    "Uk9PVCAvIG5hbWUKICAgICAgICBpZiBwYXRoLmlzX2ZpbGUoKToKICAgICAgICAgICAgY29u"
    "dGludWUKICAgICAgICBpc19mYWlsZWQgPSBmYWlsZWRfY2FwYWJpbGl0eSA9PSBjYXBhYmls"
    "aXR5X2lkCiAgICAgICAgYmxvY2tlZF9ieSA9IE5vbmUKICAgICAgICBpZiBub3QgaXNfZmFp"
    "bGVkOgogICAgICAgICAgICBibG9ja2VkX2J5ID0gZmFpbGVkX2NhcGFiaWxpdHkgb3IgZmFp"
    "bHVyZV9jbGFzcyBvciAiVVBTVFJFQU1fUFJFQ09ORElUSU9OIgogICAgICAgIHdyaXRlX2pz"
    "b24oCiAgICAgICAgICAgIHBhdGgsCiAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICJz"
    "Y2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgICAgICAgICAiY2FwYWJpbGl0eV9p"
    "ZCI6IGNhcGFiaWxpdHlfaWQsCiAgICAgICAgICAgICAgICAic3RhdHVzIjogIkZBSUxFRCIg"
    "aWYgaXNfZmFpbGVkIGVsc2UgIk5PVF9SVU4iLAogICAgICAgICAgICAgICAgImRlY2lzaW9u"
    "X3N0YXRlIjogKEJlaGF2aW9yU3RhdGUuRkFJTC52YWx1ZSBpZiBpc19mYWlsZWQgZWxzZSBO"
    "b25lKSwKICAgICAgICAgICAgICAgICJibG9ja2VkX2J5IjogYmxvY2tlZF9ieSwKICAgICAg"
    "ICAgICAgICAgICJmYWlsdXJlX2NsYXNzIjogZmFpbHVyZV9jbGFzcywKICAgICAgICAgICAg"
    "ICAgICJjb21wbGV0ZWRfY2FwYWJpbGl0aWVzX2JlZm9yZV90ZXJtaW5hbF9zdGF0ZSI6IGNv"
    "bXBsZXRlZCwKICAgICAgICAgICAgICAgICJnbG9iYWxfbW9kZWxfcmVxdWVzdF9jb3VudCI6"
    "IGNvdW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdLAogICAgICAgICAgICAgICAgInJhd19wcm9t"
    "cHRfbG9nZ2VkIjogRmFsc2UsCiAgICAgICAgICAgICAgICAicmF3X291dHB1dF9sb2dnZWQi"
    "OiBGYWxzZSwKICAgICAgICAgICAgICAgICJwdWJsaWNfZXZpZGVuY2VfdXNlZF9hc19zZW1h"
    "bnRpY19pbnB1dCI6IEZhbHNlLAogICAgICAgICAgICB9LAogICAgICAgICkKCgpkZWYgZW5z"
    "dXJlX3J1bnRpbWVfc291cmNlX2lkZW50aXR5X3JlcG9ydCgKICAgIGZhaWx1cmVfY29kZTog"
    "c3RyIHwgTm9uZSwKKSAtPiBOb25lOgogICAgcGF0aCA9IE9VVFBVVF9ST09UIC8gInJ1bnRp"
    "bWVfc291cmNlX2lkZW50aXR5X3JlcG9ydF92MS5qc29uIgogICAgaWYgcGF0aC5pc19maWxl"
    "KCk6CiAgICAgICAgcmV0dXJuCiAgICB3cml0ZV9qc29uKAogICAgICAgIHBhdGgsCiAgICAg"
    "ICAgewogICAgICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAg"
    "ICAicmVwb3J0X2lkIjogKAogICAgICAgICAgICAgICAgImF1cmFnYXRld2F5LXA1LXA2LWV4"
    "YWN0LXJ1bnRpbWUtcmVxdWFsaWZpY2F0aW9uLXJ1bnRpbWUtc291cmNlLWlkZW50aXR5LXYx"
    "IgogICAgICAgICAgICApLAogICAgICAgICAgICAic3RhdHVzIjogIkZBSUxFRCIsCiAgICAg"
    "ICAgICAgICJkZWNpc2lvbiI6ICJFWEVDVVRFRF9SVU5USU1FX1NDUklQVF9JREVOVElUWV9V"
    "TlZFUklGSUVEIiwKICAgICAgICAgICAgIm5vdGVib29rX25hbWUiOiBOT1RFQk9PS19OQU1F"
    "LAogICAgICAgICAgICAic291cmNlX21haW5fY29tbWl0IjogU09VUkNFX01BSU5fQ09NTUlU"
    "LAogICAgICAgICAgICAiZXhlY3V0ZWRfcnVudGltZV9zY3JpcHRfc2hhMjU2IjogTm9uZSwK"
    "ICAgICAgICAgICAgIndyYXBwZXJfaGFzaF92ZXJpZmljYXRpb25fcGFzc2VkIjogRmFsc2Us"
    "CiAgICAgICAgICAgICJmYWlsdXJlX2NvZGUiOiAoZmFpbHVyZV9jb2RlIG9yICJQM19QNl9S"
    "VU5USU1FX1NPVVJDRV9JREVOVElUWV9NSVNNQVRDSCIpLAogICAgICAgIH0sCiAgICApCgoK"
    "ZGVmIGVuc3VyZV9pbnN0YWxsX3JlcG9ydChmYWlsdXJlX2NvZGU6IHN0ciB8IE5vbmUpIC0+"
    "IE5vbmU6CiAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyAicnVudGltZV9pbnN0YWxsX3JlcG9y"
    "dF92MS5qc29uIgogICAgaWYgcGF0aC5pc19maWxlKCk6CiAgICAgICAgcmV0dXJuCiAgICB3"
    "cml0ZV9qc29uKAogICAgICAgIHBhdGgsCiAgICAgICAgewogICAgICAgICAgICAic2NoZW1h"
    "X3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAicmVwb3J0X2lkIjogImF1cmFnYXRl"
    "d2F5LXA1LXA2LWV4YWN0LXJ1bnRpbWUtcmVxdWFsaWZpY2F0aW9uLXJ1bnRpbWUtaW5zdGFs"
    "bC12MSIsCiAgICAgICAgICAgICJjb21tYW5kX3JvbGUiOiAib2ZmbGluZV90YXJnZXRfcnVu"
    "dGltZV9pbnN0YWxsIiwKICAgICAgICAgICAgInN0YXR1cyI6ICJOT1RfUlVOIiwKICAgICAg"
    "ICAgICAgInByb2Nlc3Nfb3V0Y29tZSI6ICJOT1RfUlVOIiwKICAgICAgICAgICAgInJlYXNv"
    "biI6IGZhaWx1cmVfY29kZSBvciAicnVudGltZSBpbnN0YWxsYXRpb24gd2FzIG5vdCByZWFj"
    "aGVkIiwKICAgICAgICAgICAgInJldHVybmNvZGUiOiBOb25lLAogICAgICAgICAgICAidGlt"
    "ZWRfb3V0IjogRmFsc2UsCiAgICAgICAgICAgICJzdGRvdXRfdGFpbCI6ICIiLAogICAgICAg"
    "ICAgICAic3RkZXJyX3RhaWwiOiAiIiwKICAgICAgICAgICAgImZhaWx1cmVfc2lnbmFscyI6"
    "IFtdLAogICAgICAgICAgICAibmV0d29ya19hY2Nlc3NfcmVxdWVzdGVkIjogRmFsc2UsCiAg"
    "ICAgICAgICAgICJoaWRkZW5fcmV0cnlfY291bnQiOiAwLAogICAgICAgICAgICAibW9kZWxf"
    "Y29weV9jb21wbGV0ZWRfYmVmb3JlX2luc3RhbGwiOiBGYWxzZSwKICAgICAgICAgICAgInJv"
    "b3RfY2F1c2VfcmV2aWV3X3JlcXVpcmVkIjogRmFsc2UsCiAgICAgICAgfSwKICAgICkKCgpk"
    "ZWYgZW5zdXJlX2ltcG9ydF9jbG9zdXJlX3JlcG9ydChmYWlsdXJlX2NvZGU6IHN0ciB8IE5v"
    "bmUpIC0+IE5vbmU6CiAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyAicnVudGltZV9pbXBvcnRf"
    "Y2xvc3VyZV9yZXBvcnRfdjEuanNvbiIKICAgIGlmIHBhdGguaXNfZmlsZSgpOgogICAgICAg"
    "IHJldHVybgogICAgd3JpdGVfanNvbigKICAgICAgICBwYXRoLAogICAgICAgIHsKICAgICAg"
    "ICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgInJlcG9ydF9p"
    "ZCI6ICgKICAgICAgICAgICAgICAgICJhdXJhZ2F0ZXdheS1wNS1wNi1leGFjdC1ydW50aW1l"
    "LXJlcXVhbGlmaWNhdGlvbi1ydW50aW1lLWltcG9ydC1jbG9zdXJlLXYxIgogICAgICAgICAg"
    "ICApLAogICAgICAgICAgICAic3RhdHVzIjogIk5PVF9SVU4iLAogICAgICAgICAgICAicHJv"
    "Y2Vzc19vdXRjb21lIjogIk5PVF9SVU4iLAogICAgICAgICAgICAicmVhc29uIjogKGZhaWx1"
    "cmVfY29kZSBvciAicnVudGltZSBpbXBvcnQtY2xvc3VyZSBwcm9iZSB3YXMgbm90IHJlYWNo"
    "ZWQiKSwKICAgICAgICAgICAgInRhcmdldF9zaXRlIjogc3RyKFRBUkdFVF9TSVRFKSwKICAg"
    "ICAgICAgICAgInB5dGhvbnBhdGhfZXhhY3RfdGFyZ2V0X3NpdGUiOiBGYWxzZSwKICAgICAg"
    "ICAgICAgIm5lc3RlZF9pbnRlcnByZXRlcl9kZXB0aCI6IDIsCiAgICAgICAgICAgICJtb2Rl"
    "bF9sb2Fkc19jb25zdW1lZCI6IDAsCiAgICAgICAgICAgICJ3b3JrZXJfc3RhcnRzX2NvbnN1"
    "bWVkIjogMCwKICAgICAgICAgICAgIm5ldHdvcmtfYWNjZXNzX3JlcXVlc3RlZCI6IEZhbHNl"
    "LAogICAgICAgICAgICAiaGlkZGVuX3JldHJ5X2NvdW50IjogMCwKICAgICAgICB9LAogICAg"
    "KQoKCmRlZiBjbGVhbnVwX3NjcmF0Y2goKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHNu"
    "YXBzaG90X2Vycm9yX3R5cGUgPSBOb25lCiAgICBzbmFwc2hvdF9zYWZlX21lc3NhZ2UgPSBO"
    "b25lCiAgICB0cnk6CiAgICAgICAgYmVmb3JlID0gZGlyZWN0b3J5X3NuYXBzaG90KFNDUkFU"
    "Q0hfUk9PVCkKICAgIGV4Y2VwdCAoT1NFcnJvciwgUnVudGltZUVycm9yLCBWYWx1ZUVycm9y"
    "KSBhcyBlcnJvcjoKICAgICAgICBzbmFwc2hvdF9lcnJvcl90eXBlID0gdHlwZShlcnJvciku"
    "X19uYW1lX18KICAgICAgICBzbmFwc2hvdF9zYWZlX21lc3NhZ2UgPSBzYW5pdGl6ZV9leGNl"
    "cnB0KHN0cihlcnJvcikpCiAgICAgICAgYmVmb3JlID0gewogICAgICAgICAgICAiZXhpc3Rz"
    "IjogU0NSQVRDSF9ST09ULmV4aXN0cygpLAogICAgICAgICAgICAic25hcHNob3RfYXZhaWxh"
    "YmxlIjogRmFsc2UsCiAgICAgICAgfQoKICAgIGRlbGV0ZV9lcnJvcl90eXBlID0gTm9uZQog"
    "ICAgZGVsZXRlX3NhZmVfbWVzc2FnZSA9IE5vbmUKICAgIHRyeToKICAgICAgICBpZiBTQ1JB"
    "VENIX1JPT1QuZXhpc3RzKCk6CiAgICAgICAgICAgIHNodXRpbC5ybXRyZWUoU0NSQVRDSF9S"
    "T09UKQogICAgZXhjZXB0IE9TRXJyb3IgYXMgZXJyb3I6CiAgICAgICAgZGVsZXRlX2Vycm9y"
    "X3R5cGUgPSB0eXBlKGVycm9yKS5fX25hbWVfXwogICAgICAgIGRlbGV0ZV9zYWZlX21lc3Nh"
    "Z2UgPSBzYW5pdGl6ZV9leGNlcnB0KHN0cihlcnJvcikpCgogICAgc3RhdHVzID0gKAogICAg"
    "ICAgICJQQVNTRUQiCiAgICAgICAgaWYgKHNuYXBzaG90X2Vycm9yX3R5cGUgaXMgTm9uZSBh"
    "bmQgZGVsZXRlX2Vycm9yX3R5cGUgaXMgTm9uZSBhbmQgbm90IFNDUkFUQ0hfUk9PVC5leGlz"
    "dHMoKSkKICAgICAgICBlbHNlICJGQUlMRUQiCiAgICApCiAgICByZXBvcnQgPSB7CiAgICAg"
    "ICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAicmVwb3J0X2lkIjogKCJh"
    "dXJhZ2F0ZXdheS1wNS1wNi1leGFjdC1ydW50aW1lLXJlcXVhbGlmaWNhdGlvbi1zY3JhdGNo"
    "LWNsZWFudXAtdjEiKSwKICAgICAgICAic3RhdHVzIjogc3RhdHVzLAogICAgICAgICJzY3Jh"
    "dGNoX2JlZm9yZSI6IGJlZm9yZSwKICAgICAgICAic2NyYXRjaF9leGlzdHNfYWZ0ZXIiOiBT"
    "Q1JBVENIX1JPT1QuZXhpc3RzKCksCiAgICAgICAgInNuYXBzaG90X2Vycm9yX3R5cGUiOiBz"
    "bmFwc2hvdF9lcnJvcl90eXBlLAogICAgICAgICJzbmFwc2hvdF9zYWZlX21lc3NhZ2UiOiBz"
    "bmFwc2hvdF9zYWZlX21lc3NhZ2UsCiAgICAgICAgImRlbGV0ZV9lcnJvcl90eXBlIjogZGVs"
    "ZXRlX2Vycm9yX3R5cGUsCiAgICAgICAgImRlbGV0ZV9zYWZlX21lc3NhZ2UiOiBkZWxldGVf"
    "c2FmZV9tZXNzYWdlLAogICAgICAgICJzZWNvbmRhcnlfZmFpbHVyZV9vbmx5IjogVHJ1ZSwK"
    "ICAgIH0KICAgIHRyeToKICAgICAgICB3cml0ZV9qc29uKE9VVFBVVF9ST09UIC8gInNjcmF0"
    "Y2hfY2xlYW51cF9yZXBvcnRfdjEuanNvbiIsIHJlcG9ydCkKICAgICAgICByZXBvcnRbInJl"
    "cG9ydF9wZXJzaXN0ZWQiXSA9IFRydWUKICAgIGV4Y2VwdCBPU0Vycm9yOgogICAgICAgIHJl"
    "cG9ydFsicmVwb3J0X3BlcnNpc3RlZCJdID0gRmFsc2UKICAgICAgICBpZiByZXBvcnRbInN0"
    "YXR1cyJdID09ICJQQVNTRUQiOgogICAgICAgICAgICByZXBvcnRbInN0YXR1cyJdID0gIkZB"
    "SUxFRCIKICAgIHJldHVybiByZXBvcnQKCgpkZWYgc2FmZV93b3JrZXJfdGVhcmRvd24od29y"
    "a2VyOiBXb3JrZXIsIHJlYXNvbjogc3RyKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHRy"
    "eToKICAgICAgICByZXR1cm4gd29ya2VyLnN0b3BfYW5kX3JlcG9ydChyZWFzb24pCiAgICBl"
    "eGNlcHQgKAogICAgICAgIE9TRXJyb3IsCiAgICAgICAgUnVudGltZUVycm9yLAogICAgICAg"
    "IFZhbHVlRXJyb3IsCiAgICAgICAgc3VicHJvY2Vzcy5TdWJwcm9jZXNzRXJyb3IsCiAgICAp"
    "IGFzIGVycm9yOgogICAgICAgIHJldHVybiB7CiAgICAgICAgICAgICJ3b3JrZXJfaWQiOiB3"
    "b3JrZXIud29ya2VyX2lkLAogICAgICAgICAgICAid29ya2VyX2luc3RhbmNlX2lkIjogd29y"
    "a2VyLmluc3RhbmNlX2lkLAogICAgICAgICAgICAic3RhdHVzIjogIkZBSUxFRCIsCiAgICAg"
    "ICAgICAgICJyZWFzb24iOiByZWFzb24sCiAgICAgICAgICAgICJ0ZWFyZG93bl9lcnJvcl90"
    "eXBlIjogdHlwZShlcnJvcikuX19uYW1lX18sCiAgICAgICAgICAgICJ0ZWFyZG93bl9lcnJv"
    "cl9tZXNzYWdlIjogc2FuaXRpemVfZXhjZXJwdChzdHIoZXJyb3IpKVs6NTEyXSwKICAgICAg"
    "ICAgICAgInNlY29uZGFyeV9mYWlsdXJlX29ubHkiOiBUcnVlLAogICAgICAgIH0KCgpSRVBF"
    "VElUSU9OX0lNUExFTUVOVEFUSU9OX0JBU0VfQ09NTUlUOiBGaW5hbCA9ICJlM2M0Mjk2OWE4"
    "M2IwMWFhZGNmOTg5ZmQ4MDYwMDRmZWVhNzhmM2M1IgpSRVBFVElUSU9OX0RFU0lHTl9SRUNP"
    "UkRfU0hBMjU2OiBGaW5hbCA9ICgKICAgICIxZGMwMGMwYmMzNmExOTc5MjkxMDc4YjE2Yzdi"
    "NTRlZDUwMjM4NWJiNjJhNTNjNjA0M2QyNTVjNWZjZjRmYTAwIgopClJFUEVUSVRJT05fUkVR"
    "VUVTVF9PUkRFUjogRmluYWwgPSAoCiAgICAiQ09OVFJPTF8xWCIsCiAgICAiVFJFQVRNRU5U"
    "XzI0WCIsCiAgICAiVFJFQVRNRU5UXzI0WCIsCiAgICAiQ09OVFJPTF8xWCIsCiAgICAiQ09O"
    "VFJPTF8xWCIsCiAgICAiVFJFQVRNRU5UXzI0WCIsCikKUkVQRVRJVElPTl9DT05UUk9MX0NP"
    "VU5UOiBGaW5hbCA9IDEKUkVQRVRJVElPTl9UUkVBVE1FTlRfQ09VTlQ6IEZpbmFsID0gMjQK"
    "UkVQRVRJVElPTl9UUkVBVE1FTlRfVE9LRU5fQ09VTlQ6IEZpbmFsID0gODk5ClJFUEVUSVRJ"
    "T05fVFJFQVRNRU5UX1RPS0VOX1NIQTI1NjogRmluYWwgPSAoCiAgICAiNmIzNDQ0OGQwODNl"
    "Njg4MjZjYWU4NGMwNjc1ODc2YzE4ZDdhNzBmYWI1MzI5OTg2MGI3OGYyYTVhMTg5MjJiMCIK"
    "KQpSRVBFVElUSU9OX1RSRUFUTUVOVF9QQVlMT0FEX1NIQTI1NjogRmluYWwgPSAoCiAgICAi"
    "YjAzODc2M2E1YTJjYjA5ZjBhNTY1ZGQ3ZDExYWM5NTljNDJjOWM5YTUzZjBmMmQ1ZTM4NGVk"
    "YjY1MzFjM2E4ZSIKKQpSRVBFVElUSU9OX0NPTlRFWFRfQk9EWTogRmluYWwgPSAoCiAgICAi"
    "QXVyYUdhdGV3YXkgZGV0ZXJtaW5pc3RpYyBwcmVmaXgtY2FjaGUgcmVsaWFiaWxpdHkgcHJv"
    "YmUgdmFyaWFudCBBLiAiCiAgICAiVGhpcyB0ZXh0IGlzIHN5bnRoZXRpYywgY29udGFpbnMg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIG11c3QgcmVtYWluICIKICAgICJieXRlLWlkZW50aWNh"
    "bCB3aGVuZXZlciBwcmVmaXggdmFyaWFudCBBIGlzIHJlcXVlc3RlZC4gIgopClJFUEVUSVRJ"
    "T05fT1VUUFVUX05BTUVTOiBGaW5hbCA9ICgKICAgICJydW50aW1lX3NvdXJjZV9pZGVudGl0"
    "eV9yZXBvcnRfdjEuanNvbiIsCiAgICAicnVudGltZV9pbnN0YWxsX3JlcG9ydF92MS5qc29u"
    "IiwKICAgICJydW50aW1lX2Vudmlyb25tZW50X3JlcG9ydF92MS5qc29uIiwKICAgICJydW50"
    "aW1lX2ltcG9ydF9jbG9zdXJlX3JlcG9ydF92MS5qc29uIiwKICAgICJwNF9wNV9yZXBldGl0"
    "aW9uX3J1bnRpbWVfcmVhZHlfdjEuanNvbiIsCiAgICAicHJlX3JlcXVlc3RfdG9rZW5faWRl"
    "bnRpdHlfam91cm5hbF92MS5qc29uIiwKICAgICJwNF9wNV9yZXBldGl0aW9uX3JlcXVlc3Rf"
    "cmVzdWx0c192MS5qc29uIiwKICAgICJwNF9wNV9yZXBldGl0aW9uX2RlY2lzaW9uX3YxLmpz"
    "b24iLAogICAgIndvcmtlcl90ZWFyZG93bl9yZXBvcnRfdjEuanNvbiIsCiAgICAic2NyYXRj"
    "aF9jbGVhbnVwX3JlcG9ydF92MS5qc29uIiwKICAgICJmYWlsdXJlX3JlcG9ydF92MS5qc29u"
    "IiwKICAgICJwNF9wNV9yZXBldGl0aW9uX3N1bW1hcnlfdjEuanNvbiIsCiAgICAiaHVtYW5f"
    "cmVwb3J0X3YxLm1kIiwKICAgICJidW5kbGVfbWFuaWZlc3RfdjEuanNvbiIsCikKCgpkZWYg"
    "cmVwZXRpdGlvbl9jb250ZXh0KHJlcGV0aXRpb25fY291bnQ6IGludCkgLT4gc3RyOgogICAg"
    "aWYgcmVwZXRpdGlvbl9jb3VudCBub3QgaW4gewogICAgICAgIFJFUEVUSVRJT05fQ09OVFJP"
    "TF9DT1VOVCwKICAgICAgICBSRVBFVElUSU9OX1RSRUFUTUVOVF9DT1VOVCwKICAgIH06CiAg"
    "ICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NF"
    "TUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAidW5zdXBwb3J0ZWQgY2FjaGUtY29udGV4"
    "dCByZXBldGl0aW9uIGNvdW50IiwKICAgICAgICApCiAgICBvYnNlcnZlZCA9IFJFUEVUSVRJ"
    "T05fQ09OVEVYVF9CT0RZICogcmVwZXRpdGlvbl9jb3VudCArIFNZU1RFTV9QUk9NUFQKICAg"
    "IGlmIHJlcGV0aXRpb25fY291bnQgPT0gUkVQRVRJVElPTl9UUkVBVE1FTlRfQ09VTlQgYW5k"
    "IG9ic2VydmVkICE9IFNZTlRIRVRJQ19DQUNIRV9DT05URVhUX0E6CiAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAiMjR4IHRyZWF0bWVudCBjb250ZXh0IG5vIGxvbmdlciBtYXRj"
    "aGVzIHRoZSBoaXN0b3JpY2FsIHByZWZpeC1BIHBheWxvYWQiLAogICAgICAgICkKICAgIHJl"
    "dHVybiBvYnNlcnZlZAoKCmRlZiByZXBldGl0aW9uX3JlcXVlc3RfbWVzc2FnZXMoCiAgICBj"
    "b25kaXRpb25faWQ6IHN0ciwKKSAtPiBsaXN0W2RpY3Rbc3RyLCBzdHJdXToKICAgIGlmIGNv"
    "bmRpdGlvbl9pZCA9PSAiQ09OVFJPTF8xWCI6CiAgICAgICAgcmVwZXRpdGlvbl9jb3VudCA9"
    "IFJFUEVUSVRJT05fQ09OVFJPTF9DT1VOVAogICAgZWxpZiBjb25kaXRpb25faWQgPT0gIlRS"
    "RUFUTUVOVF8yNFgiOgogICAgICAgIHJlcGV0aXRpb25fY291bnQgPSBSRVBFVElUSU9OX1RS"
    "RUFUTUVOVF9DT1VOVAogICAgZWxzZToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVy"
    "ZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJ1bnN1cHBvcnRlZCByZXBldGl0aW9uLWRpZmZlcmVudGlhbCBjb25kaXRpb24iLAogICAg"
    "ICAgICkKICAgIHJldHVybiBbCiAgICAgICAgeyJyb2xlIjogInN5c3RlbSIsICJjb250ZW50"
    "IjogU1lTVEVNX1BST01QVH0sCiAgICAgICAgeyJyb2xlIjogInVzZXIiLCAiY29udGVudCI6"
    "IHJlcGV0aXRpb25fY29udGV4dChyZXBldGl0aW9uX2NvdW50KX0sCiAgICAgICAgeyJyb2xl"
    "IjogImFzc2lzdGFudCIsICJjb250ZW50IjogU1lOVEhFVElDX0FTU0lTVEFOVF9BQ0t9LAog"
    "ICAgICAgIHsicm9sZSI6ICJ1c2VyIiwgImNvbnRlbnQiOiBFWFBFQ1RFRF9PQkpFQ1RfQ0FO"
    "T05JQ0FMfSwKICAgIF0KCgpkZWYgcmVwZXRpdGlvbl9yZXF1ZXN0X3BheWxvYWQoCiAgICBj"
    "b25kaXRpb25faWQ6IHN0ciwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHJldHVybiB7"
    "CiAgICAgICAgIm1vZGVsIjogU0VSVkVEX01PREVMX05BTUUsCiAgICAgICAgIm1lc3NhZ2Vz"
    "IjogcmVwZXRpdGlvbl9yZXF1ZXN0X21lc3NhZ2VzKGNvbmRpdGlvbl9pZCksCiAgICAgICAg"
    "InRlbXBlcmF0dXJlIjogMCwKICAgICAgICAidG9wX3AiOiAxLAogICAgICAgICJyZXBldGl0"
    "aW9uX3BlbmFsdHkiOiAxLjEsCiAgICAgICAgInNlZWQiOiA3LAogICAgICAgICJtYXhfdG9r"
    "ZW5zIjogMzIsCiAgICAgICAgInN0cmVhbSI6IEZhbHNlLAogICAgfQoKCmRlZiByZXBldGl0"
    "aW9uX3Rva2VuaXplX3BheWxvYWQoCiAgICBjb25kaXRpb25faWQ6IHN0ciwKKSAtPiBkaWN0"
    "W3N0ciwgb2JqZWN0XToKICAgIHJldHVybiB7CiAgICAgICAgIm1vZGVsIjogU0VSVkVEX01P"
    "REVMX05BTUUsCiAgICAgICAgIm1lc3NhZ2VzIjogcmVwZXRpdGlvbl9yZXF1ZXN0X21lc3Nh"
    "Z2VzKGNvbmRpdGlvbl9pZCksCiAgICAgICAgImFkZF9nZW5lcmF0aW9uX3Byb21wdCI6IFRy"
    "dWUsCiAgICAgICAgImNvbnRpbnVlX2ZpbmFsX21lc3NhZ2UiOiBGYWxzZSwKICAgICAgICAi"
    "YWRkX3NwZWNpYWxfdG9rZW5zIjogRmFsc2UsCiAgICAgICAgInJldHVybl90b2tlbl9zdHJz"
    "IjogRmFsc2UsCiAgICB9CgoKZGVmIHJlcGV0aXRpb25fdG9rZW5faWRlbnRpdHkoCiAgICB3"
    "b3JrZXI6IFdvcmtlciwKICAgIGNvbmRpdGlvbl9pZDogc3RyLAopIC0+IGRpY3Rbc3RyLCBv"
    "YmplY3RdOgogICAgcmVzcG9uc2UgPSBwb3N0X2pzb24oCiAgICAgICAgZiJodHRwOi8vMTI3"
    "LjAuMC4xOnt3b3JrZXIucG9ydH0vdG9rZW5pemUiLAogICAgICAgIHJlcGV0aXRpb25fdG9r"
    "ZW5pemVfcGF5bG9hZChjb25kaXRpb25faWQpLAogICAgKQogICAgcmF3X3Rva2VucyA9IHJl"
    "c3BvbnNlLmdldCgidG9rZW5zIikKICAgIGNvdW50ID0gcmVzcG9uc2UuZ2V0KCJjb3VudCIp"
    "CiAgICBpZiAoCiAgICAgICAgbm90IGlzaW5zdGFuY2UocmF3X3Rva2VucywgbGlzdCkKICAg"
    "ICAgICBvciBpc2luc3RhbmNlKGNvdW50LCBib29sKQogICAgICAgIG9yIG5vdCBpc2luc3Rh"
    "bmNlKGNvdW50LCBpbnQpCiAgICAgICAgb3IgY291bnQgIT0gbGVuKHJhd190b2tlbnMpCiAg"
    "ICApOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFS"
    "TkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgInJlcGV0aXRpb24gdG9rZW5p"
    "emF0aW9uIHJlc3BvbnNlIHNoYXBlIGlzIGludmFsaWQiLAogICAgICAgICkKICAgIHRva2Vu"
    "czogbGlzdFtpbnRdID0gW10KICAgIGZvciByYXdfdG9rZW4gaW4gcmF3X3Rva2VuczoKICAg"
    "ICAgICBpZiBpc2luc3RhbmNlKHJhd190b2tlbiwgYm9vbCkgb3Igbm90IGlzaW5zdGFuY2Uo"
    "cmF3X3Rva2VuLCBpbnQpOgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAgICAgInJlcGV0aXRpb24gdG9rZW5pemF0aW9uIHJldHVybmVkIGEgbm9uLWludGVnZXIg"
    "dG9rZW4gaWQiLAogICAgICAgICAgICApCiAgICAgICAgaWYgcmF3X3Rva2VuIDwgMDoKICAg"
    "ICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFS"
    "TkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJyZXBldGl0aW9uIHRv"
    "a2VuaXphdGlvbiByZXR1cm5lZCBhIG5lZ2F0aXZlIHRva2VuIGlkIiwKICAgICAgICAgICAg"
    "KQogICAgICAgIHRva2Vucy5hcHBlbmQocmF3X3Rva2VuKQogICAgcmV0dXJuIHsKICAgICAg"
    "ICAidG9rZW5fY291bnQiOiBsZW4odG9rZW5zKSwKICAgICAgICAidG9rZW5fc2hhMjU2Ijog"
    "c2hhMjU2X2J5dGVzKGNhbm9uaWNhbF9qc29uKHRva2VucykuZW5jb2RlKCJ1dGYtOCIpKSwK"
    "ICAgIH0KCgpkZWYgaW5pdGlhbGl6ZV9yZXBldGl0aW9uX2pvdXJuYWwoKSAtPiBOb25lOgog"
    "ICAgaWYgUFJFX1JFUVVFU1RfVE9LRU5fSURFTlRJVFlfSk9VUk5BTC5leGlzdHMoKToKICAg"
    "ICAgICByYWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuLWlkZW50aXR5IGpv"
    "dXJuYWwgYWxyZWFkeSBleGlzdHMiKQogICAgd3JpdGVfanNvbigKICAgICAgICBQUkVfUkVR"
    "VUVTVF9UT0tFTl9JREVOVElUWV9KT1VSTkFMLAogICAgICAgIHsKICAgICAgICAgICAgInNj"
    "aGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgImpvdXJuYWxfaWQiOiAoCiAg"
    "ICAgICAgICAgICAgICAiYXVyYWdhdGV3YXktcDQtcDUtY2FjaGUtY29udGV4dC1yZXBldGl0"
    "aW9uLWRpZmZlcmVudGlhbC0iCiAgICAgICAgICAgICAgICAicHJlLXJlcXVlc3QtdG9rZW4t"
    "aWRlbnRpdHktdjEiCiAgICAgICAgICAgICksCiAgICAgICAgICAgICJlbnRyaWVzIjogW10s"
    "CiAgICAgICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgICAg"
    "ICJyYXdfbW9kZWxfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgfSwKICAgICkK"
    "CgpkZWYgcGVyc2lzdF9yZXBldGl0aW9uX3ByZV9yZXF1ZXN0X2lkZW50aXR5KAogICAgcmVx"
    "dWVzdF9vcmRpbmFsOiBpbnQsCiAgICBjb25kaXRpb25faWQ6IHN0ciwKICAgIHJlcGV0aXRp"
    "b25fY291bnQ6IGludCwKICAgIHRva2VuX2lkZW50aXR5OiBkaWN0W3N0ciwgb2JqZWN0XSwK"
    "ICAgIHBheWxvYWRfc2hhMjU2OiBzdHIsCikgLT4gTm9uZToKICAgIGpvdXJuYWwgPSBfcmVh"
    "ZF9wcmVfcmVxdWVzdF90b2tlbl9pZGVudGl0eV9qb3VybmFsKCkKICAgIGVudHJpZXMgPSBq"
    "b3VybmFsLmdldCgiZW50cmllcyIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShlbnRyaWVzLCBs"
    "aXN0KToKICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuLWlk"
    "ZW50aXR5IGpvdXJuYWwgZW50cmllcyBhcmUgaW52YWxpZCIpCiAgICBpZiByZXF1ZXN0X29y"
    "ZGluYWwgIT0gbGVuKGVudHJpZXMpICsgMToKICAgICAgICByYWlzZSBSdW50aW1lRXJyb3Io"
    "InByZS1yZXF1ZXN0IHRva2VuLWlkZW50aXR5IHJlcXVlc3Qgb3JkaW5hbCBkcmlmdGVkIikK"
    "ICAgIHRva2VuX2NvdW50ID0gdG9rZW5faWRlbnRpdHkuZ2V0KCJ0b2tlbl9jb3VudCIpCiAg"
    "ICB0b2tlbl9zaGEyNTYgPSB0b2tlbl9pZGVudGl0eS5nZXQoInRva2VuX3NoYTI1NiIpCiAg"
    "ICBpZiAoCiAgICAgICAgaXNpbnN0YW5jZSh0b2tlbl9jb3VudCwgYm9vbCkKICAgICAgICBv"
    "ciBub3QgaXNpbnN0YW5jZSh0b2tlbl9jb3VudCwgaW50KQogICAgICAgIG9yIG5vdCBpc2lu"
    "c3RhbmNlKHRva2VuX3NoYTI1Niwgc3RyKQogICAgICAgIG9yIHJlLmZ1bGxtYXRjaChyIlsw"
    "LTlhLWZdezY0fSIsIHRva2VuX3NoYTI1NikgaXMgTm9uZQogICAgICAgIG9yIHJlLmZ1bGxt"
    "YXRjaChyIlswLTlhLWZdezY0fSIsIHBheWxvYWRfc2hhMjU2KSBpcyBOb25lCiAgICApOgog"
    "ICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicHJlLXJlcXVlc3QgdG9rZW4gaWRlbnRpdHkg"
    "aXMgaW52YWxpZCIpCiAgICB3cml0ZV9qc29uKAogICAgICAgIFBSRV9SRVFVRVNUX1RPS0VO"
    "X0lERU5USVRZX0pPVVJOQUwsCiAgICAgICAgewogICAgICAgICAgICAqKmpvdXJuYWwsCiAg"
    "ICAgICAgICAgICJlbnRyaWVzIjogWwogICAgICAgICAgICAgICAgKmVudHJpZXMsCiAgICAg"
    "ICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAgICAgInJlcXVlc3Rfb3JkaW5hbCI6IHJl"
    "cXVlc3Rfb3JkaW5hbCwKICAgICAgICAgICAgICAgICAgICAiY29uZGl0aW9uX2lkIjogY29u"
    "ZGl0aW9uX2lkLAogICAgICAgICAgICAgICAgICAgICJwcmVmaXhfdmFyaWFudCI6ICJBIiwK"
    "ICAgICAgICAgICAgICAgICAgICAicmVwZXRpdGlvbl9jb3VudCI6IHJlcGV0aXRpb25fY291"
    "bnQsCiAgICAgICAgICAgICAgICAgICAgInRva2VuX2NvdW50IjogdG9rZW5fY291bnQsCiAg"
    "ICAgICAgICAgICAgICAgICAgInRva2VuX3NoYTI1NiI6IHRva2VuX3NoYTI1NiwKICAgICAg"
    "ICAgICAgICAgICAgICAicGF5bG9hZF9zaGEyNTYiOiBwYXlsb2FkX3NoYTI1NiwKICAgICAg"
    "ICAgICAgICAgICAgICAicGVyc2lzdGVkX2JlZm9yZV9tb2RlbF9yZXF1ZXN0IjogVHJ1ZSwK"
    "ICAgICAgICAgICAgICAgIH0sCiAgICAgICAgICAgIF0sCiAgICAgICAgfSwKICAgICkKCgpk"
    "ZWYgdmFsaWRhdGVfemVyb19jYWNoZV9iYXNlbGluZSgKICAgIHdvcmtlcjogV29ya2VyLAop"
    "IC0+IGRpY3Rbc3RyLCBmbG9hdF06CiAgICBiYXNlbGluZSA9IHdvcmtlci5tZXRyaWNfc25h"
    "cHNob3QoKQogICAgdmFsdWVzID0gewogICAgICAgICJwcmVmaXhfY2FjaGVfcXVlcmllcyI6"
    "IGJhc2VsaW5lLnByZWZpeF9jYWNoZV9xdWVyaWVzLAogICAgICAgICJwcmVmaXhfY2FjaGVf"
    "aGl0cyI6IGJhc2VsaW5lLnByZWZpeF9jYWNoZV9oaXRzLAogICAgICAgICJsb2NhbF9jb21w"
    "dXRlIjogYmFzZWxpbmUubG9jYWxfY29tcHV0ZSwKICAgICAgICAibG9jYWxfY2FjaGVfaGl0"
    "IjogYmFzZWxpbmUubG9jYWxfY2FjaGVfaGl0LAogICAgICAgICJleHRlcm5hbF9rdl90cmFu"
    "c2ZlciI6IGJhc2VsaW5lLmV4dGVybmFsX2t2X3RyYW5zZmVyLAogICAgICAgICJjYWNoZWRf"
    "cHJvbXB0X3Rva2VucyI6IGJhc2VsaW5lLmNhY2hlZF9wcm9tcHRfdG9rZW5zLAogICAgICAg"
    "ICJuZXdseV9jb21wdXRlZF9wcmVmaWxsX3Rva2VucyI6IGJhc2VsaW5lLm5ld2x5X2NvbXB1"
    "dGVkX3ByZWZpbGxfdG9rZW5zLAogICAgfQogICAgaWYgYW55KGFicyhmbG9hdCh2YWx1ZSkp"
    "ID4gMWUtOSBmb3IgdmFsdWUgaW4gdmFsdWVzLnZhbHVlcygpKToKICAgICAgICByYWlzZSBE"
    "aWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlA1X1NUQVJUSU5HX1NUQVRFX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAiZnJlc2ggb2JzZXJ2YXRpb24gd29ya2VyIGRpZCBub3QgYmVn"
    "aW4gYXQgYSB6ZXJvIHJlcXVlc3QvY2FjaGUgbWV0cmljIGJhc2VsaW5lIiwKICAgICAgICAp"
    "CiAgICByZXR1cm4gdmFsdWVzCgoKZGVmIHJlcGV0aXRpb25fZWRnZV9jbGFzcyhjaGFyYWN0"
    "ZXI6IHN0ciB8IE5vbmUpIC0+IHN0cjoKICAgIGlmIGNoYXJhY3RlciBpcyBOb25lOgogICAg"
    "ICAgIHJldHVybiAiTk9ORSIKICAgIGlmIGNoYXJhY3RlciBpbiAie1siOgogICAgICAgIHJl"
    "dHVybiAiSlNPTl9PUEVOIgogICAgaWYgY2hhcmFjdGVyIGluICJdfSI6CiAgICAgICAgcmV0"
    "dXJuICJKU09OX0NMT1NFIgogICAgaWYgY2hhcmFjdGVyID09ICJgIjoKICAgICAgICByZXR1"
    "cm4gIkJBQ0tUSUNLIgogICAgaWYgY2hhcmFjdGVyLmlzc3BhY2UoKToKICAgICAgICByZXR1"
    "cm4gIldISVRFU1BBQ0UiCiAgICBpZiBjaGFyYWN0ZXIuaXNhbHBoYSgpOgogICAgICAgIHJl"
    "dHVybiAiQUxQSEEiCiAgICBpZiBjaGFyYWN0ZXIuaXNkaWdpdCgpOgogICAgICAgIHJldHVy"
    "biAiRElHSVQiCiAgICByZXR1cm4gIk9USEVSIgoKCmRlZiBydW5fcmVwZXRpdGlvbl9vYnNl"
    "cnZhdGlvbigKICAgIHdvcmtlcjogV29ya2VyLAogICAgY29uZGl0aW9uX2lkOiBzdHIsCiAg"
    "ICBzZXF1ZW5jZV9pbmRleDogaW50LAogICAgY291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAop"
    "IC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgcmVwZXRpdGlvbl9jb3VudCA9ICgKICAgICAg"
    "ICBSRVBFVElUSU9OX0NPTlRST0xfQ09VTlQKICAgICAgICBpZiBjb25kaXRpb25faWQgPT0g"
    "IkNPTlRST0xfMVgiCiAgICAgICAgZWxzZSBSRVBFVElUSU9OX1RSRUFUTUVOVF9DT1VOVAog"
    "ICAgICAgIGlmIGNvbmRpdGlvbl9pZCA9PSAiVFJFQVRNRU5UXzI0WCIKICAgICAgICBlbHNl"
    "IE5vbmUKICAgICkKICAgIGlmIHJlcGV0aXRpb25fY291bnQgaXMgTm9uZToKICAgICAgICBy"
    "YWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNf"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICJ1bnN1cHBvcnRlZCByZXBldGl0aW9uLWRpZmZlcmVu"
    "dGlhbCBjb25kaXRpb24iLAogICAgICAgICkKCiAgICB0b2tlbl9pZGVudGl0eSA9IHJlcGV0"
    "aXRpb25fdG9rZW5faWRlbnRpdHkod29ya2VyLCBjb25kaXRpb25faWQpCiAgICBwYXlsb2Fk"
    "ID0gcmVwZXRpdGlvbl9yZXF1ZXN0X3BheWxvYWQoY29uZGl0aW9uX2lkKQogICAgcGF5bG9h"
    "ZF9zaGEyNTYgPSBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihwYXlsb2FkKSkKCiAgICBp"
    "ZiBjb25kaXRpb25faWQgPT0gIlRSRUFUTUVOVF8yNFgiOgogICAgICAgIGlmIHRva2VuX2lk"
    "ZW50aXR5LmdldCgidG9rZW5fY291bnQiKSAhPSBSRVBFVElUSU9OX1RSRUFUTUVOVF9UT0tF"
    "Tl9DT1VOVDoKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICIy"
    "NHggdHJlYXRtZW50IHRva2VuIGNvdW50IGRyaWZ0ZWQgYmVmb3JlIG1vZGVsIHJlcXVlc3Qi"
    "LAogICAgICAgICAgICApCiAgICAgICAgaWYgdG9rZW5faWRlbnRpdHkuZ2V0KCJ0b2tlbl9z"
    "aGEyNTYiKSAhPSBSRVBFVElUSU9OX1RSRUFUTUVOVF9UT0tFTl9TSEEyNTY6CiAgICAgICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1Nf"
    "U0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICAgICAiMjR4IHRyZWF0bWVudCB0b2tl"
    "biBpZGVudGl0eSBkcmlmdGVkIGJlZm9yZSBtb2RlbCByZXF1ZXN0IiwKICAgICAgICAgICAg"
    "KQogICAgICAgIGlmIHBheWxvYWRfc2hhMjU2ICE9IFJFUEVUSVRJT05fVFJFQVRNRU5UX1BB"
    "WUxPQURfU0hBMjU2OgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAg"
    "ICAgIjI0eCB0cmVhdG1lbnQgcGF5bG9hZCBpZGVudGl0eSBkcmlmdGVkIGJlZm9yZSBtb2Rl"
    "bCByZXF1ZXN0IiwKICAgICAgICAgICAgKQoKICAgIHJlcXVlc3Rfb3JkaW5hbCA9IGNvdW50"
    "ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdICsgMQogICAgaWYgcmVxdWVzdF9vcmRpbmFsICE9IHNl"
    "cXVlbmNlX2luZGV4OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgInBy"
    "ZS1yZXF1ZXN0IG9yZGluYWwgZG9lcyBub3QgbWF0Y2ggdGhlIGZyb3plbiBjaHJvbm9sb2d5"
    "IiwKICAgICAgICApCgogICAgcGVyc2lzdF9yZXBldGl0aW9uX3ByZV9yZXF1ZXN0X2lkZW50"
    "aXR5KAogICAgICAgIHJlcXVlc3Rfb3JkaW5hbCwKICAgICAgICBjb25kaXRpb25faWQsCiAg"
    "ICAgICAgcmVwZXRpdGlvbl9jb3VudCwKICAgICAgICB0b2tlbl9pZGVudGl0eSwKICAgICAg"
    "ICBwYXlsb2FkX3NoYTI1NiwKICAgICkKCiAgICBiYXNlbGluZSA9IHZhbGlkYXRlX3plcm9f"
    "Y2FjaGVfYmFzZWxpbmUod29ya2VyKQogICAgYmVmb3JlID0gd29ya2VyLm1ldHJpY19zbmFw"
    "c2hvdCgpCgogICAgY29uc3VtZV9hY3Rpb25zKGNvdW50ZXJzLCAibW9kZWxfcmVxdWVzdHMi"
    "KQogICAgZW5jb2RlZCA9IGNhbm9uaWNhbF9qc29uKHBheWxvYWQpLmVuY29kZSgidXRmLTgi"
    "KQogICAgcmVxdWVzdCA9IHVybGxpYi5yZXF1ZXN0LlJlcXVlc3QoCiAgICAgICAgYm91bmRl"
    "ZF9sb29wYmFjayhmImh0dHA6Ly8xMjcuMC4wLjE6e3dvcmtlci5wb3J0fS92MS9jaGF0L2Nv"
    "bXBsZXRpb25zIiksCiAgICAgICAgZGF0YT1lbmNvZGVkLAogICAgICAgIGhlYWRlcnM9eyJD"
    "b250ZW50LVR5cGUiOiAiYXBwbGljYXRpb24vanNvbiJ9LAogICAgICAgIG1ldGhvZD0iUE9T"
    "VCIsCiAgICApCiAgICB0cnk6CiAgICAgICAgd2l0aCB1cmxsaWIucmVxdWVzdC51cmxvcGVu"
    "KAogICAgICAgICAgICByZXF1ZXN0LAogICAgICAgICAgICB0aW1lb3V0PUhUVFBfVElNRU9V"
    "VF9TRUNPTkRTLAogICAgICAgICkgYXMgcmVzcG9uc2U6CiAgICAgICAgICAgIHN0YXR1c19j"
    "b2RlID0gcmVzcG9uc2Uuc3RhdHVzCiAgICAgICAgICAgIHJlc3BvbnNlX3BheWxvYWQgPSBy"
    "ZXNwb25zZS5yZWFkKCkKICAgIGV4Y2VwdCB1cmxsaWIuZXJyb3IuSFRUUEVycm9yIGFzIGVy"
    "cm9yOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVR"
    "VUVTVF9FWEVDVVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgIGYicmVwZXRpdGlvbiByZXF1"
    "ZXN0IEhUVFAgZmFpbHVyZToge2Vycm9yLmNvZGV9IiwKICAgICAgICApIGZyb20gZXJyb3IK"
    "ICAgIGV4Y2VwdCAodXJsbGliLmVycm9yLlVSTEVycm9yLCBUaW1lb3V0RXJyb3IpIGFzIGVy"
    "cm9yOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVR"
    "VUVTVF9FWEVDVVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJyZXBldGl0aW9uIHJlcXVl"
    "c3QgdHJhbnNwb3J0IGZhaWxlZCIsCiAgICAgICAgKSBmcm9tIGVycm9yCgogICAgaWYgc3Rh"
    "dHVzX2NvZGUgIT0gMjAwOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAg"
    "ICAgICAgICAiUkVRVUVTVF9FWEVDVVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJyZXBl"
    "dGl0aW9uIHJlcXVlc3QgcmV0dXJuZWQgYW4gdW5leHBlY3RlZCBIVFRQIHN0YXR1cyIsCiAg"
    "ICAgICAgKQoKICAgIHRyeToKICAgICAgICBlbnZlbG9wZSA9IGpzb24ubG9hZHMocmVzcG9u"
    "c2VfcGF5bG9hZCkKICAgIGV4Y2VwdCAoanNvbi5KU09ORGVjb2RlRXJyb3IsIFVuaWNvZGVE"
    "ZWNvZGVFcnJvcikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJy"
    "ZXBldGl0aW9uIHJlc3BvbnNlIGVudmVsb3BlIGlzIG5vdCB2YWxpZCBKU09OIiwKICAgICAg"
    "ICApIGZyb20gZXJyb3IKICAgIGlmIG5vdCBpc2luc3RhbmNlKGVudmVsb3BlLCBkaWN0KToK"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9D"
    "T05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInJlcGV0aXRpb24gcmVzcG9uc2UgZW52"
    "ZWxvcGUgcm9vdCBpcyBpbnZhbGlkIiwKICAgICAgICApCgogICAgdXNhZ2UgPSBlbnZlbG9w"
    "ZS5nZXQoInVzYWdlIikKICAgIGNob2ljZXMgPSBlbnZlbG9wZS5nZXQoImNob2ljZXMiKQog"
    "ICAgaWYgbm90IGlzaW5zdGFuY2UodXNhZ2UsIGRpY3QpOgogICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAog"
    "ICAgICAgICAgICAicmVwZXRpdGlvbiByZXNwb25zZSB1c2FnZSBpcyBtaXNzaW5nIiwKICAg"
    "ICAgICApCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2VzLCBsaXN0KSBvciBsZW4oY2hv"
    "aWNlcykgIT0gMToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAg"
    "ICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInJlcGV0aXRpb24g"
    "cmVzcG9uc2UgY2hvaWNlcyBhcmUgaW52YWxpZCIsCiAgICAgICAgKQogICAgY2hvaWNlID0g"
    "Y2hvaWNlc1swXQogICAgaWYgbm90IGlzaW5zdGFuY2UoY2hvaWNlLCBkaWN0KToKICAgICAg"
    "ICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFD"
    "VF9GQUlMVVJFIiwKICAgICAgICAgICAgInJlcGV0aXRpb24gcmVzcG9uc2UgY2hvaWNlIGlz"
    "IGludmFsaWQiLAogICAgICAgICkKICAgIG1lc3NhZ2UgPSBjaG9pY2UuZ2V0KCJtZXNzYWdl"
    "IikKICAgIGlmIG5vdCBpc2luc3RhbmNlKG1lc3NhZ2UsIGRpY3QpOgogICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAicmVwZXRpdGlvbiByZXNwb25zZSBtZXNzYWdlIGlzIGludmFs"
    "aWQiLAogICAgICAgICkKICAgIGNvbnRlbnQgPSBtZXNzYWdlLmdldCgiY29udGVudCIpCiAg"
    "ICBpZiBub3QgaXNpbnN0YW5jZShjb250ZW50LCBzdHIpOgogICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAog"
    "ICAgICAgICAgICAicmVwZXRpdGlvbiByZXNwb25zZSBjb250ZW50IGlzIG5vdCBhIHN0cmlu"
    "ZyIsCiAgICAgICAgKQoKICAgIGNvbXBsZXRpb25fdG9rZW5zID0gdXNhZ2UuZ2V0KCJjb21w"
    "bGV0aW9uX3Rva2VucyIpCiAgICBwcm9tcHRfdG9rZW5zID0gdXNhZ2UuZ2V0KCJwcm9tcHRf"
    "dG9rZW5zIikKICAgIGlmICgKICAgICAgICBpc2luc3RhbmNlKGNvbXBsZXRpb25fdG9rZW5z"
    "LCBib29sKQogICAgICAgIG9yIG5vdCBpc2luc3RhbmNlKGNvbXBsZXRpb25fdG9rZW5zLCBp"
    "bnQpCiAgICAgICAgb3Igbm90IDEgPD0gY29tcGxldGlvbl90b2tlbnMgPD0gMzIKICAgICk6"
    "CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRf"
    "Q09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJyZXBldGl0aW9uIGNvbXBsZXRpb24t"
    "dG9rZW4gYnVkZ2V0IGRyaWZ0ZWQiLAogICAgICAgICkKICAgIGlmIGlzaW5zdGFuY2UocHJv"
    "bXB0X3Rva2VucywgYm9vbCkgb3Igbm90IGlzaW5zdGFuY2UocHJvbXB0X3Rva2VucywgaW50"
    "KSBvciBwcm9tcHRfdG9rZW5zIDw9IDA6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1"
    "cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJyZXBldGl0aW9uIHByb21wdC10b2tlbiBjb3VudCBpcyBpbnZhbGlkIiwKICAgICAgICAp"
    "CgogICAgYWZ0ZXIgPSB3b3JrZXIubWV0cmljX3NuYXBzaG90KCkKICAgIGRlbHRhID0gbWV0"
    "cmljX2RlbHRhKGJlZm9yZSwgYWZ0ZXIpCgogICAgc3RyaXBwZWQgPSBjb250ZW50LnN0cmlw"
    "KCkKICAgIGZpcnN0ID0gc3RyaXBwZWRbMF0gaWYgc3RyaXBwZWQgZWxzZSBOb25lCiAgICBs"
    "YXN0ID0gc3RyaXBwZWRbLTFdIGlmIHN0cmlwcGVkIGVsc2UgTm9uZQogICAgdmFsaWRfanNv"
    "biA9IEZhbHNlCiAgICBleGFjdF9vYmplY3QgPSBGYWxzZQogICAganNvbl9lcnJvcl9saW5l"
    "OiBpbnQgfCBOb25lID0gTm9uZQogICAganNvbl9lcnJvcl9jb2x1bW46IGludCB8IE5vbmUg"
    "PSBOb25lCiAgICBqc29uX2Vycm9yX3Bvc2l0aW9uOiBpbnQgfCBOb25lID0gTm9uZQogICAg"
    "dHJ5OgogICAgICAgIHBhcnNlZCA9IGpzb24ubG9hZHMoY29udGVudCkKICAgICAgICBleHBl"
    "Y3RlZCA9IGpzb24ubG9hZHMoRVhQRUNURURfT0JKRUNUX0NBTk9OSUNBTCkKICAgICAgICB2"
    "YWxpZF9qc29uID0gVHJ1ZQogICAgICAgIGV4YWN0X29iamVjdCA9IHBhcnNlZCA9PSBleHBl"
    "Y3RlZAogICAgZXhjZXB0IGpzb24uSlNPTkRlY29kZUVycm9yIGFzIGVycm9yOgogICAgICAg"
    "IGpzb25fZXJyb3JfbGluZSA9IGVycm9yLmxpbmVubwogICAgICAgIGpzb25fZXJyb3JfY29s"
    "dW1uID0gZXJyb3IuY29sbm8KICAgICAgICBqc29uX2Vycm9yX3Bvc2l0aW9uID0gZXJyb3Iu"
    "cG9zCgogICAgd29ya2VyX3JlcG9ydCA9IHdvcmtlci5yZXBvcnQoKQogICAgcGlkID0gd29y"
    "a2VyX3JlcG9ydC5nZXQoInBpZCIpCiAgICBwcm9jZXNzX3N0YXJ0X3RpY2tzX3ZhbHVlID0g"
    "d29ya2VyX3JlcG9ydC5nZXQoInByb2Nlc3Nfc3RhcnRfdGlja3MiKQogICAgcHJvY2Vzc19p"
    "ZGVudGl0eV9zaGEyNTYgPSBzaGEyNTZfdGV4dCgKICAgICAgICBjYW5vbmljYWxfanNvbigK"
    "ICAgICAgICAgICAgewogICAgICAgICAgICAgICAgInBpZCI6IHBpZCwKICAgICAgICAgICAg"
    "ICAgICJwcm9jZXNzX3N0YXJ0X3RpY2tzIjogcHJvY2Vzc19zdGFydF90aWNrc192YWx1ZSwK"
    "ICAgICAgICAgICAgfQogICAgICAgICkKICAgICkKCiAgICByZXR1cm4gewogICAgICAgICJj"
    "b25kaXRpb25faWQiOiBjb25kaXRpb25faWQsCiAgICAgICAgInNlcXVlbmNlX2luZGV4Ijog"
    "c2VxdWVuY2VfaW5kZXgsCiAgICAgICAgInByZWZpeF92YXJpYW50IjogIkEiLAogICAgICAg"
    "ICJyZXBldGl0aW9uX2NvdW50IjogcmVwZXRpdGlvbl9jb3VudCwKICAgICAgICAid29ya2Vy"
    "X2luc3RhbmNlX2lkIjogd29ya2VyLmluc3RhbmNlX2lkLAogICAgICAgICJ3b3JrZXJfcHJv"
    "Y2Vzc19pZGVudGl0eV9zaGEyNTYiOiBwcm9jZXNzX2lkZW50aXR5X3NoYTI1NiwKICAgICAg"
    "ICAidG9rZW5fY291bnQiOiB0b2tlbl9pZGVudGl0eVsidG9rZW5fY291bnQiXSwKICAgICAg"
    "ICAidG9rZW5fc2hhMjU2IjogdG9rZW5faWRlbnRpdHlbInRva2VuX3NoYTI1NiJdLAogICAg"
    "ICAgICJwYXlsb2FkX3NoYTI1NiI6IHBheWxvYWRfc2hhMjU2LAogICAgICAgICJ6ZXJvX2Nh"
    "Y2hlX2Jhc2VsaW5lIjogVHJ1ZSwKICAgICAgICAiYmFzZWxpbmVfbWV0cmljcyI6IGJhc2Vs"
    "aW5lLAogICAgICAgICJodHRwX3N0YXR1cyI6IHN0YXR1c19jb2RlLAogICAgICAgICJyZXNw"
    "b25zZV9zaGEyNTYiOiBzaGEyNTZfdGV4dChjb250ZW50KSwKICAgICAgICAicmVzcG9uc2Vf"
    "bGVuZ3RoIjogbGVuKGNvbnRlbnQpLAogICAgICAgICJmaW5pc2hfcmVhc29uIjogY2hvaWNl"
    "LmdldCgiZmluaXNoX3JlYXNvbiIpLAogICAgICAgICJwcm9tcHRfdG9rZW5zIjogcHJvbXB0"
    "X3Rva2VucywKICAgICAgICAiY29tcGxldGlvbl90b2tlbnMiOiBjb21wbGV0aW9uX3Rva2Vu"
    "cywKICAgICAgICAidmFsaWRfanNvbiI6IHZhbGlkX2pzb24sCiAgICAgICAgImV4YWN0X29i"
    "amVjdCI6IGV4YWN0X29iamVjdCwKICAgICAgICAianNvbl9lcnJvcl9saW5lIjoganNvbl9l"
    "cnJvcl9saW5lLAogICAgICAgICJqc29uX2Vycm9yX2NvbHVtbiI6IGpzb25fZXJyb3JfY29s"
    "dW1uLAogICAgICAgICJqc29uX2Vycm9yX3Bvc2l0aW9uIjoganNvbl9lcnJvcl9wb3NpdGlv"
    "biwKICAgICAgICAiZmlyc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiOiByZXBldGl0aW9uX2Vk"
    "Z2VfY2xhc3MoZmlyc3QpLAogICAgICAgICJsYXN0X25vbl93aGl0ZXNwYWNlX2NsYXNzIjog"
    "cmVwZXRpdGlvbl9lZGdlX2NsYXNzKGxhc3QpLAogICAgICAgICJtYXJrZG93bl9mZW5jZV9k"
    "ZXRlY3RlZCI6ICJgYGAiIGluIGNvbnRlbnQsCiAgICAgICAgIm1ldHJpY19kZWx0YSI6IGFz"
    "ZGljdChkZWx0YSksCiAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAg"
    "ICAgICAicmF3X291dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgfQoKCmRlZiByZXBldGl0"
    "aW9uX3B1YmxpY19vYnNlcnZhdGlvbigKICAgIG9ic2VydmF0aW9uOiBkaWN0W3N0ciwgb2Jq"
    "ZWN0XSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHBlcm1pdHRlZCA9ICgKICAgICAg"
    "ICAiY29uZGl0aW9uX2lkIiwKICAgICAgICAic2VxdWVuY2VfaW5kZXgiLAogICAgICAgICJw"
    "cmVmaXhfdmFyaWFudCIsCiAgICAgICAgInJlcGV0aXRpb25fY291bnQiLAogICAgICAgICJ3"
    "b3JrZXJfaW5zdGFuY2VfaWQiLAogICAgICAgICJ3b3JrZXJfcHJvY2Vzc19pZGVudGl0eV9z"
    "aGEyNTYiLAogICAgICAgICJ0b2tlbl9jb3VudCIsCiAgICAgICAgInRva2VuX3NoYTI1NiIs"
    "CiAgICAgICAgInBheWxvYWRfc2hhMjU2IiwKICAgICAgICAiemVyb19jYWNoZV9iYXNlbGlu"
    "ZSIsCiAgICAgICAgImJhc2VsaW5lX21ldHJpY3MiLAogICAgICAgICJodHRwX3N0YXR1cyIs"
    "CiAgICAgICAgInJlc3BvbnNlX3NoYTI1NiIsCiAgICAgICAgInJlc3BvbnNlX2xlbmd0aCIs"
    "CiAgICAgICAgImZpbmlzaF9yZWFzb24iLAogICAgICAgICJwcm9tcHRfdG9rZW5zIiwKICAg"
    "ICAgICAiY29tcGxldGlvbl90b2tlbnMiLAogICAgICAgICJ2YWxpZF9qc29uIiwKICAgICAg"
    "ICAiZXhhY3Rfb2JqZWN0IiwKICAgICAgICAianNvbl9lcnJvcl9saW5lIiwKICAgICAgICAi"
    "anNvbl9lcnJvcl9jb2x1bW4iLAogICAgICAgICJqc29uX2Vycm9yX3Bvc2l0aW9uIiwKICAg"
    "ICAgICAiZmlyc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiLAogICAgICAgICJsYXN0X25vbl93"
    "aGl0ZXNwYWNlX2NsYXNzIiwKICAgICAgICAibWFya2Rvd25fZmVuY2VfZGV0ZWN0ZWQiLAog"
    "ICAgICAgICJtZXRyaWNfZGVsdGEiLAogICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIiwK"
    "ICAgICAgICAicmF3X291dHB1dF9yZXRhaW5lZCIsCiAgICApCiAgICByZXR1cm4ge25hbWU6"
    "IG9ic2VydmF0aW9uLmdldChuYW1lKSBmb3IgbmFtZSBpbiBwZXJtaXR0ZWR9CgoKZGVmIHJ1"
    "bl9mcmVzaF93b3JrZXJfb2JzZXJ2YXRpb24oCiAgICBtb2RlbF9ob21lOiBQYXRoLAogICAg"
    "c25hcHNob3Q6IFBhdGgsCiAgICBjb25kaXRpb25faWQ6IHN0ciwKICAgIHNlcXVlbmNlX2lu"
    "ZGV4OiBpbnQsCiAgICBjb3VudGVyczogZGljdFtzdHIsIGludF0sCiAgICB3b3JrZXJfcmVw"
    "b3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0sCiAgICB0ZWFyZG93bl9yZXBvcnRzOiBs"
    "aXN0W2RpY3Rbc3RyLCBvYmplY3RdXSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHdv"
    "cmtlciA9IFdvcmtlcigKICAgICAgICAib2JzZXJ2YXRpb25fd29ya2VyIiwKICAgICAgICAw"
    "LAogICAgICAgIDgwMDEsCiAgICAgICAgbW9kZWxfaG9tZSwKICAgICAgICBzbmFwc2hvdCwK"
    "ICAgICAgICBnZW5lcmF0aW9uPXNlcXVlbmNlX2luZGV4LAogICAgKQogICAgcHJpbWFyeV9l"
    "cnJvcjogRXhjZXB0aW9uIHwgTm9uZSA9IE5vbmUKICAgIG9ic2VydmF0aW9uOiBkaWN0W3N0"
    "ciwgb2JqZWN0XSB8IE5vbmUgPSBOb25lCiAgICByZXBvcnQ6IGRpY3Rbc3RyLCBvYmplY3Rd"
    "IHwgTm9uZSA9IE5vbmUKICAgIHRyeToKICAgICAgICB3b3JrZXIuc3RhcnQoY291bnRlcnMp"
    "CiAgICAgICAgd29ya2VyLndhaXRfcmVhZHkoKQogICAgICAgIHdvcmtlci52YWxpZGF0ZV9t"
    "b2RlbCgpCiAgICAgICAgd29ya2VyLndhaXRfYmFja2VuZF9tYXJrZXIoKQogICAgICAgIG5h"
    "dGl2ZV9vcmlnaW5zID0gdmFsaWRhdGVfbmF0aXZlX29yaWdpbl9jbG9zdXJlKHdvcmtlcikK"
    "ICAgICAgICByZXBvcnQgPSB7CiAgICAgICAgICAgICoqd29ya2VyLnJlcG9ydCgpLAogICAg"
    "ICAgICAgICAibmF0aXZlX29yaWdpbl9jbG9zdXJlIjogbmF0aXZlX29yaWdpbnMsCiAgICAg"
    "ICAgfQogICAgICAgIHdvcmtlcl9yZXBvcnRzLmFwcGVuZChyZXBvcnQpCiAgICAgICAgb2Jz"
    "ZXJ2YXRpb24gPSBydW5fcmVwZXRpdGlvbl9vYnNlcnZhdGlvbigKICAgICAgICAgICAgd29y"
    "a2VyLAogICAgICAgICAgICBjb25kaXRpb25faWQsCiAgICAgICAgICAgIHNlcXVlbmNlX2lu"
    "ZGV4LAogICAgICAgICAgICBjb3VudGVycywKICAgICAgICApCiAgICBleGNlcHQgRXhjZXB0"
    "aW9uIGFzIGVycm9yOgogICAgICAgIHByaW1hcnlfZXJyb3IgPSBlcnJvcgogICAgZmluYWxs"
    "eToKICAgICAgICB0ZWFyZG93biA9IHNhZmVfd29ya2VyX3RlYXJkb3duKAogICAgICAgICAg"
    "ICB3b3JrZXIsCiAgICAgICAgICAgIGYiUkVQRVRJVElPTl9PQlNFUlZBVElPTl97c2VxdWVu"
    "Y2VfaW5kZXh9X1RFUk1JTkFMIiwKICAgICAgICApCiAgICAgICAgdGVhcmRvd25fcmVwb3J0"
    "cy5hcHBlbmQodGVhcmRvd24pCgogICAgdGVhcmRvd25fc3RhdHVzID0gdGVhcmRvd24uZ2V0"
    "KCJzdGF0dXMiKQogICAgaWYgdGVhcmRvd25fc3RhdHVzIG5vdCBpbiB7IlBBU1NFRCIsICJO"
    "T1RfU1RBUlRFRCJ9OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAgICJmcmVzaCBvYnNlcnZhdGlv"
    "biB3b3JrZXIgdGVhcmRvd24gcHJvb2YgZmFpbGVkIiwKICAgICAgICApCiAgICBpZiBwcmlt"
    "YXJ5X2Vycm9yIGlzIG5vdCBOb25lOgogICAgICAgIHJhaXNlIHByaW1hcnlfZXJyb3IKICAg"
    "IGlmIHRlYXJkb3duX3N0YXR1cyAhPSAiUEFTU0VEIjoKICAgICAgICByYWlzZSBEaWFnbm9z"
    "dGljRmFpbHVyZSgKICAgICAgICAgICAgIlRFQVJET1dOX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAiY29tcGxldGVkIG9ic2VydmF0aW9uIGRpZCBub3QgcHJvdmUgd29ya2VyIHRlYXJkb3du"
    "IiwKICAgICAgICApCiAgICBpZiBvYnNlcnZhdGlvbiBpcyBOb25lIG9yIHJlcG9ydCBpcyBO"
    "b25lOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFS"
    "TkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgImZyZXNoIG9ic2VydmF0aW9u"
    "IGRpZCBub3QgcHJvZHVjZSBjb21wbGV0ZSBldmlkZW5jZSIsCiAgICAgICAgKQogICAgcmV0"
    "dXJuIG9ic2VydmF0aW9uCgoKZGVmIHdyaXRlX3JlcGV0aXRpb25fcmVzdWx0cygKICAgIHJl"
    "c3VsdHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAogICAgc3RhdHVzOiBzdHIsCikgLT4g"
    "Tm9uZToKICAgIHdyaXRlX2pzb24oCiAgICAgICAgT1VUUFVUX1JPT1QgLyAicDRfcDVfcmVw"
    "ZXRpdGlvbl9yZXF1ZXN0X3Jlc3VsdHNfdjEuanNvbiIsCiAgICAgICAgewogICAgICAgICAg"
    "ICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAic3RhdHVzIjogc3Rh"
    "dHVzLAogICAgICAgICAgICAic2NoZWR1bGVkX3JlcXVlc3RfY291bnQiOiBsZW4oUkVQRVRJ"
    "VElPTl9SRVFVRVNUX09SREVSKSwKICAgICAgICAgICAgIm9ic2VydmVkX3JlcXVlc3RfY291"
    "bnQiOiBsZW4ocmVzdWx0cyksCiAgICAgICAgICAgICJyZXF1ZXN0X29yZGVyIjogbGlzdChS"
    "RVBFVElUSU9OX1JFUVVFU1RfT1JERVIpLAogICAgICAgICAgICAicmVzdWx0cyI6IFtyZXBl"
    "dGl0aW9uX3B1YmxpY19vYnNlcnZhdGlvbihpdGVtKSBmb3IgaXRlbSBpbiByZXN1bHRzXSwK"
    "ICAgICAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAgICAg"
    "InJhd19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICB9LAogICAgKQoKCmRlZiBk"
    "ZWNpZGVfcmVwZXRpdGlvbl9kaWZmZXJlbnRpYWwoCiAgICByZXN1bHRzOiBsaXN0W2RpY3Rb"
    "c3RyLCBvYmplY3RdXSwKICAgIHdvcmtlcl9yZXBvcnRzOiBsaXN0W2RpY3Rbc3RyLCBvYmpl"
    "Y3RdXSwKICAgIHRlYXJkb3duX3JlcG9ydHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAog"
    "ICAgY291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgog"
    "ICAgaWYgbGVuKHJlc3VsdHMpICE9IDY6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1"
    "cmUoCiAgICAgICAgICAgICJSRVFVRVNUX1JFQ09OQ0lMSUFUSU9OX0ZBSUxVUkUiLAogICAg"
    "ICAgICAgICAicmVwZXRpdGlvbiBkaWZmZXJlbnRpYWwgcmVzdWx0IGNvdW50IGRyaWZ0ZWQi"
    "LAogICAgICAgICkKICAgIG9ic2VydmVkX29yZGVyID0gdHVwbGUoc3RyKHJvdy5nZXQoImNv"
    "bmRpdGlvbl9pZCIpKSBmb3Igcm93IGluIHJlc3VsdHMpCiAgICBpZiBvYnNlcnZlZF9vcmRl"
    "ciAhPSBSRVBFVElUSU9OX1JFUVVFU1RfT1JERVI6CiAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX1JFQ09OQ0lMSUFUSU9OX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAicmVwZXRpdGlvbiBkaWZmZXJlbnRpYWwgY2hyb25vbG9neSBkcmlm"
    "dGVkIiwKICAgICAgICApCiAgICBvYnNlcnZlZF9pbmRleGVzID0gdHVwbGUocm93LmdldCgi"
    "c2VxdWVuY2VfaW5kZXgiKSBmb3Igcm93IGluIHJlc3VsdHMpCiAgICBpZiBvYnNlcnZlZF9p"
    "bmRleGVzICE9IHR1cGxlKHJhbmdlKDEsIDcpKToKICAgICAgICByYWlzZSBEaWFnbm9zdGlj"
    "RmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJyZXBldGl0aW9uIGRpZmZlcmVudGlhbCBzZXF1ZW5jZSBpbmRleGVz"
    "IGRyaWZ0ZWQiLAogICAgICAgICkKICAgIGlmIGxlbih3b3JrZXJfcmVwb3J0cykgIT0gNiBv"
    "ciBsZW4odGVhcmRvd25fcmVwb3J0cykgIT0gNjoKICAgICAgICByYWlzZSBEaWFnbm9zdGlj"
    "RmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJmcmVzaC13b3JrZXIgZXZpZGVuY2UgY2FyZGluYWxpdHkgZHJpZnRl"
    "ZCIsCiAgICAgICAgKQogICAgaWYgYW55KGl0ZW0uZ2V0KCJzdGF0dXMiKSAhPSAiUEFTU0VE"
    "IiBmb3IgaXRlbSBpbiB0ZWFyZG93bl9yZXBvcnRzKToKICAgICAgICByYWlzZSBEaWFnbm9z"
    "dGljRmFpbHVyZSgKICAgICAgICAgICAgIlRFQVJET1dOX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAib25lIG9yIG1vcmUgb2JzZXJ2YXRpb24gdGVhcmRvd25zIGZhaWxlZCIsCiAgICAgICAg"
    "KQogICAgd29ya2VyX2lkZW50aXRpZXMgPSB7CiAgICAgICAgc3RyKHJvdy5nZXQoIndvcmtl"
    "cl9wcm9jZXNzX2lkZW50aXR5X3NoYTI1NiIpKQogICAgICAgIGZvciByb3cgaW4gcmVzdWx0"
    "cwogICAgICAgIGlmIGlzaW5zdGFuY2Uocm93LmdldCgid29ya2VyX3Byb2Nlc3NfaWRlbnRp"
    "dHlfc2hhMjU2IiksIHN0cikKICAgIH0KICAgIGlmIGxlbih3b3JrZXJfaWRlbnRpdGllcykg"
    "IT0gNjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlA1"
    "X1NUQVJUSU5HX1NUQVRFX0ZBSUxVUkUiLAogICAgICAgICAgICAiZnJlc2ggd29ya2VyIHBy"
    "b2Nlc3MgaWRlbnRpdHkgd2FzIHJldXNlZCBhY3Jvc3Mgb2JzZXJ2YXRpb25zIiwKICAgICAg"
    "ICApCiAgICBpZiBhbnkocm93LmdldCgiemVyb19jYWNoZV9iYXNlbGluZSIpIGlzIG5vdCBU"
    "cnVlIGZvciByb3cgaW4gcmVzdWx0cyk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1"
    "cmUoCiAgICAgICAgICAgICJQNV9TVEFSVElOR19TVEFURV9GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgIm9uZSBvciBtb3JlIG9ic2VydmF0aW9ucyBsYWNrZWQgYSB6ZXJvIGNhY2hlIGJhc2Vs"
    "aW5lIiwKICAgICAgICApCgogICAgZXhwZWN0ZWRfY291bnRlcnMgPSB7CiAgICAgICAgIm1v"
    "ZGVsX3JlcXVlc3RzIjogNiwKICAgICAgICAibW9kZWxfbG9hZHMiOiA2LAogICAgICAgICJ3"
    "b3JrZXJfc3RhcnRzIjogNiwKICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAg"
    "ICJuZXR3b3JrX3JlcXVlc3RzIjogMCwKICAgICAgICAiYmVuY2htYXJrX3RyYWplY3Rvcnlf"
    "cmVxdWVzdHMiOiAwLAogICAgICAgICJleHRlcm5hbF9zcGVuZCI6IDAsCiAgICB9CiAgICBm"
    "b3IgbmFtZSwgZXhwZWN0ZWQgaW4gZXhwZWN0ZWRfY291bnRlcnMuaXRlbXMoKToKICAgICAg"
    "ICBpZiBjb3VudGVycy5nZXQobmFtZSkgIT0gZXhwZWN0ZWQ6CiAgICAgICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJ"
    "QVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICAgICBmIntuYW1lfSBleHBlY3RlZCB7ZXhw"
    "ZWN0ZWR9LCBvYnNlcnZlZCB7Y291bnRlcnMuZ2V0KG5hbWUpfSIsCiAgICAgICAgICAgICkK"
    "CiAgICBjb250cm9sID0gW3JvdyBmb3Igcm93IGluIHJlc3VsdHMgaWYgcm93LmdldCgiY29u"
    "ZGl0aW9uX2lkIikgPT0gIkNPTlRST0xfMVgiXQogICAgdHJlYXRtZW50ID0gW3JvdyBmb3Ig"
    "cm93IGluIHJlc3VsdHMgaWYgcm93LmdldCgiY29uZGl0aW9uX2lkIikgPT0gIlRSRUFUTUVO"
    "VF8yNFgiXQogICAgaWYgbGVuKGNvbnRyb2wpICE9IDMgb3IgbGVuKHRyZWF0bWVudCkgIT0g"
    "MzoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVF"
    "U1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJjb25kaXRpb24gY2Fy"
    "ZGluYWxpdHkgZHJpZnRlZCIsCiAgICAgICAgKQoKICAgIHRyZWF0bWVudF90b2tlbl9wYWly"
    "cyA9IHsocm93LmdldCgidG9rZW5fY291bnQiKSwgcm93LmdldCgidG9rZW5fc2hhMjU2Iikp"
    "IGZvciByb3cgaW4gdHJlYXRtZW50fQogICAgdHJlYXRtZW50X3BheWxvYWRzID0ge3Jvdy5n"
    "ZXQoInBheWxvYWRfc2hhMjU2IikgZm9yIHJvdyBpbiB0cmVhdG1lbnR9CiAgICBpZiB0cmVh"
    "dG1lbnRfdG9rZW5fcGFpcnMgIT0gewogICAgICAgIChSRVBFVElUSU9OX1RSRUFUTUVOVF9U"
    "T0tFTl9DT1VOVCwgUkVQRVRJVElPTl9UUkVBVE1FTlRfVE9LRU5fU0hBMjU2KQogICAgfToK"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1Nf"
    "U0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICIyNHggdHJlYXRtZW50IHRva2VuIGlk"
    "ZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgKQogICAgaWYgdHJlYXRt"
    "ZW50X3BheWxvYWRzICE9IHtSRVBFVElUSU9OX1RSRUFUTUVOVF9QQVlMT0FEX1NIQTI1Nn06"
    "CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNT"
    "X1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAiMjR4IHRyZWF0bWVudCBwYXlsb2Fk"
    "IGlkZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgKQoKICAgIGNvbnRy"
    "b2xfdG9rZW5fcGFpcnMgPSB7KHJvdy5nZXQoInRva2VuX2NvdW50IiksIHJvdy5nZXQoInRv"
    "a2VuX3NoYTI1NiIpKSBmb3Igcm93IGluIGNvbnRyb2x9CiAgICBjb250cm9sX3BheWxvYWRz"
    "ID0ge3Jvdy5nZXQoInBheWxvYWRfc2hhMjU2IikgZm9yIHJvdyBpbiBjb250cm9sfQogICAg"
    "aWYgbGVuKGNvbnRyb2xfdG9rZW5fcGFpcnMpICE9IDEgb3IgbGVuKGNvbnRyb2xfcGF5bG9h"
    "ZHMpICE9IDE6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAg"
    "ICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAiMXggY29udHJvbCBp"
    "ZGVudGl0eSBpcyBub3QgaW50ZXJuYWxseSBzdGFibGUiLAogICAgICAgICkKICAgIGlmIGNv"
    "bnRyb2xfdG9rZW5fcGFpcnMgPT0gdHJlYXRtZW50X3Rva2VuX3BhaXJzOgogICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19G"
    "QUlMVVJFIiwKICAgICAgICAgICAgIjF4IGNvbnRyb2wgdG9rZW4gaWRlbnRpdHkgZG9lcyBu"
    "b3QgZGlmZmVyIGZyb20gMjR4IHRyZWF0bWVudCIsCiAgICAgICAgKQogICAgaWYgY29udHJv"
    "bF9wYXlsb2FkcyA9PSB0cmVhdG1lbnRfcGF5bG9hZHM6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAog"
    "ICAgICAgICAgICAiMXggY29udHJvbCBwYXlsb2FkIGlkZW50aXR5IGRvZXMgbm90IGRpZmZl"
    "ciBmcm9tIDI0eCB0cmVhdG1lbnQiLAogICAgICAgICkKCiAgICBjb250cm9sX2V4YWN0ID0g"
    "c3VtKHJvdy5nZXQoImV4YWN0X29iamVjdCIpIGlzIFRydWUgZm9yIHJvdyBpbiBjb250cm9s"
    "KQogICAgdHJlYXRtZW50X2V4YWN0ID0gc3VtKHJvdy5nZXQoImV4YWN0X29iamVjdCIpIGlz"
    "IFRydWUgZm9yIHJvdyBpbiB0cmVhdG1lbnQpCiAgICBjb250cm9sX3ZhbGlkX2pzb24gPSBz"
    "dW0ocm93LmdldCgidmFsaWRfanNvbiIpIGlzIFRydWUgZm9yIHJvdyBpbiBjb250cm9sKQog"
    "ICAgdHJlYXRtZW50X3ZhbGlkX2pzb24gPSBzdW0ocm93LmdldCgidmFsaWRfanNvbiIpIGlz"
    "IFRydWUgZm9yIHJvdyBpbiB0cmVhdG1lbnQpCgogICAgaWYgY29udHJvbF9leGFjdCBpbiB7"
    "MSwgMn06CiAgICAgICAgc3RhdGUgPSAiQ09OVFJPTF9OT1RfUkVMSUFCTEUiCiAgICBlbGlm"
    "IGNvbnRyb2xfZXhhY3QgPT0gMyBhbmQgdHJlYXRtZW50X2V4YWN0ID09IDA6CiAgICAgICAg"
    "c3RhdGUgPSAiTE9OR19SRVBFQVRFRF9DT05URVhUX05FQ0VTU0lUWV9TVVBQT1JURUQiCiAg"
    "ICBlbGlmIGNvbnRyb2xfZXhhY3QgPT0gMCBhbmQgdHJlYXRtZW50X2V4YWN0ID09IDA6CiAg"
    "ICAgICAgc3RhdGUgPSAiUkVQRVRJVElPTl9OT1RfTkVDRVNTQVJZIgogICAgZWxpZiBjb250"
    "cm9sX2V4YWN0ID09IDMgYW5kIHRyZWF0bWVudF9leGFjdCA9PSAzOgogICAgICAgIHN0YXRl"
    "ID0gIlJFR1JFU1NJT05fTk9UX1JFUFJPRFVDRUQiCiAgICBlbGlmIGNvbnRyb2xfZXhhY3Qg"
    "PT0gMyBhbmQgdHJlYXRtZW50X2V4YWN0IGluIHsxLCAyfToKICAgICAgICBzdGF0ZSA9ICJO"
    "T05fREVURVJNSU5JU1RJQ19PUl9BTUJJR1VPVVMiCiAgICBlbHNlOgogICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlM"
    "VVJFIiwKICAgICAgICAgICAgIm9ic2VydmVkIGJlaGF2aW9yYWwgY29tYmluYXRpb24gaXMg"
    "bm90IGNvdmVyZWQgYnkgdGhlIGZyb3plbiBkZWNpc2lvbiBjb250cmFjdCIsCiAgICAgICAg"
    "KQoKICAgIHJldHVybiB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAg"
    "ICAgICAic3RhdHVzIjogIkRFQ0lERUQiLAogICAgICAgICJkZWNpc2lvbl9zdGF0ZSI6IHN0"
    "YXRlLAogICAgICAgICJ2YXJpYWJsZV91bmRlcl90ZXN0IjogIkNBQ0hFX0NPTlRFWFRfUkVQ"
    "RVRJVElPTl9DT1VOVCIsCiAgICAgICAgImNvbnRyb2xfcmVwZXRpdGlvbl9jb3VudCI6IFJF"
    "UEVUSVRJT05fQ09OVFJPTF9DT1VOVCwKICAgICAgICAidHJlYXRtZW50X3JlcGV0aXRpb25f"
    "Y291bnQiOiBSRVBFVElUSU9OX1RSRUFUTUVOVF9DT1VOVCwKICAgICAgICAiY29udHJvbF9l"
    "eGFjdF9vYmplY3RfY291bnQiOiBjb250cm9sX2V4YWN0LAogICAgICAgICJ0cmVhdG1lbnRf"
    "ZXhhY3Rfb2JqZWN0X2NvdW50IjogdHJlYXRtZW50X2V4YWN0LAogICAgICAgICJjb250cm9s"
    "X3ZhbGlkX2pzb25fY291bnQiOiBjb250cm9sX3ZhbGlkX2pzb24sCiAgICAgICAgInRyZWF0"
    "bWVudF92YWxpZF9qc29uX2NvdW50IjogdHJlYXRtZW50X3ZhbGlkX2pzb24sCiAgICAgICAg"
    "ImZyZXNoX3dvcmtlcl9wcm9jZXNzX3Blcl9vYnNlcnZhdGlvbiI6IFRydWUsCiAgICAgICAg"
    "Indvcmtlcl9pZGVudGl0eV9jYXJkaW5hbGl0eSI6IGxlbih3b3JrZXJfaWRlbnRpdGllcyks"
    "CiAgICAgICAgInRyZWF0bWVudF9oaXN0b3JpY2FsX2lkZW50aXR5X21hdGNoZWQiOiBUcnVl"
    "LAogICAgICAgICJjb250cm9sX2ludHJhX2NvbmRpdGlvbl9pZGVudGl0eV9tYXRjaGVkIjog"
    "VHJ1ZSwKICAgICAgICAidHJlYXRtZW50X2ludHJhX2NvbmRpdGlvbl9pZGVudGl0eV9tYXRj"
    "aGVkIjogVHJ1ZSwKICAgICAgICAiY2FjaGVfdGVsZW1ldHJ5X2lzX2RpYWdub3N0aWNfb25s"
    "eSI6IFRydWUsCiAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAg"
    "ICAicmF3X291dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgfQoKCmRlZiByZXBldGl0aW9u"
    "X2ZhaWx1cmVfcmVjb3JkKAogICAgZXJyb3I6IEV4Y2VwdGlvbiwKICAgIGFjdGl2ZV9mYWls"
    "dXJlX2NvZGU6IHN0ciwKICAgIGZhaWxlZF9zdGFnZTogc3RyLAogICAgY29tcGxldGVkX3Jl"
    "cXVlc3RzOiBpbnQsCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBpZiBpc2luc3RhbmNl"
    "KGVycm9yLCBEaWFnbm9zdGljQW1iaWd1aXR5KToKICAgICAgICBmYWlsdXJlX2NsYXNzID0g"
    "ZXJyb3IuZmFpbHVyZV9jbGFzcwogICAgICAgIGRldGFpbF9jb2RlID0gTm9uZQogICAgICAg"
    "IHNhZmVfbWVzc2FnZSA9IGVycm9yLnNhZmVfbWVzc2FnZQogICAgZWxpZiBpc2luc3RhbmNl"
    "KGVycm9yLCBEaWFnbm9zdGljRmFpbHVyZSk6CiAgICAgICAgZGV0YWlsX2NvZGUgPSBlcnJv"
    "ci5lcnJvcl9jb2RlCiAgICAgICAgZmFpbHVyZV9jbGFzcyA9IGNsYXNzaWZ5X2ZhaWx1cmVf"
    "ZGV0YWlsKGRldGFpbF9jb2RlKQogICAgICAgIHNhZmVfbWVzc2FnZSA9IGVycm9yLnNhZmVf"
    "bWVzc2FnZQogICAgZWxzZToKICAgICAgICBkZXRhaWxfY29kZSA9IGFjdGl2ZV9mYWlsdXJl"
    "X2NvZGUKICAgICAgICBmYWlsdXJlX2NsYXNzID0gY2xhc3NpZnlfZmFpbHVyZV9kZXRhaWwo"
    "ZGV0YWlsX2NvZGUpCiAgICAgICAgc2FmZV9tZXNzYWdlID0gc2FuaXRpemVfZXhjZXJwdChz"
    "dHIoZXJyb3IpKVs6NTEyXSBvciB0eXBlKGVycm9yKS5fX25hbWVfXwogICAgcmV0dXJuIHsK"
    "ICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICJzdGF0dXMiOiAi"
    "RElBR05PU1RJQ19JTlZBTElEIiwKICAgICAgICAiZmFpbGVkX3N0YWdlIjogZmFpbGVkX3N0"
    "YWdlLAogICAgICAgICJjb21wbGV0ZWRfcmVxdWVzdHMiOiBjb21wbGV0ZWRfcmVxdWVzdHMs"
    "CiAgICAgICAgImZhaWx1cmVfY2xhc3MiOiBmYWlsdXJlX2NsYXNzLAogICAgICAgICJkZXRh"
    "aWxfY29kZSI6IGRldGFpbF9jb2RlLAogICAgICAgICJlcnJvcl90eXBlIjogdHlwZShlcnJv"
    "cikuX19uYW1lX18sCiAgICAgICAgInNhZmVfbWVzc2FnZSI6IHNhZmVfbWVzc2FnZSwKICAg"
    "IH0KCgpkZWYgcmVwZXRpdGlvbl9idW5kbGVfb3V0cHV0cygpIC0+IGRpY3Rbc3RyLCBvYmpl"
    "Y3RdOgogICAgcmVxdWlyZWRfYmVmb3JlX21hbmlmZXN0ID0gc2V0KFJFUEVUSVRJT05fT1VU"
    "UFVUX05BTUVTKSAtIHsiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24ifQogICAgb2JzZXJ2ZWRf"
    "YmVmb3JlX21hbmlmZXN0ID0ge3BhdGgubmFtZSBmb3IgcGF0aCBpbiBPVVRQVVRfUk9PVC5p"
    "dGVyZGlyKCkgaWYgcGF0aC5pc19maWxlKCl9CiAgICBtaXNzaW5nID0gcmVxdWlyZWRfYmVm"
    "b3JlX21hbmlmZXN0IC0gb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0CiAgICB1bmV4cGVjdGVk"
    "ID0gb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0IC0gcmVxdWlyZWRfYmVmb3JlX21hbmlmZXN0"
    "CiAgICBpZiBtaXNzaW5nIG9yIHVuZXhwZWN0ZWQ6CiAgICAgICAgcmFpc2UgUnVudGltZUVy"
    "cm9yKAogICAgICAgICAgICAicmVwZXRpdGlvbiBldmlkZW5jZSBvdXRwdXQgY29udHJhY3Qg"
    "ZHJpZnRlZDogIgogICAgICAgICAgICArIGNhbm9uaWNhbF9qc29uKAogICAgICAgICAgICAg"
    "ICAgewogICAgICAgICAgICAgICAgICAgICJtaXNzaW5nIjogc29ydGVkKG1pc3NpbmcpLAog"
    "ICAgICAgICAgICAgICAgICAgICJ1bmV4cGVjdGVkIjogc29ydGVkKHVuZXhwZWN0ZWQpLAog"
    "ICAgICAgICAgICAgICAgfQogICAgICAgICAgICApCiAgICAgICAgKQogICAgZW50cmllcyA9"
    "IFtdCiAgICBmb3IgbmFtZSBpbiBSRVBFVElUSU9OX09VVFBVVF9OQU1FUzoKICAgICAgICBw"
    "YXRoID0gT1VUUFVUX1JPT1QgLyBuYW1lCiAgICAgICAgaWYgbm90IHBhdGguaXNfZmlsZSgp"
    "OgogICAgICAgICAgICBjb250aW51ZQogICAgICAgIGVudHJpZXMuYXBwZW5kKAogICAgICAg"
    "ICAgICB7CiAgICAgICAgICAgICAgICAicGF0aCI6IG5hbWUsCiAgICAgICAgICAgICAgICAi"
    "c2hhMjU2IjogZmlsZV9zaGEyNTYocGF0aCksCiAgICAgICAgICAgICAgICAic2l6ZV9ieXRl"
    "cyI6IHBhdGguc3RhdCgpLnN0X3NpemUsCiAgICAgICAgICAgIH0KICAgICAgICApCiAgICBt"
    "YW5pZmVzdCA9IHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAg"
    "ICJkaWFnbm9zdGljX2lkIjogKCJhdXJhZ2F0ZXdheS1wNC1wNS1jYWNoZS1jb250ZXh0LXJl"
    "cGV0aXRpb24tZGlmZmVyZW50aWFsLXYxIiksCiAgICAgICAgInNvdXJjZV9tYWluX2NvbW1p"
    "dCI6IFNPVVJDRV9NQUlOX0NPTU1JVCwKICAgICAgICAibWVtYmVycyI6IFtpdGVtIGZvciBp"
    "dGVtIGluIGVudHJpZXMgaWYgaXRlbVsicGF0aCJdICE9ICJidW5kbGVfbWFuaWZlc3RfdjEu"
    "anNvbiJdLAogICAgICAgICJzY3JhdGNoX2RpcmVjdG9yaWVzX2luY2x1ZGVkIjogRmFsc2Us"
    "CiAgICAgICAgIndvcmtlcl9sb2dfZGlyZWN0b3J5X2luY2x1ZGVkIjogRmFsc2UsCiAgICAg"
    "ICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAicmF3X291dHB1dF9y"
    "ZXRhaW5lZCI6IEZhbHNlLAogICAgfQogICAgd3JpdGVfanNvbihPVVRQVVRfUk9PVCAvICJi"
    "dW5kbGVfbWFuaWZlc3RfdjEuanNvbiIsIG1hbmlmZXN0KQogICAgd2l0aCB6aXBmaWxlLlpp"
    "cEZpbGUoCiAgICAgICAgRVZJREVOQ0VfWklQLAogICAgICAgICJ3IiwKICAgICAgICBjb21w"
    "cmVzc2lvbj16aXBmaWxlLlpJUF9ERUZMQVRFRCwKICAgICAgICBjb21wcmVzc2xldmVsPTks"
    "CiAgICApIGFzIGFyY2hpdmU6CiAgICAgICAgZm9yIG5hbWUgaW4gUkVQRVRJVElPTl9PVVRQ"
    "VVRfTkFNRVM6CiAgICAgICAgICAgIHBhdGggPSBPVVRQVVRfUk9PVCAvIG5hbWUKICAgICAg"
    "ICAgICAgaWYgcGF0aC5pc19maWxlKCk6CiAgICAgICAgICAgICAgICBhcmNoaXZlLndyaXRl"
    "KHBhdGgsIGFyY25hbWU9bmFtZSkKICAgIGlmIEVWSURFTkNFX1pJUC5zdGF0KCkuc3Rfc2l6"
    "ZSA+IE1BWF9FVklERU5DRV9aSVBfQllURVM6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9y"
    "KCJyZXBldGl0aW9uIGV2aWRlbmNlIFpJUCBleGNlZWRzIHRoZSBieXRlIGJ1ZGdldCIpCiAg"
    "ICByZXR1cm4gewogICAgICAgICJldmlkZW5jZV96aXAiOiBzdHIoRVZJREVOQ0VfWklQKSwK"
    "ICAgICAgICAiZXZpZGVuY2VfemlwX3NoYTI1NiI6IGZpbGVfc2hhMjU2KEVWSURFTkNFX1pJ"
    "UCksCiAgICAgICAgImV2aWRlbmNlX3ppcF9zaXplX2J5dGVzIjogRVZJREVOQ0VfWklQLnN0"
    "YXQoKS5zdF9zaXplLAogICAgfQoKClRPS0VOX01BVENIRURfSU1QTEVNRU5UQVRJT05fQkFT"
    "RV9DT01NSVQ6IEZpbmFsID0gIjIwYTdiZDU4Yzg5MzdmZWZlOWQ0YjFjMmNhOWM3NTA3MDlk"
    "MmRjZWIiClRPS0VOX01BVENIRURfREVTSUdOX1JFQ09SRF9TSEEyNTY6IEZpbmFsID0gKAog"
    "ICAgIjg4OGJmMGEyNWE5NzRiYTJjNjI4OTJiYzk5OWZlMGM5ZjIzZDJjZjg0NWJjZDI1NDJj"
    "NjdlMmM5YmM0Y2NmMDMiCikKVE9LRU5fTUFUQ0hFRF9SRVFVRVNUX09SREVSOiBGaW5hbCA9"
    "ICgKICAgICgiQV9PUklHSU5BTF8yNFhfQU5DSE9SIiksCiAgICAoIkJfTkVVVFJBTF9SRVBF"
    "QVRFRF8yNFgiKSwKICAgICgiQ19ORVVUUkFMX0RJVkVSU0VfMjRfU0VHTUVOVCIpLAogICAg"
    "KCJCX05FVVRSQUxfUkVQRUFURURfMjRYIiksCiAgICAoIkNfTkVVVFJBTF9ESVZFUlNFXzI0"
    "X1NFR01FTlQiKSwKICAgICgiQV9PUklHSU5BTF8yNFhfQU5DSE9SIiksCiAgICAoIkNfTkVV"
    "VFJBTF9ESVZFUlNFXzI0X1NFR01FTlQiKSwKICAgICgiQV9PUklHSU5BTF8yNFhfQU5DSE9S"
    "IiksCiAgICAoIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgiKSwKKQpUT0tFTl9NQVRDSEVEX1BS"
    "T01QVF9UT0tFTl9DT1VOVDogRmluYWwgPSA4OTkKVE9LRU5fTUFUQ0hFRF9BX1RPS0VOX1NI"
    "QTI1NjogRmluYWwgPSAoCiAgICAiNmIzNDQ0OGQwODNlNjg4MjZjYWU4NGMwNjc1ODc2YzE4"
    "ZDdhNzBmYWI1MzI5OTg2MGI3OGYyYTVhMTg5MjJiMCIKKQpUT0tFTl9NQVRDSEVEX0JfVE9L"
    "RU5fU0hBMjU2OiBGaW5hbCA9ICgKICAgICIwMmYyNjc1YTA0OTBkMTZlM2EzOWRlOTYxOWFl"
    "ODY1YThmNzMwMjRiMjZjNmY5MTI2YmY0ZGVhMTk3ZDk5ZjY4IgopClRPS0VOX01BVENIRURf"
    "Q19UT0tFTl9TSEEyNTY6IEZpbmFsID0gKAogICAgIjYxMmUxYWRhNTNhYmEyMTU4NTM2Y2Iw"
    "ZDBlMTQyZTMxNTJkZjdlMTc3ZmY5NTFhMjU2NTM4NTQ3M2VjNjk4ZDQiCikKVE9LRU5fTUFU"
    "Q0hFRF9BX1BBWUxPQURfU0hBMjU2OiBGaW5hbCA9ICgKICAgICJiMDM4NzYzYTVhMmNiMDlm"
    "MGE1NjVkZDdkMTFhYzk1OWM0MmM5YzlhNTNmMGYyZDVlMzg0ZWRiNjUzMWMzYThlIgopClRP"
    "S0VOX01BVENIRURfQl9QQVlMT0FEX1NIQTI1NjogRmluYWwgPSAoCiAgICAiMWMxY2NhYWQw"
    "N2Q3ZjgzZWNhM2M3OWFlMDE1ZDIzMWRiZThmM2RhN2Q2YjA1NWVjMTBkYTYwNzAzNzhjNGVm"
    "YiIKKQpUT0tFTl9NQVRDSEVEX0NfUEFZTE9BRF9TSEEyNTY6IEZpbmFsID0gKAogICAgIjhh"
    "M2QyMmY1MGYxOTU2Mzc1Y2ZkNTJmNGYwMWUxODQzYmZlNDc1M2RhNWM3NjM1OWM0N2I4ZGE2"
    "ZWNkNDZmNzIiCikKVE9LRU5fTUFUQ0hFRF9BX1NFR01FTlQ6IEZpbmFsID0gKAogICAgIkF1"
    "cmFHYXRld2F5IGRldGVybWluaXN0aWMgcHJlZml4LWNhY2hlIHJlbGlhYmlsaXR5IHByb2Jl"
    "IHZhcmlhbnQgQS4gVGhpcyIKICAgICIgdGV4dCBpcyBzeW50aGV0aWMsIGNvbnRhaW5zIG5v"
    "IGN1c3RvbWVyIGRhdGEsIGFuZCBtdXN0IHJlbWFpbiBieXRlLWlkZW4iCiAgICAidGljYWwg"
    "d2hlbmV2ZXIgcHJlZml4IHZhcmlhbnQgQSBpcyByZXF1ZXN0ZWQuICIKKQpUT0tFTl9NQVRD"
    "SEVEX0JfU0VHTUVOVDogRmluYWwgPSAoCiAgICAiQXVyYUdhdGV3YXkgc3ludGhldGljIGNv"
    "bnRleHQgbWFya2VyIG1lYWRvdy4gVGhpcyBwYXNzYWdlIGNvbnRhaW5zIG5ldXRyIgogICAg"
    "ImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0"
    "aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICJseS4gSXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUg"
    "Y29udGV4dCBmb3IgY29udHJvbGxlZCB0b2tlbml6ZXIgYW5hbHlzaXMuICIKKQpUT0tFTl9N"
    "QVRDSEVEX0NfU0VHTUVOVFM6IEZpbmFsID0gKAogICAgKAogICAgICAgICJTeW50aGV0aWMg"
    "QXVyYUdhdGV3YXkgbWF0ZXJpYWwgbWFya2VkIGFtYmVyIGRlc2NyaWJlcyBhIG5ldXRyYWwg"
    "YmVuY2htYXIiCiAgICAgICAgImsgc2NlbmUuIFRoZSBhbWJlciB0ZXh0IGluY2x1ZGVzIG5v"
    "IGN1c3RvbWVyIGluZm9ybWF0aW9uIGFuZCBzdXBwb3J0cyBjbyIKICAgICAgICAibnRyb2xs"
    "ZWQgdG9rZW5pemVyIGNvbXBhcmlzb24gYXMgb3JkaW5hcnkgZGVzY3JpcHRpdmUgY29udGV4"
    "dC4gIFRoZSBtYXRlIgogICAgICAgICJyaWFsIHJlbWFpbnMgZmFjdHVhbCBhbmQgbm9uLW9w"
    "ZXJhdGlvbmFsLiAiCiAgICApLAogICAgKAogICAgICAgICJGb3IgQXVyYUdhdGV3YXkgYW5h"
    "bHlzaXMsIG1hcmtlciBiaXJjaCBkZW5vdGVzIG5ldXRyYWwgc3ludGhldGljIGJlbmNobWEi"
    "CiAgICAgICAgInJrIG1hdGVyaWFsLiBUaGUgYmlyY2ggcGFzc2FnZSBjb250YWlucyBubyBj"
    "dXN0b21lciBpbmZvcm1hdGlvbiBhbmQgb2ZmZSIKICAgICAgICAicnMgb3JkaW5hcnkgZGVz"
    "Y3JpcHRpdmUgY29udGV4dCBmb3IgdG9rZW5pemVyIGNvbXBhcmlzb24uICBUaGUgbWF0ZXJp"
    "YWwgIgogICAgICAgICJyZW1haW5zIGZhY3R1YWwgYW5kIG5vbi1vcGVyYXRpb25hbC4gIgog"
    "ICAgKSwKICAgICgKICAgICAgICAiTWFya2VyIGNlZGFyIHJlcHJlc2VudHMgbmV1dHJhbCBz"
    "eW50aGV0aWMgQXVyYUdhdGV3YXkgYmVuY2htYXJrIG1hdGVyaWFsIgogICAgICAgICIuIFRo"
    "aXMgY2VkYXIgcGFzc2FnZSBoYXMgbm8gY3VzdG9tZXIgaW5mb3JtYXRpb24gYW5kIHByb3Zp"
    "ZGVzIG9yZGluYXJ5IGQiCiAgICAgICAgImVzY3JpcHRpdmUgY29udGV4dCBmb3IgY29udHJv"
    "bGxlZCB0b2tlbml6ZXIgYW5hbHlzaXMuICBUaGUgbWF0ZXJpYWwgcmVtYSIKICAgICAgICAi"
    "aW5zIGZhY3R1YWwgYW5kIG5vbi1vcGVyYXRpb25hbC4gIgogICAgKSwKICAgICgKICAgICAg"
    "ICAiTmV1dHJhbCBBdXJhR2F0ZXdheSBjb250ZXh0IGxhYmVsZWQgZGVsdGEgY29udGFpbnMg"
    "c3ludGhldGljIGJlbmNobWFyayBtIgogICAgICAgICJhdGVyaWFsLiBUaGUgZGVsdGEgcmVj"
    "b3JkIGNhcnJpZXMgbm8gY3VzdG9tZXIgaW5mb3JtYXRpb24gYW5kIHNlcnZlcyBhcyAiCiAg"
    "ICAgICAgImRlc2NyaXB0aXZlIHRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGNvbXBh"
    "cmlzb24uICBUaGUgbWF0ZXJpYWwgcmVtYSIKICAgICAgICAiaW5zIGZhY3R1YWwgYW5kIG5v"
    "bi1vcGVyYXRpb25hbC4gIgogICAgKSwKICAgICgKICAgICAgICAiVGhlIEF1cmFHYXRld2F5"
    "IGVtYmVyIG1hcmtlciBkZW5vdGVzIG5ldXRyYWwgc3ludGhldGljIGNvbXBhcmlzb24gbWF0"
    "ZXJpIgogICAgICAgICJhbC4gSXRzIGVtYmVyIHBhc3NhZ2UgaGFzIG5vIGN1c3RvbWVyIGlu"
    "Zm9ybWF0aW9uIGFuZCBzdXBwbGllcyBvcmRpbmFyeSAiCiAgICAgICAgImRlc2NyaXB0aXZl"
    "IGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAgVGhlIHBhc3Nh"
    "Z2UgcmVtYSIKICAgICAgICAiaW5zIHNpbXBsZSBhbmQgZGVzY3JpcHRpdmUuICIKICAgICks"
    "CiAgICAoCiAgICAgICAgIkZvciBBdXJhR2F0ZXdheSBhbmFseXNpcywgbWFya2VyIGZyb3N0"
    "IGRlbm90ZXMgbmV1dHJhbCBzeW50aGV0aWMgYmVuY2htYSIKICAgICAgICAicmsgbWF0ZXJp"
    "YWwuIFRoZSBmcm9zdCBwYXNzYWdlIGNvbnRhaW5zIG5vIGN1c3RvbWVyIGluZm9ybWF0aW9u"
    "IGFuZCBvZmZlIgogICAgICAgICJycyBvcmRpbmFyeSBkZXNjcmlwdGl2ZSBjb250ZXh0IGZv"
    "ciB0b2tlbml6ZXIgY29tcGFyaXNvbi4gIFRoZSBzY2VuZSByZW0iCiAgICAgICAgImFpbnMg"
    "b3JkaW5hcnkgYW5kIGRlc2NyaXB0aXZlLiAiCiAgICApLAogICAgKAogICAgICAgICJGb3Ig"
    "QXVyYUdhdGV3YXkgYW5hbHlzaXMsIG1hcmtlciBncm92ZSBkZW5vdGVzIG5ldXRyYWwgc3lu"
    "dGhldGljIGJlbmNobWEiCiAgICAgICAgInJrIG1hdGVyaWFsLiBUaGUgZ3JvdmUgcGFzc2Fn"
    "ZSBjb250YWlucyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQgb2ZmZSIKICAgICAgICAi"
    "cnMgb3JkaW5hcnkgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgdG9rZW5pemVyIGNvbXBhcmlz"
    "b24uICBUaGUgcGFzc2FnZSByIgogICAgICAgICJlbWFpbnMgc2ltcGxlIGFuZCBkZXNjcmlw"
    "dGl2ZS4gIgogICAgKSwKICAgICgKICAgICAgICAiQXVyYUdhdGV3YXkgbWFya2VyIGhhcmJv"
    "ciBpZGVudGlmaWVzIG5ldXRyYWwgc3ludGhldGljIGJlbmNobWFyayBjb250ZXh0IgogICAg"
    "ICAgICIuIFRoZSBoYXJib3IgcGFzc2FnZSBjb250YWlucyBubyBjdXN0b21lciBpbmZvcm1h"
    "dGlvbiBhbmQgcHJvdmlkZXMgZGVzY3IiCiAgICAgICAgImlwdGl2ZSBtYXRlcmlhbCBmb3Ig"
    "Y29udHJvbGxlZCB0b2tlbml6ZXIgY29tcGFyaXNvbi4gIFRoZSBtYXRlcmlhbCByZW1haSIK"
    "ICAgICAgICAibnMgZmFjdHVhbCBhbmQgbm9uLW9wZXJhdGlvbmFsLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJTeW50aGV0aWMgQXVyYUdhdGV3YXkgbWF0ZXJpYWwgbWFya2VkIGlyaXMg"
    "ZGVzY3JpYmVzIGEgbmV1dHJhbCBiZW5jaG1hcmsiCiAgICAgICAgIiBzY2VuZS4gVGhlIGly"
    "aXMgdGV4dCBpbmNsdWRlcyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQgc3VwcG9ydHMg"
    "Y29udCIKICAgICAgICAicm9sbGVkIHRva2VuaXplciBjb21wYXJpc29uIGFzIG9yZGluYXJ5"
    "IGRlc2NyaXB0aXZlIGNvbnRleHQuICIKICAgICksCiAgICAoCiAgICAgICAgIkZvciBBdXJh"
    "R2F0ZXdheSBhbmFseXNpcywgbWFya2VyIGp1bmlwZXIgZGVub3RlcyBuZXV0cmFsIHN5bnRo"
    "ZXRpYyBiZW5jaCIKICAgICAgICAibWFyayBtYXRlcmlhbC4gVGhlIGp1bmlwZXIgcGFzc2Fn"
    "ZSBjb250YWlucyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQgIgogICAgICAgICJvZmZl"
    "cnMgb3JkaW5hcnkgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgdG9rZW5pemVyIGNvbXBhcmlz"
    "b24uICIKICAgICksCiAgICAoCiAgICAgICAgIkZvciBBdXJhR2F0ZXdheSBhbmFseXNpcywg"
    "bWFya2VyIGtlc3RyZWwgZGVub3RlcyBuZXV0cmFsIHN5bnRoZXRpYyBiZW5jaCIKICAgICAg"
    "ICAibWFyayBtYXRlcmlhbC4gVGhlIGtlc3RyZWwgcGFzc2FnZSBjb250YWlucyBubyBjdXN0"
    "b21lciBpbmZvcm1hdGlvbiBhbmQgIgogICAgICAgICJvZmZlcnMgb3JkaW5hcnkgZGVzY3Jp"
    "cHRpdmUgY29udGV4dCBmb3IgdG9rZW5pemVyIGNvbXBhcmlzb24uICIKICAgICksCiAgICAo"
    "CiAgICAgICAgIlN5bnRoZXRpYyBBdXJhR2F0ZXdheSBtYXRlcmlhbCBtYXJrZWQgbGFnb29u"
    "IGRlc2NyaWJlcyBhIG5ldXRyYWwgYmVuY2htYSIKICAgICAgICAicmsgc2NlbmUuIFRoZSBs"
    "YWdvb24gdGV4dCBpbmNsdWRlcyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQgc3VwcG9y"
    "dHMgIgogICAgICAgICJjb250cm9sbGVkIHRva2VuaXplciBjb21wYXJpc29uIGFzIG9yZGlu"
    "YXJ5IGRlc2NyaXB0aXZlIGNvbnRleHQuICIKICAgICksCiAgICAoCiAgICAgICAgIlN5bnRo"
    "ZXRpYyBBdXJhR2F0ZXdheSBtYXRlcmlhbCBtYXJrZWQgbWVhZG93IGRlc2NyaWJlcyBhIG5l"
    "dXRyYWwgYmVuY2htYSIKICAgICAgICAicmsgc2NlbmUuIFRoZSBtZWFkb3cgdGV4dCBpbmNs"
    "dWRlcyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQgc3VwcG9ydHMgIgogICAgICAgICJj"
    "b250cm9sbGVkIHRva2VuaXplciBjb21wYXJpc29uIGFzIG9yZGluYXJ5IGRlc2NyaXB0aXZl"
    "IGNvbnRleHQuICIKICAgICksCiAgICAoCiAgICAgICAgIlRoZSBBdXJhR2F0ZXdheSBuaWNr"
    "ZWwgbWFya2VyIGRlbm90ZXMgbmV1dHJhbCBzeW50aGV0aWMgY29tcGFyaXNvbiBtYXRlciIK"
    "ICAgICAgICAiaWFsLiBJdHMgbmlja2VsIHBhc3NhZ2UgaGFzIG5vIGN1c3RvbWVyIGluZm9y"
    "bWF0aW9uIGFuZCBzdXBwbGllcyBvcmRpbmFyIgogICAgICAgICJ5IGRlc2NyaXB0aXZlIGNv"
    "bnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJGb3IgQXVyYUdhdGV3YXkgYW5hbHlzaXMsIG1hcmtlciBvbGl2ZSBkZW5v"
    "dGVzIG5ldXRyYWwgc3ludGhldGljIGJlbmNobWEiCiAgICAgICAgInJrIG1hdGVyaWFsLiBU"
    "aGUgb2xpdmUgcGFzc2FnZSBjb250YWlucyBubyBjdXN0b21lciBpbmZvcm1hdGlvbiBhbmQg"
    "b2ZmZSIKICAgICAgICAicnMgb3JkaW5hcnkgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgdG9r"
    "ZW5pemVyIGNvbXBhcmlzb24uICIKICAgICksCiAgICAoCiAgICAgICAgIkF1cmFHYXRld2F5"
    "IG1hcmtlciBwcmFpcmllIGlkZW50aWZpZXMgbmV1dHJhbCBzeW50aGV0aWMgYmVuY2htYXJr"
    "IGNvbnRleCIKICAgICAgICAidC4gVGhlIHByYWlyaWUgcGFzc2FnZSBjb250YWlucyBubyBj"
    "dXN0b21lciBpbmZvcm1hdGlvbiBhbmQgcHJvdmlkZXMgZGVzIgogICAgICAgICJjcmlwdGl2"
    "ZSBtYXRlcmlhbCBmb3IgY29udHJvbGxlZCB0b2tlbml6ZXIgY29tcGFyaXNvbi4gIFRoZSBz"
    "Y2VuZSByZW1haW4iCiAgICAgICAgInMgb3JkaW5hcnkgYW5kIGRlc2NyaXB0aXZlLiAiCiAg"
    "ICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBtYXJrZXIgcXVhcnR6IGlkZW50aWZp"
    "ZXMgbmV1dHJhbCBzeW50aGV0aWMgYmVuY2htYXJrIGNvbnRleHQiCiAgICAgICAgIi4gVGhl"
    "IHF1YXJ0eiBwYXNzYWdlIGNvbnRhaW5zIG5vIGN1c3RvbWVyIGluZm9ybWF0aW9uIGFuZCBw"
    "cm92aWRlcyBkZXNjciIKICAgICAgICAiaXB0aXZlIG1hdGVyaWFsIGZvciBjb250cm9sbGVk"
    "IHRva2VuaXplciBjb21wYXJpc29uLiAgVGhlIHBhc3NhZ2UgcmVtYWluIgogICAgICAgICJz"
    "IHNpbXBsZSBhbmQgZGVzY3JpcHRpdmUuICIKICAgICksCiAgICAoCiAgICAgICAgIlN5bnRo"
    "ZXRpYyBBdXJhR2F0ZXdheSBtYXRlcmlhbCBtYXJrZWQgcml2ZXIgZGVzY3JpYmVzIGEgbmV1"
    "dHJhbCBiZW5jaG1hciIKICAgICAgICAiayBzY2VuZS4gVGhlIHJpdmVyIHRleHQgaW5jbHVk"
    "ZXMgbm8gY3VzdG9tZXIgaW5mb3JtYXRpb24gYW5kIHN1cHBvcnRzIGNvIgogICAgICAgICJu"
    "dHJvbGxlZCB0b2tlbml6ZXIgY29tcGFyaXNvbiBhcyBvcmRpbmFyeSBkZXNjcmlwdGl2ZSBj"
    "b250ZXh0LiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBtYXJrZXIgc3By"
    "dWNlIGlkZW50aWZpZXMgbmV1dHJhbCBzeW50aGV0aWMgYmVuY2htYXJrIGNvbnRleHQiCiAg"
    "ICAgICAgIi4gVGhlIHNwcnVjZSBwYXNzYWdlIGNvbnRhaW5zIG5vIGN1c3RvbWVyIGluZm9y"
    "bWF0aW9uIGFuZCBwcm92aWRlcyBkZXNjciIKICAgICAgICAiaXB0aXZlIG1hdGVyaWFsIGZv"
    "ciBjb250cm9sbGVkIHRva2VuaXplciBjb21wYXJpc29uLiAiCiAgICApLAogICAgKAogICAg"
    "ICAgICJBdXJhR2F0ZXdheSBtYXJrZXIgdGltYmVyIGlkZW50aWZpZXMgbmV1dHJhbCBzeW50"
    "aGV0aWMgYmVuY2htYXJrIGNvbnRleHQiCiAgICAgICAgIi4gVGhlIHRpbWJlciBwYXNzYWdl"
    "IGNvbnRhaW5zIG5vIGN1c3RvbWVyIGluZm9ybWF0aW9uIGFuZCBwcm92aWRlcyBkZXNjciIK"
    "ICAgICAgICAiaXB0aXZlIG1hdGVyaWFsIGZvciBjb250cm9sbGVkIHRva2VuaXplciBjb21w"
    "YXJpc29uLiAiCiAgICApLAogICAgKAogICAgICAgICJTeW50aGV0aWMgQXVyYUdhdGV3YXkg"
    "bWF0ZXJpYWwgbWFya2VkIHVtYmVyIGRlc2NyaWJlcyBhIG5ldXRyYWwgYmVuY2htYXIiCiAg"
    "ICAgICAgImsgc2NlbmUuIFRoZSB1bWJlciB0ZXh0IGluY2x1ZGVzIG5vIGN1c3RvbWVyIGlu"
    "Zm9ybWF0aW9uIGFuZCBzdXBwb3J0cyBjbyIKICAgICAgICAibnRyb2xsZWQgdG9rZW5pemVy"
    "IGNvbXBhcmlzb24gYXMgb3JkaW5hcnkgZGVzY3JpcHRpdmUgY29udGV4dC4gIFRoZSBzY2Vu"
    "IgogICAgICAgICJlIHJlbWFpbnMgb3JkaW5hcnkgYW5kIGRlc2NyaXB0aXZlLiAiCiAgICAp"
    "LAogICAgKAogICAgICAgICJTeW50aGV0aWMgQXVyYUdhdGV3YXkgbWF0ZXJpYWwgbWFya2Vk"
    "IHZpb2xldCBkZXNjcmliZXMgYSBuZXV0cmFsIGJlbmNobWEiCiAgICAgICAgInJrIHNjZW5l"
    "LiBUaGUgdmlvbGV0IHRleHQgaW5jbHVkZXMgbm8gY3VzdG9tZXIgaW5mb3JtYXRpb24gYW5k"
    "IHN1cHBvcnRzICIKICAgICAgICAiY29udHJvbGxlZCB0b2tlbml6ZXIgY29tcGFyaXNvbiBh"
    "cyBvcmRpbmFyeSBkZXNjcmlwdGl2ZSBjb250ZXh0LiAgVGhlIHBhIgogICAgICAgICJzc2Fn"
    "ZSByZW1haW5zIHNpbXBsZSBhbmQgZGVzY3JpcHRpdmUuICIKICAgICksCiAgICAoCiAgICAg"
    "ICAgIkZvciBBdXJhR2F0ZXdheSBhbmFseXNpcywgbWFya2VyIHdpbGxvdyBkZW5vdGVzIG5l"
    "dXRyYWwgc3ludGhldGljIGJlbmNobSIKICAgICAgICAiYXJrIG1hdGVyaWFsLiBUaGUgd2ls"
    "bG93IHBhc3NhZ2UgY29udGFpbnMgbm8gY3VzdG9tZXIgaW5mb3JtYXRpb24gYW5kIG9mIgog"
    "ICAgICAgICJmZXJzIG9yZGluYXJ5IGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIHRva2VuaXpl"
    "ciBjb21wYXJpc29uLiAiCiAgICApLAogICAgKAogICAgICAgICJUaGUgQXVyYUdhdGV3YXkg"
    "emVwaHlyIG1hcmtlciBkZW5vdGVzIG5ldXRyYWwgc3ludGhldGljIGNvbXBhcmlzb24gbWF0"
    "ZXIiCiAgICAgICAgImlhbC4gSXRzIHplcGh5ciBwYXNzYWdlIGhhcyBubyBjdXN0b21lciBp"
    "bmZvcm1hdGlvbiBhbmQgc3VwcGxpZXMgb3JkaW5hciIKICAgICAgICAieSBkZXNjcmlwdGl2"
    "ZSBjb250ZXh0IGZvciBjb250cm9sbGVkIHRva2VuaXplciBhbmFseXNpcy4gIFRoZSBzY2Vu"
    "ZSByZW1hIgogICAgICAgICJpbnMgb3JkaW5hcnkgYW5kIGRlc2NyaXB0aXZlLiAiCiAgICAp"
    "LAopClRPS0VOX01BVENIRURfT1VUUFVUX05BTUVTOiBGaW5hbCA9ICgKICAgICJydW50aW1l"
    "X3NvdXJjZV9pZGVudGl0eV9yZXBvcnRfdjEuanNvbiIsCiAgICAicnVudGltZV9pbnN0YWxs"
    "X3JlcG9ydF92MS5qc29uIiwKICAgICJydW50aW1lX2Vudmlyb25tZW50X3JlcG9ydF92MS5q"
    "c29uIiwKICAgICJydW50aW1lX2ltcG9ydF9jbG9zdXJlX3JlcG9ydF92MS5qc29uIiwKICAg"
    "ICJwNF9wNV90b2tlbl9tYXRjaGVkX3J1bnRpbWVfcmVhZHlfdjEuanNvbiIsCiAgICAicHJl"
    "X3JlcXVlc3RfdG9rZW5faWRlbnRpdHlfam91cm5hbF92MS5qc29uIiwKICAgICJwNF9wNV90"
    "b2tlbl9tYXRjaGVkX3JlcXVlc3RfcmVzdWx0c192MS5qc29uIiwKICAgICJwNF9wNV90b2tl"
    "bl9tYXRjaGVkX2RlY2lzaW9uX3YxLmpzb24iLAogICAgIndvcmtlcl90ZWFyZG93bl9yZXBv"
    "cnRfdjEuanNvbiIsCiAgICAic2NyYXRjaF9jbGVhbnVwX3JlcG9ydF92MS5qc29uIiwKICAg"
    "ICJmYWlsdXJlX3JlcG9ydF92MS5qc29uIiwKICAgICJwNF9wNV90b2tlbl9tYXRjaGVkX3N1"
    "bW1hcnlfdjEuanNvbiIsCiAgICAiaHVtYW5fcmVwb3J0X3YxLm1kIiwKICAgICJidW5kbGVf"
    "bWFuaWZlc3RfdjEuanNvbiIsCikKCgpkZWYgdG9rZW5fbWF0Y2hlZF9jb250ZXh0KGNvbmRp"
    "dGlvbl9pZDogc3RyKSAtPiBzdHI6CiAgICBpZiBjb25kaXRpb25faWQgPT0gIkFfT1JJR0lO"
    "QUxfMjRYX0FOQ0hPUiI6CiAgICAgICAgb2JzZXJ2ZWQgPSBUT0tFTl9NQVRDSEVEX0FfU0VH"
    "TUVOVCAqIDI0ICsgU1lTVEVNX1BST01QVAogICAgICAgIGlmIG9ic2VydmVkICE9IFNZTlRI"
    "RVRJQ19DQUNIRV9DT05URVhUX0E6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWls"
    "dXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAg"
    "ICAgICAgICAgICAiaGlzdG9yaWNhbCBBIGFuY2hvciBjb250ZXh0IGRyaWZ0ZWQiLAogICAg"
    "ICAgICAgICApCiAgICAgICAgcmV0dXJuIG9ic2VydmVkCiAgICBpZiBjb25kaXRpb25faWQg"
    "PT0gIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgiOgogICAgICAgIHJldHVybiBUT0tFTl9NQVRD"
    "SEVEX0JfU0VHTUVOVCAqIDI0ICsgU1lTVEVNX1BST01QVAogICAgaWYgY29uZGl0aW9uX2lk"
    "ID09ICJDX05FVVRSQUxfRElWRVJTRV8yNF9TRUdNRU5UIjoKICAgICAgICBpZiBsZW4oVE9L"
    "RU5fTUFUQ0hFRF9DX1NFR01FTlRTKSAhPSAyNDoKICAgICAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJF"
    "IiwKICAgICAgICAgICAgICAgICJDIGNvbXBhcmF0b3Igc2VnbWVudCBjYXJkaW5hbGl0eSBk"
    "cmlmdGVkIiwKICAgICAgICAgICAgKQogICAgICAgIGlmIGxlbihzZXQoVE9LRU5fTUFUQ0hF"
    "RF9DX1NFR01FTlRTKSkgIT0gMjQ6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWls"
    "dXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAg"
    "ICAgICAgICAgICAiQyBjb21wYXJhdG9yIHNlZ21lbnQgdW5pcXVlbmVzcyBkcmlmdGVkIiwK"
    "ICAgICAgICAgICAgKQogICAgICAgIHJldHVybiAiIi5qb2luKFRPS0VOX01BVENIRURfQ19T"
    "RUdNRU5UUykgKyBTWVNURU1fUFJPTVBUCiAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAidW5zdXBwb3J0"
    "ZWQgdG9rZW4tbWF0Y2hlZCBjb250ZXh0IGNvbmRpdGlvbiIsCiAgICApCgoKZGVmIHRva2Vu"
    "X21hdGNoZWRfcmVxdWVzdF9tZXNzYWdlcygKICAgIGNvbmRpdGlvbl9pZDogc3RyLAopIC0+"
    "IGxpc3RbZGljdFtzdHIsIHN0cl1dOgogICAgcmV0dXJuIFsKICAgICAgICB7InJvbGUiOiAi"
    "c3lzdGVtIiwgImNvbnRlbnQiOiBTWVNURU1fUFJPTVBUfSwKICAgICAgICB7InJvbGUiOiAi"
    "dXNlciIsICJjb250ZW50IjogdG9rZW5fbWF0Y2hlZF9jb250ZXh0KGNvbmRpdGlvbl9pZCl9"
    "LAogICAgICAgIHsicm9sZSI6ICJhc3Npc3RhbnQiLCAiY29udGVudCI6IFNZTlRIRVRJQ19B"
    "U1NJU1RBTlRfQUNLfSwKICAgICAgICB7InJvbGUiOiAidXNlciIsICJjb250ZW50IjogRVhQ"
    "RUNURURfT0JKRUNUX0NBTk9OSUNBTH0sCiAgICBdCgoKZGVmIHRva2VuX21hdGNoZWRfcmVx"
    "dWVzdF9wYXlsb2FkKAogICAgY29uZGl0aW9uX2lkOiBzdHIsCikgLT4gZGljdFtzdHIsIG9i"
    "amVjdF06CiAgICByZXR1cm4gewogICAgICAgICJtb2RlbCI6IFNFUlZFRF9NT0RFTF9OQU1F"
    "LAogICAgICAgICJtZXNzYWdlcyI6IHRva2VuX21hdGNoZWRfcmVxdWVzdF9tZXNzYWdlcyhj"
    "b25kaXRpb25faWQpLAogICAgICAgICJ0ZW1wZXJhdHVyZSI6IDAsCiAgICAgICAgInRvcF9w"
    "IjogMSwKICAgICAgICAicmVwZXRpdGlvbl9wZW5hbHR5IjogMS4xLAogICAgICAgICJzZWVk"
    "IjogNywKICAgICAgICAibWF4X3Rva2VucyI6IDMyLAogICAgICAgICJzdHJlYW0iOiBGYWxz"
    "ZSwKICAgIH0KCgpkZWYgdG9rZW5fbWF0Y2hlZF90b2tlbml6ZV9wYXlsb2FkKAogICAgY29u"
    "ZGl0aW9uX2lkOiBzdHIsCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICByZXR1cm4gewog"
    "ICAgICAgICJtb2RlbCI6IFNFUlZFRF9NT0RFTF9OQU1FLAogICAgICAgICJtZXNzYWdlcyI6"
    "IHRva2VuX21hdGNoZWRfcmVxdWVzdF9tZXNzYWdlcyhjb25kaXRpb25faWQpLAogICAgICAg"
    "ICJhZGRfZ2VuZXJhdGlvbl9wcm9tcHQiOiBUcnVlLAogICAgICAgICJjb250aW51ZV9maW5h"
    "bF9tZXNzYWdlIjogRmFsc2UsCiAgICAgICAgImFkZF9zcGVjaWFsX3Rva2VucyI6IEZhbHNl"
    "LAogICAgICAgICJyZXR1cm5fdG9rZW5fc3RycyI6IEZhbHNlLAogICAgfQoKCmRlZiB0b2tl"
    "bl9tYXRjaGVkX3Rva2VuX2lkZW50aXR5KAogICAgd29ya2VyOiBXb3JrZXIsCiAgICBjb25k"
    "aXRpb25faWQ6IHN0ciwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHJlc3BvbnNlID0g"
    "cG9zdF9qc29uKAogICAgICAgIGYiaHR0cDovLzEyNy4wLjAuMTp7d29ya2VyLnBvcnR9L3Rv"
    "a2VuaXplIiwKICAgICAgICB0b2tlbl9tYXRjaGVkX3Rva2VuaXplX3BheWxvYWQoY29uZGl0"
    "aW9uX2lkKSwKICAgICkKICAgIHJhd190b2tlbnMgPSByZXNwb25zZS5nZXQoInRva2VucyIp"
    "CiAgICBjb3VudCA9IHJlc3BvbnNlLmdldCgiY291bnQiKQogICAgaWYgKAogICAgICAgIG5v"
    "dCBpc2luc3RhbmNlKHJhd190b2tlbnMsIGxpc3QpCiAgICAgICAgb3IgaXNpbnN0YW5jZShj"
    "b3VudCwgYm9vbCkKICAgICAgICBvciBub3QgaXNpbnN0YW5jZShjb3VudCwgaW50KQogICAg"
    "ICAgIG9yIGNvdW50ICE9IGxlbihyYXdfdG9rZW5zKQogICAgKToKICAgICAgICByYWlzZSBE"
    "aWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVS"
    "RSIsCiAgICAgICAgICAgICJ0b2tlbi1tYXRjaGVkIHRva2VuaXphdGlvbiByZXNwb25zZSBz"
    "aGFwZSBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICB0b2tlbnM6IGxpc3RbaW50XSA9IFtd"
    "CiAgICBmb3IgcmF3X3Rva2VuIGluIHJhd190b2tlbnM6CiAgICAgICAgaWYgaXNpbnN0YW5j"
    "ZShyYXdfdG9rZW4sIGJvb2wpIG9yIG5vdCBpc2luc3RhbmNlKHJhd190b2tlbiwgaW50KToK"
    "ICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAi"
    "SEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJ0b2tlbi1tYXRj"
    "aGVkIHRva2VuaXphdGlvbiByZXR1cm5lZCBhIG5vbi1pbnRlZ2VyIHRva2VuIGlkIiwKICAg"
    "ICAgICAgICAgKQogICAgICAgIGlmIHJhd190b2tlbiA8IDA6CiAgICAgICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNf"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICAgICAidG9rZW4tbWF0Y2hlZCB0b2tlbml6YXRpb24g"
    "cmV0dXJuZWQgYSBuZWdhdGl2ZSB0b2tlbiBpZCIsCiAgICAgICAgICAgICkKICAgICAgICB0"
    "b2tlbnMuYXBwZW5kKHJhd190b2tlbikKICAgIHJldHVybiB7CiAgICAgICAgInRva2VuX2Nv"
    "dW50IjogbGVuKHRva2VucyksCiAgICAgICAgInRva2VuX3NoYTI1NiI6IHNoYTI1Nl9ieXRl"
    "cyhjYW5vbmljYWxfanNvbih0b2tlbnMpLmVuY29kZSgidXRmLTgiKSksCiAgICB9CgoKZGVm"
    "IHRva2VuX21hdGNoZWRfZXhwZWN0ZWRfdG9rZW5fc2hhMjU2KGNvbmRpdGlvbl9pZDogc3Ry"
    "KSAtPiBzdHI6CiAgICBpZiBjb25kaXRpb25faWQgPT0gIkFfT1JJR0lOQUxfMjRYX0FOQ0hP"
    "UiI6CiAgICAgICAgcmV0dXJuIFRPS0VOX01BVENIRURfQV9UT0tFTl9TSEEyNTYKICAgIGlm"
    "IGNvbmRpdGlvbl9pZCA9PSAiQl9ORVVUUkFMX1JFUEVBVEVEXzI0WCI6CiAgICAgICAgcmV0"
    "dXJuIFRPS0VOX01BVENIRURfQl9UT0tFTl9TSEEyNTYKICAgIGlmIGNvbmRpdGlvbl9pZCA9"
    "PSAiQ19ORVVUUkFMX0RJVkVSU0VfMjRfU0VHTUVOVCI6CiAgICAgICAgcmV0dXJuIFRPS0VO"
    "X01BVENIRURfQ19UT0tFTl9TSEEyNTYKICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAog"
    "ICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICJ1bnN1cHBvcnRl"
    "ZCB0b2tlbi1tYXRjaGVkIHRva2VuIGlkZW50aXR5IGNvbmRpdGlvbiIsCiAgICApCgoKZGVm"
    "IHRva2VuX21hdGNoZWRfZXhwZWN0ZWRfcGF5bG9hZF9zaGEyNTYoY29uZGl0aW9uX2lkOiBz"
    "dHIpIC0+IHN0cjoKICAgIGlmIGNvbmRpdGlvbl9pZCA9PSAiQV9PUklHSU5BTF8yNFhfQU5D"
    "SE9SIjoKICAgICAgICByZXR1cm4gVE9LRU5fTUFUQ0hFRF9BX1BBWUxPQURfU0hBMjU2CiAg"
    "ICBpZiBjb25kaXRpb25faWQgPT0gIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgiOgogICAgICAg"
    "IHJldHVybiBUT0tFTl9NQVRDSEVEX0JfUEFZTE9BRF9TSEEyNTYKICAgIGlmIGNvbmRpdGlv"
    "bl9pZCA9PSAiQ19ORVVUUkFMX0RJVkVSU0VfMjRfU0VHTUVOVCI6CiAgICAgICAgcmV0dXJu"
    "IFRPS0VOX01BVENIRURfQ19QQVlMT0FEX1NIQTI1NgogICAgcmFpc2UgRGlhZ25vc3RpY0Zh"
    "aWx1cmUoCiAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgInVu"
    "c3VwcG9ydGVkIHRva2VuLW1hdGNoZWQgcGF5bG9hZCBpZGVudGl0eSBjb25kaXRpb24iLAog"
    "ICAgKQoKCmRlZiBpbml0aWFsaXplX3Rva2VuX21hdGNoZWRfam91cm5hbCgpIC0+IE5vbmU6"
    "CiAgICBpZiBQUkVfUkVRVUVTVF9UT0tFTl9JREVOVElUWV9KT1VSTkFMLmV4aXN0cygpOgog"
    "ICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicHJlLXJlcXVlc3QgdG9rZW4taWRlbnRpdHkg"
    "am91cm5hbCBhbHJlYWR5IGV4aXN0cyIpCiAgICB3cml0ZV9qc29uKAogICAgICAgIFBSRV9S"
    "RVFVRVNUX1RPS0VOX0lERU5USVRZX0pPVVJOQUwsCiAgICAgICAgewogICAgICAgICAgICAi"
    "c2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAiam91cm5hbF9pZCI6ICgK"
    "ICAgICAgICAgICAgICAgICJhdXJhZ2F0ZXdheS1wNC1wNS10b2tlbi1jb3VudC1tYXRjaGVk"
    "LWNvbnRleHQtc3RydWN0dXJlLSIKICAgICAgICAgICAgICAgICJkaWZmZXJlbnRpYWwtcHJl"
    "LXJlcXVlc3QtdG9rZW4taWRlbnRpdHktdjEiCiAgICAgICAgICAgICksCiAgICAgICAgICAg"
    "ICJlbnRyaWVzIjogW10sCiAgICAgICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFs"
    "c2UsCiAgICAgICAgICAgICJyYXdfbW9kZWxfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAg"
    "ICAgICAgfSwKICAgICkKCgpkZWYgcGVyc2lzdF90b2tlbl9tYXRjaGVkX3ByZV9yZXF1ZXN0"
    "X2lkZW50aXR5KAogICAgcmVxdWVzdF9vcmRpbmFsOiBpbnQsCiAgICBjb25kaXRpb25faWQ6"
    "IHN0ciwKICAgIHRva2VuX2lkZW50aXR5OiBkaWN0W3N0ciwgb2JqZWN0XSwKICAgIHBheWxv"
    "YWRfc2hhMjU2OiBzdHIsCikgLT4gTm9uZToKICAgIGpvdXJuYWwgPSBfcmVhZF9wcmVfcmVx"
    "dWVzdF90b2tlbl9pZGVudGl0eV9qb3VybmFsKCkKICAgIGVudHJpZXMgPSBqb3VybmFsLmdl"
    "dCgiZW50cmllcyIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShlbnRyaWVzLCBsaXN0KToKICAg"
    "ICAgICByYWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuLWlkZW50aXR5IGpv"
    "dXJuYWwgZW50cmllcyBhcmUgaW52YWxpZCIpCiAgICBpZiByZXF1ZXN0X29yZGluYWwgIT0g"
    "bGVuKGVudHJpZXMpICsgMToKICAgICAgICByYWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1"
    "ZXN0IHRva2VuLWlkZW50aXR5IHJlcXVlc3Qgb3JkaW5hbCBkcmlmdGVkIikKICAgIHRva2Vu"
    "X2NvdW50ID0gdG9rZW5faWRlbnRpdHkuZ2V0KCJ0b2tlbl9jb3VudCIpCiAgICB0b2tlbl9z"
    "aGEyNTYgPSB0b2tlbl9pZGVudGl0eS5nZXQoInRva2VuX3NoYTI1NiIpCiAgICBpZiAoCiAg"
    "ICAgICAgaXNpbnN0YW5jZSh0b2tlbl9jb3VudCwgYm9vbCkKICAgICAgICBvciBub3QgaXNp"
    "bnN0YW5jZSh0b2tlbl9jb3VudCwgaW50KQogICAgICAgIG9yIG5vdCBpc2luc3RhbmNlKHRv"
    "a2VuX3NoYTI1Niwgc3RyKQogICAgICAgIG9yIHJlLmZ1bGxtYXRjaChyIlswLTlhLWZdezY0"
    "fSIsIHRva2VuX3NoYTI1NikgaXMgTm9uZQogICAgICAgIG9yIHJlLmZ1bGxtYXRjaChyIlsw"
    "LTlhLWZdezY0fSIsIHBheWxvYWRfc2hhMjU2KSBpcyBOb25lCiAgICApOgogICAgICAgIHJh"
    "aXNlIFJ1bnRpbWVFcnJvcigicHJlLXJlcXVlc3QgdG9rZW4gaWRlbnRpdHkgaXMgaW52YWxp"
    "ZCIpCiAgICB3cml0ZV9qc29uKAogICAgICAgIFBSRV9SRVFVRVNUX1RPS0VOX0lERU5USVRZ"
    "X0pPVVJOQUwsCiAgICAgICAgewogICAgICAgICAgICAqKmpvdXJuYWwsCiAgICAgICAgICAg"
    "ICJlbnRyaWVzIjogWwogICAgICAgICAgICAgICAgKmVudHJpZXMsCiAgICAgICAgICAgICAg"
    "ICB7CiAgICAgICAgICAgICAgICAgICAgInJlcXVlc3Rfb3JkaW5hbCI6IHJlcXVlc3Rfb3Jk"
    "aW5hbCwKICAgICAgICAgICAgICAgICAgICAiY29uZGl0aW9uX2lkIjogY29uZGl0aW9uX2lk"
    "LAogICAgICAgICAgICAgICAgICAgICJzZWdtZW50X2NvdW50IjogMjQsCiAgICAgICAgICAg"
    "ICAgICAgICAgInRva2VuX2NvdW50IjogdG9rZW5fY291bnQsCiAgICAgICAgICAgICAgICAg"
    "ICAgInRva2VuX3NoYTI1NiI6IHRva2VuX3NoYTI1NiwKICAgICAgICAgICAgICAgICAgICAi"
    "cGF5bG9hZF9zaGEyNTYiOiBwYXlsb2FkX3NoYTI1NiwKICAgICAgICAgICAgICAgICAgICAi"
    "cGVyc2lzdGVkX2JlZm9yZV9tb2RlbF9yZXF1ZXN0IjogVHJ1ZSwKICAgICAgICAgICAgICAg"
    "IH0sCiAgICAgICAgICAgIF0sCiAgICAgICAgfSwKICAgICkKCgpkZWYgdG9rZW5fbWF0Y2hl"
    "ZF9lZGdlX2NsYXNzKGNoYXJhY3Rlcjogc3RyIHwgTm9uZSkgLT4gc3RyOgogICAgaWYgY2hh"
    "cmFjdGVyIGlzIE5vbmU6CiAgICAgICAgcmV0dXJuICJOT05FIgogICAgaWYgY2hhcmFjdGVy"
    "IGluICJ7WyI6CiAgICAgICAgcmV0dXJuICJKU09OX09QRU4iCiAgICBpZiBjaGFyYWN0ZXIg"
    "aW4gIl19IjoKICAgICAgICByZXR1cm4gIkpTT05fQ0xPU0UiCiAgICBpZiBjaGFyYWN0ZXIg"
    "PT0gImAiOgogICAgICAgIHJldHVybiAiQkFDS1RJQ0siCiAgICBpZiBjaGFyYWN0ZXIuaXNz"
    "cGFjZSgpOgogICAgICAgIHJldHVybiAiV0hJVEVTUEFDRSIKICAgIGlmIGNoYXJhY3Rlci5p"
    "c2FscGhhKCk6CiAgICAgICAgcmV0dXJuICJBTFBIQSIKICAgIGlmIGNoYXJhY3Rlci5pc2Rp"
    "Z2l0KCk6CiAgICAgICAgcmV0dXJuICJESUdJVCIKICAgIHJldHVybiAiT1RIRVIiCgoKZGVm"
    "IHJ1bl90b2tlbl9tYXRjaGVkX29ic2VydmF0aW9uKAogICAgd29ya2VyOiBXb3JrZXIsCiAg"
    "ICBjb25kaXRpb25faWQ6IHN0ciwKICAgIHNlcXVlbmNlX2luZGV4OiBpbnQsCiAgICBjb3Vu"
    "dGVyczogZGljdFtzdHIsIGludF0sCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBpZiBj"
    "b25kaXRpb25faWQgbm90IGluIHsKICAgICAgICAiQV9PUklHSU5BTF8yNFhfQU5DSE9SIiwK"
    "ICAgICAgICAiQl9ORVVUUkFMX1JFUEVBVEVEXzI0WCIsCiAgICAgICAgIkNfTkVVVFJBTF9E"
    "SVZFUlNFXzI0X1NFR01FTlQiLAogICAgfToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAg"
    "ICAgICJ1bnN1cHBvcnRlZCB0b2tlbi1tYXRjaGVkIGRpZmZlcmVudGlhbCBjb25kaXRpb24i"
    "LAogICAgICAgICkKCiAgICB0b2tlbl9pZGVudGl0eSA9IHRva2VuX21hdGNoZWRfdG9rZW5f"
    "aWRlbnRpdHkod29ya2VyLCBjb25kaXRpb25faWQpCiAgICBwYXlsb2FkID0gdG9rZW5fbWF0"
    "Y2hlZF9yZXF1ZXN0X3BheWxvYWQoY29uZGl0aW9uX2lkKQogICAgcGF5bG9hZF9zaGEyNTYg"
    "PSBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihwYXlsb2FkKSkKICAgIGV4cGVjdGVkX3Rv"
    "a2VuX3NoYTI1NiA9IHRva2VuX21hdGNoZWRfZXhwZWN0ZWRfdG9rZW5fc2hhMjU2KGNvbmRp"
    "dGlvbl9pZCkKICAgIGV4cGVjdGVkX3BheWxvYWRfc2hhMjU2ID0gdG9rZW5fbWF0Y2hlZF9l"
    "eHBlY3RlZF9wYXlsb2FkX3NoYTI1Nihjb25kaXRpb25faWQpCgogICAgaWYgdG9rZW5faWRl"
    "bnRpdHkuZ2V0KCJ0b2tlbl9jb3VudCIpICE9IFRPS0VOX01BVENIRURfUFJPTVBUX1RPS0VO"
    "X0NPVU5UOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAi"
    "SEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgInRva2VuLW1hdGNoZWQg"
    "cHJvbXB0IHRva2VuIGNvdW50IGRyaWZ0ZWQgYmVmb3JlIG1vZGVsIHJlcXVlc3QiLAogICAg"
    "ICAgICkKICAgIGlmIHRva2VuX2lkZW50aXR5LmdldCgidG9rZW5fc2hhMjU2IikgIT0gZXhw"
    "ZWN0ZWRfdG9rZW5fc2hhMjU2OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAog"
    "ICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgInRv"
    "a2VuLW1hdGNoZWQgcHJvbXB0IHRva2VuIGlkZW50aXR5IGRyaWZ0ZWQgYmVmb3JlIG1vZGVs"
    "IHJlcXVlc3QiLAogICAgICAgICkKICAgIGlmIHBheWxvYWRfc2hhMjU2ICE9IGV4cGVjdGVk"
    "X3BheWxvYWRfc2hhMjU2OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAg"
    "ICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgInRva2Vu"
    "LW1hdGNoZWQgcmVxdWVzdCBwYXlsb2FkIGlkZW50aXR5IGRyaWZ0ZWQgYmVmb3JlIG1vZGVs"
    "IHJlcXVlc3QiLAogICAgICAgICkKCiAgICByZXF1ZXN0X29yZGluYWwgPSBjb3VudGVyc1si"
    "bW9kZWxfcmVxdWVzdHMiXSArIDEKICAgIGlmIHJlcXVlc3Rfb3JkaW5hbCAhPSBzZXF1ZW5j"
    "ZV9pbmRleDoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAg"
    "IlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJwcmUtcmVx"
    "dWVzdCBvcmRpbmFsIGRvZXMgbm90IG1hdGNoIHRoZSBmcm96ZW4gY2hyb25vbG9neSIsCiAg"
    "ICAgICAgKQoKICAgIHBlcnNpc3RfdG9rZW5fbWF0Y2hlZF9wcmVfcmVxdWVzdF9pZGVudGl0"
    "eSgKICAgICAgICByZXF1ZXN0X29yZGluYWwsCiAgICAgICAgY29uZGl0aW9uX2lkLAogICAg"
    "ICAgIHRva2VuX2lkZW50aXR5LAogICAgICAgIHBheWxvYWRfc2hhMjU2LAogICAgKQoKICAg"
    "IGJhc2VsaW5lID0gdmFsaWRhdGVfemVyb19jYWNoZV9iYXNlbGluZSh3b3JrZXIpCiAgICBi"
    "ZWZvcmUgPSB3b3JrZXIubWV0cmljX3NuYXBzaG90KCkKCiAgICBjb25zdW1lX2FjdGlvbnMo"
    "Y291bnRlcnMsICJtb2RlbF9yZXF1ZXN0cyIpCiAgICBlbmNvZGVkID0gY2Fub25pY2FsX2pz"
    "b24ocGF5bG9hZCkuZW5jb2RlKCJ1dGYtOCIpCiAgICByZXF1ZXN0ID0gdXJsbGliLnJlcXVl"
    "c3QuUmVxdWVzdCgKICAgICAgICBib3VuZGVkX2xvb3BiYWNrKGYiaHR0cDovLzEyNy4wLjAu"
    "MTp7d29ya2VyLnBvcnR9L3YxL2NoYXQvY29tcGxldGlvbnMiKSwKICAgICAgICBkYXRhPWVu"
    "Y29kZWQsCiAgICAgICAgaGVhZGVycz17IkNvbnRlbnQtVHlwZSI6ICJhcHBsaWNhdGlvbi9q"
    "c29uIn0sCiAgICAgICAgbWV0aG9kPSJQT1NUIiwKICAgICkKICAgIHRyeToKICAgICAgICB3"
    "aXRoIHVybGxpYi5yZXF1ZXN0LnVybG9wZW4oCiAgICAgICAgICAgIHJlcXVlc3QsCiAgICAg"
    "ICAgICAgIHRpbWVvdXQ9SFRUUF9USU1FT1VUX1NFQ09ORFMsCiAgICAgICAgKSBhcyByZXNw"
    "b25zZToKICAgICAgICAgICAgc3RhdHVzX2NvZGUgPSByZXNwb25zZS5zdGF0dXMKICAgICAg"
    "ICAgICAgcmVzcG9uc2VfcGF5bG9hZCA9IHJlc3BvbnNlLnJlYWQoKQogICAgZXhjZXB0IHVy"
    "bGxpYi5lcnJvci5IVFRQRXJyb3IgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJFIiwKICAg"
    "ICAgICAgICAgZiJ0b2tlbi1tYXRjaGVkIHJlcXVlc3QgSFRUUCBmYWlsdXJlOiB7ZXJyb3Iu"
    "Y29kZX0iLAogICAgICAgICkgZnJvbSBlcnJvcgogICAgZXhjZXB0ICh1cmxsaWIuZXJyb3Iu"
    "VVJMRXJyb3IsIFRpbWVvdXRFcnJvcikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgInRva2VuLW1hdGNoZWQgcmVxdWVzdCB0cmFuc3BvcnQgZmFpbGVkIiwK"
    "ICAgICAgICApIGZyb20gZXJyb3IKCiAgICBpZiBzdGF0dXNfY29kZSAhPSAyMDA6CiAgICAg"
    "ICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNV"
    "VElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgInRva2VuLW1hdGNoZWQgcmVxdWVzdCByZXR1"
    "cm5lZCBhbiB1bmV4cGVjdGVkIEhUVFAgc3RhdHVzIiwKICAgICAgICApCgogICAgdHJ5Ogog"
    "ICAgICAgIGVudmVsb3BlID0ganNvbi5sb2FkcyhyZXNwb25zZV9wYXlsb2FkKQogICAgZXhj"
    "ZXB0IChqc29uLkpTT05EZWNvZGVFcnJvciwgVW5pY29kZURlY29kZUVycm9yKSBhcyBlcnJv"
    "cjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBV"
    "VF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInRva2VuLW1hdGNoZWQgcmVzcG9u"
    "c2UgZW52ZWxvcGUgaXMgbm90IHZhbGlkIEpTT04iLAogICAgICAgICkgZnJvbSBlcnJvcgog"
    "ICAgaWYgbm90IGlzaW5zdGFuY2UoZW52ZWxvcGUsIGRpY3QpOgogICAgICAgIHJhaXNlIERp"
    "YWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAidG9rZW4tbWF0Y2hlZCByZXNwb25zZSBlbnZlbG9wZSByb290IGlz"
    "IGludmFsaWQiLAogICAgICAgICkKCiAgICB1c2FnZSA9IGVudmVsb3BlLmdldCgidXNhZ2Ui"
    "KQogICAgY2hvaWNlcyA9IGVudmVsb3BlLmdldCgiY2hvaWNlcyIpCiAgICBpZiBub3QgaXNp"
    "bnN0YW5jZSh1c2FnZSwgZGljdCk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJ0"
    "b2tlbi1tYXRjaGVkIHJlc3BvbnNlIHVzYWdlIGlzIG1pc3NpbmciLAogICAgICAgICkKICAg"
    "IGlmIG5vdCBpc2luc3RhbmNlKGNob2ljZXMsIGxpc3QpIG9yIGxlbihjaG9pY2VzKSAhPSAx"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVU"
    "X0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAidG9rZW4tbWF0Y2hlZCByZXNwb25z"
    "ZSBjaG9pY2VzIGFyZSBpbnZhbGlkIiwKICAgICAgICApCiAgICBjaG9pY2UgPSBjaG9pY2Vz"
    "WzBdCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2UsIGRpY3QpOgogICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAidG9rZW4tbWF0Y2hlZCByZXNwb25zZSBjaG9pY2UgaXMgaW52"
    "YWxpZCIsCiAgICAgICAgKQogICAgbWVzc2FnZSA9IGNob2ljZS5nZXQoIm1lc3NhZ2UiKQog"
    "ICAgaWYgbm90IGlzaW5zdGFuY2UobWVzc2FnZSwgZGljdCk6CiAgICAgICAgcmFpc2UgRGlh"
    "Z25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJ0b2tlbi1tYXRjaGVkIHJlc3BvbnNlIG1lc3NhZ2UgaXMgaW52YWxp"
    "ZCIsCiAgICAgICAgKQogICAgY29udGVudCA9IG1lc3NhZ2UuZ2V0KCJjb250ZW50IikKICAg"
    "IGlmIG5vdCBpc2luc3RhbmNlKGNvbnRlbnQsIHN0cik6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAg"
    "ICAgICAgICAgICJ0b2tlbi1tYXRjaGVkIHJlc3BvbnNlIGNvbnRlbnQgaXMgbm90IGEgc3Ry"
    "aW5nIiwKICAgICAgICApCgogICAgY29tcGxldGlvbl90b2tlbnMgPSB1c2FnZS5nZXQoImNv"
    "bXBsZXRpb25fdG9rZW5zIikKICAgIHByb21wdF90b2tlbnMgPSB1c2FnZS5nZXQoInByb21w"
    "dF90b2tlbnMiKQogICAgaWYgKAogICAgICAgIGlzaW5zdGFuY2UoY29tcGxldGlvbl90b2tl"
    "bnMsIGJvb2wpCiAgICAgICAgb3Igbm90IGlzaW5zdGFuY2UoY29tcGxldGlvbl90b2tlbnMs"
    "IGludCkKICAgICAgICBvciBub3QgMSA8PSBjb21wbGV0aW9uX3Rva2VucyA8PSAzMgogICAg"
    "KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBV"
    "VF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInRva2VuLW1hdGNoZWQgY29tcGxl"
    "dGlvbi10b2tlbiBidWRnZXQgZHJpZnRlZCIsCiAgICAgICAgKQogICAgaWYgKAogICAgICAg"
    "IGlzaW5zdGFuY2UocHJvbXB0X3Rva2VucywgYm9vbCkKICAgICAgICBvciBub3QgaXNpbnN0"
    "YW5jZShwcm9tcHRfdG9rZW5zLCBpbnQpCiAgICAgICAgb3IgcHJvbXB0X3Rva2VucyAhPSBU"
    "T0tFTl9NQVRDSEVEX1BST01QVF9UT0tFTl9DT1VOVAogICAgKToKICAgICAgICByYWlzZSBE"
    "aWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJF"
    "IiwKICAgICAgICAgICAgInRva2VuLW1hdGNoZWQgcmVzcG9uc2UgcHJvbXB0LXRva2VuIGNv"
    "dW50IGRyaWZ0ZWQiLAogICAgICAgICkKCiAgICBhZnRlciA9IHdvcmtlci5tZXRyaWNfc25h"
    "cHNob3QoKQogICAgZGVsdGEgPSBtZXRyaWNfZGVsdGEoYmVmb3JlLCBhZnRlcikKCiAgICBz"
    "dHJpcHBlZCA9IGNvbnRlbnQuc3RyaXAoKQogICAgZmlyc3QgPSBzdHJpcHBlZFswXSBpZiBz"
    "dHJpcHBlZCBlbHNlIE5vbmUKICAgIGxhc3QgPSBzdHJpcHBlZFstMV0gaWYgc3RyaXBwZWQg"
    "ZWxzZSBOb25lCiAgICB2YWxpZF9qc29uID0gRmFsc2UKICAgIGV4YWN0X29iamVjdCA9IEZh"
    "bHNlCiAgICBqc29uX2Vycm9yX2xpbmU6IGludCB8IE5vbmUgPSBOb25lCiAgICBqc29uX2Vy"
    "cm9yX2NvbHVtbjogaW50IHwgTm9uZSA9IE5vbmUKICAgIGpzb25fZXJyb3JfcG9zaXRpb246"
    "IGludCB8IE5vbmUgPSBOb25lCiAgICB0cnk6CiAgICAgICAgcGFyc2VkID0ganNvbi5sb2Fk"
    "cyhjb250ZW50KQogICAgICAgIGV4cGVjdGVkID0ganNvbi5sb2FkcyhFWFBFQ1RFRF9PQkpF"
    "Q1RfQ0FOT05JQ0FMKQogICAgICAgIHZhbGlkX2pzb24gPSBUcnVlCiAgICAgICAgZXhhY3Rf"
    "b2JqZWN0ID0gcGFyc2VkID09IGV4cGVjdGVkCiAgICBleGNlcHQganNvbi5KU09ORGVjb2Rl"
    "RXJyb3IgYXMgZXJyb3I6CiAgICAgICAganNvbl9lcnJvcl9saW5lID0gZXJyb3IubGluZW5v"
    "CiAgICAgICAganNvbl9lcnJvcl9jb2x1bW4gPSBlcnJvci5jb2xubwogICAgICAgIGpzb25f"
    "ZXJyb3JfcG9zaXRpb24gPSBlcnJvci5wb3MKCiAgICB3b3JrZXJfcmVwb3J0ID0gd29ya2Vy"
    "LnJlcG9ydCgpCiAgICBwaWQgPSB3b3JrZXJfcmVwb3J0LmdldCgicGlkIikKICAgIHByb2Nl"
    "c3Nfc3RhcnRfdGlja3NfdmFsdWUgPSB3b3JrZXJfcmVwb3J0LmdldCgicHJvY2Vzc19zdGFy"
    "dF90aWNrcyIpCiAgICBwcm9jZXNzX2lkZW50aXR5X3NoYTI1NiA9IHNoYTI1Nl90ZXh0KAog"
    "ICAgICAgIGNhbm9uaWNhbF9qc29uKAogICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAi"
    "cGlkIjogcGlkLAogICAgICAgICAgICAgICAgInByb2Nlc3Nfc3RhcnRfdGlja3MiOiBwcm9j"
    "ZXNzX3N0YXJ0X3RpY2tzX3ZhbHVlLAogICAgICAgICAgICB9CiAgICAgICAgKQogICAgKQoK"
    "ICAgIHJldHVybiB7CiAgICAgICAgImNvbmRpdGlvbl9pZCI6IGNvbmRpdGlvbl9pZCwKICAg"
    "ICAgICAic2VxdWVuY2VfaW5kZXgiOiBzZXF1ZW5jZV9pbmRleCwKICAgICAgICAic2VnbWVu"
    "dF9jb3VudCI6IDI0LAogICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiOiB3b3JrZXIuaW5z"
    "dGFuY2VfaWQsCiAgICAgICAgIndvcmtlcl9wcm9jZXNzX2lkZW50aXR5X3NoYTI1NiI6IHBy"
    "b2Nlc3NfaWRlbnRpdHlfc2hhMjU2LAogICAgICAgICJ0b2tlbl9jb3VudCI6IHRva2VuX2lk"
    "ZW50aXR5WyJ0b2tlbl9jb3VudCJdLAogICAgICAgICJ0b2tlbl9zaGEyNTYiOiB0b2tlbl9p"
    "ZGVudGl0eVsidG9rZW5fc2hhMjU2Il0sCiAgICAgICAgInBheWxvYWRfc2hhMjU2IjogcGF5"
    "bG9hZF9zaGEyNTYsCiAgICAgICAgInplcm9fY2FjaGVfYmFzZWxpbmUiOiBUcnVlLAogICAg"
    "ICAgICJiYXNlbGluZV9tZXRyaWNzIjogYmFzZWxpbmUsCiAgICAgICAgImh0dHBfc3RhdHVz"
    "Ijogc3RhdHVzX2NvZGUsCiAgICAgICAgInJlc3BvbnNlX3NoYTI1NiI6IHNoYTI1Nl90ZXh0"
    "KGNvbnRlbnQpLAogICAgICAgICJyZXNwb25zZV9sZW5ndGgiOiBsZW4oY29udGVudCksCiAg"
    "ICAgICAgImZpbmlzaF9yZWFzb24iOiBjaG9pY2UuZ2V0KCJmaW5pc2hfcmVhc29uIiksCiAg"
    "ICAgICAgInByb21wdF90b2tlbnMiOiBwcm9tcHRfdG9rZW5zLAogICAgICAgICJjb21wbGV0"
    "aW9uX3Rva2VucyI6IGNvbXBsZXRpb25fdG9rZW5zLAogICAgICAgICJ2YWxpZF9qc29uIjog"
    "dmFsaWRfanNvbiwKICAgICAgICAiZXhhY3Rfb2JqZWN0IjogZXhhY3Rfb2JqZWN0LAogICAg"
    "ICAgICJqc29uX2Vycm9yX2xpbmUiOiBqc29uX2Vycm9yX2xpbmUsCiAgICAgICAgImpzb25f"
    "ZXJyb3JfY29sdW1uIjoganNvbl9lcnJvcl9jb2x1bW4sCiAgICAgICAgImpzb25fZXJyb3Jf"
    "cG9zaXRpb24iOiBqc29uX2Vycm9yX3Bvc2l0aW9uLAogICAgICAgICJmaXJzdF9ub25fd2hp"
    "dGVzcGFjZV9jbGFzcyI6IHRva2VuX21hdGNoZWRfZWRnZV9jbGFzcyhmaXJzdCksCiAgICAg"
    "ICAgImxhc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiOiB0b2tlbl9tYXRjaGVkX2VkZ2VfY2xh"
    "c3MobGFzdCksCiAgICAgICAgIm1hcmtkb3duX2ZlbmNlX2RldGVjdGVkIjogImBgYCIgaW4g"
    "Y29udGVudCwKICAgICAgICAibWV0cmljX2RlbHRhIjogYXNkaWN0KGRlbHRhKSwKICAgICAg"
    "ICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICJyYXdfb3V0cHV0X3Jl"
    "dGFpbmVkIjogRmFsc2UsCiAgICB9CgoKZGVmIHRva2VuX21hdGNoZWRfcHVibGljX29ic2Vy"
    "dmF0aW9uKAogICAgb2JzZXJ2YXRpb246IGRpY3Rbc3RyLCBvYmplY3RdLAopIC0+IGRpY3Rb"
    "c3RyLCBvYmplY3RdOgogICAgcGVybWl0dGVkID0gKAogICAgICAgICJjb25kaXRpb25faWQi"
    "LAogICAgICAgICJzZXF1ZW5jZV9pbmRleCIsCiAgICAgICAgInNlZ21lbnRfY291bnQiLAog"
    "ICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiLAogICAgICAgICJ3b3JrZXJfcHJvY2Vzc19p"
    "ZGVudGl0eV9zaGEyNTYiLAogICAgICAgICJ0b2tlbl9jb3VudCIsCiAgICAgICAgInRva2Vu"
    "X3NoYTI1NiIsCiAgICAgICAgInBheWxvYWRfc2hhMjU2IiwKICAgICAgICAiemVyb19jYWNo"
    "ZV9iYXNlbGluZSIsCiAgICAgICAgImJhc2VsaW5lX21ldHJpY3MiLAogICAgICAgICJodHRw"
    "X3N0YXR1cyIsCiAgICAgICAgInJlc3BvbnNlX3NoYTI1NiIsCiAgICAgICAgInJlc3BvbnNl"
    "X2xlbmd0aCIsCiAgICAgICAgImZpbmlzaF9yZWFzb24iLAogICAgICAgICJwcm9tcHRfdG9r"
    "ZW5zIiwKICAgICAgICAiY29tcGxldGlvbl90b2tlbnMiLAogICAgICAgICJ2YWxpZF9qc29u"
    "IiwKICAgICAgICAiZXhhY3Rfb2JqZWN0IiwKICAgICAgICAianNvbl9lcnJvcl9saW5lIiwK"
    "ICAgICAgICAianNvbl9lcnJvcl9jb2x1bW4iLAogICAgICAgICJqc29uX2Vycm9yX3Bvc2l0"
    "aW9uIiwKICAgICAgICAiZmlyc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiLAogICAgICAgICJs"
    "YXN0X25vbl93aGl0ZXNwYWNlX2NsYXNzIiwKICAgICAgICAibWFya2Rvd25fZmVuY2VfZGV0"
    "ZWN0ZWQiLAogICAgICAgICJtZXRyaWNfZGVsdGEiLAogICAgICAgICJyYXdfcHJvbXB0X3Jl"
    "dGFpbmVkIiwKICAgICAgICAicmF3X291dHB1dF9yZXRhaW5lZCIsCiAgICApCiAgICByZXR1"
    "cm4ge25hbWU6IG9ic2VydmF0aW9uLmdldChuYW1lKSBmb3IgbmFtZSBpbiBwZXJtaXR0ZWR9"
    "CgoKZGVmIHJ1bl90b2tlbl9tYXRjaGVkX2ZyZXNoX3dvcmtlcl9vYnNlcnZhdGlvbigKICAg"
    "IG1vZGVsX2hvbWU6IFBhdGgsCiAgICBzbmFwc2hvdDogUGF0aCwKICAgIGNvbmRpdGlvbl9p"
    "ZDogc3RyLAogICAgc2VxdWVuY2VfaW5kZXg6IGludCwKICAgIGNvdW50ZXJzOiBkaWN0W3N0"
    "ciwgaW50XSwKICAgIHdvcmtlcl9yZXBvcnRzOiBsaXN0W2RpY3Rbc3RyLCBvYmplY3RdXSwK"
    "ICAgIHRlYXJkb3duX3JlcG9ydHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAopIC0+IGRp"
    "Y3Rbc3RyLCBvYmplY3RdOgogICAgd29ya2VyID0gV29ya2VyKAogICAgICAgICJvYnNlcnZh"
    "dGlvbl93b3JrZXIiLAogICAgICAgIDAsCiAgICAgICAgODAwMSwKICAgICAgICBtb2RlbF9o"
    "b21lLAogICAgICAgIHNuYXBzaG90LAogICAgICAgIGdlbmVyYXRpb249c2VxdWVuY2VfaW5k"
    "ZXgsCiAgICApCiAgICBwcmltYXJ5X2Vycm9yOiBFeGNlcHRpb24gfCBOb25lID0gTm9uZQog"
    "ICAgb2JzZXJ2YXRpb246IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZSA9IE5vbmUKICAgIHJl"
    "cG9ydDogZGljdFtzdHIsIG9iamVjdF0gfCBOb25lID0gTm9uZQogICAgdHJ5OgogICAgICAg"
    "IHdvcmtlci5zdGFydChjb3VudGVycykKICAgICAgICB3b3JrZXIud2FpdF9yZWFkeSgpCiAg"
    "ICAgICAgd29ya2VyLnZhbGlkYXRlX21vZGVsKCkKICAgICAgICB3b3JrZXIud2FpdF9iYWNr"
    "ZW5kX21hcmtlcigpCiAgICAgICAgbmF0aXZlX29yaWdpbnMgPSB2YWxpZGF0ZV9uYXRpdmVf"
    "b3JpZ2luX2Nsb3N1cmUod29ya2VyKQogICAgICAgIHJlcG9ydCA9IHsKICAgICAgICAgICAg"
    "Kip3b3JrZXIucmVwb3J0KCksCiAgICAgICAgICAgICJuYXRpdmVfb3JpZ2luX2Nsb3N1cmUi"
    "OiBuYXRpdmVfb3JpZ2lucywKICAgICAgICB9CiAgICAgICAgd29ya2VyX3JlcG9ydHMuYXBw"
    "ZW5kKHJlcG9ydCkKICAgICAgICBvYnNlcnZhdGlvbiA9IHJ1bl90b2tlbl9tYXRjaGVkX29i"
    "c2VydmF0aW9uKAogICAgICAgICAgICB3b3JrZXIsCiAgICAgICAgICAgIGNvbmRpdGlvbl9p"
    "ZCwKICAgICAgICAgICAgc2VxdWVuY2VfaW5kZXgsCiAgICAgICAgICAgIGNvdW50ZXJzLAog"
    "ICAgICAgICkKICAgIGV4Y2VwdCBFeGNlcHRpb24gYXMgZXJyb3I6CiAgICAgICAgcHJpbWFy"
    "eV9lcnJvciA9IGVycm9yCiAgICBmaW5hbGx5OgogICAgICAgIHRlYXJkb3duID0gc2FmZV93"
    "b3JrZXJfdGVhcmRvd24oCiAgICAgICAgICAgIHdvcmtlciwKICAgICAgICAgICAgZiJUT0tF"
    "Tl9NQVRDSEVEX09CU0VSVkFUSU9OX3tzZXF1ZW5jZV9pbmRleH1fVEVSTUlOQUwiLAogICAg"
    "ICAgICkKICAgICAgICB0ZWFyZG93bl9yZXBvcnRzLmFwcGVuZCh0ZWFyZG93bikKCiAgICB0"
    "ZWFyZG93bl9zdGF0dXMgPSB0ZWFyZG93bi5nZXQoInN0YXR1cyIpCiAgICBpZiB0ZWFyZG93"
    "bl9zdGF0dXMgbm90IGluIHsiUEFTU0VEIiwgIk5PVF9TVEFSVEVEIn06CiAgICAgICAgcmFp"
    "c2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJURUFSRE9XTl9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgImZyZXNoIG9ic2VydmF0aW9uIHdvcmtlciB0ZWFyZG93biBwcm9vZiBm"
    "YWlsZWQiLAogICAgICAgICkKICAgIGlmIHByaW1hcnlfZXJyb3IgaXMgbm90IE5vbmU6CiAg"
    "ICAgICAgcmFpc2UgcHJpbWFyeV9lcnJvcgogICAgaWYgdGVhcmRvd25fc3RhdHVzICE9ICJQ"
    "QVNTRUQiOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAi"
    "VEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAgICJjb21wbGV0ZWQgb2JzZXJ2YXRpb24g"
    "ZGlkIG5vdCBwcm92ZSB3b3JrZXIgdGVhcmRvd24iLAogICAgICAgICkKICAgIGlmIG9ic2Vy"
    "dmF0aW9uIGlzIE5vbmUgb3IgcmVwb3J0IGlzIE5vbmU6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAog"
    "ICAgICAgICAgICAiZnJlc2ggb2JzZXJ2YXRpb24gZGlkIG5vdCBwcm9kdWNlIGNvbXBsZXRl"
    "IGV2aWRlbmNlIiwKICAgICAgICApCiAgICByZXR1cm4gb2JzZXJ2YXRpb24KCgpkZWYgd3Jp"
    "dGVfdG9rZW5fbWF0Y2hlZF9yZXN1bHRzKAogICAgcmVzdWx0czogbGlzdFtkaWN0W3N0ciwg"
    "b2JqZWN0XV0sCiAgICBzdGF0dXM6IHN0ciwKKSAtPiBOb25lOgogICAgd3JpdGVfanNvbigK"
    "ICAgICAgICBPVVRQVVRfUk9PVCAvICJwNF9wNV90b2tlbl9tYXRjaGVkX3JlcXVlc3RfcmVz"
    "dWx0c192MS5qc29uIiwKICAgICAgICB7CiAgICAgICAgICAgICJzY2hlbWFfdmVyc2lvbiI6"
    "ICIxLjAuMCIsCiAgICAgICAgICAgICJzdGF0dXMiOiBzdGF0dXMsCiAgICAgICAgICAgICJz"
    "Y2hlZHVsZWRfcmVxdWVzdF9jb3VudCI6IGxlbihUT0tFTl9NQVRDSEVEX1JFUVVFU1RfT1JE"
    "RVIpLAogICAgICAgICAgICAib2JzZXJ2ZWRfcmVxdWVzdF9jb3VudCI6IGxlbihyZXN1bHRz"
    "KSwKICAgICAgICAgICAgInJlcXVlc3Rfb3JkZXIiOiBsaXN0KFRPS0VOX01BVENIRURfUkVR"
    "VUVTVF9PUkRFUiksCiAgICAgICAgICAgICJyZXN1bHRzIjogW3Rva2VuX21hdGNoZWRfcHVi"
    "bGljX29ic2VydmF0aW9uKGl0ZW0pIGZvciBpdGVtIGluIHJlc3VsdHNdLAogICAgICAgICAg"
    "ICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICAgICAicmF3X291dHB1"
    "dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgIH0sCiAgICApCgoKZGVmIHRva2VuX21hdGNo"
    "ZWRfY29uZGl0aW9uX3N0YXR1cyhleGFjdF9jb3VudDogaW50KSAtPiBzdHI6CiAgICBpZiBl"
    "eGFjdF9jb3VudCA9PSAzOgogICAgICAgIHJldHVybiAiM19PRl8zX0VYQUNUX09CSkVDVF9U"
    "UlVFIgogICAgaWYgZXhhY3RfY291bnQgPT0gMDoKICAgICAgICByZXR1cm4gIjBfT0ZfM19F"
    "WEFDVF9PQkpFQ1RfVFJVRSIKICAgIGlmIGV4YWN0X2NvdW50IGluIHsxLCAyfToKICAgICAg"
    "ICByZXR1cm4gIjFfT1JfMl9PRl8zX0VYQUNUX09CSkVDVF9UUlVFIgogICAgcmFpc2UgRGlh"
    "Z25vc3RpY0ZhaWx1cmUoCiAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAg"
    "ICAgICAgImNvbmRpdGlvbiBleGFjdC1vYmplY3QgY291bnQgaXMgb3V0c2lkZSB0aGUgZnJv"
    "emVuIGVuZHBvaW50IGNvbnRyYWN0IiwKICAgICkKCgpkZWYgZGVjaWRlX3Rva2VuX21hdGNo"
    "ZWRfZGlmZmVyZW50aWFsKAogICAgcmVzdWx0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0s"
    "CiAgICB3b3JrZXJfcmVwb3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0sCiAgICB0ZWFy"
    "ZG93bl9yZXBvcnRzOiBsaXN0W2RpY3Rbc3RyLCBvYmplY3RdXSwKICAgIGNvdW50ZXJzOiBk"
    "aWN0W3N0ciwgaW50XSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIGxlbihyZXN1"
    "bHRzKSAhPSA5OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAg"
    "ICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgInRva2Vu"
    "LW1hdGNoZWQgZGlmZmVyZW50aWFsIHJlc3VsdCBjb3VudCBkcmlmdGVkIiwKICAgICAgICAp"
    "CiAgICBvYnNlcnZlZF9vcmRlciA9IHR1cGxlKHN0cihyb3cuZ2V0KCJjb25kaXRpb25faWQi"
    "KSkgZm9yIHJvdyBpbiByZXN1bHRzKQogICAgaWYgb2JzZXJ2ZWRfb3JkZXIgIT0gVE9LRU5f"
    "TUFUQ0hFRF9SRVFVRVNUX09SREVSOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJl"
    "KAogICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAg"
    "ICAgICAgInRva2VuLW1hdGNoZWQgZGlmZmVyZW50aWFsIGNocm9ub2xvZ3kgZHJpZnRlZCIs"
    "CiAgICAgICAgKQogICAgb2JzZXJ2ZWRfaW5kZXhlcyA9IHR1cGxlKHJvdy5nZXQoInNlcXVl"
    "bmNlX2luZGV4IikgZm9yIHJvdyBpbiByZXN1bHRzKQogICAgaWYgb2JzZXJ2ZWRfaW5kZXhl"
    "cyAhPSB0dXBsZShyYW5nZSgxLCAxMCkpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWls"
    "dXJlKAogICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAg"
    "ICAgICAgICAgInRva2VuLW1hdGNoZWQgZGlmZmVyZW50aWFsIHNlcXVlbmNlIGluZGV4ZXMg"
    "ZHJpZnRlZCIsCiAgICAgICAgKQogICAgaWYgbGVuKHdvcmtlcl9yZXBvcnRzKSAhPSA5IG9y"
    "IGxlbih0ZWFyZG93bl9yZXBvcnRzKSAhPSA5OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNG"
    "YWlsdXJlKAogICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgImZyZXNoLXdvcmtlciBldmlkZW5jZSBjYXJkaW5hbGl0eSBkcmlmdGVk"
    "IiwKICAgICAgICApCiAgICBpZiBhbnkoaXRlbS5nZXQoInN0YXR1cyIpICE9ICJQQVNTRUQi"
    "IGZvciBpdGVtIGluIHRlYXJkb3duX3JlcG9ydHMpOgogICAgICAgIHJhaXNlIERpYWdub3N0"
    "aWNGYWlsdXJlKAogICAgICAgICAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJvbmUgb3IgbW9yZSBvYnNlcnZhdGlvbiB0ZWFyZG93bnMgZmFpbGVkIiwKICAgICAgICAp"
    "CiAgICB3b3JrZXJfaWRlbnRpdGllcyA9IHsKICAgICAgICBzdHIocm93LmdldCgid29ya2Vy"
    "X3Byb2Nlc3NfaWRlbnRpdHlfc2hhMjU2IikpCiAgICAgICAgZm9yIHJvdyBpbiByZXN1bHRz"
    "CiAgICAgICAgaWYgaXNpbnN0YW5jZShyb3cuZ2V0KCJ3b3JrZXJfcHJvY2Vzc19pZGVudGl0"
    "eV9zaGEyNTYiKSwgc3RyKQogICAgfQogICAgaWYgbGVuKHdvcmtlcl9pZGVudGl0aWVzKSAh"
    "PSA5OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUDVf"
    "U1RBUlRJTkdfU1RBVEVfRkFJTFVSRSIsCiAgICAgICAgICAgICJmcmVzaCB3b3JrZXIgcHJv"
    "Y2VzcyBpZGVudGl0eSB3YXMgcmV1c2VkIGFjcm9zcyBvYnNlcnZhdGlvbnMiLAogICAgICAg"
    "ICkKICAgIGlmIGFueShyb3cuZ2V0KCJ6ZXJvX2NhY2hlX2Jhc2VsaW5lIikgaXMgbm90IFRy"
    "dWUgZm9yIHJvdyBpbiByZXN1bHRzKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVy"
    "ZSgKICAgICAgICAgICAgIlA1X1NUQVJUSU5HX1NUQVRFX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAib25lIG9yIG1vcmUgb2JzZXJ2YXRpb25zIGxhY2tlZCBhIHplcm8gY2FjaGUgYmFzZWxp"
    "bmUiLAogICAgICAgICkKCiAgICBleHBlY3RlZF9jb3VudGVycyA9IHsKICAgICAgICAibW9k"
    "ZWxfcmVxdWVzdHMiOiA5LAogICAgICAgICJtb2RlbF9sb2FkcyI6IDksCiAgICAgICAgIndv"
    "cmtlcl9zdGFydHMiOiA5LAogICAgICAgICJoaWRkZW5fcmV0cmllcyI6IDAsCiAgICAgICAg"
    "Im5ldHdvcmtfcmVxdWVzdHMiOiAwLAogICAgICAgICJiZW5jaG1hcmtfdHJhamVjdG9yeV9y"
    "ZXF1ZXN0cyI6IDAsCiAgICAgICAgImV4dGVybmFsX3NwZW5kIjogMCwKICAgIH0KICAgIGZv"
    "ciBuYW1lLCBleHBlY3RlZCBpbiBleHBlY3RlZF9jb3VudGVycy5pdGVtcygpOgogICAgICAg"
    "IGlmIGNvdW50ZXJzLmdldChuYW1lKSAhPSBleHBlY3RlZDoKICAgICAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElB"
    "VElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgICAgIGYie25hbWV9IGV4cGVjdGVkIHtleHBl"
    "Y3RlZH0sIG9ic2VydmVkIHtjb3VudGVycy5nZXQobmFtZSl9IiwKICAgICAgICAgICAgKQoK"
    "ICAgIGNvbmRpdGlvbl9yb3dzID0gewogICAgICAgIGNvbmRpdGlvbl9pZDogW3JvdyBmb3Ig"
    "cm93IGluIHJlc3VsdHMgaWYgcm93LmdldCgiY29uZGl0aW9uX2lkIikgPT0gY29uZGl0aW9u"
    "X2lkXQogICAgICAgIGZvciBjb25kaXRpb25faWQgaW4gKAogICAgICAgICAgICAiQV9PUklH"
    "SU5BTF8yNFhfQU5DSE9SIiwKICAgICAgICAgICAgIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgi"
    "LAogICAgICAgICAgICAiQ19ORVVUUkFMX0RJVkVSU0VfMjRfU0VHTUVOVCIsCiAgICAgICAg"
    "KQogICAgfQogICAgaWYgYW55KGxlbihyb3dzKSAhPSAzIGZvciByb3dzIGluIGNvbmRpdGlv"
    "bl9yb3dzLnZhbHVlcygpKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJ0b2tlbi1tYXRjaGVkIGNvbmRpdGlvbiBjYXJkaW5hbGl0eSBkcmlmdGVkIiwKICAgICAg"
    "ICApCgogICAgZXhwZWN0ZWRfaWRlbnRpdGllcyA9IHsKICAgICAgICAiQV9PUklHSU5BTF8y"
    "NFhfQU5DSE9SIjogKAogICAgICAgICAgICBUT0tFTl9NQVRDSEVEX0FfVE9LRU5fU0hBMjU2"
    "LAogICAgICAgICAgICBUT0tFTl9NQVRDSEVEX0FfUEFZTE9BRF9TSEEyNTYsCiAgICAgICAg"
    "KSwKICAgICAgICAiQl9ORVVUUkFMX1JFUEVBVEVEXzI0WCI6ICgKICAgICAgICAgICAgVE9L"
    "RU5fTUFUQ0hFRF9CX1RPS0VOX1NIQTI1NiwKICAgICAgICAgICAgVE9LRU5fTUFUQ0hFRF9C"
    "X1BBWUxPQURfU0hBMjU2LAogICAgICAgICksCiAgICAgICAgIkNfTkVVVFJBTF9ESVZFUlNF"
    "XzI0X1NFR01FTlQiOiAoCiAgICAgICAgICAgIFRPS0VOX01BVENIRURfQ19UT0tFTl9TSEEy"
    "NTYsCiAgICAgICAgICAgIFRPS0VOX01BVENIRURfQ19QQVlMT0FEX1NIQTI1NiwKICAgICAg"
    "ICApLAogICAgfQogICAgZm9yIGNvbmRpdGlvbl9pZCwgcm93cyBpbiBjb25kaXRpb25fcm93"
    "cy5pdGVtcygpOgogICAgICAgIHRva2VuX3BhaXJzID0geyhyb3cuZ2V0KCJ0b2tlbl9jb3Vu"
    "dCIpLCByb3cuZ2V0KCJ0b2tlbl9zaGEyNTYiKSkgZm9yIHJvdyBpbiByb3dzfQogICAgICAg"
    "IHBheWxvYWRzID0ge3Jvdy5nZXQoInBheWxvYWRfc2hhMjU2IikgZm9yIHJvdyBpbiByb3dz"
    "fQogICAgICAgIGV4cGVjdGVkX3Rva2VuX3NoYTI1NiwgZXhwZWN0ZWRfcGF5bG9hZF9zaGEy"
    "NTYgPSBleHBlY3RlZF9pZGVudGl0aWVzW2NvbmRpdGlvbl9pZF0KICAgICAgICBpZiB0b2tl"
    "bl9wYWlycyAhPSB7KFRPS0VOX01BVENIRURfUFJPTVBUX1RPS0VOX0NPVU5ULCBleHBlY3Rl"
    "ZF90b2tlbl9zaGEyNTYpfToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgICAgICJ0b2tlbi1tYXRjaGVkIGNvbmRpdGlvbiB0b2tlbiBpZGVudGl0eSBmYWlsZWQg"
    "cmVjb25jaWxpYXRpb24iLAogICAgICAgICAgICApCiAgICAgICAgaWYgcGF5bG9hZHMgIT0g"
    "e2V4cGVjdGVkX3BheWxvYWRfc2hhMjU2fToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgICAgICJ0b2tlbi1tYXRjaGVkIGNvbmRpdGlvbiBwYXlsb2FkIGlkZW50"
    "aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgICAgICkKCiAgICBleGFjdF9j"
    "b3VudHMgPSB7CiAgICAgICAgY29uZGl0aW9uX2lkOiBzdW0ocm93LmdldCgiZXhhY3Rfb2Jq"
    "ZWN0IikgaXMgVHJ1ZSBmb3Igcm93IGluIHJvd3MpCiAgICAgICAgZm9yIGNvbmRpdGlvbl9p"
    "ZCwgcm93cyBpbiBjb25kaXRpb25fcm93cy5pdGVtcygpCiAgICB9CiAgICB2YWxpZF9qc29u"
    "X2NvdW50cyA9IHsKICAgICAgICBjb25kaXRpb25faWQ6IHN1bShyb3cuZ2V0KCJ2YWxpZF9q"
    "c29uIikgaXMgVHJ1ZSBmb3Igcm93IGluIHJvd3MpCiAgICAgICAgZm9yIGNvbmRpdGlvbl9p"
    "ZCwgcm93cyBpbiBjb25kaXRpb25fcm93cy5pdGVtcygpCiAgICB9CiAgICBjb25kaXRpb25f"
    "c3RhdHVzZXMgPSB7CiAgICAgICAgY29uZGl0aW9uX2lkOiB0b2tlbl9tYXRjaGVkX2NvbmRp"
    "dGlvbl9zdGF0dXMoZXhhY3RfY291bnQpCiAgICAgICAgZm9yIGNvbmRpdGlvbl9pZCwgZXhh"
    "Y3RfY291bnQgaW4gZXhhY3RfY291bnRzLml0ZW1zKCkKICAgIH0KCiAgICBhX2V4YWN0ID0g"
    "ZXhhY3RfY291bnRzWyJBX09SSUdJTkFMXzI0WF9BTkNIT1IiXQogICAgYl9leGFjdCA9IGV4"
    "YWN0X2NvdW50c1siQl9ORVVUUkFMX1JFUEVBVEVEXzI0WCJdCiAgICBjX2V4YWN0ID0gZXhh"
    "Y3RfY291bnRzWyJDX05FVVRSQUxfRElWRVJTRV8yNF9TRUdNRU5UIl0KCiAgICBpZiBhbnko"
    "dmFsdWUgaW4gezEsIDJ9IGZvciB2YWx1ZSBpbiBleGFjdF9jb3VudHMudmFsdWVzKCkpOgog"
    "ICAgICAgIHN0YXRlID0gIlVOU1RBQkxFX05PX01FQ0hBTklTVElDX0NMQUlNIgogICAgZWxp"
    "ZiBhX2V4YWN0ICE9IDA6CiAgICAgICAgc3RhdGUgPSAiQU5DSE9SX05PTlJFUFJPRFVDVElP"
    "Tl9JTlZBTElEQVRFU19NRUNIQU5JU1RJQ19JTkZFUkVOQ0UiCiAgICBlbGlmIGJfZXhhY3Qg"
    "PT0gMyBhbmQgY19leGFjdCA9PSAzOgogICAgICAgIHN0YXRlID0gIlJFUEVBVEVEX0lOU1RS"
    "VUNUSU9OX0xJS0VfU0VNQU5USUNfQU1QTElGSUNBVElPTl9TVFJPTkdMWV9JTVBMSUNBVEVE"
    "IgogICAgZWxpZiBiX2V4YWN0ID09IDAgYW5kIGNfZXhhY3QgPT0gMzoKICAgICAgICBzdGF0"
    "ZSA9ICJISUdIX0VYQUNUX1RPS0VOX1BBVFRFUk5fUkVQRVRJVElPTl9TVFJPTkdMWV9JTVBM"
    "SUNBVEVEIgogICAgZWxpZiBiX2V4YWN0ID09IDAgYW5kIGNfZXhhY3QgPT0gMDoKICAgICAg"
    "ICBzdGF0ZSA9ICJTSEFSRURfTE9OR19DT05URVhUX0ZBQ1RPUl9SRU1BSU5TX0xJVkUiCiAg"
    "ICBlbGlmIGJfZXhhY3QgPT0gMyBhbmQgY19leGFjdCA9PSAwOgogICAgICAgIHN0YXRlID0g"
    "IkRJVkVSU0VfQ09NUEFSQVRPUl9TUEVDSUZJQ19FRkZFQ1RfT0JTRVJWRUQiCiAgICBlbHNl"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFSTkVT"
    "U19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgIm9ic2VydmVkIGJlaGF2aW9yYWwg"
    "Y29tYmluYXRpb24gaXMgbm90IGNvdmVyZWQgYnkgdGhlIGZyb3plbiBkZWNpc2lvbiBjb250"
    "cmFjdCIsCiAgICAgICAgKQoKICAgIHJldHVybiB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9u"
    "IjogIjEuMC4wIiwKICAgICAgICAic3RhdHVzIjogIkRFQ0lERUQiLAogICAgICAgICJkZWNp"
    "c2lvbl9zdGF0ZSI6IHN0YXRlLAogICAgICAgICJwcmltYXJ5X2VuZHBvaW50IjogImV4YWN0"
    "X29iamVjdCIsCiAgICAgICAgImNvbmRpdGlvbl9leGFjdF9vYmplY3RfY291bnRzIjogZXhh"
    "Y3RfY291bnRzLAogICAgICAgICJjb25kaXRpb25fdmFsaWRfanNvbl9jb3VudHMiOiB2YWxp"
    "ZF9qc29uX2NvdW50cywKICAgICAgICAiY29uZGl0aW9uX2VuZHBvaW50X3N0YXR1c2VzIjog"
    "Y29uZGl0aW9uX3N0YXR1c2VzLAogICAgICAgICJhbmNob3JfcmVxdWlyZWRfZXhhY3Rfb2Jq"
    "ZWN0X2NvdW50IjogMCwKICAgICAgICAiYW5jaG9yX3JlcHJvZHVjZWQiOiBhX2V4YWN0ID09"
    "IDAsCiAgICAgICAgIm1lY2hhbmlzdGljX2luZmVyZW5jZV9wZXJtaXR0ZWQiOiAoCiAgICAg"
    "ICAgICAgIGFfZXhhY3QgPT0gMCBhbmQgYWxsKHZhbHVlIG5vdCBpbiB7MSwgMn0gZm9yIHZh"
    "bHVlIGluIGV4YWN0X2NvdW50cy52YWx1ZXMoKSkKICAgICAgICApLAogICAgICAgICJmcmVz"
    "aF93b3JrZXJfcHJvY2Vzc19wZXJfb2JzZXJ2YXRpb24iOiBUcnVlLAogICAgICAgICJ3b3Jr"
    "ZXJfaWRlbnRpdHlfY2FyZGluYWxpdHkiOiBsZW4od29ya2VyX2lkZW50aXRpZXMpLAogICAg"
    "ICAgICJhbGxfY29uZGl0aW9uX3Rva2VuX2lkZW50aXRpZXNfbWF0Y2hlZCI6IFRydWUsCiAg"
    "ICAgICAgImFsbF9jb25kaXRpb25fcGF5bG9hZF9pZGVudGl0aWVzX21hdGNoZWQiOiBUcnVl"
    "LAogICAgICAgICJiX3RvX2NfcmVzaWR1YWxfbGV4aWNhbF9ub3ZlbHR5X2JvdW5kZWQiOiBU"
    "cnVlLAogICAgICAgICJjYWNoZV90ZWxlbWV0cnlfaXNfZGlhZ25vc3RpY19vbmx5IjogVHJ1"
    "ZSwKICAgICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICJyYXdf"
    "b3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICB9CgoKZGVmIHRva2VuX21hdGNoZWRfZmFp"
    "bHVyZV9yZWNvcmQoCiAgICBlcnJvcjogRXhjZXB0aW9uLAogICAgYWN0aXZlX2ZhaWx1cmVf"
    "Y29kZTogc3RyLAogICAgZmFpbGVkX3N0YWdlOiBzdHIsCiAgICBjb21wbGV0ZWRfcmVxdWVz"
    "dHM6IGludCwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIGlzaW5zdGFuY2UoZXJy"
    "b3IsIERpYWdub3N0aWNBbWJpZ3VpdHkpOgogICAgICAgIGZhaWx1cmVfY2xhc3MgPSBlcnJv"
    "ci5mYWlsdXJlX2NsYXNzCiAgICAgICAgZGV0YWlsX2NvZGUgPSBOb25lCiAgICAgICAgc2Fm"
    "ZV9tZXNzYWdlID0gZXJyb3Iuc2FmZV9tZXNzYWdlCiAgICBlbGlmIGlzaW5zdGFuY2UoZXJy"
    "b3IsIERpYWdub3N0aWNGYWlsdXJlKToKICAgICAgICBkZXRhaWxfY29kZSA9IGVycm9yLmVy"
    "cm9yX2NvZGUKICAgICAgICBmYWlsdXJlX2NsYXNzID0gY2xhc3NpZnlfZmFpbHVyZV9kZXRh"
    "aWwoZGV0YWlsX2NvZGUpCiAgICAgICAgc2FmZV9tZXNzYWdlID0gZXJyb3Iuc2FmZV9tZXNz"
    "YWdlCiAgICBlbHNlOgogICAgICAgIGRldGFpbF9jb2RlID0gYWN0aXZlX2ZhaWx1cmVfY29k"
    "ZQogICAgICAgIGZhaWx1cmVfY2xhc3MgPSBjbGFzc2lmeV9mYWlsdXJlX2RldGFpbChkZXRh"
    "aWxfY29kZSkKICAgICAgICBzYWZlX21lc3NhZ2UgPSBzYW5pdGl6ZV9leGNlcnB0KHN0cihl"
    "cnJvcikpWzo1MTJdIG9yIHR5cGUoZXJyb3IpLl9fbmFtZV9fCiAgICByZXR1cm4gewogICAg"
    "ICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgInN0YXR1cyI6ICJESUFH"
    "Tk9TVElDX0lOVkFMSUQiLAogICAgICAgICJmYWlsZWRfc3RhZ2UiOiBmYWlsZWRfc3RhZ2Us"
    "CiAgICAgICAgImNvbXBsZXRlZF9yZXF1ZXN0cyI6IGNvbXBsZXRlZF9yZXF1ZXN0cywKICAg"
    "ICAgICAiZmFpbHVyZV9jbGFzcyI6IGZhaWx1cmVfY2xhc3MsCiAgICAgICAgImRldGFpbF9j"
    "b2RlIjogZGV0YWlsX2NvZGUsCiAgICAgICAgImVycm9yX3R5cGUiOiB0eXBlKGVycm9yKS5f"
    "X25hbWVfXywKICAgICAgICAic2FmZV9tZXNzYWdlIjogc2FmZV9tZXNzYWdlLAogICAgfQoK"
    "CmRlZiB0b2tlbl9tYXRjaGVkX2J1bmRsZV9vdXRwdXRzKCkgLT4gZGljdFtzdHIsIG9iamVj"
    "dF06CiAgICByZXF1aXJlZF9iZWZvcmVfbWFuaWZlc3QgPSBzZXQoVE9LRU5fTUFUQ0hFRF9P"
    "VVRQVVRfTkFNRVMpIC0geyJidW5kbGVfbWFuaWZlc3RfdjEuanNvbiJ9CiAgICBvYnNlcnZl"
    "ZF9iZWZvcmVfbWFuaWZlc3QgPSB7cGF0aC5uYW1lIGZvciBwYXRoIGluIE9VVFBVVF9ST09U"
    "Lml0ZXJkaXIoKSBpZiBwYXRoLmlzX2ZpbGUoKX0KICAgIG1pc3NpbmcgPSByZXF1aXJlZF9i"
    "ZWZvcmVfbWFuaWZlc3QgLSBvYnNlcnZlZF9iZWZvcmVfbWFuaWZlc3QKICAgIHVuZXhwZWN0"
    "ZWQgPSBvYnNlcnZlZF9iZWZvcmVfbWFuaWZlc3QgLSByZXF1aXJlZF9iZWZvcmVfbWFuaWZl"
    "c3QKICAgIGlmIG1pc3Npbmcgb3IgdW5leHBlY3RlZDoKICAgICAgICByYWlzZSBSdW50aW1l"
    "RXJyb3IoCiAgICAgICAgICAgICJ0b2tlbi1tYXRjaGVkIGV2aWRlbmNlIG91dHB1dCBjb250"
    "cmFjdCBkcmlmdGVkOiAiCiAgICAgICAgICAgICsgY2Fub25pY2FsX2pzb24oCiAgICAgICAg"
    "ICAgICAgICB7CiAgICAgICAgICAgICAgICAgICAgIm1pc3NpbmciOiBzb3J0ZWQobWlzc2lu"
    "ZyksCiAgICAgICAgICAgICAgICAgICAgInVuZXhwZWN0ZWQiOiBzb3J0ZWQodW5leHBlY3Rl"
    "ZCksCiAgICAgICAgICAgICAgICB9CiAgICAgICAgICAgICkKICAgICAgICApCiAgICBlbnRy"
    "aWVzID0gW10KICAgIGZvciBuYW1lIGluIFRPS0VOX01BVENIRURfT1VUUFVUX05BTUVTOgog"
    "ICAgICAgIHBhdGggPSBPVVRQVVRfUk9PVCAvIG5hbWUKICAgICAgICBpZiBub3QgcGF0aC5p"
    "c19maWxlKCk6CiAgICAgICAgICAgIGNvbnRpbnVlCiAgICAgICAgZW50cmllcy5hcHBlbmQo"
    "CiAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICJwYXRoIjogbmFtZSwKICAgICAgICAg"
    "ICAgICAgICJzaGEyNTYiOiBmaWxlX3NoYTI1NihwYXRoKSwKICAgICAgICAgICAgICAgICJz"
    "aXplX2J5dGVzIjogcGF0aC5zdGF0KCkuc3Rfc2l6ZSwKICAgICAgICAgICAgfQogICAgICAg"
    "ICkKICAgIG1hbmlmZXN0ID0gewogICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIs"
    "CiAgICAgICAgImRpYWdub3N0aWNfaWQiOiAoCiAgICAgICAgICAgICJhdXJhZ2F0ZXdheS1w"
    "NC1wNS10b2tlbi1jb3VudC1tYXRjaGVkLWNvbnRleHQtc3RydWN0dXJlLWRpZmZlcmVudGlh"
    "bC12MSIKICAgICAgICApLAogICAgICAgICJzb3VyY2VfbWFpbl9jb21taXQiOiBTT1VSQ0Vf"
    "TUFJTl9DT01NSVQsCiAgICAgICAgIm1lbWJlcnMiOiBbaXRlbSBmb3IgaXRlbSBpbiBlbnRy"
    "aWVzIGlmIGl0ZW1bInBhdGgiXSAhPSAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iXSwKICAg"
    "ICAgICAic2NyYXRjaF9kaXJlY3Rvcmllc19pbmNsdWRlZCI6IEZhbHNlLAogICAgICAgICJ3"
    "b3JrZXJfbG9nX2RpcmVjdG9yeV9pbmNsdWRlZCI6IEZhbHNlLAogICAgICAgICJyYXdfcHJv"
    "bXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgInJhd19vdXRwdXRfcmV0YWluZWQiOiBG"
    "YWxzZSwKICAgIH0KICAgIHdyaXRlX2pzb24oT1VUUFVUX1JPT1QgLyAiYnVuZGxlX21hbmlm"
    "ZXN0X3YxLmpzb24iLCBtYW5pZmVzdCkKICAgIHdpdGggemlwZmlsZS5aaXBGaWxlKAogICAg"
    "ICAgIEVWSURFTkNFX1pJUCwKICAgICAgICAidyIsCiAgICAgICAgY29tcHJlc3Npb249emlw"
    "ZmlsZS5aSVBfREVGTEFURUQsCiAgICAgICAgY29tcHJlc3NsZXZlbD05LAogICAgKSBhcyBh"
    "cmNoaXZlOgogICAgICAgIGZvciBuYW1lIGluIFRPS0VOX01BVENIRURfT1VUUFVUX05BTUVT"
    "OgogICAgICAgICAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyBuYW1lCiAgICAgICAgICAgIGlm"
    "IHBhdGguaXNfZmlsZSgpOgogICAgICAgICAgICAgICAgYXJjaGl2ZS53cml0ZShwYXRoLCBh"
    "cmNuYW1lPW5hbWUpCiAgICBpZiBFVklERU5DRV9aSVAuc3RhdCgpLnN0X3NpemUgPiBNQVhf"
    "RVZJREVOQ0VfWklQX0JZVEVTOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigidG9rZW4t"
    "bWF0Y2hlZCBldmlkZW5jZSBaSVAgZXhjZWVkcyB0aGUgYnl0ZSBidWRnZXQiKQogICAgcmV0"
    "dXJuIHsKICAgICAgICAiZXZpZGVuY2VfemlwIjogc3RyKEVWSURFTkNFX1pJUCksCiAgICAg"
    "ICAgImV2aWRlbmNlX3ppcF9zaGEyNTYiOiBmaWxlX3NoYTI1NihFVklERU5DRV9aSVApLAog"
    "ICAgICAgICJldmlkZW5jZV96aXBfc2l6ZV9ieXRlcyI6IEVWSURFTkNFX1pJUC5zdGF0KCku"
    "c3Rfc2l6ZSwKICAgIH0KCgpCX1ZTX0RfSU1QTEVNRU5UQVRJT05fQkFTRV9DT01NSVQ6IEZp"
    "bmFsID0gImNmZTM2NDA3ZmU3ZDdiMWM3MTkzOGQyMTg4ODY4NzJlNDdhMWJlMzkiCkJfVlNf"
    "RF9ERVNJR05fUkVDT1JEX1NIQTI1NjogRmluYWwgPSAoCiAgICAiMmUwNzY1MTY4MWQ5OGQ2"
    "MDRmMGUwZjZiNGU4OTY0OTA2ZjM5YjhiZmEwZTQ4YjhmOGZhOGU5ZGU0MzFlN2VmOSIKKQpC"
    "X1ZTX0RfQl9DT05ESVRJT046IEZpbmFsID0gIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgiCkJf"
    "VlNfRF9EX0NPTkRJVElPTjogRmluYWwgPSAiRF9ORVVUUkFMX01BUktFUl9ESVZFUlNJRklF"
    "RF8yNFhfQ1VNVUxBVElWRV9MRU5HVEhfTE9DS0VEIgpCX1ZTX0RfUkVRVUVTVF9PUkRFUjog"
    "RmluYWwgPSAoCiAgICAiQl9ORVVUUkFMX1JFUEVBVEVEXzI0WCIsCiAgICAiRF9ORVVUUkFM"
    "X01BUktFUl9ESVZFUlNJRklFRF8yNFhfQ1VNVUxBVElWRV9MRU5HVEhfTE9DS0VEIiwKICAg"
    "ICJEX05FVVRSQUxfTUFSS0VSX0RJVkVSU0lGSUVEXzI0WF9DVU1VTEFUSVZFX0xFTkdUSF9M"
    "T0NLRUQiLAogICAgIkJfTkVVVFJBTF9SRVBFQVRFRF8yNFgiLAogICAgIkJfTkVVVFJBTF9S"
    "RVBFQVRFRF8yNFgiLAogICAgIkRfTkVVVFJBTF9NQVJLRVJfRElWRVJTSUZJRURfMjRYX0NV"
    "TVVMQVRJVkVfTEVOR1RIX0xPQ0tFRCIsCikKQl9WU19EX1BST01QVF9UT0tFTl9DT1VOVDog"
    "RmluYWwgPSA4OTkKQl9WU19EX0JfVE9LRU5fU0hBMjU2OiBGaW5hbCA9ICIwMmYyNjc1YTA0"
    "OTBkMTZlM2EzOWRlOTYxOWFlODY1YThmNzMwMjRiMjZjNmY5MTI2YmY0ZGVhMTk3ZDk5ZjY4"
    "IgpCX1ZTX0RfQl9QQVlMT0FEX1NIQTI1NjogRmluYWwgPSAiMWMxY2NhYWQwN2Q3ZjgzZWNh"
    "M2M3OWFlMDE1ZDIzMWRiZThmM2RhN2Q2YjA1NWVjMTBkYTYwNzAzNzhjNGVmYiIKQl9WU19E"
    "X0RfVE9LRU5fU0hBMjU2OiBGaW5hbCA9ICI4NzhlY2MwNTdmYmM5Mjc2NGM3YjhiZGRjMzAy"
    "NGUxMjcyMDQ3MGI4NGE3MmQ5NzRlZjY3N2MxNmQxZTM3ZTIxIgpCX1ZTX0RfRF9QQVlMT0FE"
    "X1NIQTI1NjogRmluYWwgPSAiMDcyOGU4NjMyZTQ2OTRjZDY3MGU0NzI3NTExNTRkMzhkY2Fj"
    "YzM0MDcxZDc0ZTFjYWFkOGVjZTY2MDhjODAxMCIKQl9WU19EX0JfU0VHTUVOVFM6IEZpbmFs"
    "ID0gKAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJr"
    "ZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5"
    "bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRl"
    "c3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNv"
    "bnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93"
    "LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBt"
    "YXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4"
    "dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9y"
    "IGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAg"
    "ICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBh"
    "c3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAg"
    "ICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xs"
    "ZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0"
    "ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29u"
    "dGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9t"
    "ZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHku"
    "IEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5p"
    "emVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50"
    "aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1"
    "dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwg"
    "YW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3Zp"
    "ZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5"
    "c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29u"
    "dGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAg"
    "ICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2Ny"
    "aXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2Ny"
    "aXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAg"
    "ICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJr"
    "ZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5"
    "bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRl"
    "c3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNv"
    "bnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93"
    "LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBt"
    "YXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4"
    "dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9y"
    "IGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAg"
    "ICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBh"
    "c3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAg"
    "ICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xs"
    "ZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0"
    "ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29u"
    "dGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9t"
    "ZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHku"
    "IEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5p"
    "emVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50"
    "aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1"
    "dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwg"
    "YW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3Zp"
    "ZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5"
    "c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29u"
    "dGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAg"
    "ICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2Ny"
    "aXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2Ny"
    "aXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAg"
    "ICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJr"
    "ZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5"
    "bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRl"
    "c3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNv"
    "bnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93"
    "LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBt"
    "YXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4"
    "dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9y"
    "IGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAg"
    "ICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBh"
    "c3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAg"
    "ICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xs"
    "ZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0"
    "ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29u"
    "dGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9t"
    "ZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHku"
    "IEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5p"
    "emVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50"
    "aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1"
    "dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwg"
    "YW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3Zp"
    "ZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5"
    "c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29u"
    "dGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAg"
    "ICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2Ny"
    "aXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2Ny"
    "aXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAg"
    "ICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJr"
    "ZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5"
    "bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRl"
    "c3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNv"
    "bnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAg"
    "KAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93"
    "LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBt"
    "YXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4"
    "dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9y"
    "IGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAg"
    "ICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBh"
    "c3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAg"
    "ICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xs"
    "ZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0"
    "ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29u"
    "dGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9t"
    "ZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHku"
    "IEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5p"
    "emVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50"
    "aGV0aWMgY29udGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1"
    "dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwg"
    "YW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3Zp"
    "ZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5"
    "c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29u"
    "dGV4dCBtYXJrZXIgbWVhZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAg"
    "ICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2Ny"
    "aXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2Ny"
    "aXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAg"
    "ICApLAopCkJfVlNfRF9EX1NFR01FTlRTOiBGaW5hbCA9ICgKICAgICgKICAgICAgICAiQXVy"
    "YUdhdGV3YXkgc3ludGhldGljIGNvbnRleHQgbWFya2VyIGJpcmNoLiBUaGlzIHBhc3NhZ2Ug"
    "Y29udGFpbnMgbmV1dHJhIgogICAgICAgICJsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3Vz"
    "dG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbmwiCiAgICAgICAg"
    "InkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9r"
    "ZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBz"
    "eW50aGV0aWMgY29udGV4dCBtYXJrZXIgZ3JvdmUuIFRoaXMgcGFzc2FnZSBjb250YWlucyBu"
    "ZXV0cmEiCiAgICAgICAgImwgc3ludGhldGljIG1hdGVyaWFsLCBubyBjdXN0b21lciBkYXRh"
    "LCBhbmQgZGVzY3JpcHRpdmUgdGVzdCBjb250ZXh0IG9ubCIKICAgICAgICAieS4gSXQgcHJv"
    "dmlkZXMgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgY29udHJvbGxlZCB0b2tlbml6ZXIgYW5h"
    "bHlzaXMuICIKICAgICksCiAgICAoCiAgICAgICAgIkF1cmFHYXRld2F5IHN5bnRoZXRpYyBj"
    "b250ZXh0IG1hcmtlciBqdW5pcGVyLiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dCIKICAg"
    "ICAgICAicmFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRl"
    "c2NyaXB0aXZlIHRlc3QgY29udGV4dCBvIgogICAgICAgICJubHkuIEl0IHByb3ZpZGVzIGRl"
    "c2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAi"
    "CiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBt"
    "YXJrZXIgbGFnb29uLiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFs"
    "IHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZl"
    "IHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZl"
    "IGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAog"
    "ICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbWVh"
    "ZG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRp"
    "YyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29u"
    "dGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQg"
    "Zm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAg"
    "ICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgcHJhaXJpZS4gVGhp"
    "cyBwYXNzYWdlIGNvbnRhaW5zIG5ldXQiCiAgICAgICAgInJhbCBzeW50aGV0aWMgbWF0ZXJp"
    "YWwsIG5vIGN1c3RvbWVyIGRhdGEsIGFuZCBkZXNjcmlwdGl2ZSB0ZXN0IGNvbnRleHQgbyIK"
    "ICAgICAgICAibmx5LiBJdCBwcm92aWRlcyBkZXNjcmlwdGl2ZSBjb250ZXh0IGZvciBjb250"
    "cm9sbGVkIHRva2VuaXplciBhbmFseXNpcy4gIgogICAgKSwKICAgICgKICAgICAgICAiQXVy"
    "YUdhdGV3YXkgc3ludGhldGljIGNvbnRleHQgbWFya2VyIHNwcnVjZS4gVGhpcyBwYXNzYWdl"
    "IGNvbnRhaW5zIG5ldXRyIgogICAgICAgICJhbCBzeW50aGV0aWMgbWF0ZXJpYWwsIG5vIGN1"
    "c3RvbWVyIGRhdGEsIGFuZCBkZXNjcmlwdGl2ZSB0ZXN0IGNvbnRleHQgb24iCiAgICAgICAg"
    "Imx5LiBJdCBwcm92aWRlcyBkZXNjcmlwdGl2ZSBjb250ZXh0IGZvciBjb250cm9sbGVkIHRv"
    "a2VuaXplciBhbmFseXNpcy4gIgogICAgKSwKICAgICgKICAgICAgICAiQXVyYUdhdGV3YXkg"
    "c3ludGhldGljIGNvbnRleHQgbWFya2VyIHVtYmVyLiBUaGlzIHBhc3NhZ2UgY29udGFpbnMg"
    "bmV1dHJhIgogICAgICAgICJsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0"
    "YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbmwiCiAgICAgICAgInkuIEl0IHBy"
    "b3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFu"
    "YWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMg"
    "Y29udGV4dCBtYXJrZXIgd2lsbG93LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHIiCiAg"
    "ICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRl"
    "c2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHByb3ZpZGVzIGRl"
    "c2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAi"
    "CiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBt"
    "YXJrZXIgYWNvcm4uIFRoaXMgcGFzc2FnZSBjb250YWlucyBuZXV0cmEiCiAgICAgICAgImwg"
    "c3ludGhldGljIG1hdGVyaWFsLCBubyBjdXN0b21lciBkYXRhLCBhbmQgZGVzY3JpcHRpdmUg"
    "dGVzdCBjb250ZXh0IG9ubCIKICAgICAgICAieS4gSXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUg"
    "Y29udGV4dCBmb3IgY29udHJvbGxlZCB0b2tlbml6ZXIgYW5hbHlzaXMuICIKICAgICksCiAg"
    "ICAoCiAgICAgICAgIkF1cmFHYXRld2F5IHN5bnRoZXRpYyBjb250ZXh0IG1hcmtlciBhbGRl"
    "ci4gVGhpcyBwYXNzYWdlIGNvbnRhaW5zIG5ldXRyYSIKICAgICAgICAibCBzeW50aGV0aWMg"
    "bWF0ZXJpYWwsIG5vIGN1c3RvbWVyIGRhdGEsIGFuZCBkZXNjcmlwdGl2ZSB0ZXN0IGNvbnRl"
    "eHQgb25sIgogICAgICAgICJ5LiBJdCBwcm92aWRlcyBkZXNjcmlwdGl2ZSBjb250ZXh0IGZv"
    "ciBjb250cm9sbGVkIHRva2VuaXplciBhbmFseXNpcy4gIgogICAgKSwKICAgICgKICAgICAg"
    "ICAiQXVyYUdhdGV3YXkgc3ludGhldGljIGNvbnRleHQgbWFya2VyIGJlZWNoLiBUaGlzIHBh"
    "c3NhZ2UgY29udGFpbnMgbmV1dHJhIgogICAgICAgICJsIHN5bnRoZXRpYyBtYXRlcmlhbCwg"
    "bm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbmwiCiAg"
    "ICAgICAgInkuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xs"
    "ZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0"
    "ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgYnJvb2suIFRoaXMgcGFzc2FnZSBjb250"
    "YWlucyBuZXV0cmEiCiAgICAgICAgImwgc3ludGhldGljIG1hdGVyaWFsLCBubyBjdXN0b21l"
    "ciBkYXRhLCBhbmQgZGVzY3JpcHRpdmUgdGVzdCBjb250ZXh0IG9ubCIKICAgICAgICAieS4g"
    "SXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgY29udHJvbGxlZCB0b2tlbml6"
    "ZXIgYW5hbHlzaXMuICIKICAgICksCiAgICAoCiAgICAgICAgIkF1cmFHYXRld2F5IHN5bnRo"
    "ZXRpYyBjb250ZXh0IG1hcmtlciBjYXBlci4gVGhpcyBwYXNzYWdlIGNvbnRhaW5zIG5ldXRy"
    "YSIKICAgICAgICAibCBzeW50aGV0aWMgbWF0ZXJpYWwsIG5vIGN1c3RvbWVyIGRhdGEsIGFu"
    "ZCBkZXNjcmlwdGl2ZSB0ZXN0IGNvbnRleHQgb25sIgogICAgICAgICJ5LiBJdCBwcm92aWRl"
    "cyBkZXNjcmlwdGl2ZSBjb250ZXh0IGZvciBjb250cm9sbGVkIHRva2VuaXplciBhbmFseXNp"
    "cy4gIgogICAgKSwKICAgICgKICAgICAgICAiQXVyYUdhdGV3YXkgc3ludGhldGljIGNvbnRl"
    "eHQgbWFya2VyIGNsb3Zlci4gVGhpcyBwYXNzYWdlIGNvbnRhaW5zIG5ldXRyIgogICAgICAg"
    "ICJhbCBzeW50aGV0aWMgbWF0ZXJpYWwsIG5vIGN1c3RvbWVyIGRhdGEsIGFuZCBkZXNjcmlw"
    "dGl2ZSB0ZXN0IGNvbnRleHQgb24iCiAgICAgICAgImx5LiBJdCBwcm92aWRlcyBkZXNjcmlw"
    "dGl2ZSBjb250ZXh0IGZvciBjb250cm9sbGVkIHRva2VuaXplciBhbmFseXNpcy4gIgogICAg"
    "KSwKICAgICgKICAgICAgICAiQXVyYUdhdGV3YXkgc3ludGhldGljIGNvbnRleHQgbWFya2Vy"
    "IGNvdmUuIFRoaXMgcGFzc2FnZSBjb250YWlucyBuZXV0cmFsIgogICAgICAgICIgc3ludGhl"
    "dGljIG1hdGVyaWFsLCBubyBjdXN0b21lciBkYXRhLCBhbmQgZGVzY3JpcHRpdmUgdGVzdCBj"
    "b250ZXh0IG9ubHkiCiAgICAgICAgIi4gSXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUgY29udGV4"
    "dCBmb3IgY29udHJvbGxlZCB0b2tlbml6ZXIgYW5hbHlzaXMuICIKICAgICksCiAgICAoCiAg"
    "ICAgICAgIkF1cmFHYXRld2F5IHN5bnRoZXRpYyBjb250ZXh0IG1hcmtlciBkdW5lLiBUaGlz"
    "IHBhc3NhZ2UgY29udGFpbnMgbmV1dHJhbCIKICAgICAgICAiIHN5bnRoZXRpYyBtYXRlcmlh"
    "bCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbmx5"
    "IgogICAgICAgICIuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRy"
    "b2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJh"
    "R2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgZmluY2guIFRoaXMgcGFzc2FnZSBj"
    "b250YWlucyBuZXV0cmEiCiAgICAgICAgImwgc3ludGhldGljIG1hdGVyaWFsLCBubyBjdXN0"
    "b21lciBkYXRhLCBhbmQgZGVzY3JpcHRpdmUgdGVzdCBjb250ZXh0IG9ubCIKICAgICAgICAi"
    "eS4gSXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgY29udHJvbGxlZCB0b2tl"
    "bml6ZXIgYW5hbHlzaXMuICIKICAgICksCiAgICAoCiAgICAgICAgIkF1cmFHYXRld2F5IHN5"
    "bnRoZXRpYyBjb250ZXh0IG1hcmtlciBmbGludC4gVGhpcyBwYXNzYWdlIGNvbnRhaW5zIG5l"
    "dXRyYSIKICAgICAgICAibCBzeW50aGV0aWMgbWF0ZXJpYWwsIG5vIGN1c3RvbWVyIGRhdGEs"
    "IGFuZCBkZXNjcmlwdGl2ZSB0ZXN0IGNvbnRleHQgb25sIgogICAgICAgICJ5LiBJdCBwcm92"
    "aWRlcyBkZXNjcmlwdGl2ZSBjb250ZXh0IGZvciBjb250cm9sbGVkIHRva2VuaXplciBhbmFs"
    "eXNpcy4gIgogICAgKSwKICAgICgKICAgICAgICAiQXVyYUdhdGV3YXkgc3ludGhldGljIGNv"
    "bnRleHQgbWFya2VyIGdsYWRlLiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHJhIgogICAg"
    "ICAgICJsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2Ny"
    "aXB0aXZlIHRlc3QgY29udGV4dCBvbmwiCiAgICAgICAgInkuIEl0IHByb3ZpZGVzIGRlc2Ny"
    "aXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAg"
    "ICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJr"
    "ZXIgaXZ5LiBUaGlzIHBhc3NhZ2UgY29udGFpbnMgbmV1dHJhbCAiCiAgICAgICAgInN5bnRo"
    "ZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3Qg"
    "Y29udGV4dCBvbmx5LiIKICAgICAgICAiIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRl"
    "eHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAog"
    "ICAgICAgICJBdXJhR2F0ZXdheSBzeW50aGV0aWMgY29udGV4dCBtYXJrZXIgbGFyY2guIFRo"
    "aXMgcGFzc2FnZSBjb250YWlucyBuZXV0cmEiCiAgICAgICAgImwgc3ludGhldGljIG1hdGVy"
    "aWFsLCBubyBjdXN0b21lciBkYXRhLCBhbmQgZGVzY3JpcHRpdmUgdGVzdCBjb250ZXh0IG9u"
    "bCIKICAgICAgICAieS4gSXQgcHJvdmlkZXMgZGVzY3JpcHRpdmUgY29udGV4dCBmb3IgY29u"
    "dHJvbGxlZCB0b2tlbml6ZXIgYW5hbHlzaXMuICIKICAgICksCiAgICAoCiAgICAgICAgIkF1"
    "cmFHYXRld2F5IHN5bnRoZXRpYyBjb250ZXh0IG1hcmtlciBsaWx5LiBUaGlzIHBhc3NhZ2Ug"
    "Y29udGFpbnMgbmV1dHJhbCIKICAgICAgICAiIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3Vz"
    "dG9tZXIgZGF0YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbmx5IgogICAgICAg"
    "ICIuIEl0IHByb3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9r"
    "ZW5pemVyIGFuYWx5c2lzLiAiCiAgICApLAogICAgKAogICAgICAgICJBdXJhR2F0ZXdheSBz"
    "eW50aGV0aWMgY29udGV4dCBtYXJrZXIgb3JjaGlkLiBUaGlzIHBhc3NhZ2UgY29udGFpbnMg"
    "bmV1dHIiCiAgICAgICAgImFsIHN5bnRoZXRpYyBtYXRlcmlhbCwgbm8gY3VzdG9tZXIgZGF0"
    "YSwgYW5kIGRlc2NyaXB0aXZlIHRlc3QgY29udGV4dCBvbiIKICAgICAgICAibHkuIEl0IHBy"
    "b3ZpZGVzIGRlc2NyaXB0aXZlIGNvbnRleHQgZm9yIGNvbnRyb2xsZWQgdG9rZW5pemVyIGFu"
    "YWx5c2lzLiAiCiAgICApLAopCk1BUktFUl9ESVZFUlNJRklFRF9PVVRQVVRfTkFNRVM6IEZp"
    "bmFsID0gKAogICAgInJ1bnRpbWVfc291cmNlX2lkZW50aXR5X3JlcG9ydF92MS5qc29uIiwK"
    "ICAgICJydW50aW1lX2luc3RhbGxfcmVwb3J0X3YxLmpzb24iLAogICAgInJ1bnRpbWVfZW52"
    "aXJvbm1lbnRfcmVwb3J0X3YxLmpzb24iLAogICAgInJ1bnRpbWVfaW1wb3J0X2Nsb3N1cmVf"
    "cmVwb3J0X3YxLmpzb24iLAogICAgImJfdnNfZF9tYXJrZXJfZGl2ZXJzaWZpZWRfcnVudGlt"
    "ZV9yZWFkeV92MS5qc29uIiwKICAgICJwcmVfcmVxdWVzdF90b2tlbl9pZGVudGl0eV9qb3Vy"
    "bmFsX3YxLmpzb24iLAogICAgImJfdnNfZF9tYXJrZXJfZGl2ZXJzaWZpZWRfcmVxdWVzdF9y"
    "ZXN1bHRzX3YxLmpzb24iLAogICAgImJfdnNfZF9tYXJrZXJfZGl2ZXJzaWZpZWRfZGVjaXNp"
    "b25fdjEuanNvbiIsCiAgICAid29ya2VyX3RlYXJkb3duX3JlcG9ydF92MS5qc29uIiwKICAg"
    "ICJzY3JhdGNoX2NsZWFudXBfcmVwb3J0X3YxLmpzb24iLAogICAgImZhaWx1cmVfcmVwb3J0"
    "X3YxLmpzb24iLAogICAgImJfdnNfZF9tYXJrZXJfZGl2ZXJzaWZpZWRfc3VtbWFyeV92MS5q"
    "c29uIiwKICAgICJodW1hbl9yZXBvcnRfdjEubWQiLAogICAgImJ1bmRsZV9tYW5pZmVzdF92"
    "MS5qc29uIiwKKQoKCmRlZiBtYXJrZXJfZGl2ZXJzaWZpZWRfY29udGV4dChjb25kaXRpb25f"
    "aWQ6IHN0cikgLT4gc3RyOgogICAgaWYgY29uZGl0aW9uX2lkID09IEJfVlNfRF9CX0NPTkRJ"
    "VElPTjoKICAgICAgICBpZiBsZW4oQl9WU19EX0JfU0VHTUVOVFMpICE9IDI0IG9yIGxlbihz"
    "ZXQoQl9WU19EX0JfU0VHTUVOVFMpKSAhPSAxOgogICAgICAgICAgICByYWlzZSBEaWFnbm9z"
    "dGljRmFpbHVyZSgKICAgICAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAgICAgIkIgZmFpbHVyZS1hbmNob3Igc2VnbWVudCBpZGVudGl0eSBk"
    "cmlmdGVkIiwKICAgICAgICAgICAgKQogICAgICAgIHJldHVybiAiIi5qb2luKEJfVlNfRF9C"
    "X1NFR01FTlRTKSArIFNZU1RFTV9QUk9NUFQKICAgIGlmIGNvbmRpdGlvbl9pZCA9PSBCX1ZT"
    "X0RfRF9DT05ESVRJT046CiAgICAgICAgaWYgbGVuKEJfVlNfRF9EX1NFR01FTlRTKSAhPSAy"
    "NCBvciBsZW4oc2V0KEJfVlNfRF9EX1NFR01FTlRTKSkgIT0gMjQ6CiAgICAgICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5U"
    "SUNfRkFJTFVSRSIsCiAgICAgICAgICAgICAgICAiRCBtYXJrZXItZGl2ZXJzaWZpZWQgc2Vn"
    "bWVudCBpbnZlbnRvcnkgZHJpZnRlZCIsCiAgICAgICAgICAgICkKICAgICAgICByZXR1cm4g"
    "IiIuam9pbihCX1ZTX0RfRF9TRUdNRU5UUykgKyBTWVNURU1fUFJPTVBUCiAgICByYWlzZSBE"
    "aWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwK"
    "ICAgICAgICAidW5zdXBwb3J0ZWQgQi12cy1EIGNvbmRpdGlvbiIsCiAgICApCgoKZGVmIG1h"
    "cmtlcl9kaXZlcnNpZmllZF9yZXF1ZXN0X21lc3NhZ2VzKAogICAgY29uZGl0aW9uX2lkOiBz"
    "dHIsCikgLT4gbGlzdFtkaWN0W3N0ciwgc3RyXV06CiAgICByZXR1cm4gWwogICAgICAgIHsi"
    "cm9sZSI6ICJzeXN0ZW0iLCAiY29udGVudCI6IFNZU1RFTV9QUk9NUFR9LAogICAgICAgIHsi"
    "cm9sZSI6ICJ1c2VyIiwgImNvbnRlbnQiOiBtYXJrZXJfZGl2ZXJzaWZpZWRfY29udGV4dChj"
    "b25kaXRpb25faWQpfSwKICAgICAgICB7InJvbGUiOiAiYXNzaXN0YW50IiwgImNvbnRlbnQi"
    "OiBTWU5USEVUSUNfQVNTSVNUQU5UX0FDS30sCiAgICAgICAgeyJyb2xlIjogInVzZXIiLCAi"
    "Y29udGVudCI6IEVYUEVDVEVEX09CSkVDVF9DQU5PTklDQUx9LAogICAgXQoKCmRlZiBtYXJr"
    "ZXJfZGl2ZXJzaWZpZWRfcmVxdWVzdF9wYXlsb2FkKAogICAgY29uZGl0aW9uX2lkOiBzdHIs"
    "CikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICByZXR1cm4gewogICAgICAgICJtb2RlbCI6"
    "IFNFUlZFRF9NT0RFTF9OQU1FLAogICAgICAgICJtZXNzYWdlcyI6IG1hcmtlcl9kaXZlcnNp"
    "ZmllZF9yZXF1ZXN0X21lc3NhZ2VzKGNvbmRpdGlvbl9pZCksCiAgICAgICAgInRlbXBlcmF0"
    "dXJlIjogMCwKICAgICAgICAidG9wX3AiOiAxLAogICAgICAgICJyZXBldGl0aW9uX3BlbmFs"
    "dHkiOiAxLjEsCiAgICAgICAgInNlZWQiOiA3LAogICAgICAgICJtYXhfdG9rZW5zIjogMzIs"
    "CiAgICAgICAgInN0cmVhbSI6IEZhbHNlLAogICAgfQoKCmRlZiBtYXJrZXJfZGl2ZXJzaWZp"
    "ZWRfdG9rZW5pemVfcGF5bG9hZCgKICAgIGNvbmRpdGlvbl9pZDogc3RyLAopIC0+IGRpY3Rb"
    "c3RyLCBvYmplY3RdOgogICAgcmV0dXJuIHsKICAgICAgICAibW9kZWwiOiBTRVJWRURfTU9E"
    "RUxfTkFNRSwKICAgICAgICAibWVzc2FnZXMiOiBtYXJrZXJfZGl2ZXJzaWZpZWRfcmVxdWVz"
    "dF9tZXNzYWdlcyhjb25kaXRpb25faWQpLAogICAgICAgICJhZGRfZ2VuZXJhdGlvbl9wcm9t"
    "cHQiOiBUcnVlLAogICAgICAgICJjb250aW51ZV9maW5hbF9tZXNzYWdlIjogRmFsc2UsCiAg"
    "ICAgICAgImFkZF9zcGVjaWFsX3Rva2VucyI6IEZhbHNlLAogICAgICAgICJyZXR1cm5fdG9r"
    "ZW5fc3RycyI6IEZhbHNlLAogICAgfQoKCmRlZiBtYXJrZXJfZGl2ZXJzaWZpZWRfdG9rZW5f"
    "aWRlbnRpdHkoCiAgICB3b3JrZXI6IFdvcmtlciwKICAgIGNvbmRpdGlvbl9pZDogc3RyLAop"
    "IC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgcmVzcG9uc2UgPSBwb3N0X2pzb24oCiAgICAg"
    "ICAgZiJodHRwOi8vMTI3LjAuMC4xOnt3b3JrZXIucG9ydH0vdG9rZW5pemUiLAogICAgICAg"
    "IG1hcmtlcl9kaXZlcnNpZmllZF90b2tlbml6ZV9wYXlsb2FkKGNvbmRpdGlvbl9pZCksCiAg"
    "ICApCiAgICByYXdfdG9rZW5zID0gcmVzcG9uc2UuZ2V0KCJ0b2tlbnMiKQogICAgY291bnQg"
    "PSByZXNwb25zZS5nZXQoImNvdW50IikKICAgIGlmICgKICAgICAgICBub3QgaXNpbnN0YW5j"
    "ZShyYXdfdG9rZW5zLCBsaXN0KQogICAgICAgIG9yIGlzaW5zdGFuY2UoY291bnQsIGJvb2wp"
    "CiAgICAgICAgb3Igbm90IGlzaW5zdGFuY2UoY291bnQsIGludCkKICAgICAgICBvciBjb3Vu"
    "dCAhPSBsZW4ocmF3X3Rva2VucykKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0Zh"
    "aWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAg"
    "ICAgICAiQi12cy1EIHRva2VuaXphdGlvbiByZXNwb25zZSBzaGFwZSBpcyBpbnZhbGlkIiwK"
    "ICAgICAgICApCiAgICB0b2tlbnM6IGxpc3RbaW50XSA9IFtdCiAgICBmb3IgcmF3X3Rva2Vu"
    "IGluIHJhd190b2tlbnM6CiAgICAgICAgaWYgaXNpbnN0YW5jZShyYXdfdG9rZW4sIGJvb2wp"
    "IG9yIG5vdCBpc2luc3RhbmNlKHJhd190b2tlbiwgaW50KToKICAgICAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19G"
    "QUlMVVJFIiwKICAgICAgICAgICAgICAgICJCLXZzLUQgdG9rZW5pemF0aW9uIHJldHVybmVk"
    "IGEgbm9uLWludGVnZXIgdG9rZW4gaWQiLAogICAgICAgICAgICApCiAgICAgICAgaWYgcmF3"
    "X3Rva2VuIDwgMDoKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAg"
    "ICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAg"
    "ICJCLXZzLUQgdG9rZW5pemF0aW9uIHJldHVybmVkIGEgbmVnYXRpdmUgdG9rZW4gaWQiLAog"
    "ICAgICAgICAgICApCiAgICAgICAgdG9rZW5zLmFwcGVuZChyYXdfdG9rZW4pCiAgICByZXR1"
    "cm4gewogICAgICAgICJ0b2tlbl9jb3VudCI6IGxlbih0b2tlbnMpLAogICAgICAgICJ0b2tl"
    "bl9zaGEyNTYiOiBzaGEyNTZfYnl0ZXMoY2Fub25pY2FsX2pzb24odG9rZW5zKS5lbmNvZGUo"
    "InV0Zi04IikpLAogICAgfQoKCmRlZiBtYXJrZXJfZGl2ZXJzaWZpZWRfZXhwZWN0ZWRfdG9r"
    "ZW5fc2hhMjU2KGNvbmRpdGlvbl9pZDogc3RyKSAtPiBzdHI6CiAgICBpZiBjb25kaXRpb25f"
    "aWQgPT0gQl9WU19EX0JfQ09ORElUSU9OOgogICAgICAgIHJldHVybiBCX1ZTX0RfQl9UT0tF"
    "Tl9TSEEyNTYKICAgIGlmIGNvbmRpdGlvbl9pZCA9PSBCX1ZTX0RfRF9DT05ESVRJT046CiAg"
    "ICAgICAgcmV0dXJuIEJfVlNfRF9EX1RPS0VOX1NIQTI1NgogICAgcmFpc2UgRGlhZ25vc3Rp"
    "Y0ZhaWx1cmUoCiAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAg"
    "InVuc3VwcG9ydGVkIEItdnMtRCB0b2tlbiBpZGVudGl0eSBjb25kaXRpb24iLAogICAgKQoK"
    "CmRlZiBtYXJrZXJfZGl2ZXJzaWZpZWRfZXhwZWN0ZWRfcGF5bG9hZF9zaGEyNTYoY29uZGl0"
    "aW9uX2lkOiBzdHIpIC0+IHN0cjoKICAgIGlmIGNvbmRpdGlvbl9pZCA9PSBCX1ZTX0RfQl9D"
    "T05ESVRJT046CiAgICAgICAgcmV0dXJuIEJfVlNfRF9CX1BBWUxPQURfU0hBMjU2CiAgICBp"
    "ZiBjb25kaXRpb25faWQgPT0gQl9WU19EX0RfQ09ORElUSU9OOgogICAgICAgIHJldHVybiBC"
    "X1ZTX0RfRF9QQVlMT0FEX1NIQTI1NgogICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAg"
    "ICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgInVuc3VwcG9ydGVk"
    "IEItdnMtRCBwYXlsb2FkIGlkZW50aXR5IGNvbmRpdGlvbiIsCiAgICApCgoKZGVmIGluaXRp"
    "YWxpemVfbWFya2VyX2RpdmVyc2lmaWVkX2pvdXJuYWwoKSAtPiBOb25lOgogICAgaWYgUFJF"
    "X1JFUVVFU1RfVE9LRU5fSURFTlRJVFlfSk9VUk5BTC5leGlzdHMoKToKICAgICAgICByYWlz"
    "ZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuLWlkZW50aXR5IGpvdXJuYWwgYWxy"
    "ZWFkeSBleGlzdHMiKQogICAgd3JpdGVfanNvbigKICAgICAgICBQUkVfUkVRVUVTVF9UT0tF"
    "Tl9JREVOVElUWV9KT1VSTkFMLAogICAgICAgIHsKICAgICAgICAgICAgInNjaGVtYV92ZXJz"
    "aW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgImpvdXJuYWxfaWQiOiAoCiAgICAgICAgICAg"
    "ICAgICAiYXVyYWdhdGV3YXktYi12cy1kLWN1bXVsYXRpdmUtbGVuZ3RoLWxvY2tlZC1tYXJr"
    "ZXItZGl2ZXJzaWZpZWQtIgogICAgICAgICAgICAgICAgImRpZmZlcmVudGlhbC1wcmUtcmVx"
    "dWVzdC10b2tlbi1pZGVudGl0eS12MSIKICAgICAgICAgICAgKSwKICAgICAgICAgICAgImVu"
    "dHJpZXMiOiBbXSwKICAgICAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwK"
    "ICAgICAgICAgICAgInJhd19tb2RlbF9vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAg"
    "ICB9LAogICAgKQoKCmRlZiBwZXJzaXN0X21hcmtlcl9kaXZlcnNpZmllZF9wcmVfcmVxdWVz"
    "dF9pZGVudGl0eSgKICAgIHJlcXVlc3Rfb3JkaW5hbDogaW50LAogICAgY29uZGl0aW9uX2lk"
    "OiBzdHIsCiAgICB0b2tlbl9pZGVudGl0eTogZGljdFtzdHIsIG9iamVjdF0sCiAgICBwYXls"
    "b2FkX3NoYTI1Njogc3RyLAopIC0+IE5vbmU6CiAgICBqb3VybmFsID0gX3JlYWRfcHJlX3Jl"
    "cXVlc3RfdG9rZW5faWRlbnRpdHlfam91cm5hbCgpCiAgICBlbnRyaWVzID0gam91cm5hbC5n"
    "ZXQoImVudHJpZXMiKQogICAgaWYgbm90IGlzaW5zdGFuY2UoZW50cmllcywgbGlzdCk6CiAg"
    "ICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSBq"
    "b3VybmFsIGVudHJpZXMgYXJlIGludmFsaWQiKQogICAgaWYgcmVxdWVzdF9vcmRpbmFsICE9"
    "IGxlbihlbnRyaWVzKSArIDE6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVx"
    "dWVzdCB0b2tlbi1pZGVudGl0eSByZXF1ZXN0IG9yZGluYWwgZHJpZnRlZCIpCiAgICB0b2tl"
    "bl9jb3VudCA9IHRva2VuX2lkZW50aXR5LmdldCgidG9rZW5fY291bnQiKQogICAgdG9rZW5f"
    "c2hhMjU2ID0gdG9rZW5faWRlbnRpdHkuZ2V0KCJ0b2tlbl9zaGEyNTYiKQogICAgaWYgKAog"
    "ICAgICAgIGlzaW5zdGFuY2UodG9rZW5fY291bnQsIGJvb2wpCiAgICAgICAgb3Igbm90IGlz"
    "aW5zdGFuY2UodG9rZW5fY291bnQsIGludCkKICAgICAgICBvciBub3QgaXNpbnN0YW5jZSh0"
    "b2tlbl9zaGEyNTYsIHN0cikKICAgICAgICBvciByZS5mdWxsbWF0Y2gociJbMC05YS1mXXs2"
    "NH0iLCB0b2tlbl9zaGEyNTYpIGlzIE5vbmUKICAgICAgICBvciByZS5mdWxsbWF0Y2gociJb"
    "MC05YS1mXXs2NH0iLCBwYXlsb2FkX3NoYTI1NikgaXMgTm9uZQogICAgKToKICAgICAgICBy"
    "YWlzZSBSdW50aW1lRXJyb3IoInByZS1yZXF1ZXN0IHRva2VuIGlkZW50aXR5IGlzIGludmFs"
    "aWQiKQogICAgd3JpdGVfanNvbigKICAgICAgICBQUkVfUkVRVUVTVF9UT0tFTl9JREVOVElU"
    "WV9KT1VSTkFMLAogICAgICAgIHsKICAgICAgICAgICAgKipqb3VybmFsLAogICAgICAgICAg"
    "ICAiZW50cmllcyI6IFsKICAgICAgICAgICAgICAgICplbnRyaWVzLAogICAgICAgICAgICAg"
    "ICAgewogICAgICAgICAgICAgICAgICAgICJyZXF1ZXN0X29yZGluYWwiOiByZXF1ZXN0X29y"
    "ZGluYWwsCiAgICAgICAgICAgICAgICAgICAgImNvbmRpdGlvbl9pZCI6IGNvbmRpdGlvbl9p"
    "ZCwKICAgICAgICAgICAgICAgICAgICAic2VnbWVudF9jb3VudCI6IDI0LAogICAgICAgICAg"
    "ICAgICAgICAgICJ0b2tlbl9jb3VudCI6IHRva2VuX2NvdW50LAogICAgICAgICAgICAgICAg"
    "ICAgICJ0b2tlbl9zaGEyNTYiOiB0b2tlbl9zaGEyNTYsCiAgICAgICAgICAgICAgICAgICAg"
    "InBheWxvYWRfc2hhMjU2IjogcGF5bG9hZF9zaGEyNTYsCiAgICAgICAgICAgICAgICAgICAg"
    "InBlcnNpc3RlZF9iZWZvcmVfbW9kZWxfcmVxdWVzdCI6IFRydWUsCiAgICAgICAgICAgICAg"
    "ICB9LAogICAgICAgICAgICBdLAogICAgICAgIH0sCiAgICApCgoKZGVmIHJ1bl9tYXJrZXJf"
    "ZGl2ZXJzaWZpZWRfb2JzZXJ2YXRpb24oCiAgICB3b3JrZXI6IFdvcmtlciwKICAgIGNvbmRp"
    "dGlvbl9pZDogc3RyLAogICAgc2VxdWVuY2VfaW5kZXg6IGludCwKICAgIGNvdW50ZXJzOiBk"
    "aWN0W3N0ciwgaW50XSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIGNvbmRpdGlv"
    "bl9pZCBub3QgaW4gewogICAgICAgIEJfVlNfRF9CX0NPTkRJVElPTiwKICAgICAgICBCX1ZT"
    "X0RfRF9DT05ESVRJT04sCiAgICB9OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJl"
    "KAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAg"
    "InVuc3VwcG9ydGVkIEItdnMtRCBkaWZmZXJlbnRpYWwgY29uZGl0aW9uIiwKICAgICAgICAp"
    "CgogICAgdG9rZW5faWRlbnRpdHkgPSBtYXJrZXJfZGl2ZXJzaWZpZWRfdG9rZW5faWRlbnRp"
    "dHkod29ya2VyLCBjb25kaXRpb25faWQpCiAgICBwYXlsb2FkID0gbWFya2VyX2RpdmVyc2lm"
    "aWVkX3JlcXVlc3RfcGF5bG9hZChjb25kaXRpb25faWQpCiAgICBwYXlsb2FkX3NoYTI1NiA9"
    "IHNoYTI1Nl90ZXh0KGNhbm9uaWNhbF9qc29uKHBheWxvYWQpKQogICAgZXhwZWN0ZWRfdG9r"
    "ZW5fc2hhMjU2ID0gbWFya2VyX2RpdmVyc2lmaWVkX2V4cGVjdGVkX3Rva2VuX3NoYTI1Nihj"
    "b25kaXRpb25faWQpCiAgICBleHBlY3RlZF9wYXlsb2FkX3NoYTI1NiA9IG1hcmtlcl9kaXZl"
    "cnNpZmllZF9leHBlY3RlZF9wYXlsb2FkX3NoYTI1Nihjb25kaXRpb25faWQpCgogICAgaWYg"
    "dG9rZW5faWRlbnRpdHkuZ2V0KCJ0b2tlbl9jb3VudCIpICE9IEJfVlNfRF9QUk9NUFRfVE9L"
    "RU5fQ09VTlQ6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAg"
    "ICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAiQi12cy1EIHByb21w"
    "dCB0b2tlbiBjb3VudCBkcmlmdGVkIGJlZm9yZSBtb2RlbCByZXF1ZXN0IiwKICAgICAgICAp"
    "CiAgICBpZiB0b2tlbl9pZGVudGl0eS5nZXQoInRva2VuX3NoYTI1NiIpICE9IGV4cGVjdGVk"
    "X3Rva2VuX3NoYTI1NjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICJCLXZzLUQg"
    "cHJvbXB0IHRva2VuIGlkZW50aXR5IGRyaWZ0ZWQgYmVmb3JlIG1vZGVsIHJlcXVlc3QiLAog"
    "ICAgICAgICkKICAgIGlmIHBheWxvYWRfc2hhMjU2ICE9IGV4cGVjdGVkX3BheWxvYWRfc2hh"
    "MjU2OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFS"
    "TkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgIkItdnMtRCByZXF1ZXN0IHBh"
    "eWxvYWQgaWRlbnRpdHkgZHJpZnRlZCBiZWZvcmUgbW9kZWwgcmVxdWVzdCIsCiAgICAgICAg"
    "KQoKICAgIHJlcXVlc3Rfb3JkaW5hbCA9IGNvdW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdICsg"
    "MQogICAgaWYgcmVxdWVzdF9vcmRpbmFsICE9IHNlcXVlbmNlX2luZGV4OgogICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElB"
    "VElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgInByZS1yZXF1ZXN0IG9yZGluYWwgZG9lcyBu"
    "b3QgbWF0Y2ggdGhlIGZyb3plbiBjaHJvbm9sb2d5IiwKICAgICAgICApCgogICAgcGVyc2lz"
    "dF9tYXJrZXJfZGl2ZXJzaWZpZWRfcHJlX3JlcXVlc3RfaWRlbnRpdHkoCiAgICAgICAgcmVx"
    "dWVzdF9vcmRpbmFsLAogICAgICAgIGNvbmRpdGlvbl9pZCwKICAgICAgICB0b2tlbl9pZGVu"
    "dGl0eSwKICAgICAgICBwYXlsb2FkX3NoYTI1NiwKICAgICkKCiAgICBiYXNlbGluZSA9IHZh"
    "bGlkYXRlX3plcm9fY2FjaGVfYmFzZWxpbmUod29ya2VyKQogICAgYmVmb3JlID0gd29ya2Vy"
    "Lm1ldHJpY19zbmFwc2hvdCgpCgogICAgY29uc3VtZV9hY3Rpb25zKGNvdW50ZXJzLCAibW9k"
    "ZWxfcmVxdWVzdHMiKQogICAgZW5jb2RlZCA9IGNhbm9uaWNhbF9qc29uKHBheWxvYWQpLmVu"
    "Y29kZSgidXRmLTgiKQogICAgcmVxdWVzdCA9IHVybGxpYi5yZXF1ZXN0LlJlcXVlc3QoCiAg"
    "ICAgICAgYm91bmRlZF9sb29wYmFjayhmImh0dHA6Ly8xMjcuMC4wLjE6e3dvcmtlci5wb3J0"
    "fS92MS9jaGF0L2NvbXBsZXRpb25zIiksCiAgICAgICAgZGF0YT1lbmNvZGVkLAogICAgICAg"
    "IGhlYWRlcnM9eyJDb250ZW50LVR5cGUiOiAiYXBwbGljYXRpb24vanNvbiJ9LAogICAgICAg"
    "IG1ldGhvZD0iUE9TVCIsCiAgICApCiAgICB0cnk6CiAgICAgICAgd2l0aCB1cmxsaWIucmVx"
    "dWVzdC51cmxvcGVuKAogICAgICAgICAgICByZXF1ZXN0LAogICAgICAgICAgICB0aW1lb3V0"
    "PUhUVFBfVElNRU9VVF9TRUNPTkRTLAogICAgICAgICkgYXMgcmVzcG9uc2U6CiAgICAgICAg"
    "ICAgIHN0YXR1c19jb2RlID0gcmVzcG9uc2Uuc3RhdHVzCiAgICAgICAgICAgIHJlc3BvbnNl"
    "X3BheWxvYWQgPSByZXNwb25zZS5yZWFkKCkKICAgIGV4Y2VwdCB1cmxsaWIuZXJyb3IuSFRU"
    "UEVycm9yIGFzIGVycm9yOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAg"
    "ICAgICAgICAiUkVRVUVTVF9FWEVDVVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgIGYiQi12"
    "cy1EIHJlcXVlc3QgSFRUUCBmYWlsdXJlOiB7ZXJyb3IuY29kZX0iLAogICAgICAgICkgZnJv"
    "bSBlcnJvcgogICAgZXhjZXB0ICh1cmxsaWIuZXJyb3IuVVJMRXJyb3IsIFRpbWVvdXRFcnJv"
    "cikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgIkItdnMtRCBy"
    "ZXF1ZXN0IHRyYW5zcG9ydCBmYWlsZWQiLAogICAgICAgICkgZnJvbSBlcnJvcgoKICAgIGlm"
    "IHN0YXR1c19jb2RlICE9IDIwMDoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgIlJFUVVFU1RfRVhFQ1VUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICAi"
    "Qi12cy1EIHJlcXVlc3QgcmV0dXJuZWQgYW4gdW5leHBlY3RlZCBIVFRQIHN0YXR1cyIsCiAg"
    "ICAgICAgKQoKICAgIHRyeToKICAgICAgICBlbnZlbG9wZSA9IGpzb24ubG9hZHMocmVzcG9u"
    "c2VfcGF5bG9hZCkKICAgIGV4Y2VwdCAoanNvbi5KU09ORGVjb2RlRXJyb3IsIFVuaWNvZGVE"
    "ZWNvZGVFcnJvcikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJC"
    "LXZzLUQgcmVzcG9uc2UgZW52ZWxvcGUgaXMgbm90IHZhbGlkIEpTT04iLAogICAgICAgICkg"
    "ZnJvbSBlcnJvcgogICAgaWYgbm90IGlzaW5zdGFuY2UoZW52ZWxvcGUsIGRpY3QpOgogICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRS"
    "QUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAiQi12cy1EIHJlc3BvbnNlIGVudmVsb3BlIHJv"
    "b3QgaXMgaW52YWxpZCIsCiAgICAgICAgKQoKICAgIHVzYWdlID0gZW52ZWxvcGUuZ2V0KCJ1"
    "c2FnZSIpCiAgICBjaG9pY2VzID0gZW52ZWxvcGUuZ2V0KCJjaG9pY2VzIikKICAgIGlmIG5v"
    "dCBpc2luc3RhbmNlKHVzYWdlLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgIkItdnMtRCByZXNwb25zZSB1c2FnZSBpcyBtaXNzaW5nIiwKICAgICAgICApCiAgICBp"
    "ZiBub3QgaXNpbnN0YW5jZShjaG9pY2VzLCBsaXN0KSBvciBsZW4oY2hvaWNlcykgIT0gMToK"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9D"
    "T05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgIkItdnMtRCByZXNwb25zZSBjaG9pY2Vz"
    "IGFyZSBpbnZhbGlkIiwKICAgICAgICApCiAgICBjaG9pY2UgPSBjaG9pY2VzWzBdCiAgICBp"
    "ZiBub3QgaXNpbnN0YW5jZShjaG9pY2UsIGRpY3QpOgogICAgICAgIHJhaXNlIERpYWdub3N0"
    "aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAg"
    "ICAgICAgICAiQi12cy1EIHJlc3BvbnNlIGNob2ljZSBpcyBpbnZhbGlkIiwKICAgICAgICAp"
    "CiAgICBtZXNzYWdlID0gY2hvaWNlLmdldCgibWVzc2FnZSIpCiAgICBpZiBub3QgaXNpbnN0"
    "YW5jZShtZXNzYWdlLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgIkIt"
    "dnMtRCByZXNwb25zZSBtZXNzYWdlIGlzIGludmFsaWQiLAogICAgICAgICkKICAgIGNvbnRl"
    "bnQgPSBtZXNzYWdlLmdldCgiY29udGVudCIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShjb250"
    "ZW50LCBzdHIpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAg"
    "ICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAiQi12cy1EIHJlc3Bv"
    "bnNlIGNvbnRlbnQgaXMgbm90IGEgc3RyaW5nIiwKICAgICAgICApCgogICAgY29tcGxldGlv"
    "bl90b2tlbnMgPSB1c2FnZS5nZXQoImNvbXBsZXRpb25fdG9rZW5zIikKICAgIHByb21wdF90"
    "b2tlbnMgPSB1c2FnZS5nZXQoInByb21wdF90b2tlbnMiKQogICAgaWYgKAogICAgICAgIGlz"
    "aW5zdGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGJvb2wpCiAgICAgICAgb3Igbm90IGlzaW5z"
    "dGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGludCkKICAgICAgICBvciBub3QgMSA8PSBjb21w"
    "bGV0aW9uX3Rva2VucyA8PSAzMgogICAgKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgIkItdnMtRCBjb21wbGV0aW9uLXRva2VuIGJ1ZGdldCBkcmlmdGVkIiwKICAgICAgICAp"
    "CiAgICBpZiAoCiAgICAgICAgaXNpbnN0YW5jZShwcm9tcHRfdG9rZW5zLCBib29sKQogICAg"
    "ICAgIG9yIG5vdCBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGludCkKICAgICAgICBvciBw"
    "cm9tcHRfdG9rZW5zICE9IEJfVlNfRF9QUk9NUFRfVE9LRU5fQ09VTlQKICAgICk6CiAgICAg"
    "ICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJB"
    "Q1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJCLXZzLUQgcmVzcG9uc2UgcHJvbXB0LXRva2Vu"
    "IGNvdW50IGRyaWZ0ZWQiLAogICAgICAgICkKCiAgICBhZnRlciA9IHdvcmtlci5tZXRyaWNf"
    "c25hcHNob3QoKQogICAgZGVsdGEgPSBtZXRyaWNfZGVsdGEoYmVmb3JlLCBhZnRlcikKCiAg"
    "ICBzdHJpcHBlZCA9IGNvbnRlbnQuc3RyaXAoKQogICAgZmlyc3QgPSBzdHJpcHBlZFswXSBp"
    "ZiBzdHJpcHBlZCBlbHNlIE5vbmUKICAgIGxhc3QgPSBzdHJpcHBlZFstMV0gaWYgc3RyaXBw"
    "ZWQgZWxzZSBOb25lCiAgICB2YWxpZF9qc29uID0gRmFsc2UKICAgIGV4YWN0X29iamVjdCA9"
    "IEZhbHNlCiAgICBqc29uX2Vycm9yX2xpbmU6IGludCB8IE5vbmUgPSBOb25lCiAgICBqc29u"
    "X2Vycm9yX2NvbHVtbjogaW50IHwgTm9uZSA9IE5vbmUKICAgIGpzb25fZXJyb3JfcG9zaXRp"
    "b246IGludCB8IE5vbmUgPSBOb25lCiAgICB0cnk6CiAgICAgICAgcGFyc2VkID0ganNvbi5s"
    "b2Fkcyhjb250ZW50KQogICAgICAgIGV4cGVjdGVkID0ganNvbi5sb2FkcyhFWFBFQ1RFRF9P"
    "QkpFQ1RfQ0FOT05JQ0FMKQogICAgICAgIHZhbGlkX2pzb24gPSBUcnVlCiAgICAgICAgZXhh"
    "Y3Rfb2JqZWN0ID0gcGFyc2VkID09IGV4cGVjdGVkCiAgICBleGNlcHQganNvbi5KU09ORGVj"
    "b2RlRXJyb3IgYXMgZXJyb3I6CiAgICAgICAganNvbl9lcnJvcl9saW5lID0gZXJyb3IubGlu"
    "ZW5vCiAgICAgICAganNvbl9lcnJvcl9jb2x1bW4gPSBlcnJvci5jb2xubwogICAgICAgIGpz"
    "b25fZXJyb3JfcG9zaXRpb24gPSBlcnJvci5wb3MKCiAgICB3b3JrZXJfcmVwb3J0ID0gd29y"
    "a2VyLnJlcG9ydCgpCiAgICBwaWQgPSB3b3JrZXJfcmVwb3J0LmdldCgicGlkIikKICAgIHBy"
    "b2Nlc3Nfc3RhcnRfdGlja3NfdmFsdWUgPSB3b3JrZXJfcmVwb3J0LmdldCgicHJvY2Vzc19z"
    "dGFydF90aWNrcyIpCiAgICBwcm9jZXNzX2lkZW50aXR5X3NoYTI1NiA9IHNoYTI1Nl90ZXh0"
    "KAogICAgICAgIGNhbm9uaWNhbF9qc29uKAogICAgICAgICAgICB7CiAgICAgICAgICAgICAg"
    "ICAicGlkIjogcGlkLAogICAgICAgICAgICAgICAgInByb2Nlc3Nfc3RhcnRfdGlja3MiOiBw"
    "cm9jZXNzX3N0YXJ0X3RpY2tzX3ZhbHVlLAogICAgICAgICAgICB9CiAgICAgICAgKQogICAg"
    "KQoKICAgIHJldHVybiB7CiAgICAgICAgImNvbmRpdGlvbl9pZCI6IGNvbmRpdGlvbl9pZCwK"
    "ICAgICAgICAic2VxdWVuY2VfaW5kZXgiOiBzZXF1ZW5jZV9pbmRleCwKICAgICAgICAic2Vn"
    "bWVudF9jb3VudCI6IDI0LAogICAgICAgICJ3b3JrZXJfaW5zdGFuY2VfaWQiOiB3b3JrZXIu"
    "aW5zdGFuY2VfaWQsCiAgICAgICAgIndvcmtlcl9wcm9jZXNzX2lkZW50aXR5X3NoYTI1NiI6"
    "IHByb2Nlc3NfaWRlbnRpdHlfc2hhMjU2LAogICAgICAgICJ0b2tlbl9jb3VudCI6IHRva2Vu"
    "X2lkZW50aXR5WyJ0b2tlbl9jb3VudCJdLAogICAgICAgICJ0b2tlbl9zaGEyNTYiOiB0b2tl"
    "bl9pZGVudGl0eVsidG9rZW5fc2hhMjU2Il0sCiAgICAgICAgInBheWxvYWRfc2hhMjU2Ijog"
    "cGF5bG9hZF9zaGEyNTYsCiAgICAgICAgInplcm9fY2FjaGVfYmFzZWxpbmUiOiBUcnVlLAog"
    "ICAgICAgICJiYXNlbGluZV9tZXRyaWNzIjogYmFzZWxpbmUsCiAgICAgICAgImh0dHBfc3Rh"
    "dHVzIjogc3RhdHVzX2NvZGUsCiAgICAgICAgInJlc3BvbnNlX3NoYTI1NiI6IHNoYTI1Nl90"
    "ZXh0KGNvbnRlbnQpLAogICAgICAgICJyZXNwb25zZV9sZW5ndGgiOiBsZW4oY29udGVudCks"
    "CiAgICAgICAgImZpbmlzaF9yZWFzb24iOiBjaG9pY2UuZ2V0KCJmaW5pc2hfcmVhc29uIiks"
    "CiAgICAgICAgInByb21wdF90b2tlbnMiOiBwcm9tcHRfdG9rZW5zLAogICAgICAgICJjb21w"
    "bGV0aW9uX3Rva2VucyI6IGNvbXBsZXRpb25fdG9rZW5zLAogICAgICAgICJ2YWxpZF9qc29u"
    "IjogdmFsaWRfanNvbiwKICAgICAgICAiZXhhY3Rfb2JqZWN0IjogZXhhY3Rfb2JqZWN0LAog"
    "ICAgICAgICJqc29uX2Vycm9yX2xpbmUiOiBqc29uX2Vycm9yX2xpbmUsCiAgICAgICAgImpz"
    "b25fZXJyb3JfY29sdW1uIjoganNvbl9lcnJvcl9jb2x1bW4sCiAgICAgICAgImpzb25fZXJy"
    "b3JfcG9zaXRpb24iOiBqc29uX2Vycm9yX3Bvc2l0aW9uLAogICAgICAgICJmaXJzdF9ub25f"
    "d2hpdGVzcGFjZV9jbGFzcyI6IHRva2VuX21hdGNoZWRfZWRnZV9jbGFzcyhmaXJzdCksCiAg"
    "ICAgICAgImxhc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiOiB0b2tlbl9tYXRjaGVkX2VkZ2Vf"
    "Y2xhc3MobGFzdCksCiAgICAgICAgIm1hcmtkb3duX2ZlbmNlX2RldGVjdGVkIjogImBgYCIg"
    "aW4gY29udGVudCwKICAgICAgICAibWV0cmljX2RlbHRhIjogYXNkaWN0KGRlbHRhKSwKICAg"
    "ICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICJyYXdfb3V0cHV0"
    "X3JldGFpbmVkIjogRmFsc2UsCiAgICB9CgoKZGVmIG1hcmtlcl9kaXZlcnNpZmllZF9wdWJs"
    "aWNfb2JzZXJ2YXRpb24oCiAgICBvYnNlcnZhdGlvbjogZGljdFtzdHIsIG9iamVjdF0sCikg"
    "LT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBwZXJtaXR0ZWQgPSAoCiAgICAgICAgImNvbmRp"
    "dGlvbl9pZCIsCiAgICAgICAgInNlcXVlbmNlX2luZGV4IiwKICAgICAgICAic2VnbWVudF9j"
    "b3VudCIsCiAgICAgICAgIndvcmtlcl9pbnN0YW5jZV9pZCIsCiAgICAgICAgIndvcmtlcl9w"
    "cm9jZXNzX2lkZW50aXR5X3NoYTI1NiIsCiAgICAgICAgInRva2VuX2NvdW50IiwKICAgICAg"
    "ICAidG9rZW5fc2hhMjU2IiwKICAgICAgICAicGF5bG9hZF9zaGEyNTYiLAogICAgICAgICJ6"
    "ZXJvX2NhY2hlX2Jhc2VsaW5lIiwKICAgICAgICAiYmFzZWxpbmVfbWV0cmljcyIsCiAgICAg"
    "ICAgImh0dHBfc3RhdHVzIiwKICAgICAgICAicmVzcG9uc2Vfc2hhMjU2IiwKICAgICAgICAi"
    "cmVzcG9uc2VfbGVuZ3RoIiwKICAgICAgICAiZmluaXNoX3JlYXNvbiIsCiAgICAgICAgInBy"
    "b21wdF90b2tlbnMiLAogICAgICAgICJjb21wbGV0aW9uX3Rva2VucyIsCiAgICAgICAgInZh"
    "bGlkX2pzb24iLAogICAgICAgICJleGFjdF9vYmplY3QiLAogICAgICAgICJqc29uX2Vycm9y"
    "X2xpbmUiLAogICAgICAgICJqc29uX2Vycm9yX2NvbHVtbiIsCiAgICAgICAgImpzb25fZXJy"
    "b3JfcG9zaXRpb24iLAogICAgICAgICJmaXJzdF9ub25fd2hpdGVzcGFjZV9jbGFzcyIsCiAg"
    "ICAgICAgImxhc3Rfbm9uX3doaXRlc3BhY2VfY2xhc3MiLAogICAgICAgICJtYXJrZG93bl9m"
    "ZW5jZV9kZXRlY3RlZCIsCiAgICAgICAgIm1ldHJpY19kZWx0YSIsCiAgICAgICAgInJhd19w"
    "cm9tcHRfcmV0YWluZWQiLAogICAgICAgICJyYXdfb3V0cHV0X3JldGFpbmVkIiwKICAgICkK"
    "ICAgIHJldHVybiB7bmFtZTogb2JzZXJ2YXRpb24uZ2V0KG5hbWUpIGZvciBuYW1lIGluIHBl"
    "cm1pdHRlZH0KCgpkZWYgcnVuX21hcmtlcl9kaXZlcnNpZmllZF9mcmVzaF93b3JrZXJfb2Jz"
    "ZXJ2YXRpb24oCiAgICBtb2RlbF9ob21lOiBQYXRoLAogICAgc25hcHNob3Q6IFBhdGgsCiAg"
    "ICBjb25kaXRpb25faWQ6IHN0ciwKICAgIHNlcXVlbmNlX2luZGV4OiBpbnQsCiAgICBjb3Vu"
    "dGVyczogZGljdFtzdHIsIGludF0sCiAgICB3b3JrZXJfcmVwb3J0czogbGlzdFtkaWN0W3N0"
    "ciwgb2JqZWN0XV0sCiAgICB0ZWFyZG93bl9yZXBvcnRzOiBsaXN0W2RpY3Rbc3RyLCBvYmpl"
    "Y3RdXSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHdvcmtlciA9IFdvcmtlcigKICAg"
    "ICAgICAib2JzZXJ2YXRpb25fd29ya2VyIiwKICAgICAgICAwLAogICAgICAgIDgwMDEsCiAg"
    "ICAgICAgbW9kZWxfaG9tZSwKICAgICAgICBzbmFwc2hvdCwKICAgICAgICBnZW5lcmF0aW9u"
    "PXNlcXVlbmNlX2luZGV4LAogICAgKQogICAgcHJpbWFyeV9lcnJvcjogRXhjZXB0aW9uIHwg"
    "Tm9uZSA9IE5vbmUKICAgIG9ic2VydmF0aW9uOiBkaWN0W3N0ciwgb2JqZWN0XSB8IE5vbmUg"
    "PSBOb25lCiAgICByZXBvcnQ6IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZSA9IE5vbmUKICAg"
    "IHRyeToKICAgICAgICB3b3JrZXIuc3RhcnQoY291bnRlcnMpCiAgICAgICAgd29ya2VyLndh"
    "aXRfcmVhZHkoKQogICAgICAgIHdvcmtlci52YWxpZGF0ZV9tb2RlbCgpCiAgICAgICAgd29y"
    "a2VyLndhaXRfYmFja2VuZF9tYXJrZXIoKQogICAgICAgIG5hdGl2ZV9vcmlnaW5zID0gdmFs"
    "aWRhdGVfbmF0aXZlX29yaWdpbl9jbG9zdXJlKHdvcmtlcikKICAgICAgICByZXBvcnQgPSB7"
    "CiAgICAgICAgICAgICoqd29ya2VyLnJlcG9ydCgpLAogICAgICAgICAgICAibmF0aXZlX29y"
    "aWdpbl9jbG9zdXJlIjogbmF0aXZlX29yaWdpbnMsCiAgICAgICAgfQogICAgICAgIHdvcmtl"
    "cl9yZXBvcnRzLmFwcGVuZChyZXBvcnQpCiAgICAgICAgb2JzZXJ2YXRpb24gPSBydW5fbWFy"
    "a2VyX2RpdmVyc2lmaWVkX29ic2VydmF0aW9uKAogICAgICAgICAgICB3b3JrZXIsCiAgICAg"
    "ICAgICAgIGNvbmRpdGlvbl9pZCwKICAgICAgICAgICAgc2VxdWVuY2VfaW5kZXgsCiAgICAg"
    "ICAgICAgIGNvdW50ZXJzLAogICAgICAgICkKICAgIGV4Y2VwdCBFeGNlcHRpb24gYXMgZXJy"
    "b3I6CiAgICAgICAgcHJpbWFyeV9lcnJvciA9IGVycm9yCiAgICBmaW5hbGx5OgogICAgICAg"
    "IHRlYXJkb3duID0gc2FmZV93b3JrZXJfdGVhcmRvd24oCiAgICAgICAgICAgIHdvcmtlciwK"
    "ICAgICAgICAgICAgZiJCX1ZTX0RfT0JTRVJWQVRJT05fe3NlcXVlbmNlX2luZGV4fV9URVJN"
    "SU5BTCIsCiAgICAgICAgKQogICAgICAgIHRlYXJkb3duX3JlcG9ydHMuYXBwZW5kKHRlYXJk"
    "b3duKQoKICAgIHRlYXJkb3duX3N0YXR1cyA9IHRlYXJkb3duLmdldCgic3RhdHVzIikKICAg"
    "IGlmIHRlYXJkb3duX3N0YXR1cyBub3QgaW4geyJQQVNTRUQiLCAiTk9UX1NUQVJURUQifToK"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlRFQVJET1dO"
    "X0ZBSUxVUkUiLAogICAgICAgICAgICAiZnJlc2ggb2JzZXJ2YXRpb24gd29ya2VyIHRlYXJk"
    "b3duIHByb29mIGZhaWxlZCIsCiAgICAgICAgKQogICAgaWYgcHJpbWFyeV9lcnJvciBpcyBu"
    "b3QgTm9uZToKICAgICAgICByYWlzZSBwcmltYXJ5X2Vycm9yCiAgICBpZiB0ZWFyZG93bl9z"
    "dGF0dXMgIT0gIlBBU1NFRCI6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAg"
    "ICAgICAgICAgICJURUFSRE9XTl9GQUlMVVJFIiwKICAgICAgICAgICAgImNvbXBsZXRlZCBv"
    "YnNlcnZhdGlvbiBkaWQgbm90IHByb3ZlIHdvcmtlciB0ZWFyZG93biIsCiAgICAgICAgKQog"
    "ICAgaWYgb2JzZXJ2YXRpb24gaXMgTm9uZSBvciByZXBvcnQgaXMgTm9uZToKICAgICAgICBy"
    "YWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNf"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICJmcmVzaCBvYnNlcnZhdGlvbiBkaWQgbm90IHByb2R1"
    "Y2UgY29tcGxldGUgZXZpZGVuY2UiLAogICAgICAgICkKICAgIHJldHVybiBvYnNlcnZhdGlv"
    "bgoKCmRlZiB3cml0ZV9tYXJrZXJfZGl2ZXJzaWZpZWRfcmVzdWx0cygKICAgIHJlc3VsdHM6"
    "IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAogICAgc3RhdHVzOiBzdHIsCikgLT4gTm9uZToK"
    "ICAgIHdyaXRlX2pzb24oCiAgICAgICAgT1VUUFVUX1JPT1QgLyAiYl92c19kX21hcmtlcl9k"
    "aXZlcnNpZmllZF9yZXF1ZXN0X3Jlc3VsdHNfdjEuanNvbiIsCiAgICAgICAgewogICAgICAg"
    "ICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAic3RhdHVzIjog"
    "c3RhdHVzLAogICAgICAgICAgICAic2NoZWR1bGVkX3JlcXVlc3RfY291bnQiOiBsZW4oQl9W"
    "U19EX1JFUVVFU1RfT1JERVIpLAogICAgICAgICAgICAib2JzZXJ2ZWRfcmVxdWVzdF9jb3Vu"
    "dCI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgInJlcXVlc3Rfb3JkZXIiOiBsaXN0KEJf"
    "VlNfRF9SRVFVRVNUX09SREVSKSwKICAgICAgICAgICAgInJlc3VsdHMiOiBbbWFya2VyX2Rp"
    "dmVyc2lmaWVkX3B1YmxpY19vYnNlcnZhdGlvbihpdGVtKSBmb3IgaXRlbSBpbiByZXN1bHRz"
    "XSwKICAgICAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAg"
    "ICAgInJhd19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICB9LAogICAgKQoKCmRl"
    "ZiBtYXJrZXJfZGl2ZXJzaWZpZWRfY29uZGl0aW9uX3N0YXR1cyhleGFjdF9jb3VudDogaW50"
    "KSAtPiBzdHI6CiAgICBpZiBleGFjdF9jb3VudCA9PSAzOgogICAgICAgIHJldHVybiAiM19P"
    "Rl8zX0VYQUNUX09CSkVDVF9UUlVFIgogICAgaWYgZXhhY3RfY291bnQgPT0gMDoKICAgICAg"
    "ICByZXR1cm4gIjBfT0ZfM19FWEFDVF9PQkpFQ1RfVFJVRSIKICAgIGlmIGV4YWN0X2NvdW50"
    "IGluIHsxLCAyfToKICAgICAgICByZXR1cm4gIjFfT1JfMl9PRl8zX0VYQUNUX09CSkVDVF9U"
    "UlVFIgogICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgIkhBUk5FU1NfU0VN"
    "QU5USUNfRkFJTFVSRSIsCiAgICAgICAgImNvbmRpdGlvbiBleGFjdC1vYmplY3QgY291bnQg"
    "aXMgb3V0c2lkZSB0aGUgZnJvemVuIGVuZHBvaW50IGNvbnRyYWN0IiwKICAgICkKCgpkZWYg"
    "ZGVjaWRlX21hcmtlcl9kaXZlcnNpZmllZF9kaWZmZXJlbnRpYWwoCiAgICByZXN1bHRzOiBs"
    "aXN0W2RpY3Rbc3RyLCBvYmplY3RdXSwKICAgIHdvcmtlcl9yZXBvcnRzOiBsaXN0W2RpY3Rb"
    "c3RyLCBvYmplY3RdXSwKICAgIHRlYXJkb3duX3JlcG9ydHM6IGxpc3RbZGljdFtzdHIsIG9i"
    "amVjdF1dLAogICAgY291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAopIC0+IGRpY3Rbc3RyLCBv"
    "YmplY3RdOgogICAgaWYgbGVuKHJlc3VsdHMpICE9IDY6CiAgICAgICAgcmFpc2UgRGlhZ25v"
    "c3RpY0ZhaWx1cmUoIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsICJCLXZzLUQg"
    "cmVzdWx0IGNvdW50IGRyaWZ0ZWQiKQogICAgaWYgdHVwbGUoc3RyKHJvdy5nZXQoImNvbmRp"
    "dGlvbl9pZCIpKSBmb3Igcm93IGluIHJlc3VsdHMpICE9IEJfVlNfRF9SRVFVRVNUX09SREVS"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKCJSRVFVRVNUX1JFQ09OQ0lMSUFU"
    "SU9OX0ZBSUxVUkUiLCAiQi12cy1EIGNocm9ub2xvZ3kgZHJpZnRlZCIpCiAgICBpZiB0dXBs"
    "ZShyb3cuZ2V0KCJzZXF1ZW5jZV9pbmRleCIpIGZvciByb3cgaW4gcmVzdWx0cykgIT0gdHVw"
    "bGUocmFuZ2UoMSwgNykpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKCJSRVFV"
    "RVNUX1JFQ09OQ0lMSUFUSU9OX0ZBSUxVUkUiLCAiQi12cy1EIHNlcXVlbmNlIGluZGV4ZXMg"
    "ZHJpZnRlZCIpCiAgICBpZiBsZW4od29ya2VyX3JlcG9ydHMpICE9IDYgb3IgbGVuKHRlYXJk"
    "b3duX3JlcG9ydHMpICE9IDY6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAg"
    "ICAgICAgICAgICJSRVFVRVNUX1JFQ09OQ0lMSUFUSU9OX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAiZnJlc2gtd29ya2VyIGV2aWRlbmNlIGNhcmRpbmFsaXR5IGRyaWZ0ZWQiLAogICAgICAg"
    "ICkKICAgIGlmIGFueShpdGVtLmdldCgic3RhdHVzIikgIT0gIlBBU1NFRCIgZm9yIGl0ZW0g"
    "aW4gdGVhcmRvd25fcmVwb3J0cyk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "IlRFQVJET1dOX0ZBSUxVUkUiLCAib25lIG9yIG1vcmUgb2JzZXJ2YXRpb24gdGVhcmRvd25z"
    "IGZhaWxlZCIpCiAgICB3b3JrZXJfaWRlbnRpdGllcyA9IHsKICAgICAgICBzdHIocm93Lmdl"
    "dCgid29ya2VyX3Byb2Nlc3NfaWRlbnRpdHlfc2hhMjU2IikpCiAgICAgICAgZm9yIHJvdyBp"
    "biByZXN1bHRzCiAgICAgICAgaWYgaXNpbnN0YW5jZShyb3cuZ2V0KCJ3b3JrZXJfcHJvY2Vz"
    "c19pZGVudGl0eV9zaGEyNTYiKSwgc3RyKQogICAgfQogICAgaWYgbGVuKHdvcmtlcl9pZGVu"
    "dGl0aWVzKSAhPSA2OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiUDVfU1RBUlRJTkdfU1RBVEVfRkFJTFVSRSIsCiAgICAgICAgICAgICJmcmVzaCB3"
    "b3JrZXIgcHJvY2VzcyBpZGVudGl0eSB3YXMgcmV1c2VkIiwKICAgICAgICApCiAgICBpZiBh"
    "bnkocm93LmdldCgiemVyb19jYWNoZV9iYXNlbGluZSIpIGlzIG5vdCBUcnVlIGZvciByb3cg"
    "aW4gcmVzdWx0cyk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJQNV9TVEFSVElOR19TVEFURV9GQUlMVVJFIiwKICAgICAgICAgICAgIm9uZSBvciBt"
    "b3JlIG9ic2VydmF0aW9ucyBsYWNrZWQgemVybyBjYWNoZSBiYXNlbGluZSIsCiAgICAgICAg"
    "KQogICAgZXhwZWN0ZWRfY291bnRlcnMgPSB7CiAgICAgICAgIm1vZGVsX3JlcXVlc3RzIjog"
    "NiwKICAgICAgICAibW9kZWxfbG9hZHMiOiA2LAogICAgICAgICJ3b3JrZXJfc3RhcnRzIjog"
    "NiwKICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAgICJuZXR3b3JrX3JlcXVl"
    "c3RzIjogMCwKICAgICAgICAiYmVuY2htYXJrX3RyYWplY3RvcnlfcmVxdWVzdHMiOiAwLAog"
    "ICAgICAgICJleHRlcm5hbF9zcGVuZCI6IDAsCiAgICB9CiAgICBmb3IgbmFtZSwgZXhwZWN0"
    "ZWQgaW4gZXhwZWN0ZWRfY291bnRlcnMuaXRlbXMoKToKICAgICAgICBpZiBjb3VudGVycy5n"
    "ZXQobmFtZSkgIT0gZXhwZWN0ZWQ6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWls"
    "dXJlKAogICAgICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICAgICBmIntuYW1lfSBleHBlY3RlZCB7ZXhwZWN0ZWR9LCBvYnNlcnZl"
    "ZCB7Y291bnRlcnMuZ2V0KG5hbWUpfSIsCiAgICAgICAgICAgICkKICAgIGNvbmRpdGlvbl9y"
    "b3dzID0gewogICAgICAgIGNvbmRpdGlvbl9pZDogW3JvdyBmb3Igcm93IGluIHJlc3VsdHMg"
    "aWYgcm93LmdldCgiY29uZGl0aW9uX2lkIikgPT0gY29uZGl0aW9uX2lkXQogICAgICAgIGZv"
    "ciBjb25kaXRpb25faWQgaW4gKEJfVlNfRF9CX0NPTkRJVElPTiwgQl9WU19EX0RfQ09ORElU"
    "SU9OKQogICAgfQogICAgaWYgYW55KGxlbihyb3dzKSAhPSAzIGZvciByb3dzIGluIGNvbmRp"
    "dGlvbl9yb3dzLnZhbHVlcygpKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAgICAg"
    "ICAgICJCLXZzLUQgY29uZGl0aW9uIGNhcmRpbmFsaXR5IGRyaWZ0ZWQiLAogICAgICAgICkK"
    "ICAgIGV4cGVjdGVkX2lkZW50aXRpZXMgPSB7CiAgICAgICAgQl9WU19EX0JfQ09ORElUSU9O"
    "OiAoQl9WU19EX0JfVE9LRU5fU0hBMjU2LCBCX1ZTX0RfQl9QQVlMT0FEX1NIQTI1NiksCiAg"
    "ICAgICAgQl9WU19EX0RfQ09ORElUSU9OOiAoQl9WU19EX0RfVE9LRU5fU0hBMjU2LCBCX1ZT"
    "X0RfRF9QQVlMT0FEX1NIQTI1NiksCiAgICB9CiAgICBmb3IgY29uZGl0aW9uX2lkLCByb3dz"
    "IGluIGNvbmRpdGlvbl9yb3dzLml0ZW1zKCk6CiAgICAgICAgZXhwZWN0ZWRfdG9rZW4sIGV4"
    "cGVjdGVkX3BheWxvYWQgPSBleHBlY3RlZF9pZGVudGl0aWVzW2NvbmRpdGlvbl9pZF0KICAg"
    "ICAgICBvYnNlcnZlZF90b2tlbnMgPSB7KHJvdy5nZXQoInRva2VuX2NvdW50IiksIHJvdy5n"
    "ZXQoInRva2VuX3NoYTI1NiIpKSBmb3Igcm93IGluIHJvd3N9CiAgICAgICAgaWYgb2JzZXJ2"
    "ZWRfdG9rZW5zICE9IHsoQl9WU19EX1BST01QVF9UT0tFTl9DT1VOVCwgZXhwZWN0ZWRfdG9r"
    "ZW4pfToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAg"
    "ICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJCLXZz"
    "LUQgY29uZGl0aW9uIHRva2VuIGlkZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAg"
    "ICAgICAgICAgICkKICAgICAgICBpZiB7cm93LmdldCgicGF5bG9hZF9zaGEyNTYiKSBmb3Ig"
    "cm93IGluIHJvd3N9ICE9IHtleHBlY3RlZF9wYXlsb2FkfToKICAgICAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19G"
    "QUlMVVJFIiwKICAgICAgICAgICAgICAgICJCLXZzLUQgY29uZGl0aW9uIHBheWxvYWQgaWRl"
    "bnRpdHkgZmFpbGVkIHJlY29uY2lsaWF0aW9uIiwKICAgICAgICAgICAgKQogICAgZXhhY3Rf"
    "Y291bnRzID0gewogICAgICAgIGNvbmRpdGlvbl9pZDogc3VtKHJvdy5nZXQoImV4YWN0X29i"
    "amVjdCIpIGlzIFRydWUgZm9yIHJvdyBpbiByb3dzKQogICAgICAgIGZvciBjb25kaXRpb25f"
    "aWQsIHJvd3MgaW4gY29uZGl0aW9uX3Jvd3MuaXRlbXMoKQogICAgfQogICAgdmFsaWRfanNv"
    "bl9jb3VudHMgPSB7CiAgICAgICAgY29uZGl0aW9uX2lkOiBzdW0ocm93LmdldCgidmFsaWRf"
    "anNvbiIpIGlzIFRydWUgZm9yIHJvdyBpbiByb3dzKQogICAgICAgIGZvciBjb25kaXRpb25f"
    "aWQsIHJvd3MgaW4gY29uZGl0aW9uX3Jvd3MuaXRlbXMoKQogICAgfQogICAgc3RhdHVzZXMg"
    "PSB7CiAgICAgICAgY29uZGl0aW9uX2lkOiBtYXJrZXJfZGl2ZXJzaWZpZWRfY29uZGl0aW9u"
    "X3N0YXR1cyhleGFjdF9jb3VudCkKICAgICAgICBmb3IgY29uZGl0aW9uX2lkLCBleGFjdF9j"
    "b3VudCBpbiBleGFjdF9jb3VudHMuaXRlbXMoKQogICAgfQogICAgYl9leGFjdCA9IGV4YWN0"
    "X2NvdW50c1tCX1ZTX0RfQl9DT05ESVRJT05dCiAgICBkX2V4YWN0ID0gZXhhY3RfY291bnRz"
    "W0JfVlNfRF9EX0NPTkRJVElPTl0KICAgIGlmIGJfZXhhY3QgIT0gMDoKICAgICAgICBzdGF0"
    "ZSA9ICJCX0FOQ0hPUl9OT05SRVBST0RVQ1RJT05fSU5WQUxJREFURVNfSU5GRVJFTkNFIgog"
    "ICAgZWxpZiBkX2V4YWN0IGluIHsxLCAyfToKICAgICAgICBzdGF0ZSA9ICJEX0NPTkRJVElP"
    "Tl9VTlNUQUJMRV9OT19NRUNIQU5JU1RJQ19DTEFJTSIKICAgIGVsaWYgZF9leGFjdCA9PSAz"
    "OgogICAgICAgIHN0YXRlID0gIk1BUktFUl9ESVZFUlNJRklDQVRJT05fUkVTVE9SRVNfQkVI"
    "QVZJT1JfVU5ERVJfQ1VNVUxBVElWRV9MRU5HVEhfTE9DSyIKICAgIGVsaWYgZF9leGFjdCA9"
    "PSAwOgogICAgICAgIHN0YXRlID0gIk1BUktFUl9ESVZFUlNJRklDQVRJT05fSU5TVUZGSUNJ"
    "RU5UX0FUX0RfUkVQRVRJVElPTl9MRVZFTCIKICAgIGVsc2U6CiAgICAgICAgcmFpc2UgRGlh"
    "Z25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAiQi12cy1EIG91dGNvbWUgaXMgb3V0c2lkZSB0aGUgZnJvemVuIGRl"
    "Y2lzaW9uIGNvbnRyYWN0IiwKICAgICAgICApCiAgICByZXR1cm4gewogICAgICAgICJzY2hl"
    "bWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgInN0YXR1cyI6ICJERUNJREVEIiwKICAg"
    "ICAgICAiZGVjaXNpb25fc3RhdGUiOiBzdGF0ZSwKICAgICAgICAicHJpbWFyeV9lbmRwb2lu"
    "dCI6ICJleGFjdF9vYmplY3QiLAogICAgICAgICJjb25kaXRpb25fZXhhY3Rfb2JqZWN0X2Nv"
    "dW50cyI6IGV4YWN0X2NvdW50cywKICAgICAgICAiY29uZGl0aW9uX3ZhbGlkX2pzb25fY291"
    "bnRzIjogdmFsaWRfanNvbl9jb3VudHMsCiAgICAgICAgImNvbmRpdGlvbl9lbmRwb2ludF9z"
    "dGF0dXNlcyI6IHN0YXR1c2VzLAogICAgICAgICJiX2FuY2hvcl9yZXF1aXJlZF9leGFjdF9v"
    "YmplY3RfY291bnQiOiAwLAogICAgICAgICJiX2FuY2hvcl9yZXByb2R1Y2VkIjogYl9leGFj"
    "dCA9PSAwLAogICAgICAgICJtZWNoYW5pc3RpY19pbmZlcmVuY2VfcGVybWl0dGVkIjogYl9l"
    "eGFjdCA9PSAwIGFuZCBkX2V4YWN0IGluIHswLCAzfSwKICAgICAgICAiZnJlc2hfd29ya2Vy"
    "X3Byb2Nlc3NfcGVyX29ic2VydmF0aW9uIjogVHJ1ZSwKICAgICAgICAid29ya2VyX2lkZW50"
    "aXR5X2NhcmRpbmFsaXR5IjogbGVuKHdvcmtlcl9pZGVudGl0aWVzKSwKICAgICAgICAiYWxs"
    "X2NvbmRpdGlvbl90b2tlbl9pZGVudGl0aWVzX21hdGNoZWQiOiBUcnVlLAogICAgICAgICJh"
    "bGxfY29uZGl0aW9uX3BheWxvYWRfaWRlbnRpdGllc19tYXRjaGVkIjogVHJ1ZSwKICAgICAg"
    "ICAiY29tcGxldGVfY3VtdWxhdGl2ZV9wcm9tcHRfdG9rZW5fcHJvZmlsZV9sb2NrZWQiOiBU"
    "cnVlLAogICAgICAgICJ0ZXh0X2JvdW5kYXJ5X3Rva2VuX2JvdW5kYXJ5X2Fzc3VtcHRpb25f"
    "dXNlZCI6IEZhbHNlLAogICAgICAgICJtYXJrZXJfbGV4aWNhbF9zZW1hbnRpY19ub3ZlbHR5"
    "X2JvdW5kZWRfbm90X2VsaW1pbmF0ZWQiOiBUcnVlLAogICAgICAgICJleGFjdF9uZ3JhbV9i"
    "bG9ja19hbmRfcGVyaW9kaWNpdHlfZWZmZWN0c19ub3RfaW5kaXZpZHVhbGx5X2lzb2xhdGVk"
    "IjogVHJ1ZSwKICAgICAgICAiY2FjaGVfdGVsZW1ldHJ5X2lzX2RpYWdub3N0aWNfb25seSI6"
    "IFRydWUsCiAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAi"
    "cmF3X291dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgfQoKCmRlZiBtYXJrZXJfZGl2ZXJz"
    "aWZpZWRfZmFpbHVyZV9yZWNvcmQoCiAgICBlcnJvcjogRXhjZXB0aW9uLAogICAgYWN0aXZl"
    "X2ZhaWx1cmVfY29kZTogc3RyLAogICAgZmFpbGVkX3N0YWdlOiBzdHIsCiAgICBjb21wbGV0"
    "ZWRfcmVxdWVzdHM6IGludCwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIGlzaW5z"
    "dGFuY2UoZXJyb3IsIERpYWdub3N0aWNBbWJpZ3VpdHkpOgogICAgICAgIGZhaWx1cmVfY2xh"
    "c3MgPSBlcnJvci5mYWlsdXJlX2NsYXNzCiAgICAgICAgZGV0YWlsX2NvZGUgPSBOb25lCiAg"
    "ICAgICAgc2FmZV9tZXNzYWdlID0gZXJyb3Iuc2FmZV9tZXNzYWdlCiAgICBlbGlmIGlzaW5z"
    "dGFuY2UoZXJyb3IsIERpYWdub3N0aWNGYWlsdXJlKToKICAgICAgICBkZXRhaWxfY29kZSA9"
    "IGVycm9yLmVycm9yX2NvZGUKICAgICAgICBmYWlsdXJlX2NsYXNzID0gY2xhc3NpZnlfZmFp"
    "bHVyZV9kZXRhaWwoZGV0YWlsX2NvZGUpCiAgICAgICAgc2FmZV9tZXNzYWdlID0gZXJyb3Iu"
    "c2FmZV9tZXNzYWdlCiAgICBlbHNlOgogICAgICAgIGRldGFpbF9jb2RlID0gYWN0aXZlX2Zh"
    "aWx1cmVfY29kZQogICAgICAgIGZhaWx1cmVfY2xhc3MgPSBjbGFzc2lmeV9mYWlsdXJlX2Rl"
    "dGFpbChkZXRhaWxfY29kZSkKICAgICAgICBzYWZlX21lc3NhZ2UgPSBzYW5pdGl6ZV9leGNl"
    "cnB0KHN0cihlcnJvcikpWzo1MTJdIG9yIHR5cGUoZXJyb3IpLl9fbmFtZV9fCiAgICByZXR1"
    "cm4gewogICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgInN0YXR1"
    "cyI6ICJESUFHTk9TVElDX0lOVkFMSUQiLAogICAgICAgICJmYWlsZWRfc3RhZ2UiOiBmYWls"
    "ZWRfc3RhZ2UsCiAgICAgICAgImNvbXBsZXRlZF9yZXF1ZXN0cyI6IGNvbXBsZXRlZF9yZXF1"
    "ZXN0cywKICAgICAgICAiZmFpbHVyZV9jbGFzcyI6IGZhaWx1cmVfY2xhc3MsCiAgICAgICAg"
    "ImRldGFpbF9jb2RlIjogZGV0YWlsX2NvZGUsCiAgICAgICAgImVycm9yX3R5cGUiOiB0eXBl"
    "KGVycm9yKS5fX25hbWVfXywKICAgICAgICAic2FmZV9tZXNzYWdlIjogc2FmZV9tZXNzYWdl"
    "LAogICAgfQoKCmRlZiBtYXJrZXJfZGl2ZXJzaWZpZWRfYnVuZGxlX291dHB1dHMoKSAtPiBk"
    "aWN0W3N0ciwgb2JqZWN0XToKICAgIHJlcXVpcmVkX2JlZm9yZV9tYW5pZmVzdCA9IHNldChN"
    "QVJLRVJfRElWRVJTSUZJRURfT1VUUFVUX05BTUVTKSAtIHsiYnVuZGxlX21hbmlmZXN0X3Yx"
    "Lmpzb24ifQogICAgb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0ID0ge3BhdGgubmFtZSBmb3Ig"
    "cGF0aCBpbiBPVVRQVVRfUk9PVC5pdGVyZGlyKCkgaWYgcGF0aC5pc19maWxlKCl9CiAgICBt"
    "aXNzaW5nID0gcmVxdWlyZWRfYmVmb3JlX21hbmlmZXN0IC0gb2JzZXJ2ZWRfYmVmb3JlX21h"
    "bmlmZXN0CiAgICB1bmV4cGVjdGVkID0gb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0IC0gcmVx"
    "dWlyZWRfYmVmb3JlX21hbmlmZXN0CiAgICBpZiBtaXNzaW5nIG9yIHVuZXhwZWN0ZWQ6CiAg"
    "ICAgICAgcmFpc2UgUnVudGltZUVycm9yKAogICAgICAgICAgICAiQi12cy1EIGV2aWRlbmNl"
    "IG91dHB1dCBjb250cmFjdCBkcmlmdGVkOiAiCiAgICAgICAgICAgICsgY2Fub25pY2FsX2pz"
    "b24oCiAgICAgICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAgICAgIm1pc3NpbmciOiBz"
    "b3J0ZWQobWlzc2luZyksCiAgICAgICAgICAgICAgICAgICAgInVuZXhwZWN0ZWQiOiBzb3J0"
    "ZWQodW5leHBlY3RlZCksCiAgICAgICAgICAgICAgICB9CiAgICAgICAgICAgICkKICAgICAg"
    "ICApCiAgICBlbnRyaWVzID0gW10KICAgIGZvciBuYW1lIGluIE1BUktFUl9ESVZFUlNJRklF"
    "RF9PVVRQVVRfTkFNRVM6CiAgICAgICAgcGF0aCA9IE9VVFBVVF9ST09UIC8gbmFtZQogICAg"
    "ICAgIGlmIG5vdCBwYXRoLmlzX2ZpbGUoKToKICAgICAgICAgICAgY29udGludWUKICAgICAg"
    "ICBlbnRyaWVzLmFwcGVuZCgKICAgICAgICAgICAgewogICAgICAgICAgICAgICAgInBhdGgi"
    "OiBuYW1lLAogICAgICAgICAgICAgICAgInNoYTI1NiI6IGZpbGVfc2hhMjU2KHBhdGgpLAog"
    "ICAgICAgICAgICAgICAgInNpemVfYnl0ZXMiOiBwYXRoLnN0YXQoKS5zdF9zaXplLAogICAg"
    "ICAgICAgICB9CiAgICAgICAgKQogICAgbWFuaWZlc3QgPSB7CiAgICAgICAgInNjaGVtYV92"
    "ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAiZGlhZ25vc3RpY19pZCI6ICgKICAgICAgICAg"
    "ICAgImF1cmFnYXRld2F5LWItdnMtZC1jdW11bGF0aXZlLWxlbmd0aC1sb2NrZWQtbWFya2Vy"
    "LWRpdmVyc2lmaWVkLWRpZmZlcmVudGlhbC12MSIKICAgICAgICApLAogICAgICAgICJzb3Vy"
    "Y2VfbWFpbl9jb21taXQiOiBTT1VSQ0VfTUFJTl9DT01NSVQsCiAgICAgICAgIm1lbWJlcnMi"
    "OiBbaXRlbSBmb3IgaXRlbSBpbiBlbnRyaWVzIGlmIGl0ZW1bInBhdGgiXSAhPSAiYnVuZGxl"
    "X21hbmlmZXN0X3YxLmpzb24iXSwKICAgICAgICAic2NyYXRjaF9kaXJlY3Rvcmllc19pbmNs"
    "dWRlZCI6IEZhbHNlLAogICAgICAgICJ3b3JrZXJfbG9nX2RpcmVjdG9yeV9pbmNsdWRlZCI6"
    "IEZhbHNlLAogICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAg"
    "InJhd19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgIH0KICAgIHdyaXRlX2pzb24oT1VU"
    "UFVUX1JPT1QgLyAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iLCBtYW5pZmVzdCkKICAgIHdp"
    "dGggemlwZmlsZS5aaXBGaWxlKAogICAgICAgIEVWSURFTkNFX1pJUCwKICAgICAgICAidyIs"
    "CiAgICAgICAgY29tcHJlc3Npb249emlwZmlsZS5aSVBfREVGTEFURUQsCiAgICAgICAgY29t"
    "cHJlc3NsZXZlbD05LAogICAgKSBhcyBhcmNoaXZlOgogICAgICAgIGZvciBuYW1lIGluIE1B"
    "UktFUl9ESVZFUlNJRklFRF9PVVRQVVRfTkFNRVM6CiAgICAgICAgICAgIHBhdGggPSBPVVRQ"
    "VVRfUk9PVCAvIG5hbWUKICAgICAgICAgICAgaWYgcGF0aC5pc19maWxlKCk6CiAgICAgICAg"
    "ICAgICAgICBhcmNoaXZlLndyaXRlKHBhdGgsIGFyY25hbWU9bmFtZSkKICAgIGlmIEVWSURF"
    "TkNFX1pJUC5zdGF0KCkuc3Rfc2l6ZSA+IE1BWF9FVklERU5DRV9aSVBfQllURVM6CiAgICAg"
    "ICAgcmFpc2UgUnVudGltZUVycm9yKCJCLXZzLUQgZXZpZGVuY2UgWklQIGV4Y2VlZHMgdGhl"
    "IGJ5dGUgYnVkZ2V0IikKICAgIHJldHVybiB7CiAgICAgICAgImV2aWRlbmNlX3ppcCI6IHN0"
    "cihFVklERU5DRV9aSVApLAogICAgICAgICJldmlkZW5jZV96aXBfc2hhMjU2IjogZmlsZV9z"
    "aGEyNTYoRVZJREVOQ0VfWklQKSwKICAgICAgICAiZXZpZGVuY2VfemlwX3NpemVfYnl0ZXMi"
    "OiBFVklERU5DRV9aSVAuc3RhdCgpLnN0X3NpemUsCiAgICB9CgoKQzRfUVVBTElGSUNBVElP"
    "Tl9JRDogRmluYWwgPSAiQ0FOT05JQ0FMX1NZTlRIRVRJQ19QUkVGSVhfQzRfQkVIQVZJT1JB"
    "TF9RVUFMSUZJQ0FUSU9OX1YxIgpDNF9DQU5PTklDQUxfQ09SUFVTX1ZFUlNJT046IEZpbmFs"
    "ID0gIkNBTk9OSUNBTF9TWU5USEVUSUNfUFJFRklYX0NPUlBVU19WMSIKQzRfQ0FOT05JQ0FM"
    "X0NPUlBVU19TSEEyNTY6IEZpbmFsID0gKAogICAgIjE0MGU4MTU3ZGE4ODNlMDdmMmQ3NmQ0"
    "ZjUxNmVjMmJlZWM5NjFmZWZiNjM5Yjg1MDljYzhmM2E2MjM5ZDE0ZTkiCikKQzRfRlVMTF9Q"
    "Uk9NUFRfVE9LRU5fU0hBMjU2OiBGaW5hbCA9ICgKICAgICJmMDA5YjE0OWIwYjhjY2YwOGE0"
    "MjMzNDZlMTczNmJlODE5MjcwOTU5MDdiNTIyMWUwNzBlNTlmZmM2ZDg3ZjRjIgopCkM0X1JF"
    "VVNBQkxFX1BSRUZJWF9UT0tFTl9TSEEyNTY6IEZpbmFsID0gKAogICAgImYyOWFmNTRjYTQ2"
    "MjQ5ZmE2M2M3ZmQ4OWRhNDRjYTM3NWQ2NGYxODNmOGQ0NjNiM2E0MzY3ODMxODg5MGRmYjEi"
    "CikKQzRfUkVRVUVTVF9QQVlMT0FEX1NIQTI1NjogRmluYWwgPSAoCiAgICAiYTg4OGMxN2Vk"
    "OGU4MjM2MGZkZDQ2ZDBiYjY4MzNkYjlkYjJkYzNmYmJmYjE0ZDg2MWY3MDYzYjI3MTA2Mzc4"
    "OCIKKQpDNF9FWFBFQ1RFRF9PQkpFQ1RfU0hBMjU2OiBGaW5hbCA9ICgKICAgICI0NDhmYWQz"
    "ZDNhYzVjMmYxMWY0YzA5YjBkZjFlN2U2MjM3Y2UyYTA5MTg1Zjk5NTAzOTQ2MzExODc1ZjVl"
    "MTEzIgopCkM0X0ZJTkFMX1VTRVJfQk9VTkRBUllfU0VOVElORUw6IEZpbmFsID0gIl9fQVVS"
    "QUdBVEVXQVlfRklOQUxfVVNFUl9CT1VOREFSWV9TRU5USU5FTF9WMV9fIgpDNF9DQU5PTklD"
    "QUxfQ09OVEVYVDogRmluYWwgPSAoCiAgICAiQSBxdWlldCBzeW50aGV0aWMgbGFuZHNjYXBl"
    "IHN0cmV0Y2hlcyBhY3Jvc3MgYSBicm9hZCBwbGFpbiB3aGVyZSBsb3cgaGlsIgogICAgImxz"
    "IG1lZXQgYSBwYWxlIGhvcml6b24uIFRoZSBncm91bmQgYWx0ZXJuYXRlcyBiZXR3ZWVuIHNt"
    "b290aCBzdG9uZSwgZHJ5ICIKICAgICJncmFzcywgZGFyayBzb2lsLCBhbmQgbmFycm93IGJh"
    "bmRzIG9mIHNhbmQuIFNtYWxsIGNsdXN0ZXJzIG9mIHRyZWVzIGFwcGUiCiAgICAiYXIgYXQg"
    "aXJyZWd1bGFyIGRpc3RhbmNlcywgd2hpbGUgb3BlbiBzcGFjZXMgcmVtYWluIHdpZGUgZW5v"
    "dWdoIHRvIGtlZXAgIgogICAgImVhY2ggZmVhdHVyZSB2aXN1YWxseSBkaXN0aW5jdC4gTm90"
    "aGluZyBpbiB0aGUgc2NlbmUgYmVsb25ncyB0byBhIHBlcnNvbiIKICAgICIsIG9yZ2FuaXph"
    "dGlvbiwgb3IgcmVhbCBsb2NhdGlvbi4gSXQgaXMgc2ltcGx5IGEgZmljdGlvbmFsIGVudmly"
    "b25tZW50IGQiCiAgICAiZXNjcmliZWQgZm9yIG5ldXRyYWwgY29udGV4dC5cblxuTmVhciB0"
    "aGUgd2VzdGVybiBlZGdlLCBhIHNoYWxsb3cgc3RyZWFtIGIiCiAgICAiZW5kcyBhcm91bmQg"
    "cm91bmRlZCByb2NrcyBhbmQgcGFzc2VzIGJlbmVhdGggYSBzaW1wbGUgd29vZGVuIGZvb3Ri"
    "cmlkZ2UuIgogICAgIiBSZWVkcyBncm93IGFsb25nIG9uZSBiYW5rLCBtb3NzIGNvdmVycyBh"
    "IGZldyBzaGFkZWQgc3RvbmVzLCBhbmQgY2xlYXIgdyIKICAgICJhdGVyIHJlZmxlY3RzIHBh"
    "dGNoZXMgb2Ygc2t5IGJldHdlZW4gbW92aW5nIGNsb3Vkcy4gRmFydGhlciBkb3duc3RyZWFt"
    "LCAiCiAgICAidGhlIGNoYW5uZWwgYnJvYWRlbnMgaW50byBhIGNhbG0gcG9vbCBiZWZvcmUg"
    "bmFycm93aW5nIGFnYWluIGJlc2lkZSBhIGdyIgogICAgImF2ZWwgc2hlbGYuXG5cbkJleW9u"
    "ZCB0aGUgd2F0ZXIsIGEgbWl4ZWQgZ3JvdmUgY29udGFpbnMgY2VkYXIsIHdpbGxvdywgYmly"
    "IgogICAgImNoLCBhbmQgcGluZSBhbW9uZyBzbWFsbGVyIHNocnVicy4gRmFsbGVuIGxlYXZl"
    "cyBjb2xsZWN0IGluIHVuZXZlbiBwb2NrZSIKICAgICJ0cyB3aGVyZSB0aGUgZ3JvdW5kIGRp"
    "cHMgc2xpZ2h0bHksIHdoaWxlIGV4cG9zZWQgcm9vdHMgY3Jvc3Mgc2VjdGlvbnMgb2YiCiAg"
    "ICAiIGNvbXBhY3QgZWFydGguIFNvbWUgYnJhbmNoZXMgZXh0ZW5kIHRvd2FyZCBvcGVuIHN1"
    "bmxpZ2h0IGFuZCBvdGhlcnMgcmVtIgogICAgImFpbiBiZW5lYXRoIGEgbG9vc2UgY2Fub3B5"
    "LiBUaGUgYXJyYW5nZW1lbnQgaGFzIG5vIGVuY29kZWQgbWVhbmluZyBhbmQgZCIKICAgICJv"
    "ZXMgbm90IGNvcnJlc3BvbmQgdG8gYSByZWFsIG1hcC4gSXQgaXMgYSBmYWJyaWNhdGVkIGRl"
    "c2NyaXB0aW9uIHdob3NlIGQiCiAgICAiZXRhaWxzIGV4aXN0IG9ubHkgd2l0aGluIHRoaXMg"
    "cGFzc2FnZS5cblxuQSBsb3cgcmlkZ2UgcmlzZXMgdG8gdGhlIG5vcnRoIGEiCiAgICAibmQg"
    "aXMgY292ZXJlZCB3aXRoIHNjYXR0ZXJlZCBzdG9uZXMgb2YgZGlmZmVyZW50IHNoYXBlcyBh"
    "bmQgdGV4dHVyZXMuIFNvIgogICAgIm1lIHN1cmZhY2VzIGFyZSBjb2Fyc2UsIG90aGVycyBh"
    "cmUgc21vb3RoLCBhbmQgYSBmZXcgc2hvdyB0aGluIG1pbmVyYWwgYiIKICAgICJhbmRzLiBX"
    "aW5kIGhhcyBnYXRoZXJlZCBmaW5lIGR1c3QgYWxvbmcgc2hlbHRlcmVkIGVkZ2VzLCBsZWF2"
    "aW5nIG5lYXJieSAiCiAgICAicm9jayBmYWNlcyByZWxhdGl2ZWx5IGNsZWFuLiBUaGUgc2xv"
    "cGUgY2hhbmdlcyBncmFkdWFsbHkgcmF0aGVyIHRoYW4gZm9sIgogICAgImxvd2luZyBhIHJl"
    "cGVhdGVkIHBhdHRlcm4uIEF0IHRoZSBjcmVzdCwgdGhlIGxhbmQgb3BlbnMgaW50byBhIGJy"
    "b2FkIHZpZSIKICAgICJ3IG9mIHRoZSBwbGFpbiB3aXRob3V0IHJldmVhbGluZyBhbnkgbmFt"
    "ZWQgdG93biwgcm9hZCwgcHJvcGVydHksIG9yIGxhbmQiCiAgICAibWFyay5cblxuQWNyb3Nz"
    "IHRoZSBjZW50cmFsIGZpZWxkLCBzZXZlcmFsIG9yZGluYXJ5IHN0cnVjdHVyZXMgcHJvdmlk"
    "ZSBhZGQiCiAgICAiaXRpb25hbCB2aXN1YWwgdmFyaWV0eS4gQSBuYXJyb3cgc2hlbHRlciBo"
    "YXMgYSBmbGF0IHJvb2Ygc3VwcG9ydGVkIGJ5IHBsIgogICAgImFpbiBwb3N0cywgYW4gZW1w"
    "dHkgcGxhdGZvcm0gcmVzdHMgYmVzaWRlIGEgc2hvcnQgcGF0aCwgYW5kIGEgc21hbGwgc3Rv"
    "ciIKICAgICJhZ2Ugc2hlZCBzdGFuZHMgbmVhciBhIHBhdGNoIG9mIGdyYXZlbC4gVGhlaXIg"
    "bWF0ZXJpYWxzIGluY2x1ZGUgdW50cmVhdGUiCiAgICAiZCB0aW1iZXIsIGR1bGwgbWV0YWws"
    "IGNsYXkgYnJpY2ssIGFuZCB0cmFuc2x1Y2VudCBwYW5lbHMuIE5vbmUgb2YgdGhlc2UgIgog"
    "ICAgInN0cnVjdHVyZXMgaGFzIGEgY29tbWVyY2lhbCBwdXJwb3NlLCBvd25lciwgYWRkcmVz"
    "cywgc2VyaWFsIG51bWJlciwgb3IgbyIKICAgICJwZXJhdGlvbmFsIHJvbGUuIFRoZXkgYXJl"
    "IG5ldXRyYWwgZmljdGlvbmFsIG9iamVjdHMgYXJyYW5nZWQgaW5zaWRlIHRoZSAiCiAgICAi"
    "c2FtZSBzeW50aGV0aWMgc2V0dGluZy5cblxuVGhlIGVhc3Rlcm4gc2lkZSBjaGFuZ2VzIGdy"
    "YWR1YWxseSBmcm9tIG9wZW4gZ3IiCiAgICAiYXNzIHRvIGZpcm1lciBncm91bmQgd2l0aCBz"
    "Y2F0dGVyZWQgcGViYmxlcyBhbmQgbG93IGZsb3dlcmluZyBwbGFudHMuIEEgIgogICAgImxp"
    "bmUgb2YgdGFsbGVyIHZlZ2V0YXRpb24gYnJlYWtzIHRoZSB3aW5kIHdpdGhvdXQgZm9ybWlu"
    "ZyBhIHN0cmljdCBib3VuZCIKICAgICJhcnkuIE5lYXJieSwgYSBzaGFsbG93IGRlcHJlc3Np"
    "b24gaG9sZHMgcmFpbndhdGVyIGFmdGVyIHBhc3Npbmcgd2VhdGhlciwiCiAgICAiIGFuZCB0"
    "aGUgZGFtcCBzb2lsIGFyb3VuZCBpdCBpcyBkYXJrZXIgdGhhbiB0aGUgc3Vycm91bmRpbmcg"
    "ZWFydGguIEluc2VjIgogICAgInRzIG1vdmUgYmV0d2VlbiBwbGFudHMsIGJpcmRzIGNyb3Nz"
    "IG92ZXJoZWFkLCBhbmQgZGlzdGFudCBicmFuY2hlcyBzaGlmdCIKICAgICIgZ2VudGx5Llxu"
    "XG5MaWdodCBjaGFuZ2VzIGFjcm9zcyB0aGUgc2NlbmUgYXMgY2xvdWRzIHBhc3Mgb3Zlcmhl"
    "YWQuIEJyaWdodCIKICAgICIgYXJlYXMgbW92ZSBvdmVyIHRoZSBmaWVsZCwgdGhlbiBzb2Z0"
    "ZW4gYXMgdGhpbiBzaGFkZSByZWFjaGVzIHRoZSByaWRnZSAiCiAgICAiYW5kIHN0cmVhbS4g"
    "VGhlIHRlbXBlcmF0dXJlIGlzIG1pbGQsIHZpc2liaWxpdHkgaXMgY2xlYXIsIGFuZCB0aGVy"
    "ZSBpcyBuIgogICAgIm8gc3Rvcm0sIGhhemFyZCwgZW1lcmdlbmN5LCBvciB1bnVzdWFsIGFj"
    "dGl2aXR5LiBTaGFkb3dzIHZhcnkgaW4gbGVuZ3RoICIKICAgICJiZWNhdXNlIG9iamVjdHMg"
    "c3RhbmQgYXQgZGlmZmVyZW50IGFuZ2xlcyBhbmQgaGVpZ2h0cy4gVGhlIHNjZW5lIHJlbWFp"
    "bnMiCiAgICAiIGNhbG0sIGZpY3Rpb25hbCwgYW5kIGludGVybmFsbHkgY29uc2lzdGVudCB3"
    "aXRob3V0IGRlcGVuZGluZyBvbiBhIHJlcGVhIgogICAgInRlZCBzZW50ZW5jZSBmcmFtZSBv"
    "ciBhIHNlcXVlbmNlIG9mIG51bWJlcmVkIHNlY3Rpb25zLlxuXG5DbG9zZSB0byB0aGUgZm9v"
    "IgogICAgInRicmlkZ2UsIHRoZSBwYXRoIGRpdmlkZXMgbmF0dXJhbGx5IGFyb3VuZCBhIGxh"
    "cmdlIHN0b25lIGFuZCByZWpvaW5zIGJlcyIKICAgICJpZGUgYSBzdGFuZCBvZiB0YWxsIGdy"
    "YXNzLiBPbmUgYnJhbmNoIGlzIHBhY2tlZCBlYXJ0aCwgd2hpbGUgdGhlIG90aGVyIGMiCiAg"
    "ICAib250YWlucyBsb29zZSBncmF2ZWwgbWl4ZWQgd2l0aCBzbWFsbCBwYWxlIGZyYWdtZW50"
    "cy4gTmVpdGhlciBicmFuY2ggbGVhIgogICAgImRzIHRvIGEgZGVzdGluYXRpb24gd2l0aCBh"
    "IG5hbWUuIFRoZXkgc2ltcGx5IHByb3ZpZGUgdHdvIGRpZmZlcmVudCBzdXJmYSIKICAgICJj"
    "ZXMgd2l0aGluIHRoZSBpbWFnaW5hcnkgbGFuZHNjYXBlLiBOZWFyYnksIGEgd2VhdGhlcmVk"
    "IGJlbmNoIGZhY2VzIHRoZSAiCiAgICAid2F0ZXIsIGJ1dCBubyBwbGFxdWUsIG1lc3NhZ2Us"
    "IGluc2NyaXB0aW9uLCBvciBwZXJzb25hbCBkZXRhaWwgYXBwZWFycyBvIgogICAgIm4gaXQu"
    "XG5cbkZhcnRoZXIgc291dGgsIHRoZSBwbGFpbiBiZWNvbWVzIHNsaWdodGx5IHVuZXZlbiBh"
    "bmQgaW5jbHVkZXMgc2hhIgogICAgImxsb3cgdGVycmFjZXMgc2hhcGVkIGJ5IG9yZGluYXJ5"
    "IGVyb3Npb24uIFNob3J0IGdyYXNzZXMgZ3JvdyBkZW5zZWx5IG9uICIKICAgICJsZXZlbCBn"
    "cm91bmQgYW5kIHRoaW4gb3V0IHdoZXJlIHN0b25lcyBicmVhayB0aGUgc3VyZmFjZS4gQSBm"
    "ZXcgaXNvbGF0ZWQiCiAgICAiIHNocnVicyBvY2N1cHkgc2hlbHRlcmVkIHBvY2tldHMsIHdo"
    "aWxlIGV4cG9zZWQgYXJlYXMgcmVtYWluIG1vc3RseSBiYXJlIgogICAgIi4gVGhlIGNoYW5n"
    "ZSBpbiB0ZXJyYWluIGlzIGdyYWR1YWwsIG5vdCBzeW1tZXRyaWNhbCwgYW5kIG5vdCBkZXJp"
    "dmVkIGZybyIKICAgICJtIGNvcHlpbmcgb25lIGFycmFuZ2VtZW50IG1hbnkgdGltZXMuXG5c"
    "bkF0IHRoZSBvdXRlciBlZGdlIG9mIHRoZSBzZXR0aW5nLCIKICAgICIgZGlzdGFudCBoaWxs"
    "cyBmYWRlIGludG8gYSBsaWdodCBoYXplLiBUaGVpciBvdXRsaW5lcyBhcmUgc29mdCBhbmQg"
    "aXJyZWciCiAgICAidWxhciwgd2l0aCBubyBkaXN0aW5jdGl2ZSBmZWF0dXJlIGludGVuZGVk"
    "IHRvIGlkZW50aWZ5IGEgcmVhbCBwbGFjZS4gVGhlIgogICAgIiBza3kgYWJvdmUgdGhlbSBj"
    "b250YWlucyBicm9hZCBjbG91ZCBsYXllcnMgc2VwYXJhdGVkIGJ5IGNsZWFyIGdhcHMuIE5v"
    "ICIKICAgICJhaXJjcmFmdCwgdmVoaWNsZXMsIGFkdmVydGlzZW1lbnRzLCBpbnRlcmZhY2Vz"
    "LCBkb2N1bWVudHMsIG9yIHNjcmVlbnMgYXAiCiAgICAicGVhciBhbnl3aGVyZSBpbiB0aGUg"
    "c2NlbmUuIFJldHVybiBvbmx5IHRoZSBleGFjdCBKU09OIG9iamVjdCBzdXBwbGllZCBpIgog"
    "ICAgIm4gdGhlIGZpbmFsIHVzZXIgbWVzc2FnZSwgd2l0aCBubyBtYXJrZG93biBvciBhZGRp"
    "dGlvbmFsIHRleHQuIgopCkM0X0ZVTExfUFJPTVBUX1RPS0VOX0NPVU5UOiBGaW5hbCA9IDg5"
    "OQpDNF9SRVVTQUJMRV9QUkVGSVhfVE9LRU5fQ09VTlQ6IEZpbmFsID0gODgwCkM0X09CU0VS"
    "VkFUSU9OX0NPVU5UOiBGaW5hbCA9IDMKQzRfT1VUUFVUX05BTUVTOiBGaW5hbCA9ICgKICAg"
    "ICJydW50aW1lX3NvdXJjZV9pZGVudGl0eV9yZXBvcnRfdjEuanNvbiIsCiAgICAicnVudGlt"
    "ZV9pbnN0YWxsX3JlcG9ydF92MS5qc29uIiwKICAgICJydW50aW1lX2Vudmlyb25tZW50X3Jl"
    "cG9ydF92MS5qc29uIiwKICAgICJydW50aW1lX2ltcG9ydF9jbG9zdXJlX3JlcG9ydF92MS5q"
    "c29uIiwKICAgICJjNF9ydW50aW1lX3JlYWR5X3YxLmpzb24iLAogICAgInByZV9yZXF1ZXN0"
    "X3Rva2VuX2lkZW50aXR5X2pvdXJuYWxfdjEuanNvbiIsCiAgICAiYzRfcmVxdWVzdF9yZXN1"
    "bHRzX3YxLmpzb24iLAogICAgImM0X2RlY2lzaW9uX3YxLmpzb24iLAogICAgIndvcmtlcl90"
    "ZWFyZG93bl9yZXBvcnRfdjEuanNvbiIsCiAgICAic2NyYXRjaF9jbGVhbnVwX3JlcG9ydF92"
    "MS5qc29uIiwKICAgICJmYWlsdXJlX3JlcG9ydF92MS5qc29uIiwKICAgICJjNF9zdW1tYXJ5"
    "X3YxLmpzb24iLAogICAgImh1bWFuX3JlcG9ydF92MS5tZCIsCiAgICAiYnVuZGxlX21hbmlm"
    "ZXN0X3YxLmpzb24iLAopCgoKY2xhc3MgQzREdXBsaWNhdGVLZXlFcnJvcihWYWx1ZUVycm9y"
    "KToKICAgIHBhc3MKCgpPUkRFUl9ESUFHTk9TVElDX0lEOiBGaW5hbCA9ICJDNF9QQVJBR1JB"
    "UEhfT1JERVJfQkVIQVZJT1JBTF9ESUZGRVJFTlRJQUxfVjEiCk9SREVSX0NPTlRST0xfQ09O"
    "RElUSU9OOiBGaW5hbCA9ICJDT05UUk9MX09SSUdJTkFMX0M0IgpPUkRFUl9UUkVBVE1FTlRf"
    "Q09ORElUSU9OOiBGaW5hbCA9ICJUUkVBVE1FTlRfUkVWRVJTRURfTUlERExFX0VJR0hUIgpP"
    "UkRFUl9SRVFVRVNUX09SREVSOiBGaW5hbCA9ICgKICAgIE9SREVSX0NPTlRST0xfQ09ORElU"
    "SU9OLAogICAgT1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTiwKICAgIE9SREVSX1RSRUFUTUVO"
    "VF9DT05ESVRJT04sCiAgICBPUkRFUl9DT05UUk9MX0NPTkRJVElPTiwKICAgIE9SREVSX0NP"
    "TlRST0xfQ09ORElUSU9OLAogICAgT1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTiwKKQpPUkRF"
    "Ul9QUk9NUFRfVE9LRU5fQ09VTlQ6IEZpbmFsID0gODk5Ck9SREVSX0ZJTkFMX1VTRVJfQk9V"
    "TkRBUlk6IEZpbmFsID0gODgwCk9SREVSX0NPTlRST0xfVE9LRU5fU0hBMjU2OiBGaW5hbCA9"
    "ICgKICAgICJmMDA5YjE0OWIwYjhjY2YwOGE0MjMzNDZlMTczNmJlODE5MjcwOTU5MDdiNTIy"
    "MWUwNzBlNTlmZmM2ZDg3ZjRjIgopCk9SREVSX1RSRUFUTUVOVF9UT0tFTl9TSEEyNTY6IEZp"
    "bmFsID0gKAogICAgIjE0ZDZhNjg1NmZmYjVjNGNhYTRhNGVkMjI5ZmEwYzk0YWMwNmI4NmZi"
    "ZWY0NzNiZTAwMWRkNmQ4ZTM2OThjY2UiCikKT1JERVJfQ09OVFJPTF9QQVlMT0FEX1NIQTI1"
    "NjogRmluYWwgPSAoCiAgICAiYTg4OGMxN2VkOGU4MjM2MGZkZDQ2ZDBiYjY4MzNkYjlkYjJk"
    "YzNmYmJmYjE0ZDg2MWY3MDYzYjI3MTA2Mzc4OCIKKQpPUkRFUl9UUkVBVE1FTlRfUEFZTE9B"
    "RF9TSEEyNTY6IEZpbmFsID0gKAogICAgIjQ3YzUxOWMyNGVmZDQwZTNiYWI0YmZhMmVhZWMx"
    "Y2YzZDYyYzkxYTY0ODg3MGU2MzE3MjE2MjU1NjdmMjBiNWUiCikKT1JERVJfSElTVE9SSUNB"
    "TF9DT05UUk9MX1BBUlNFRF9PQkpFQ1RfU0hBMjU2OiBGaW5hbCA9ICgKICAgICJmYjhjYmZk"
    "ZTBmZmVmZjQ4YzQ3NzNjZWU5NWM1NzZmODIxYjIyZjg0YjAwZGMxMDU5NDEwODU2NTAyMjU2"
    "YWJhIgopCk9SREVSX1RSRUFUTUVOVF9QQVJBR1JBUEhfT1JERVI6IEZpbmFsID0gKDEsIDks"
    "IDgsIDcsIDYsIDUsIDQsIDMsIDIsIDEwKQpPUkRFUl9PVVRQVVRfTkFNRVM6IEZpbmFsID0g"
    "KAogICAgInJ1bnRpbWVfc291cmNlX2lkZW50aXR5X3JlcG9ydF92MS5qc29uIiwKICAgICJy"
    "dW50aW1lX2luc3RhbGxfcmVwb3J0X3YxLmpzb24iLAogICAgInJ1bnRpbWVfZW52aXJvbm1l"
    "bnRfcmVwb3J0X3YxLmpzb24iLAogICAgInJ1bnRpbWVfaW1wb3J0X2Nsb3N1cmVfcmVwb3J0"
    "X3YxLmpzb24iLAogICAgImM0X3BhcmFncmFwaF9vcmRlcl9ydW50aW1lX3JlYWR5X3YxLmpz"
    "b24iLAogICAgInByZV9yZXF1ZXN0X3Rva2VuX2lkZW50aXR5X2pvdXJuYWxfdjEuanNvbiIs"
    "CiAgICAiYzRfcGFyYWdyYXBoX29yZGVyX3JlcXVlc3RfcmVzdWx0c192MS5qc29uIiwKICAg"
    "ICJjNF9wYXJhZ3JhcGhfb3JkZXJfZGVjaXNpb25fdjEuanNvbiIsCiAgICAid29ya2VyX3Rl"
    "YXJkb3duX3JlcG9ydF92MS5qc29uIiwKICAgICJzY3JhdGNoX2NsZWFudXBfcmVwb3J0X3Yx"
    "Lmpzb24iLAogICAgImZhaWx1cmVfcmVwb3J0X3YxLmpzb24iLAogICAgImM0X3BhcmFncmFw"
    "aF9vcmRlcl9zdW1tYXJ5X3YxLmpzb24iLAogICAgImh1bWFuX3JlcG9ydF92MS5tZCIsCiAg"
    "ICAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iLAopCgoKZGVmIG9yZGVyX2NvbnRleHQoY29u"
    "ZGl0aW9uX2lkOiBzdHIpIC0+IHN0cjoKICAgIGlmIG5vdCBDNF9DQU5PTklDQUxfQ09OVEVY"
    "VC5lbmRzd2l0aChTWVNURU1fUFJPTVBUKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAg"
    "ICAgICJjYW5vbmljYWwgQzQgaW5zdHJ1Y3Rpb24gc3VmZml4IGRyaWZ0ZWQiLAogICAgICAg"
    "ICkKCiAgICBiZWZvcmVfaW5zdHJ1Y3Rpb24gPSBDNF9DQU5PTklDQUxfQ09OVEVYVFs6IC1s"
    "ZW4oU1lTVEVNX1BST01QVCldCiAgICBib2R5ID0gYmVmb3JlX2luc3RydWN0aW9uLnJzdHJp"
    "cCgiIFx0XHJcbiIpCiAgICBzZXBhcmF0b3IgPSBiZWZvcmVfaW5zdHJ1Y3Rpb25bbGVuKGJv"
    "ZHkpIDpdCgogICAgaWYgc2VwYXJhdG9yICE9ICIgIjoKICAgICAgICByYWlzZSBEaWFnbm9z"
    "dGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAg"
    "ICAgICAgICAgICJjYW5vbmljYWwgQzQgaW5zdHJ1Y3Rpb24gc2VwYXJhdG9yIGRyaWZ0ZWQi"
    "LAogICAgICAgICkKCiAgICBwYXJhZ3JhcGhzID0gYm9keS5zcGxpdCgiXG5cbiIpCiAgICBp"
    "ZiBsZW4ocGFyYWdyYXBocykgIT0gMTA6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1"
    "cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAiY2Fub25pY2FsIEM0IHBhcmFncmFwaCBjb3VudCBkcmlmdGVkIiwKICAgICAgICApCgog"
    "ICAgaWYgY29uZGl0aW9uX2lkID09IE9SREVSX0NPTlRST0xfQ09ORElUSU9OOgogICAgICAg"
    "IHJldHVybiBDNF9DQU5PTklDQUxfQ09OVEVYVAoKICAgIGlmIGNvbmRpdGlvbl9pZCA9PSBP"
    "UkRFUl9UUkVBVE1FTlRfQ09ORElUSU9OOgogICAgICAgIHJlb3JkZXJlZCA9IFsKICAgICAg"
    "ICAgICAgcGFyYWdyYXBoc1swXSwKICAgICAgICAgICAgKnJldmVyc2VkKHBhcmFncmFwaHNb"
    "MTo5XSksCiAgICAgICAgICAgIHBhcmFncmFwaHNbOV0sCiAgICAgICAgXQogICAgICAgIGNh"
    "bmRpZGF0ZSA9ICJcblxuIi5qb2luKHJlb3JkZXJlZCkgKyBzZXBhcmF0b3IgKyBTWVNURU1f"
    "UFJPTVBUCiAgICAgICAgaWYgbGVuKGNhbmRpZGF0ZSkgIT0gbGVuKEM0X0NBTk9OSUNBTF9D"
    "T05URVhUKToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJw"
    "YXJhZ3JhcGgtb3JkZXIgaW50ZXJ2ZW50aW9uIGNoYW5nZWQgY2hhcmFjdGVyIGNvdW50IiwK"
    "ICAgICAgICAgICAgKQogICAgICAgIHJldHVybiBjYW5kaWRhdGUKCiAgICByYWlzZSBEaWFn"
    "bm9zdGljRmFpbHVyZSgKICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAg"
    "ICAgICAidW5zdXBwb3J0ZWQgcGFyYWdyYXBoLW9yZGVyIGNvbmRpdGlvbiIsCiAgICApCgoK"
    "ZGVmIG9yZGVyX3JlcXVlc3RfbWVzc2FnZXMoCiAgICBjb25kaXRpb25faWQ6IHN0ciwKICAg"
    "IGZpbmFsX3VzZXJfY29udGVudDogc3RyLAopIC0+IGxpc3RbZGljdFtzdHIsIHN0cl1dOgog"
    "ICAgcmV0dXJuIFsKICAgICAgICB7InJvbGUiOiAic3lzdGVtIiwgImNvbnRlbnQiOiBTWVNU"
    "RU1fUFJPTVBUfSwKICAgICAgICB7InJvbGUiOiAidXNlciIsICJjb250ZW50Ijogb3JkZXJf"
    "Y29udGV4dChjb25kaXRpb25faWQpfSwKICAgICAgICB7InJvbGUiOiAiYXNzaXN0YW50Iiwg"
    "ImNvbnRlbnQiOiBTWU5USEVUSUNfQVNTSVNUQU5UX0FDS30sCiAgICAgICAgeyJyb2xlIjog"
    "InVzZXIiLCAiY29udGVudCI6IGZpbmFsX3VzZXJfY29udGVudH0sCiAgICBdCgoKZGVmIG9y"
    "ZGVyX3JlcXVlc3RfcGF5bG9hZCgKICAgIGNvbmRpdGlvbl9pZDogc3RyLAopIC0+IGRpY3Rb"
    "c3RyLCBvYmplY3RdOgogICAgcmV0dXJuIHsKICAgICAgICAibW9kZWwiOiBTRVJWRURfTU9E"
    "RUxfTkFNRSwKICAgICAgICAibWVzc2FnZXMiOiBvcmRlcl9yZXF1ZXN0X21lc3NhZ2VzKAog"
    "ICAgICAgICAgICBjb25kaXRpb25faWQsCiAgICAgICAgICAgIEVYUEVDVEVEX09CSkVDVF9D"
    "QU5PTklDQUwsCiAgICAgICAgKSwKICAgICAgICAidGVtcGVyYXR1cmUiOiAwLAogICAgICAg"
    "ICJ0b3BfcCI6IDEsCiAgICAgICAgInJlcGV0aXRpb25fcGVuYWx0eSI6IDEuMSwKICAgICAg"
    "ICAic2VlZCI6IDcsCiAgICAgICAgIm1heF90b2tlbnMiOiAzMiwKICAgICAgICAic3RyZWFt"
    "IjogRmFsc2UsCiAgICB9CgoKZGVmIG9yZGVyX3Rva2VuaXplX3BheWxvYWQoCiAgICBjb25k"
    "aXRpb25faWQ6IHN0ciwKICAgIGZpbmFsX3VzZXJfY29udGVudDogc3RyLAopIC0+IGRpY3Rb"
    "c3RyLCBvYmplY3RdOgogICAgcmV0dXJuIHsKICAgICAgICAibW9kZWwiOiBTRVJWRURfTU9E"
    "RUxfTkFNRSwKICAgICAgICAibWVzc2FnZXMiOiBvcmRlcl9yZXF1ZXN0X21lc3NhZ2VzKAog"
    "ICAgICAgICAgICBjb25kaXRpb25faWQsCiAgICAgICAgICAgIGZpbmFsX3VzZXJfY29udGVu"
    "dCwKICAgICAgICApLAogICAgICAgICJhZGRfZ2VuZXJhdGlvbl9wcm9tcHQiOiBUcnVlLAog"
    "ICAgICAgICJjb250aW51ZV9maW5hbF9tZXNzYWdlIjogRmFsc2UsCiAgICAgICAgImFkZF9z"
    "cGVjaWFsX3Rva2VucyI6IEZhbHNlLAogICAgICAgICJyZXR1cm5fdG9rZW5fc3RycyI6IEZh"
    "bHNlLAogICAgfQoKCmRlZiBvcmRlcl90b2tlbl9zZXF1ZW5jZSgKICAgIHdvcmtlcjogV29y"
    "a2VyLAogICAgY29uZGl0aW9uX2lkOiBzdHIsCiAgICBmaW5hbF91c2VyX2NvbnRlbnQ6IHN0"
    "ciwKKSAtPiB0dXBsZVtpbnQsIC4uLl06CiAgICByZXNwb25zZSA9IHBvc3RfanNvbigKICAg"
    "ICAgICBmImh0dHA6Ly8xMjcuMC4wLjE6e3dvcmtlci5wb3J0fS90b2tlbml6ZSIsCiAgICAg"
    "ICAgb3JkZXJfdG9rZW5pemVfcGF5bG9hZCgKICAgICAgICAgICAgY29uZGl0aW9uX2lkLAog"
    "ICAgICAgICAgICBmaW5hbF91c2VyX2NvbnRlbnQsCiAgICAgICAgKSwKICAgICkKICAgIHJh"
    "d190b2tlbnMgPSByZXNwb25zZS5nZXQoInRva2VucyIpCiAgICBjb3VudCA9IHJlc3BvbnNl"
    "LmdldCgiY291bnQiKQogICAgaWYgKAogICAgICAgIG5vdCBpc2luc3RhbmNlKHJhd190b2tl"
    "bnMsIGxpc3QpCiAgICAgICAgb3IgaXNpbnN0YW5jZShjb3VudCwgYm9vbCkKICAgICAgICBv"
    "ciBub3QgaXNpbnN0YW5jZShjb3VudCwgaW50KQogICAgICAgIG9yIGNvdW50ICE9IGxlbihy"
    "YXdfdG9rZW5zKQogICAgKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICJwYXJh"
    "Z3JhcGgtb3JkZXIgdG9rZW5pemF0aW9uIHJlc3BvbnNlIHNoYXBlIGlzIGludmFsaWQiLAog"
    "ICAgICAgICkKCiAgICB0b2tlbnM6IGxpc3RbaW50XSA9IFtdCiAgICBmb3IgcmF3X3Rva2Vu"
    "IGluIHJhd190b2tlbnM6CiAgICAgICAgaWYgaXNpbnN0YW5jZShyYXdfdG9rZW4sIGJvb2wp"
    "IG9yIG5vdCBpc2luc3RhbmNlKHJhd190b2tlbiwgaW50KToKICAgICAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19G"
    "QUlMVVJFIiwKICAgICAgICAgICAgICAgICJwYXJhZ3JhcGgtb3JkZXIgdG9rZW5pemF0aW9u"
    "IHJldHVybmVkIGEgbm9uLWludGVnZXIgdG9rZW4gaWQiLAogICAgICAgICAgICApCiAgICAg"
    "ICAgaWYgcmF3X3Rva2VuIDwgMDoKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1"
    "cmUoCiAgICAgICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAg"
    "ICAgICAgICAgICJwYXJhZ3JhcGgtb3JkZXIgdG9rZW5pemF0aW9uIHJldHVybmVkIGEgbmVn"
    "YXRpdmUgdG9rZW4gaWQiLAogICAgICAgICAgICApCiAgICAgICAgdG9rZW5zLmFwcGVuZChy"
    "YXdfdG9rZW4pCgogICAgcmV0dXJuIHR1cGxlKHRva2VucykKCgpkZWYgb3JkZXJfZXhwZWN0"
    "ZWRfdG9rZW5fc2hhMjU2KGNvbmRpdGlvbl9pZDogc3RyKSAtPiBzdHI6CiAgICBpZiBjb25k"
    "aXRpb25faWQgPT0gT1JERVJfQ09OVFJPTF9DT05ESVRJT046CiAgICAgICAgcmV0dXJuIE9S"
    "REVSX0NPTlRST0xfVE9LRU5fU0hBMjU2CiAgICBpZiBjb25kaXRpb25faWQgPT0gT1JERVJf"
    "VFJFQVRNRU5UX0NPTkRJVElPTjoKICAgICAgICByZXR1cm4gT1JERVJfVFJFQVRNRU5UX1RP"
    "S0VOX1NIQTI1NgogICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgIkhBUk5F"
    "U1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgInVuc3VwcG9ydGVkIHBhcmFncmFwaC1v"
    "cmRlciB0b2tlbiBpZGVudGl0eSBjb25kaXRpb24iLAogICAgKQoKCmRlZiBvcmRlcl9leHBl"
    "Y3RlZF9wYXlsb2FkX3NoYTI1Nihjb25kaXRpb25faWQ6IHN0cikgLT4gc3RyOgogICAgaWYg"
    "Y29uZGl0aW9uX2lkID09IE9SREVSX0NPTlRST0xfQ09ORElUSU9OOgogICAgICAgIHJldHVy"
    "biBPUkRFUl9DT05UUk9MX1BBWUxPQURfU0hBMjU2CiAgICBpZiBjb25kaXRpb25faWQgPT0g"
    "T1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTjoKICAgICAgICByZXR1cm4gT1JERVJfVFJFQVRN"
    "RU5UX1BBWUxPQURfU0hBMjU2CiAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAidW5zdXBwb3J0ZWQgcGFy"
    "YWdyYXBoLW9yZGVyIHBheWxvYWQgaWRlbnRpdHkgY29uZGl0aW9uIiwKICAgICkKCgpkZWYg"
    "b3JkZXJfdG9rZW5faWRlbnRpdHkoCiAgICB3b3JrZXI6IFdvcmtlciwKICAgIGNvbmRpdGlv"
    "bl9pZDogc3RyLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgY2Fub25pY2FsX3Rva2Vu"
    "cyA9IG9yZGVyX3Rva2VuX3NlcXVlbmNlKAogICAgICAgIHdvcmtlciwKICAgICAgICBjb25k"
    "aXRpb25faWQsCiAgICAgICAgRVhQRUNURURfT0JKRUNUX0NBTk9OSUNBTCwKICAgICkKICAg"
    "IHNlbnRpbmVsX3Rva2VucyA9IG9yZGVyX3Rva2VuX3NlcXVlbmNlKAogICAgICAgIHdvcmtl"
    "ciwKICAgICAgICBjb25kaXRpb25faWQsCiAgICAgICAgQzRfRklOQUxfVVNFUl9CT1VOREFS"
    "WV9TRU5USU5FTCwKICAgICkKICAgIHJldXNhYmxlX3Rva2VucyA9IGM0X2xvbmdlc3RfY29t"
    "bW9uX3ByZWZpeCgKICAgICAgICBjYW5vbmljYWxfdG9rZW5zLAogICAgICAgIHNlbnRpbmVs"
    "X3Rva2VucywKICAgICkKCiAgICBjYW5vbmljYWxfc2hhMjU2ID0gc2hhMjU2X2J5dGVzKGNh"
    "bm9uaWNhbF9qc29uKGxpc3QoY2Fub25pY2FsX3Rva2VucykpLmVuY29kZSgidXRmLTgiKSkK"
    "ICAgIHJldXNhYmxlX3NoYTI1NiA9IHNoYTI1Nl9ieXRlcyhjYW5vbmljYWxfanNvbihsaXN0"
    "KHJldXNhYmxlX3Rva2VucykpLmVuY29kZSgidXRmLTgiKSkKCiAgICBpZiBsZW4oY2Fub25p"
    "Y2FsX3Rva2VucykgIT0gT1JERVJfUFJPTVBUX1RPS0VOX0NPVU5UOgogICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlM"
    "VVJFIiwKICAgICAgICAgICAgInBhcmFncmFwaC1vcmRlciBwcm9tcHQgdG9rZW4gY291bnQg"
    "ZHJpZnRlZCIsCiAgICAgICAgKQogICAgaWYgY2Fub25pY2FsX3NoYTI1NiAhPSBvcmRlcl9l"
    "eHBlY3RlZF90b2tlbl9zaGEyNTYoY29uZGl0aW9uX2lkKToKICAgICAgICByYWlzZSBEaWFn"
    "bm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJwYXJhZ3JhcGgtb3JkZXIgcHJvbXB0IHRva2VuIGlkZW50aXR5IGRy"
    "aWZ0ZWQiLAogICAgICAgICkKICAgIGlmIGxlbihyZXVzYWJsZV90b2tlbnMpICE9IE9SREVS"
    "X0ZJTkFMX1VTRVJfQk9VTkRBUlk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAi"
    "cGFyYWdyYXBoLW9yZGVyIGZpbmFsLXVzZXIgYm91bmRhcnkgZHJpZnRlZCIsCiAgICAgICAg"
    "KQoKICAgIGlmICgKICAgICAgICBjb25kaXRpb25faWQgPT0gT1JERVJfQ09OVFJPTF9DT05E"
    "SVRJT04KICAgICAgICBhbmQgcmV1c2FibGVfc2hhMjU2ICE9IEM0X1JFVVNBQkxFX1BSRUZJ"
    "WF9UT0tFTl9TSEEyNTYKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUo"
    "CiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAi"
    "Y29udHJvbCByZXVzYWJsZS1wcmVmaXggaWRlbnRpdHkgZHJpZnRlZCIsCiAgICAgICAgKQoK"
    "ICAgIHJldHVybiB7CiAgICAgICAgInRva2VuX2NvdW50IjogbGVuKGNhbm9uaWNhbF90b2tl"
    "bnMpLAogICAgICAgICJ0b2tlbl9zaGEyNTYiOiBjYW5vbmljYWxfc2hhMjU2LAogICAgICAg"
    "ICJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fY291bnQiOiBsZW4ocmV1c2FibGVfdG9rZW5zKSwK"
    "ICAgICAgICAicmV1c2FibGVfcHJlZml4X3Rva2VuX3NoYTI1NiI6IHJldXNhYmxlX3NoYTI1"
    "NiwKICAgICAgICAic2VudGluZWxfcHJvbXB0X3Rva2VuX2NvdW50IjogbGVuKHNlbnRpbmVs"
    "X3Rva2VucyksCiAgICAgICAgInNlbnRpbmVsX3Byb21wdF90b2tlbl9zaGEyNTYiOiBzaGEy"
    "NTZfYnl0ZXMoCiAgICAgICAgICAgIGNhbm9uaWNhbF9qc29uKGxpc3Qoc2VudGluZWxfdG9r"
    "ZW5zKSkuZW5jb2RlKCJ1dGYtOCIpCiAgICAgICAgKSwKICAgICAgICAiZmlyc3RfZGl2ZXJn"
    "ZW50X3Rva2VuX2luZGV4IjogbGVuKHJldXNhYmxlX3Rva2VucyksCiAgICB9CgoKZGVmIGlu"
    "aXRpYWxpemVfb3JkZXJfam91cm5hbCgpIC0+IE5vbmU6CiAgICBpZiBQUkVfUkVRVUVTVF9U"
    "T0tFTl9JREVOVElUWV9KT1VSTkFMLmV4aXN0cygpOgogICAgICAgIHJhaXNlIFJ1bnRpbWVF"
    "cnJvcigicHJlLXJlcXVlc3QgdG9rZW4taWRlbnRpdHkgam91cm5hbCBhbHJlYWR5IGV4aXN0"
    "cyIpCiAgICB3cml0ZV9qc29uKAogICAgICAgIFBSRV9SRVFVRVNUX1RPS0VOX0lERU5USVRZ"
    "X0pPVVJOQUwsCiAgICAgICAgewogICAgICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4w"
    "LjAiLAogICAgICAgICAgICAiam91cm5hbF9pZCI6ICgKICAgICAgICAgICAgICAgICJhdXJh"
    "Z2F0ZXdheS1jNC1wYXJhZ3JhcGgtb3JkZXItYmVoYXZpb3JhbC0iCiAgICAgICAgICAgICAg"
    "ICAiZGlmZmVyZW50aWFsLXByZS1yZXF1ZXN0LXRva2VuLWlkZW50aXR5LXYxIgogICAgICAg"
    "ICAgICApLAogICAgICAgICAgICAiZGlhZ25vc3RpY19pZCI6IE9SREVSX0RJQUdOT1NUSUNf"
    "SUQsCiAgICAgICAgICAgICJlbnRyaWVzIjogW10sCiAgICAgICAgICAgICJyYXdfcHJvbXB0"
    "X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgICAgICJyYXdfbW9kZWxfb3V0cHV0X3JldGFp"
    "bmVkIjogRmFsc2UsCiAgICAgICAgfSwKICAgICkKCgpkZWYgcGVyc2lzdF9vcmRlcl9wcmVf"
    "cmVxdWVzdF9pZGVudGl0eSgKICAgIHJlcXVlc3Rfb3JkaW5hbDogaW50LAogICAgY29uZGl0"
    "aW9uX2lkOiBzdHIsCiAgICB0b2tlbl9pZGVudGl0eTogZGljdFtzdHIsIG9iamVjdF0sCiAg"
    "ICBwYXlsb2FkX3NoYTI1Njogc3RyLAopIC0+IE5vbmU6CiAgICBqb3VybmFsID0gX3JlYWRf"
    "cHJlX3JlcXVlc3RfdG9rZW5faWRlbnRpdHlfam91cm5hbCgpCiAgICBlbnRyaWVzID0gam91"
    "cm5hbC5nZXQoImVudHJpZXMiKQogICAgaWYgbm90IGlzaW5zdGFuY2UoZW50cmllcywgbGlz"
    "dCk6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVu"
    "dGl0eSBqb3VybmFsIGVudHJpZXMgYXJlIGludmFsaWQiKQogICAgaWYgcmVxdWVzdF9vcmRp"
    "bmFsICE9IGxlbihlbnRyaWVzKSArIDE6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJw"
    "cmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSByZXF1ZXN0IG9yZGluYWwgZHJpZnRlZCIpCiAg"
    "ICBpZiBwYXlsb2FkX3NoYTI1NiAhPSBvcmRlcl9leHBlY3RlZF9wYXlsb2FkX3NoYTI1Nihj"
    "b25kaXRpb25faWQpOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicGFyYWdyYXBoLW9y"
    "ZGVyIHJlcXVlc3QgcGF5bG9hZCBpZGVudGl0eSBkcmlmdGVkIikKCiAgICB3cml0ZV9qc29u"
    "KAogICAgICAgIFBSRV9SRVFVRVNUX1RPS0VOX0lERU5USVRZX0pPVVJOQUwsCiAgICAgICAg"
    "ewogICAgICAgICAgICAqKmpvdXJuYWwsCiAgICAgICAgICAgICJlbnRyaWVzIjogWwogICAg"
    "ICAgICAgICAgICAgKmVudHJpZXMsCiAgICAgICAgICAgICAgICB7CiAgICAgICAgICAgICAg"
    "ICAgICAgInJlcXVlc3Rfb3JkaW5hbCI6IHJlcXVlc3Rfb3JkaW5hbCwKICAgICAgICAgICAg"
    "ICAgICAgICAiY29uZGl0aW9uX2lkIjogY29uZGl0aW9uX2lkLAogICAgICAgICAgICAgICAg"
    "ICAgICJ0b2tlbl9jb3VudCI6IHRva2VuX2lkZW50aXR5WyJ0b2tlbl9jb3VudCJdLAogICAg"
    "ICAgICAgICAgICAgICAgICJ0b2tlbl9zaGEyNTYiOiB0b2tlbl9pZGVudGl0eVsidG9rZW5f"
    "c2hhMjU2Il0sCiAgICAgICAgICAgICAgICAgICAgInJldXNhYmxlX3ByZWZpeF90b2tlbl9j"
    "b3VudCI6ICh0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJlZml4X3Rva2VuX2NvdW50Il0p"
    "LAogICAgICAgICAgICAgICAgICAgICJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hhMjU2Ijog"
    "KAogICAgICAgICAgICAgICAgICAgICAgICB0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJl"
    "Zml4X3Rva2VuX3NoYTI1NiJdCiAgICAgICAgICAgICAgICAgICAgKSwKICAgICAgICAgICAg"
    "ICAgICAgICAicGF5bG9hZF9zaGEyNTYiOiBwYXlsb2FkX3NoYTI1NiwKICAgICAgICAgICAg"
    "ICAgICAgICAicGVyc2lzdGVkX2JlZm9yZV9tb2RlbF9yZXF1ZXN0IjogVHJ1ZSwKICAgICAg"
    "ICAgICAgICAgIH0sCiAgICAgICAgICAgIF0sCiAgICAgICAgfSwKICAgICkKCgpkZWYgd3Jp"
    "dGVfb3JkZXJfcmVzdWx0cygKICAgIHJlc3VsdHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1d"
    "LAogICAgc3RhdHVzOiBzdHIsCikgLT4gTm9uZToKICAgIHdyaXRlX2pzb24oCiAgICAgICAg"
    "T1VUUFVUX1JPT1QgLyAiYzRfcGFyYWdyYXBoX29yZGVyX3JlcXVlc3RfcmVzdWx0c192MS5q"
    "c29uIiwKICAgICAgICB7CiAgICAgICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIs"
    "CiAgICAgICAgICAgICJkaWFnbm9zdGljX2lkIjogT1JERVJfRElBR05PU1RJQ19JRCwKICAg"
    "ICAgICAgICAgInN0YXR1cyI6IHN0YXR1cywKICAgICAgICAgICAgInNjaGVkdWxlZF9yZXF1"
    "ZXN0X2NvdW50IjogbGVuKE9SREVSX1JFUVVFU1RfT1JERVIpLAogICAgICAgICAgICAib2Jz"
    "ZXJ2ZWRfcmVxdWVzdF9jb3VudCI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgInJlcXVl"
    "c3Rfb3JkZXIiOiBsaXN0KE9SREVSX1JFUVVFU1RfT1JERVIpLAogICAgICAgICAgICAicmVz"
    "dWx0cyI6IFtvcmRlcl9wdWJsaWNfb2JzZXJ2YXRpb24oaXRlbSkgZm9yIGl0ZW0gaW4gcmVz"
    "dWx0c10sCiAgICAgICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAg"
    "ICAgICAgICJyYXdfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgfSwKICAgICkK"
    "CgpkZWYgb3JkZXJfY29uZGl0aW9uX3N0YXR1cyhleGFjdF9jb3VudDogaW50KSAtPiBzdHI6"
    "CiAgICBpZiBleGFjdF9jb3VudCA9PSAzOgogICAgICAgIHJldHVybiAiM19PRl8zX0VYQUNU"
    "X09CSkVDVF9UUlVFIgogICAgaWYgZXhhY3RfY291bnQgPT0gMDoKICAgICAgICByZXR1cm4g"
    "IjBfT0ZfM19FWEFDVF9PQkpFQ1RfVFJVRSIKICAgIGlmIGV4YWN0X2NvdW50IGluIHsxLCAy"
    "fToKICAgICAgICByZXR1cm4gIjFfT1JfMl9PRl8zX0VYQUNUX09CSkVDVF9UUlVFIgogICAg"
    "cmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJ"
    "TFVSRSIsCiAgICAgICAgImNvbmRpdGlvbiBleGFjdC1vYmplY3QgY291bnQgaXMgb3V0c2lk"
    "ZSB0aGUgZnJvemVuIGVuZHBvaW50IiwKICAgICkKCgpkZWYgZGVjaWRlX29yZGVyX2RpZmZl"
    "cmVudGlhbCgKICAgIHJlc3VsdHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAogICAgd29y"
    "a2VyX3JlcG9ydHM6IGxpc3RbZGljdFtzdHIsIG9iamVjdF1dLAogICAgdGVhcmRvd25fcmVw"
    "b3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0sCiAgICBjb3VudGVyczogZGljdFtzdHIs"
    "IGludF0sCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBpZiBsZW4ocmVzdWx0cykgIT0g"
    "NjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVF"
    "U1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJwYXJhZ3JhcGgtb3Jk"
    "ZXIgcmVzdWx0IGNvdW50IGRyaWZ0ZWQiLAogICAgICAgICkKICAgIGlmIHR1cGxlKHN0cihy"
    "b3cuZ2V0KCJjb25kaXRpb25faWQiKSkgZm9yIHJvdyBpbiByZXN1bHRzKSAhPSBPUkRFUl9S"
    "RVFVRVNUX09SREVSOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgInBh"
    "cmFncmFwaC1vcmRlciBjaHJvbm9sb2d5IGRyaWZ0ZWQiLAogICAgICAgICkKICAgIGlmIHR1"
    "cGxlKHJvdy5nZXQoInNlcXVlbmNlX2luZGV4IikgZm9yIHJvdyBpbiByZXN1bHRzKSAhPSB0"
    "dXBsZShyYW5nZSgxLCA3KSk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAg"
    "ICAgICAgICAgICJSRVFVRVNUX1JFQ09OQ0lMSUFUSU9OX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAicGFyYWdyYXBoLW9yZGVyIHNlcXVlbmNlIGluZGV4ZXMgZHJpZnRlZCIsCiAgICAgICAg"
    "KQogICAgaWYgbGVuKHdvcmtlcl9yZXBvcnRzKSAhPSA2IG9yIGxlbih0ZWFyZG93bl9yZXBv"
    "cnRzKSAhPSA2OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAg"
    "ICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgImZyZXNo"
    "LXdvcmtlciBldmlkZW5jZSBjYXJkaW5hbGl0eSBkcmlmdGVkIiwKICAgICAgICApCiAgICBp"
    "ZiBhbnkoaXRlbS5nZXQoInN0YXR1cyIpICE9ICJQQVNTRUQiIGZvciBpdGVtIGluIHRlYXJk"
    "b3duX3JlcG9ydHMpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAg"
    "ICAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAgICJvbmUgb3IgbW9yZSBwYXJh"
    "Z3JhcGgtb3JkZXIgdGVhcmRvd25zIGZhaWxlZCIsCiAgICAgICAgKQoKICAgIHdvcmtlcl9p"
    "ZGVudGl0aWVzID0gewogICAgICAgIHN0cihyb3cuZ2V0KCJ3b3JrZXJfcHJvY2Vzc19pZGVu"
    "dGl0eV9zaGEyNTYiKSkKICAgICAgICBmb3Igcm93IGluIHJlc3VsdHMKICAgICAgICBpZiBp"
    "c2luc3RhbmNlKAogICAgICAgICAgICByb3cuZ2V0KCJ3b3JrZXJfcHJvY2Vzc19pZGVudGl0"
    "eV9zaGEyNTYiKSwKICAgICAgICAgICAgc3RyLAogICAgICAgICkKICAgIH0KICAgIGlmIGxl"
    "bih3b3JrZXJfaWRlbnRpdGllcykgIT0gNjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIlA1X1NUQVJUSU5HX1NUQVRFX0ZBSUxVUkUiLAogICAgICAg"
    "ICAgICAiZnJlc2ggd29ya2VyIHByb2Nlc3MgaWRlbnRpdHkgd2FzIHJldXNlZCIsCiAgICAg"
    "ICAgKQoKICAgIGlmIGFueShyb3cuZ2V0KCJ6ZXJvX2NhY2hlX2Jhc2VsaW5lIikgaXMgbm90"
    "IFRydWUgZm9yIHJvdyBpbiByZXN1bHRzKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIlA1X1NUQVJUSU5HX1NUQVRFX0ZBSUxVUkUiLAogICAgICAg"
    "ICAgICAib25lIG9yIG1vcmUgb2JzZXJ2YXRpb25zIGxhY2tlZCB6ZXJvIGNhY2hlIGJhc2Vs"
    "aW5lIiwKICAgICAgICApCgogICAgZXhwZWN0ZWRfY291bnRlcnMgPSB7CiAgICAgICAgIm1v"
    "ZGVsX3JlcXVlc3RzIjogNiwKICAgICAgICAibW9kZWxfbG9hZHMiOiA2LAogICAgICAgICJ3"
    "b3JrZXJfc3RhcnRzIjogNiwKICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAg"
    "ICJuZXR3b3JrX3JlcXVlc3RzIjogMCwKICAgICAgICAiYmVuY2htYXJrX3RyYWplY3Rvcnlf"
    "cmVxdWVzdHMiOiAwLAogICAgICAgICJleHRlcm5hbF9zcGVuZCI6IDAsCiAgICB9CiAgICBm"
    "b3IgbmFtZSwgZXhwZWN0ZWQgaW4gZXhwZWN0ZWRfY291bnRlcnMuaXRlbXMoKToKICAgICAg"
    "ICBpZiBjb3VudGVycy5nZXQobmFtZSkgIT0gZXhwZWN0ZWQ6CiAgICAgICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJ"
    "QVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICAgICBmIntuYW1lfSBleHBlY3RlZCB7ZXhw"
    "ZWN0ZWR9LCBvYnNlcnZlZCB7Y291bnRlcnMuZ2V0KG5hbWUpfSIsCiAgICAgICAgICAgICkK"
    "CiAgICByb3dzX2J5X2NvbmRpdGlvbiA9IHsKICAgICAgICBjb25kaXRpb25faWQ6IFtyb3cg"
    "Zm9yIHJvdyBpbiByZXN1bHRzIGlmIHJvdy5nZXQoImNvbmRpdGlvbl9pZCIpID09IGNvbmRp"
    "dGlvbl9pZF0KICAgICAgICBmb3IgY29uZGl0aW9uX2lkIGluICgKICAgICAgICAgICAgT1JE"
    "RVJfQ09OVFJPTF9DT05ESVRJT04sCiAgICAgICAgICAgIE9SREVSX1RSRUFUTUVOVF9DT05E"
    "SVRJT04sCiAgICAgICAgKQogICAgfQogICAgaWYgYW55KGxlbihyb3dzKSAhPSAzIGZvciBy"
    "b3dzIGluIHJvd3NfYnlfY29uZGl0aW9uLnZhbHVlcygpKToKICAgICAgICByYWlzZSBEaWFn"
    "bm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJ"
    "TFVSRSIsCiAgICAgICAgICAgICJjb25kaXRpb24gY2FyZGluYWxpdHkgZHJpZnRlZCIsCiAg"
    "ICAgICAgKQoKICAgIGV4cGVjdGVkX3Rva2VucyA9IHsKICAgICAgICBPUkRFUl9DT05UUk9M"
    "X0NPTkRJVElPTjogT1JERVJfQ09OVFJPTF9UT0tFTl9TSEEyNTYsCiAgICAgICAgT1JERVJf"
    "VFJFQVRNRU5UX0NPTkRJVElPTjogT1JERVJfVFJFQVRNRU5UX1RPS0VOX1NIQTI1NiwKICAg"
    "IH0KICAgIGV4cGVjdGVkX3BheWxvYWRzID0gewogICAgICAgIE9SREVSX0NPTlRST0xfQ09O"
    "RElUSU9OOiBPUkRFUl9DT05UUk9MX1BBWUxPQURfU0hBMjU2LAogICAgICAgIE9SREVSX1RS"
    "RUFUTUVOVF9DT05ESVRJT046IE9SREVSX1RSRUFUTUVOVF9QQVlMT0FEX1NIQTI1NiwKICAg"
    "IH0KCiAgICBmb3IgY29uZGl0aW9uX2lkLCByb3dzIGluIHJvd3NfYnlfY29uZGl0aW9uLml0"
    "ZW1zKCk6CiAgICAgICAgdG9rZW5faWRlbnRpdGllcyA9IHsKICAgICAgICAgICAgKAogICAg"
    "ICAgICAgICAgICAgcm93LmdldCgidG9rZW5fY291bnQiKSwKICAgICAgICAgICAgICAgIHJv"
    "dy5nZXQoInRva2VuX3NoYTI1NiIpLAogICAgICAgICAgICAgICAgcm93LmdldCgicmV1c2Fi"
    "bGVfcHJlZml4X3Rva2VuX2NvdW50IiksCiAgICAgICAgICAgICkKICAgICAgICAgICAgZm9y"
    "IHJvdyBpbiByb3dzCiAgICAgICAgfQogICAgICAgIGlmIHRva2VuX2lkZW50aXRpZXMgIT0g"
    "ewogICAgICAgICAgICAoCiAgICAgICAgICAgICAgICBPUkRFUl9QUk9NUFRfVE9LRU5fQ09V"
    "TlQsCiAgICAgICAgICAgICAgICBleHBlY3RlZF90b2tlbnNbY29uZGl0aW9uX2lkXSwKICAg"
    "ICAgICAgICAgICAgIE9SREVSX0ZJTkFMX1VTRVJfQk9VTkRBUlksCiAgICAgICAgICAgICkK"
    "ICAgICAgICB9OgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAg"
    "ImNvbmRpdGlvbiB0b2tlbiBpZGVudGl0eSBmYWlsZWQgcmVjb25jaWxpYXRpb24iLAogICAg"
    "ICAgICAgICApCgogICAgICAgIGlmIHtyb3cuZ2V0KCJwYXlsb2FkX3NoYTI1NiIpIGZvciBy"
    "b3cgaW4gcm93c30gIT0ge2V4cGVjdGVkX3BheWxvYWRzW2NvbmRpdGlvbl9pZF19OgogICAg"
    "ICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgICAgICJIQVJO"
    "RVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgImNvbmRpdGlvbiBwYXls"
    "b2FkIGlkZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgICAgICkKCiAg"
    "ICAgICAgaWYgYW55KAogICAgICAgICAgICByb3cuZ2V0KCJodHRwX3N0YXR1cyIpICE9IDIw"
    "MAogICAgICAgICAgICBvciByb3cuZ2V0KCJmaW5pc2hfcmVhc29uIikgIT0gInN0b3AiCiAg"
    "ICAgICAgICAgIG9yIHJvdy5nZXQoInJlc3BvbnNlX2NvbXBsZXRlIikgaXMgbm90IFRydWUK"
    "ICAgICAgICAgICAgb3Igcm93LmdldCgid29ya2VyX2hlYWx0aF9hZnRlcl9yZXF1ZXN0Iikg"
    "aXMgbm90IFRydWUKICAgICAgICAgICAgb3Igcm93LmdldCgidGVhcmRvd25fc3RhdHVzIikg"
    "IT0gIlBBU1NFRCIKICAgICAgICAgICAgZm9yIHJvdyBpbiByb3dzCiAgICAgICAgKToKICAg"
    "ICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiT1VU"
    "UFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgImhlYWx0aHkgb2JzZXJ2"
    "YXRpb24gY29udHJhY3QgZHJpZnRlZCIsCiAgICAgICAgICAgICkKCiAgICBjb250cm9sX3Jv"
    "d3MgPSByb3dzX2J5X2NvbmRpdGlvbltPUkRFUl9DT05UUk9MX0NPTkRJVElPTl0KICAgIHRy"
    "ZWF0bWVudF9yb3dzID0gcm93c19ieV9jb25kaXRpb25bT1JERVJfVFJFQVRNRU5UX0NPTkRJ"
    "VElPTl0KCiAgICBjb250cm9sX2V4YWN0ID0gc3VtKHJvdy5nZXQoImV4YWN0X29iamVjdCIp"
    "IGlzIFRydWUgZm9yIHJvdyBpbiBjb250cm9sX3Jvd3MpCiAgICB0cmVhdG1lbnRfZXhhY3Qg"
    "PSBzdW0ocm93LmdldCgiZXhhY3Rfb2JqZWN0IikgaXMgVHJ1ZSBmb3Igcm93IGluIHRyZWF0"
    "bWVudF9yb3dzKQoKICAgIGNvbnRyb2xfdmFsaWRfanNvbiA9IHN1bShyb3cuZ2V0KCJ2YWxp"
    "ZF9qc29uIikgaXMgVHJ1ZSBmb3Igcm93IGluIGNvbnRyb2xfcm93cykKICAgIHRyZWF0bWVu"
    "dF92YWxpZF9qc29uID0gc3VtKHJvdy5nZXQoInZhbGlkX2pzb24iKSBpcyBUcnVlIGZvciBy"
    "b3cgaW4gdHJlYXRtZW50X3Jvd3MpCgogICAgY29udHJvbF9wYXJzZWQgPSB7cm93LmdldCgi"
    "Y2Fub25pY2FsX3BhcnNlZF9vYmplY3Rfc2hhMjU2IikgZm9yIHJvdyBpbiBjb250cm9sX3Jv"
    "d3N9CiAgICB0cmVhdG1lbnRfcGFyc2VkID0ge3Jvdy5nZXQoImNhbm9uaWNhbF9wYXJzZWRf"
    "b2JqZWN0X3NoYTI1NiIpIGZvciByb3cgaW4gdHJlYXRtZW50X3Jvd3N9CgogICAgY29udHJv"
    "bF9hbmNob3JfcmVwcm9kdWNlZCA9ICgKICAgICAgICBjb250cm9sX2V4YWN0ID09IDAKICAg"
    "ICAgICBhbmQgY29udHJvbF92YWxpZF9qc29uID09IDMKICAgICAgICBhbmQgY29udHJvbF9w"
    "YXJzZWQgPT0ge09SREVSX0hJU1RPUklDQUxfQ09OVFJPTF9QQVJTRURfT0JKRUNUX1NIQTI1"
    "Nn0KICAgICkKCiAgICBpZiBub3QgY29udHJvbF9hbmNob3JfcmVwcm9kdWNlZDoKICAgICAg"
    "ICBzdGF0ZSA9ICJDT05UUk9MX0FOQ0hPUl9OT05SRVBST0RVQ1RJT05fSU5WQUxJREFURVNf"
    "SU5GRVJFTkNFIgogICAgZWxpZiB0cmVhdG1lbnRfZXhhY3QgPT0gMzoKICAgICAgICBzdGF0"
    "ZSA9ICJPUkRFUl9JTlRFUlZFTlRJT05fUkVTVE9SRVNfQkVIQVZJT1IiCiAgICBlbGlmICgK"
    "ICAgICAgICB0cmVhdG1lbnRfZXhhY3QgPT0gMAogICAgICAgIGFuZCB0cmVhdG1lbnRfdmFs"
    "aWRfanNvbiA9PSAzCiAgICAgICAgYW5kIHRyZWF0bWVudF9wYXJzZWQgPT0ge09SREVSX0hJ"
    "U1RPUklDQUxfQ09OVFJPTF9QQVJTRURfT0JKRUNUX1NIQTI1Nn0KICAgICk6CiAgICAgICAg"
    "c3RhdGUgPSAiT1JERVJfSU5URVJWRU5USU9OX0RPRVNfTk9UX0NIQU5HRV9PQlNFUlZFRF9Q"
    "SEVOT1RZUEUiCiAgICBlbGlmICgKICAgICAgICB0cmVhdG1lbnRfZXhhY3QgPT0gMAogICAg"
    "ICAgIGFuZCB0cmVhdG1lbnRfdmFsaWRfanNvbiA9PSAzCiAgICAgICAgYW5kIGxlbih0cmVh"
    "dG1lbnRfcGFyc2VkKSA9PSAxCiAgICAgICAgYW5kIE5vbmUgbm90IGluIHRyZWF0bWVudF9w"
    "YXJzZWQKICAgICAgICBhbmQgdHJlYXRtZW50X3BhcnNlZCAhPSB7T1JERVJfSElTVE9SSUNB"
    "TF9DT05UUk9MX1BBUlNFRF9PQkpFQ1RfU0hBMjU2fQogICAgKToKICAgICAgICBzdGF0ZSA9"
    "ICJPUkRFUl9JTlRFUlZFTlRJT05fQ0hBTkdFU19GQUlMVVJFX1BIRU5PVFlQRSIKICAgIGVs"
    "c2U6CiAgICAgICAgc3RhdGUgPSAiT1JERVJfSU5URVJWRU5USU9OX0VGRkVDVF9BTUJJR1VP"
    "VVMiCgogICAgcmV0dXJuIHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAog"
    "ICAgICAgICJzdGF0dXMiOiAiREVDSURFRCIsCiAgICAgICAgImRpYWdub3N0aWNfaWQiOiBP"
    "UkRFUl9ESUFHTk9TVElDX0lELAogICAgICAgICJvYnNlcnZlZF90ZXJtaW5hbF9zdGF0ZSI6"
    "IHN0YXRlLAogICAgICAgICJwcmltYXJ5X2VuZHBvaW50IjogImV4YWN0X29iamVjdCIsCiAg"
    "ICAgICAgImNvbmRpdGlvbl9leGFjdF9vYmplY3RfY291bnRzIjogewogICAgICAgICAgICBP"
    "UkRFUl9DT05UUk9MX0NPTkRJVElPTjogY29udHJvbF9leGFjdCwKICAgICAgICAgICAgT1JE"
    "RVJfVFJFQVRNRU5UX0NPTkRJVElPTjogdHJlYXRtZW50X2V4YWN0LAogICAgICAgIH0sCiAg"
    "ICAgICAgImNvbmRpdGlvbl92YWxpZF9qc29uX2NvdW50cyI6IHsKICAgICAgICAgICAgT1JE"
    "RVJfQ09OVFJPTF9DT05ESVRJT046IGNvbnRyb2xfdmFsaWRfanNvbiwKICAgICAgICAgICAg"
    "T1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTjogdHJlYXRtZW50X3ZhbGlkX2pzb24sCiAgICAg"
    "ICAgfSwKICAgICAgICAiY29uZGl0aW9uX2VuZHBvaW50X3N0YXR1c2VzIjogewogICAgICAg"
    "ICAgICBPUkRFUl9DT05UUk9MX0NPTkRJVElPTjogb3JkZXJfY29uZGl0aW9uX3N0YXR1cyhj"
    "b250cm9sX2V4YWN0KSwKICAgICAgICAgICAgT1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTjog"
    "b3JkZXJfY29uZGl0aW9uX3N0YXR1cyh0cmVhdG1lbnRfZXhhY3QpLAogICAgICAgIH0sCiAg"
    "ICAgICAgImNvbnRyb2xfYW5jaG9yX3JlcHJvZHVjZWQiOiBjb250cm9sX2FuY2hvcl9yZXBy"
    "b2R1Y2VkLAogICAgICAgICJjb250cm9sX2hpc3RvcmljYWxfcGFyc2VkX2lkZW50aXR5X3Jl"
    "cXVpcmVkIjogVHJ1ZSwKICAgICAgICAiZnJlc2hfd29ya2VyX3Byb2Nlc3NfcGVyX29ic2Vy"
    "dmF0aW9uIjogVHJ1ZSwKICAgICAgICAid29ya2VyX2lkZW50aXR5X2NhcmRpbmFsaXR5Ijog"
    "bGVuKHdvcmtlcl9pZGVudGl0aWVzKSwKICAgICAgICAiYWxsX2NvbmRpdGlvbl90b2tlbl9p"
    "ZGVudGl0aWVzX21hdGNoZWQiOiBUcnVlLAogICAgICAgICJhbGxfY29uZGl0aW9uX3BheWxv"
    "YWRfaWRlbnRpdGllc19tYXRjaGVkIjogVHJ1ZSwKICAgICAgICAic3RhdGljX3Rva2VuX211"
    "bHRpc2V0X3ByZW1pc2VfcmVleGVjdXRlZCI6IEZhbHNlLAogICAgICAgICJwYXJhZ3JhcGhf"
    "b3JkZXJfcm9vdF9jYXVzZV9lc3RhYmxpc2hlZCI6IEZhbHNlLAogICAgICAgICJwNV9yZXF1"
    "YWxpZmllZCI6IEZhbHNlLAogICAgICAgICJwNl9yZXF1YWxpZmllZCI6IEZhbHNlLAogICAg"
    "ICAgICJmaW5hbF9hYmNfbWVhc3VyZWQiOiBGYWxzZSwKICAgICAgICAicmF3X3Byb21wdF9y"
    "ZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICJyYXdfb3V0cHV0X3JldGFpbmVkIjogRmFsc2Us"
    "CiAgICB9CgoKZGVmIHJ1bl9vcmRlcl9vYnNlcnZhdGlvbigKICAgIHdvcmtlcjogV29ya2Vy"
    "LAogICAgY29uZGl0aW9uX2lkOiBzdHIsCiAgICBzZXF1ZW5jZV9pbmRleDogaW50LAogICAg"
    "Y291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAopIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAg"
    "dG9rZW5faWRlbnRpdHkgPSBvcmRlcl90b2tlbl9pZGVudGl0eSh3b3JrZXIsIGNvbmRpdGlv"
    "bl9pZCkKICAgIHBheWxvYWQgPSBvcmRlcl9yZXF1ZXN0X3BheWxvYWQoY29uZGl0aW9uX2lk"
    "KQogICAgcGF5bG9hZF9zaGEyNTYgPSBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihwYXls"
    "b2FkKSkKCiAgICBpZiBwYXlsb2FkX3NoYTI1NiAhPSBvcmRlcl9leHBlY3RlZF9wYXlsb2Fk"
    "X3NoYTI1Nihjb25kaXRpb25faWQpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJl"
    "KAogICAgICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAg"
    "InBhcmFncmFwaC1vcmRlciByZXF1ZXN0IHBheWxvYWQgaWRlbnRpdHkgZHJpZnRlZCBiZWZv"
    "cmUgbW9kZWwgcmVxdWVzdCIsCiAgICAgICAgKQoKICAgIHJlcXVlc3Rfb3JkaW5hbCA9IGNv"
    "dW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdICsgMQogICAgaWYgcmVxdWVzdF9vcmRpbmFsICE9"
    "IHNlcXVlbmNlX2luZGV4OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAg"
    "ICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAg"
    "InBhcmFncmFwaC1vcmRlciBwcmUtcmVxdWVzdCBvcmRpbmFsIGRyaWZ0ZWQiLAogICAgICAg"
    "ICkKCiAgICBwZXJzaXN0X29yZGVyX3ByZV9yZXF1ZXN0X2lkZW50aXR5KAogICAgICAgIHJl"
    "cXVlc3Rfb3JkaW5hbCwKICAgICAgICBjb25kaXRpb25faWQsCiAgICAgICAgdG9rZW5faWRl"
    "bnRpdHksCiAgICAgICAgcGF5bG9hZF9zaGEyNTYsCiAgICApCgogICAgYmFzZWxpbmUgPSB2"
    "YWxpZGF0ZV96ZXJvX2NhY2hlX2Jhc2VsaW5lKHdvcmtlcikKICAgIHplcm9fY2FjaGVfcmVj"
    "ZWlwdF9zaGEyNTYgPSBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihiYXNlbGluZSkpCiAg"
    "ICBiZWZvcmUgPSB3b3JrZXIubWV0cmljX3NuYXBzaG90KCkKCiAgICBjb25zdW1lX2FjdGlv"
    "bnMoY291bnRlcnMsICJtb2RlbF9yZXF1ZXN0cyIpCiAgICBlbmNvZGVkID0gY2Fub25pY2Fs"
    "X2pzb24ocGF5bG9hZCkuZW5jb2RlKCJ1dGYtOCIpCiAgICByZXF1ZXN0ID0gdXJsbGliLnJl"
    "cXVlc3QuUmVxdWVzdCgKICAgICAgICBib3VuZGVkX2xvb3BiYWNrKGYiaHR0cDovLzEyNy4w"
    "LjAuMTp7d29ya2VyLnBvcnR9L3YxL2NoYXQvY29tcGxldGlvbnMiKSwKICAgICAgICBkYXRh"
    "PWVuY29kZWQsCiAgICAgICAgaGVhZGVycz17IkNvbnRlbnQtVHlwZSI6ICJhcHBsaWNhdGlv"
    "bi9qc29uIn0sCiAgICAgICAgbWV0aG9kPSJQT1NUIiwKICAgICkKCiAgICB0cnk6CiAgICAg"
    "ICAgd2l0aCB1cmxsaWIucmVxdWVzdC51cmxvcGVuKAogICAgICAgICAgICByZXF1ZXN0LAog"
    "ICAgICAgICAgICB0aW1lb3V0PUhUVFBfVElNRU9VVF9TRUNPTkRTLAogICAgICAgICkgYXMg"
    "cmVzcG9uc2U6CiAgICAgICAgICAgIHN0YXR1c19jb2RlID0gcmVzcG9uc2Uuc3RhdHVzCiAg"
    "ICAgICAgICAgIHJlc3BvbnNlX3BheWxvYWQgPSByZXNwb25zZS5yZWFkKCkKICAgIGV4Y2Vw"
    "dCB1cmxsaWIuZXJyb3IuSFRUUEVycm9yIGFzIGVycm9yOgogICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVRVUVTVF9FWEVDVVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgIGYicGFyYWdyYXBoLW9yZGVyIHJlcXVlc3QgSFRUUCBmYWlsdXJlOiB7"
    "ZXJyb3IuY29kZX0iLAogICAgICAgICkgZnJvbSBlcnJvcgogICAgZXhjZXB0ICh1cmxsaWIu"
    "ZXJyb3IuVVJMRXJyb3IsIFRpbWVvdXRFcnJvcikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2Ug"
    "RGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNVVElPTl9GQUlM"
    "VVJFIiwKICAgICAgICAgICAgInBhcmFncmFwaC1vcmRlciByZXF1ZXN0IHRyYW5zcG9ydCBm"
    "YWlsZWQiLAogICAgICAgICkgZnJvbSBlcnJvcgoKICAgIGlmIHN0YXR1c19jb2RlICE9IDIw"
    "MDoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVF"
    "U1RfRVhFQ1VUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIHJl"
    "cXVlc3QgcmV0dXJuZWQgYW4gdW5leHBlY3RlZCBIVFRQIHN0YXR1cyIsCiAgICAgICAgKQoK"
    "ICAgIHRyeToKICAgICAgICBlbnZlbG9wZSA9IGpzb24ubG9hZHMocmVzcG9uc2VfcGF5bG9h"
    "ZCkKICAgIGV4Y2VwdCAoanNvbi5KU09ORGVjb2RlRXJyb3IsIFVuaWNvZGVEZWNvZGVFcnJv"
    "cikgYXMgZXJyb3I6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJwYXJhZ3JhcGgt"
    "b3JkZXIgcmVzcG9uc2UgZW52ZWxvcGUgaXMgbm90IHZhbGlkIEpTT04iLAogICAgICAgICkg"
    "ZnJvbSBlcnJvcgogICAgaWYgbm90IGlzaW5zdGFuY2UoZW52ZWxvcGUsIGRpY3QpOgogICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRS"
    "QUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIHJlc3BvbnNlIGVu"
    "dmVsb3BlIHJvb3QgaXMgaW52YWxpZCIsCiAgICAgICAgKQoKICAgIHVzYWdlID0gZW52ZWxv"
    "cGUuZ2V0KCJ1c2FnZSIpCiAgICBjaG9pY2VzID0gZW52ZWxvcGUuZ2V0KCJjaG9pY2VzIikK"
    "ICAgIGlmIG5vdCBpc2luc3RhbmNlKHVzYWdlLCBkaWN0KToKICAgICAgICByYWlzZSBEaWFn"
    "bm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgInBhcmFncmFwaC1vcmRlciByZXNwb25zZSB1c2FnZSBpcyBtaXNzaW5n"
    "IiwKICAgICAgICApCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2VzLCBsaXN0KSBvciBs"
    "ZW4oY2hvaWNlcykgIT0gMToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAg"
    "ICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInBhcmFn"
    "cmFwaC1vcmRlciByZXNwb25zZSBjaG9pY2VzIGFyZSBpbnZhbGlkIiwKICAgICAgICApCiAg"
    "ICBjaG9pY2UgPSBjaG9pY2VzWzBdCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2UsIGRp"
    "Y3QpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VU"
    "UFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIHJl"
    "c3BvbnNlIGNob2ljZSBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICBtZXNzYWdlID0gY2hv"
    "aWNlLmdldCgibWVzc2FnZSIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShtZXNzYWdlLCBkaWN0"
    "KToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIk9VVFBV"
    "VF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgInBhcmFncmFwaC1vcmRlciByZXNw"
    "b25zZSBtZXNzYWdlIGlzIGludmFsaWQiLAogICAgICAgICkKICAgIGNvbnRlbnQgPSBtZXNz"
    "YWdlLmdldCgiY29udGVudCIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShjb250ZW50LCBzdHIp"
    "OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVU"
    "X0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIHJlc3Bv"
    "bnNlIGNvbnRlbnQgaXMgbm90IGEgc3RyaW5nIiwKICAgICAgICApCgogICAgY29tcGxldGlv"
    "bl90b2tlbnMgPSB1c2FnZS5nZXQoImNvbXBsZXRpb25fdG9rZW5zIikKICAgIHByb21wdF90"
    "b2tlbnMgPSB1c2FnZS5nZXQoInByb21wdF90b2tlbnMiKQogICAgaWYgKAogICAgICAgIGlz"
    "aW5zdGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGJvb2wpCiAgICAgICAgb3Igbm90IGlzaW5z"
    "dGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGludCkKICAgICAgICBvciBub3QgMSA8PSBjb21w"
    "bGV0aW9uX3Rva2VucyA8PSAzMgogICAgKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgIkM0IGNvbXBsZXRpb24tdG9rZW4gYnVkZ2V0IGRyaWZ0ZWQiLAogICAgICAgICkKICAg"
    "IGlmICgKICAgICAgICBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGJvb2wpCiAgICAgICAg"
    "b3Igbm90IGlzaW5zdGFuY2UocHJvbXB0X3Rva2VucywgaW50KQogICAgICAgIG9yIHByb21w"
    "dF90b2tlbnMgIT0gT1JERVJfUFJPTVBUX1RPS0VOX0NPVU5UCiAgICApOgogICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZB"
    "SUxVUkUiLAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIHJlc3BvbnNlIHByb21wdC10"
    "b2tlbiBjb3VudCBkcmlmdGVkIiwKICAgICAgICApCgogICAgYWZ0ZXIgPSB3b3JrZXIubWV0"
    "cmljX3NuYXBzaG90KCkKICAgIGRlbHRhID0gbWV0cmljX2RlbHRhKGJlZm9yZSwgYWZ0ZXIp"
    "CiAgICB3b3JrZXJfaGVhbHRoX2FmdGVyX3JlcXVlc3QgPSBUcnVlCgogICAgZmluaXNoX3Jl"
    "YXNvbiA9IGNob2ljZS5nZXQoImZpbmlzaF9yZWFzb24iKQogICAgcHJvamVjdGlvbiA9IGM0"
    "X3Jlc3BvbnNlX3Byb2plY3Rpb24oCiAgICAgICAgY29udGVudCwKICAgICAgICBmaW5pc2hf"
    "cmVhc29uLAogICAgKQoKICAgIHdvcmtlcl9yZXBvcnQgPSB3b3JrZXIucmVwb3J0KCkKICAg"
    "IHBpZCA9IHdvcmtlcl9yZXBvcnQuZ2V0KCJwaWQiKQogICAgcHJvY2Vzc19zdGFydF90aWNr"
    "c192YWx1ZSA9IHdvcmtlcl9yZXBvcnQuZ2V0KCJwcm9jZXNzX3N0YXJ0X3RpY2tzIikKICAg"
    "IHByb2Nlc3NfaWRlbnRpdHlfc2hhMjU2ID0gc2hhMjU2X3RleHQoCiAgICAgICAgY2Fub25p"
    "Y2FsX2pzb24oCiAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICJwaWQiOiBwaWQsCiAg"
    "ICAgICAgICAgICAgICAicHJvY2Vzc19zdGFydF90aWNrcyI6IHByb2Nlc3Nfc3RhcnRfdGlj"
    "a3NfdmFsdWUsCiAgICAgICAgICAgIH0KICAgICAgICApCiAgICApCgogICAgcmV0dXJuIHsK"
    "ICAgICAgICAib2JzZXJ2YXRpb25faWQiOiBmIk9SREVSX3tjb25kaXRpb25faWR9X3tzZXF1"
    "ZW5jZV9pbmRleH0iLAogICAgICAgICJjb25kaXRpb25faWQiOiBjb25kaXRpb25faWQsCiAg"
    "ICAgICAgInJlcXVlc3Rfb3JkaW5hbCI6IHJlcXVlc3Rfb3JkaW5hbCwKICAgICAgICAic2Vx"
    "dWVuY2VfaW5kZXgiOiBzZXF1ZW5jZV9pbmRleCwKICAgICAgICAid29ya2VyX2luc3RhbmNl"
    "X2lkIjogd29ya2VyLmluc3RhbmNlX2lkLAogICAgICAgICJ3b3JrZXJfcHJvY2Vzc19pZGVu"
    "dGl0eV9zaGEyNTYiOiBwcm9jZXNzX2lkZW50aXR5X3NoYTI1NiwKICAgICAgICAidG9rZW5f"
    "Y291bnQiOiB0b2tlbl9pZGVudGl0eVsidG9rZW5fY291bnQiXSwKICAgICAgICAidG9rZW5f"
    "c2hhMjU2IjogdG9rZW5faWRlbnRpdHlbInRva2VuX3NoYTI1NiJdLAogICAgICAgICJyZXVz"
    "YWJsZV9wcmVmaXhfdG9rZW5fY291bnQiOiAodG9rZW5faWRlbnRpdHlbInJldXNhYmxlX3By"
    "ZWZpeF90b2tlbl9jb3VudCJdKSwKICAgICAgICAicmV1c2FibGVfcHJlZml4X3Rva2VuX3No"
    "YTI1NiI6ICh0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJlZml4X3Rva2VuX3NoYTI1NiJd"
    "KSwKICAgICAgICAicGF5bG9hZF9zaGEyNTYiOiBwYXlsb2FkX3NoYTI1NiwKICAgICAgICAi"
    "emVyb19jYWNoZV9iYXNlbGluZSI6IFRydWUsCiAgICAgICAgInplcm9fY2FjaGVfYmFzZWxp"
    "bmVfcmVjZWlwdF9zaGEyNTYiOiAoemVyb19jYWNoZV9yZWNlaXB0X3NoYTI1NiksCiAgICAg"
    "ICAgImh0dHBfc3RhdHVzIjogc3RhdHVzX2NvZGUsCiAgICAgICAgInJlc3BvbnNlX3NoYTI1"
    "NiI6IHNoYTI1Nl90ZXh0KGNvbnRlbnQpLAogICAgICAgICJyZXNwb25zZV9sZW5ndGgiOiBs"
    "ZW4oY29udGVudCksCiAgICAgICAgImZpbmlzaF9yZWFzb24iOiBmaW5pc2hfcmVhc29uLAog"
    "ICAgICAgICJwcm9tcHRfdG9rZW5zIjogcHJvbXB0X3Rva2VucywKICAgICAgICAiY29tcGxl"
    "dGlvbl90b2tlbnMiOiBjb21wbGV0aW9uX3Rva2VucywKICAgICAgICAqKnByb2plY3Rpb24s"
    "CiAgICAgICAgIndvcmtlcl9oZWFsdGhfYWZ0ZXJfcmVxdWVzdCI6IHdvcmtlcl9oZWFsdGhf"
    "YWZ0ZXJfcmVxdWVzdCwKICAgICAgICAicmVxdWVzdF9lcnJvciI6IE5vbmUsCiAgICAgICAg"
    "InRyYW5zcG9ydF9lcnJvciI6IE5vbmUsCiAgICAgICAgIm1ldHJpY19kZWx0YSI6IGFzZGlj"
    "dChkZWx0YSksCiAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAg"
    "ICAicmF3X291dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgfQoKCmRlZiBvcmRlcl9wdWJs"
    "aWNfb2JzZXJ2YXRpb24oCiAgICBvYnNlcnZhdGlvbjogZGljdFtzdHIsIG9iamVjdF0sCikg"
    "LT4gZGljdFtzdHIsIG9iamVjdF06CiAgICBwZXJtaXR0ZWQgPSAoCiAgICAgICAgIm9ic2Vy"
    "dmF0aW9uX2lkIiwKICAgICAgICAiY29uZGl0aW9uX2lkIiwKICAgICAgICAicmVxdWVzdF9v"
    "cmRpbmFsIiwKICAgICAgICAic2VxdWVuY2VfaW5kZXgiLAogICAgICAgICJ3b3JrZXJfaW5z"
    "dGFuY2VfaWQiLAogICAgICAgICJ3b3JrZXJfcHJvY2Vzc19pZGVudGl0eV9zaGEyNTYiLAog"
    "ICAgICAgICJ3b3JrZXJfc3RhcnRfcmVjZWlwdF9zaGEyNTYiLAogICAgICAgICJ0b2tlbl9j"
    "b3VudCIsCiAgICAgICAgInRva2VuX3NoYTI1NiIsCiAgICAgICAgInJldXNhYmxlX3ByZWZp"
    "eF90b2tlbl9jb3VudCIsCiAgICAgICAgInJldXNhYmxlX3ByZWZpeF90b2tlbl9zaGEyNTYi"
    "LAogICAgICAgICJwYXlsb2FkX3NoYTI1NiIsCiAgICAgICAgInplcm9fY2FjaGVfYmFzZWxp"
    "bmUiLAogICAgICAgICJ6ZXJvX2NhY2hlX2Jhc2VsaW5lX3JlY2VpcHRfc2hhMjU2IiwKICAg"
    "ICAgICAiaHR0cF9zdGF0dXMiLAogICAgICAgICJyZXNwb25zZV9zaGEyNTYiLAogICAgICAg"
    "ICJyZXNwb25zZV9sZW5ndGgiLAogICAgICAgICJmaW5pc2hfcmVhc29uIiwKICAgICAgICAi"
    "cHJvbXB0X3Rva2VucyIsCiAgICAgICAgImNvbXBsZXRpb25fdG9rZW5zIiwKICAgICAgICAi"
    "dmFsaWRfanNvbiIsCiAgICAgICAgImR1cGxpY2F0ZV9rZXlfZGV0ZWN0ZWQiLAogICAgICAg"
    "ICJqc29uX2Vycm9yX2xpbmUiLAogICAgICAgICJqc29uX2Vycm9yX2NvbHVtbiIsCiAgICAg"
    "ICAgImpzb25fZXJyb3JfcG9zaXRpb24iLAogICAgICAgICJqc29uX3Jvb3RfdHlwZSIsCiAg"
    "ICAgICAgInBhcnNlZF9rZXlfc2V0IiwKICAgICAgICAicHJvYmVfanNvbl90eXBlIiwKICAg"
    "ICAgICAidmFsdWVfanNvbl90eXBlIiwKICAgICAgICAicHJvYmVfZXhhY3QiLAogICAgICAg"
    "ICJ2YWx1ZV9leGFjdCIsCiAgICAgICAgImNhbm9uaWNhbF9wYXJzZWRfb2JqZWN0X3NoYTI1"
    "NiIsCiAgICAgICAgImNhbm9uaWNhbF9leHBlY3RlZF9vYmplY3Rfc2hhMjU2IiwKICAgICAg"
    "ICAibGVhZGluZ19ub25fd2hpdGVzcGFjZV9jb250ZW50X2RldGVjdGVkIiwKICAgICAgICAi"
    "dHJhaWxpbmdfbm9uX3doaXRlc3BhY2VfY29udGVudF9kZXRlY3RlZCIsCiAgICAgICAgIm1h"
    "cmtkb3duX2ZlbmNlX2RldGVjdGVkIiwKICAgICAgICAicmVzcG9uc2VfY29tcGxldGUiLAog"
    "ICAgICAgICJleGFjdF9vYmplY3QiLAogICAgICAgICJ3b3JrZXJfaGVhbHRoX2FmdGVyX3Jl"
    "cXVlc3QiLAogICAgICAgICJ0ZWFyZG93bl9zdGF0dXMiLAogICAgICAgICJyZXF1ZXN0X2Vy"
    "cm9yIiwKICAgICAgICAidHJhbnNwb3J0X2Vycm9yIiwKICAgICAgICAibWV0cmljX2RlbHRh"
    "IiwKICAgICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCIsCiAgICAgICAgInJhd19vdXRwdXRf"
    "cmV0YWluZWQiLAogICAgKQogICAgcmV0dXJuIHtuYW1lOiBvYnNlcnZhdGlvbi5nZXQobmFt"
    "ZSkgZm9yIG5hbWUgaW4gcGVybWl0dGVkfQoKCmRlZiBydW5fb3JkZXJfZnJlc2hfd29ya2Vy"
    "X29ic2VydmF0aW9uKAogICAgbW9kZWxfaG9tZTogUGF0aCwKICAgIHNuYXBzaG90OiBQYXRo"
    "LAogICAgY29uZGl0aW9uX2lkOiBzdHIsCiAgICBzZXF1ZW5jZV9pbmRleDogaW50LAogICAg"
    "Y291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAogICAgd29ya2VyX3JlcG9ydHM6IGxpc3RbZGlj"
    "dFtzdHIsIG9iamVjdF1dLAogICAgdGVhcmRvd25fcmVwb3J0czogbGlzdFtkaWN0W3N0ciwg"
    "b2JqZWN0XV0sCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICB3b3JrZXIgPSBXb3JrZXIo"
    "CiAgICAgICAgIm9ic2VydmF0aW9uX3dvcmtlciIsCiAgICAgICAgMCwKICAgICAgICA4MDAx"
    "LAogICAgICAgIG1vZGVsX2hvbWUsCiAgICAgICAgc25hcHNob3QsCiAgICAgICAgZ2VuZXJh"
    "dGlvbj1zZXF1ZW5jZV9pbmRleCwKICAgICkKICAgIHByaW1hcnlfZXJyb3I6IEV4Y2VwdGlv"
    "biB8IE5vbmUgPSBOb25lCiAgICBvYnNlcnZhdGlvbjogZGljdFtzdHIsIG9iamVjdF0gfCBO"
    "b25lID0gTm9uZQogICAgcmVwb3J0OiBkaWN0W3N0ciwgb2JqZWN0XSB8IE5vbmUgPSBOb25l"
    "CgogICAgdHJ5OgogICAgICAgIHdvcmtlci5zdGFydChjb3VudGVycykKICAgICAgICB3b3Jr"
    "ZXIud2FpdF9yZWFkeSgpCiAgICAgICAgd29ya2VyLnZhbGlkYXRlX21vZGVsKCkKICAgICAg"
    "ICB3b3JrZXIud2FpdF9iYWNrZW5kX21hcmtlcigpCiAgICAgICAgbmF0aXZlX29yaWdpbnMg"
    "PSB2YWxpZGF0ZV9uYXRpdmVfb3JpZ2luX2Nsb3N1cmUod29ya2VyKQogICAgICAgIHJlcG9y"
    "dCA9IHsKICAgICAgICAgICAgKip3b3JrZXIucmVwb3J0KCksCiAgICAgICAgICAgICJuYXRp"
    "dmVfb3JpZ2luX2Nsb3N1cmUiOiBuYXRpdmVfb3JpZ2lucywKICAgICAgICB9CiAgICAgICAg"
    "d29ya2VyX3JlcG9ydHMuYXBwZW5kKHJlcG9ydCkKICAgICAgICBvYnNlcnZhdGlvbiA9IHJ1"
    "bl9vcmRlcl9vYnNlcnZhdGlvbigKICAgICAgICAgICAgd29ya2VyLAogICAgICAgICAgICBj"
    "b25kaXRpb25faWQsCiAgICAgICAgICAgIHNlcXVlbmNlX2luZGV4LAogICAgICAgICAgICBj"
    "b3VudGVycywKICAgICAgICApCiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGVycm9yOgogICAg"
    "ICAgIHByaW1hcnlfZXJyb3IgPSBlcnJvcgogICAgZmluYWxseToKICAgICAgICB0ZWFyZG93"
    "biA9IHNhZmVfd29ya2VyX3RlYXJkb3duKAogICAgICAgICAgICB3b3JrZXIsCiAgICAgICAg"
    "ICAgIGYiT1JERVJfe2NvbmRpdGlvbl9pZH1fe3NlcXVlbmNlX2luZGV4fV9URVJNSU5BTCIs"
    "CiAgICAgICAgKQogICAgICAgIHRlYXJkb3duX3JlcG9ydHMuYXBwZW5kKHRlYXJkb3duKQoK"
    "ICAgIHRlYXJkb3duX3N0YXR1cyA9IHRlYXJkb3duLmdldCgic3RhdHVzIikKICAgIGlmIHRl"
    "YXJkb3duX3N0YXR1cyBub3QgaW4geyJQQVNTRUQiLCAiTk9UX1NUQVJURUQifToKICAgICAg"
    "ICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlRFQVJET1dOX0ZBSUxV"
    "UkUiLAogICAgICAgICAgICAiZnJlc2ggcGFyYWdyYXBoLW9yZGVyIG9ic2VydmF0aW9uIHdv"
    "cmtlciB0ZWFyZG93biBwcm9vZiBmYWlsZWQiLAogICAgICAgICkKICAgIGlmIHByaW1hcnlf"
    "ZXJyb3IgaXMgbm90IE5vbmU6CiAgICAgICAgcmFpc2UgcHJpbWFyeV9lcnJvcgogICAgaWYg"
    "dGVhcmRvd25fc3RhdHVzICE9ICJQQVNTRUQiOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNG"
    "YWlsdXJlKAogICAgICAgICAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAgICJj"
    "b21wbGV0ZWQgcGFyYWdyYXBoLW9yZGVyIG9ic2VydmF0aW9uIGRpZCBub3QgcHJvdmUgd29y"
    "a2VyIHRlYXJkb3duIiwKICAgICAgICApCiAgICBpZiBvYnNlcnZhdGlvbiBpcyBOb25lIG9y"
    "IHJlcG9ydCBpcyBOb25lOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAg"
    "ICAgICAgICAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIiwKICAgICAgICAgICAgImZyZXNo"
    "IHBhcmFncmFwaC1vcmRlciBvYnNlcnZhdGlvbiBkaWQgbm90IHByb2R1Y2UgY29tcGxldGUg"
    "ZXZpZGVuY2UiLAogICAgICAgICkKCiAgICByZXR1cm4gewogICAgICAgICoqb2JzZXJ2YXRp"
    "b24sCiAgICAgICAgIndvcmtlcl9zdGFydF9yZWNlaXB0X3NoYTI1NiI6IHNoYTI1Nl90ZXh0"
    "KGNhbm9uaWNhbF9qc29uKHJlcG9ydCkpLAogICAgICAgICJ0ZWFyZG93bl9zdGF0dXMiOiB0"
    "ZWFyZG93bl9zdGF0dXMsCiAgICB9CgoKZGVmIG9yZGVyX2J1bmRsZV9vdXRwdXRzKCkgLT4g"
    "ZGljdFtzdHIsIG9iamVjdF06CiAgICByZXF1aXJlZF9iZWZvcmVfbWFuaWZlc3QgPSBzZXQo"
    "T1JERVJfT1VUUFVUX05BTUVTKSAtIHsiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24ifQogICAg"
    "b2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0ID0ge3BhdGgubmFtZSBmb3IgcGF0aCBpbiBPVVRQ"
    "VVRfUk9PVC5pdGVyZGlyKCkgaWYgcGF0aC5pc19maWxlKCl9CiAgICBtaXNzaW5nID0gcmVx"
    "dWlyZWRfYmVmb3JlX21hbmlmZXN0IC0gb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0CiAgICB1"
    "bmV4cGVjdGVkID0gb2JzZXJ2ZWRfYmVmb3JlX21hbmlmZXN0IC0gcmVxdWlyZWRfYmVmb3Jl"
    "X21hbmlmZXN0CiAgICBpZiBtaXNzaW5nIG9yIHVuZXhwZWN0ZWQ6CiAgICAgICAgcmFpc2Ug"
    "UnVudGltZUVycm9yKAogICAgICAgICAgICAicGFyYWdyYXBoLW9yZGVyIGV2aWRlbmNlIG91"
    "dHB1dCBjb250cmFjdCBkcmlmdGVkOiAiCiAgICAgICAgICAgIGYibWlzc2luZz17c29ydGVk"
    "KG1pc3NpbmcpfSAiCiAgICAgICAgICAgIGYidW5leHBlY3RlZD17c29ydGVkKHVuZXhwZWN0"
    "ZWQpfSIKICAgICAgICApCgogICAgZW50cmllczogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0g"
    "PSBbXQogICAgZm9yIG5hbWUgaW4gT1JERVJfT1VUUFVUX05BTUVTOgogICAgICAgIGlmIG5h"
    "bWUgPT0gImJ1bmRsZV9tYW5pZmVzdF92MS5qc29uIjoKICAgICAgICAgICAgY29udGludWUK"
    "ICAgICAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyBuYW1lCiAgICAgICAgZW50cmllcy5hcHBl"
    "bmQoCiAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICJwYXRoIjogbmFtZSwKICAgICAg"
    "ICAgICAgICAgICJzaGEyNTYiOiBmaWxlX3NoYTI1NihwYXRoKSwKICAgICAgICAgICAgICAg"
    "ICJzaXplX2J5dGVzIjogcGF0aC5zdGF0KCkuc3Rfc2l6ZSwKICAgICAgICAgICAgfQogICAg"
    "ICAgICkKCiAgICBtYW5pZmVzdCA9IHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4w"
    "LjAiLAogICAgICAgICJkaWFnbm9zdGljX2lkIjogT1JERVJfRElBR05PU1RJQ19JRCwKICAg"
    "ICAgICAic291cmNlX21haW5fY29tbWl0IjogU09VUkNFX01BSU5fQ09NTUlULAogICAgICAg"
    "ICJtZW1iZXJzIjogZW50cmllcywKICAgICAgICAic2NyYXRjaF9kaXJlY3Rvcmllc19pbmNs"
    "dWRlZCI6IEZhbHNlLAogICAgICAgICJ3b3JrZXJfbG9nX2RpcmVjdG9yeV9pbmNsdWRlZCI6"
    "IEZhbHNlLAogICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAg"
    "InJhd19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgIH0KICAgIHdyaXRlX2pzb24oCiAg"
    "ICAgICAgT1VUUFVUX1JPT1QgLyAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iLAogICAgICAg"
    "IG1hbmlmZXN0LAogICAgKQoKICAgIHdpdGggemlwZmlsZS5aaXBGaWxlKAogICAgICAgIEVW"
    "SURFTkNFX1pJUCwKICAgICAgICAidyIsCiAgICAgICAgY29tcHJlc3Npb249emlwZmlsZS5a"
    "SVBfREVGTEFURUQsCiAgICApIGFzIGFyY2hpdmU6CiAgICAgICAgZm9yIG5hbWUgaW4gT1JE"
    "RVJfT1VUUFVUX05BTUVTOgogICAgICAgICAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyBuYW1l"
    "CiAgICAgICAgICAgIGFyY2hpdmUud3JpdGUocGF0aCwgYXJjbmFtZT1uYW1lKQoKICAgIGlm"
    "IG5vdCBFVklERU5DRV9aSVAuaXNfZmlsZSgpIG9yIEVWSURFTkNFX1pJUC5zdGF0KCkuc3Rf"
    "c2l6ZSA+IE1BWF9FVklERU5DRV9aSVBfQllURVM6CiAgICAgICAgcmFpc2UgUnVudGltZUVy"
    "cm9yKCJwYXJhZ3JhcGgtb3JkZXIgZXZpZGVuY2UgWklQIGlzIHVuYXZhaWxhYmxlIG9yIG92"
    "ZXJzaXplZCIpCgogICAgcmV0dXJuIHsKICAgICAgICAiYnVuZGxlX3N0YXR1cyI6ICJQQVNT"
    "RUQiLAogICAgICAgICJidW5kbGVfbWFuaWZlc3Rfc2hhMjU2IjogZmlsZV9zaGEyNTYoT1VU"
    "UFVUX1JPT1QgLyAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iKSwKICAgICAgICAiZXZpZGVu"
    "Y2VfemlwX3NoYTI1NiI6IGZpbGVfc2hhMjU2KEVWSURFTkNFX1pJUCksCiAgICAgICAgImV2"
    "aWRlbmNlX3ppcF9zaXplX2J5dGVzIjogRVZJREVOQ0VfWklQLnN0YXQoKS5zdF9zaXplLAog"
    "ICAgfQoKCmRlZiBjNF9yZXF1ZXN0X21lc3NhZ2VzKGZpbmFsX3VzZXJfY29udGVudDogc3Ry"
    "KSAtPiBsaXN0W2RpY3Rbc3RyLCBzdHJdXToKICAgIHJldHVybiBbCiAgICAgICAgeyJyb2xl"
    "IjogInN5c3RlbSIsICJjb250ZW50IjogU1lTVEVNX1BST01QVH0sCiAgICAgICAgeyJyb2xl"
    "IjogInVzZXIiLCAiY29udGVudCI6IEM0X0NBTk9OSUNBTF9DT05URVhUfSwKICAgICAgICB7"
    "InJvbGUiOiAiYXNzaXN0YW50IiwgImNvbnRlbnQiOiBTWU5USEVUSUNfQVNTSVNUQU5UX0FD"
    "S30sCiAgICAgICAgeyJyb2xlIjogInVzZXIiLCAiY29udGVudCI6IGZpbmFsX3VzZXJfY29u"
    "dGVudH0sCiAgICBdCgoKZGVmIGM0X3JlcXVlc3RfcGF5bG9hZCgpIC0+IGRpY3Rbc3RyLCBv"
    "YmplY3RdOgogICAgcmV0dXJuIHsKICAgICAgICAibW9kZWwiOiBTRVJWRURfTU9ERUxfTkFN"
    "RSwKICAgICAgICAibWVzc2FnZXMiOiBjNF9yZXF1ZXN0X21lc3NhZ2VzKEVYUEVDVEVEX09C"
    "SkVDVF9DQU5PTklDQUwpLAogICAgICAgICJ0ZW1wZXJhdHVyZSI6IDAsCiAgICAgICAgInRv"
    "cF9wIjogMSwKICAgICAgICAicmVwZXRpdGlvbl9wZW5hbHR5IjogMS4xLAogICAgICAgICJz"
    "ZWVkIjogNywKICAgICAgICAibWF4X3Rva2VucyI6IDMyLAogICAgICAgICJzdHJlYW0iOiBG"
    "YWxzZSwKICAgIH0KCgpkZWYgYzRfdG9rZW5pemVfcGF5bG9hZChmaW5hbF91c2VyX2NvbnRl"
    "bnQ6IHN0cikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICByZXR1cm4gewogICAgICAgICJt"
    "b2RlbCI6IFNFUlZFRF9NT0RFTF9OQU1FLAogICAgICAgICJtZXNzYWdlcyI6IGM0X3JlcXVl"
    "c3RfbWVzc2FnZXMoZmluYWxfdXNlcl9jb250ZW50KSwKICAgICAgICAiYWRkX2dlbmVyYXRp"
    "b25fcHJvbXB0IjogVHJ1ZSwKICAgICAgICAiY29udGludWVfZmluYWxfbWVzc2FnZSI6IEZh"
    "bHNlLAogICAgICAgICJhZGRfc3BlY2lhbF90b2tlbnMiOiBGYWxzZSwKICAgICAgICAicmV0"
    "dXJuX3Rva2VuX3N0cnMiOiBGYWxzZSwKICAgIH0KCgpkZWYgYzRfdG9rZW5fc2VxdWVuY2Uo"
    "CiAgICB3b3JrZXI6IFdvcmtlciwKICAgIGZpbmFsX3VzZXJfY29udGVudDogc3RyLAopIC0+"
    "IHR1cGxlW2ludCwgLi4uXToKICAgIHJlc3BvbnNlID0gcG9zdF9qc29uKAogICAgICAgIGYi"
    "aHR0cDovLzEyNy4wLjAuMTp7d29ya2VyLnBvcnR9L3Rva2VuaXplIiwKICAgICAgICBjNF90"
    "b2tlbml6ZV9wYXlsb2FkKGZpbmFsX3VzZXJfY29udGVudCksCiAgICApCiAgICByYXdfdG9r"
    "ZW5zID0gcmVzcG9uc2UuZ2V0KCJ0b2tlbnMiKQogICAgY291bnQgPSByZXNwb25zZS5nZXQo"
    "ImNvdW50IikKICAgIGlmICgKICAgICAgICBub3QgaXNpbnN0YW5jZShyYXdfdG9rZW5zLCBs"
    "aXN0KQogICAgICAgIG9yIGlzaW5zdGFuY2UoY291bnQsIGJvb2wpCiAgICAgICAgb3Igbm90"
    "IGlzaW5zdGFuY2UoY291bnQsIGludCkKICAgICAgICBvciBjb3VudCAhPSBsZW4ocmF3X3Rv"
    "a2VucykKICAgICk6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAiQzQgdG9rZW5p"
    "emF0aW9uIHJlc3BvbnNlIHNoYXBlIGlzIGludmFsaWQiLAogICAgICAgICkKICAgIHRva2Vu"
    "czogbGlzdFtpbnRdID0gW10KICAgIGZvciByYXdfdG9rZW4gaW4gcmF3X3Rva2VuczoKICAg"
    "ICAgICBpZiBpc2luc3RhbmNlKHJhd190b2tlbiwgYm9vbCkgb3Igbm90IGlzaW5zdGFuY2Uo"
    "cmF3X3Rva2VuLCBpbnQpOgogICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAg"
    "ICAgICAgIkM0IHRva2VuaXphdGlvbiByZXR1cm5lZCBhIG5vbi1pbnRlZ2VyIHRva2VuIGlk"
    "IiwKICAgICAgICAgICAgKQogICAgICAgIGlmIHJhd190b2tlbiA8IDA6CiAgICAgICAgICAg"
    "IHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VN"
    "QU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICAgICAiQzQgdG9rZW5pemF0aW9uIHJldHVy"
    "bmVkIGEgbmVnYXRpdmUgdG9rZW4gaWQiLAogICAgICAgICAgICApCiAgICAgICAgdG9rZW5z"
    "LmFwcGVuZChyYXdfdG9rZW4pCiAgICByZXR1cm4gdHVwbGUodG9rZW5zKQoKCmRlZiBjNF9s"
    "b25nZXN0X2NvbW1vbl9wcmVmaXgoCiAgICBsZWZ0OiB0dXBsZVtpbnQsIC4uLl0sCiAgICBy"
    "aWdodDogdHVwbGVbaW50LCAuLi5dLAopIC0+IHR1cGxlW2ludCwgLi4uXToKICAgIGNvdW50"
    "ID0gMAogICAgbGltaXQgPSBtaW4obGVuKGxlZnQpLCBsZW4ocmlnaHQpKQogICAgd2hpbGUg"
    "Y291bnQgPCBsaW1pdCBhbmQgbGVmdFtjb3VudF0gPT0gcmlnaHRbY291bnRdOgogICAgICAg"
    "IGNvdW50ICs9IDEKICAgIHJldHVybiBsZWZ0Wzpjb3VudF0KCgpkZWYgYzRfdG9rZW5faWRl"
    "bnRpdHkod29ya2VyOiBXb3JrZXIpIC0+IGRpY3Rbc3RyLCBvYmplY3RdOgogICAgY2Fub25p"
    "Y2FsX3Rva2VucyA9IGM0X3Rva2VuX3NlcXVlbmNlKAogICAgICAgIHdvcmtlciwKICAgICAg"
    "ICBFWFBFQ1RFRF9PQkpFQ1RfQ0FOT05JQ0FMLAogICAgKQogICAgc2VudGluZWxfdG9rZW5z"
    "ID0gYzRfdG9rZW5fc2VxdWVuY2UoCiAgICAgICAgd29ya2VyLAogICAgICAgIEM0X0ZJTkFM"
    "X1VTRVJfQk9VTkRBUllfU0VOVElORUwsCiAgICApCiAgICByZXVzYWJsZV90b2tlbnMgPSBj"
    "NF9sb25nZXN0X2NvbW1vbl9wcmVmaXgoCiAgICAgICAgY2Fub25pY2FsX3Rva2VucywKICAg"
    "ICAgICBzZW50aW5lbF90b2tlbnMsCiAgICApCiAgICBjYW5vbmljYWxfc2hhID0gc2hhMjU2"
    "X2J5dGVzKGNhbm9uaWNhbF9qc29uKGxpc3QoY2Fub25pY2FsX3Rva2VucykpLmVuY29kZSgi"
    "dXRmLTgiKSkKICAgIHJldXNhYmxlX3NoYSA9IHNoYTI1Nl9ieXRlcyhjYW5vbmljYWxfanNv"
    "bihsaXN0KHJldXNhYmxlX3Rva2VucykpLmVuY29kZSgidXRmLTgiKSkKICAgIGlmICgKICAg"
    "ICAgICBsZW4oY2Fub25pY2FsX3Rva2VucykgIT0gQzRfRlVMTF9QUk9NUFRfVE9LRU5fQ09V"
    "TlQKICAgICAgICBvciBjYW5vbmljYWxfc2hhICE9IEM0X0ZVTExfUFJPTVBUX1RPS0VOX1NI"
    "QTI1NgogICAgKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAg"
    "ICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICJDNCBmdWxsLXBy"
    "b21wdCB0b2tlbiBpZGVudGl0eSBkcmlmdGVkIGJlZm9yZSBtb2RlbCByZXF1ZXN0IiwKICAg"
    "ICAgICApCiAgICBpZiAoCiAgICAgICAgbGVuKHJldXNhYmxlX3Rva2VucykgIT0gQzRfUkVV"
    "U0FCTEVfUFJFRklYX1RPS0VOX0NPVU5UCiAgICAgICAgb3IgcmV1c2FibGVfc2hhICE9IEM0"
    "X1JFVVNBQkxFX1BSRUZJWF9UT0tFTl9TSEEyNTYKICAgICk6CiAgICAgICAgcmFpc2UgRGlh"
    "Z25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUi"
    "LAogICAgICAgICAgICAiQzQgcmV1c2FibGUtcHJlZml4IHRva2VuIGlkZW50aXR5IGRyaWZ0"
    "ZWQgYmVmb3JlIG1vZGVsIHJlcXVlc3QiLAogICAgICAgICkKICAgIGlmIGxlbihyZXVzYWJs"
    "ZV90b2tlbnMpID49IGxlbihjYW5vbmljYWxfdG9rZW5zKToKICAgICAgICByYWlzZSBEaWFn"
    "bm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJDNCBmaW5hbC11c2VyIGJvdW5kYXJ5IGRpZCBub3QgZGl2ZXJnZSIs"
    "CiAgICAgICAgKQogICAgcmV0dXJuIHsKICAgICAgICAidG9rZW5fY291bnQiOiBsZW4oY2Fu"
    "b25pY2FsX3Rva2VucyksCiAgICAgICAgInRva2VuX3NoYTI1NiI6IGNhbm9uaWNhbF9zaGEs"
    "CiAgICAgICAgInJldXNhYmxlX3ByZWZpeF90b2tlbl9jb3VudCI6IGxlbihyZXVzYWJsZV90"
    "b2tlbnMpLAogICAgICAgICJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hhMjU2IjogcmV1c2Fi"
    "bGVfc2hhLAogICAgICAgICJzZW50aW5lbF9wcm9tcHRfdG9rZW5fY291bnQiOiBsZW4oc2Vu"
    "dGluZWxfdG9rZW5zKSwKICAgICAgICAic2VudGluZWxfcHJvbXB0X3Rva2VuX3NoYTI1NiI6"
    "IHNoYTI1Nl9ieXRlcygKICAgICAgICAgICAgY2Fub25pY2FsX2pzb24obGlzdChzZW50aW5l"
    "bF90b2tlbnMpKS5lbmNvZGUoInV0Zi04IikKICAgICAgICApLAogICAgICAgICJmaXJzdF9k"
    "aXZlcmdlbnRfdG9rZW5faW5kZXgiOiBsZW4ocmV1c2FibGVfdG9rZW5zKSwKICAgIH0KCgpk"
    "ZWYgaW5pdGlhbGl6ZV9jNF9qb3VybmFsKCkgLT4gTm9uZToKICAgIGlmIFBSRV9SRVFVRVNU"
    "X1RPS0VOX0lERU5USVRZX0pPVVJOQUwuZXhpc3RzKCk6CiAgICAgICAgcmFpc2UgUnVudGlt"
    "ZUVycm9yKCJwcmUtcmVxdWVzdCB0b2tlbi1pZGVudGl0eSBqb3VybmFsIGFscmVhZHkgZXhp"
    "c3RzIikKICAgIHdyaXRlX2pzb24oCiAgICAgICAgUFJFX1JFUVVFU1RfVE9LRU5fSURFTlRJ"
    "VFlfSk9VUk5BTCwKICAgICAgICB7CiAgICAgICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIx"
    "LjAuMCIsCiAgICAgICAgICAgICJqb3VybmFsX2lkIjogKAogICAgICAgICAgICAgICAgImF1"
    "cmFnYXRld2F5LWNhbm9uaWNhbC1zeW50aGV0aWMtcHJlZml4LWM0LXByZS1yZXF1ZXN0LXRv"
    "a2VuLWlkZW50aXR5LXYxIgogICAgICAgICAgICApLAogICAgICAgICAgICAicXVhbGlmaWNh"
    "dGlvbl9pZCI6IEM0X1FVQUxJRklDQVRJT05fSUQsCiAgICAgICAgICAgICJlbnRyaWVzIjog"
    "W10sCiAgICAgICAgICAgICJyYXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAg"
    "ICAgICJyYXdfbW9kZWxfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgfSwKICAg"
    "ICkKCgpkZWYgcGVyc2lzdF9jNF9wcmVfcmVxdWVzdF9pZGVudGl0eSgKICAgIHJlcXVlc3Rf"
    "b3JkaW5hbDogaW50LAogICAgdG9rZW5faWRlbnRpdHk6IGRpY3Rbc3RyLCBvYmplY3RdLAog"
    "ICAgcGF5bG9hZF9zaGEyNTY6IHN0ciwKKSAtPiBOb25lOgogICAgam91cm5hbCA9IF9yZWFk"
    "X3ByZV9yZXF1ZXN0X3Rva2VuX2lkZW50aXR5X2pvdXJuYWwoKQogICAgZW50cmllcyA9IGpv"
    "dXJuYWwuZ2V0KCJlbnRyaWVzIikKICAgIGlmIG5vdCBpc2luc3RhbmNlKGVudHJpZXMsIGxp"
    "c3QpOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigicHJlLXJlcXVlc3QgdG9rZW4taWRl"
    "bnRpdHkgam91cm5hbCBlbnRyaWVzIGFyZSBpbnZhbGlkIikKICAgIGlmIHJlcXVlc3Rfb3Jk"
    "aW5hbCAhPSBsZW4oZW50cmllcykgKyAxOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigi"
    "cHJlLXJlcXVlc3QgdG9rZW4taWRlbnRpdHkgcmVxdWVzdCBvcmRpbmFsIGRyaWZ0ZWQiKQog"
    "ICAgaWYgcGF5bG9hZF9zaGEyNTYgIT0gQzRfUkVRVUVTVF9QQVlMT0FEX1NIQTI1NjoKICAg"
    "ICAgICByYWlzZSBSdW50aW1lRXJyb3IoIkM0IHJlcXVlc3QgcGF5bG9hZCBpZGVudGl0eSBk"
    "cmlmdGVkIikKICAgIHdyaXRlX2pzb24oCiAgICAgICAgUFJFX1JFUVVFU1RfVE9LRU5fSURF"
    "TlRJVFlfSk9VUk5BTCwKICAgICAgICB7CiAgICAgICAgICAgICoqam91cm5hbCwKICAgICAg"
    "ICAgICAgImVudHJpZXMiOiBbCiAgICAgICAgICAgICAgICAqZW50cmllcywKICAgICAgICAg"
    "ICAgICAgIHsKICAgICAgICAgICAgICAgICAgICAicmVxdWVzdF9vcmRpbmFsIjogcmVxdWVz"
    "dF9vcmRpbmFsLAogICAgICAgICAgICAgICAgICAgICJvYnNlcnZhdGlvbl9pZCI6IGYiQzRf"
    "T0JTRVJWQVRJT05fe3JlcXVlc3Rfb3JkaW5hbH0iLAogICAgICAgICAgICAgICAgICAgICJ0"
    "b2tlbl9jb3VudCI6IHRva2VuX2lkZW50aXR5WyJ0b2tlbl9jb3VudCJdLAogICAgICAgICAg"
    "ICAgICAgICAgICJ0b2tlbl9zaGEyNTYiOiB0b2tlbl9pZGVudGl0eVsidG9rZW5fc2hhMjU2"
    "Il0sCiAgICAgICAgICAgICAgICAgICAgInJldXNhYmxlX3ByZWZpeF90b2tlbl9jb3VudCI6"
    "ICh0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJlZml4X3Rva2VuX2NvdW50Il0pLAogICAg"
    "ICAgICAgICAgICAgICAgICJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hhMjU2IjogKAogICAg"
    "ICAgICAgICAgICAgICAgICAgICB0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJlZml4X3Rv"
    "a2VuX3NoYTI1NiJdCiAgICAgICAgICAgICAgICAgICAgKSwKICAgICAgICAgICAgICAgICAg"
    "ICAicGF5bG9hZF9zaGEyNTYiOiBwYXlsb2FkX3NoYTI1NiwKICAgICAgICAgICAgICAgICAg"
    "ICAicGVyc2lzdGVkX2JlZm9yZV9tb2RlbF9yZXF1ZXN0IjogVHJ1ZSwKICAgICAgICAgICAg"
    "ICAgIH0sCiAgICAgICAgICAgIF0sCiAgICAgICAgfSwKICAgICkKCgpkZWYgYzRfb2JqZWN0"
    "X3BhaXJzKAogICAgcGFpcnM6IGxpc3RbdHVwbGVbc3RyLCBvYmplY3RdXSwKKSAtPiBkaWN0"
    "W3N0ciwgb2JqZWN0XToKICAgIHJlc3VsdDogZGljdFtzdHIsIG9iamVjdF0gPSB7fQogICAg"
    "Zm9yIGtleSwgdmFsdWUgaW4gcGFpcnM6CiAgICAgICAgaWYga2V5IGluIHJlc3VsdDoKICAg"
    "ICAgICAgICAgcmFpc2UgQzREdXBsaWNhdGVLZXlFcnJvcihrZXkpCiAgICAgICAgcmVzdWx0"
    "W2tleV0gPSB2YWx1ZQogICAgcmV0dXJuIHJlc3VsdAoKCmRlZiBjNF9qc29uX3R5cGUodmFs"
    "dWU6IG9iamVjdCkgLT4gc3RyOgogICAgaWYgdmFsdWUgaXMgTm9uZToKICAgICAgICByZXR1"
    "cm4gIm51bGwiCiAgICBpZiBpc2luc3RhbmNlKHZhbHVlLCBib29sKToKICAgICAgICByZXR1"
    "cm4gImJvb2xlYW4iCiAgICBpZiB0eXBlKHZhbHVlKSBpcyBpbnQ6CiAgICAgICAgcmV0dXJu"
    "ICJpbnRlZ2VyIgogICAgaWYgaXNpbnN0YW5jZSh2YWx1ZSwgZmxvYXQpOgogICAgICAgIHJl"
    "dHVybiAibnVtYmVyIgogICAgaWYgaXNpbnN0YW5jZSh2YWx1ZSwgc3RyKToKICAgICAgICBy"
    "ZXR1cm4gInN0cmluZyIKICAgIGlmIGlzaW5zdGFuY2UodmFsdWUsIGxpc3QpOgogICAgICAg"
    "IHJldHVybiAiYXJyYXkiCiAgICBpZiBpc2luc3RhbmNlKHZhbHVlLCBkaWN0KToKICAgICAg"
    "ICByZXR1cm4gIm9iamVjdCIKICAgIHJldHVybiAidW5rbm93biIKCgpkZWYgYzRfcmVzcG9u"
    "c2VfcHJvamVjdGlvbigKICAgIGNvbnRlbnQ6IHN0ciwKICAgIGZpbmlzaF9yZWFzb246IG9i"
    "amVjdCwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIHN0cmlwcGVkID0gY29udGVudC5z"
    "dHJpcCgpCiAgICBmaXJzdCA9IHN0cmlwcGVkWzBdIGlmIHN0cmlwcGVkIGVsc2UgTm9uZQog"
    "ICAgbGFzdCA9IHN0cmlwcGVkWy0xXSBpZiBzdHJpcHBlZCBlbHNlIE5vbmUKICAgIGxlYWRp"
    "bmdfY29udGVudCA9IGZpcnN0ICE9ICJ7IgogICAgdHJhaWxpbmdfY29udGVudCA9IGxhc3Qg"
    "IT0gIn0iCiAgICBtYXJrZG93bl9mZW5jZSA9ICJgYGAiIGluIGNvbnRlbnQKCiAgICB2YWxp"
    "ZF9qc29uID0gRmFsc2UKICAgIGR1cGxpY2F0ZV9rZXlfZGV0ZWN0ZWQgPSBGYWxzZQogICAg"
    "cGFyc2VkOiBvYmplY3QgPSBOb25lCiAgICBqc29uX2Vycm9yX2xpbmU6IGludCB8IE5vbmUg"
    "PSBOb25lCiAgICBqc29uX2Vycm9yX2NvbHVtbjogaW50IHwgTm9uZSA9IE5vbmUKICAgIGpz"
    "b25fZXJyb3JfcG9zaXRpb246IGludCB8IE5vbmUgPSBOb25lCgogICAgdHJ5OgogICAgICAg"
    "IHBhcnNlZCA9IGpzb24ubG9hZHMoCiAgICAgICAgICAgIGNvbnRlbnQsCiAgICAgICAgICAg"
    "IG9iamVjdF9wYWlyc19ob29rPWM0X29iamVjdF9wYWlycywKICAgICAgICApCiAgICAgICAg"
    "dmFsaWRfanNvbiA9IFRydWUKICAgIGV4Y2VwdCBDNER1cGxpY2F0ZUtleUVycm9yOgogICAg"
    "ICAgIGR1cGxpY2F0ZV9rZXlfZGV0ZWN0ZWQgPSBUcnVlCiAgICBleGNlcHQganNvbi5KU09O"
    "RGVjb2RlRXJyb3IgYXMgZXJyb3I6CiAgICAgICAganNvbl9lcnJvcl9saW5lID0gZXJyb3Iu"
    "bGluZW5vCiAgICAgICAganNvbl9lcnJvcl9jb2x1bW4gPSBlcnJvci5jb2xubwogICAgICAg"
    "IGpzb25fZXJyb3JfcG9zaXRpb24gPSBlcnJvci5wb3MKCiAgICBwYXJzZWRfa2V5X3NldDog"
    "bGlzdFtzdHJdID0gW10KICAgIHByb2JlX2pzb25fdHlwZTogc3RyIHwgTm9uZSA9IE5vbmUK"
    "ICAgIHZhbHVlX2pzb25fdHlwZTogc3RyIHwgTm9uZSA9IE5vbmUKICAgIHByb2JlX2V4YWN0"
    "ID0gRmFsc2UKICAgIHZhbHVlX2V4YWN0ID0gRmFsc2UKICAgIGNhbm9uaWNhbF9wYXJzZWRf"
    "b2JqZWN0X3NoYTI1Njogc3RyIHwgTm9uZSA9IE5vbmUKCiAgICBpZiB2YWxpZF9qc29uOgog"
    "ICAgICAgIGNhbm9uaWNhbF9wYXJzZWRfb2JqZWN0X3NoYTI1NiA9IHNoYTI1Nl90ZXh0KGNh"
    "bm9uaWNhbF9qc29uKHBhcnNlZCkpCiAgICAgICAgaWYgaXNpbnN0YW5jZShwYXJzZWQsIGRp"
    "Y3QpOgogICAgICAgICAgICBwYXJzZWRfa2V5X3NldCA9IHNvcnRlZChrZXkgZm9yIGtleSBp"
    "biBwYXJzZWQgaWYgaXNpbnN0YW5jZShrZXksIHN0cikpCiAgICAgICAgICAgIHByb2JlID0g"
    "cGFyc2VkLmdldCgicHJvYmUiKQogICAgICAgICAgICB2YWx1ZSA9IHBhcnNlZC5nZXQoInZh"
    "bHVlIikKICAgICAgICAgICAgcHJvYmVfanNvbl90eXBlID0gYzRfanNvbl90eXBlKHByb2Jl"
    "KQogICAgICAgICAgICB2YWx1ZV9qc29uX3R5cGUgPSBjNF9qc29uX3R5cGUodmFsdWUpCiAg"
    "ICAgICAgICAgIHByb2JlX2V4YWN0ID0gaXNpbnN0YW5jZShwcm9iZSwgc3RyKSBhbmQgcHJv"
    "YmUgPT0gImV4YWN0LXJ1bnRpbWUtcDUtcDYiCiAgICAgICAgICAgIHZhbHVlX2V4YWN0ID0g"
    "aXNpbnN0YW5jZSh2YWx1ZSwgaW50KSBhbmQgbm90IGlzaW5zdGFuY2UodmFsdWUsIGJvb2wp"
    "IGFuZCB2YWx1ZSA9PSAxCgogICAgcmVzcG9uc2VfY29tcGxldGUgPSBmaW5pc2hfcmVhc29u"
    "ID09ICJzdG9wIgogICAgZXhhY3Rfa2V5X3NldCA9IHBhcnNlZF9rZXlfc2V0ID09IFsicHJv"
    "YmUiLCAidmFsdWUiXQogICAgZXhhY3Rfb2JqZWN0ID0gKAogICAgICAgIHZhbGlkX2pzb24K"
    "ICAgICAgICBhbmQgbm90IGR1cGxpY2F0ZV9rZXlfZGV0ZWN0ZWQKICAgICAgICBhbmQgaXNp"
    "bnN0YW5jZShwYXJzZWQsIGRpY3QpCiAgICAgICAgYW5kIGV4YWN0X2tleV9zZXQKICAgICAg"
    "ICBhbmQgcHJvYmVfZXhhY3QKICAgICAgICBhbmQgdmFsdWVfZXhhY3QKICAgICAgICBhbmQg"
    "bm90IGxlYWRpbmdfY29udGVudAogICAgICAgIGFuZCBub3QgdHJhaWxpbmdfY29udGVudAog"
    "ICAgICAgIGFuZCBub3QgbWFya2Rvd25fZmVuY2UKICAgICAgICBhbmQgcmVzcG9uc2VfY29t"
    "cGxldGUKICAgICkKCiAgICByZXR1cm4gewogICAgICAgICJ2YWxpZF9qc29uIjogdmFsaWRf"
    "anNvbiwKICAgICAgICAiZHVwbGljYXRlX2tleV9kZXRlY3RlZCI6IGR1cGxpY2F0ZV9rZXlf"
    "ZGV0ZWN0ZWQsCiAgICAgICAgImpzb25fZXJyb3JfbGluZSI6IGpzb25fZXJyb3JfbGluZSwK"
    "ICAgICAgICAianNvbl9lcnJvcl9jb2x1bW4iOiBqc29uX2Vycm9yX2NvbHVtbiwKICAgICAg"
    "ICAianNvbl9lcnJvcl9wb3NpdGlvbiI6IGpzb25fZXJyb3JfcG9zaXRpb24sCiAgICAgICAg"
    "Impzb25fcm9vdF90eXBlIjogYzRfanNvbl90eXBlKHBhcnNlZCkgaWYgdmFsaWRfanNvbiBl"
    "bHNlIE5vbmUsCiAgICAgICAgInBhcnNlZF9rZXlfc2V0IjogcGFyc2VkX2tleV9zZXQsCiAg"
    "ICAgICAgInByb2JlX2pzb25fdHlwZSI6IHByb2JlX2pzb25fdHlwZSwKICAgICAgICAidmFs"
    "dWVfanNvbl90eXBlIjogdmFsdWVfanNvbl90eXBlLAogICAgICAgICJwcm9iZV9leGFjdCI6"
    "IHByb2JlX2V4YWN0LAogICAgICAgICJ2YWx1ZV9leGFjdCI6IHZhbHVlX2V4YWN0LAogICAg"
    "ICAgICJjYW5vbmljYWxfcGFyc2VkX29iamVjdF9zaGEyNTYiOiAoY2Fub25pY2FsX3BhcnNl"
    "ZF9vYmplY3Rfc2hhMjU2KSwKICAgICAgICAiY2Fub25pY2FsX2V4cGVjdGVkX29iamVjdF9z"
    "aGEyNTYiOiAoQzRfRVhQRUNURURfT0JKRUNUX1NIQTI1NiksCiAgICAgICAgImxlYWRpbmdf"
    "bm9uX3doaXRlc3BhY2VfY29udGVudF9kZXRlY3RlZCI6IGxlYWRpbmdfY29udGVudCwKICAg"
    "ICAgICAidHJhaWxpbmdfbm9uX3doaXRlc3BhY2VfY29udGVudF9kZXRlY3RlZCI6IHRyYWls"
    "aW5nX2NvbnRlbnQsCiAgICAgICAgIm1hcmtkb3duX2ZlbmNlX2RldGVjdGVkIjogbWFya2Rv"
    "d25fZmVuY2UsCiAgICAgICAgInJlc3BvbnNlX2NvbXBsZXRlIjogcmVzcG9uc2VfY29tcGxl"
    "dGUsCiAgICAgICAgImV4YWN0X29iamVjdCI6IGV4YWN0X29iamVjdCwKICAgIH0KCgpkZWYg"
    "cnVuX2M0X29ic2VydmF0aW9uKAogICAgd29ya2VyOiBXb3JrZXIsCiAgICBzZXF1ZW5jZV9p"
    "bmRleDogaW50LAogICAgY291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAopIC0+IGRpY3Rbc3Ry"
    "LCBvYmplY3RdOgogICAgdG9rZW5faWRlbnRpdHkgPSBjNF90b2tlbl9pZGVudGl0eSh3b3Jr"
    "ZXIpCiAgICBwYXlsb2FkID0gYzRfcmVxdWVzdF9wYXlsb2FkKCkKICAgIHBheWxvYWRfc2hh"
    "MjU2ID0gc2hhMjU2X3RleHQoY2Fub25pY2FsX2pzb24ocGF5bG9hZCkpCgogICAgaWYgcGF5"
    "bG9hZF9zaGEyNTYgIT0gQzRfUkVRVUVTVF9QQVlMT0FEX1NIQTI1NjoKICAgICAgICByYWlz"
    "ZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJ"
    "TFVSRSIsCiAgICAgICAgICAgICJDNCByZXF1ZXN0IHBheWxvYWQgaWRlbnRpdHkgZHJpZnRl"
    "ZCBiZWZvcmUgbW9kZWwgcmVxdWVzdCIsCiAgICAgICAgKQoKICAgIHJlcXVlc3Rfb3JkaW5h"
    "bCA9IGNvdW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdICsgMQogICAgaWYgcmVxdWVzdF9vcmRp"
    "bmFsICE9IHNlcXVlbmNlX2luZGV4OgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJl"
    "KAogICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAg"
    "ICAgICAgIkM0IHByZS1yZXF1ZXN0IG9yZGluYWwgZHJpZnRlZCIsCiAgICAgICAgKQoKICAg"
    "IHBlcnNpc3RfYzRfcHJlX3JlcXVlc3RfaWRlbnRpdHkoCiAgICAgICAgcmVxdWVzdF9vcmRp"
    "bmFsLAogICAgICAgIHRva2VuX2lkZW50aXR5LAogICAgICAgIHBheWxvYWRfc2hhMjU2LAog"
    "ICAgKQoKICAgIGJhc2VsaW5lID0gdmFsaWRhdGVfemVyb19jYWNoZV9iYXNlbGluZSh3b3Jr"
    "ZXIpCiAgICB6ZXJvX2NhY2hlX3JlY2VpcHRfc2hhMjU2ID0gc2hhMjU2X3RleHQoY2Fub25p"
    "Y2FsX2pzb24oYmFzZWxpbmUpKQogICAgYmVmb3JlID0gd29ya2VyLm1ldHJpY19zbmFwc2hv"
    "dCgpCgogICAgY29uc3VtZV9hY3Rpb25zKGNvdW50ZXJzLCAibW9kZWxfcmVxdWVzdHMiKQog"
    "ICAgZW5jb2RlZCA9IGNhbm9uaWNhbF9qc29uKHBheWxvYWQpLmVuY29kZSgidXRmLTgiKQog"
    "ICAgcmVxdWVzdCA9IHVybGxpYi5yZXF1ZXN0LlJlcXVlc3QoCiAgICAgICAgYm91bmRlZF9s"
    "b29wYmFjayhmImh0dHA6Ly8xMjcuMC4wLjE6e3dvcmtlci5wb3J0fS92MS9jaGF0L2NvbXBs"
    "ZXRpb25zIiksCiAgICAgICAgZGF0YT1lbmNvZGVkLAogICAgICAgIGhlYWRlcnM9eyJDb250"
    "ZW50LVR5cGUiOiAiYXBwbGljYXRpb24vanNvbiJ9LAogICAgICAgIG1ldGhvZD0iUE9TVCIs"
    "CiAgICApCgogICAgdHJ5OgogICAgICAgIHdpdGggdXJsbGliLnJlcXVlc3QudXJsb3BlbigK"
    "ICAgICAgICAgICAgcmVxdWVzdCwKICAgICAgICAgICAgdGltZW91dD1IVFRQX1RJTUVPVVRf"
    "U0VDT05EUywKICAgICAgICApIGFzIHJlc3BvbnNlOgogICAgICAgICAgICBzdGF0dXNfY29k"
    "ZSA9IHJlc3BvbnNlLnN0YXR1cwogICAgICAgICAgICByZXNwb25zZV9wYXlsb2FkID0gcmVz"
    "cG9uc2UucmVhZCgpCiAgICBleGNlcHQgdXJsbGliLmVycm9yLkhUVFBFcnJvciBhcyBlcnJv"
    "cjoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVF"
    "U1RfRVhFQ1VUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICBmIkM0IHJlcXVlc3QgSFRUUCBm"
    "YWlsdXJlOiB7ZXJyb3IuY29kZX0iLAogICAgICAgICkgZnJvbSBlcnJvcgogICAgZXhjZXB0"
    "ICh1cmxsaWIuZXJyb3IuVVJMRXJyb3IsIFRpbWVvdXRFcnJvcikgYXMgZXJyb3I6CiAgICAg"
    "ICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJSRVFVRVNUX0VYRUNV"
    "VElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgIkM0IHJlcXVlc3QgdHJhbnNwb3J0IGZhaWxl"
    "ZCIsCiAgICAgICAgKSBmcm9tIGVycm9yCgogICAgaWYgc3RhdHVzX2NvZGUgIT0gMjAwOgog"
    "ICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVRVUVTVF9F"
    "WEVDVVRJT05fRkFJTFVSRSIsCiAgICAgICAgICAgICJDNCByZXF1ZXN0IHJldHVybmVkIGFu"
    "IHVuZXhwZWN0ZWQgSFRUUCBzdGF0dXMiLAogICAgICAgICkKCiAgICB0cnk6CiAgICAgICAg"
    "ZW52ZWxvcGUgPSBqc29uLmxvYWRzKHJlc3BvbnNlX3BheWxvYWQpCiAgICBleGNlcHQgKGpz"
    "b24uSlNPTkRlY29kZUVycm9yLCBVbmljb2RlRGVjb2RlRXJyb3IpIGFzIGVycm9yOgogICAg"
    "ICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRS"
    "QUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAiQzQgcmVzcG9uc2UgZW52ZWxvcGUgaXMgbm90"
    "IHZhbGlkIEpTT04iLAogICAgICAgICkgZnJvbSBlcnJvcgogICAgaWYgbm90IGlzaW5zdGFu"
    "Y2UoZW52ZWxvcGUsIGRpY3QpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAog"
    "ICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAgICAgICAiQzQg"
    "cmVzcG9uc2UgZW52ZWxvcGUgcm9vdCBpcyBpbnZhbGlkIiwKICAgICAgICApCgogICAgdXNh"
    "Z2UgPSBlbnZlbG9wZS5nZXQoInVzYWdlIikKICAgIGNob2ljZXMgPSBlbnZlbG9wZS5nZXQo"
    "ImNob2ljZXMiKQogICAgaWYgbm90IGlzaW5zdGFuY2UodXNhZ2UsIGRpY3QpOgogICAgICAg"
    "IHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNU"
    "X0ZBSUxVUkUiLAogICAgICAgICAgICAiQzQgcmVzcG9uc2UgdXNhZ2UgaXMgbWlzc2luZyIs"
    "CiAgICAgICAgKQogICAgaWYgbm90IGlzaW5zdGFuY2UoY2hvaWNlcywgbGlzdCkgb3IgbGVu"
    "KGNob2ljZXMpICE9IDE6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAg"
    "ICAgICAgICJPVVRQVVRfQ09OVFJBQ1RfRkFJTFVSRSIsCiAgICAgICAgICAgICJDNCByZXNw"
    "b25zZSBjaG9pY2VzIGFyZSBpbnZhbGlkIiwKICAgICAgICApCiAgICBjaG9pY2UgPSBjaG9p"
    "Y2VzWzBdCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2UsIGRpY3QpOgogICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZB"
    "SUxVUkUiLAogICAgICAgICAgICAiQzQgcmVzcG9uc2UgY2hvaWNlIGlzIGludmFsaWQiLAog"
    "ICAgICAgICkKICAgIG1lc3NhZ2UgPSBjaG9pY2UuZ2V0KCJtZXNzYWdlIikKICAgIGlmIG5v"
    "dCBpc2luc3RhbmNlKG1lc3NhZ2UsIGRpY3QpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNG"
    "YWlsdXJlKAogICAgICAgICAgICAiT1VUUFVUX0NPTlRSQUNUX0ZBSUxVUkUiLAogICAgICAg"
    "ICAgICAiQzQgcmVzcG9uc2UgbWVzc2FnZSBpcyBpbnZhbGlkIiwKICAgICAgICApCiAgICBj"
    "b250ZW50ID0gbWVzc2FnZS5nZXQoImNvbnRlbnQiKQogICAgaWYgbm90IGlzaW5zdGFuY2Uo"
    "Y29udGVudCwgc3RyKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAg"
    "ICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAgICAgIkM0IHJlc3Bv"
    "bnNlIGNvbnRlbnQgaXMgbm90IGEgc3RyaW5nIiwKICAgICAgICApCgogICAgY29tcGxldGlv"
    "bl90b2tlbnMgPSB1c2FnZS5nZXQoImNvbXBsZXRpb25fdG9rZW5zIikKICAgIHByb21wdF90"
    "b2tlbnMgPSB1c2FnZS5nZXQoInByb21wdF90b2tlbnMiKQogICAgaWYgKAogICAgICAgIGlz"
    "aW5zdGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGJvb2wpCiAgICAgICAgb3Igbm90IGlzaW5z"
    "dGFuY2UoY29tcGxldGlvbl90b2tlbnMsIGludCkKICAgICAgICBvciBub3QgMSA8PSBjb21w"
    "bGV0aW9uX3Rva2VucyA8PSAzMgogICAgKToKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFp"
    "bHVyZSgKICAgICAgICAgICAgIk9VVFBVVF9DT05UUkFDVF9GQUlMVVJFIiwKICAgICAgICAg"
    "ICAgIkM0IGNvbXBsZXRpb24tdG9rZW4gYnVkZ2V0IGRyaWZ0ZWQiLAogICAgICAgICkKICAg"
    "IGlmICgKICAgICAgICBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGJvb2wpCiAgICAgICAg"
    "b3Igbm90IGlzaW5zdGFuY2UocHJvbXB0X3Rva2VucywgaW50KQogICAgICAgIG9yIHByb21w"
    "dF90b2tlbnMgIT0gQzRfRlVMTF9QUk9NUFRfVE9LRU5fQ09VTlQKICAgICk6CiAgICAgICAg"
    "cmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICJPVVRQVVRfQ09OVFJBQ1Rf"
    "RkFJTFVSRSIsCiAgICAgICAgICAgICJDNCByZXNwb25zZSBwcm9tcHQtdG9rZW4gY291bnQg"
    "ZHJpZnRlZCIsCiAgICAgICAgKQoKICAgIGFmdGVyID0gd29ya2VyLm1ldHJpY19zbmFwc2hv"
    "dCgpCiAgICBkZWx0YSA9IG1ldHJpY19kZWx0YShiZWZvcmUsIGFmdGVyKQogICAgd29ya2Vy"
    "X2hlYWx0aF9hZnRlcl9yZXF1ZXN0ID0gVHJ1ZQoKICAgIGZpbmlzaF9yZWFzb24gPSBjaG9p"
    "Y2UuZ2V0KCJmaW5pc2hfcmVhc29uIikKICAgIHByb2plY3Rpb24gPSBjNF9yZXNwb25zZV9w"
    "cm9qZWN0aW9uKAogICAgICAgIGNvbnRlbnQsCiAgICAgICAgZmluaXNoX3JlYXNvbiwKICAg"
    "ICkKCiAgICB3b3JrZXJfcmVwb3J0ID0gd29ya2VyLnJlcG9ydCgpCiAgICBwaWQgPSB3b3Jr"
    "ZXJfcmVwb3J0LmdldCgicGlkIikKICAgIHByb2Nlc3Nfc3RhcnRfdGlja3NfdmFsdWUgPSB3"
    "b3JrZXJfcmVwb3J0LmdldCgicHJvY2Vzc19zdGFydF90aWNrcyIpCiAgICBwcm9jZXNzX2lk"
    "ZW50aXR5X3NoYTI1NiA9IHNoYTI1Nl90ZXh0KAogICAgICAgIGNhbm9uaWNhbF9qc29uKAog"
    "ICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAicGlkIjogcGlkLAogICAgICAgICAgICAg"
    "ICAgInByb2Nlc3Nfc3RhcnRfdGlja3MiOiBwcm9jZXNzX3N0YXJ0X3RpY2tzX3ZhbHVlLAog"
    "ICAgICAgICAgICB9CiAgICAgICAgKQogICAgKQoKICAgIHJldHVybiB7CiAgICAgICAgIm9i"
    "c2VydmF0aW9uX2lkIjogZiJDNF9PQlNFUlZBVElPTl97c2VxdWVuY2VfaW5kZXh9IiwKICAg"
    "ICAgICAicmVxdWVzdF9vcmRpbmFsIjogcmVxdWVzdF9vcmRpbmFsLAogICAgICAgICJzZXF1"
    "ZW5jZV9pbmRleCI6IHNlcXVlbmNlX2luZGV4LAogICAgICAgICJ3b3JrZXJfaW5zdGFuY2Vf"
    "aWQiOiB3b3JrZXIuaW5zdGFuY2VfaWQsCiAgICAgICAgIndvcmtlcl9wcm9jZXNzX2lkZW50"
    "aXR5X3NoYTI1NiI6IHByb2Nlc3NfaWRlbnRpdHlfc2hhMjU2LAogICAgICAgICJ0b2tlbl9j"
    "b3VudCI6IHRva2VuX2lkZW50aXR5WyJ0b2tlbl9jb3VudCJdLAogICAgICAgICJ0b2tlbl9z"
    "aGEyNTYiOiB0b2tlbl9pZGVudGl0eVsidG9rZW5fc2hhMjU2Il0sCiAgICAgICAgInJldXNh"
    "YmxlX3ByZWZpeF90b2tlbl9jb3VudCI6ICh0b2tlbl9pZGVudGl0eVsicmV1c2FibGVfcHJl"
    "Zml4X3Rva2VuX2NvdW50Il0pLAogICAgICAgICJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hh"
    "MjU2IjogKHRva2VuX2lkZW50aXR5WyJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hhMjU2Il0p"
    "LAogICAgICAgICJwYXlsb2FkX3NoYTI1NiI6IHBheWxvYWRfc2hhMjU2LAogICAgICAgICJ6"
    "ZXJvX2NhY2hlX2Jhc2VsaW5lIjogVHJ1ZSwKICAgICAgICAiemVyb19jYWNoZV9iYXNlbGlu"
    "ZV9yZWNlaXB0X3NoYTI1NiI6ICh6ZXJvX2NhY2hlX3JlY2VpcHRfc2hhMjU2KSwKICAgICAg"
    "ICAiaHR0cF9zdGF0dXMiOiBzdGF0dXNfY29kZSwKICAgICAgICAicmVzcG9uc2Vfc2hhMjU2"
    "Ijogc2hhMjU2X3RleHQoY29udGVudCksCiAgICAgICAgInJlc3BvbnNlX2xlbmd0aCI6IGxl"
    "bihjb250ZW50KSwKICAgICAgICAiZmluaXNoX3JlYXNvbiI6IGZpbmlzaF9yZWFzb24sCiAg"
    "ICAgICAgInByb21wdF90b2tlbnMiOiBwcm9tcHRfdG9rZW5zLAogICAgICAgICJjb21wbGV0"
    "aW9uX3Rva2VucyI6IGNvbXBsZXRpb25fdG9rZW5zLAogICAgICAgICoqcHJvamVjdGlvbiwK"
    "ICAgICAgICAid29ya2VyX2hlYWx0aF9hZnRlcl9yZXF1ZXN0Ijogd29ya2VyX2hlYWx0aF9h"
    "ZnRlcl9yZXF1ZXN0LAogICAgICAgICJyZXF1ZXN0X2Vycm9yIjogTm9uZSwKICAgICAgICAi"
    "dHJhbnNwb3J0X2Vycm9yIjogTm9uZSwKICAgICAgICAibWV0cmljX2RlbHRhIjogYXNkaWN0"
    "KGRlbHRhKSwKICAgICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAg"
    "ICJyYXdfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICB9CgoKZGVmIGM0X3B1YmxpY19v"
    "YnNlcnZhdGlvbigKICAgIG9ic2VydmF0aW9uOiBkaWN0W3N0ciwgb2JqZWN0XSwKKSAtPiBk"
    "aWN0W3N0ciwgb2JqZWN0XToKICAgIHBlcm1pdHRlZCA9ICgKICAgICAgICAib2JzZXJ2YXRp"
    "b25faWQiLAogICAgICAgICJyZXF1ZXN0X29yZGluYWwiLAogICAgICAgICJzZXF1ZW5jZV9p"
    "bmRleCIsCiAgICAgICAgIndvcmtlcl9pbnN0YW5jZV9pZCIsCiAgICAgICAgIndvcmtlcl9w"
    "cm9jZXNzX2lkZW50aXR5X3NoYTI1NiIsCiAgICAgICAgIndvcmtlcl9zdGFydF9yZWNlaXB0"
    "X3NoYTI1NiIsCiAgICAgICAgInRva2VuX2NvdW50IiwKICAgICAgICAidG9rZW5fc2hhMjU2"
    "IiwKICAgICAgICAicmV1c2FibGVfcHJlZml4X3Rva2VuX2NvdW50IiwKICAgICAgICAicmV1"
    "c2FibGVfcHJlZml4X3Rva2VuX3NoYTI1NiIsCiAgICAgICAgInBheWxvYWRfc2hhMjU2IiwK"
    "ICAgICAgICAiemVyb19jYWNoZV9iYXNlbGluZSIsCiAgICAgICAgInplcm9fY2FjaGVfYmFz"
    "ZWxpbmVfcmVjZWlwdF9zaGEyNTYiLAogICAgICAgICJodHRwX3N0YXR1cyIsCiAgICAgICAg"
    "InJlc3BvbnNlX3NoYTI1NiIsCiAgICAgICAgInJlc3BvbnNlX2xlbmd0aCIsCiAgICAgICAg"
    "ImZpbmlzaF9yZWFzb24iLAogICAgICAgICJwcm9tcHRfdG9rZW5zIiwKICAgICAgICAiY29t"
    "cGxldGlvbl90b2tlbnMiLAogICAgICAgICJ2YWxpZF9qc29uIiwKICAgICAgICAiZHVwbGlj"
    "YXRlX2tleV9kZXRlY3RlZCIsCiAgICAgICAgImpzb25fZXJyb3JfbGluZSIsCiAgICAgICAg"
    "Impzb25fZXJyb3JfY29sdW1uIiwKICAgICAgICAianNvbl9lcnJvcl9wb3NpdGlvbiIsCiAg"
    "ICAgICAgImpzb25fcm9vdF90eXBlIiwKICAgICAgICAicGFyc2VkX2tleV9zZXQiLAogICAg"
    "ICAgICJwcm9iZV9qc29uX3R5cGUiLAogICAgICAgICJ2YWx1ZV9qc29uX3R5cGUiLAogICAg"
    "ICAgICJwcm9iZV9leGFjdCIsCiAgICAgICAgInZhbHVlX2V4YWN0IiwKICAgICAgICAiY2Fu"
    "b25pY2FsX3BhcnNlZF9vYmplY3Rfc2hhMjU2IiwKICAgICAgICAiY2Fub25pY2FsX2V4cGVj"
    "dGVkX29iamVjdF9zaGEyNTYiLAogICAgICAgICJsZWFkaW5nX25vbl93aGl0ZXNwYWNlX2Nv"
    "bnRlbnRfZGV0ZWN0ZWQiLAogICAgICAgICJ0cmFpbGluZ19ub25fd2hpdGVzcGFjZV9jb250"
    "ZW50X2RldGVjdGVkIiwKICAgICAgICAibWFya2Rvd25fZmVuY2VfZGV0ZWN0ZWQiLAogICAg"
    "ICAgICJyZXNwb25zZV9jb21wbGV0ZSIsCiAgICAgICAgImV4YWN0X29iamVjdCIsCiAgICAg"
    "ICAgIndvcmtlcl9oZWFsdGhfYWZ0ZXJfcmVxdWVzdCIsCiAgICAgICAgInRlYXJkb3duX3N0"
    "YXR1cyIsCiAgICAgICAgInJlcXVlc3RfZXJyb3IiLAogICAgICAgICJ0cmFuc3BvcnRfZXJy"
    "b3IiLAogICAgICAgICJtZXRyaWNfZGVsdGEiLAogICAgICAgICJyYXdfcHJvbXB0X3JldGFp"
    "bmVkIiwKICAgICAgICAicmF3X291dHB1dF9yZXRhaW5lZCIsCiAgICApCiAgICByZXR1cm4g"
    "e25hbWU6IG9ic2VydmF0aW9uLmdldChuYW1lKSBmb3IgbmFtZSBpbiBwZXJtaXR0ZWR9CgoK"
    "ZGVmIHJ1bl9jNF9mcmVzaF93b3JrZXJfb2JzZXJ2YXRpb24oCiAgICBtb2RlbF9ob21lOiBQ"
    "YXRoLAogICAgc25hcHNob3Q6IFBhdGgsCiAgICBzZXF1ZW5jZV9pbmRleDogaW50LAogICAg"
    "Y291bnRlcnM6IGRpY3Rbc3RyLCBpbnRdLAogICAgd29ya2VyX3JlcG9ydHM6IGxpc3RbZGlj"
    "dFtzdHIsIG9iamVjdF1dLAogICAgdGVhcmRvd25fcmVwb3J0czogbGlzdFtkaWN0W3N0ciwg"
    "b2JqZWN0XV0sCikgLT4gZGljdFtzdHIsIG9iamVjdF06CiAgICB3b3JrZXIgPSBXb3JrZXIo"
    "CiAgICAgICAgIm9ic2VydmF0aW9uX3dvcmtlciIsCiAgICAgICAgMCwKICAgICAgICA4MDAx"
    "LAogICAgICAgIG1vZGVsX2hvbWUsCiAgICAgICAgc25hcHNob3QsCiAgICAgICAgZ2VuZXJh"
    "dGlvbj1zZXF1ZW5jZV9pbmRleCwKICAgICkKICAgIHByaW1hcnlfZXJyb3I6IEV4Y2VwdGlv"
    "biB8IE5vbmUgPSBOb25lCiAgICBvYnNlcnZhdGlvbjogZGljdFtzdHIsIG9iamVjdF0gfCBO"
    "b25lID0gTm9uZQogICAgcmVwb3J0OiBkaWN0W3N0ciwgb2JqZWN0XSB8IE5vbmUgPSBOb25l"
    "CgogICAgdHJ5OgogICAgICAgIHdvcmtlci5zdGFydChjb3VudGVycykKICAgICAgICB3b3Jr"
    "ZXIud2FpdF9yZWFkeSgpCiAgICAgICAgd29ya2VyLnZhbGlkYXRlX21vZGVsKCkKICAgICAg"
    "ICB3b3JrZXIud2FpdF9iYWNrZW5kX21hcmtlcigpCiAgICAgICAgbmF0aXZlX29yaWdpbnMg"
    "PSB2YWxpZGF0ZV9uYXRpdmVfb3JpZ2luX2Nsb3N1cmUod29ya2VyKQogICAgICAgIHJlcG9y"
    "dCA9IHsKICAgICAgICAgICAgKip3b3JrZXIucmVwb3J0KCksCiAgICAgICAgICAgICJuYXRp"
    "dmVfb3JpZ2luX2Nsb3N1cmUiOiBuYXRpdmVfb3JpZ2lucywKICAgICAgICB9CiAgICAgICAg"
    "d29ya2VyX3JlcG9ydHMuYXBwZW5kKHJlcG9ydCkKICAgICAgICBvYnNlcnZhdGlvbiA9IHJ1"
    "bl9jNF9vYnNlcnZhdGlvbigKICAgICAgICAgICAgd29ya2VyLAogICAgICAgICAgICBzZXF1"
    "ZW5jZV9pbmRleCwKICAgICAgICAgICAgY291bnRlcnMsCiAgICAgICAgKQogICAgZXhjZXB0"
    "IEV4Y2VwdGlvbiBhcyBlcnJvcjoKICAgICAgICBwcmltYXJ5X2Vycm9yID0gZXJyb3IKICAg"
    "IGZpbmFsbHk6CiAgICAgICAgdGVhcmRvd24gPSBzYWZlX3dvcmtlcl90ZWFyZG93bigKICAg"
    "ICAgICAgICAgd29ya2VyLAogICAgICAgICAgICBmIkM0X09CU0VSVkFUSU9OX3tzZXF1ZW5j"
    "ZV9pbmRleH1fVEVSTUlOQUwiLAogICAgICAgICkKICAgICAgICB0ZWFyZG93bl9yZXBvcnRz"
    "LmFwcGVuZCh0ZWFyZG93bikKCiAgICB0ZWFyZG93bl9zdGF0dXMgPSB0ZWFyZG93bi5nZXQo"
    "InN0YXR1cyIpCiAgICBpZiB0ZWFyZG93bl9zdGF0dXMgbm90IGluIHsiUEFTU0VEIiwgIk5P"
    "VF9TVEFSVEVEIn06CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICJURUFSRE9XTl9GQUlMVVJFIiwKICAgICAgICAgICAgImZyZXNoIEM0IG9ic2VydmF0"
    "aW9uIHdvcmtlciB0ZWFyZG93biBwcm9vZiBmYWlsZWQiLAogICAgICAgICkKICAgIGlmIHBy"
    "aW1hcnlfZXJyb3IgaXMgbm90IE5vbmU6CiAgICAgICAgcmFpc2UgcHJpbWFyeV9lcnJvcgog"
    "ICAgaWYgdGVhcmRvd25fc3RhdHVzICE9ICJQQVNTRUQiOgogICAgICAgIHJhaXNlIERpYWdu"
    "b3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAg"
    "ICAgICJjb21wbGV0ZWQgQzQgb2JzZXJ2YXRpb24gZGlkIG5vdCBwcm92ZSB3b3JrZXIgdGVh"
    "cmRvd24iLAogICAgICAgICkKICAgIGlmIG9ic2VydmF0aW9uIGlzIE5vbmUgb3IgcmVwb3J0"
    "IGlzIE5vbmU6CiAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAg"
    "ICJIQVJORVNTX1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAiZnJlc2ggQzQgb2Jz"
    "ZXJ2YXRpb24gZGlkIG5vdCBwcm9kdWNlIGNvbXBsZXRlIGV2aWRlbmNlIiwKICAgICAgICAp"
    "CgogICAgcmV0dXJuIHsKICAgICAgICAqKm9ic2VydmF0aW9uLAogICAgICAgICJ3b3JrZXJf"
    "c3RhcnRfcmVjZWlwdF9zaGEyNTYiOiBzaGEyNTZfdGV4dChjYW5vbmljYWxfanNvbihyZXBv"
    "cnQpKSwKICAgICAgICAidGVhcmRvd25fc3RhdHVzIjogdGVhcmRvd25fc3RhdHVzLAogICAg"
    "fQoKCmRlZiB3cml0ZV9jNF9yZXN1bHRzKAogICAgcmVzdWx0czogbGlzdFtkaWN0W3N0ciwg"
    "b2JqZWN0XV0sCiAgICBzdGF0dXM6IHN0ciwKKSAtPiBOb25lOgogICAgd3JpdGVfanNvbigK"
    "ICAgICAgICBPVVRQVVRfUk9PVCAvICJjNF9yZXF1ZXN0X3Jlc3VsdHNfdjEuanNvbiIsCiAg"
    "ICAgICAgewogICAgICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAg"
    "ICAgICAicXVhbGlmaWNhdGlvbl9pZCI6IEM0X1FVQUxJRklDQVRJT05fSUQsCiAgICAgICAg"
    "ICAgICJzdGF0dXMiOiBzdGF0dXMsCiAgICAgICAgICAgICJzY2hlZHVsZWRfcmVxdWVzdF9j"
    "b3VudCI6IEM0X09CU0VSVkFUSU9OX0NPVU5ULAogICAgICAgICAgICAib2JzZXJ2ZWRfcmVx"
    "dWVzdF9jb3VudCI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgInJlc3VsdHMiOiBbYzRf"
    "cHVibGljX29ic2VydmF0aW9uKGl0ZW0pIGZvciBpdGVtIGluIHJlc3VsdHNdLAogICAgICAg"
    "ICAgICAicmF3X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICAgICAicmF3X291"
    "dHB1dF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgIH0sCiAgICApCgoKZGVmIGRlY2lkZV9j"
    "NF9xdWFsaWZpY2F0aW9uKAogICAgcmVzdWx0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0s"
    "CiAgICB3b3JrZXJfcmVwb3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0sCiAgICB0ZWFy"
    "ZG93bl9yZXBvcnRzOiBsaXN0W2RpY3Rbc3RyLCBvYmplY3RdXSwKICAgIGNvdW50ZXJzOiBk"
    "aWN0W3N0ciwgaW50XSwKKSAtPiBkaWN0W3N0ciwgb2JqZWN0XToKICAgIGlmIGxlbihyZXN1"
    "bHRzKSAhPSBDNF9PQlNFUlZBVElPTl9DT1VOVDoKICAgICAgICByYWlzZSBEaWFnbm9zdGlj"
    "RmFpbHVyZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICJDNCByZXN1bHQgY291bnQgZHJpZnRlZCIsCiAgICAgICAgKQogICAg"
    "ZXhwZWN0ZWRfb3JkaW5hbHMgPSB0dXBsZShyYW5nZSgxLCBDNF9PQlNFUlZBVElPTl9DT1VO"
    "VCArIDEpKQogICAgb2JzZXJ2ZWRfb3JkaW5hbHMgPSB0dXBsZShyb3cuZ2V0KCJyZXF1ZXN0"
    "X29yZGluYWwiKSBmb3Igcm93IGluIHJlc3VsdHMpCiAgICBpZiBvYnNlcnZlZF9vcmRpbmFs"
    "cyAhPSBleHBlY3RlZF9vcmRpbmFsczoKICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVy"
    "ZSgKICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIsCiAgICAg"
    "ICAgICAgICJDNCByZXF1ZXN0IGNocm9ub2xvZ3kgZHJpZnRlZCIsCiAgICAgICAgKQogICAg"
    "aWYgbGVuKHdvcmtlcl9yZXBvcnRzKSAhPSAzIG9yIGxlbih0ZWFyZG93bl9yZXBvcnRzKSAh"
    "PSAzOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAiUkVR"
    "VUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgIkM0IGZyZXNoLXdv"
    "cmtlciBldmlkZW5jZSBjYXJkaW5hbGl0eSBkcmlmdGVkIiwKICAgICAgICApCiAgICBpZiBh"
    "bnkoaXRlbS5nZXQoInN0YXR1cyIpICE9ICJQQVNTRUQiIGZvciBpdGVtIGluIHRlYXJkb3du"
    "X3JlcG9ydHMpOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAg"
    "ICAiVEVBUkRPV05fRkFJTFVSRSIsCiAgICAgICAgICAgICJvbmUgb3IgbW9yZSBDNCBvYnNl"
    "cnZhdGlvbiB0ZWFyZG93bnMgZmFpbGVkIiwKICAgICAgICApCgogICAgd29ya2VyX2lkZW50"
    "aXRpZXMgPSB7CiAgICAgICAgc3RyKHJvdy5nZXQoIndvcmtlcl9wcm9jZXNzX2lkZW50aXR5"
    "X3NoYTI1NiIpKQogICAgICAgIGZvciByb3cgaW4gcmVzdWx0cwogICAgICAgIGlmIGlzaW5z"
    "dGFuY2UoCiAgICAgICAgICAgIHJvdy5nZXQoIndvcmtlcl9wcm9jZXNzX2lkZW50aXR5X3No"
    "YTI1NiIpLAogICAgICAgICAgICBzdHIsCiAgICAgICAgKQogICAgfQogICAgaWYgbGVuKHdv"
    "cmtlcl9pZGVudGl0aWVzKSAhPSAzOgogICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJl"
    "KAogICAgICAgICAgICAiUDVfU1RBUlRJTkdfU1RBVEVfRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJmcmVzaCB3b3JrZXIgcHJvY2VzcyBpZGVudGl0eSB3YXMgcmV1c2VkIiwKICAgICAgICAp"
    "CgogICAgZXhwZWN0ZWRfY291bnRlcnMgPSB7CiAgICAgICAgIm1vZGVsX3JlcXVlc3RzIjog"
    "MywKICAgICAgICAibW9kZWxfbG9hZHMiOiAzLAogICAgICAgICJ3b3JrZXJfc3RhcnRzIjog"
    "MywKICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAgICJuZXR3b3JrX3JlcXVl"
    "c3RzIjogMCwKICAgICAgICAiYmVuY2htYXJrX3RyYWplY3RvcnlfcmVxdWVzdHMiOiAwLAog"
    "ICAgICAgICJleHRlcm5hbF9zcGVuZCI6IDAsCiAgICB9CiAgICBmb3IgbmFtZSwgZXhwZWN0"
    "ZWQgaW4gZXhwZWN0ZWRfY291bnRlcnMuaXRlbXMoKToKICAgICAgICBpZiBjb3VudGVycy5n"
    "ZXQobmFtZSkgIT0gZXhwZWN0ZWQ6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWls"
    "dXJlKAogICAgICAgICAgICAgICAgIlJFUVVFU1RfUkVDT05DSUxJQVRJT05fRkFJTFVSRSIs"
    "CiAgICAgICAgICAgICAgICBmIntuYW1lfSBleHBlY3RlZCB7ZXhwZWN0ZWR9LCBvYnNlcnZl"
    "ZCB7Y291bnRlcnMuZ2V0KG5hbWUpfSIsCiAgICAgICAgICAgICkKCiAgICByZXF1aXJlZF9m"
    "aWVsZHMgPSB7CiAgICAgICAgIm9ic2VydmF0aW9uX2lkIiwKICAgICAgICAicmVxdWVzdF9v"
    "cmRpbmFsIiwKICAgICAgICAid29ya2VyX3N0YXJ0X3JlY2VpcHRfc2hhMjU2IiwKICAgICAg"
    "ICAiemVyb19jYWNoZV9iYXNlbGluZV9yZWNlaXB0X3NoYTI1NiIsCiAgICAgICAgInJlc3Bv"
    "bnNlX3NoYTI1NiIsCiAgICAgICAgInJlc3BvbnNlX2xlbmd0aCIsCiAgICAgICAgImZpbmlz"
    "aF9yZWFzb24iLAogICAgICAgICJwcm9tcHRfdG9rZW5zIiwKICAgICAgICAiY29tcGxldGlv"
    "bl90b2tlbnMiLAogICAgICAgICJ2YWxpZF9qc29uIiwKICAgICAgICAiZHVwbGljYXRlX2tl"
    "eV9kZXRlY3RlZCIsCiAgICAgICAgImpzb25fZXJyb3JfbGluZSIsCiAgICAgICAgImpzb25f"
    "ZXJyb3JfY29sdW1uIiwKICAgICAgICAianNvbl9lcnJvcl9wb3NpdGlvbiIsCiAgICAgICAg"
    "Im1hcmtkb3duX2ZlbmNlX2RldGVjdGVkIiwKICAgICAgICAibGVhZGluZ19ub25fd2hpdGVz"
    "cGFjZV9jb250ZW50X2RldGVjdGVkIiwKICAgICAgICAidHJhaWxpbmdfbm9uX3doaXRlc3Bh"
    "Y2VfY29udGVudF9kZXRlY3RlZCIsCiAgICAgICAgInBhcnNlZF9rZXlfc2V0IiwKICAgICAg"
    "ICAiY2Fub25pY2FsX3BhcnNlZF9vYmplY3Rfc2hhMjU2IiwKICAgICAgICAicmVzcG9uc2Vf"
    "Y29tcGxldGUiLAogICAgICAgICJleGFjdF9vYmplY3QiLAogICAgICAgICJjYW5vbmljYWxf"
    "ZXhwZWN0ZWRfb2JqZWN0X3NoYTI1NiIsCiAgICAgICAgInJlcXVlc3RfZXJyb3IiLAogICAg"
    "ICAgICJ0cmFuc3BvcnRfZXJyb3IiLAogICAgICAgICJ3b3JrZXJfaGVhbHRoX2FmdGVyX3Jl"
    "cXVlc3QiLAogICAgICAgICJ0ZWFyZG93bl9zdGF0dXMiLAogICAgfQoKICAgIGZvciByb3cg"
    "aW4gcmVzdWx0czoKICAgICAgICBpZiByZXF1aXJlZF9maWVsZHMgLSBzZXQocm93KToKICAg"
    "ICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAgICAgICAiRVZJ"
    "REVOQ0VfUFJPSkVDVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgICAgICJDNCBvYnNlcnZh"
    "dGlvbiBldmlkZW5jZSBmaWVsZCBzZXQgaXMgaW5jb21wbGV0ZSIsCiAgICAgICAgICAgICkK"
    "ICAgICAgICBpZiByb3cuZ2V0KCJ6ZXJvX2NhY2hlX2Jhc2VsaW5lIikgaXMgbm90IFRydWU6"
    "CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAg"
    "IlA1X1NUQVJUSU5HX1NUQVRFX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgIkM0IG9ic2Vy"
    "dmF0aW9uIGxhY2tlZCB6ZXJvIGNhY2hlIGJhc2VsaW5lIiwKICAgICAgICAgICAgKQogICAg"
    "ICAgIGlmICgKICAgICAgICAgICAgcm93LmdldCgidG9rZW5fY291bnQiKSAhPSBDNF9GVUxM"
    "X1BST01QVF9UT0tFTl9DT1VOVAogICAgICAgICAgICBvciByb3cuZ2V0KCJ0b2tlbl9zaGEy"
    "NTYiKSAhPSBDNF9GVUxMX1BST01QVF9UT0tFTl9TSEEyNTYKICAgICAgICApOgogICAgICAg"
    "ICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgKICAgICAgICAgICAgICAgICJIQVJORVNT"
    "X1NFTUFOVElDX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgIkM0IGZ1bGwtcHJvbXB0IHRv"
    "a2VuIGlkZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgICAgICkKICAg"
    "ICAgICBpZiAoCiAgICAgICAgICAgIHJvdy5nZXQoInJldXNhYmxlX3ByZWZpeF90b2tlbl9j"
    "b3VudCIpICE9IEM0X1JFVVNBQkxFX1BSRUZJWF9UT0tFTl9DT1VOVAogICAgICAgICAgICBv"
    "ciByb3cuZ2V0KCJyZXVzYWJsZV9wcmVmaXhfdG9rZW5fc2hhMjU2IikgIT0gQzRfUkVVU0FC"
    "TEVfUFJFRklYX1RPS0VOX1NIQTI1NgogICAgICAgICk6CiAgICAgICAgICAgIHJhaXNlIERp"
    "YWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5FU1NfU0VNQU5USUNfRkFJ"
    "TFVSRSIsCiAgICAgICAgICAgICAgICAiQzQgcmV1c2FibGUtcHJlZml4IGlkZW50aXR5IGZh"
    "aWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgICAgICkKICAgICAgICBpZiByb3cuZ2V0"
    "KCJwYXlsb2FkX3NoYTI1NiIpICE9IEM0X1JFUVVFU1RfUEFZTE9BRF9TSEEyNTY6CiAgICAg"
    "ICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIkhBUk5F"
    "U1NfU0VNQU5USUNfRkFJTFVSRSIsCiAgICAgICAgICAgICAgICAiQzQgcmVxdWVzdCBwYXls"
    "b2FkIGlkZW50aXR5IGZhaWxlZCByZWNvbmNpbGlhdGlvbiIsCiAgICAgICAgICAgICkKICAg"
    "ICAgICBpZiByb3cuZ2V0KCJ3b3JrZXJfaGVhbHRoX2FmdGVyX3JlcXVlc3QiKSBpcyBub3Qg"
    "VHJ1ZToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAgICAg"
    "ICAgICAiV09SS0VSX0lERU5USVRZX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgIkM0IHdv"
    "cmtlciB3YXMgbm90IGhlYWx0aHkgYWZ0ZXIgaXRzIHJlcXVlc3QiLAogICAgICAgICAgICAp"
    "CiAgICAgICAgaWYgcm93LmdldCgidGVhcmRvd25fc3RhdHVzIikgIT0gIlBBU1NFRCI6CiAg"
    "ICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlRF"
    "QVJET1dOX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAgIkM0IG9ic2VydmF0aW9uIHRlYXJk"
    "b3duIHN0YXR1cyBkcmlmdGVkIiwKICAgICAgICAgICAgKQogICAgICAgIGlmIHJvdy5nZXQo"
    "InJhd19wcm9tcHRfcmV0YWluZWQiKSBpcyBub3QgRmFsc2U6CiAgICAgICAgICAgIHJhaXNl"
    "IERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlAzX1A2X1BSSVZBQ1lfQk9V"
    "TkRBUllfVklPTEFUSU9OIiwKICAgICAgICAgICAgICAgICJDNCByYXcgcHJvbXB0IHJldGVu"
    "dGlvbiBib3VuZGFyeSBkcmlmdGVkIiwKICAgICAgICAgICAgKQogICAgICAgIGlmIHJvdy5n"
    "ZXQoInJhd19vdXRwdXRfcmV0YWluZWQiKSBpcyBub3QgRmFsc2U6CiAgICAgICAgICAgIHJh"
    "aXNlIERpYWdub3N0aWNGYWlsdXJlKAogICAgICAgICAgICAgICAgIlAzX1A2X1BSSVZBQ1lf"
    "Qk9VTkRBUllfVklPTEFUSU9OIiwKICAgICAgICAgICAgICAgICJDNCByYXcgb3V0cHV0IHJl"
    "dGVudGlvbiBib3VuZGFyeSBkcmlmdGVkIiwKICAgICAgICAgICAgKQoKICAgIGV4YWN0X2Nv"
    "dW50ID0gc3VtKHJvdy5nZXQoImV4YWN0X29iamVjdCIpIGlzIFRydWUgZm9yIHJvdyBpbiBy"
    "ZXN1bHRzKQogICAgc3RhdGUgPSAiUVVBTElGSUVEIiBpZiBleGFjdF9jb3VudCA9PSAzIGVs"
    "c2UgIk5PVF9RVUFMSUZJRUQiCgogICAgcmV0dXJuIHsKICAgICAgICAic2NoZW1hX3ZlcnNp"
    "b24iOiAiMS4wLjAiLAogICAgICAgICJzdGF0dXMiOiAiREVDSURFRCIsCiAgICAgICAgInF1"
    "YWxpZmljYXRpb25faWQiOiBDNF9RVUFMSUZJQ0FUSU9OX0lELAogICAgICAgICJvYnNlcnZl"
    "ZF90ZXJtaW5hbF9zdGF0ZSI6IHN0YXRlLAogICAgICAgICJleGFjdF9vYmplY3RfY291bnQi"
    "OiBleGFjdF9jb3VudCwKICAgICAgICAicmVxdWlyZWRfZXhhY3Rfb2JqZWN0X2NvdW50Ijog"
    "MywKICAgICAgICAib2JzZXJ2YXRpb25fY291bnQiOiAzLAogICAgICAgICJjb21wbGV0ZV9i"
    "ZWhhdmlvcmFsX3J1biI6IFRydWUsCiAgICAgICAgImZyZXNoX3dvcmtlcl9wcm9jZXNzX3Bl"
    "cl9vYnNlcnZhdGlvbiI6IFRydWUsCiAgICAgICAgIndvcmtlcl9pZGVudGl0eV9jYXJkaW5h"
    "bGl0eSI6IGxlbih3b3JrZXJfaWRlbnRpdGllcyksCiAgICAgICAgImZ1bGxfcHJvbXB0X3Rv"
    "a2VuX2NvdW50IjogQzRfRlVMTF9QUk9NUFRfVE9LRU5fQ09VTlQsCiAgICAgICAgImZ1bGxf"
    "cHJvbXB0X3Rva2VuX3NoYTI1NiI6IEM0X0ZVTExfUFJPTVBUX1RPS0VOX1NIQTI1NiwKICAg"
    "ICAgICAicmV1c2FibGVfcHJlZml4X3Rva2VuX2NvdW50IjogKEM0X1JFVVNBQkxFX1BSRUZJ"
    "WF9UT0tFTl9DT1VOVCksCiAgICAgICAgInJldXNhYmxlX3ByZWZpeF90b2tlbl9zaGEyNTYi"
    "OiAoQzRfUkVVU0FCTEVfUFJFRklYX1RPS0VOX1NIQTI1NiksCiAgICAgICAgImNhbm9uaWNh"
    "bF9yZXF1ZXN0X3BheWxvYWRfc2hhMjU2IjogKEM0X1JFUVVFU1RfUEFZTE9BRF9TSEEyNTYp"
    "LAogICAgICAgICJzdHJpY3RfZHVwbGljYXRlX2tleV9yZWplY3Rpb24iOiBUcnVlLAogICAg"
    "ICAgICJzdHJpY3RfaW50ZWdlcl92YWx1ZV92YWxpZGF0aW9uIjogVHJ1ZSwKICAgICAgICAi"
    "ZmluaXNoX3JlYXNvbl9zdG9wX3JlcXVpcmVkIjogVHJ1ZSwKICAgICAgICAicXVhbGlmaWNh"
    "dGlvbl9hY2NlcHRlZF9ieV9yZXBvc2l0b3J5IjogRmFsc2UsCiAgICAgICAgInA1X3JlcXVh"
    "bGlmaWVkIjogRmFsc2UsCiAgICAgICAgInA2X3JlcXVhbGlmaWVkIjogRmFsc2UsCiAgICAg"
    "ICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAicmF3X291dHB1dF9y"
    "ZXRhaW5lZCI6IEZhbHNlLAogICAgfQoKCmRlZiBjNF9idW5kbGVfb3V0cHV0cygpIC0+IGRp"
    "Y3Rbc3RyLCBvYmplY3RdOgogICAgcmVxdWlyZWRfYmVmb3JlX21hbmlmZXN0ID0gc2V0KEM0"
    "X09VVFBVVF9OQU1FUykgLSB7ImJ1bmRsZV9tYW5pZmVzdF92MS5qc29uIn0KICAgIG9ic2Vy"
    "dmVkX2JlZm9yZV9tYW5pZmVzdCA9IHtwYXRoLm5hbWUgZm9yIHBhdGggaW4gT1VUUFVUX1JP"
    "T1QuaXRlcmRpcigpIGlmIHBhdGguaXNfZmlsZSgpfQogICAgbWlzc2luZyA9IHJlcXVpcmVk"
    "X2JlZm9yZV9tYW5pZmVzdCAtIG9ic2VydmVkX2JlZm9yZV9tYW5pZmVzdAogICAgdW5leHBl"
    "Y3RlZCA9IG9ic2VydmVkX2JlZm9yZV9tYW5pZmVzdCAtIHJlcXVpcmVkX2JlZm9yZV9tYW5p"
    "ZmVzdAogICAgaWYgbWlzc2luZyBvciB1bmV4cGVjdGVkOgogICAgICAgIHJhaXNlIFJ1bnRp"
    "bWVFcnJvcigKICAgICAgICAgICAgIkM0IGV2aWRlbmNlIG91dHB1dCBjb250cmFjdCBkcmlm"
    "dGVkOiAiCiAgICAgICAgICAgIGYibWlzc2luZz17c29ydGVkKG1pc3NpbmcpfSAiCiAgICAg"
    "ICAgICAgIGYidW5leHBlY3RlZD17c29ydGVkKHVuZXhwZWN0ZWQpfSIKICAgICAgICApCgog"
    "ICAgZW50cmllczogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0gPSBbXQogICAgZm9yIG5hbWUg"
    "aW4gQzRfT1VUUFVUX05BTUVTOgogICAgICAgIGlmIG5hbWUgPT0gImJ1bmRsZV9tYW5pZmVz"
    "dF92MS5qc29uIjoKICAgICAgICAgICAgY29udGludWUKICAgICAgICBwYXRoID0gT1VUUFVU"
    "X1JPT1QgLyBuYW1lCiAgICAgICAgZW50cmllcy5hcHBlbmQoCiAgICAgICAgICAgIHsKICAg"
    "ICAgICAgICAgICAgICJwYXRoIjogbmFtZSwKICAgICAgICAgICAgICAgICJzaGEyNTYiOiBm"
    "aWxlX3NoYTI1NihwYXRoKSwKICAgICAgICAgICAgICAgICJzaXplX2J5dGVzIjogcGF0aC5z"
    "dGF0KCkuc3Rfc2l6ZSwKICAgICAgICAgICAgfQogICAgICAgICkKCiAgICBtYW5pZmVzdCA9"
    "IHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICJxdWFsaWZp"
    "Y2F0aW9uX2lkIjogQzRfUVVBTElGSUNBVElPTl9JRCwKICAgICAgICAic291cmNlX21haW5f"
    "Y29tbWl0IjogU09VUkNFX01BSU5fQ09NTUlULAogICAgICAgICJtZW1iZXJzIjogZW50cmll"
    "cywKICAgICAgICAic2NyYXRjaF9kaXJlY3Rvcmllc19pbmNsdWRlZCI6IEZhbHNlLAogICAg"
    "ICAgICJ3b3JrZXJfbG9nX2RpcmVjdG9yeV9pbmNsdWRlZCI6IEZhbHNlLAogICAgICAgICJy"
    "YXdfcHJvbXB0X3JldGFpbmVkIjogRmFsc2UsCiAgICAgICAgInJhd19vdXRwdXRfcmV0YWlu"
    "ZWQiOiBGYWxzZSwKICAgIH0KICAgIHdyaXRlX2pzb24oCiAgICAgICAgT1VUUFVUX1JPT1Qg"
    "LyAiYnVuZGxlX21hbmlmZXN0X3YxLmpzb24iLAogICAgICAgIG1hbmlmZXN0LAogICAgKQoK"
    "ICAgIHdpdGggemlwZmlsZS5aaXBGaWxlKAogICAgICAgIEVWSURFTkNFX1pJUCwKICAgICAg"
    "ICAidyIsCiAgICAgICAgY29tcHJlc3Npb249emlwZmlsZS5aSVBfREVGTEFURUQsCiAgICAp"
    "IGFzIGFyY2hpdmU6CiAgICAgICAgZm9yIG5hbWUgaW4gQzRfT1VUUFVUX05BTUVTOgogICAg"
    "ICAgICAgICBwYXRoID0gT1VUUFVUX1JPT1QgLyBuYW1lCiAgICAgICAgICAgIGFyY2hpdmUu"
    "d3JpdGUocGF0aCwgYXJjbmFtZT1uYW1lKQoKICAgIGlmIG5vdCBFVklERU5DRV9aSVAuaXNf"
    "ZmlsZSgpIG9yIEVWSURFTkNFX1pJUC5zdGF0KCkuc3Rfc2l6ZSA+IE1BWF9FVklERU5DRV9a"
    "SVBfQllURVM6CiAgICAgICAgcmFpc2UgUnVudGltZUVycm9yKCJDNCBldmlkZW5jZSBaSVAg"
    "aXMgdW5hdmFpbGFibGUgb3Igb3ZlcnNpemVkIikKCiAgICByZXR1cm4gewogICAgICAgICJi"
    "dW5kbGVfc3RhdHVzIjogIlBBU1NFRCIsCiAgICAgICAgImJ1bmRsZV9tYW5pZmVzdF9zaGEy"
    "NTYiOiBmaWxlX3NoYTI1NihPVVRQVVRfUk9PVCAvICJidW5kbGVfbWFuaWZlc3RfdjEuanNv"
    "biIpLAogICAgICAgICJldmlkZW5jZV96aXBfc2hhMjU2IjogZmlsZV9zaGEyNTYoRVZJREVO"
    "Q0VfWklQKSwKICAgICAgICAiZXZpZGVuY2VfemlwX3NpemVfYnl0ZXMiOiBFVklERU5DRV9a"
    "SVAuc3RhdCgpLnN0X3NpemUsCiAgICB9CgoKZGVmIG1haW4oKSAtPiBpbnQ6CiAgICBpZiBP"
    "VVRQVVRfUk9PVC5leGlzdHMoKSBvciBTQ1JBVENIX1JPT1QuZXhpc3RzKCkgb3IgRVZJREVO"
    "Q0VfWklQLmV4aXN0cygpOgogICAgICAgIHJhaXNlIFJ1bnRpbWVFcnJvcigiQzQgb3V0cHV0"
    "LCBzY3JhdGNoLCBvciBldmlkZW5jZSBwYXRoIGFscmVhZHkgZXhpc3RzIikKCiAgICBPVVRQ"
    "VVRfUk9PVC5ta2RpcihwYXJlbnRzPVRydWUpCiAgICBMT0dfUk9PVC5ta2RpcigpCiAgICBT"
    "Q1JBVENIX1JPT1QubWtkaXIoKQoKICAgIGNvdW50ZXJzID0gewogICAgICAgICJydW50aW1l"
    "X2luc3RhbGxfYXR0ZW1wdHMiOiAwLAogICAgICAgICJydW50aW1lX2ltcG9ydF9jbG9zdXJl"
    "X3Byb2JlcyI6IDAsCiAgICAgICAgIm1vZGVsX2xvYWRzIjogMCwKICAgICAgICAid29ya2Vy"
    "X3N0YXJ0cyI6IDAsCiAgICAgICAgIm1vZGVsX3JlcXVlc3RzIjogMCwKICAgICAgICAiYmVu"
    "Y2htYXJrX3RyYWplY3RvcnlfcmVxdWVzdHMiOiAwLAogICAgICAgICJuZXR3b3JrX3JlcXVl"
    "c3RzIjogMCwKICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAgICJleHRlcm5h"
    "bF9zcGVuZCI6IDAsCiAgICB9CiAgICByZXN1bHRzOiBsaXN0W2RpY3Rbc3RyLCBvYmplY3Rd"
    "XSA9IFtdCiAgICB3b3JrZXJfcmVwb3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0gPSBb"
    "XQogICAgdGVhcmRvd25fcmVwb3J0czogbGlzdFtkaWN0W3N0ciwgb2JqZWN0XV0gPSBbXQog"
    "ICAgZGVjaXNpb246IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZSA9IE5vbmUKICAgIGZhaWx1"
    "cmU6IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZSA9IE5vbmUKICAgIGF1dGhvcml6YXRpb246"
    "IGRpY3Rbc3RyLCBvYmplY3RdIHwgTm9uZSA9IE5vbmUKICAgIGNsZWFudXA6IGRpY3Rbc3Ry"
    "LCBvYmplY3RdID0gewogICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAg"
    "ICAgInN0YXR1cyI6ICJOT1RfUlVOIiwKICAgIH0KICAgIGFjdGl2ZV9mYWlsdXJlX2NvZGUg"
    "PSAiUDNfUDZfUlVOVElNRV9TT1VSQ0VfSURFTlRJVFlfTUlTTUFUQ0giCiAgICBmYWlsZWRf"
    "c3RhZ2UgPSAiUlVOVElNRV9TT1VSQ0VfSURFTlRJVFkiCgogICAgdHJ5OgogICAgICAgIGZh"
    "aWxlZF9zdGFnZSA9ICJQUkVfUkVRVUVTVF9KT1VSTkFMIgogICAgICAgIGFjdGl2ZV9mYWls"
    "dXJlX2NvZGUgPSAiRVZJREVOQ0VfUFJPSkVDVElPTl9GQUlMVVJFIgogICAgICAgIGluaXRp"
    "YWxpemVfb3JkZXJfam91cm5hbCgpCgogICAgICAgIGZhaWxlZF9zdGFnZSA9ICJSVU5USU1F"
    "X1NPVVJDRV9JREVOVElUWSIKICAgICAgICBhY3RpdmVfZmFpbHVyZV9jb2RlID0gIlAzX1A2"
    "X1JVTlRJTUVfU09VUkNFX0lERU5USVRZX01JU01BVENIIgogICAgICAgIHNvdXJjZV9pZGVu"
    "dGl0eSA9IHdyaXRlX3J1bnRpbWVfc291cmNlX2lkZW50aXR5X3JlcG9ydCgpCgogICAgICAg"
    "IGZhaWxlZF9zdGFnZSA9ICJQUklWQUNZX0JPVU5EQVJZIgogICAgICAgIGFjdGl2ZV9mYWls"
    "dXJlX2NvZGUgPSAiUDNfUDZfUFJJVkFDWV9CT1VOREFSWV9WSU9MQVRJT04iCiAgICAgICAg"
    "cmVxdWlyZV9wcml2YXRlX2Vudmlyb25tZW50KCkKCiAgICAgICAgZmFpbGVkX3N0YWdlID0g"
    "IkFVVEhPUklaQVRJT04iCiAgICAgICAgYWN0aXZlX2ZhaWx1cmVfY29kZSA9ICJBVVRIT1JJ"
    "VFlfRkFJTFVSRSIKICAgICAgICBhdXRob3JpemF0aW9uID0gcmVxdWlyZV90cmFuc2FjdGlv"
    "bl9ib3VuZF9jb250ZXh0KCkKCiAgICAgICAgZmFpbGVkX3N0YWdlID0gIldIRUVMSE9VU0Ui"
    "CiAgICAgICAgYWN0aXZlX2ZhaWx1cmVfY29kZSA9ICJQM19QNl9XSEVFTEhPVVNFX0lOVkFM"
    "SUQiCiAgICAgICAgd2hlZWxob3VzZSA9IGRpc2NvdmVyX29uZV9kaXJlY3RvcnkoUlVOVElN"
    "RV9PVVRQVVRfRElSRUNUT1JZKQogICAgICAgIHZhbGlkYXRlX3doZWVsaG91c2Uod2hlZWxo"
    "b3VzZSkKCiAgICAgICAgZmFpbGVkX3N0YWdlID0gIk1PREVMX1NOQVBTSE9UIgogICAgICAg"
    "IGFjdGl2ZV9mYWlsdXJlX2NvZGUgPSAiUDNfUDZfTU9ERUxfSURFTlRJVFlfTUlTTUFUQ0gi"
    "CiAgICAgICAgc291cmNlX3NuYXBzaG90ID0gZGlzY292ZXJfbW9kZWxfc25hcHNob3QoKQoK"
    "ICAgICAgICBmYWlsZWRfc3RhZ2UgPSAiUlVOVElNRV9JTlNUQUxMIgogICAgICAgIGFjdGl2"
    "ZV9mYWlsdXJlX2NvZGUgPSAiUDNfUDZfUlVOVElNRV9JTlNUQUxMX0ZBSUxFRCIKICAgICAg"
    "ICBpbnN0YWxsX3J1bnRpbWUod2hlZWxob3VzZSwgY291bnRlcnMpCgogICAgICAgIGZhaWxl"
    "ZF9zdGFnZSA9ICJSVU5USU1FX0lERU5USVRZIgogICAgICAgIGFjdGl2ZV9mYWlsdXJlX2Nv"
    "ZGUgPSAiUDNfUDZfUExBVEZPUk1fSURFTlRJVFlfTUlTTUFUQ0giCiAgICAgICAgcnVudGlt"
    "ZV9pZGVudGl0eSA9IHZhbGlkYXRlX3RhcmdldF9ydW50aW1lKCkKICAgICAgICBydW50aW1l"
    "X2Vudmlyb25tZW50ID0gcHJvY2Vzc190cmVlX2Vudmlyb25tZW50KAogICAgICAgICAgICAw"
    "LAogICAgICAgICAgICBTQ1JBVENIX1JPT1QgLyAiZW52aXJvbm1lbnRfcmVwb3J0X21vZGVs"
    "X2hvbWUiLAogICAgICAgICkKICAgICAgICBlbnZpcm9ubWVudF9yZXBvcnQgPSBydW50aW1l"
    "X2Vudmlyb25tZW50X3JlcG9ydChydW50aW1lX2Vudmlyb25tZW50KQogICAgICAgIGlmIGVu"
    "dmlyb25tZW50X3JlcG9ydFsicHJvaGliaXRlZF9zdHViX3BhdGhfcHJlc2VudCJdIGlzIG5v"
    "dCBGYWxzZToKICAgICAgICAgICAgcmFpc2UgRGlhZ25vc3RpY0ZhaWx1cmUoCiAgICAgICAg"
    "ICAgICAgICAiTU9ERUxfQ09OU1RSVUNUSU9OX0ZBSUxVUkUiLAogICAgICAgICAgICAgICAg"
    "ImV4YWN0LXJ1bnRpbWUgZW52aXJvbm1lbnQgcmV0YWluZWQgYSBDVURBIHN0dWIgcGF0aCIs"
    "CiAgICAgICAgICAgICkKICAgICAgICBpZiBlbnZpcm9ubWVudF9yZXBvcnRbImxkX3ByZWxv"
    "YWRfYWJzZW50Il0gaXMgbm90IFRydWU6CiAgICAgICAgICAgIHJhaXNlIERpYWdub3N0aWNG"
    "YWlsdXJlKAogICAgICAgICAgICAgICAgIk1PREVMX0NPTlNUUlVDVElPTl9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgICAgICJleGFjdC1ydW50aW1lIGVudmlyb25tZW50IHJldGFpbmVkIExE"
    "X1BSRUxPQUQiLAogICAgICAgICAgICApCiAgICAgICAgd3JpdGVfanNvbigKICAgICAgICAg"
    "ICAgT1VUUFVUX1JPT1QgLyAicnVudGltZV9lbnZpcm9ubWVudF9yZXBvcnRfdjEuanNvbiIs"
    "CiAgICAgICAgICAgIGVudmlyb25tZW50X3JlcG9ydCwKICAgICAgICApCgogICAgICAgIGZh"
    "aWxlZF9zdGFnZSA9ICJJTVBPUlRfQ0xPU1VSRSIKICAgICAgICBhY3RpdmVfZmFpbHVyZV9j"
    "b2RlID0gIlAzX1A2X1JVTlRJTUVfUFJPQ0VTU19UUkVFX0lNUE9SVF9DTE9TVVJFX0ZBSUxF"
    "RCIKICAgICAgICB2YWxpZGF0ZV9wcm9jZXNzX3RyZWVfaW1wb3J0X2Nsb3N1cmUoY291bnRl"
    "cnMpCgogICAgICAgIGZhaWxlZF9zdGFnZSA9ICJNT0RFTF9IT01FIgogICAgICAgIGFjdGl2"
    "ZV9mYWlsdXJlX2NvZGUgPSAiUDNfUDZfTU9ERUxfSURFTlRJVFlfTUlTTUFUQ0giCiAgICAg"
    "ICAgbW9kZWxfaG9tZSwgc25hcHNob3QgPSBwcmVwYXJlX21vZGVsX2hvbWUoc291cmNlX3Nu"
    "YXBzaG90KQoKICAgICAgICB3cml0ZV9qc29uKAogICAgICAgICAgICBPVVRQVVRfUk9PVCAv"
    "ICJjNF9wYXJhZ3JhcGhfb3JkZXJfcnVudGltZV9yZWFkeV92MS5qc29uIiwKICAgICAgICAg"
    "ICAgewogICAgICAgICAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAg"
    "ICAgICAgICAgICJzdGF0dXMiOiAiUEFTU0VEIiwKICAgICAgICAgICAgICAgICJkZWNpc2lv"
    "biI6ICJTVEFUSUNfUlVOVElNRV9QUkVSRVFVSVNJVEVTX1JFQUxJWkVEIiwKICAgICAgICAg"
    "ICAgICAgICJkaWFnbm9zdGljX2lkIjogT1JERVJfRElBR05PU1RJQ19JRCwKICAgICAgICAg"
    "ICAgICAgICJydW50aW1lX2lkZW50aXR5IjogcnVudGltZV9pZGVudGl0eSwKICAgICAgICAg"
    "ICAgICAgICJydW50aW1lX3NvdXJjZV9pZGVudGl0eSI6IHNvdXJjZV9pZGVudGl0eSwKICAg"
    "ICAgICAgICAgICAgICJtb2RlbF9yZXBvc2l0b3J5IjogTU9ERUxfUkVQT1NJVE9SWSwKICAg"
    "ICAgICAgICAgICAgICJtb2RlbF9yZXZpc2lvbiI6IE1PREVMX1JFVklTSU9OLAogICAgICAg"
    "ICAgICAgICAgIm1vZGVsX3NuYXBzaG90X3NoYTI1NiI6IE1PREVMX1NOQVBTSE9UX1NIQTI1"
    "NiwKICAgICAgICAgICAgICAgICJiYWNrZW5kIjogRVhQRUNURURfQkFDS0VORCwKICAgICAg"
    "ICAgICAgICAgICJwcmVmaXhfY2FjaGVfZW5hYmxlZCI6IFRydWUsCiAgICAgICAgICAgICAg"
    "ICAiY2FjaGVfYmxvY2tfc2l6ZSI6IENBQ0hFX0JMT0NLX1NJWkUsCiAgICAgICAgICAgICAg"
    "ICAic2NoZWR1bGVkX3dvcmtlcl9zdGFydHMiOiA2LAogICAgICAgICAgICAgICAgInNjaGVk"
    "dWxlZF9tb2RlbF9sb2FkcyI6IDYsCiAgICAgICAgICAgICAgICAic2NoZWR1bGVkX21vZGVs"
    "X3JlcXVlc3RzIjogNiwKICAgICAgICAgICAgICAgICJmcmVzaF93b3JrZXJfcHJvY2Vzc19w"
    "ZXJfb2JzZXJ2YXRpb24iOiBUcnVlLAogICAgICAgICAgICAgICAgInByb21wdF90b2tlbl9j"
    "b3VudF9wZXJfY29uZGl0aW9uIjogT1JERVJfUFJPTVBUX1RPS0VOX0NPVU5ULAogICAgICAg"
    "ICAgICAgICAgImNvbnRyb2xfcHJvbXB0X3Rva2VuX3NoYTI1NiI6IE9SREVSX0NPTlRST0xf"
    "VE9LRU5fU0hBMjU2LAogICAgICAgICAgICAgICAgInRyZWF0bWVudF9wcm9tcHRfdG9rZW5f"
    "c2hhMjU2IjogT1JERVJfVFJFQVRNRU5UX1RPS0VOX1NIQTI1NiwKICAgICAgICAgICAgICAg"
    "ICJmaW5hbF91c2VyX2JvdW5kYXJ5X3Blcl9jb25kaXRpb24iOiBPUkRFUl9GSU5BTF9VU0VS"
    "X0JPVU5EQVJZLAogICAgICAgICAgICAgICAgImNvbmRpdGlvbl9yZXF1ZXN0X3BheWxvYWRf"
    "c2hhMjU2IjogewogICAgICAgICAgICAgICAgICAgIE9SREVSX0NPTlRST0xfQ09ORElUSU9O"
    "OiBPUkRFUl9DT05UUk9MX1BBWUxPQURfU0hBMjU2LAogICAgICAgICAgICAgICAgICAgIE9S"
    "REVSX1RSRUFUTUVOVF9DT05ESVRJT046IE9SREVSX1RSRUFUTUVOVF9QQVlMT0FEX1NIQTI1"
    "NiwKICAgICAgICAgICAgICAgIH0sCiAgICAgICAgICAgICAgICAicmF3X3Byb21wdF9yZXRh"
    "aW5lZCI6IEZhbHNlLAogICAgICAgICAgICAgICAgInJhd19vdXRwdXRfcmV0YWluZWQiOiBG"
    "YWxzZSwKICAgICAgICAgICAgfSwKICAgICAgICApCgogICAgICAgIGZvciBzZXF1ZW5jZV9p"
    "bmRleCwgY29uZGl0aW9uX2lkIGluIGVudW1lcmF0ZShPUkRFUl9SRVFVRVNUX09SREVSLCBz"
    "dGFydD0xKToKICAgICAgICAgICAgZmFpbGVkX3N0YWdlID0gZiJPQlNFUlZBVElPTl97c2Vx"
    "dWVuY2VfaW5kZXh9IgogICAgICAgICAgICBhY3RpdmVfZmFpbHVyZV9jb2RlID0gIldPUktF"
    "Ul9TVEFSVFVQX0ZBSUxVUkUiCiAgICAgICAgICAgIG9ic2VydmF0aW9uID0gcnVuX29yZGVy"
    "X2ZyZXNoX3dvcmtlcl9vYnNlcnZhdGlvbigKICAgICAgICAgICAgICAgIG1vZGVsX2hvbWUs"
    "CiAgICAgICAgICAgICAgICBzbmFwc2hvdCwKICAgICAgICAgICAgICAgIGNvbmRpdGlvbl9p"
    "ZCwKICAgICAgICAgICAgICAgIHNlcXVlbmNlX2luZGV4LAogICAgICAgICAgICAgICAgY291"
    "bnRlcnMsCiAgICAgICAgICAgICAgICB3b3JrZXJfcmVwb3J0cywKICAgICAgICAgICAgICAg"
    "IHRlYXJkb3duX3JlcG9ydHMsCiAgICAgICAgICAgICkKICAgICAgICAgICAgcmVzdWx0cy5h"
    "cHBlbmQob2JzZXJ2YXRpb24pCiAgICAgICAgICAgIHdyaXRlX29yZGVyX3Jlc3VsdHMocmVz"
    "dWx0cywgIklOX1BST0dSRVNTIikKCiAgICAgICAgZmFpbGVkX3N0YWdlID0gIkFDVElPTl9S"
    "RUNPTkNJTElBVElPTiIKICAgICAgICBhY3RpdmVfZmFpbHVyZV9jb2RlID0gIlJFUVVFU1Rf"
    "UkVDT05DSUxJQVRJT05fRkFJTFVSRSIKICAgICAgICBleHBlY3RlZF9jb3VudGVycyA9IHsK"
    "ICAgICAgICAgICAgInJ1bnRpbWVfaW5zdGFsbF9hdHRlbXB0cyI6IDEsCiAgICAgICAgICAg"
    "ICJydW50aW1lX2ltcG9ydF9jbG9zdXJlX3Byb2JlcyI6IDEsCiAgICAgICAgICAgICJtb2Rl"
    "bF9sb2FkcyI6IDYsCiAgICAgICAgICAgICJ3b3JrZXJfc3RhcnRzIjogNiwKICAgICAgICAg"
    "ICAgIm1vZGVsX3JlcXVlc3RzIjogNiwKICAgICAgICAgICAgImJlbmNobWFya190cmFqZWN0"
    "b3J5X3JlcXVlc3RzIjogMCwKICAgICAgICAgICAgIm5ldHdvcmtfcmVxdWVzdHMiOiAwLAog"
    "ICAgICAgICAgICAiaGlkZGVuX3JldHJpZXMiOiAwLAogICAgICAgICAgICAiZXh0ZXJuYWxf"
    "c3BlbmQiOiAwLAogICAgICAgIH0KICAgICAgICBmb3IgbmFtZSwgZXhwZWN0ZWQgaW4gZXhw"
    "ZWN0ZWRfY291bnRlcnMuaXRlbXMoKToKICAgICAgICAgICAgaWYgY291bnRlcnNbbmFtZV0g"
    "IT0gZXhwZWN0ZWQ6CiAgICAgICAgICAgICAgICByYWlzZSBEaWFnbm9zdGljRmFpbHVyZSgK"
    "ICAgICAgICAgICAgICAgICAgICAiUkVRVUVTVF9SRUNPTkNJTElBVElPTl9GQUlMVVJFIiwK"
    "ICAgICAgICAgICAgICAgICAgICBmIntuYW1lfSBleHBlY3RlZCB7ZXhwZWN0ZWR9LCBvYnNl"
    "cnZlZCB7Y291bnRlcnNbbmFtZV19IiwKICAgICAgICAgICAgICAgICkKCiAgICAgICAgZmFp"
    "bGVkX3N0YWdlID0gIkM0X0RFQ0lTSU9OIgogICAgICAgIGFjdGl2ZV9mYWlsdXJlX2NvZGUg"
    "PSAiSEFSTkVTU19TRU1BTlRJQ19GQUlMVVJFIgogICAgICAgIGRlY2lzaW9uID0gZGVjaWRl"
    "X29yZGVyX2RpZmZlcmVudGlhbCgKICAgICAgICAgICAgcmVzdWx0cywKICAgICAgICAgICAg"
    "d29ya2VyX3JlcG9ydHMsCiAgICAgICAgICAgIHRlYXJkb3duX3JlcG9ydHMsCiAgICAgICAg"
    "ICAgIGNvdW50ZXJzLAogICAgICAgICkKCiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGVycm9y"
    "OgogICAgICAgIGlmIGlzaW5zdGFuY2UoZXJyb3IsIERpYWdub3N0aWNBbWJpZ3VpdHkpOgog"
    "ICAgICAgICAgICBmYWlsdXJlX2NsYXNzID0gZXJyb3IuZmFpbHVyZV9jbGFzcwogICAgICAg"
    "ICAgICBkZXRhaWxfY29kZSA9IE5vbmUKICAgICAgICAgICAgc2FmZV9tZXNzYWdlID0gZXJy"
    "b3Iuc2FmZV9tZXNzYWdlCiAgICAgICAgZWxpZiBpc2luc3RhbmNlKGVycm9yLCBEaWFnbm9z"
    "dGljRmFpbHVyZSk6CiAgICAgICAgICAgIGRldGFpbF9jb2RlID0gZXJyb3IuZXJyb3JfY29k"
    "ZQogICAgICAgICAgICBmYWlsdXJlX2NsYXNzID0gY2xhc3NpZnlfZmFpbHVyZV9kZXRhaWwo"
    "ZGV0YWlsX2NvZGUpCiAgICAgICAgICAgIHNhZmVfbWVzc2FnZSA9IGVycm9yLnNhZmVfbWVz"
    "c2FnZQogICAgICAgIGVsc2U6CiAgICAgICAgICAgIGRldGFpbF9jb2RlID0gYWN0aXZlX2Zh"
    "aWx1cmVfY29kZQogICAgICAgICAgICBmYWlsdXJlX2NsYXNzID0gY2xhc3NpZnlfZmFpbHVy"
    "ZV9kZXRhaWwoZGV0YWlsX2NvZGUpCiAgICAgICAgICAgIHNhZmVfbWVzc2FnZSA9IHNhbml0"
    "aXplX2V4Y2VycHQoc3RyKGVycm9yKSlbOjUxMl0gb3IgdHlwZShlcnJvcikuX19uYW1lX18K"
    "ICAgICAgICBmYWlsdXJlID0gewogICAgICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4w"
    "LjAiLAogICAgICAgICAgICAic3RhdHVzIjogIklOVkFMSURfRVhFQ1VUSU9OIiwKICAgICAg"
    "ICAgICAgImZhaWxlZF9zdGFnZSI6IGZhaWxlZF9zdGFnZSwKICAgICAgICAgICAgImNvbXBs"
    "ZXRlZF9yZXF1ZXN0cyI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgImZhaWx1cmVfY2xh"
    "c3MiOiBmYWlsdXJlX2NsYXNzLAogICAgICAgICAgICAiZGV0YWlsX2NvZGUiOiBkZXRhaWxf"
    "Y29kZSwKICAgICAgICAgICAgImVycm9yX3R5cGUiOiB0eXBlKGVycm9yKS5fX25hbWVfXywK"
    "ICAgICAgICAgICAgInNhZmVfbWVzc2FnZSI6IHNhZmVfbWVzc2FnZSwKICAgICAgICB9Cgog"
    "ICAgdGVhcmRvd25fZmFpbHVyZXMgPSB0dXBsZSgKICAgICAgICBpdGVtIGZvciBpdGVtIGlu"
    "IHRlYXJkb3duX3JlcG9ydHMgaWYgaXRlbS5nZXQoInN0YXR1cyIpIG5vdCBpbiB7IlBBU1NF"
    "RCIsICJOT1RfU1RBUlRFRCJ9CiAgICApCiAgICB0ZWFyZG93bl9zdGF0dXMgPSAiTk9UX1JV"
    "TiIKICAgIGlmIHRlYXJkb3duX3JlcG9ydHM6CiAgICAgICAgdGVhcmRvd25fc3RhdHVzID0g"
    "IlBBU1NFRCIKICAgIGlmIHRlYXJkb3duX2ZhaWx1cmVzOgogICAgICAgIHRlYXJkb3duX3N0"
    "YXR1cyA9ICJGQUlMRUQiCgogICAgd3JpdGVfanNvbigKICAgICAgICBPVVRQVVRfUk9PVCAv"
    "ICJ3b3JrZXJfdGVhcmRvd25fcmVwb3J0X3YxLmpzb24iLAogICAgICAgIHsKICAgICAgICAg"
    "ICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgInN0YXR1cyI6IHRl"
    "YXJkb3duX3N0YXR1cywKICAgICAgICAgICAgInNjaGVkdWxlZF93b3JrZXJfY291bnQiOiA2"
    "LAogICAgICAgICAgICAib2JzZXJ2ZWRfdGVhcmRvd25fY291bnQiOiBsZW4odGVhcmRvd25f"
    "cmVwb3J0cyksCiAgICAgICAgICAgICJ3b3JrZXJfdGVhcmRvd25zIjogdGVhcmRvd25fcmVw"
    "b3J0cywKICAgICAgICAgICAgImZyZXNoX3dvcmtlcl9wcm9jZXNzX3Blcl9vYnNlcnZhdGlv"
    "biI6IFRydWUsCiAgICAgICAgICAgICJhbGxfY29tcGxldGVkX29ic2VydmF0aW9uc190b3Ju"
    "X2Rvd24iOiAoCiAgICAgICAgICAgICAgICBub3QgdGVhcmRvd25fZmFpbHVyZXMgYW5kIGxl"
    "bih0ZWFyZG93bl9yZXBvcnRzKSA+PSBsZW4ocmVzdWx0cykKICAgICAgICAgICAgKSwKICAg"
    "ICAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAgICAgInJh"
    "d19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwKICAgICAgICB9LAogICAgKQoKICAgIGlmIHRl"
    "YXJkb3duX2ZhaWx1cmVzIGFuZCBmYWlsdXJlIGlzIE5vbmU6CiAgICAgICAgZmFpbHVyZSA9"
    "IHsKICAgICAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAg"
    "InN0YXR1cyI6ICJJTlZBTElEX0VYRUNVVElPTiIsCiAgICAgICAgICAgICJmYWlsZWRfc3Rh"
    "Z2UiOiAiV09SS0VSX1RFQVJET1dOIiwKICAgICAgICAgICAgImNvbXBsZXRlZF9yZXF1ZXN0"
    "cyI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgImZhaWx1cmVfY2xhc3MiOiAiVEVBUkRP"
    "V05fRkFJTFVSRSIsCiAgICAgICAgICAgICJkZXRhaWxfY29kZSI6ICJURUFSRE9XTl9GQUlM"
    "VVJFIiwKICAgICAgICAgICAgImVycm9yX3R5cGUiOiAiV29ya2VyVGVhcmRvd25GYWlsdXJl"
    "IiwKICAgICAgICAgICAgInNhZmVfbWVzc2FnZSI6ICJ3b3JrZXIgdGVhcmRvd24gcHJvb2Yg"
    "ZmFpbGVkIiwKICAgICAgICB9CgogICAgdHJ5OgogICAgICAgIGNsZWFudXAgPSBjbGVhbnVw"
    "X3NjcmF0Y2goKQogICAgZXhjZXB0IEV4Y2VwdGlvbiBhcyBlcnJvcjoKICAgICAgICBjbGVh"
    "bnVwID0gewogICAgICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAg"
    "ICAgICAic3RhdHVzIjogIkZBSUxFRCIsCiAgICAgICAgICAgICJzY3JhdGNoX2V4aXN0c19h"
    "ZnRlciI6IFNDUkFUQ0hfUk9PVC5leGlzdHMoKSwKICAgICAgICAgICAgImVycm9yX3R5cGUi"
    "OiB0eXBlKGVycm9yKS5fX25hbWVfXywKICAgICAgICAgICAgInNhZmVfbWVzc2FnZSI6IChz"
    "YW5pdGl6ZV9leGNlcnB0KHN0cihlcnJvcikpWzo1MTJdIG9yIHR5cGUoZXJyb3IpLl9fbmFt"
    "ZV9fKSwKICAgICAgICB9CiAgICAgICAgd3JpdGVfanNvbigKICAgICAgICAgICAgT1VUUFVU"
    "X1JPT1QgLyAic2NyYXRjaF9jbGVhbnVwX3JlcG9ydF92MS5qc29uIiwKICAgICAgICAgICAg"
    "Y2xlYW51cCwKICAgICAgICApCgogICAgaWYgY2xlYW51cC5nZXQoInN0YXR1cyIpICE9ICJQ"
    "QVNTRUQiIGFuZCBmYWlsdXJlIGlzIE5vbmU6CiAgICAgICAgZmFpbHVyZSA9IHsKICAgICAg"
    "ICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAgICAgInN0YXR1cyI6"
    "ICJJTlZBTElEX0VYRUNVVElPTiIsCiAgICAgICAgICAgICJmYWlsZWRfc3RhZ2UiOiAiU0NS"
    "QVRDSF9DTEVBTlVQIiwKICAgICAgICAgICAgImNvbXBsZXRlZF9yZXF1ZXN0cyI6IGxlbihy"
    "ZXN1bHRzKSwKICAgICAgICAgICAgImZhaWx1cmVfY2xhc3MiOiAiVEVBUkRPV05fRkFJTFVS"
    "RSIsCiAgICAgICAgICAgICJkZXRhaWxfY29kZSI6ICJQM19QNl9TQ1JBVENIX0NMRUFOVVBf"
    "RkFJTEVEIiwKICAgICAgICAgICAgImVycm9yX3R5cGUiOiBzdHIoY2xlYW51cC5nZXQoImVy"
    "cm9yX3R5cGUiKSksCiAgICAgICAgICAgICJzYWZlX21lc3NhZ2UiOiBzdHIoY2xlYW51cC5n"
    "ZXQoInNhZmVfbWVzc2FnZSIpKSwKICAgICAgICB9CgogICAgcmVzdWx0X3N0YXR1cyA9ICJO"
    "T1RfUlVOIgogICAgaWYgcmVzdWx0czoKICAgICAgICByZXN1bHRfc3RhdHVzID0gIlBBUlRJ"
    "QUwiCiAgICBpZiBsZW4ocmVzdWx0cykgPT0gNiBhbmQgZmFpbHVyZSBpcyBOb25lOgogICAg"
    "ICAgIHJlc3VsdF9zdGF0dXMgPSAiQ09NUExFVEUiCiAgICB3cml0ZV9vcmRlcl9yZXN1bHRz"
    "KHJlc3VsdHMsIHJlc3VsdF9zdGF0dXMpCgogICAgaWYgZGVjaXNpb24gaXMgbm90IE5vbmUg"
    "YW5kIGZhaWx1cmUgaXMgTm9uZToKICAgICAgICB3cml0ZV9qc29uKAogICAgICAgICAgICBP"
    "VVRQVVRfUk9PVCAvICJjNF9wYXJhZ3JhcGhfb3JkZXJfZGVjaXNpb25fdjEuanNvbiIsCiAg"
    "ICAgICAgICAgIGRlY2lzaW9uLAogICAgICAgICkKICAgIGVsc2U6CiAgICAgICAgd3JpdGVf"
    "anNvbigKICAgICAgICAgICAgT1VUUFVUX1JPT1QgLyAiYzRfcGFyYWdyYXBoX29yZGVyX2Rl"
    "Y2lzaW9uX3YxLmpzb24iLAogICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAic2NoZW1h"
    "X3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAgICAgInN0YXR1cyI6ICJOT1RfRVZB"
    "TFVBVEVEIiwKICAgICAgICAgICAgICAgICJkaWFnbm9zdGljX2lkIjogT1JERVJfRElBR05P"
    "U1RJQ19JRCwKICAgICAgICAgICAgICAgICJvYnNlcnZlZF90ZXJtaW5hbF9zdGF0ZSI6ICJJ"
    "TlZBTElEX0VYRUNVVElPTiIsCiAgICAgICAgICAgICAgICAiYmxvY2tlZF9ieSI6IChOb25l"
    "IGlmIGZhaWx1cmUgaXMgTm9uZSBlbHNlIGZhaWx1cmUuZ2V0KCJmYWlsdXJlX2NsYXNzIikp"
    "LAogICAgICAgICAgICAgICAgInF1YWxpZmljYXRpb25fYWNjZXB0ZWRfYnlfcmVwb3NpdG9y"
    "eSI6IEZhbHNlLAogICAgICAgICAgICAgICAgInJhd19wcm9tcHRfcmV0YWluZWQiOiBGYWxz"
    "ZSwKICAgICAgICAgICAgICAgICJyYXdfb3V0cHV0X3JldGFpbmVkIjogRmFsc2UsCiAgICAg"
    "ICAgICAgIH0sCiAgICAgICAgKQoKICAgIHRlcm1pbmFsX2ZhaWx1cmVfY2xhc3M6IHN0ciB8"
    "IE5vbmUgPSAoCiAgICAgICAgTm9uZSBpZiBmYWlsdXJlIGlzIE5vbmUgZWxzZSBzdHIoZmFp"
    "bHVyZS5nZXQoImZhaWx1cmVfY2xhc3MiKSkKICAgICkKICAgIGVuc3VyZV9ydW50aW1lX3Nv"
    "dXJjZV9pZGVudGl0eV9yZXBvcnQodGVybWluYWxfZmFpbHVyZV9jbGFzcykKICAgIGVuc3Vy"
    "ZV9pbnN0YWxsX3JlcG9ydCh0ZXJtaW5hbF9mYWlsdXJlX2NsYXNzKQogICAgZW5zdXJlX2lt"
    "cG9ydF9jbG9zdXJlX3JlcG9ydCh0ZXJtaW5hbF9mYWlsdXJlX2NsYXNzKQoKICAgIGVudmly"
    "b25tZW50X3BhdGggPSBPVVRQVVRfUk9PVCAvICJydW50aW1lX2Vudmlyb25tZW50X3JlcG9y"
    "dF92MS5qc29uIgogICAgaWYgbm90IGVudmlyb25tZW50X3BhdGguaXNfZmlsZSgpOgogICAg"
    "ICAgIHdyaXRlX2pzb24oCiAgICAgICAgICAgIGVudmlyb25tZW50X3BhdGgsCiAgICAgICAg"
    "ICAgIHsKICAgICAgICAgICAgICAgICJzY2hlbWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAg"
    "ICAgICAgICAgICAic3RhdHVzIjogIk5PVF9SVU4iLAogICAgICAgICAgICAgICAgImJsb2Nr"
    "ZWRfYnkiOiAodGVybWluYWxfZmFpbHVyZV9jbGFzcyBvciAiVVBTVFJFQU1fUFJFQ09ORElU"
    "SU9OIiksCiAgICAgICAgICAgICAgICAicmF3X2Vudmlyb25tZW50X3JldGFpbmVkIjogRmFs"
    "c2UsCiAgICAgICAgICAgIH0sCiAgICAgICAgKQoKICAgIHJlYWR5X3BhdGggPSBPVVRQVVRf"
    "Uk9PVCAvICJjNF9wYXJhZ3JhcGhfb3JkZXJfcnVudGltZV9yZWFkeV92MS5qc29uIgogICAg"
    "aWYgbm90IHJlYWR5X3BhdGguaXNfZmlsZSgpOgogICAgICAgIHdyaXRlX2pzb24oCiAgICAg"
    "ICAgICAgIHJlYWR5X3BhdGgsCiAgICAgICAgICAgIHsKICAgICAgICAgICAgICAgICJzY2hl"
    "bWFfdmVyc2lvbiI6ICIxLjAuMCIsCiAgICAgICAgICAgICAgICAic3RhdHVzIjogIk5PVF9S"
    "VU4iLAogICAgICAgICAgICAgICAgImRpYWdub3N0aWNfaWQiOiBPUkRFUl9ESUFHTk9TVElD"
    "X0lELAogICAgICAgICAgICAgICAgImJsb2NrZWRfYnkiOiAodGVybWluYWxfZmFpbHVyZV9j"
    "bGFzcyBvciAiVVBTVFJFQU1fUFJFQ09ORElUSU9OIiksCiAgICAgICAgICAgICAgICAicmF3"
    "X3Byb21wdF9yZXRhaW5lZCI6IEZhbHNlLAogICAgICAgICAgICAgICAgInJhd19vdXRwdXRf"
    "cmV0YWluZWQiOiBGYWxzZSwKICAgICAgICAgICAgfSwKICAgICAgICApCgogICAgaWYgZmFp"
    "bHVyZSBpcyBOb25lOgogICAgICAgIHdyaXRlX2pzb24oCiAgICAgICAgICAgIE9VVFBVVF9S"
    "T09UIC8gImZhaWx1cmVfcmVwb3J0X3YxLmpzb24iLAogICAgICAgICAgICB7CiAgICAgICAg"
    "ICAgICAgICAic2NoZW1hX3ZlcnNpb24iOiAiMS4wLjAiLAogICAgICAgICAgICAgICAgInN0"
    "YXR1cyI6ICJOT1RfQVBQTElDQUJMRSIsCiAgICAgICAgICAgICAgICAiZmFpbHVyZV9jbGFz"
    "cyI6IE5vbmUsCiAgICAgICAgICAgICAgICAiZGV0YWlsX2NvZGUiOiBOb25lLAogICAgICAg"
    "ICAgICAgICAgImVycm9yX3R5cGUiOiBOb25lLAogICAgICAgICAgICAgICAgInNhZmVfbWVz"
    "c2FnZSI6IE5vbmUsCiAgICAgICAgICAgICAgICAiY29tcGxldGVkX3JlcXVlc3RzIjogbGVu"
    "KHJlc3VsdHMpLAogICAgICAgICAgICAgICAgInRlYXJkb3duX3N0YXR1cyI6IHRlYXJkb3du"
    "X3N0YXR1cywKICAgICAgICAgICAgfSwKICAgICAgICApCiAgICBlbHNlOgogICAgICAgIHdy"
    "aXRlX2pzb24oCiAgICAgICAgICAgIE9VVFBVVF9ST09UIC8gImZhaWx1cmVfcmVwb3J0X3Yx"
    "Lmpzb24iLAogICAgICAgICAgICB7CiAgICAgICAgICAgICAgICAqKmZhaWx1cmUsCiAgICAg"
    "ICAgICAgICAgICAidGVhcmRvd25fc3RhdHVzIjogdGVhcmRvd25fc3RhdHVzLAogICAgICAg"
    "ICAgICB9LAogICAgICAgICkKCiAgICBleGVjdXRpb25fdmFsaWQgPSBkZWNpc2lvbiBpcyBu"
    "b3QgTm9uZSBhbmQgZmFpbHVyZSBpcyBOb25lCiAgICBvYnNlcnZlZF90ZXJtaW5hbF9zdGF0"
    "ZTogb2JqZWN0ID0gIklOVkFMSURfRVhFQ1VUSU9OIgogICAgaWYgZXhlY3V0aW9uX3ZhbGlk"
    "IGFuZCBkZWNpc2lvbiBpcyBub3QgTm9uZToKICAgICAgICBvYnNlcnZlZF90ZXJtaW5hbF9z"
    "dGF0ZSA9IGRlY2lzaW9uLmdldCgib2JzZXJ2ZWRfdGVybWluYWxfc3RhdGUiKQoKICAgIHN1"
    "bW1hcnkgPSB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwKICAgICAgICAi"
    "ZGlhZ25vc3RpY19pZCI6IE9SREVSX0RJQUdOT1NUSUNfSUQsCiAgICAgICAgInN0YXR1cyI6"
    "ICgiUVVBTElGSUNBVElPTl9FWEVDVVRJT05fQ09NUExFVEUiIGlmIGV4ZWN1dGlvbl92YWxp"
    "ZCBlbHNlICJJTlZBTElEX0VYRUNVVElPTiIpLAogICAgICAgICJvYnNlcnZlZF90ZXJtaW5h"
    "bF9zdGF0ZSI6IG9ic2VydmVkX3Rlcm1pbmFsX3N0YXRlLAogICAgICAgICJxdWFsaWZpY2F0"
    "aW9uX2FjY2VwdGVkX2J5X3JlcG9zaXRvcnkiOiBGYWxzZSwKICAgICAgICAiYXV0aG9yaXph"
    "dGlvbiI6IGF1dGhvcml6YXRpb24sCiAgICAgICAgImNvbXBsZXRlZF9yZXF1ZXN0cyI6IGxl"
    "bihyZXN1bHRzKSwKICAgICAgICAic2NoZWR1bGVkX3JlcXVlc3RzIjogNiwKICAgICAgICAi"
    "ZnJlc2hfd29ya2VyX3Byb2Nlc3NfcGVyX29ic2VydmF0aW9uIjogVHJ1ZSwKICAgICAgICAi"
    "cHJvbXB0X3Rva2VuX2NvdW50X3Blcl9jb25kaXRpb24iOiBPUkRFUl9QUk9NUFRfVE9LRU5f"
    "Q09VTlQsCiAgICAgICAgImNvbnRyb2xfcHJvbXB0X3Rva2VuX3NoYTI1NiI6IE9SREVSX0NP"
    "TlRST0xfVE9LRU5fU0hBMjU2LAogICAgICAgICJ0cmVhdG1lbnRfcHJvbXB0X3Rva2VuX3No"
    "YTI1NiI6IE9SREVSX1RSRUFUTUVOVF9UT0tFTl9TSEEyNTYsCiAgICAgICAgImZpbmFsX3Vz"
    "ZXJfYm91bmRhcnlfcGVyX2NvbmRpdGlvbiI6IE9SREVSX0ZJTkFMX1VTRVJfQk9VTkRBUlks"
    "CiAgICAgICAgImNvbmRpdGlvbl9yZXF1ZXN0X3BheWxvYWRfc2hhMjU2IjogewogICAgICAg"
    "ICAgICBPUkRFUl9DT05UUk9MX0NPTkRJVElPTjogT1JERVJfQ09OVFJPTF9QQVlMT0FEX1NI"
    "QTI1NiwKICAgICAgICAgICAgT1JERVJfVFJFQVRNRU5UX0NPTkRJVElPTjogT1JERVJfVFJF"
    "QVRNRU5UX1BBWUxPQURfU0hBMjU2LAogICAgICAgIH0sCiAgICAgICAgIndvcmtlcl9zdGFy"
    "dHMiOiBjb3VudGVyc1sid29ya2VyX3N0YXJ0cyJdLAogICAgICAgICJtb2RlbF9sb2FkcyI6"
    "IGNvdW50ZXJzWyJtb2RlbF9sb2FkcyJdLAogICAgICAgICJtb2RlbF9yZXF1ZXN0cyI6IGNv"
    "dW50ZXJzWyJtb2RlbF9yZXF1ZXN0cyJdLAogICAgICAgICJoaWRkZW5fcmV0cmllcyI6IGNv"
    "dW50ZXJzWyJoaWRkZW5fcmV0cmllcyJdLAogICAgICAgICJleHRlcm5hbF9uZXR3b3JrX3Jl"
    "cXVlc3RzIjogY291bnRlcnNbIm5ldHdvcmtfcmVxdWVzdHMiXSwKICAgICAgICAiZXh0ZXJu"
    "YWxfc3BlbmQiOiBjb3VudGVyc1siZXh0ZXJuYWxfc3BlbmQiXSwKICAgICAgICAidGVhcmRv"
    "d25fc3RhdHVzIjogdGVhcmRvd25fc3RhdHVzLAogICAgICAgICJzY3JhdGNoX2NsZWFudXBf"
    "c3RhdHVzIjogY2xlYW51cC5nZXQoInN0YXR1cyIpLAogICAgICAgICJyYXdfcHJvbXB0X3Jl"
    "dGFpbmVkIjogRmFsc2UsCiAgICAgICAgInJhd19vdXRwdXRfcmV0YWluZWQiOiBGYWxzZSwK"
    "ICAgICAgICAicGFyYWdyYXBoX29yZGVyX3JlcG9zaXRvcnlfc3RhdGVfYWR2YW5jZWQiOiBG"
    "YWxzZSwKICAgICAgICAicDVfcmVxdWFsaWZpZWQiOiBGYWxzZSwKICAgICAgICAicDZfcmVx"
    "dWFsaWZpZWQiOiBGYWxzZSwKICAgICAgICAiZmluYWxfYWJjX21lYXN1cmVkIjogRmFsc2Us"
    "CiAgICAgICAgInByb2R1Y3Rpb25fcmVhZGluZXNzX2VzdGFibGlzaGVkIjogRmFsc2UsCiAg"
    "ICAgICAgIm5leHRfZ2F0ZSI6ICgiUFJFU0VSVkVfQU5EX0RJU1BPU0lUSU9OX0M0X1BBUkFH"
    "UkFQSF9PUkRFUl9CRUhBVklPUkFMX0RJRkZFUkVOVElBTF9WMSIpLAogICAgfQogICAgd3Jp"
    "dGVfanNvbigKICAgICAgICBPVVRQVVRfUk9PVCAvICJjNF9wYXJhZ3JhcGhfb3JkZXJfc3Vt"
    "bWFyeV92MS5qc29uIiwKICAgICAgICBzdW1tYXJ5LAogICAgKQoKICAgIGh1bWFuID0gKAog"
    "ICAgICAgICIjIEF1cmFHYXRld2F5IEM0IFBhcmFncmFwaC1PcmRlciBCZWhhdmlvcmFsICIK"
    "ICAgICAgICAiRGlmZmVyZW50aWFsIFYxXG5cbiIKICAgICAgICBmIi0gRXhlY3V0aW9uIHN0"
    "YXR1czoge3N1bW1hcnlbJ3N0YXR1cyddfVxuIgogICAgICAgIGYiLSBPYnNlcnZlZCB0ZXJt"
    "aW5hbCBzdGF0ZTogIgogICAgICAgIGYie3N1bW1hcnlbJ29ic2VydmVkX3Rlcm1pbmFsX3N0"
    "YXRlJ119XG4iCiAgICAgICAgZiItIENvbXBsZXRlZCByZXF1ZXN0czoge2xlbihyZXN1bHRz"
    "KX0gLyA2XG4iCiAgICAgICAgZiItIFdvcmtlciBzdGFydHM6IHtjb3VudGVyc1snd29ya2Vy"
    "X3N0YXJ0cyddfSAvIDZcbiIKICAgICAgICBmIi0gTW9kZWwgbG9hZHM6IHtjb3VudGVyc1sn"
    "bW9kZWxfbG9hZHMnXX0gLyA2XG4iCiAgICAgICAgZiItIFdvcmtlciB0ZWFyZG93bjoge3Rl"
    "YXJkb3duX3N0YXR1c31cbiIKICAgICAgICBmIi0gU2NyYXRjaCBjbGVhbnVwOiB7Y2xlYW51"
    "cC5nZXQoJ3N0YXR1cycpfVxuIgogICAgICAgICItIERpZmZlcmVudGlhbCBhY2NlcHRlZCBi"
    "eSByZXBvc2l0b3J5OiBmYWxzZVxuIgogICAgICAgICItIFJhdyBwcm9tcHRzIHJldGFpbmVk"
    "OiBmYWxzZVxuIgogICAgICAgICItIFJhdyBtb2RlbCBvdXRwdXRzIHJldGFpbmVkOiBmYWxz"
    "ZVxuIgogICAgICAgICItIEhpZGRlbiByZXRyaWVzOiAwXG4iCiAgICAgICAgIi0gUmVwbGFj"
    "ZW1lbnQgcmVxdWVzdHM6IDBcbiIKICAgICAgICAiLSBQNS9QNiBhcmUgb3V0c2lkZSB0aGlz"
    "IGRpYWdub3N0aWMuXG4iCiAgICAgICAgIi0gRmluYWwgQS9CL0Mgd2FzIG5vdCBtZWFzdXJl"
    "ZC5cbiIKICAgICAgICAiLSBQcm9kdWN0aW9uIHJlYWRpbmVzcyBpcyBub3QgZXN0YWJsaXNo"
    "ZWQuXG4iCiAgICApCiAgICB3cml0ZV90ZXh0KAogICAgICAgIE9VVFBVVF9ST09UIC8gImh1"
    "bWFuX3JlcG9ydF92MS5tZCIsCiAgICAgICAgaHVtYW4sCiAgICApCgogICAgdHJ5OgogICAg"
    "ICAgIGJ1bmRsZSA9IG9yZGVyX2J1bmRsZV9vdXRwdXRzKCkKICAgIGV4Y2VwdCBFeGNlcHRp"
    "b24gYXMgZXJyb3I6CiAgICAgICAgYmVoYXZpb3JhbF9vYnNlcnZlZF90ZXJtaW5hbF9zdGF0"
    "ZSA9IHN1bW1hcnlbIm9ic2VydmVkX3Rlcm1pbmFsX3N0YXRlIl0KICAgICAgICBwYWNrYWdp"
    "bmdfZmFpbHVyZSA9IHsKICAgICAgICAgICAgInNjaGVtYV92ZXJzaW9uIjogIjEuMC4wIiwK"
    "ICAgICAgICAgICAgInN0YXR1cyI6ICJJTlZBTElEX0VYRUNVVElPTiIsCiAgICAgICAgICAg"
    "ICJmYWlsZWRfc3RhZ2UiOiAiRVZJREVOQ0VfUEFDS0FHSU5HIiwKICAgICAgICAgICAgImNv"
    "bXBsZXRlZF9yZXF1ZXN0cyI6IGxlbihyZXN1bHRzKSwKICAgICAgICAgICAgImZhaWx1cmVf"
    "Y2xhc3MiOiAiRVZJREVOQ0VfUFJPSkVDVElPTl9GQUlMVVJFIiwKICAgICAgICAgICAgImRl"
    "dGFpbF9jb2RlIjogIkVWSURFTkNFX1BST0pFQ1RJT05fRkFJTFVSRSIsCiAgICAgICAgICAg"
    "ICJlcnJvcl90eXBlIjogdHlwZShlcnJvcikuX19uYW1lX18sCiAgICAgICAgICAgICJzYWZl"
    "X21lc3NhZ2UiOiAoc2FuaXRpemVfZXhjZXJwdChzdHIoZXJyb3IpKVs6NTEyXSBvciB0eXBl"
    "KGVycm9yKS5fX25hbWVfXyksCiAgICAgICAgICAgICJ0ZWFyZG93bl9zdGF0dXMiOiB0ZWFy"
    "ZG93bl9zdGF0dXMsCiAgICAgICAgICAgICJiZWhhdmlvcmFsX29ic2VydmVkX3Rlcm1pbmFs"
    "X3N0YXRlIjogKGJlaGF2aW9yYWxfb2JzZXJ2ZWRfdGVybWluYWxfc3RhdGUpLAogICAgICAg"
    "IH0KICAgICAgICB3cml0ZV9qc29uKAogICAgICAgICAgICBPVVRQVVRfUk9PVCAvICJmYWls"
    "dXJlX3JlcG9ydF92MS5qc29uIiwKICAgICAgICAgICAgcGFja2FnaW5nX2ZhaWx1cmUsCiAg"
    "ICAgICAgKQogICAgICAgIGludmFsaWRfc3VtbWFyeSA9IHsKICAgICAgICAgICAgKipzdW1t"
    "YXJ5LAogICAgICAgICAgICAic3RhdHVzIjogIklOVkFMSURfRVhFQ1VUSU9OIiwKICAgICAg"
    "ICAgICAgIm9ic2VydmVkX3Rlcm1pbmFsX3N0YXRlIjogIklOVkFMSURfRVhFQ1VUSU9OIiwK"
    "ICAgICAgICAgICAgImJlaGF2aW9yYWxfb2JzZXJ2ZWRfdGVybWluYWxfc3RhdGUiOiAoYmVo"
    "YXZpb3JhbF9vYnNlcnZlZF90ZXJtaW5hbF9zdGF0ZSksCiAgICAgICAgICAgICJldmlkZW5j"
    "ZV9wYWNrYWdpbmdfc3RhdHVzIjogIkZBSUxFRCIsCiAgICAgICAgfQogICAgICAgIHdyaXRl"
    "X2pzb24oCiAgICAgICAgICAgIE9VVFBVVF9ST09UIC8gImM0X3BhcmFncmFwaF9vcmRlcl9z"
    "dW1tYXJ5X3YxLmpzb24iLAogICAgICAgICAgICBpbnZhbGlkX3N1bW1hcnksCiAgICAgICAg"
    "KQogICAgICAgIHRlcm1pbmFsID0gewogICAgICAgICAgICAqKmludmFsaWRfc3VtbWFyeSwK"
    "ICAgICAgICAgICAgImJ1bmRsZV9zdGF0dXMiOiAiRkFJTEVEIiwKICAgICAgICAgICAgImJ1"
    "bmRsZV9lcnJvcl90eXBlIjogdHlwZShlcnJvcikuX19uYW1lX18sCiAgICAgICAgICAgICJi"
    "ZWhhdmlvcmFsX2RlY2lzaW9uX3ByZXNlcnZlZF9iZWZvcmVfYnVuZGxlIjogKGRlY2lzaW9u"
    "IGlzIG5vdCBOb25lKSwKICAgICAgICB9CiAgICAgICAgcHJpbnQoY2Fub25pY2FsX2pzb24o"
    "dGVybWluYWwpKQogICAgICAgIHJldHVybiAyCgogICAgdGVybWluYWwgPSB7CiAgICAgICAg"
    "KipzdW1tYXJ5LAogICAgICAgICoqYnVuZGxlLAogICAgfQogICAgcHJpbnQoY2Fub25pY2Fs"
    "X2pzb24odGVybWluYWwpKQogICAgcmV0dXJuIDAgaWYgZXhlY3V0aW9uX3ZhbGlkIGVsc2Ug"
    "MgoKCmlmIF9fbmFtZV9fID09ICJfX21haW5fXyI6CiAgICByYWlzZSBTeXN0ZW1FeGl0KG1h"
    "aW4oKSkK"
)
_TRANSACTION_ID = "19fc03a6b4ca74a025f8d9b8cd21be7e1cb14a4e776995760c679a581340f122"
_ISSUER_MERGE_COMMIT = "aaa86888df2b8a9c3d01680e1a1e4b8ebba1ecc7"
_ISSUER_SOURCE_SHA256 = "ecaccc674478cf5b8d379b456b462b6431e071f3567cfb5e98c245ca1fd3f030"
_RUNTIME_PAYLOAD_SHA256 = "1d055dfab9f83a2706f5335b4529df98d45e45de5210a8c6c21c2b91e6a72df0"
_GENERATOR_CONTRACT_SHA256 = "7a62445e48a51f32c49aa76d378127a8339a91e22e0c22e4de0150f7571b9f23"
_RUNTIME_CONTRACT_SHA256 = "47ec64a6cd025645442bf2b2333da53cc0f34567f085f8ce9936a5bcd3520b6a"
_BUDGET_CONTRACT_SHA256 = "639c480f91c5262d34325e79ac71ccb7812960b66c9883818aec2775087f6b81"
_EXPERIMENT_CONTRACT_SHA256 = "dd67be30fa97c2852448dabdebb1468b333df55ffe029ad21800f6d099fda623"
_PLATFORM_CONTRACT_SHA256 = "7560409e66f9f75e9528cedf86f23f603f75b81ac3f0152b4928d2b435b5ea66"

_DESIGN_MERGE_COMMIT = "2e0e09b65023397f30d73a406f3f60e7090c85b1"
_DESIGN_RECORD_SHA256 = "8305ebb153f962015c28de98bd6fcf6feeb202482163c6cce3f0caf08cc3d143"
_IMPLEMENTATION_MERGE_COMMIT = "7e037596de1a74038583a85ed81d46ec12debbac"
_IMPLEMENTATION_REVIEW_SHA256 = "355a6b7f7871e648d8bfaf4c7841e9e6346f9b59eba65ac98c00b55d940d2595"
_IMPLEMENTATION_RECORD_SHA256 = "c563bf012c7ec587089b7b28af5074207a389c5fb7381b9c1213299d3b489386"


def _canonical_json_bytes(value: object) -> bytes:
    return json.dumps(
        value,
        ensure_ascii=True,
        separators=(",", ":"),
        sort_keys=True,
    ).encode("utf-8")


def _sha256(payload: bytes) -> str:
    return hashlib.sha256(payload).hexdigest()


def _parse_time(value: str) -> datetime:
    observed = datetime.fromisoformat(value)
    if observed.tzinfo is None or observed.utcoffset() is None:
        raise RuntimeError("authorization timestamp must be timezone-aware")
    return observed.astimezone(UTC)


def _observed_gpu_count() -> int:
    completed = subprocess.run(
        ["nvidia-smi", "-L"],
        check=False,
        capture_output=True,
        text=True,
    )
    if completed.returncode != 0:
        raise RuntimeError("unable to observe GPU topology")
    return sum(1 for line in completed.stdout.splitlines() if line.strip().startswith("GPU "))


def _decode_authorization() -> tuple[dict[str, object], bytes]:
    payload = base64.b64decode(
        _AUTHORIZATION_B64.encode("ascii"),
        validate=True,
    )
    parsed = json.loads(payload)
    if not isinstance(parsed, dict):
        raise RuntimeError("embedded authorization must be one JSON object")
    if payload != _canonical_json_bytes(parsed):
        raise RuntimeError("embedded authorization is not canonical")
    return parsed, payload


def _require_contract_hash(
    authorization: dict[str, object],
    field: str,
    component: str,
    expected_sha256: str,
) -> None:
    if authorization.get(field) != expected_sha256:
        raise RuntimeError(f"embedded {field} drifted")
    value = authorization.get(component)
    if not isinstance(value, dict):
        raise RuntimeError(f"embedded {component} contract is missing")
    if _sha256(_canonical_json_bytes(value)) != expected_sha256:
        raise RuntimeError(f"embedded {component} contract identity drifted")


def admit(now: datetime | None = None) -> dict[str, object]:
    envelope, _ = _decode_authorization()
    transaction_id = envelope.get("transaction_id")
    authorization = envelope.get("authorization")
    if transaction_id != _TRANSACTION_ID or not isinstance(authorization, dict):
        raise RuntimeError("embedded authorization envelope drifted")

    canonical_authorization = _canonical_json_bytes(authorization)
    runtime_payload = base64.b64decode(
        _RUNTIME_PAYLOAD_B64.encode("ascii"),
        validate=True,
    )

    required = {
        "schema_version": "1.0.0",
        "decision": "AUTHORIZED",
        "lifecycle": "ISSUED",
        "scope": "C4_PARAGRAPH_ORDER_BEHAVIORAL_DIFFERENTIAL_V1",
        "issuer_merge_commit": _ISSUER_MERGE_COMMIT,
        "authorization_design_merge_commit": _DESIGN_MERGE_COMMIT,
        "authorization_design_record_sha256": _DESIGN_RECORD_SHA256,
        "implementation_merge_commit": _IMPLEMENTATION_MERGE_COMMIT,
        "implementation_review_sha256": _IMPLEMENTATION_REVIEW_SHA256,
        "implementation_record_sha256": _IMPLEMENTATION_RECORD_SHA256,
        "issuer_source_sha256": _ISSUER_SOURCE_SHA256,
        "runtime_payload_sha256": _RUNTIME_PAYLOAD_SHA256,
        "generator_contract_sha256": _GENERATOR_CONTRACT_SHA256,
        "platform_observation_control_id": ("PERSIST_DURABLE_OBSERVATION_BEFORE_SAVE_AND_RUN_ALL"),
        "durable_platform_observation_required": True,
        "runtime_execution_authorized": True,
        "single_use": True,
        "every_terminal_attempt_consumes_authorization": True,
        "unchanged_replay_authorized": False,
        "authorization_reusable": False,
        "runtime_anti_replay_established": False,
        "threshold_search_authorized": False,
        "runtime_remediation_authorized": False,
        "p5_p6_requalification_authorized": False,
        "north_star_abc_effect_claim_authorized": False,
        "paragraph_order_root_cause_claim_authorized": False,
    }
    drift = tuple(key for key, expected in required.items() if authorization.get(key) != expected)
    if drift:
        raise RuntimeError("embedded authorization contract drift: " + ",".join(drift))

    if _sha256(canonical_authorization) != _TRANSACTION_ID:
        raise RuntimeError("transaction identity mismatch")
    if _sha256(runtime_payload) != _RUNTIME_PAYLOAD_SHA256:
        raise RuntimeError("runtime payload identity mismatch")

    _require_contract_hash(
        authorization,
        "runtime_contract_sha256",
        "runtime",
        _RUNTIME_CONTRACT_SHA256,
    )
    _require_contract_hash(
        authorization,
        "budget_contract_sha256",
        "budget",
        _BUDGET_CONTRACT_SHA256,
    )
    _require_contract_hash(
        authorization,
        "experiment_contract_sha256",
        "experiment",
        _EXPERIMENT_CONTRACT_SHA256,
    )
    _require_contract_hash(
        authorization,
        "platform_contract_sha256",
        "required_platform",
        _PLATFORM_CONTRACT_SHA256,
    )

    observed = (now or datetime.now(UTC)).astimezone(UTC)
    issued_at = _parse_time(str(authorization["issued_at"]))
    expires_at = _parse_time(str(authorization["expires_at"]))
    if observed < issued_at or observed >= expires_at:
        raise RuntimeError("authorization is outside live admission window")
    if _observed_gpu_count() != 2:
        raise RuntimeError("machine-observable GPU topology does not match authorization")

    return {
        "status": "C4_PARAGRAPH_ORDER_TRANSACTION_BOUND_RUNTIME_ADMISSION_VALID",
        "transaction_id": _TRANSACTION_ID,
        "issuer_merge_commit": _ISSUER_MERGE_COMMIT,
        "runtime_payload_sha256": _RUNTIME_PAYLOAD_SHA256,
        "authorization_live_at_admission": True,
        "observed_gpu_count": 2,
        "internet_policy_bound": True,
        "network_probe_performed": False,
        "platform_observation_receipt_runtime_input": False,
    }


def _write_json(path: Path, value: object) -> None:
    path.write_bytes(_canonical_json_bytes(value) + b"\n")


def _record_primary_failure(primary: BaseException) -> None:
    record = {
        "status": "PRIMARY_FAILURE_CAPTURED",
        "transaction_id": _TRANSACTION_ID,
        "exception_type": type(primary).__name__,
        "safe_message": str(primary)[:1000],
    }
    try:
        _write_json(
            Path("c4_paragraph_order_behavioral_differential_primary_failure_v1.json"),
            record,
        )
    except BaseException as reporting_error:
        print(
            "SECONDARY_FAILURE_REPORTING_ERROR=" + type(reporting_error).__name__,
            file=sys.stderr,
        )


def execute_bound_payload() -> None:
    admission = admit()
    _write_json(
        Path("c4_paragraph_order_behavioral_differential_transaction_bound_admission_v1.json"),
        admission,
    )
    runtime_payload = base64.b64decode(
        _RUNTIME_PAYLOAD_B64.encode("ascii"),
        validate=True,
    )
    namespace = {
        "__name__": "__main__",
        "__file__": "<c4-paragraph-order-differential-runtime>",
        "AURAGATEWAY_TRANSACTION_ID": _TRANSACTION_ID,
        "EXECUTED_RUNTIME_SCRIPT_SHA256": _RUNTIME_PAYLOAD_SHA256,
    }
    try:
        exec(
            compile(
                runtime_payload,
                "<c4-paragraph-order-differential-runtime>",
                "exec",
            ),
            namespace,
            namespace,
        )
    except SystemExit as signal:
        if signal.code in (None, 0):
            return
        _record_primary_failure(signal)
        raise
    except BaseException as primary:
        _record_primary_failure(primary)
        raise


if __name__ == "__main__":
    execute_bound_payload()
```
