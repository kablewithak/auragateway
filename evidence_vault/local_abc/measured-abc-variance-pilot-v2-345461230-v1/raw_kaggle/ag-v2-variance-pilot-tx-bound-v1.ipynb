```python
"""Generated AuraGateway variance-pilot successor V2 transaction-bound wrapper."""

from __future__ import annotations

import base64
import hashlib
import json
import subprocess
import sys
import types
from datetime import UTC, datetime
from typing import Any

_R2_RUNTIME_B64 = ""
_R2_RUNTIME_SHA256 = "7f820f1b1195dd2877d4cd197fdc10b79c4e86490e98597aab8bae09cd4a3afc"
_OUTPUT_ADMISSION_RUNTIME_B64 = "IiIiU3RkbGliLW9ubHkgb3V0cHV0IGFkbWlzc2lvbiBhbmQgYXRvbWljIGhpc3RvcnkgbXV0YXRpb24gZm9yIHZhcmlhbmNlLXBpbG90IFYyLiIiIgoKZnJvbSBfX2Z1dHVyZV9fIGltcG9ydCBhbm5vdGF0aW9ucwoKaW1wb3J0IGpzb24KZnJvbSBkYXRhY2xhc3NlcyBpbXBvcnQgZGF0YWNsYXNzCmZyb20gdHlwaW5nIGltcG9ydCBGaW5hbCwgTmV2ZXIsIFR5cGVBbGlhcywgY2FzdAoKSnNvbk9iamVjdDogVHlwZUFsaWFzID0gZGljdFtzdHIsIG9iamVjdF0KSnNvbkhpc3Rvcnk6IFR5cGVBbGlhcyA9IGxpc3RbZGljdFtzdHIsIHN0cl1dCgpTVVBQT1JURURfU1BFQ19WRVJTSU9OOiBGaW5hbCA9ICIxLjAuMCIKCgpjbGFzcyBSdW50aW1lT3V0cHV0QWRtaXNzaW9uRXJyb3IoUnVudGltZUVycm9yKToKICAgICIiIk1ldGFkYXRhLXNhZmUgc3RhbmRhbG9uZSBvdXRwdXQtYWRtaXNzaW9uIGZhaWx1cmUuIiIiCgogICAgZGVmIF9faW5pdF9fKHNlbGYsIGVycm9yX2NvZGU6IHN0ciwgc2FmZV9tZXNzYWdlOiBzdHIpIC0+IE5vbmU6CiAgICAgICAgc3VwZXIoKS5fX2luaXRfXyhzYWZlX21lc3NhZ2UpCiAgICAgICAgc2VsZi5lcnJvcl9jb2RlID0gZXJyb3JfY29kZQogICAgICAgIHNlbGYuc2FmZV9tZXNzYWdlID0gc2FmZV9tZXNzYWdlCgoKQGRhdGFjbGFzcyhmcm96ZW49VHJ1ZSkKY2xhc3MgQWRtaXR0ZWRUZXJtaW5hbE91dHB1dDoKICAgICIiIkNhbm9uaWNhbCBhZG1pdHRlZCB0ZXJtaW5hbC1kZWNpc2lvbiBvdXRwdXQuIiIiCgogICAgcGF5bG9hZDogSnNvbk9iamVjdAogICAgY2Fub25pY2FsX2pzb246IHN0cgoKCmRlZiBjYW5vbmljYWxfanNvbih2YWx1ZTogb2JqZWN0KSAtPiBzdHI6CiAgICAiIiJSZXR1cm4gZGV0ZXJtaW5pc3RpYyBKU09OIGZvciBhZG1pdHRlZCBhc3Npc3RhbnQgc3RhdGUuIiIiCgogICAgcmV0dXJuIGpzb24uZHVtcHMoCiAgICAgICAgdmFsdWUsCiAgICAgICAgZW5zdXJlX2FzY2lpPVRydWUsCiAgICAgICAgc2VwYXJhdG9ycz0oIiwiLCAiOiIpLAogICAgICAgIHNvcnRfa2V5cz1UcnVlLAogICAgKQoKCmRlZiBfZmFpbChjb2RlOiBzdHIsIG1lc3NhZ2U6IHN0cikgLT4gTmV2ZXI6CiAgICByYWlzZSBSdW50aW1lT3V0cHV0QWRtaXNzaW9uRXJyb3IoY29kZSwgbWVzc2FnZSkKCgpkZWYgX2FzX29iamVjdCh2YWx1ZTogb2JqZWN0LCBjb2RlOiBzdHIsIG1lc3NhZ2U6IHN0cikgLT4gSnNvbk9iamVjdDoKICAgIGlmIG5vdCBpc2luc3RhbmNlKHZhbHVlLCBkaWN0KToKICAgICAgICBfZmFpbChjb2RlLCBtZXNzYWdlKQogICAgaWYgYW55KG5vdCBpc2luc3RhbmNlKGtleSwgc3RyKSBmb3Iga2V5IGluIHZhbHVlKToKICAgICAgICBfZmFpbChjb2RlLCBtZXNzYWdlKQogICAgcmV0dXJuIGNhc3QoSnNvbk9iamVjdCwgdmFsdWUpCgoKZGVmIF9hc19zdHJpbmdfbGlzdCh2YWx1ZTogb2JqZWN0LCBmaWVsZF9uYW1lOiBzdHIpIC0+IGxpc3Rbc3RyXToKICAgIGlmIG5vdCBpc2luc3RhbmNlKHZhbHVlLCBsaXN0KSBvciBhbnkobm90IGlzaW5zdGFuY2UoaXRlbSwgc3RyKSBmb3IgaXRlbSBpbiB2YWx1ZSk6CiAgICAgICAgX2ZhaWwoCiAgICAgICAgICAgICJWMl9PVVRQVVRfRklFTERfVFlQRV9JTlZBTElEIiwKICAgICAgICAgICAgZiJmaWVsZCB7ZmllbGRfbmFtZX0gbXVzdCBiZSBhbiBhcnJheSBvZiBzdHJpbmdzIiwKICAgICAgICApCiAgICByZXR1cm4gY2FzdChsaXN0W3N0cl0sIHZhbHVlKQoKCmRlZiBfdmFsaWRhdGVfZmllbGRfc3BlYyhmaWVsZF9zcGVjOiBKc29uT2JqZWN0KSAtPiBOb25lOgogICAgaWYgc2V0KGZpZWxkX3NwZWMpICE9IHsKICAgICAgICAic2NoZW1hX3ZlcnNpb24iLAogICAgICAgICJraW5kIiwKICAgICAgICAiYWxsb3dlZF92YWx1ZXMiLAogICAgICAgICJtaW5pbXVtX2xlbmd0aCIsCiAgICAgICAgIm1pbmltdW1faXRlbXMiLAogICAgICAgICJkZWZhdWx0X2VtcHR5X2FycmF5IiwKICAgIH06CiAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiZmllbGQgc3BlY2lmaWNhdGlvbiBrZXlzIGRyaWZ0ZWQiKQogICAgaWYgZmllbGRfc3BlYy5nZXQoInNjaGVtYV92ZXJzaW9uIikgIT0gU1VQUE9SVEVEX1NQRUNfVkVSU0lPTjoKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJmaWVsZCBzcGVjaWZpY2F0aW9uIHZlcnNpb24gaXMgdW5zdXBwb3J0ZWQiKQogICAga2luZCA9IGZpZWxkX3NwZWMuZ2V0KCJraW5kIikKICAgIGFsbG93ZWRfdmFsdWVzID0gZmllbGRfc3BlYy5nZXQoImFsbG93ZWRfdmFsdWVzIikKICAgIG1pbmltdW1fbGVuZ3RoID0gZmllbGRfc3BlYy5nZXQoIm1pbmltdW1fbGVuZ3RoIikKICAgIG1pbmltdW1faXRlbXMgPSBmaWVsZF9zcGVjLmdldCgibWluaW11bV9pdGVtcyIpCiAgICBkZWZhdWx0X2VtcHR5X2FycmF5ID0gZmllbGRfc3BlYy5nZXQoImRlZmF1bHRfZW1wdHlfYXJyYXkiKQoKICAgIGlmIG5vdCBpc2luc3RhbmNlKGFsbG93ZWRfdmFsdWVzLCBsaXN0KSBvciBhbnkoCiAgICAgICAgbm90IGlzaW5zdGFuY2UoaXRlbSwgc3RyKSBmb3IgaXRlbSBpbiBhbGxvd2VkX3ZhbHVlcwogICAgKToKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJmaWVsZCBhbGxvd2VkLXZhbHVlcyBjb250cmFjdCBpcyBpbnZhbGlkIikKICAgIGlmIGxlbihhbGxvd2VkX3ZhbHVlcykgIT0gbGVuKHNldChjYXN0KGxpc3Rbc3RyXSwgYWxsb3dlZF92YWx1ZXMpKSk6CiAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiZmllbGQgYWxsb3dlZC12YWx1ZXMgY29udHJhY3QgaGFzIGR1cGxpY2F0ZXMiKQogICAgaWYgKAogICAgICAgIG5vdCBpc2luc3RhbmNlKG1pbmltdW1fbGVuZ3RoLCBpbnQpCiAgICAgICAgb3IgaXNpbnN0YW5jZShtaW5pbXVtX2xlbmd0aCwgYm9vbCkKICAgICAgICBvciBtaW5pbXVtX2xlbmd0aCA8IDAKICAgICk6CiAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiZmllbGQgbWluaW11bS1sZW5ndGggY29udHJhY3QgaXMgaW52YWxpZCIpCiAgICBpZiAoCiAgICAgICAgbm90IGlzaW5zdGFuY2UobWluaW11bV9pdGVtcywgaW50KQogICAgICAgIG9yIGlzaW5zdGFuY2UobWluaW11bV9pdGVtcywgYm9vbCkKICAgICAgICBvciBtaW5pbXVtX2l0ZW1zIDwgMAogICAgKToKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJmaWVsZCBtaW5pbXVtLWl0ZW1zIGNvbnRyYWN0IGlzIGludmFsaWQiKQogICAgaWYgbm90IGlzaW5zdGFuY2UoZGVmYXVsdF9lbXB0eV9hcnJheSwgYm9vbCk6CiAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiZmllbGQgZGVmYXVsdCBjb250cmFjdCBpcyBpbnZhbGlkIikKICAgIGlmIGtpbmQgPT0gInN0cmluZyI6CiAgICAgICAgaWYgbWluaW11bV9pdGVtcyAhPSAwIG9yIGRlZmF1bHRfZW1wdHlfYXJyYXk6CiAgICAgICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgInN0cmluZyBmaWVsZCBjb250cmFjdCBpcyBpbmNvbnNpc3RlbnQiKQogICAgICAgIHJldHVybgogICAgaWYga2luZCA9PSAic3RyaW5nX2FycmF5IjoKICAgICAgICBpZiBhbGxvd2VkX3ZhbHVlcyBvciBtaW5pbXVtX2xlbmd0aCAhPSAwOgogICAgICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJzdHJpbmctYXJyYXkgZmllbGQgY29udHJhY3QgaXMgaW5jb25zaXN0ZW50IikKICAgICAgICByZXR1cm4KICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImZpZWxkIGtpbmQgaXMgdW5zdXBwb3J0ZWQiKQoKCmRlZiBfdmFsaWRhdGVfZmllbGQoZmllbGRfbmFtZTogc3RyLCB2YWx1ZTogb2JqZWN0LCBmaWVsZF9zcGVjOiBKc29uT2JqZWN0KSAtPiBOb25lOgogICAgX3ZhbGlkYXRlX2ZpZWxkX3NwZWMoZmllbGRfc3BlYykKICAgIGtpbmQgPSBmaWVsZF9zcGVjWyJraW5kIl0KICAgIGFsbG93ZWRfdmFsdWVzID0gY2FzdChsaXN0W3N0cl0sIGZpZWxkX3NwZWNbImFsbG93ZWRfdmFsdWVzIl0pCiAgICBtaW5pbXVtX2xlbmd0aCA9IGNhc3QoaW50LCBmaWVsZF9zcGVjWyJtaW5pbXVtX2xlbmd0aCJdKQogICAgbWluaW11bV9pdGVtcyA9IGNhc3QoaW50LCBmaWVsZF9zcGVjWyJtaW5pbXVtX2l0ZW1zIl0pCgogICAgaWYga2luZCA9PSAic3RyaW5nIjoKICAgICAgICBpZiBub3QgaXNpbnN0YW5jZSh2YWx1ZSwgc3RyKToKICAgICAgICAgICAgX2ZhaWwoIlYyX09VVFBVVF9GSUVMRF9UWVBFX0lOVkFMSUQiLCBmImZpZWxkIHtmaWVsZF9uYW1lfSBtdXN0IGJlIGEgc3RyaW5nIikKICAgICAgICBpZiBsZW4odmFsdWUpIDwgbWluaW11bV9sZW5ndGg6CiAgICAgICAgICAgIF9mYWlsKAogICAgICAgICAgICAgICAgIlYyX09VVFBVVF9GSUVMRF9MRU5HVEhfSU5WQUxJRCIsCiAgICAgICAgICAgICAgICBmImZpZWxkIHtmaWVsZF9uYW1lfSBpcyBzaG9ydGVyIHRoYW4gdGhlIGZyb3plbiBtaW5pbXVtIiwKICAgICAgICAgICAgKQogICAgICAgIGlmIGFsbG93ZWRfdmFsdWVzIGFuZCB2YWx1ZSBub3QgaW4gYWxsb3dlZF92YWx1ZXM6CiAgICAgICAgICAgIF9mYWlsKAogICAgICAgICAgICAgICAgIlYyX09VVFBVVF9GSUVMRF9WQUxVRV9JTlZBTElEIiwKICAgICAgICAgICAgICAgIGYiZmllbGQge2ZpZWxkX25hbWV9IHZpb2xhdGVzIHRoZSBmcm96ZW4gdmFsdWUgc2V0IiwKICAgICAgICAgICAgKQogICAgICAgIHJldHVybgoKICAgIGlmIGtpbmQgPT0gInN0cmluZ19hcnJheSI6CiAgICAgICAgaXRlbXMgPSBfYXNfc3RyaW5nX2xpc3QodmFsdWUsIGZpZWxkX25hbWUpCiAgICAgICAgaWYgbGVuKGl0ZW1zKSA8IG1pbmltdW1faXRlbXM6CiAgICAgICAgICAgIF9mYWlsKAogICAgICAgICAgICAgICAgIlYyX09VVFBVVF9GSUVMRF9JVEVNU19JTlZBTElEIiwKICAgICAgICAgICAgICAgIGYiZmllbGQge2ZpZWxkX25hbWV9IGhhcyB0b28gZmV3IGl0ZW1zIiwKICAgICAgICAgICAgKQogICAgICAgIHJldHVybgoKICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImZpZWxkIGtpbmQgaXMgdW5zdXBwb3J0ZWQiKQoKCmRlZiB2YWxpZGF0ZV9hZG1pc3Npb25fc3BlYyhzcGVjOiBvYmplY3QpIC0+IEpzb25PYmplY3Q6CiAgICAiIiJWYWxpZGF0ZSB0aGUgY29tcGFjdCBnZW5lcmF0ZWQgYWRtaXNzaW9uIGNvbnRyYWN0IGJlZm9yZSB1c2luZyBpdC4iIiIKCiAgICByb290ID0gX2FzX29iamVjdChzcGVjLCAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gc3BlYyByb290IG11c3QgYmUgYW4gb2JqZWN0IikKICAgIGlmIHNldChyb290KSAhPSB7CiAgICAgICAgInNjaGVtYV92ZXJzaW9uIiwKICAgICAgICAic2VtYW50aWNfY29udHJhY3QiLAogICAgICAgICJkaXNjcmltaW5hdG9yX2ZpZWxkIiwKICAgICAgICAidmFyaWFudHMiLAogICAgfToKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gc3BlYyByb290IGtleXMgZHJpZnRlZCIpCiAgICBpZiByb290LmdldCgic2NoZW1hX3ZlcnNpb24iKSAhPSBTVVBQT1JURURfU1BFQ19WRVJTSU9OOgogICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImFkbWlzc2lvbiBzcGVjIHZlcnNpb24gaXMgdW5zdXBwb3J0ZWQiKQogICAgaWYgcm9vdC5nZXQoInNlbWFudGljX2NvbnRyYWN0IikgIT0gIlRlcm1pbmFsRGVjaXNpb25PdXRwdXQiOgogICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImFkbWlzc2lvbiBzZW1hbnRpYyBjb250cmFjdCBpcyBpbnZhbGlkIikKICAgIGlmIHJvb3QuZ2V0KCJkaXNjcmltaW5hdG9yX2ZpZWxkIikgIT0gImRlY2lzaW9uIjoKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gZGlzY3JpbWluYXRvciBtdXN0IGJlIGRlY2lzaW9uIikKICAgIHZhcmlhbnRzID0gcm9vdC5nZXQoInZhcmlhbnRzIikKICAgIGlmIG5vdCBpc2luc3RhbmNlKHZhcmlhbnRzLCBsaXN0KSBvciBsZW4odmFyaWFudHMpICE9IDQ6CiAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiYWRtaXNzaW9uIHNwZWMgbXVzdCBjb250YWluIGZvdXIgdmFyaWFudHMiKQoKICAgIGRlY2lzaW9uczogc2V0W3N0cl0gPSBzZXQoKQogICAgZGVjaXNpb25fb3JkZXI6IGxpc3Rbc3RyXSA9IFtdCiAgICBmb3IgcmF3X3ZhcmlhbnQgaW4gdmFyaWFudHM6CiAgICAgICAgdmFyaWFudCA9IF9hc19vYmplY3QoCiAgICAgICAgICAgIHJhd192YXJpYW50LAogICAgICAgICAgICAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsCiAgICAgICAgICAgICJhZG1pc3Npb24gdmFyaWFudCBtdXN0IGJlIGFuIG9iamVjdCIsCiAgICAgICAgKQogICAgICAgIGlmIHNldCh2YXJpYW50KSAhPSB7InNjaGVtYV92ZXJzaW9uIiwgImRlY2lzaW9uIiwgImZpZWxkcyIsICJyZXF1aXJlZF9maWVsZHMifToKICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiYWRtaXNzaW9uIHZhcmlhbnQga2V5cyBkcmlmdGVkIikKICAgICAgICBpZiB2YXJpYW50LmdldCgic2NoZW1hX3ZlcnNpb24iKSAhPSBTVVBQT1JURURfU1BFQ19WRVJTSU9OOgogICAgICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gdmFyaWFudCB2ZXJzaW9uIGlzIHVuc3VwcG9ydGVkIikKICAgICAgICBkZWNpc2lvbiA9IHZhcmlhbnQuZ2V0KCJkZWNpc2lvbiIpCiAgICAgICAgZmllbGRzID0gdmFyaWFudC5nZXQoImZpZWxkcyIpCiAgICAgICAgcmVxdWlyZWRfZmllbGRzID0gdmFyaWFudC5nZXQoInJlcXVpcmVkX2ZpZWxkcyIpCiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2UoZGVjaXNpb24sIHN0cikgb3Igbm90IGRlY2lzaW9uOgogICAgICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gdmFyaWFudCBkZWNpc2lvbiBpcyBpbnZhbGlkIikKICAgICAgICBpZiBkZWNpc2lvbiBpbiBkZWNpc2lvbnM6CiAgICAgICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImFkbWlzc2lvbiBkZWNpc2lvbnMgbXVzdCBiZSB1bmlxdWUiKQogICAgICAgIGRlY2lzaW9ucy5hZGQoZGVjaXNpb24pCiAgICAgICAgZGVjaXNpb25fb3JkZXIuYXBwZW5kKGRlY2lzaW9uKQogICAgICAgIGlmIG5vdCBpc2luc3RhbmNlKGZpZWxkcywgZGljdCkgb3Igbm90IGZpZWxkczoKICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiYWRtaXNzaW9uIHZhcmlhbnQgZmllbGRzIGFyZSBpbnZhbGlkIikKICAgICAgICBpZiBhbnkobm90IGlzaW5zdGFuY2Uoa2V5LCBzdHIpIGZvciBrZXkgaW4gZmllbGRzKToKICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAiYWRtaXNzaW9uIGZpZWxkIG5hbWVzIG11c3QgYmUgc3RyaW5ncyIpCiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2UocmVxdWlyZWRfZmllbGRzLCBsaXN0KSBvciBhbnkoCiAgICAgICAgICAgIG5vdCBpc2luc3RhbmNlKGl0ZW0sIHN0cikgZm9yIGl0ZW0gaW4gcmVxdWlyZWRfZmllbGRzCiAgICAgICAgKToKICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAicmVxdWlyZWQgZmllbGRzIGNvbnRyYWN0IGlzIGludmFsaWQiKQogICAgICAgIHR5cGVkX3JlcXVpcmVkID0gY2FzdChsaXN0W3N0cl0sIHJlcXVpcmVkX2ZpZWxkcykKICAgICAgICB0eXBlZF9maWVsZHMgPSBjYXN0KGRpY3Rbc3RyLCBvYmplY3RdLCBmaWVsZHMpCiAgICAgICAgaWYgbGVuKHR5cGVkX3JlcXVpcmVkKSAhPSBsZW4oc2V0KHR5cGVkX3JlcXVpcmVkKSk6CiAgICAgICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgInJlcXVpcmVkIGZpZWxkcyBjb250YWluIGR1cGxpY2F0ZXMiKQogICAgICAgIGlmIG5vdCBzZXQodHlwZWRfcmVxdWlyZWQpLmlzc3Vic2V0KHNldCh0eXBlZF9maWVsZHMpKToKICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAicmVxdWlyZWQgZmllbGRzIGFyZSBub3QgZGVmaW5lZCIpCiAgICAgICAgZGVjaXNpb25fZmllbGQgPSBfYXNfb2JqZWN0KAogICAgICAgICAgICB0eXBlZF9maWVsZHMuZ2V0KCJkZWNpc2lvbiIpLAogICAgICAgICAgICAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsCiAgICAgICAgICAgICJkZWNpc2lvbiBmaWVsZCBzcGVjaWZpY2F0aW9uIG11c3QgYmUgYW4gb2JqZWN0IiwKICAgICAgICApCiAgICAgICAgaWYgZGVjaXNpb25fZmllbGQuZ2V0KCJhbGxvd2VkX3ZhbHVlcyIpICE9IFtkZWNpc2lvbl06CiAgICAgICAgICAgIF9mYWlsKCJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwgImRlY2lzaW9uIGZpZWxkIGRvZXMgbm90IG1hdGNoIGl0cyB2YXJpYW50IikKICAgICAgICBmb3IgZmllbGRfbmFtZSwgcmF3X2ZpZWxkX3NwZWMgaW4gdHlwZWRfZmllbGRzLml0ZW1zKCk6CiAgICAgICAgICAgIGZpZWxkX3NwZWMgPSBfYXNfb2JqZWN0KAogICAgICAgICAgICAgICAgcmF3X2ZpZWxkX3NwZWMsCiAgICAgICAgICAgICAgICAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsCiAgICAgICAgICAgICAgICAiZmllbGQgc3BlY2lmaWNhdGlvbiBtdXN0IGJlIGFuIG9iamVjdCIsCiAgICAgICAgICAgICkKICAgICAgICAgICAgX3ZhbGlkYXRlX2ZpZWxkX3NwZWMoZmllbGRfc3BlYykKICAgICAgICAgICAgaWYgKAogICAgICAgICAgICAgICAgZmllbGRfc3BlYy5nZXQoImRlZmF1bHRfZW1wdHlfYXJyYXkiKSBpcyBUcnVlCiAgICAgICAgICAgICAgICBhbmQgZmllbGRfbmFtZSBpbiB0eXBlZF9yZXF1aXJlZAogICAgICAgICAgICApOgogICAgICAgICAgICAgICAgX2ZhaWwoIlYyX0FETUlTU0lPTl9TUEVDX0lOVkFMSUQiLCAicmVxdWlyZWQgZmllbGRzIGNhbm5vdCB1c2UgcnVudGltZSBkZWZhdWx0cyIpCiAgICBpZiBkZWNpc2lvbnMgIT0geyJhbnN3ZXIiLCAiY2xhcmlmeSIsICJlc2NhbGF0ZSIsICJyZWZ1c2UifToKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gZGVjaXNpb24gc2V0IGRyaWZ0ZWQiKQogICAgaWYgZGVjaXNpb25fb3JkZXIgIT0gWyJhbnN3ZXIiLCAiY2xhcmlmeSIsICJlc2NhbGF0ZSIsICJyZWZ1c2UiXToKICAgICAgICBfZmFpbCgiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsICJhZG1pc3Npb24gZGVjaXNpb24gb3JkZXIgZHJpZnRlZCIpCiAgICByZXR1cm4gcm9vdAoKCmRlZiBhZG1pdF90ZXJtaW5hbF9vdXRwdXQoY29udGVudDogc3RyLCBzcGVjOiBvYmplY3QpIC0+IEFkbWl0dGVkVGVybWluYWxPdXRwdXQ6CiAgICAiIiJQYXJzZSBhbmQgdmFsaWRhdGUgb25lIG1vZGVsIG91dHB1dCBhZ2FpbnN0IHRoZSBnZW5lcmF0ZWQgc3RhbmRhbG9uZSBjb250cmFjdC4iIiIKCiAgICByb290ID0gdmFsaWRhdGVfYWRtaXNzaW9uX3NwZWMoc3BlYykKICAgIHRyeToKICAgICAgICBwYXJzZWQgPSBqc29uLmxvYWRzKGNvbnRlbnQpCiAgICBleGNlcHQganNvbi5KU09ORGVjb2RlRXJyb3IgYXMgZXhjOgogICAgICAgIHJhaXNlIFJ1bnRpbWVPdXRwdXRBZG1pc3Npb25FcnJvcigKICAgICAgICAgICAgIlYyX09VVFBVVF9KU09OX0lOVkFMSUQiLAogICAgICAgICAgICAibW9kZWwgb3V0cHV0IGlzIG5vdCB2YWxpZCBKU09OIiwKICAgICAgICApIGZyb20gZXhjCiAgICBwYXlsb2FkID0gX2FzX29iamVjdCgKICAgICAgICBwYXJzZWQsCiAgICAgICAgIlYyX09VVFBVVF9ST09UX0lOVkFMSUQiLAogICAgICAgICJtb2RlbCBvdXRwdXQgcm9vdCBtdXN0IGJlIGEgSlNPTiBvYmplY3QiLAogICAgKQogICAgZGVjaXNpb24gPSBwYXlsb2FkLmdldCgiZGVjaXNpb24iKQogICAgaWYgbm90IGlzaW5zdGFuY2UoZGVjaXNpb24sIHN0cik6CiAgICAgICAgX2ZhaWwoIlYyX09VVFBVVF9ESVNDUklNSU5BVE9SX0lOVkFMSUQiLCAiZGVjaXNpb24gZGlzY3JpbWluYXRvciBpcyBtaXNzaW5nIG9yIGludmFsaWQiKQoKICAgIHZhcmlhbnRzID0gY2FzdChsaXN0W29iamVjdF0sIHJvb3RbInZhcmlhbnRzIl0pCiAgICBzZWxlY3RlZDogSnNvbk9iamVjdCB8IE5vbmUgPSBOb25lCiAgICBmb3IgcmF3X3ZhcmlhbnQgaW4gdmFyaWFudHM6CiAgICAgICAgdmFyaWFudCA9IF9hc19vYmplY3QoCiAgICAgICAgICAgIHJhd192YXJpYW50LAogICAgICAgICAgICAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsCiAgICAgICAgICAgICJhZG1pc3Npb24gdmFyaWFudCBtdXN0IGJlIGFuIG9iamVjdCIsCiAgICAgICAgKQogICAgICAgIGlmIHZhcmlhbnQuZ2V0KCJkZWNpc2lvbiIpID09IGRlY2lzaW9uOgogICAgICAgICAgICBzZWxlY3RlZCA9IHZhcmlhbnQKICAgICAgICAgICAgYnJlYWsKICAgIGlmIHNlbGVjdGVkIGlzIE5vbmU6CiAgICAgICAgX2ZhaWwoIlYyX09VVFBVVF9ESVNDUklNSU5BVE9SX0lOVkFMSUQiLCAiZGVjaXNpb24gZGlzY3JpbWluYXRvciBpcyB1bnN1cHBvcnRlZCIpCgogICAgZmllbGRzID0gY2FzdChkaWN0W3N0ciwgb2JqZWN0XSwgc2VsZWN0ZWRbImZpZWxkcyJdKQogICAgcmVxdWlyZWQgPSBjYXN0KGxpc3Rbc3RyXSwgc2VsZWN0ZWRbInJlcXVpcmVkX2ZpZWxkcyJdKQogICAgdW5rbm93biA9IHNvcnRlZChzZXQocGF5bG9hZCkgLSBzZXQoZmllbGRzKSkKICAgIGlmIHVua25vd246CiAgICAgICAgX2ZhaWwoIlYyX09VVFBVVF9FWFRSQV9GSUVMRFMiLCAibW9kZWwgb3V0cHV0IGNvbnRhaW5zIGZpZWxkcyBvdXRzaWRlIHRoZSBmcm96ZW4gdmFyaWFudCIpCiAgICBtaXNzaW5nID0gc29ydGVkKHNldChyZXF1aXJlZCkgLSBzZXQocGF5bG9hZCkpCiAgICBpZiBtaXNzaW5nOgogICAgICAgIF9mYWlsKCJWMl9PVVRQVVRfUkVRVUlSRURfRklFTERfTUlTU0lORyIsICJtb2RlbCBvdXRwdXQgaXMgbWlzc2luZyByZXF1aXJlZCBmaWVsZHMiKQoKICAgIG5vcm1hbGl6ZWQgPSBkaWN0KHBheWxvYWQpCiAgICBmb3IgZmllbGRfbmFtZSwgcmF3X2ZpZWxkX3NwZWMgaW4gZmllbGRzLml0ZW1zKCk6CiAgICAgICAgZmllbGRfc3BlYyA9IF9hc19vYmplY3QoCiAgICAgICAgICAgIHJhd19maWVsZF9zcGVjLAogICAgICAgICAgICAiVjJfQURNSVNTSU9OX1NQRUNfSU5WQUxJRCIsCiAgICAgICAgICAgICJmaWVsZCBzcGVjaWZpY2F0aW9uIG11c3QgYmUgYW4gb2JqZWN0IiwKICAgICAgICApCiAgICAgICAgaWYgZmllbGRfbmFtZSBub3QgaW4gbm9ybWFsaXplZCBhbmQgZmllbGRfc3BlYy5nZXQoImRlZmF1bHRfZW1wdHlfYXJyYXkiKSBpcyBUcnVlOgogICAgICAgICAgICBub3JtYWxpemVkW2ZpZWxkX25hbWVdID0gW10KCiAgICBmb3IgZmllbGRfbmFtZSwgdmFsdWUgaW4gbm9ybWFsaXplZC5pdGVtcygpOgogICAgICAgIGZpZWxkX3NwZWMgPSBfYXNfb2JqZWN0KAogICAgICAgICAgICBmaWVsZHNbZmllbGRfbmFtZV0sCiAgICAgICAgICAgICJWMl9BRE1JU1NJT05fU1BFQ19JTlZBTElEIiwKICAgICAgICAgICAgImZpZWxkIHNwZWNpZmljYXRpb24gbXVzdCBiZSBhbiBvYmplY3QiLAogICAgICAgICkKICAgICAgICBfdmFsaWRhdGVfZmllbGQoZmllbGRfbmFtZSwgdmFsdWUsIGZpZWxkX3NwZWMpCgogICAgY2Fub25pY2FsID0gY2Fub25pY2FsX2pzb24obm9ybWFsaXplZCkKICAgIGNhbm9uaWNhbF9wYXlsb2FkID0gY2FzdChKc29uT2JqZWN0LCBqc29uLmxvYWRzKGNhbm9uaWNhbCkpCiAgICByZXR1cm4gQWRtaXR0ZWRUZXJtaW5hbE91dHB1dChwYXlsb2FkPWNhbm9uaWNhbF9wYXlsb2FkLCBjYW5vbmljYWxfanNvbj1jYW5vbmljYWwpCgoKZGVmIF9leHRyYWN0X3Jlc3BvbnNlX2NvbnRlbnQocmVzcG9uc2U6IG9iamVjdCkgLT4gc3RyOgogICAgZW52ZWxvcGUgPSBfYXNfb2JqZWN0KAogICAgICAgIHJlc3BvbnNlLAogICAgICAgICJWMl9SRVNQT05TRV9FTlZFTE9QRV9JTlZBTElEIiwKICAgICAgICAicmVzcG9uc2UgZW52ZWxvcGUgbXVzdCBiZSBhbiBvYmplY3QiLAogICAgKQogICAgY2hvaWNlcyA9IGVudmVsb3BlLmdldCgiY2hvaWNlcyIpCiAgICBpZiBub3QgaXNpbnN0YW5jZShjaG9pY2VzLCBsaXN0KSBvciBsZW4oY2hvaWNlcykgIT0gMToKICAgICAgICBfZmFpbCgiVjJfUkVTUE9OU0VfRU5WRUxPUEVfSU5WQUxJRCIsICJyZXNwb25zZSBtdXN0IGNvbnRhaW4gZXhhY3RseSBvbmUgY2hvaWNlIikKICAgIGNob2ljZSA9IF9hc19vYmplY3QoCiAgICAgICAgY2hvaWNlc1swXSwKICAgICAgICAiVjJfUkVTUE9OU0VfRU5WRUxPUEVfSU5WQUxJRCIsCiAgICAgICAgInJlc3BvbnNlIGNob2ljZSBtdXN0IGJlIGFuIG9iamVjdCIsCiAgICApCiAgICBmaW5pc2hfcmVhc29uID0gY2hvaWNlLmdldCgiZmluaXNoX3JlYXNvbiIpCiAgICBpZiBmaW5pc2hfcmVhc29uID09ICJsZW5ndGgiOgogICAgICAgIF9mYWlsKCJWMl9PVVRQVVRfVFJVTkNBVEVEIiwgImZpbmlzaF9yZWFzb24gbGVuZ3RoIGlzIGEgaGFyZCBvdXRwdXQtY29udHJhY3QgZmFpbHVyZSIpCiAgICBpZiBmaW5pc2hfcmVhc29uICE9ICJzdG9wIjoKICAgICAgICBfZmFpbCgiVjJfRklOSVNIX1JFQVNPTl9JTlZBTElEIiwgImZpbmlzaF9yZWFzb24gbXVzdCBiZSBzdG9wIikKICAgIG1lc3NhZ2UgPSBfYXNfb2JqZWN0KAogICAgICAgIGNob2ljZS5nZXQoIm1lc3NhZ2UiKSwKICAgICAgICAiVjJfUkVTUE9OU0VfRU5WRUxPUEVfSU5WQUxJRCIsCiAgICAgICAgInJlc3BvbnNlIG1lc3NhZ2UgbXVzdCBiZSBhbiBvYmplY3QiLAogICAgKQogICAgY29udGVudCA9IG1lc3NhZ2UuZ2V0KCJjb250ZW50IikKICAgIGlmIG5vdCBpc2luc3RhbmNlKGNvbnRlbnQsIHN0cik6CiAgICAgICAgX2ZhaWwoIlYyX1JFU1BPTlNFX0VOVkVMT1BFX0lOVkFMSUQiLCAicmVzcG9uc2UgY29udGVudCBtdXN0IGJlIGEgc3RyaW5nIikKICAgIHJldHVybiBjb250ZW50CgoKZGVmIGFkbWl0X3Jlc3BvbnNlKHJlc3BvbnNlOiBvYmplY3QsIHNwZWM6IG9iamVjdCkgLT4gQWRtaXR0ZWRUZXJtaW5hbE91dHB1dDoKICAgICIiIkFkbWl0IGFuIE9wZW5BSS1jb21wYXRpYmxlIHJlc3BvbnNlIG9ubHkgYWZ0ZXIgZW52ZWxvcGUgYW5kIGZpbmlzaC1yZWFzb24gY2hlY2tzLiIiIgoKICAgIHJldHVybiBhZG1pdF90ZXJtaW5hbF9vdXRwdXQoX2V4dHJhY3RfcmVzcG9uc2VfY29udGVudChyZXNwb25zZSksIHNwZWMpCgoKZGVmIGFkbWl0X2FuZF9jb21taXRfdHVybigKICAgIGhpc3Rvcnk6IEpzb25IaXN0b3J5LAogICAgdXNlcl9jb250ZW50OiBzdHIsCiAgICByZXNwb25zZTogb2JqZWN0LAogICAgc3BlYzogb2JqZWN0LAopIC0+IEFkbWl0dGVkVGVybWluYWxPdXRwdXQ6CiAgICAiIiJNdXRhdGUgaGlzdG9yeSBhdG9taWNhbGx5IG9ubHkgYWZ0ZXIgdGVybWluYWwgb3V0cHV0IGhhcyBiZWVuIGZ1bGx5IGFkbWl0dGVkLiIiIgoKICAgIGlmIG5vdCBpc2luc3RhbmNlKHVzZXJfY29udGVudCwgc3RyKSBvciBub3QgdXNlcl9jb250ZW50OgogICAgICAgIF9mYWlsKCJWMl9VU0VSX0NPTlRFTlRfSU5WQUxJRCIsICJjdXJyZW50IHVzZXIgY29udGVudCBtdXN0IGJlIGEgbm9uLWVtcHR5IHN0cmluZyIpCiAgICBhZG1pdHRlZCA9IGFkbWl0X3Jlc3BvbnNlKHJlc3BvbnNlLCBzcGVjKQogICAgaGlzdG9yeS5leHRlbmQoCiAgICAgICAgKAogICAgICAgICAgICB7InJvbGUiOiAidXNlciIsICJjb250ZW50IjogdXNlcl9jb250ZW50fSwKICAgICAgICAgICAgeyJyb2xlIjogImFzc2lzdGFudCIsICJjb250ZW50IjogYWRtaXR0ZWQuY2Fub25pY2FsX2pzb259LAogICAgICAgICkKICAgICkKICAgIHJldHVybiBhZG1pdHRlZAo="
_OUTPUT_ADMISSION_RUNTIME_SHA256 = "6ed477bf2a3bfa3dd7092f7ebf6b6d040fbc0a208ec9193e62ea5ce141f38c54"
_STANDALONE_RUNTIME_B64 = "IiIiU3RhbmRhbG9uZSBWMiBydW50aW1lIHdpdGggZXhhY3QgY3VycmVudC1yZXF1ZXN0IGFuZCBwcm9zcGVjdGl2ZSBuZXh0LXJlcXVlc3QgYnVkZ2V0IGd1YXJkcy4iIiIKCmZyb20gX19mdXR1cmVfXyBpbXBvcnQgYW5ub3RhdGlvbnMKCmZyb20gY29sbGVjdGlvbnMuYWJjIGltcG9ydCBDYWxsYWJsZQpmcm9tIGRhdGFjbGFzc2VzIGltcG9ydCBkYXRhY2xhc3MKZnJvbSB0eXBpbmcgaW1wb3J0IEZpbmFsLCBUeXBlQWxpYXMKCmZyb20gYXVyYWdhdGV3YXkubG9jYWxfYWJjLm1lYXN1cmVkX2FiY192YXJpYW5jZV9waWxvdF92Ml9vdXRwdXRfYWRtaXNzaW9uX3J1bnRpbWUgaW1wb3J0ICgKICAgIFJ1bnRpbWVPdXRwdXRBZG1pc3Npb25FcnJvciwKICAgIGFkbWl0X3Jlc3BvbnNlLAopCgpKc29uSGlzdG9yeTogVHlwZUFsaWFzID0gbGlzdFtkaWN0W3N0ciwgc3RyXV0KTWVzc2FnZUxpc3Q6IFR5cGVBbGlhcyA9IGxpc3RbZGljdFtzdHIsIHN0cl1dCk1BWF9NT0RFTF9MRU46IEZpbmFsID0gNDA5NgpNQVhfT1VUUFVUX1RPS0VOUzogRmluYWwgPSAyNTYKCgpjbGFzcyBTdGFuZGFsb25lUnVudGltZVYyRXJyb3IoUnVudGltZUVycm9yKToKICAgICIiIk1ldGFkYXRhLXNhZmUgZmFpbC1jbG9zZWQgVjIgcnVudGltZSBmYWlsdXJlLiIiIgoKICAgIGRlZiBfX2luaXRfXyhzZWxmLCBlcnJvcl9jb2RlOiBzdHIsIHNhZmVfbWVzc2FnZTogc3RyKSAtPiBOb25lOgogICAgICAgIHN1cGVyKCkuX19pbml0X18oc2FmZV9tZXNzYWdlKQogICAgICAgIHNlbGYuZXJyb3JfY29kZSA9IGVycm9yX2NvZGUKICAgICAgICBzZWxmLnNhZmVfbWVzc2FnZSA9IHNhZmVfbWVzc2FnZQoKCkBkYXRhY2xhc3MoZnJvemVuPVRydWUpCmNsYXNzIFJ1bnRpbWVUdXJuUmVxdWVzdDoKICAgIHJlcXVlc3RfaWQ6IHN0cgogICAgdXNlcl9jb250ZW50OiBzdHIKICAgIHdvcmtlcl9pZDogc3RyCgoKQGRhdGFjbGFzcyhmcm96ZW49VHJ1ZSkKY2xhc3MgVG9rZW5CdWRnZXRPYnNlcnZhdGlvbjoKICAgIHJlcXVlc3RfaWQ6IHN0cgogICAgcHJvbXB0X3Rva2VuX2NvdW50OiBpbnQKICAgIG1heF9vdXRwdXRfdG9rZW5zOiBpbnQgPSBNQVhfT1VUUFVUX1RPS0VOUwogICAgbWF4X21vZGVsX2xlbjogaW50ID0gTUFYX01PREVMX0xFTgoKCkBkYXRhY2xhc3MoZnJvemVuPVRydWUpCmNsYXNzIFRyYWplY3RvcnlFeGVjdXRpb25SZXN1bHQ6CiAgICBjb21wbGV0ZWRfdHVybl9jb3VudDogaW50CiAgICByZXF1ZXN0X2F0dGVtcHRfY291bnQ6IGludAogICAgZmFpbGVkOiBib29sCiAgICBmYWlsdXJlX2NvZGU6IHN0ciB8IE5vbmUKCgpUb2tlbkNvdW50ZXI6IFR5cGVBbGlhcyA9IENhbGxhYmxlW1tNZXNzYWdlTGlzdF0sIGludF0KTWVzc2FnZVJlbmRlcmVyOiBUeXBlQWxpYXMgPSBDYWxsYWJsZVtbUnVudGltZVR1cm5SZXF1ZXN0LCBKc29uSGlzdG9yeV0sIE1lc3NhZ2VMaXN0XQpSZXNwb25zZVNlbmRlcjogVHlwZUFsaWFzID0gQ2FsbGFibGVbW1J1bnRpbWVUdXJuUmVxdWVzdCwgTWVzc2FnZUxpc3RdLCBvYmplY3RdCgoKZGVmIF92YWxpZGF0ZV9tZXNzYWdlcyhtZXNzYWdlczogTWVzc2FnZUxpc3QpIC0+IE5vbmU6CiAgICBpZiBub3QgaXNpbnN0YW5jZShtZXNzYWdlcywgbGlzdCkgb3Igbm90IG1lc3NhZ2VzOgogICAgICAgIHJhaXNlIFN0YW5kYWxvbmVSdW50aW1lVjJFcnJvcigKICAgICAgICAgICAgIlYyX1JVTlRJTUVfTUVTU0FHRVNfSU5WQUxJRCIsCiAgICAgICAgICAgICJyZW5kZXJlZCByZXF1ZXN0IG1lc3NhZ2VzIG11c3QgYmUgYSBub24tZW1wdHkgbGlzdCIsCiAgICAgICAgKQogICAgZm9yIG1lc3NhZ2UgaW4gbWVzc2FnZXM6CiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2UobWVzc2FnZSwgZGljdCkgb3Igc2V0KG1lc3NhZ2UpICE9IHsicm9sZSIsICJjb250ZW50In06CiAgICAgICAgICAgIHJhaXNlIFN0YW5kYWxvbmVSdW50aW1lVjJFcnJvcigKICAgICAgICAgICAgICAgICJWMl9SVU5USU1FX01FU1NBR0VTX0lOVkFMSUQiLAogICAgICAgICAgICAgICAgInJlbmRlcmVkIHJlcXVlc3QgbWVzc2FnZSBzaGFwZSBpcyBpbnZhbGlkIiwKICAgICAgICAgICAgKQogICAgICAgIGlmIG5vdCBpc2luc3RhbmNlKG1lc3NhZ2UuZ2V0KCJyb2xlIiksIHN0cikgb3Igbm90IG1lc3NhZ2VbInJvbGUiXToKICAgICAgICAgICAgcmFpc2UgU3RhbmRhbG9uZVJ1bnRpbWVWMkVycm9yKAogICAgICAgICAgICAgICAgIlYyX1JVTlRJTUVfTUVTU0FHRVNfSU5WQUxJRCIsCiAgICAgICAgICAgICAgICAicmVuZGVyZWQgcmVxdWVzdCByb2xlIGlzIGludmFsaWQiLAogICAgICAgICAgICApCiAgICAgICAgaWYgbm90IGlzaW5zdGFuY2UobWVzc2FnZS5nZXQoImNvbnRlbnQiKSwgc3RyKToKICAgICAgICAgICAgcmFpc2UgU3RhbmRhbG9uZVJ1bnRpbWVWMkVycm9yKAogICAgICAgICAgICAgICAgIlYyX1JVTlRJTUVfTUVTU0FHRVNfSU5WQUxJRCIsCiAgICAgICAgICAgICAgICAicmVuZGVyZWQgcmVxdWVzdCBjb250ZW50IGlzIGludmFsaWQiLAogICAgICAgICAgICApCgoKZGVmIGNoZWNrX3Rva2VuX2J1ZGdldCgKICAgIHJlcXVlc3RfaWQ6IHN0ciwKICAgIG1lc3NhZ2VzOiBNZXNzYWdlTGlzdCwKICAgIHRva2VuX2NvdW50ZXI6IFRva2VuQ291bnRlciwKKSAtPiBUb2tlbkJ1ZGdldE9ic2VydmF0aW9uOgogICAgaWYgbm90IGlzaW5zdGFuY2UocmVxdWVzdF9pZCwgc3RyKSBvciBub3QgcmVxdWVzdF9pZDoKICAgICAgICByYWlzZSBTdGFuZGFsb25lUnVudGltZVYyRXJyb3IoCiAgICAgICAgICAgICJWMl9SVU5USU1FX1JFUVVFU1RfSURfSU5WQUxJRCIsCiAgICAgICAgICAgICJyZXF1ZXN0IGlkZW50aXR5IG11c3QgYmUgYSBub24tZW1wdHkgc3RyaW5nIiwKICAgICAgICApCiAgICBfdmFsaWRhdGVfbWVzc2FnZXMobWVzc2FnZXMpCiAgICBwcm9tcHRfdG9rZW5zID0gdG9rZW5fY291bnRlcihtZXNzYWdlcykKICAgIGlmIG5vdCBpc2luc3RhbmNlKHByb21wdF90b2tlbnMsIGludCkgb3IgaXNpbnN0YW5jZShwcm9tcHRfdG9rZW5zLCBib29sKSBvciBwcm9tcHRfdG9rZW5zIDwgMDoKICAgICAgICByYWlzZSBTdGFuZGFsb25lUnVudGltZVYyRXJyb3IoCiAgICAgICAgICAgICJWMl9SVU5USU1FX1RPS0VOX0NPVU5UX0lOVkFMSUQiLAogICAgICAgICAgICAiYWNjZXB0ZWQgdG9rZW5pemVyIHJldHVybmVkIGFuIGludmFsaWQgcHJvbXB0LXRva2VuIGNvdW50IiwKICAgICAgICApCiAgICBpZiBwcm9tcHRfdG9rZW5zICsgTUFYX09VVFBVVF9UT0tFTlMgPiBNQVhfTU9ERUxfTEVOOgogICAgICAgIHJhaXNlIFN0YW5kYWxvbmVSdW50aW1lVjJFcnJvcigKICAgICAgICAgICAgIlYyX1JVTlRJTUVfVE9LRU5fQlVER0VUX0VYQ0VFREVEIiwKICAgICAgICAgICAgInByb21wdCBwbHVzIGZyb3plbiBvdXRwdXQgYnVkZ2V0IGV4Y2VlZHMgbWF4X21vZGVsX2xlbiIsCiAgICAgICAgKQogICAgcmV0dXJuIFRva2VuQnVkZ2V0T2JzZXJ2YXRpb24ocmVxdWVzdF9pZD1yZXF1ZXN0X2lkLCBwcm9tcHRfdG9rZW5fY291bnQ9cHJvbXB0X3Rva2VucykKCgpkZWYgX2NhbmRpZGF0ZV9oaXN0b3J5KAogICAgaGlzdG9yeTogSnNvbkhpc3RvcnksCiAgICB1c2VyX2NvbnRlbnQ6IHN0ciwKICAgIGFzc2lzdGFudF9jb250ZW50OiBzdHIsCikgLT4gSnNvbkhpc3Rvcnk6CiAgICByZXR1cm4gWwogICAgICAgICooeyJyb2xlIjogaXRlbVsicm9sZSJdLCAiY29udGVudCI6IGl0ZW1bImNvbnRlbnQiXX0gZm9yIGl0ZW0gaW4gaGlzdG9yeSksCiAgICAgICAgeyJyb2xlIjogInVzZXIiLCAiY29udGVudCI6IHVzZXJfY29udGVudH0sCiAgICAgICAgeyJyb2xlIjogImFzc2lzdGFudCIsICJjb250ZW50IjogYXNzaXN0YW50X2NvbnRlbnR9LAogICAgXQoKCmRlZiBfcHJvc3BlY3RpdmVfbmV4dF90dXJuX2NoZWNrKAogICAgKiwKICAgIG5leHRfcmVxdWVzdDogUnVudGltZVR1cm5SZXF1ZXN0LAogICAgY2FuZGlkYXRlX2hpc3Rvcnk6IEpzb25IaXN0b3J5LAogICAgcmVuZGVyX21lc3NhZ2VzOiBNZXNzYWdlUmVuZGVyZXIsCiAgICB0b2tlbl9jb3VudGVyOiBUb2tlbkNvdW50ZXIsCikgLT4gTm9uZToKICAgIG5leHRfbWVzc2FnZXMgPSByZW5kZXJfbWVzc2FnZXMobmV4dF9yZXF1ZXN0LCBjYW5kaWRhdGVfaGlzdG9yeSkKICAgIHRyeToKICAgICAgICBjaGVja190b2tlbl9idWRnZXQobmV4dF9yZXF1ZXN0LnJlcXVlc3RfaWQsIG5leHRfbWVzc2FnZXMsIHRva2VuX2NvdW50ZXIpCiAgICBleGNlcHQgU3RhbmRhbG9uZVJ1bnRpbWVWMkVycm9yIGFzIGV4YzoKICAgICAgICBpZiBleGMuZXJyb3JfY29kZSA9PSAiVjJfUlVOVElNRV9UT0tFTl9CVURHRVRfRVhDRUVERUQiOgogICAgICAgICAgICByYWlzZSBTdGFuZGFsb25lUnVudGltZVYyRXJyb3IoCiAgICAgICAgICAgICAgICAiVjJfUlVOVElNRV9SRUFDSEFCTEVfUFJPTVBUX0JVREdFVF9SRUpFQ1RFRCIsCiAgICAgICAgICAgICAgICAiYWRtaXR0ZWQgb3V0cHV0IHdvdWxkIG1ha2UgdGhlIG5leHQgcmVhY2hhYmxlIHByb21wdCBleGNlZWQgYnVkZ2V0IiwKICAgICAgICAgICAgKSBmcm9tIGV4YwogICAgICAgIHJhaXNlCgoKZGVmIGV4ZWN1dGVfdHVybigKICAgICosCiAgICByZXF1ZXN0OiBSdW50aW1lVHVyblJlcXVlc3QsCiAgICBoaXN0b3J5OiBKc29uSGlzdG9yeSwKICAgIGFkbWlzc2lvbl9zcGVjOiBvYmplY3QsCiAgICByZW5kZXJfbWVzc2FnZXM6IE1lc3NhZ2VSZW5kZXJlciwKICAgIHRva2VuX2NvdW50ZXI6IFRva2VuQ291bnRlciwKICAgIHNlbmRfcmVzcG9uc2U6IFJlc3BvbnNlU2VuZGVyLAogICAgbmV4dF9yZXF1ZXN0OiBSdW50aW1lVHVyblJlcXVlc3QgfCBOb25lID0gTm9uZSwKKSAtPiBUb2tlbkJ1ZGdldE9ic2VydmF0aW9uOgogICAgaWYgbm90IHJlcXVlc3QudXNlcl9jb250ZW50OgogICAgICAgIHJhaXNlIFN0YW5kYWxvbmVSdW50aW1lVjJFcnJvcigKICAgICAgICAgICAgIlYyX1JVTlRJTUVfVVNFUl9DT05URU5UX0lOVkFMSUQiLAogICAgICAgICAgICAicnVudGltZSB0dXJuIHVzZXIgY29udGVudCBtdXN0IGJlIG5vbi1lbXB0eSIsCiAgICAgICAgKQogICAgbWVzc2FnZXMgPSByZW5kZXJfbWVzc2FnZXMocmVxdWVzdCwgaGlzdG9yeSkKICAgIG9ic2VydmF0aW9uID0gY2hlY2tfdG9rZW5fYnVkZ2V0KHJlcXVlc3QucmVxdWVzdF9pZCwgbWVzc2FnZXMsIHRva2VuX2NvdW50ZXIpCiAgICByZXNwb25zZSA9IHNlbmRfcmVzcG9uc2UocmVxdWVzdCwgbWVzc2FnZXMpCiAgICBhZG1pdHRlZCA9IGFkbWl0X3Jlc3BvbnNlKHJlc3BvbnNlLCBhZG1pc3Npb25fc3BlYykKICAgIGNhbmRpZGF0ZSA9IF9jYW5kaWRhdGVfaGlzdG9yeShoaXN0b3J5LCByZXF1ZXN0LnVzZXJfY29udGVudCwgYWRtaXR0ZWQuY2Fub25pY2FsX2pzb24pCiAgICBpZiBuZXh0X3JlcXVlc3QgaXMgbm90IE5vbmU6CiAgICAgICAgX3Byb3NwZWN0aXZlX25leHRfdHVybl9jaGVjaygKICAgICAgICAgICAgbmV4dF9yZXF1ZXN0PW5leHRfcmVxdWVzdCwKICAgICAgICAgICAgY2FuZGlkYXRlX2hpc3Rvcnk9Y2FuZGlkYXRlLAogICAgICAgICAgICByZW5kZXJfbWVzc2FnZXM9cmVuZGVyX21lc3NhZ2VzLAogICAgICAgICAgICB0b2tlbl9jb3VudGVyPXRva2VuX2NvdW50ZXIsCiAgICAgICAgKQogICAgaGlzdG9yeS5leHRlbmQoY2FuZGlkYXRlW2xlbihoaXN0b3J5KSA6XSkKICAgIHJldHVybiBvYnNlcnZhdGlvbgoKCmRlZiBleGVjdXRlX3RyYWplY3RvcnkoCiAgICAqLAogICAgcmVxdWVzdHM6IHR1cGxlW1J1bnRpbWVUdXJuUmVxdWVzdCwgUnVudGltZVR1cm5SZXF1ZXN0LCBSdW50aW1lVHVyblJlcXVlc3QsIFJ1bnRpbWVUdXJuUmVxdWVzdF0sCiAgICBoaXN0b3J5OiBKc29uSGlzdG9yeSwKICAgIGFkbWlzc2lvbl9zcGVjOiBvYmplY3QsCiAgICByZW5kZXJfbWVzc2FnZXM6IE1lc3NhZ2VSZW5kZXJlciwKICAgIHRva2VuX2NvdW50ZXI6IFRva2VuQ291bnRlciwKICAgIHNlbmRfcmVzcG9uc2U6IFJlc3BvbnNlU2VuZGVyLAopIC0+IFRyYWplY3RvcnlFeGVjdXRpb25SZXN1bHQ6CiAgICBjb21wbGV0ZWQgPSAwCiAgICBhdHRlbXB0cyA9IDAKICAgIGZvciBpbmRleCwgcmVxdWVzdCBpbiBlbnVtZXJhdGUocmVxdWVzdHMpOgogICAgICAgIHRyeToKICAgICAgICAgICAgbWVzc2FnZXMgPSByZW5kZXJfbWVzc2FnZXMocmVxdWVzdCwgaGlzdG9yeSkKICAgICAgICAgICAgY2hlY2tfdG9rZW5fYnVkZ2V0KHJlcXVlc3QucmVxdWVzdF9pZCwgbWVzc2FnZXMsIHRva2VuX2NvdW50ZXIpCiAgICAgICAgICAgIGF0dGVtcHRzICs9IDEKICAgICAgICAgICAgcmVzcG9uc2UgPSBzZW5kX3Jlc3BvbnNlKHJlcXVlc3QsIG1lc3NhZ2VzKQogICAgICAgICAgICBhZG1pdHRlZCA9IGFkbWl0X3Jlc3BvbnNlKHJlc3BvbnNlLCBhZG1pc3Npb25fc3BlYykKICAgICAgICAgICAgY2FuZGlkYXRlID0gX2NhbmRpZGF0ZV9oaXN0b3J5KGhpc3RvcnksIHJlcXVlc3QudXNlcl9jb250ZW50LCBhZG1pdHRlZC5jYW5vbmljYWxfanNvbikKICAgICAgICAgICAgaWYgaW5kZXggPCBsZW4ocmVxdWVzdHMpIC0gMToKICAgICAgICAgICAgICAgIF9wcm9zcGVjdGl2ZV9uZXh0X3R1cm5fY2hlY2soCiAgICAgICAgICAgICAgICAgICAgbmV4dF9yZXF1ZXN0PXJlcXVlc3RzW2luZGV4ICsgMV0sCiAgICAgICAgICAgICAgICAgICAgY2FuZGlkYXRlX2hpc3Rvcnk9Y2FuZGlkYXRlLAogICAgICAgICAgICAgICAgICAgIHJlbmRlcl9tZXNzYWdlcz1yZW5kZXJfbWVzc2FnZXMsCiAgICAgICAgICAgICAgICAgICAgdG9rZW5fY291bnRlcj10b2tlbl9jb3VudGVyLAogICAgICAgICAgICAgICAgKQogICAgICAgICAgICBoaXN0b3J5LmV4dGVuZChjYW5kaWRhdGVbbGVuKGhpc3RvcnkpIDpdKQogICAgICAgIGV4Y2VwdCBTdGFuZGFsb25lUnVudGltZVYyRXJyb3IgYXMgZXhjOgogICAgICAgICAgICByZXR1cm4gVHJhamVjdG9yeUV4ZWN1dGlvblJlc3VsdCgKICAgICAgICAgICAgICAgIGNvbXBsZXRlZF90dXJuX2NvdW50PWNvbXBsZXRlZCwKICAgICAgICAgICAgICAgIHJlcXVlc3RfYXR0ZW1wdF9jb3VudD1hdHRlbXB0cywKICAgICAgICAgICAgICAgIGZhaWxlZD1UcnVlLAogICAgICAgICAgICAgICAgZmFpbHVyZV9jb2RlPWV4Yy5lcnJvcl9jb2RlLAogICAgICAgICAgICApCiAgICAgICAgZXhjZXB0IFJ1bnRpbWVPdXRwdXRBZG1pc3Npb25FcnJvciBhcyBleGM6CiAgICAgICAgICAgIHJldHVybiBUcmFqZWN0b3J5RXhlY3V0aW9uUmVzdWx0KAogICAgICAgICAgICAgICAgY29tcGxldGVkX3R1cm5fY291bnQ9Y29tcGxldGVkLAogICAgICAgICAgICAgICAgcmVxdWVzdF9hdHRlbXB0X2NvdW50PWF0dGVtcHRzLAogICAgICAgICAgICAgICAgZmFpbGVkPVRydWUsCiAgICAgICAgICAgICAgICBmYWlsdXJlX2NvZGU9ZXhjLmVycm9yX2NvZGUsCiAgICAgICAgICAgICkKICAgICAgICBjb21wbGV0ZWQgKz0gMQogICAgcmV0dXJuIFRyYWplY3RvcnlFeGVjdXRpb25SZXN1bHQoCiAgICAgICAgY29tcGxldGVkX3R1cm5fY291bnQ9Y29tcGxldGVkLAogICAgICAgIHJlcXVlc3RfYXR0ZW1wdF9jb3VudD1hdHRlbXB0cywKICAgICAgICBmYWlsZWQ9RmFsc2UsCiAgICAgICAgZmFpbHVyZV9jb2RlPU5vbmUsCiAgICApCg=="
_STANDALONE_RUNTIME_SHA256 = "879ba2812f3602860639572332b18d1fae67a8c4105b749568dc489f5cfa8cf8"
_LIVE_SEMANTICS_RUNTIME_B64 = ""
_LIVE_SEMANTICS_RUNTIME_SHA256 = "a3c10f8f9304dddb22f58d95028dfec69768fec9bcab5a6a44e4a61856aaaea5"
_REQUEST_ADAPTER_B64 = ""
_REQUEST_ADAPTER_SHA256 = "88f1a3b3d988219dac150998a83b0ba3b676ec47e5fff8da1afbce5651aa50fb"
_TRANSACTION_RUNTIME_B64 = ""
_TRANSACTION_RUNTIME_SHA256 = "796c6cfc2fa615a5b28b4de013e3b2bea45c758cd96c039fe6464fce637b89ce"
_MATERIAL_B64 = ""
_MATERIAL_SHA256 = "5948126418a4862a29ce6b93a1809012304d69fbfc32e2de67d35cf6d33719d7"

_AURAGATEWAY_PACKAGE = "auragateway"
_LOCAL_ABC_PACKAGE = "auragateway.local_abc"
_OUTPUT_ADMISSION_MODULE = (
    "auragateway.local_abc.measured_abc_variance_pilot_v2_output_admission_runtime"
)
_STANDALONE_MODULE = "auragateway.local_abc.measured_abc_variance_pilot_v2_standalone_runtime_v2"
_LIVE_SEMANTICS_MODULE = (
    "auragateway.local_abc.measured_abc_variance_pilot_v2_live_semantics_runtime_v1"
)
_REQUEST_ADAPTER_MODULE = (
    "auragateway.local_abc.measured_abc_variance_pilot_v2_accepted_runtime_request_adapter_v1"
)
_R2_RUNTIME_MODULE = "auragateway_r2_runtime_library"
_TRANSACTION_RUNTIME_MODULE = (
    "auragateway.local_abc.measured_abc_variance_pilot_v2_transaction_bound_runtime_v1"
)

_LIVE_EXECUTION_ENABLED = True
_AUTHORIZATION_B64 = "eyJhdXRob3JpemF0aW9uIjp7ImF1dGhvcml6YXRpb25faWQiOiJ2YXJpYW5jZS1waWxvdC12Mi10eC1jYWQ3ZWM1NGQ5ZTc3OTI3NWZkNDQwNTlhNjVjOGY3YyIsImF1dGhvcml6YXRpb25fcHJvZHVjZXJfbm90ZWJvb2tzIjowLCJhdXRob3JpemF0aW9uX3JldXNhYmxlIjpmYWxzZSwiYXV0aG9yaXphdGlvbl9zcGVjaWZpY19rYWdnbGVfaW5wdXRzIjowLCJib3VuZF9hcnRpZmFjdF9jb3VudCI6MTksImJ1ZGdldCI6eyJtYXhpbXVtX2V4dGVybmFsX25ldHdvcmtfcmVxdWVzdHMiOjAsIm1heGltdW1fZXh0ZXJuYWxfc3BlbmQiOjAsIm1heGltdW1faGlkZGVuX3JldHJpZXMiOjAsIm1heGltdW1fa2FnZ2xlX3Nlc3Npb25zIjoxLCJtYXhpbXVtX21vZGVsX2xvYWRzIjoyLCJtYXhpbXVtX25ldXRyYWxfcXVhbGlmaWNhdGlvbl9yZXF1ZXN0cyI6MjAsIm1heGltdW1fb3V0cHV0X3Rva2Vuc19wZXJfcmVxdWVzdCI6MjU2LCJtYXhpbXVtX3BpbG90X3JlcXVlc3RfYXR0ZW1wdHMiOjIxNiwibWF4aW11bV9waWxvdF90cmFqZWN0b3J5X2NvdW50Ijo1NCwibWF4aW11bV9waWxvdF90dXJuX2NvdW50IjoyMTYsIm1heGltdW1fcHJldHJlYXRtZW50X3JlcXVlc3RzIjoyNCwibWF4aW11bV9yZXBsYWNlbWVudF9jYXNlcyI6MCwibWF4aW11bV9ydW50aW1lX2luc3RhbGxfYXR0ZW1wdHMiOjEsIm1heGltdW1fc2F2ZV9hbmRfcnVuX2FsbF9hY3Rpb25zIjoxLCJtYXhpbXVtX3NjaGVtYV9jYW5hcnlfcmVxdWVzdHMiOjIsIm1heGltdW1fdG90YWxfbW9kZWxfcmVxdWVzdHMiOjI0MCwibWF4aW11bV93YXJtdXBfcmVxdWVzdHMiOjIsIm1heGltdW1fd29ya2VyX3N0YXJ0cyI6MiwibWF4aW11bV93b3JrZXJfdGVhcmRvd25zIjoyfSwiZGVjaXNpb24iOiJBVVRIT1JJWkVEIiwiZXhwaXJlc19hdCI6IjIwMjYtMDgtMjdUMjI6MjQ6MjguNzI1OTkyWiIsImZpbmFsX21lYXN1cmVkX2FiY19leGVjdXRpb25fYXV0aG9yaXplZCI6ZmFsc2UsImZyZXNoX3Bvc3RfYXJ0aWZhY3RfcGxhdGZvcm1fb2JzZXJ2YXRpb25fcmVxdWlyZWQiOnRydWUsImlzc3VlZF9hdCI6IjIwMjYtMDgtMjdUMTk6MjQ6MjguNzI1OTkyWiIsImlzc3Vlcl9tZXJnZV9jb21taXQiOiI1NjNhYTk5OTU4ZDc1ZDNjZjA5ZWI4M2RjNmVhMzA2MmU4MzA4YTkwIiwiaXNzdWVyX3NvdXJjZV9zaGEyNTYiOiJmYzQ0MWIzMzdkNjE2MjYyYjBlZTEwMjE2NjMxNDMzM2ExM2RlNGI5YmQyNGIyY2MxNTUwODRmOTU5NzE0NGE4IiwibGlmZWN5Y2xlIjoiSVNTVUVEIiwibWFudWFsX2NvbmZpcm1hdGlvbl9qc29uX2ZpbGVzIjowLCJtYXRlcmlhbF9zaGEyNTYiOiI1OTQ4MTI2NDE4YTQ4NjJhMjljZTZiOTNhMTgwOTAxMjMwNGQ2OWZiZmMzMmUyZGU2N2QzNWNmNmQzMzcxOWQ3IiwicGlsb3RfZXhlY3V0aW9uX2F1dGhvcml6ZWQiOnRydWUsInBsYXRmb3JtX29ic2VydmF0aW9uX3J1bnRpbWVfaW5wdXQiOmZhbHNlLCJyZW5kZXJlZF93cmFwcGVyX3NoYTI1NiI6IjViYzI2YjIyMGM3YjM0ZGE3ZDYzNDY4NmFmOTdkNjdiMGQyNjY2Zjg0YWM2ZjhkOWIxZjIxNDE2M2MxN2NiNDEiLCJyZXF1aXJlZF9wbGF0Zm9ybSI6eyJhY2NlbGVyYXRvciI6IlQ0X1gyIiwiYWxsb2NhdGVkX2dwdV9jb3VudCI6MiwiY3JlZGVudGlhbHNfcGVybWl0dGVkIjpmYWxzZSwiY3VzdG9tZXJfZGF0YV9wZXJtaXR0ZWQiOmZhbHNlLCJleHRlcm5hbF9uZXR3b3JrX2FjY2Vzc19wZXJtaXR0ZWQiOmZhbHNlLCJmcmVzaF9wb3N0X2FydGlmYWN0X29ic2VydmF0aW9uX3JlcXVpcmVkIjp0cnVlLCJpbnRlcm5ldF9lbmFibGVkIjpmYWxzZSwib2JzZXJ2YXRpb25fcHJlY2VkZXNfc2F2ZV9hbmRfcnVuX2FsbCI6dHJ1ZSwib2JzZXJ2YXRpb25fcnVudGltZV9pbnB1dCI6ZmFsc2V9LCJydW50aW1lIjp7ImF0dGVudGlvbl9iYWNrZW5kIjoiVFJJVE9OX0FUVE4iLCJjdWRhX3ZhcmlhbnQiOiJjdTEyOSIsImdwdV90b3BvbG9neSI6IlQ0X1gyIiwibWF4X21vZGVsX2xlbiI6NDA5NiwibW9kZWxfcmVwb3NpdG9yeSI6IlF3ZW4vUXdlbjIuNS0wLjVCLUluc3RydWN0IiwibW9kZWxfcmV2aXNpb24iOiI3YWU1NTc2MDRhZGY2N2JlNTA0MTdmNTljMmMyZjE2N2RlZjlhNzc1IiwicHl0aG9uIjoiMy4xMiIsInJlcXVpcmVkX25hdGl2ZV9tb2R1bGUiOiJ2bGxtLl9DX3N0YWJsZV9saWJ0b3JjaCIsInRvcmNoIjoiMi4xMS4wK2N1MTI5IiwidG9yY2hfY3VkYV92ZXJzaW9uIjoiMTIuOSIsInRyYW5zZm9ybWVycyI6IjUuMTQuMSIsInRyaXRvbiI6IjMuNi4wIiwidmxsbV9kaXN0cmlidXRpb24iOiIwLjI1LjErY3UxMjkiLCJ2bGxtX3B1YmxpY19zZW1hbnRpY192ZXJzaW9uIjoiMC4yNS4xIn0sInJ1bnRpbWVfYW50aV9yZXBsYXlfZXN0YWJsaXNoZWQiOmZhbHNlLCJzY2hlbWFfdmVyc2lvbiI6IjEuMC4wIiwic2NvcGUiOiJNRUFTVVJFRF9BQkNfVkFSSUFOQ0VfUElMT1RfU1VDQ0VTU09SX1YyX1RSQU5TQUNUSU9OX0JPVU5EX1YxIiwic2luZ2xlX3VzZSI6dHJ1ZSwic3RhdGljX2F1dGhvcml0eV9iaW5kaW5nX3JlY29yZF9zaGEyNTYiOiIyMTMzMTAzMzE0OWRjN2M3OWE5NGFhMjdkZmQ4MzFiNTI1MTg1ZmRjMGE5ZjMyOTE2MzA2NTIyNTk4Y2RlMzRmIiwidHJhbnNhY3Rpb25fcnVudGltZV9zaGEyNTYiOiI3OTZjNmNmYzJmYTYxNWE1YjI4YjRkZTAxM2UzYjJiZWE0NWM3NThjZDk2YzAzOWZlNjQ2NGZjZTYzN2I4OWNlIiwidW5jaGFuZ2VkX3JlcGxheV9hdXRob3JpemVkIjpmYWxzZX0sInRyYW5zYWN0aW9uX2lkIjoiNDM0MWNhZmFjODEyNDVkNDMzYTY4MGRiMGJjOWM2MmVjYWJkYmYxZDI3OWMwZGRjMGExOTc0MWViNDRjN2Q4YiJ9"
_TRANSACTION_ID = "4341cafac81245d433a680db0bc9c62ecabdbf1d279c0ddc0a19741eb44c7d8b"
_ISSUER_MERGE_COMMIT = "563aa99958d75d3cf09eb83dc6ea3062e8308a90"
_ISSUER_SOURCE_SHA256 = "fc441b337d616262b0ee102166314333a13de4b9bd24b2cc155084f9597144a8"
_STATIC_BINDING_RECORD_SHA256 = "21331033149dc7c79a94aa27dfd831b525185fdc0a9f32916306522598cde34f"
_EXPECTED_RUNTIME_B64 = "eyJhdHRlbnRpb25fYmFja2VuZCI6IlRSSVRPTl9BVFROIiwiY3VkYV92YXJpYW50IjoiY3UxMjkiLCJncHVfdG9wb2xvZ3kiOiJUNF9YMiIsIm1heF9tb2RlbF9sZW4iOjQwOTYsIm1vZGVsX3JlcG9zaXRvcnkiOiJRd2VuL1F3ZW4yLjUtMC41Qi1JbnN0cnVjdCIsIm1vZGVsX3JldmlzaW9uIjoiN2FlNTU3NjA0YWRmNjdiZTUwNDE3ZjU5YzJjMmYxNjdkZWY5YTc3NSIsInB5dGhvbiI6IjMuMTIiLCJyZXF1aXJlZF9uYXRpdmVfbW9kdWxlIjoidmxsbS5fQ19zdGFibGVfbGlidG9yY2giLCJ0b3JjaCI6IjIuMTEuMCtjdTEyOSIsInRvcmNoX2N1ZGFfdmVyc2lvbiI6IjEyLjkiLCJ0cmFuc2Zvcm1lcnMiOiI1LjE0LjEiLCJ0cml0b24iOiIzLjYuMCIsInZsbG1fZGlzdHJpYnV0aW9uIjoiMC4yNS4xK2N1MTI5IiwidmxsbV9wdWJsaWNfc2VtYW50aWNfdmVyc2lvbiI6IjAuMjUuMSJ9"
_EXPECTED_BUDGET_B64 = "eyJtYXhpbXVtX2V4dGVybmFsX25ldHdvcmtfcmVxdWVzdHMiOjAsIm1heGltdW1fZXh0ZXJuYWxfc3BlbmQiOjAsIm1heGltdW1faGlkZGVuX3JldHJpZXMiOjAsIm1heGltdW1fa2FnZ2xlX3Nlc3Npb25zIjoxLCJtYXhpbXVtX21vZGVsX2xvYWRzIjoyLCJtYXhpbXVtX25ldXRyYWxfcXVhbGlmaWNhdGlvbl9yZXF1ZXN0cyI6MjAsIm1heGltdW1fb3V0cHV0X3Rva2Vuc19wZXJfcmVxdWVzdCI6MjU2LCJtYXhpbXVtX3BpbG90X3JlcXVlc3RfYXR0ZW1wdHMiOjIxNiwibWF4aW11bV9waWxvdF90cmFqZWN0b3J5X2NvdW50Ijo1NCwibWF4aW11bV9waWxvdF90dXJuX2NvdW50IjoyMTYsIm1heGltdW1fcHJldHJlYXRtZW50X3JlcXVlc3RzIjoyNCwibWF4aW11bV9yZXBsYWNlbWVudF9jYXNlcyI6MCwibWF4aW11bV9ydW50aW1lX2luc3RhbGxfYXR0ZW1wdHMiOjEsIm1heGltdW1fc2F2ZV9hbmRfcnVuX2FsbF9hY3Rpb25zIjoxLCJtYXhpbXVtX3NjaGVtYV9jYW5hcnlfcmVxdWVzdHMiOjIsIm1heGltdW1fdG90YWxfbW9kZWxfcmVxdWVzdHMiOjI0MCwibWF4aW11bV93YXJtdXBfcmVxdWVzdHMiOjIsIm1heGltdW1fd29ya2VyX3N0YXJ0cyI6MiwibWF4aW11bV93b3JrZXJfdGVhcmRvd25zIjoyfQ=="
_EXPECTED_PLATFORM_B64 = "eyJhY2NlbGVyYXRvciI6IlQ0X1gyIiwiYWxsb2NhdGVkX2dwdV9jb3VudCI6MiwiY3JlZGVudGlhbHNfcGVybWl0dGVkIjpmYWxzZSwiY3VzdG9tZXJfZGF0YV9wZXJtaXR0ZWQiOmZhbHNlLCJleHRlcm5hbF9uZXR3b3JrX2FjY2Vzc19wZXJtaXR0ZWQiOmZhbHNlLCJmcmVzaF9wb3N0X2FydGlmYWN0X29ic2VydmF0aW9uX3JlcXVpcmVkIjp0cnVlLCJpbnRlcm5ldF9lbmFibGVkIjpmYWxzZSwib2JzZXJ2YXRpb25fcHJlY2VkZXNfc2F2ZV9hbmRfcnVuX2FsbCI6dHJ1ZSwib2JzZXJ2YXRpb25fcnVudGltZV9pbnB1dCI6ZmFsc2V9"


def _parse_time(value: object) -> datetime:
    if not isinstance(value, str):
        raise RuntimeError("authorization timestamp is invalid")
    observed = datetime.fromisoformat(value)
    if observed.tzinfo is None or observed.utcoffset() is None:
        raise RuntimeError("authorization timestamp must be timezone-aware")
    return observed.astimezone(UTC)


def _gpu_count() -> int:
    completed = subprocess.run(
        ["nvidia-smi", "-L"],
        check=False,
        capture_output=True,
        text=True,
    )
    if completed.returncode != 0:
        raise RuntimeError("unable to observe GPU topology")
    return sum(line.strip().startswith("GPU ") for line in completed.stdout.splitlines())


def admit(now: datetime | None = None) -> dict[str, object]:
    envelope = json.loads(_decode(_AUTHORIZATION_B64))
    if not isinstance(envelope, dict):
        raise RuntimeError("embedded authorization envelope is invalid")
    authorization = envelope.get("authorization")
    if envelope.get("transaction_id") != _TRANSACTION_ID or not isinstance(
        authorization, dict
    ):
        raise RuntimeError("embedded transaction identity drifted")
    if _sha256(_canonical_json(authorization).encode("utf-8")) != _TRANSACTION_ID:
        raise RuntimeError("inner authorization transaction identity mismatch")

    required = {
        "schema_version": "1.0.0",
        "decision": "AUTHORIZED",
        "lifecycle": "ISSUED",
        "scope": "MEASURED_ABC_VARIANCE_PILOT_SUCCESSOR_V2_TRANSACTION_BOUND_V1",
        "issuer_merge_commit": _ISSUER_MERGE_COMMIT,
        "issuer_source_sha256": _ISSUER_SOURCE_SHA256,
        "static_authority_binding_record_sha256": _STATIC_BINDING_RECORD_SHA256,
        "rendered_wrapper_sha256": "5bc26b220c7b34da7d634686af97d67b0d2666f84ac6f8d9b1f214163c17cb41",
        "bound_artifact_count": 19,
        "transaction_runtime_sha256": _TRANSACTION_RUNTIME_SHA256,
        "material_sha256": _MATERIAL_SHA256,
        "pilot_execution_authorized": True,
        "final_measured_abc_execution_authorized": False,
        "single_use": True,
        "authorization_reusable": False,
        "unchanged_replay_authorized": False,
        "runtime_anti_replay_established": False,
        "authorization_specific_kaggle_inputs": 0,
        "authorization_producer_notebooks": 0,
        "manual_confirmation_json_files": 0,
        "fresh_post_artifact_platform_observation_required": True,
        "platform_observation_runtime_input": False,
    }
    drift = [
        key for key, expected in required.items() if authorization.get(key) != expected
    ]
    if drift:
        raise RuntimeError("embedded V2 authorization drift: " + ",".join(drift))

    expected_runtime = json.loads(_decode(_EXPECTED_RUNTIME_B64))
    expected_budget = json.loads(_decode(_EXPECTED_BUDGET_B64))
    expected_platform = json.loads(_decode(_EXPECTED_PLATFORM_B64))
    if authorization.get("runtime") != expected_runtime:
        raise RuntimeError("embedded V2 runtime contract drifted")
    if authorization.get("budget") != expected_budget:
        raise RuntimeError("embedded V2 execution budget drifted")
    if authorization.get("required_platform") != expected_platform:
        raise RuntimeError("embedded V2 platform contract drifted")

    observed = (now or datetime.now(UTC)).astimezone(UTC)
    issued = _parse_time(authorization.get("issued_at"))
    expires = _parse_time(authorization.get("expires_at"))
    if observed < issued or observed >= expires:
        raise RuntimeError("authorization is outside its live window")
    if _gpu_count() != 2:
        raise RuntimeError("machine-observable GPU topology is not T4 x2")

    return {
        "status": "VARIANCE_PILOT_SUCCESSOR_V2_TRANSACTION_BOUND_RUNTIME_ADMISSION_VALID",
        "transaction_id": _TRANSACTION_ID,
        "issuer_merge_commit": _ISSUER_MERGE_COMMIT,
        "static_authority_binding_record_sha256": _STATIC_BINDING_RECORD_SHA256,
        "rendered_wrapper_sha256": "5bc26b220c7b34da7d634686af97d67b0d2666f84ac6f8d9b1f214163c17cb41",
        "bound_artifact_count": 19,
        "observed_gpu_count": 2,
        "network_probe_performed": False,
        "platform_observation_receipt_runtime_input": False,
        "live_execution_enabled": True,
    }


def _canonical_json(value: object) -> str:
    return json.dumps(value, ensure_ascii=True, separators=(",", ":"), sort_keys=True)


def _sha256(value: bytes) -> str:
    return hashlib.sha256(value).hexdigest()


def _decode(value: str) -> bytes:
    return base64.b64decode(value.encode("ascii"), validate=True)


def _payload(value: str, expected_sha256: str, label: str) -> bytes:
    raw = _decode(value)
    if _sha256(raw) != expected_sha256:
        raise RuntimeError(label + " identity mismatch")
    return raw


def _create_package(name: str, created: list[str]) -> types.ModuleType:
    if name in sys.modules:
        raise RuntimeError("governed rehearsal package is already registered: " + name)
    package = types.ModuleType(name)
    package.__package__ = name
    package.__path__ = []
    package.__file__ = "<embedded-package:" + name + ">"
    sys.modules[name] = package
    created.append(name)
    parent_name, _, child_name = name.rpartition(".")
    if parent_name:
        parent = sys.modules.get(parent_name)
        if parent is None:
            raise RuntimeError("governed rehearsal parent package is missing: " + parent_name)
        setattr(parent, child_name, package)
    return package


def _load_module(
    name: str,
    filename: str,
    source: bytes,
    created: list[str],
    seed: dict[str, object] | None = None,
) -> types.ModuleType:
    if name in sys.modules:
        raise RuntimeError("governed rehearsal module is already registered: " + name)
    module = types.ModuleType(name)
    module.__file__ = filename
    module.__package__ = name.rpartition(".")[0]
    if seed:
        module.__dict__.update(seed)
    sys.modules[name] = module
    created.append(name)
    try:
        exec(compile(source, filename, "exec"), module.__dict__, module.__dict__)
    except BaseException:
        sys.modules.pop(name, None)
        created.remove(name)
        raise
    parent_name, _, child_name = name.rpartition(".")
    if parent_name:
        parent = sys.modules.get(parent_name)
        if parent is not None:
            setattr(parent, child_name, module)
    return module


def _cleanup(created: list[str]) -> None:
    for name in reversed(created):
        sys.modules.pop(name, None)


def _execute_runtime_source(source: bytes) -> int:
    namespace: dict[str, object] = {"__name__": "__main__", "__file__": "<system-exit-probe>"}
    try:
        exec(compile(source, "<system-exit-probe>", "exec"), namespace, namespace)
    except SystemExit as exc:
        if exc.code in (None, 0):
            return 0
        raise
    return 0


def _bootstrap_failure_cleanup_probe() -> bool:
    name = "_auragateway_v2_bootstrap_failure_probe"
    sys.modules.pop(name, None)
    created: list[str] = []
    try:
        _load_module(
            name,
            "<bootstrap-failure-probe>",
            b"raise RuntimeError('expected rehearsal probe')\n",
            created,
        )
    except RuntimeError as exc:
        if str(exc) != "expected rehearsal probe":
            raise
    else:
        raise RuntimeError("bootstrap failure probe unexpectedly succeeded")
    return name not in sys.modules and not created


def _realize_graph() -> tuple[dict[str, types.ModuleType], list[str], dict[str, object]]:
    payloads = {
        "r2": _payload(_R2_RUNTIME_B64, _R2_RUNTIME_SHA256, "R2 runtime"),
        "admission": _payload(
            _OUTPUT_ADMISSION_RUNTIME_B64,
            _OUTPUT_ADMISSION_RUNTIME_SHA256,
            "output-admission runtime",
        ),
        "standalone": _payload(
            _STANDALONE_RUNTIME_B64,
            _STANDALONE_RUNTIME_SHA256,
            "standalone runtime",
        ),
        "live": _payload(
            _LIVE_SEMANTICS_RUNTIME_B64,
            _LIVE_SEMANTICS_RUNTIME_SHA256,
            "live-semantics runtime",
        ),
        "adapter": _payload(
            _REQUEST_ADAPTER_B64,
            _REQUEST_ADAPTER_SHA256,
            "request-adapter runtime",
        ),
        "transaction": _payload(
            _TRANSACTION_RUNTIME_B64,
            _TRANSACTION_RUNTIME_SHA256,
            "transaction runtime",
        ),
    }
    material_raw = _payload(_MATERIAL_B64, _MATERIAL_SHA256, "V2 transaction material")
    material = json.loads(material_raw)
    if not isinstance(material, dict):
        raise RuntimeError("V2 transaction material root is invalid")

    created: list[str] = []
    modules: dict[str, types.ModuleType] = {}
    try:
        _create_package(_AURAGATEWAY_PACKAGE, created)
        _create_package(_LOCAL_ABC_PACKAGE, created)

        modules["admission"] = _load_module(
            _OUTPUT_ADMISSION_MODULE,
            "<embedded-v2-output-admission-runtime>",
            payloads["admission"],
            created,
        )
        modules["standalone"] = _load_module(
            _STANDALONE_MODULE,
            "<embedded-v2-standalone-runtime>",
            payloads["standalone"],
            created,
        )
        modules["live"] = _load_module(
            _LIVE_SEMANTICS_MODULE,
            "<embedded-v2-live-semantics-runtime>",
            payloads["live"],
            created,
        )
        modules["adapter"] = _load_module(
            _REQUEST_ADAPTER_MODULE,
            "<embedded-v2-request-adapter-runtime>",
            payloads["adapter"],
            created,
        )
        modules["r2"] = _load_module(
            _R2_RUNTIME_MODULE,
            "<embedded-accepted-r2-runtime>",
            payloads["r2"],
            created,
        )
        modules["transaction"] = _load_module(
            _TRANSACTION_RUNTIME_MODULE,
            "<embedded-v2-transaction-runtime>",
            payloads["transaction"],
            created,
            {
                "AURAGATEWAY_R2_RUNTIME": modules["r2"].__dict__,
                "AURAGATEWAY_V2_LIVE_SEMANTICS": modules["live"].__dict__,
                "AURAGATEWAY_V2_REQUEST_ADAPTER": modules["adapter"].__dict__,
                "AURAGATEWAY_V2_STANDALONE_RUNTIME": modules["standalone"].__dict__,
                "AURAGATEWAY_V2_OUTPUT_ADMISSION_RUNTIME": modules["admission"].__dict__,
                "AURAGATEWAY_V2_MATERIAL": material,
                "AURAGATEWAY_TRANSACTION_ID": _TRANSACTION_ID,
                "EXECUTED_RUNTIME_SCRIPT_SHA256": _TRANSACTION_RUNTIME_SHA256,
            },
        )
    except BaseException:
        _cleanup(created)
        raise
    return modules, created, material


def rehearse() -> dict[str, object]:
    if _LIVE_EXECUTION_ENABLED:
        raise RuntimeError("structural rehearsal wrapper unexpectedly enables live execution")

    modules, created, material = _realize_graph()
    try:
        admission = modules["admission"].__dict__
        standalone = modules["standalone"].__dict__
        live = modules["live"].__dict__
        adapter = modules["adapter"].__dict__
        r2 = modules["r2"].__dict__
        transaction = modules["transaction"].__dict__

        admission["validate_admission_spec"](material["standalone_admission_spec"])
        validated_material = transaction["validate_material"](material)
        if not isinstance(validated_material, dict):
            raise RuntimeError("transaction runtime material validation returned an invalid result")

        static_prompt = live["build_static_system_prompt"](
            material["compiler_spec"],
            material["standalone_admission_spec"],
        )
        if not isinstance(static_prompt, str) or not static_prompt:
            raise RuntimeError("live-semantics static prompt rehearsal failed")

        request_budget = adapter["RequestBudget"]()
        if (
            request_budget.maximum_total_model_requests != 240
            or request_budget.attempted_model_requests != 0
        ):
            raise RuntimeError("request-adapter budget rehearsal drifted")

        module_checks = (
            (admission["AdmittedTerminalOutput"], _OUTPUT_ADMISSION_MODULE),
            (standalone["RuntimeTurnRequest"], _STANDALONE_MODULE),
            (adapter["RequestReceipt"], _REQUEST_ADAPTER_MODULE),
            (r2["MetricSnapshotObservation"], _R2_RUNTIME_MODULE),
        )
        for value, expected_module in module_checks:
            if getattr(value, "__module__", None) != expected_module:
                raise RuntimeError("dataclass module identity rehearsal failed")

        if _execute_runtime_source(b"raise SystemExit(0)\n") != 0:
            raise RuntimeError("SystemExit(0) rehearsal failed")
        nonzero_propagated = False
        try:
            _execute_runtime_source(b"raise SystemExit(3)\n")
        except SystemExit as exc:
            nonzero_propagated = exc.code == 3
        if not nonzero_propagated:
            raise RuntimeError("nonzero SystemExit rehearsal failed")

        cleanup_probe = _bootstrap_failure_cleanup_probe()
        if not cleanup_probe:
            raise RuntimeError("bootstrap failure cleanup rehearsal failed")

        return {
            "status": "V2_TRANSACTION_WRAPPER_STRUCTURAL_REHEARSAL_PASS",
            "loaded_runtime_module_count": len(modules),
            "material_validated": True,
            "dataclass_module_identity_validated": True,
            "package_import_graph_validated": True,
            "request_budget_validated": True,
            "system_exit_zero_handled": True,
            "nonzero_system_exit_propagated": True,
            "bootstrap_failure_cleanup_validated": True,
            "live_execution_enabled": False,
            "model_requests_performed": 0,
            "gpu_execution_performed": False,
            "kaggle_execution_performed": False,
            "new_execution_authorized": False,
        }
    finally:
        _cleanup(created)


def main() -> int:
    result = rehearse()
    print(_canonical_json(result))
    return 0


def execute() -> int:
    admission = admit()
    print(_canonical_json(admission))
    modules, created, _material = _realize_graph()
    try:
        result = modules["transaction"].__dict__["main"]()
        if result not in (None, 0):
            raise SystemExit(result)
        return 0
    finally:
        _cleanup(created)


if __name__ == "__main__":
    raise SystemExit(execute())
```
